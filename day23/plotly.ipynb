## 🎯 Plotly'nin YBS'deki Rolü

> **"Bir grafiğe bakmak ile grafikle etkileşime girmek arasında ciddi bir fark vardır."**

### Klasik Yaklaşım vs YBS Yaklaşımı

| Özellik | Klasik Grafik (Matplotlib/Seaborn) | Plotly (Karar Destek) |
|---------|-----------------------------------|----------------------|
| **Amaç** | Veriyi göstermek | Kullanıcıyla konuşmak |
| **Etkileşim** | Statik, sadece bakılır | Dinamik, keşfedilir |
| **Kullanım** | Rapor, sunum, makale | Dashboard, yönetici paneli, web uygulaması |
| **Soru sorma** | Yeni grafik gerekir | Grafiğin üstünde yapılır |
| **Karar süreci** | "Bakıp tahmin et" | "Keşfedip doğrula" |

---

## 💡 YBS Örneği: Rapor Okumak vs Sistemi Yönetmek

**Senaryo:** E-ticaret platformunuzda öğle saatlerinde sipariş sayısı artıyor ve sistem yavaşlıyor.

## 🎯 İlk Plotly Grafiğimiz: Express vs Objects

Aynı grafiği iki farklı yöntemle çizelim ve farkı görelim.

````python
# Basit veri: E-ticaret platformunda 5 ürün kategorisi için aylık satış

kategoriler = ['Elektronik', 'Giyim', 'Ev & Yaşam', 'Kitap & Hobi', 'Spor']
satis_adet = [450, 320, 280, 190, 160]

# DataFrame oluşturalım (Plotly Express bunu sever)
df_satis = pd.DataFrame({
    'Kategori': kategoriler,
    'Satış Adet': satis_adet
})

print("📊 Veri Hazır:")
print(df_satis)
````

### 🚀 Yöntem 1: Plotly Express (Hızlı ve Kolay)

````python
# Plotly Express ile TEK SATIRDA bar chart

fig_express = px.bar(
    df_satis,  # Veri kaynağı
    x='Kategori',  # X ekseni
    y='Satış Adet',  # Y ekseni
    title='Ürün Kategorilerine Göre Aylık Satış - Plotly Express',
    color='Satış Adet',  # Renklendirme (otomatik gradient)
    color_continuous_scale='Blues',  # Mavi tonları
    text='Satış Adet'  # Bar üzerinde değer göster
)

# Metin formatı (bar üzerindeki sayılar)
fig_express.update_traces(textposition='outside')

# Layout düzenleme
fig_express.update_layout(
    height=500,
    showlegend=False
)

fig_express.show()

print("\n💡 Plotly Express: Sadece 1 satır kod, otomatik güzel tasarım!")
````

### 🎨 Yöntem 2: Graph Objects (Detaylı Kontrol)

````python
# Graph Objects ile ADIM ADIM bar chart

# 1. Figure objesi oluştur
fig_objects = go.Figure()

# 2. Bar trace ekle
fig_objects.add_trace(
    go.Bar(
        x=df_satis['Kategori'],
        y=df_satis['Satış Adet'],
        text=df_satis['Satış Adet'],
        textposition='outside',
        marker=dict(
            color=df_satis['Satış Adet'],  # Renk veri tabanlı
            colorscale='Greens',  # Yeşil tonları
            showscale=True,  # Renk skalası göster
            colorbar=dict(title='Adet')
        ),
        hovertemplate='<b>%{x}</b><br>Satış: %{y} adet<extra></extra>'  # Özel hover
    )
)

# 3. Layout ayarları (her detay kontrol edilebilir)
fig_objects.update_layout(
    title=dict(
        text='Ürün Kategorilerine Göre Aylık Satış - Graph Objects',
        font=dict(size=18, color='#2F4F4F')
    ),
    xaxis=dict(
        title='Kategoriler',
        tickangle=-45
    ),
    yaxis=dict(
        title='Satış Adedi',
        gridcolor='lightgray'
    ),
    height=500,
    template='plotly_white'
)

fig_objects.show()

print("\n🎨 Graph Objects: Daha fazla kod, ama HER detayı kontrol edebiliriz!")
````

## 🔨 **Veri Seti Oluşturma**

12,356 satırlık gerçekçi e-ticaret verisi oluşturacağız. Bu veri 3 aylık (90 gün) operasyonu temsil eder.

### 💡 YBS Perspektifi
Gerçek sistemlerde veri **temiz** değildir. Bizim senaryomuzda:
- ✅ Farklı müşteri segmentleri farklı davranır
- ✅ Kurumsal müşteriler daha yüksek tutar harcar
- ✅ Teslimat süresi uzadıkça memnuniyet düşer
- ✅ Bazı kargo firmaları daha hızlı teslimat yapar
- ✅ Hafta sonları sipariş yoğunluğu artar

````python
# 12,356 satırlık gerçekçi e-ticaret veri seti oluşturma
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

# Rastgelelik için seed (tekrarlanabilir sonuçlar için)
np.random.seed(42)

# Veri boyutu
n_records = 12356

print(f"🔨 {n_records:,} satırlık veri seti oluşturuluyor...")

# 1. Tarih ve Saat - 3 aylık veri (90 gün)
baslangic = datetime(2025, 11, 1, 8, 0, 0)
tarih_saat = []

# Saatlik dağılım olasılıkları (toplamı 1.0 olmalı)
saat_olasilik = np.array([0.01, 0.01, 0.01, 0.01, 0.01, 0.01, 0.02, 0.03,  # 00-07
                          0.05, 0.06, 0.07, 0.08, 0.09, 0.09, 0.08, 0.07,  # 08-15
                          0.06, 0.05, 0.04, 0.04, 0.03, 0.03, 0.02, 0.01])  # 16-23
# Normalize et (toplamı tam 1.0 yap)
saat_olasilik = saat_olasilik / saat_olasilik.sum()

for _ in range(n_records):
    # Hafta içi yoğunluk daha fazla, hafta sonu biraz düşük
    gun_offset = int(np.random.randint(0, 90))
    saat_offset = int(np.random.choice(range(24), p=saat_olasilik))
    dakika_offset = int(np.random.randint(0, 60))
    saniye_offset = int(np.random.randint(0, 60))
    
    tarih = baslangic + timedelta(
        days=gun_offset,
        hours=saat_offset,
        minutes=dakika_offset,
        seconds=saniye_offset
    )
    tarih_saat.append(tarih)

# 2. Müşteri Tipi - %75 Bireysel, %25 Kurumsal
musteri_tipi = np.random.choice(
    ['Bireysel', 'Kurumsal'],
    size=n_records,
    p=[0.75, 0.25]
)

# 3. Ürün Kategorisi
urun_kategori = np.random.choice(
    ['Elektronik', 'Giyim', 'Ev & Yaşam', 'Kitap & Hobi', 'Spor'],
    size=n_records,
    p=[0.30, 0.25, 0.20, 0.15, 0.10]
)

# 4. Ödeme Yöntemi
odeme_yontemi = np.random.choice(
    ['Kredi Kartı', 'Havale', 'Kapıda Ödeme'],
    size=n_records,
    p=[0.60, 0.25, 0.15]
)

# 5. Kargo Firması
kargo_firma = np.random.choice(
    ['Hızlı Kargo', 'Süper Kargo', 'Express Lojistik'],
    size=n_records,
    p=[0.45, 0.35, 0.20]
)

# 6. Bölge
bolge = np.random.choice(
    ['İstanbul', 'Ankara', 'İzmir', 'Diğer Şehirler'],
    size=n_records,
    p=[0.40, 0.20, 0.15, 0.25]
)

# 7. Sipariş Tutarı - Müşteri tipine ve kategoriye göre değişken
siparis_tutar = []
for i in range(n_records):
    # Baz fiyat kategoriye göre
    if urun_kategori[i] == 'Elektronik':
        baz = 500
        std = 800
    elif urun_kategori[i] == 'Giyim':
        baz = 150
        std = 100
    elif urun_kategori[i] == 'Ev & Yaşam':
        baz = 200
        std = 150
    elif urun_kategori[i] == 'Kitap & Hobi':
        baz = 80
        std = 50
    else:  # Spor
        baz = 250
        std = 200
    
    # Kurumsal müşteriler 2-3x daha fazla harcar
    if musteri_tipi[i] == 'Kurumsal':
        baz *= 2.5
        std *= 2
    
    tutar = np.random.normal(baz, std)
    tutar = max(20, tutar)  # Minimum 20 TL
    siparis_tutar.append(round(tutar, 2))

siparis_tutar = np.array(siparis_tutar)

# 8. Teslimat Süresi - Kargo firmasına ve bölgeye göre değişken
teslimat_sure = []
for i in range(n_records):
    # Baz süre kargo firmasına göre
    if kargo_firma[i] == 'Hızlı Kargo':
        baz_sure = 2
    elif kargo_firma[i] == 'Süper Kargo':
        baz_sure = 3
    else:  # Express Lojistik
        baz_sure = 2
    
    # Bölgeye göre ekleme
    if bolge[i] == 'İstanbul':
        ekleme = 0
    elif bolge[i] in ['Ankara', 'İzmir']:
        ekleme = 1
    else:
        ekleme = np.random.randint(1, 3)
    
    sure = baz_sure + ekleme + np.random.randint(-1, 2)
    sure = max(1, min(7, sure))  # 1-7 gün arası
    teslimat_sure.append(sure)

teslimat_sure = np.array(teslimat_sure)

# 9. Sipariş Durumu - Çoğunlukla tamamlanır
# Yüksek tutarlı siparişlerde iade oranı biraz daha yüksek
siparis_durum = []
for i in range(n_records):
    if siparis_tutar[i] > 2000:
        durum = np.random.choice(
            ['Tamamlandı', 'İade', 'İptal'],
            p=[0.82, 0.12, 0.06]
        )
    else:
        durum = np.random.choice(
            ['Tamamlandı', 'İade', 'İptal'],
            p=[0.90, 0.06, 0.04]
        )
    siparis_durum.append(durum)

siparis_durum = np.array(siparis_durum)

# 10. Memnuniyet Puanı - Sadece tamamlanan siparişler için
# Teslimat süresi uzadıkça memnuniyet düşer
memnuniyet_puan = []
for i in range(n_records):
    if siparis_durum[i] == 'Tamamlandı':
        # Baz puan: 4.5
        baz_puan = 4.5
        
        # Teslimat süresine göre azalma
        sure_ceza = (teslimat_sure[i] - 2) * 0.2
        
        # Kargo firmasına göre bonus
        if kargo_firma[i] == 'Hızlı Kargo':
            firma_bonus = 0.2
        elif kargo_firma[i] == 'Express Lojistik':
            firma_bonus = 0.3
        else:
            firma_bonus = 0
        
        puan = baz_puan - sure_ceza + firma_bonus + np.random.normal(0, 0.3)
        puan = max(1.0, min(5.0, puan))  # 1-5 arası
        memnuniyet_puan.append(round(puan, 1))
    else:
        # İade veya iptal edilmiş siparişler için düşük puan
        if siparis_durum[i] == 'İade':
            puan = np.random.uniform(2.0, 3.5)
        else:  # İptal
            puan = np.random.uniform(1.0, 2.5)
        memnuniyet_puan.append(round(puan, 1))

memnuniyet_puan = np.array(memnuniyet_puan)

# DataFrame oluşturma
df = pd.DataFrame({
    'siparis_id': range(1001, 1001 + n_records),
    'tarih_saat': tarih_saat,
    'musteri_tipi': musteri_tipi,
    'urun_kategori': urun_kategori,
    'siparis_tutar': siparis_tutar,
    'odeme_yontemi': odeme_yontemi,
    'kargo_firma': kargo_firma,
    'teslimat_sure': teslimat_sure,
    'siparis_durum': siparis_durum,
    'memnuniyet_puan': memnuniyet_puan,
    'bolge': bolge
})

# Tarihe göre sıralama (kronolojik)
df = df.sort_values('tarih_saat').reset_index(drop=True)

# Tarih bilgileri ekleme (analiz için)
df['tarih'] = df['tarih_saat'].dt.date
df['saat'] = df['tarih_saat'].dt.hour
df['gun_adi'] = df['tarih_saat'].dt.day_name()
# Hafta numarası için en uyumlu yöntem
df['hafta'] = df['tarih_saat'].dt.strftime('%U').astype(int)

print("✅ Veri seti başarıyla oluşturuldu!")
print(f"📊 Toplam kayıt: {len(df):,}")
print(f"📅 Tarih aralığı: {df['tarih_saat'].min()} - {df['tarih_saat'].max()}")

# Veri setini CSV olarak kaydet (locale dosya olarak)
csv_dosya_yolu = 'eticaret_siparis_verisi.csv'
df.to_csv(csv_dosya_yolu, index=False, encoding='utf-8-sig')
print(f"\n💾 Veri seti CSV olarak kaydedildi: {csv_dosya_yolu}")
print(f"📁 Dosya boyutu: {len(df):,} satır x {len(df.columns)} sütun")
````

````python
# Temel İş Metrikleri (KPI) Hesaplama

print("🎯 E-TİCARET SİSTEMİ - TEMEL İŞ METRİKLERİ")
print("=" * 70)

# 1. Toplam Gelir
toplam_gelir = df['siparis_tutar'].sum()
print(f"\n💰 Toplam Gelir: {toplam_gelir:,.2f} TL")

# 2. Ortalama Sepet Değeri (Average Order Value - AOV)
ortalama_sepet = df['siparis_tutar'].mean()
print(f"🛒 Ortalama Sepet Değeri: {ortalama_sepet:,.2f} TL")

# 3. Sipariş Tamamlanma Oranı (Order Completion Rate)
tamamlanma_orani = (df['siparis_durum'] == 'Tamamlandı').sum() / len(df) * 100
print(f"✅ Sipariş Tamamlanma Oranı: %{tamamlanma_orani:.2f}")

# 4. İade Oranı (Return Rate)
iade_orani = (df['siparis_durum'] == 'İade').sum() / len(df) * 100
print(f"↩️ İade Oranı: %{iade_orani:.2f}")

# 5. İptal Oranı (Cancellation Rate)
iptal_orani = (df['siparis_durum'] == 'İptal').sum() / len(df) * 100
print(f"❌ İptal Oranı: %{iptal_orani:.2f}")

# 6. Ortalama Teslimat Süresi
ortalama_teslimat = df['teslimat_sure'].mean()
print(f"📦 Ortalama Teslimat Süresi: {ortalama_teslimat:.1f} gün")

# 7. Ortalama Müşteri Memnuniyeti
ortalama_memnuniyet = df['memnuniyet_puan'].mean()
print(f"⭐ Ortalama Memnuniyet Puanı: {ortalama_memnuniyet:.2f} / 5.00")

# 8. Günlük Ortalama Sipariş
gunluk_siparis = len(df) / 90  # 90 gün
print(f"📊 Günlük Ortalama Sipariş: {gunluk_siparis:.0f} adet")

# 9. Müşteri Segmentasyonu
print(f"\n👥 MÜŞTERİ SEGMENTİ ANALİZİ:")
print("-" * 70)
segment_analiz = df.groupby('musteri_tipi').agg({
    'siparis_tutar': ['count', 'sum', 'mean']
}).round(2)
segment_analiz.columns = ['Sipariş Sayısı', 'Toplam Gelir (TL)', 'Ortalama Tutar (TL)']
print(segment_analiz)

# 10. En Karlı Ürün Kategorisi
print(f"\n📦 ÜRÜN KATEGORİ PERFORMANSI:")
print("-" * 70)
kategori_analiz = df.groupby('urun_kategori').agg({
    'siparis_tutar': ['count', 'sum', 'mean']
}).round(2)
kategori_analiz.columns = ['Sipariş Sayısı', 'Toplam Gelir (TL)', 'Ortalama Tutar (TL)']
kategori_analiz = kategori_analiz.sort_values('Toplam Gelir (TL)', ascending=False)
print(kategori_analiz)

print("\n" + "=" * 70)
print("✅ İş metrikleri hesaplandı!")
````

# 4️⃣ **Scatter Plot (Dağılım Grafiği) ile Anomali Tespiti** 🔍

## 🎯 **YBS Sorusu: "Hangi siparişler normal dışı?"**

**Scatter Plot**, iki sayısal değişken arasındaki ilişkiyi gösterir. Anomaliler (outliers) kolayca fark edilir!

### 📊 **Veriyi Hazırlayalım**

````python
# Tamamlanmış siparişleri filtreleyelim
scatter_data = df[df['siparis_durum'] == 'Tamamlandı'].copy()

# Anomali tespiti için eşik değerler belirleyelim
# Normal dağılımda: ortalama ± 2*standart sapma
teslimat_ort = scatter_data['teslimat_sure'].mean()
teslimat_std = scatter_data['teslimat_sure'].std()
memnuniyet_ort = scatter_data['memnuniyet_puan'].mean()

# Anomali kategorisi oluştur
def anomali_tespit(row):
    """
    Anomali Tespiti Kuralları:
    1. Teslimat çok uzun (>ortalama + 2*std) AMA memnuniyet yüksek (>4) → Şüpheli
    2. Teslimat hızlı (<ortalama - 1*std) AMA memnuniyet düşük (<3) → Şüpheli
    3. Diğerleri → Normal
    """
    teslimat = row['teslimat_sure']
    memnuniyet = row['memnuniyet_puan']
    
    if teslimat > (teslimat_ort + 2*teslimat_std) and memnuniyet > 4.0:
        return 'Anomali: Uzun Teslimat × Yüksek Memnuniyet'
    elif teslimat < (teslimat_ort - teslimat_std) and memnuniyet < 3.0:
        return 'Anomali: Hızlı Teslimat × Düşük Memnuniyet'
    else:
        return 'Normal'

scatter_data['Anomali Durumu'] = scatter_data.apply(anomali_tespit, axis=1)

print("🔍 Anomali Analizi:")
print(scatter_data['Anomali Durumu'].value_counts())
print(f"\n📊 Toplam Sipariş: {len(scatter_data)}")
print(f"⚠️ Anomali Sayısı: {len(scatter_data[scatter_data['Anomali Durumu'] != 'Normal'])}")
print(f"📌 Anomali Oranı: {100 * len(scatter_data[scatter_data['Anomali Durumu'] != 'Normal']) / len(scatter_data):.2f}%")
````

````python
# Plotly Express Scatter Plot - Anomali Tespiti

fig = px.scatter(
    scatter_data,
    x='teslimat_sure',
    y='memnuniyet_puan',
    color='Anomali Durumu',
    size='siparis_tutar',  # Sipariş tutarı ne kadar büyükse nokta o kadar büyük
    hover_data=['siparis_id', 'urun_kategori', 'musteri_tipi', 'kargo_firma', 'siparis_tutar'],
    title='Teslimat Süresi vs Memnuniyet Puanı - Anomali Analizi',
    labels={
        'teslimat_sure': 'Teslimat Süresi (Gün)',
        'memnuniyet_puan': 'Müşteri Memnuniyeti (1-5)',
        'Anomali Durumu': 'Anomali Durumu',
        'siparis_tutar': 'Sipariş Tutarı (TL)'
    },
    color_discrete_map={
        'Normal': '#2ecc71',  # Yeşil - Normal
        'Anomali: Uzun Teslimat × Yüksek Memnuniyet': '#e74c3c',  # Kırmızı
        'Anomali: Hızlı Teslimat × Düşük Memnuniyet': '#f39c12'  # Turuncu
    },
    opacity=0.7
)

# Grafik özelleştirme
fig.update_traces(
    marker=dict(
        line=dict(width=1, color='white')
    ),
    hovertemplate='<b>Sipariş: %{customdata[0]}</b><br>' +
                  'Teslimat: %{x} gün<br>' +
                  'Memnuniyet: %{y}/5<br>' +
                  'Kategori: %{customdata[1]}<br>' +
                  'Müşteri: %{customdata[2]}<br>' +
                  'Kargo: %{customdata[3]}<br>' +
                  'Tutar: %{customdata[4]:.0f} TL<extra></extra>'
)

fig.update_layout(
    height=700,
    template='plotly_white',
    font=dict(size=12),
    title_font=dict(size=16, color='#2c3e50'),
    xaxis=dict(
        showgrid=True,
        gridwidth=1,
        gridcolor='lightgray'
    ),
    yaxis=dict(
        showgrid=True,
        gridwidth=1,
        gridcolor='lightgray'
    ),
    legend=dict(
        title=dict(text='Anomali Durumu'),
        orientation='v',
        yanchor='top',
        y=1,
        xanchor='left',
        x=1.02
    )
)

# Ortalama çizgilerini ekleyelim
fig.add_hline(
    y=memnuniyet_ort, 
    line_dash="dash", 
    line_color="blue",
    annotation_text=f"Ort. Memnuniyet: {memnuniyet_ort:.2f}",
    annotation_position="right"
)

fig.add_vline(
    x=teslimat_ort,
    line_dash="dash",
    line_color="red",
    annotation_text=f"Ort. Teslimat: {teslimat_ort:.1f} gün",
    annotation_position="top"
)

fig.show()

print("\n✨ Bu Grafikteki İNTERAKTİF Özellikler:")
print("🔍 ZOOM IN: Anomali noktalarını yakından inceleyin")
print("👆 HOVER: Her noktanın detaylarını görün (Sipariş ID, Kategori, Kargo)")
print("🎨 LEGEND: Tıklayarak Normal/Anomali noktalarını göster/gizle")
print("📊 SIZE: Nokta boyutu = Sipariş tutarı (Büyük nokta = Yüksek tutar)")
print("📏 REFERANS ÇİZGİLERİ: Mavi=Ortalama Memnuniyet, Kırmızı=Ortalama Teslimat")
````

# 5️⃣ **Box Plot ve Violin Plot ile Dağılım Analizi** 📦🎻

## 🎯 YBS Sorusu: "Hangi kargo firması daha tutarlı?"

**İş Senaryosu:**  
Operasyon Müdürü diyor ki:

> "3 kargo firmamız var. Hangisi daha güvenilir? Sadece ortalamaya bakmak yetmez!  
> Teslimat sürelerinin **kararlılığını** da görmem lazım. Hangisi tahmin edilebilir?"

**Box Plot ve Violin Plot**, verilerin dağılımını gösterir:
- **Medyan** (ortanca değer)
- **Çeyrekler** (Q1, Q3)
- **Outliers** (aykırı değerler)
- **Dağılım şekli** (Violin sayesinde)

### 📊 **Veriyi Hazırlayalım**

````python
# Kargo firmaları için teslimat süresi istatistikleri

kargo_istatistik = df.groupby('kargo_firma').agg({
    'teslimat_sure': ['mean', 'median', 'std', 'min', 'max', 'count']
}).round(2)

kargo_istatistik.columns = ['Ortalama', 'Medyan', 'Std Sapma', 'Min', 'Max', 'Sipariş Sayısı']
kargo_istatistik = kargo_istatistik.sort_values('Ortalama')

print("📦 Kargo Firmaları - Teslimat Süresi İstatistikleri:")
print("="*70)
print(kargo_istatistik)
print("\n💡 İpucu:")
print("   - Düşük STD SAPMA = Tutarlı teslimat (Tahmin edilebilir)")
print("   - Yüksek STD SAPMA = Değişken teslimat (Risk var)")
print("   - Min-Max ARALIK = Teslimat aralığının genişliği")
````

### 📦 **Box Plot - Klasik Dağılım Grafiği**

````python
# Plotly Express ile Box Plot

fig = px.box(
    df,
    x='kargo_firma',
    y='teslimat_sure',
    color='kargo_firma',
    title='Kargo Firmalarına Göre Teslimat Süresi Dağılımı - Box Plot',
    labels={
        'kargo_firma': 'Kargo Firması',
        'teslimat_sure': 'Teslimat Süresi (Gün)'
    },
    color_discrete_sequence=['#3498db', '#e74c3c', '#f39c12'],
    points='outliers',  # Sadece outlier noktaları göster
    hover_data=['siparis_id', 'urun_kategori', 'siparis_durum']
)

fig.update_traces(
    marker=dict(size=5, opacity=0.6),
    boxmean='sd'  # Ortalama ve standart sapma çizgilerini göster
)

fig.update_layout(
    height=600,
    template='plotly_white',
    font=dict(size=12),
    title_font=dict(size=16, color='#2c3e50'),
    showlegend=False,
    xaxis=dict(title_font=dict(size=14)),
    yaxis=dict(title_font=dict(size=14), showgrid=True, gridcolor='lightgray')
)

fig.show()

print("\n📦 Box Plot Nasıl Okunur?")
print("┌─────────────────────────────────────────┐")
print("│  ○  ← Outlier (Aykırı değer)           │")
print("│  ┬  ← Max (Q3 + 1.5*IQR)                │")
print("│  ┴  ← Q3 (3. Çeyrek - %75)              │")
print("│ ███ ← BOX (IQR - Interquartile Range)   │")
print("│ ━━━ ← Medyan (%50)                      │")
print("│ ███                                     │")
print("│  ┬  ← Q1 (1. Çeyrek - %25)              │")
print("│  ┴  ← Min (Q1 - 1.5*IQR)                │")
print("│  ○  ← Outlier                           │")
print("└─────────────────────────────────────────┘")
print("\n🎯 YBS Yorumu:")
print("   Box ne kadar KISA ise, teslimat o kadar TUTARLI!")
print("   Outlier çok ise, bazen BÜYÜK GECİKMELER yaşanıyor!")
````

### 🎻 **Violin Plot - Daha Detaylı Dağılım Grafiği**

**Fark:** Violin Plot, Box Plot'a göre dağılımın **şeklini** de gösterir. Hangi değerlerde yoğunlaşma var?

````python
# Plotly Express ile Violin Plot

fig = px.violin(
    df,
    x='kargo_firma',
    y='teslimat_sure',
    color='kargo_firma',
    title='Kargo Firmalarına Göre Teslimat Süresi Dağılımı - Violin Plot',
    labels={
        'kargo_firma': 'Kargo Firması',
        'teslimat_sure': 'Teslimat Süresi (Gün)'
    },
    color_discrete_sequence=['#3498db', '#e74c3c', '#f39c12'],
    box=True,  # İçinde box plot da göster
    points='outliers',  # Outlier noktaları göster
    hover_data=['siparis_id', 'urun_kategori']
)

fig.update_traces(
    marker=dict(size=4, opacity=0.5),
    meanline_visible=True  # Ortalama çizgisini göster
)

fig.update_layout(
    height=600,
    template='plotly_white',
    font=dict(size=12),
    title_font=dict(size=16, color='#2c3e50'),
    showlegend=False,
    xaxis=dict(title_font=dict(size=14)),
    yaxis=dict(title_font=dict(size=14), showgrid=True, gridcolor='lightgray')
)

fig.show()

print("\n🎻 Violin Plot Nasıl Okunur?")
print("┌─────────────────────────────────────────┐")
print("│        ╱╲  ← Geniş = Yoğun bölge        │")
print("│       ╱  ╲ ← Buradan çok veri var!      │")
print("│      ╱ ██ ╲ ← Box plot (içinde)         │")
print("│     ╱  ━━  ╲ ← Medyan çizgisi           │")
print("│    ╱   ██   ╲                           │")
print("│   ╱          ╲ ← Dar = Az veri          │")
print("│  ╱            ╲                         │")
print("└─────────────────────────────────────────┘")
print("\n💡 Violin vs Box:")
print("   📦 Box Plot: ÇEYREKLERİ gösterir (25%, 50%, 75%)")
print("   🎻 Violin Plot: DAĞILIM ŞEKLİNİ gösterir (yoğunluk)")
print("\n🎯 YBS Yorumu:")
print("   Violin şekli SİMETRİK ise → Normal dağılım var")
print("   Violin ÇOK KABARIKKSA → Çoğu sipariş bu sürede geliyor")
````

### 🔄 **Bonus: Box + Violin Yan Yana Karşılaştırma**

**Aynı grafikte hem Box hem Violin gösterelim!**

````python
# Graph Objects ile Box ve Violin yan yana

from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Box Plot', 'Violin Plot'),
    horizontal_spacing=0.12
)

renkler = {'Hızlı Kargo': '#3498db', 'Süper Kargo': '#e74c3c', 'Express Lojistik': '#f39c12'}

# Box Plot (Sol)
for kargo in df['kargo_firma'].unique():
    veri_filtre = df[df['kargo_firma'] == kargo]
    
    fig.add_trace(
        go.Box(
            y=veri_filtre['teslimat_sure'],
            name=kargo,
            marker_color=renkler[kargo],
            boxmean='sd',
            showlegend=False
        ),
        row=1, col=1
    )

# Violin Plot (Sağ)
for kargo in df['kargo_firma'].unique():
    veri_filtre = df[df['kargo_firma'] == kargo]
    
    fig.add_trace(
        go.Violin(
            y=veri_filtre['teslimat_sure'],
            name=kargo,
            marker_color=renkler[kargo],
            box_visible=True,
            meanline_visible=True,
            showlegend=True
        ),
        row=1, col=2
    )

# Layout
fig.update_layout(
    title_text='Kargo Firmaları: Box vs Violin Karşılaştırması',
    title_font=dict(size=16, color='#2c3e50'),
    height=600,
    template='plotly_white',
    font=dict(size=11),
    legend=dict(
        title=dict(text='Kargo Firması'),
        orientation='h',
        yanchor='bottom',
        y=-0.2,
        xanchor='center',
        x=0.5
    )
)

fig.update_yaxes(title_text='Teslimat Süresi (Gün)', row=1, col=1)
fig.update_yaxes(title_text='Teslimat Süresi (Gün)', row=1, col=2)

fig.show()

print("\n🔄 İki Görsel Bir Arada:")
print("✅ SOL (Box): Çeyrekler ve outlier'ları NET gösterir")
print("✅ SAĞ (Violin): Dağılımın şeklini ve yoğunluğu gösterir")
print("\n💡 YBS: İkisini birlikte kullanmak en iyi sonucu verir!")
````

# 8️⃣ **Heatmap ile Korelasyon ve Pivot Analizi** 🔥

## 🎯 YBS Sorusu: "Hangi değişkenler birbiriyle ilişkili?"

**İş Senaryosu:**  
Veri Analisti diyor ki:

> Sipariş tutarı ile ödeme yöntemi arasında bir pattern var mı?  
> Hangi gün-saat kombinasyonunda en çok sipariş geliyor?"

**Heatmap**, iki boyutlu veriyi renklerle gösterir. Koyu renk = yüksek değer, açık renk = düşük değer.  
**YBS açısından:** İlişki keşfi, pivot analizi, performans matrisi!

---

### 🔥 Korelasyon Heatmap'i

````python
# Sayısal değişkenler için korelasyon matrisi hesaplayalım
sayisal_kolonlar = ['siparis_tutar', 'teslimat_sure', 'memnuniyet_puan', 'saat']

korelasyon_matrix = df[sayisal_kolonlar].corr()

print("🔗 Korelasyon Matrisi:")
print(korelasyon_matrix)
print("\n💡 Korelasyon değerleri:")
print("   +1.0 = Tam pozitif ilişki")
print("    0.0 = İlişki yok")
print("   -1.0 = Tam negatif ilişki")
````

````python
# Plotly Express ile Korelasyon Heatmap

fig = px.imshow(
    korelasyon_matrix,
    text_auto='.2f',  # Değerleri göster, 2 ondalık
    aspect='auto',
    color_continuous_scale='RdBu_r',  # Kırmızı-Mavi renk skalası
    title='Sayısal Değişkenler Korelasyon Matrisi',
    labels={'color': 'Korelasyon'}
)

fig.update_xaxes(side='bottom')

fig.update_layout(
    height=550,
    font=dict(size=12),
    title_font=dict(size=16, color='#2c3e50'),
)

fig.show()

print("\n🔥 Heatmap Nasıl Okunur?")
print("✅ Koyu Kırmızı: Güçlü pozitif korelasyon (+1'e yakın)")
print("✅ Beyaz/Açık: Zayıf/Yok korelasyon (0'a yakın)")
print("✅ Koyu Mavi: Güçlü negatif korelasyon (-1'e yakın)")
print("\n🎯 YBS Yorumu:")
print("   teslimat_sure ↔ memnuniyet_puan negatif ise:")
print("      → Teslimat uzayınca memnuniyet düşüyor!")
print("   Buna göre lojistik iyileştirmesi yapılmalı!")
````

### 📅 **Pivot Heatmap: Gün-Saat Sipariş Yoğunluğu**

**Soru:** Hangi gün ve saatte en çok sipariş geliyor?

````python
# Gün ve saate göre sipariş sayısı pivot table
gun_saat_pivot = df.groupby(['gun_adi', 'saat']).size().reset_index(name='Sipariş Sayısı')

# Pivot tabloya çevir
pivot_table = gun_saat_pivot.pivot(index='gun_adi', columns='saat', values='Sipariş Sayısı')

# Günleri sırala (Pazartesi'den başlasın)
gun_sirala = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
pivot_table = pivot_table.reindex(gun_sirala)

print("📊 Gün-Saat Pivot Tablosu:")
print(pivot_table.head())
print(f"\n📏 Boyut: {pivot_table.shape[0]} gün × {pivot_table.shape[1]} saat")
````

````python
# Plotly ile Pivot Heatmap

# Gün isimlerini Türkçeleştirelim
gun_ceviri = {
    'Monday': 'Pazartesi',
    'Tuesday': 'Salı',
    'Wednesday': 'Çarşamba',
    'Thursday': 'Perşembe',
    'Friday': 'Cuma',
    'Saturday': 'Cumartesi',
    'Sunday': 'Pazar'
}

pivot_table_tr = pivot_table.copy()
pivot_table_tr.index = [gun_ceviri[gun] for gun in pivot_table_tr.index]

fig = px.imshow(
    pivot_table_tr,
    labels={'x': 'Saat', 'y': 'Gün', 'color': 'Sipariş Sayısı'},
    x=pivot_table_tr.columns,
    y=pivot_table_tr.index,
    color_continuous_scale='YlOrRd',  # Sarı-Turuncu-Kırmızı
    title='Gün-Saat Bazında Sipariş Yoğunluğu Haritası',
    aspect='auto'
)

fig.update_xaxes(side='bottom', dtick=1)
fig.update_yaxes(dtick=1)

fig.update_traces(
    hovertemplate='<b>%{y}</b><br>Saat: %{x}:00<br>Sipariş: %{z}<extra></extra>'
)

fig.update_layout(
    height=500,
    font=dict(size=12),
    title_font=dict(size=16, color='#2c3e50'),
)

fig.show()

print("\n🔥 Heatmap Yorumlama:")
print("✅ Koyu Kırmızı Kareler = Pik saatler (En yoğun)")
print("✅ Sarı Kareler = Orta yoğunluk")
print("✅ Beyaz/Açık Kareler = Boş saatler")
print("\n💡 YBS Stratejisi:")
print("   Koyu kırmızı saatlere → Daha fazla çalışan")
print("   Açık sarı saatlere → Kampanya yapın")
print("   Cuma-Cumartesi yoğunsa → Hafta sonu indirimleri")
````
