### 📏 2.3 · Hata Ölçüleri: Başarıyı Nasıl Ölçeceğiz?

Hedef MPa cinsinden sürekli bir sayı olduğu için dört ölçüt kullanıyoruz:

| Ölçüt | Formül | Anlamı |
|:--|:--|:--|
| **MAE** (ortalama mutlak hata) | $\frac{1}{n}\sum \lvert y-\hat y\rvert$ | "Ortalama kaç MPa yanılıyoruz?" **Ana ölçütümüz**: yorumu doğrudan MPa |
| **RMSE** | $\sqrt{\frac{1}{n}\sum (y-\hat y)^2}$ | Büyük hataları daha çok cezalandırır |
| **R²** | $1-\frac{\sum (y-\hat y)^2}{\sum (y-\bar y)^2}$ | Varyansın ne kadarını açıkladık? (1 = kusursuz, 0 = hep ortalama) |
| **MdAPE** | $\operatorname{medyan}\!\left(\frac{\lvert y-\hat y\rvert}{y}\right)$ | Tipik **yüzde** hata |

Referans noktası olarak, **hep eğitim ortalamasını söyleyen** "taban model" kullanacağız: hiçbir model bundan kötü olmamalı.

Modele geçmeden önce dört soruyu soruyoruz: **eksik değer var mı? Tekrarlayan satır var mı? Fiziksel olarak imkânsız değer var mı? Satırlar birbirinden bağımsız mı?** Sonuncusu bu veride **en önemli** sorudur.

```python
kalite = pd.DataFrame({
    "eksik değer": ham.isna().sum(),
    "sıfır değer": (ham == 0).sum(),
    "sıfır oranı %": ((ham == 0).mean() * 100).round(1),
    "minimum": ham.min(), "maksimum": ham.max(),
})
display(kalite)

tam_tekrar = ham.duplicated().sum()
toplam_kutle = ham[HAM_OZELLIKLER[:7]].sum(axis=1)               # 1 m³ betonun toplam bileşen kütlesi (yoğunluk kontrolü)
print(f"\nTam tekrar eden satır sayısı          : {tam_tekrar}")
print(f"Bileşen toplamı (kg/m³) aralığı       : {toplam_kutle.min():.0f} – {toplam_kutle.max():.0f}  (taze beton yoğunluğu ≈ 2.200–2.550 beklenir)")
print(f"Negatif bileşen / yaş ≤ 0 var mı?     : {(ham[HAM_OZELLIKLER[:7]] < 0).any().any()} / {(ham['yas_gun'] <= 0).any()}")
rapor_kaydet(kalite, "veri_kalitesi")
```

**Okuma:** Eksik değer yok; negatif ya da imkânsız değer yok; bileşenlerin toplamı gerçekçi bir beton yoğunluğu aralığında kalıyor. Cüruf, uçucu kül ve akışkanlaştırıcıdaki **sıfırlar eksik veri değil**: o katkı reçetede **kullanılmamış** demektir (bu yüzden doldurma / silme yapmıyoruz).

### 🔗 3.1 · Gizli Yapı: Aynı Reçete, Farklı Yaşlar

1.030 satır gerçekten 1.030 **farklı** beton mu? Yedi bileşenin hepsi aynı olan satırları **tek bir reçete** sayalım:

```python
RECETE_KOLONLARI = HAM_OZELLIKLER[:7]
ham["recete_id"] = ham.groupby(RECETE_KOLONLARI).ngroup()          # aynı 7 bileşen = aynı reçete (numaralandır)

n_recete = ham["recete_id"].nunique()
grup_boyut = ham.groupby("recete_id").size()
print(f"Satır sayısı: {len(ham):,} · Farklı reçete sayısı: {n_recete}  →  reçete başına ortalama {len(ham) / n_recete:.2f} satır")
print(f"Reçetelerin %{(grup_boyut == 1).mean() * 100:.0f}'i tek ölçümlü; en kalabalık reçetede {grup_boyut.max()} ölçüm var.\n")

fig, eksenler = plt.subplots(1, 3, figsize=(16, 4))
grup_boyut.value_counts().sort_index().plot.bar(ax=eksenler[0], color=RENK["mor"], width=0.8)
eksenler[0].set(title="Reçete başına kaç ölçüm var?", xlabel="Ölçüm sayısı", ylabel="Reçete sayısı")
ham["yas_gun"].value_counts().sort_index().plot.bar(ax=eksenler[1], color=RENK["turuncu"], width=0.8)
eksenler[1].set(title="Numune yaşları (gün)", xlabel="Yaş", ylabel="Satır sayısı")
ornek = ham[ham["recete_id"] == grup_boyut.idxmax()].sort_values("yas_gun")
eksenler[2].plot(ornek["yas_gun"], ornek["dayanim"], "o-", color=RENK["yesil"], lw=2.4, ms=7)
eksenler[2].set(title=f"Tek bir reçete ({len(ornek)} ölçüm): dayanım yaşla artıyor", xlabel="Yaş (gün)", ylabel="Dayanım (MPa)", xscale="log")
fig.tight_layout()
kaydet(fig, "01_veri_kalitesi_ve_recete_aileleri")
plt.show()
```

### 🧪 5.1 · Deney: Rastgele Bölme Skoru Ne Kadar Şişiriyor?

Aynı modelleri **10 farklı tohumla**, iki bölme yöntemiyle (**rastgele** ve **reçete bazlı**) eğitip test hatalarına bakıyoruz.

```python
X_tum = ham[HAM_OZELLIKLER]
y_tum = ham["dayanim"]
gruplar_tum = ham["recete_id"]


def hat(model, olcekle=False, ozellik=True):
    """Sızıntısız pipeline: [özellik mühendisliği] → [ölçekleme] → model. Hepsi yalnızca eğitim verisine uydurulur."""
    adimlar = []
    if ozellik:
        adimlar.append(("ozellik", FunctionTransformer(muhendislik)))
    if olcekle:
        adimlar.append(("olcek", StandardScaler()))
    adimlar.append(("model", model))
    return Pipeline(adimlar)


sizinti_modelleri = {
    "Ridge (doğrusal)": hat(Ridge(alpha=1.0), olcekle=True),
    "Random Forest": hat(RandomForestRegressor(n_estimators=200, random_state=RANDOM_STATE, n_jobs=4)),
    "Gradient Boosting": hat(GradientBoostingRegressor(random_state=RANDOM_STATE)),
}
satirlar = []
for tohum in range(10):
    bolmeler = {
        "rastgele": next(iter([train_test_split(np.arange(len(ham)), test_size=0.2, random_state=tohum)])),
        "reçete bazlı": next(GroupShuffleSplit(1, test_size=0.2, random_state=tohum).split(X_tum, y_tum, gruplar_tum)),
    }
    for bolme_ad, (e_idx, t_idx) in bolmeler.items():
        for ad, model in sizinti_modelleri.items():
            m = clone(model).fit(X_tum.iloc[e_idx], y_tum.iloc[e_idx])
            s = metrikler(y_tum.iloc[t_idx], m.predict(X_tum.iloc[t_idx]))
            satirlar.append({"tohum": tohum, "bölme": bolme_ad, "model": ad, **s})

sizinti = pd.DataFrame(satirlar)
ozet_sizinti = sizinti.groupby(["model", "bölme"])[["MAE", "R²"]].mean().unstack().round(3)
display(ozet_sizinti)
rapor_kaydet(sizinti, "sizinti_deneyi", index=False)

fig, eksenler = plt.subplots(1, 2, figsize=(14, 4.4))
sns.boxplot(data=sizinti, x="model", y="MAE", hue="bölme", palette=[RENK["pembe"], RENK["yesil"]], ax=eksenler[0])
eksenler[0].set(title="Test MAE: rastgele bölme daha 'başarılı' görünür", xlabel="", ylabel="MAE (MPa)")
sns.boxplot(data=sizinti, x="model", y="R²", hue="bölme", palette=[RENK["pembe"], RENK["yesil"]], ax=eksenler[1])
eksenler[1].set(title="Test R²", xlabel="", ylabel="R²")
fig.tight_layout()
kaydet(fig, "06_sizinti_deneyi")
plt.show()
```

### ✂️ 5.2 · Nihai Bölme ve Çapraz Doğrulama Düzeni

Karar: **%80 eğitim / %20 test**, ama **reçeteye göre** (aynı reçete iki tarafa düşmez). Eğitim içindeki **5 katlı çapraz doğrulama** da reçeteye göre yapılır (`GroupKFold`, karıştırmalı). Test seti **model seçiminde kullanılmaz**; yalnızca raporlama içindir.

```python
egitim_idx, test_idx = next(GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=RANDOM_STATE).split(X_tum, y_tum, gruplar_tum))

X_egitim, X_test = X_tum.iloc[egitim_idx].reset_index(drop=True), X_tum.iloc[test_idx].reset_index(drop=True)
y_egitim, y_test = y_tum.iloc[egitim_idx].reset_index(drop=True), y_tum.iloc[test_idx].reset_index(drop=True)
g_egitim, g_test = gruplar_tum.iloc[egitim_idx].reset_index(drop=True), gruplar_tum.iloc[test_idx].reset_index(drop=True)
X_egitim_id = X_egitim.assign(recete_id=g_egitim)                 # yalnızca grup bilinçli yığın modeli için (bölüm 7)
X_test_id = X_test.assign(recete_id=g_test)

CV = GroupKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)           # tüm modeller aynı katları kullanır

assert set(g_egitim).isdisjoint(set(g_test)), "Aynı reçete hem eğitimde hem testte!"
print(f"Eğitim: {len(X_egitim)} satır / {g_egitim.nunique()} reçete  |  Test: {len(X_test)} satır / {g_test.nunique()} reçete")
print(f"Ortak reçete sayısı (olmamalı): {len(set(g_egitim) & set(g_test))}")
ks = stats.ks_2samp(y_egitim, y_test)
print(f"Dayanım dağılımı eğitim vs test → ort. {y_egitim.mean():.1f} / {y_test.mean():.1f} MPa · KS p-değeri {ks.pvalue:.2f} (küçük değilse benzer dağılım)")
print("Taban model (eğitim ortalaması) test MAE:", round(metrikler(y_test, np.full(len(y_test), y_egitim.mean()))["MAE"], 2), "MPa")
```

<p align="center">
  <img src="assets/proje_06.svg" width="100%" alt="Bölüm 6 — Pipeline ve Deneyler">
</p>

Ön işleme adımlarını **tek bir `Pipeline`** içine koyduk (bölüm 5.1'deki `hat` fonksiyonu): özellik mühendisliği → (gerekirse) ölçekleme → model. Böylece her çapraz doğrulama katında **yalnızca eğitim katına** uydurulur; sızıntı olmaz.

Önce şu soruların yanıtını **deneyle** arıyoruz (hepsi eğitim setinde, reçete bazlı 5 katlı CV ile):

1. 🧪 Türetilmiş **özellikler** (su/çimento, su/bağlayıcı, bağlayıcı, log yaş, agrega oranı) işe yarıyor mu?
2. 📏 **Ölçekleme** hangi modelleri etkiliyor?
3. 📐 **Hedefin logaritmasını** almak yardımcı mı?
4. 🗑️ 25 **tekrar eden satırı** silmek bir şey değiştiriyor mu?

```python
def cv_mae(model, X=None, y=None):
    """Reçete bazlı 5 katlı CV ile ortalama MAE (MPa). Her çağrıda aynı katlar kullanılır."""
    X = X_egitim if X is None else X
    y = y_egitim if y is None else y
    skorlar = cross_validate(model, X, y, groups=g_egitim, cv=CV, scoring="neg_mean_absolute_error", n_jobs=1)["test_score"]
    return -skorlar.mean()


def log_hedef(model):
    """Hedefi log ile dönüştürüp tahmini tekrar MPa'ya çeviren sarmalayıcı."""
    return TransformedTargetRegressor(regressor=model, func=np.log, inverse_func=np.exp)


# ── Deney 1: özellik mühendisliği (ham 8 sütun mu, 13 sütun mu?) ─────────────
d1 = []
for ad, model, olcek in [("Ridge", Ridge(1.0), True), ("SVR (RBF)", SVR(C=10), True),
                         ("Random Forest", RandomForestRegressor(200, random_state=RANDOM_STATE, n_jobs=4), False),
                         ("Gradient Boosting", GradientBoostingRegressor(random_state=RANDOM_STATE), False)]:
    d1.append({"model": ad, "ham 8 sütun": cv_mae(hat(clone(model), olcek, ozellik=False)), "+ mühendislik (13 sütun)": cv_mae(hat(clone(model), olcek, ozellik=True))})
d1 = pd.DataFrame(d1).set_index("model")
d1["iyileşme %"] = ((d1["ham 8 sütun"] - d1["+ mühendislik (13 sütun)"]) / d1["ham 8 sütun"] * 100)
display(d1.round(3))
```

```python
# ── Deney 2: ölçekleme (ağaçlar etkilenmez; mesafe/çekirdek/sinir ağı etkilenir) ─────────
d2 = []
for ad, model in [("KNN", KNeighborsRegressor(5)), ("SVR (RBF)", SVR(C=10)),
                  ("MLP", MLPRegressor(hidden_layer_sizes=(64, 32), max_iter=2000, random_state=RANDOM_STATE)),
                  ("Gradient Boosting", GradientBoostingRegressor(random_state=RANDOM_STATE))]:
    d2.append({"model": ad, "ölçeksiz": cv_mae(hat(clone(model), False)), "StandardScaler": cv_mae(hat(clone(model), True))})
d2 = pd.DataFrame(d2).set_index("model")

# ── Deney 3: log hedef ──────────────────────────────────────────────────
d3 = []
for ad, model, olcek in [("Ridge", Ridge(1.0), True), ("SVR (RBF)", SVR(C=10), True),
                         ("Random Forest", RandomForestRegressor(200, random_state=RANDOM_STATE, n_jobs=4), False),
                         ("Gradient Boosting", GradientBoostingRegressor(random_state=RANDOM_STATE), False)]:
    d3.append({"model": ad, "ham hedef (MPa)": cv_mae(hat(clone(model), olcek)), "log hedef": cv_mae(hat(log_hedef(clone(model)), olcek))})
d3 = pd.DataFrame(d3).set_index("model")

# ── Deney 4: tekrar eden satırları eğitimden çıkarmak (doğrulama katı olduğu gibi kalır) ──
tekrar_mask = X_egitim.assign(y=y_egitim).duplicated().to_numpy()
hatalar = {"tekrarlar dahil": [], "tekrarlar eğitimden çıkarıldı": []}
for e_i, v_i in CV.split(X_egitim, y_egitim, g_egitim):
    for ad, secici in [("tekrarlar dahil", e_i), ("tekrarlar eğitimden çıkarıldı", e_i[~tekrar_mask[e_i]])]:
        m = hat(GradientBoostingRegressor(random_state=RANDOM_STATE)).fit(X_egitim.iloc[secici], y_egitim.iloc[secici])
        hatalar[ad].append(mean_absolute_error(y_egitim.iloc[v_i], m.predict(X_egitim.iloc[v_i])))
d4 = pd.DataFrame({k: [np.mean(v)] for k, v in hatalar.items()}, index=["Gradient Boosting"])

display(d2.round(3)); display(d3.round(3)); display(d4.round(3))
rapor_kaydet(d1.round(4), "deney_ozellik_muhendisligi"); rapor_kaydet(d2.round(4), "deney_olcekleme")
rapor_kaydet(d3.round(4), "deney_log_hedef"); rapor_kaydet(d4.round(4), "deney_tekrar_satirlar")

fig, eksenler = plt.subplots(1, 3, figsize=(17, 4.2))
d1[["ham 8 sütun", "+ mühendislik (13 sütun)"]].plot.bar(ax=eksenler[0], color=[RENK["gri"], RENK["yesil"]], rot=15, width=0.75)
eksenler[0].set(title="Özellik mühendisliği (CV MAE, MPa)", xlabel="", ylabel="MAE")
d2.plot.bar(ax=eksenler[1], color=[RENK["gri"], RENK["mavi"]], rot=15, width=0.75)
eksenler[1].set(title="Ölçekleme (CV MAE, MPa)", xlabel="", ylabel="MAE")
d3.plot.bar(ax=eksenler[2], color=[RENK["gri"], RENK["mor"]], rot=15, width=0.75)
eksenler[2].set(title="Log hedef (CV MAE, MPa)", xlabel="", ylabel="MAE")
fig.tight_layout()
kaydet(fig, "07_onisleme_deneyleri")
plt.show()
```

### 🧱 7.1 · Grup Bilinçli Yığın (Stacking) Modeli

`scikit-learn`'ün `StackingRegressor`'ı üst modeli eğitmek için tahminleri **iç çapraz doğrulamayla** üretir; ama bu iç CV **reçete gruplarını bilmez**. Yani aynı reçetenin satırları iç CV'de de bölünür ve üst model, ezberleyen modellere **fazla güvenmeyi** öğrenir (sızıntı!). Bu yüzden iç CV'yi `GroupKFold` ile yapan küçük bir yığın sınıfı yazıyoruz:

```python
class GrupBilincliYigin(BaseEstimator, RegressorMixin):
    """Yığın (stacking): taban modellerin 'katlı dışı' tahminleri GroupKFold ile üretilir; üst model Ridge (pozitif ağırlık).

    Girdi tablosunda `recete_id` sütunu bulunur; taban modellere bu sütun verilmez.
    """

    def __init__(self, tabanlar, n_kat=5, grup_kolonu="recete_id"):
        self.tabanlar, self.n_kat, self.grup_kolonu = tabanlar, n_kat, grup_kolonu

    def fit(self, X, y):
        g = X[self.grup_kolonu].to_numpy()
        Xf = X.drop(columns=self.grup_kolonu)
        kat = GroupKFold(self.n_kat, shuffle=True, random_state=RANDOM_STATE)
        oof = np.column_stack([cross_val_predict(clone(m), Xf, y, groups=g, cv=kat) for _, m in self.tabanlar])
        self.ust_ = Ridge(alpha=1.0, positive=True).fit(oof, y)               # üst model: tahminleri harmanlayan ağırlıklar
        self.tabanlar_ = [(ad, clone(m).fit(Xf, y)) for ad, m in self.tabanlar]
        return self

    def predict(self, X):
        Xf = X.drop(columns=self.grup_kolonu)
        return self.ust_.predict(np.column_stack([m.predict(Xf) for _, m in self.tabanlar_]))
```

```python
SKOR = {"mae": "neg_mean_absolute_error", "rmse": "neg_root_mean_squared_error", "r2": "r2"}
satirlar, FOLD_MAE = [], {}

for ad, (aile, model) in YARIS.items():
    Xg = X_egitim_id if ad in ID_GEREKLI else X_egitim
    t0 = time.time()
    r = cross_validate(model, Xg, y_egitim, groups=g_egitim, cv=CV, scoring=SKOR, return_train_score=True, n_jobs=1)
    FOLD_MAE[ad] = -r["test_mae"]
    satirlar.append({"model": ad, "aile": aile, "CV MAE": -r["test_mae"].mean(), "CV MAE std": r["test_mae"].std(),
                     "CV RMSE": -r["test_rmse"].mean(), "CV R²": r["test_r2"].mean(),
                     "eğitim MAE": -r["train_mae"].mean(), "süre (sn/kat)": r["fit_time"].mean()})
    print(f"{ad:<28} CV MAE {-r['test_mae'].mean():6.3f} ± {r['test_mae'].std():.3f}   R² {r['test_r2'].mean():.3f}   ({time.time() - t0:5.1f} sn)")

yaris = pd.DataFrame(satirlar).set_index("model")
yaris["aşırı öğrenme makası"] = yaris["CV MAE"] - yaris["eğitim MAE"]
yaris = yaris.sort_values("CV MAE")
yaris["sıra"] = np.arange(1, len(yaris) + 1)
rapor_kaydet(yaris.round(4), "model_karsilastirma")
pd.DataFrame(FOLD_MAE).to_csv(RAPOR_DIR / "cv_fold_mae.csv", index_label="kat")
```

Tablodaki sıralama kesin bir hüküm mü? İki soru soruyoruz:

2. **Aradaki fark gerçek mi, şans mı?** Test noktaları reçete gruplarından geldiği için **reçete bazlı bootstrap** kullanıyoruz: satırları değil **reçeteleri** yeniden örnekliyoruz.

```python
TEST_TAHMIN, test_satirlar = {}, []
for ad, (aile, model) in YARIS.items():
    Xe, Xt = (X_egitim_id, X_test_id) if ad in ID_GEREKLI else (X_egitim, X_test)
    m = clone(model).fit(Xe, y_egitim)
    TEST_TAHMIN[ad] = m.predict(Xt)
    test_satirlar.append({"model": ad, **{f"test {k}": v for k, v in metrikler(y_test, TEST_TAHMIN[ad]).items()}})

yaris_test = pd.DataFrame(test_satirlar).set_index("model")
birlesik = yaris.join(yaris_test)
birlesik["test sırası"] = birlesik["test MAE"].rank().astype(int)
rapor_kaydet(birlesik.round(4), "model_karsilastirma_test")

rho = stats.spearmanr(birlesik["CV MAE"], birlesik["test MAE"])
print(f"CV sırası ile test sırası arasındaki Spearman korelasyonu: ρ = {rho.statistic:.3f}")
display(birlesik[["sıra", "test sırası", "CV MAE", "test MAE", "test RMSE", "test R²", "test MdAPE %"]].round(3))
```

```python
fig, eksenler = plt.subplots(1, 2, figsize=(15, 5.4))
for aile, alt in birlesik.drop(index="Taban (ortalama)").groupby("aile"):
    eksenler[0].scatter(alt["CV MAE"], alt["test MAE"], s=90, color=AILE_RENK[aile], label=aile, edgecolor="white", zorder=3)
for ad, satir in birlesik.drop(index="Taban (ortalama)").iterrows():
    if satir["sıra"] <= 5 or ad in ("Gradient Boosting", "Gauss Süreci"):
        eksenler[0].annotate(ad, (satir["CV MAE"], satir["test MAE"]), fontsize=8, xytext=(4, 4), textcoords="offset points")
tum = birlesik.drop(index="Taban (ortalama)")
lim = [min(tum["CV MAE"].min(), tum["test MAE"].min()) - 0.2, max(tum["CV MAE"].max(), tum["test MAE"].max()) + 0.2]
eksenler[0].plot(lim, lim, color="black", ls=":", lw=1)
eksenler[0].set(xlabel="CV MAE (MPa)", ylabel="Test MAE (MPa)", title=f"CV ↔ test (Spearman ρ = {rho.statistic:.2f})", xlim=lim, ylim=lim)
eksenler[0].legend(fontsize=8, ncol=2)

# ── eşleştirilmiş reçete-bootstrap: her model, CV'de en iyi modele karşı ──
def grup_bootstrap_fark(hata_a, hata_b, gruplar, n=3000, tohum=0):
    """MAE(a) − MAE(b) farkının reçete bazlı bootstrap güven aralığı (a, b aynı test satırlarında)."""
    df = pd.DataFrame({"a": np.abs(hata_a), "b": np.abs(hata_b), "g": np.asarray(gruplar)})
    topla = df.groupby("g").agg(a=("a", "sum"), b=("b", "sum"), n=("a", "size"))
    rng = np.random.default_rng(tohum)
    secim = rng.integers(0, len(topla), size=(n, len(topla)))
    A, B, N = topla["a"].to_numpy()[secim].sum(1), topla["b"].to_numpy()[secim].sum(1), topla["n"].to_numpy()[secim].sum(1)
    fark = A / N - B / N
    return fark.mean(), np.percentile(fark, 2.5), np.percentile(fark, 97.5)


en_iyi_cv = yaris.index[0]
fark_satirlari = []
for ad in yaris.index[1:12]:
    ort, alt, ust = grup_bootstrap_fark(TEST_TAHMIN[ad] - y_test, TEST_TAHMIN[en_iyi_cv] - y_test, g_test)
    fark_satirlari.append({"model": ad, "MAE farkı (model − en iyi CV)": ort, "alt %95": alt, "üst %95": ust,
                           "fark anlamlı mı?": "evet" if (alt > 0 or ust < 0) else "hayır (belirsiz)"})
fark_tablo = pd.DataFrame(fark_satirlari).set_index("model")
rapor_kaydet(fark_tablo.round(4), "eslestirilmis_fark")

y_pos = np.arange(len(fark_tablo))[::-1]
eksenler[1].errorbar(fark_tablo["MAE farkı (model − en iyi CV)"], y_pos,
                     xerr=[fark_tablo["MAE farkı (model − en iyi CV)"] - fark_tablo["alt %95"], fark_tablo["üst %95"] - fark_tablo["MAE farkı (model − en iyi CV)"]],
                     fmt="o", color=RENK["mor"], ecolor=RENK["gri"], capsize=3)
eksenler[1].axvline(0, color="black", lw=1)
eksenler[1].set(yticks=y_pos, yticklabels=fark_tablo.index, xlabel="Test MAE farkı (MPa) · >0 = en iyi CV modelinden daha kötü",
                title=f"Reçete bazlı bootstrap: '{en_iyi_cv}' ile fark")
fig.tight_layout()
kaydet(fig, "10_cv_test_ve_fark")
plt.show()
display(fark_tablo.round(3))
```

Gradient Boosting, Friedman'ın (2001) çerçevesiyle tanıtılan ve beton dayanımı yazınında da başarılı bulunan bir yöntemdir (Rathakrishnan vd., 2022). Yarışta varsayılan Gradient Boosting ise 10. sıradaydı. Peki **neden**, ve düğmelerini çevirince ne olur? Bu bölümde yalnızca **eğitim setinde, reçete bazlı 5 katlı CV** ile çalışıyoruz (test setine dokunmuyoruz). Her deneyde modeli **tek seferde 1.500 ağaca** kadar büyütüp `staged_predict` ile her turdaki doğrulama hatasını okuyoruz.

<p align="center">
  <img src="assets/diyagram_gb_dugmeleri.svg" width="100%" alt="Gradient Boosting düğmeleri">
</p>

```python
from joblib import Parallel, delayed

F_e = muhendislik(X_egitim).to_numpy()              # pipeline'ın ürettiği 13 sütunlu matris (özellik adımı satır bazlıdır, kat içinde tekrar uydurmaya gerek yok)
y_np = y_egitim.to_numpy()
KATLAR = list(CV.split(X_egitim, y_egitim, g_egitim))


def _kat(params, e_i, v_i, N):
    m = GradientBoostingRegressor(n_estimators=N, random_state=RANDOM_STATE, **params).fit(F_e[e_i], y_np[e_i])
    v = np.array([mean_absolute_error(y_np[v_i], p) for p in m.staged_predict(F_e[v_i])])       # her turdaki doğrulama MAE'si
    t = np.array([mean_absolute_error(y_np[e_i], p) for p in m.staged_predict(F_e[e_i])])       # her turdaki eğitim MAE'si
    return v, t


def gb_egri(N=1500, **params):
    """5 katın ortalaması: (doğrulama MAE eğrisi, doğrulama std, eğitim MAE eğrisi). Kat hesapları paralel."""
    sonuc = Parallel(n_jobs=4)(delayed(_kat)(params, e, v, N) for e, v in KATLAR)
    V, T = np.array([r[0] for r in sonuc]), np.array([r[1] for r in sonuc])
    return V.mean(0), V.std(0), T.mean(0)


def egri_ozeti(v, tolerans=1.03):
    """En iyi tur, en iyi MAE ve 'güvenli bölge' (en iyiye %3 yakın tur sayısı)."""
    return {"en iyi tur": int(v.argmin()) + 1, "en iyi CV MAE": v.min(), "güvenli bölge (tur)": int((v <= v.min() * tolerans).sum())}


# ── 9.1: varsayılan düğmelerle (hız 0.1, derinlik 3) 1.500 tura kadar ──────────
v0, s0, t0_ = gb_egri(N=1500, learning_rate=0.1, max_depth=3)
o0 = egri_ozeti(v0)
print(f"Varsayılan düğmeler (η=0.1, derinlik 3): en iyi tur {o0['en iyi tur']}, en iyi CV MAE {o0['en iyi CV MAE']:.3f} MPa  (varsayılan 100. turdaki: {v0[99]:.3f})")

fig, ax = plt.subplots(figsize=(9.5, 4.4))
tur = np.arange(1, 1501)
ax.plot(tur, t0_, color=RENK["mavi"], lw=2.4, label="eğitim MAE")
ax.plot(tur, v0, color=RENK["pembe"], lw=2.4, label="doğrulama MAE (CV)")
ax.fill_between(tur, v0 - s0, v0 + s0, color=RENK["pembe"], alpha=0.15)
ax.axvline(100, color="black", ls=":", lw=1); ax.text(110, ax.get_ylim()[1] * 0.93, "sklearn varsayılanı: 100 ağaç", fontsize=9)
ax.axvline(o0["en iyi tur"], color=RENK["yesil"], ls="--", lw=1.4, label=f"en iyi tur = {o0['en iyi tur']}")
ax.set(xscale="log", xlabel="Ağaç sayısı (log ölçek)", ylabel="MAE (MPa)", title="Varsayılan düğmelerle büyütürsek: 100 ağaç yeterli mi?")
ax.legend()
fig.tight_layout()
kaydet(fig, "12_gb_varsayilan_egri")
plt.show()
```

### 🪜 9.3 · Derinlik, Rastgelelik ve Kayıp İşlevi

Üç düğmeyi birlikte tarıyoruz (hepsi η = 0,05, 1.500 ağaçlık büyüme; her hücre kendi **en iyi turunda** değerlendirilir):

- 🪜 **Derinlik** (`max_depth`): 1, 2, 3, 4, 6, 8
- 🎲 **Rastgelelik:** `subsample` ∈ {1; 0,7; 0,5} × `max_features` ∈ {1,0; 0,6} (stokastik gradyan artırma; Friedman, 2002)
- ⚖️ **Kayıp:** `squared_error`, `absolute_error`, `huber`

```python
# ── derinlik ──
d_satir = []
for d in (1, 2, 3, 4, 6, 8):
    v, s, t = gb_egri(N=1500, learning_rate=0.05, max_depth=d)
    d_satir.append({"derinlik": d, **egri_ozeti(v), "eğitim MAE (en iyi turda)": t[v.argmin()]})
derinlik_tablosu = pd.DataFrame(d_satir).set_index("derinlik")

# ── alt örnekleme × özellik örnekleme ──
r_satir = []
for alt in (1.0, 0.7, 0.5):
    for mf in (1.0, 0.6):
        v, s, t = gb_egri(N=1500, learning_rate=0.05, max_depth=4, subsample=alt, max_features=mf)
        r_satir.append({"subsample": alt, "max_features": mf, **egri_ozeti(v)})
rastgele_tablosu = pd.DataFrame(r_satir)

# ── kayıp işlevi ──
k_satir = []
for kayip in ("squared_error", "absolute_error", "huber"):
    v, s, t = gb_egri(N=1500, learning_rate=0.05, max_depth=4, loss=kayip)
    k_satir.append({"kayıp": kayip, **egri_ozeti(v)})
kayip_tablosu = pd.DataFrame(k_satir).set_index("kayıp")

display(derinlik_tablosu.round(3)); display(rastgele_tablosu.round(3)); display(kayip_tablosu.round(3))
rapor_kaydet(derinlik_tablosu.round(4), "gb_derinlik"); rapor_kaydet(rastgele_tablosu.round(4), "gb_rastgelelik", index=False); rapor_kaydet(kayip_tablosu.round(4), "gb_kayip")

fig, eksenler = plt.subplots(1, 3, figsize=(17, 4.4))
eksenler[0].bar(derinlik_tablosu.index.astype(str), derinlik_tablosu["en iyi CV MAE"], color=RENK["turuncu"], width=0.7)
for i, (d, satir) in enumerate(derinlik_tablosu.iterrows()):
    eksenler[0].text(i, satir["en iyi CV MAE"] + 0.03, f"{satir['en iyi CV MAE']:.2f}\n({satir['en iyi tur']} ağaç)", ha="center", fontsize=8)
eksenler[0].set(title="Derinlik (en iyi turda CV MAE)", xlabel="max_depth", ylabel="MAE (MPa)", ylim=(3.0, derinlik_tablosu["en iyi CV MAE"].max() + 0.5))
piv = rastgele_tablosu.pivot(index="subsample", columns="max_features", values="en iyi CV MAE")
sns.heatmap(piv, annot=True, fmt=".3f", cmap="YlGn_r", ax=eksenler[1], cbar=False)
eksenler[1].set(title="Rastgelelik: subsample × max_features", xlabel="max_features", ylabel="subsample")
eksenler[2].bar(kayip_tablosu.index, kayip_tablosu["en iyi CV MAE"], color=[RENK["gri"], RENK["mavi"], RENK["mor"]], width=0.6)
for i, (k, satir) in enumerate(kayip_tablosu.iterrows()):
    eksenler[2].text(i, satir["en iyi CV MAE"] + 0.03, f"{satir['en iyi CV MAE']:.2f}", ha="center", fontsize=9)
eksenler[2].set(title="Kayıp işlevi (en iyi turda CV MAE)", ylabel="MAE (MPa)", ylim=(3.0, kayip_tablosu["en iyi CV MAE"].max() + 0.5))
fig.tight_layout()
kaydet(fig, "14_gb_derinlik_rastgelelik_kayip")
plt.show()
```

- ⚠️ **Seçim yanlılığı:** 30 adaydan en iyisini seçince, o adayın CV skoru **iyimser** olur (şansın da payı vardır). Bu yüzden her ayarlı modeli **taze katlarda** (üç farklı tohumla yeniden karıştırılmış reçete bazlı CV) tekrar ölçüyoruz ve **final seçimini taze skora göre** yapıyoruz.

```python
def ayar_prefix(log=False):
    return "model__regressor__" if log else "model__"


def gb_uzay(log=False):
    p = ayar_prefix(log)
    return {p + "n_estimators": randint(200, 1000), p + "learning_rate": loguniform(0.01, 0.3), p + "max_depth": randint(2, 8),
            p + "subsample": uniform(0.5, 0.5), p + "min_samples_leaf": randint(1, 12), p + "max_features": uniform(0.4, 0.6),
            p + "loss": ["squared_error", "huber"]}


AYAR = {   # ad: (tahmin edici, arama uzayı, aday sayısı)
    "Gradient Boosting (ayarlı)":           (hat(GradientBoostingRegressor(random_state=RANDOM_STATE)), gb_uzay(False), 30),
    "Gradient Boosting · log hedef (ayarlı)": (hat(log_hedef(GradientBoostingRegressor(random_state=RANDOM_STATE))), gb_uzay(True), 30),
    "Random Forest (ayarlı)":               (hat(RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=1)),
                                             {"model__n_estimators": randint(200, 600), "model__max_depth": [None, 10, 15, 20], "model__min_samples_leaf": randint(1, 6),
                                              "model__max_features": uniform(0.3, 0.7)}, 30),
    "Extra Trees (ayarlı)":                 (hat(ExtraTreesRegressor(random_state=RANDOM_STATE, n_jobs=1)),
                                             {"model__n_estimators": randint(200, 800), "model__max_depth": [None, 12, 20], "model__min_samples_leaf": randint(1, 5),
                                              "model__max_features": uniform(0.4, 0.6)}, 30),
    "HistGradientBoosting (ayarlı)":        (hat(HistGradientBoostingRegressor(random_state=RANDOM_STATE)),
                                             {"model__learning_rate": loguniform(0.02, 0.3), "model__max_iter": randint(200, 1000), "model__max_leaf_nodes": randint(8, 64),
                                              "model__min_samples_leaf": randint(3, 30), "model__l2_regularization": loguniform(1e-3, 10), "model__max_features": uniform(0.4, 0.6)}, 30),
    "XGBoost (ayarlı)":                     (hat(XGBRegressor(random_state=RANDOM_STATE, n_jobs=1)),
                                             {"model__n_estimators": randint(200, 1000), "model__learning_rate": loguniform(0.01, 0.3), "model__max_depth": randint(2, 9),
                                              "model__subsample": uniform(0.5, 0.5), "model__colsample_bytree": uniform(0.4, 0.6), "model__min_child_weight": randint(1, 8),
                                              "model__reg_lambda": loguniform(0.1, 20)}, 30),
    "LightGBM (ayarlı)":                    (hat(LGBMRegressor(random_state=RANDOM_STATE, n_jobs=1, verbose=-1, subsample_freq=1)),
                                             {"model__n_estimators": randint(200, 1000), "model__learning_rate": loguniform(0.01, 0.3), "model__num_leaves": randint(6, 64),
                                              "model__min_child_samples": randint(3, 30), "model__subsample": uniform(0.5, 0.5), "model__colsample_bytree": uniform(0.4, 0.6),
                                              "model__reg_lambda": loguniform(0.01, 20)}, 30),
    "CatBoost (ayarlı)":                    (hat(CatBoostRegressor(random_seed=RANDOM_STATE, thread_count=1, verbose=0)),
                                             {"model__iterations": randint(300, 1000), "model__learning_rate": loguniform(0.02, 0.3), "model__depth": randint(3, 9),
                                              "model__l2_leaf_reg": loguniform(1, 20)}, 20),
    "AdaBoost (ayarlı)":                    (hat(AdaBoostRegressor(estimator=DecisionTreeRegressor(), random_state=RANDOM_STATE)),
                                             {"model__n_estimators": randint(30, 300), "model__learning_rate": loguniform(0.03, 1.0), "model__loss": ["linear", "square", "exponential"],
                                              "model__estimator__max_depth": randint(3, 14)}, 30),
    "SVR (ayarlı)":                         (hat(SVR(), True),
                                             {"model__C": loguniform(1, 300), "model__epsilon": loguniform(0.05, 3), "model__gamma": loguniform(1e-3, 0.3)}, 30),
}
print(f"{len(AYAR)} model ayarlanacak · toplam aday: {sum(n for _, _, n in AYAR.values())} × 5 kat")
```

```python
ARAMALAR, ayar_satirlari = {}, []
for ad, (model, uzay, n_iter) in AYAR.items():
    t0 = time.time()
    arama = RandomizedSearchCV(model, uzay, n_iter=n_iter, scoring="neg_mean_absolute_error", cv=CV, random_state=RANDOM_STATE, n_jobs=4, refit=True)
    arama.fit(X_egitim, y_egitim, groups=g_egitim)
    ARAMALAR[ad] = arama
    varsayilan = {"Gradient Boosting (ayarlı)": "Gradient Boosting", "Gradient Boosting · log hedef (ayarlı)": "Gradient Boosting"}.get(ad, {"SVR (ayarlı)": "SVR (RBF)"}.get(ad, ad.replace(" (ayarlı)", "")))
    ayar_satirlari.append({"model": ad, "varsayılan CV MAE": yaris.loc[varsayilan, "CV MAE"], "ayarlı CV MAE (seçim katları)": -arama.best_score_,
                           "süre (sn)": time.time() - t0})
    print(f"{ad:<42} varsayılan {yaris.loc[varsayilan, 'CV MAE']:.3f} → ayarlı {-arama.best_score_:.3f}   ({time.time() - t0:5.0f} sn)")
```

### 🔍 10.1 · Taze Katlarda Doğrulama ve Final Adayları

Her ayarlı modelin **en iyi adayını**, aramada **kullanılmamış üç yeni katlama düzeniyle** yeniden ölçüyoruz. Bu skor seçim yanlılığından arınmıştır.

```python
TAZE_TOHUMLAR = (7, 11, 13)
taze_katlar = [GroupKFold(5, shuffle=True, random_state=s) for s in TAZE_TOHUMLAR]

for satir in ayar_satirlari:
    en_iyi = ARAMALAR[satir["model"]].best_estimator_
    skorlar = [-cross_validate(clone(en_iyi), X_egitim, y_egitim, groups=g_egitim, cv=k, scoring="neg_mean_absolute_error", n_jobs=1)["test_score"].mean() for k in taze_katlar]
    satir["taze katlarda CV MAE"] = np.mean(skorlar)
    satir["iyileşme % (taze)"] = (satir["varsayılan CV MAE"] - satir["taze katlarda CV MAE"]) / satir["varsayılan CV MAE"] * 100
    satir["seçim yanlılığı (taze − seçim)"] = satir["taze katlarda CV MAE"] - satir["ayarlı CV MAE (seçim katları)"]
    satir["en iyi parametreler"] = json.dumps({k.replace("model__", "").replace("regressor__", ""): (round(float(v), 4) if isinstance(v, (float, np.floating)) else (int(v) if isinstance(v, (int, np.integer)) else v))
                                               for k, v in ARAMALAR[satir["model"]].best_params_.items()}, ensure_ascii=False)

ayar_tablosu = pd.DataFrame(ayar_satirlari).set_index("model").sort_values("taze katlarda CV MAE")
rapor_kaydet(ayar_tablosu.round(4), "ayar_sonuclari")
display(ayar_tablosu.drop(columns="en iyi parametreler").round(3))
```

"Model neye bakıyor?" sorusunu **üç farklı gözlükle** yanıtlıyoruz (hepsi boosting ailesinin en iyisi `FINAL_GB` üzerinde):

1. 🌲 **MDI** (*mean decrease in impurity*): ağaçların bölmelerde sağladığı toplam iyileşme; hızlıdır ama **eğitim verisine bağlıdır** ve çok değerli (sürekli) özelliklere kayırmalıdır.
2. 🎲 **Permütasyon önemi:** bir sütunu **karıştırınca** test hatası ne kadar artıyor? Modelden bağımsız, test verisinde ölçülür; ama **ilişkili** özellikleri olduğundan küçük gösterebilir.
3. 🧯 **Sütunu çıkarıp yeniden eğitme (ablasyon):** bilgiyi gerçekten kaldırır; en dürüst ama en pahalı yöntem.

```python
GB_FIT = clone(GB_FINAL_MODEL).fit(X_egitim, y_egitim)

def agac_modeli(pipe):
    """Pipeline'daki ağaç tabanlı modeli (varsa log hedef sarmalayıcısının içinden) döndürür."""
    m = pipe.named_steps["model"]
    return m.regressor_ if hasattr(m, "regressor_") else m

ozellik_adlari = list(muhendislik(X_egitim.head(2)).columns)
agac = agac_modeli(GB_FIT)
if hasattr(agac, "feature_importances_"):
    mdi = pd.Series(agac.feature_importances_, index=ozellik_adlari).sort_values()
else:                                                              # HistGradientBoosting MDI sunmaz: eğitim verisinde permütasyonla yedekle
    _p = permutation_importance(agac, muhendislik(X_egitim), y_egitim, scoring="neg_mean_absolute_error", n_repeats=10, random_state=RANDOM_STATE)
    mdi = pd.Series(np.clip(_p.importances_mean, 0, None), index=ozellik_adlari).sort_values()
mdi = mdi / mdi.sum()

perm = permutation_importance(GB_FIT, X_test, y_test, scoring="neg_mean_absolute_error", n_repeats=30, random_state=RANDOM_STATE, n_jobs=1)
perm_tablo = pd.DataFrame({"ort. MAE artışı (MPa)": perm.importances_mean, "std": perm.importances_std}, index=X_test.columns).sort_values("ort. MAE artışı (MPa)")

fig, eksenler = plt.subplots(1, 2, figsize=(15, 5.2))
mdi.plot.barh(ax=eksenler[0], color=RENK["turuncu"])
eksenler[0].set(title=f"MDI önemi · {GB_FINAL_AD}\n(13 özellik, türetilmişler dahil)", xlabel="Önem payı")
eksenler[1].barh(perm_tablo.index, perm_tablo["ort. MAE artışı (MPa)"], xerr=perm_tablo["std"], color=RENK["mor"], ecolor=RENK["gri"], capsize=3)
eksenler[1].set(title="Permütasyon önemi (test seti, 30 tekrar)\n(8 ham özellik: karıştırınca MAE ne kadar artıyor?)", xlabel="MAE artışı (MPa)")
fig.tight_layout()
kaydet(fig, "18_ozellik_onemi_mdi_permutasyon")
plt.show()
rapor_kaydet(mdi.rename("mdi_payi").to_frame(), "ozellik_onemi_mdi"); rapor_kaydet(perm_tablo.round(4), "ozellik_onemi_permutasyon")
display(perm_tablo.sort_values("ort. MAE artışı (MPa)", ascending=False).round(3))
```

### 🧯 11.1 · Ablasyon: Bilgiyi Gerçekten Çıkarınca Ne Oluyor?

Her ham özelliği (türetilmiş akrabalarıyla birlikte) **sabit bir değere** (eğitim ortalaması) çekip modeli **yeniden eğitiyoruz** (taze katlarla CV). Hatadaki artış, o sütunun **tek başına taşıdığı** bilgiyi gösterir; yedek akrabası varsa artış küçük kalır.

```python
def ablasyon_cv(kolon):
    X_a = X_egitim.copy()
    X_a[kolon] = X_egitim[kolon].mean()                                  # sütunu ortalamaya sabitle: bilgi yok
    return np.mean([-cross_validate(clone(GB_FINAL_MODEL), X_a, y_egitim, groups=g_egitim, cv=k, scoring="neg_mean_absolute_error", n_jobs=1)["test_score"].mean()
                    for k in taze_katlar[:2]])

temel = np.mean([-cross_validate(clone(GB_FINAL_MODEL), X_egitim, y_egitim, groups=g_egitim, cv=k, scoring="neg_mean_absolute_error", n_jobs=1)["test_score"].mean() for k in taze_katlar[:2]])
abl = pd.DataFrame({"kolon": list(X_egitim.columns), "CV MAE (sütunsuz)": [ablasyon_cv(k) for k in X_egitim.columns]}).set_index("kolon")
abl["MAE artışı (MPa)"] = abl["CV MAE (sütunsuz)"] - temel
abl = abl.sort_values("MAE artışı (MPa)", ascending=False)
print(f"Tüm özelliklerle CV MAE: {temel:.3f} MPa")
display(abl.round(3))
rapor_kaydet(abl.round(4), "ablasyon")

fig, ax = plt.subplots(figsize=(8.4, 4.2))
abl["MAE artışı (MPa)"].sort_values().plot.barh(ax=ax, color=RENK["pembe"])
ax.set(title="Sütunu çıkarınca CV MAE ne kadar artıyor?", xlabel="MAE artışı (MPa)")
fig.tight_layout()
kaydet(fig, "19_ablasyon")
plt.show()
```

### 🧭 11.3 · Fizik Kuralını Modele Öğretmek: Monotonluk Kısıtı

```python
hist_p = {k.replace("model__", ""): v for k, v in ARAMALAR["HistGradientBoosting (ayarlı)"].best_params_.items()}
KISIT = {"cimento": 1, "yas_gun": 1, "log_yas": 1, "su": -1, "su_cimento": -1, "su_baglayici": -1, "baglayici": 1}      # +1: artan, −1: azalan

def hist_model(kisitli):
    return hat(HistGradientBoostingRegressor(random_state=RANDOM_STATE, monotonic_cst=(KISIT if kisitli else None), **hist_p))

mono_satir = []
for ad, kisitli in [("kısıtsız", False), ("monotonluk kısıtlı", True)]:
    skor = [-cross_validate(hist_model(kisitli), X_egitim, y_egitim, groups=g_egitim, cv=k, scoring="neg_mean_absolute_error", n_jobs=1)["test_score"].mean() for k in taze_katlar]
    mono_satir.append({"model": f"HistGB {ad}", "taze CV MAE": np.mean(skor)})
mono = pd.DataFrame(mono_satir).set_index("model")
display(mono.round(3))
rapor_kaydet(mono.round(4), "monotonluk_deneyi")

kisitli_fit, kisitsiz_fit = hist_model(True).fit(X_egitim, y_egitim), hist_model(False).fit(X_egitim, y_egitim)
fig, eksenler = plt.subplots(1, 3, figsize=(16, 4.4))
for ax, (kolon, baslik) in zip(eksenler, [("cimento", "Çimento (kg/m³)"), ("su", "Su (kg/m³)"), ("yas_gun", "Yaş (gün)")]):
    for model, renk, etiket in [(kisitsiz_fit, RENK["gri"], "kısıtsız"), (kisitli_fit, RENK["yesil"], "monotonluk kısıtlı")]:
        pdv = partial_dependence(model, X_pdp, [kolon], grid_resolution=40, kind="average")
        ax.plot(pdv["grid_values"][0], pdv["average"][0], color=renk, lw=2.8, label=etiket)
    ax.set(xlabel=baslik, ylabel="Tahmin (MPa)"); ax.legend()
    if kolon == "yas_gun":
        ax.set_xscale("log")
fig.suptitle("Kısmi bağımlılık: kısıt, eğriyi fiziksel olarak tutarlı kılıyor", fontweight="bold", y=1.03)
fig.tight_layout()
kaydet(fig, "22_monotonluk_pdp")
plt.show()
```

| Yöntem | Fikir | Gradient Boosting ile bağı |
|:--|:--|:--|
| 🎯 **Kuantil Gradient Boosting** | `loss="quantile"` ile **alt (%5) ve üst (%95) kuantili** doğrudan tahmin et (Koenker & Bassett, 1978) | Kayıp işlevini değiştirmek: bölüm 6'daki "eğimi öğren" fikri |
| 🧪 **Bölünmüş konformal (global)** | Eğitimde hiç görülmeyen *kalibrasyon* reçetelerindeki **mutlak hatanın %90 kuantili** kadar pay ekle (Lei vd., 2018) | Her modelle çalışır; kapsama **garantisi** (sonlu örneklem) verir |
| 🧩 **Mondrian konformal (yaş dilimine göre)** | Aynısı, ama **her yaş diliminde ayrı** kalibre | Heterojen hata varsa kapsamayı her dilimde dengeler |

(Konformal yöntemler için bkz. Angelopoulos & Bates, 2023. Garanti, kalibrasyon ve test örneklerinin **birbirinin yerine geçebilir** olmasına dayanır; reçete grupları bölündüğü için bunu elimizden geldiğince koruyoruz.)

### 🎯 13.1 · Kuantil Gradient Boosting

```python
gb_p = {k.replace("model__", ""): v for k, v in ARAMALAR["Gradient Boosting (ayarlı)"].best_params_.items() if k != "model__loss"}   # ayarlı GB düğmeleri (kayıp hariç)
KUANTILLER = [0.05, 0.1, 0.25, 0.5, 0.75, 0.9, 0.95]
kuantil_tahmin = {}
for q in KUANTILLER:
    mq = hat(GradientBoostingRegressor(loss="quantile", alpha=q, random_state=RANDOM_STATE, **gb_p)).fit(X_egitim, y_egitim)
    kuantil_tahmin[q] = mq.predict(X_test)
    print(f"q = {q:.2f}  → test satırlarının %{(y_test.to_numpy() <= kuantil_tahmin[q]).mean() * 100:5.1f}'i bu tahminin altında kaldı (beklenen %{q * 100:.0f})")
alt_q, ust_q = np.minimum(kuantil_tahmin[0.05], kuantil_tahmin[0.95]), np.maximum(kuantil_tahmin[0.05], kuantil_tahmin[0.95])
kuantil_kapsama = ((y_test.to_numpy() >= alt_q) & (y_test.to_numpy() <= ust_q)).mean()
print(f"\n%90 kuantil aralığının test kapsaması: %{kuantil_kapsama * 100:.1f} · ortalama genişlik {np.mean(ust_q - alt_q):.1f} MPa")
```

### 🧪 13.2 · Bölünmüş Konformal (Global ve Yaş Dilimine Göre)

Eğitim setinin **reçetelerini** %75 / %25 böleriz: model %75'te eğitilir, kalan %25 (**kalibrasyon**, 100'e yakın reçete) modelin gerçek hata dağılımını ölçer. Aralık = $\hat y \pm \hat q$; burada $\hat q$, kalibrasyon mutlak hatalarının $\lceil (n+1)(1-\alpha)\rceil/n$ kuantilidir.

```python
ALFA = 0.10                                                          # %90 kapsama hedefi
prop_idx, kal_idx = next(GroupShuffleSplit(1, test_size=0.25, random_state=RANDOM_STATE).split(X_egitim, y_egitim, g_egitim))
konformal_model = clone(FINAL_MODEL).fit(X_egitim.iloc[prop_idx], y_egitim.iloc[prop_idx])
kal_tahmin = konformal_model.predict(X_egitim.iloc[kal_idx])
kal_mutlak = np.abs(y_egitim.iloc[kal_idx].to_numpy() - kal_tahmin)
test_nokta = konformal_model.predict(X_test)                         # konformal aralığın merkezi: aynı (kalibre edilen) model
y_t = y_test.to_numpy()


def konformal_q(skorlar, alfa=ALFA):
    n = len(skorlar)
    seviye = min(1.0, np.ceil((n + 1) * (1 - alfa)) / n)
    return np.quantile(skorlar, seviye, method="higher")


YAS_SINIRLARI, YAS_ETIKET = [0, 7, 28, 400], ["≤ 7 gün", "8–28 gün", "> 28 gün"]
kal_yas = pd.cut(X_egitim.iloc[kal_idx]["yas_gun"], YAS_SINIRLARI, labels=YAS_ETIKET).to_numpy()
test_yas = pd.cut(X_test["yas_gun"], YAS_SINIRLARI, labels=YAS_ETIKET).to_numpy()

q_global = konformal_q(kal_mutlak)
q_yas = {e: konformal_q(kal_mutlak[kal_yas == e]) for e in YAS_ETIKET}
q_test_yas = np.array([q_yas[e] for e in test_yas])
print(f"Kalibrasyon: {len(kal_idx)} satır / {g_egitim.iloc[kal_idx].nunique()} reçete")
print(f"Global q̂ = ±{q_global:.2f} MPa · yaş dilimi q̂: " + " · ".join(f"{e}: ±{q:.2f}" for e, q in q_yas.items()))

aralıklar = {
    "Kuantil GB (%5–%95)":          (alt_q, ust_q),
    "Konformal · global":           (test_nokta - q_global, test_nokta + q_global),
    "Konformal · Mondrian (yaş)":   (test_nokta - q_test_yas, test_nokta + q_test_yas),
}
kapsama_satir = []
for ad, (a, u) in aralıklar.items():
    icinde = (y_t >= a) & (y_t <= u)
    satir = {"yöntem": ad, "genel kapsama %": icinde.mean() * 100, "ort. genişlik (MPa)": np.mean(u - a)}
    for e in YAS_ETIKET:
        satir[f"kapsama % · {e}"] = icinde[test_yas == e].mean() * 100
    kapsama_satir.append(satir)
kapsama = pd.DataFrame(kapsama_satir).set_index("yöntem")
display(kapsama.round(1))
rapor_kaydet(kapsama.round(3), "tahmin_araligi_kapsama")
```

### 🛡️ 13.3 · Güvenlik Payına Dayalı Karar Kuralı

Bir reçetenin **"en az T MPa vereceğini"** iddia etmek istiyoruz. **Dayanımı olduğundan yüksek tahmin etmek**, düşük tahmin etmekten çok daha **tehlikeli**dir (yapı güvenliği). İki kuralı karşılaştırıyoruz:

- 🎯 **Nokta kuralı:** tahmin $\hat y \ge T$ ise "yeterli" de.
- 🛡️ **Güvenli kural:** **alt sınır** $\hat y - \hat q \ge T$ ise "yeterli" de (Mondrian konformal alt sınır).

Ölçülen: kabul edilen reçeteler arasında **gerçekte T'nin altında kalanların oranı** (yanlış kabul) ve yeterli reçetelerin **kaçını kaçırdığımız**.

```python
alt_sinir = test_nokta - q_test_yas
guvenlik = []
for T in (20, 30, 40):
    gercek_yeterli = y_t >= T
    for kural, kabul in [("nokta: tahmin ≥ T", test_nokta >= T), ("güvenli: alt sınır ≥ T", alt_sinir >= T)]:
        guvenlik.append({"T (MPa)": T, "kural": kural, "kabul edilen": int(kabul.sum()),
                         "yanlış kabul (gerçek < T)": int((kabul & ~gercek_yeterli).sum()),
                         "yanlış kabul oranı %": (kabul & ~gercek_yeterli).sum() / max(kabul.sum(), 1) * 100,
                         "kaçırılan yeterli reçete %": ((~kabul) & gercek_yeterli).sum() / max(gercek_yeterli.sum(), 1) * 100})
guvenlik = pd.DataFrame(guvenlik).set_index(["T (MPa)", "kural"])
display(guvenlik.round(1))
rapor_kaydet(guvenlik.round(3), "guvenlik_kurali")

# ── Tek bir eşik için pay taraması: ne kadar pay, ne kadar kaçırma? ──
T0 = 30
paylar = np.linspace(0, 12, 49)
yk, kc = [], []
gy = y_t >= T0
for pay in paylar:
    kabul = (test_nokta - pay) >= T0
    yk.append((kabul & ~gy).sum() / max(kabul.sum(), 1) * 100); kc.append(((~kabul) & gy).sum() / max(gy.sum(), 1) * 100)
fig, ax = plt.subplots(figsize=(9.2, 4.4))
ax.plot(paylar, yk, color=RENK["pembe"], lw=2.6, label="yanlış kabul oranı % (tehlikeli)")
ax.plot(paylar, kc, color=RENK["mavi"], lw=2.6, label="kaçırılan yeterli reçete % (maliyet)")
ax.axvline(q_global, color="black", ls="--", lw=1, label=f"global konformal q̂ = {q_global:.1f} MPa")
ax.set(xlabel="Güvenlik payı (MPa) — kabul kuralı: tahmin − pay ≥ 30", ylabel="%", title="Pay büyüdükçe yanlış kabul azalır, ama yeterli reçeteleri de kaçırırız")
ax.legend()
fig.tight_layout()
kaydet(fig, "27_guvenlik_payi_egrisi")
plt.show()
```

> ⚠️ **Bu bölüm bir keşif egzersizidir, bir tasarım önerisi değildir.** Model **fizik yasası değil, veri özetidir**: optimizasyon, modelin **en iyimser hata yaptığı** yerleri bulma eğilimindedir ("kazananın laneti"). Bu yüzden (1) aramayı **verinin desteklediği bölgeyle** sınırlıyoruz, (2) **güvenlik payı** uyguluyoruz ve (3) çıkan adayları **mutlaka kırma deneyiyle** doğrulanacak "aday" sayıyoruz.

**Arama kuralları:**

| Kural | Değer |
|:--|:--|
| Hedef | 28 günde **alt sınır ≥ 40 MPa** (tahmin − Mondrian konformal pay) |
| Amaç | **Çimento** kg/m³'ü en aza indir |
| Kapsam | Her bileşen eğitim verisinin **%1–%99 aralığında**; toplam kütle ve su/bağlayıcı oranı da gözlenen aralıkta |
| Veri desteği | En yakın 5 eğitim reçetesine **ortalama uzaklık**, eğitim reçetelerinin kendi komşuluk uzaklıklarının **%90'ını** aşmamalı |

```python
from sklearn.neighbors import NearestNeighbors

HEDEF_T, HEDEF_YAS = 40.0, 28
egitim_recete = X_egitim.assign(rid=g_egitim).drop_duplicates("rid")[RECETE_KOLONLARI]            # 341 farklı eğitim reçetesi
olcek_r = StandardScaler().fit(egitim_recete)
komsu = NearestNeighbors(n_neighbors=6).fit(olcek_r.transform(egitim_recete))
d_kendi = komsu.kneighbors(olcek_r.transform(egitim_recete))[0][:, 1:6].mean(axis=1)                # kendisi hariç 5 komşuya ort. uzaklık
DESTEK_ESIGI = np.quantile(d_kendi, 0.90)

alt, ust = egitim_recete.quantile(0.01), egitim_recete.quantile(0.99)
rng = np.random.default_rng(RANDOM_STATE)
N_ADAY = 200_000
aday = pd.DataFrame({k: rng.uniform(alt[k], ust[k], N_ADAY) for k in RECETE_KOLONLARI})
aday["yas_gun"] = HEDEF_YAS

tum = aday[RECETE_KOLONLARI].sum(axis=1)
egitim_toplam = egitim_recete.sum(axis=1)
su_bag = aday["su"] / (aday["cimento"] + aday["curuf"] + aday["kul"])
egitim_su_bag = egitim_recete["su"] / (egitim_recete[["cimento", "curuf", "kul"]].sum(axis=1))
fiziksel = tum.between(egitim_toplam.quantile(0.01), egitim_toplam.quantile(0.99)) & su_bag.between(egitim_su_bag.quantile(0.01), egitim_su_bag.quantile(0.99))
aday = aday[fiziksel].reset_index(drop=True)

uzaklik = komsu.kneighbors(olcek_r.transform(aday[RECETE_KOLONLARI]), n_neighbors=5)[0].mean(axis=1)
aday["veri_uzakligi"] = uzaklik
destekli = aday[aday["veri_uzakligi"] <= DESTEK_ESIGI].copy().reset_index(drop=True)
print(f"Üretilen aday: {N_ADAY:,} → fiziksel kurallara uyan: {len(aday):,} → veri tarafından desteklenen: {len(destekli):,}")

PAY_28 = q_yas["8–28 gün"]                                               # 28. gün için Mondrian konformal pay (bölüm 13)
destekli["tahmin"] = FINAL_FIT.predict(destekli[HAM_OZELLIKLER])
destekli["alt_sinir"] = destekli["tahmin"] - PAY_28
uygun = destekli[destekli["alt_sinir"] >= HEDEF_T].sort_values("cimento")
print(f"Alt sınırı ≥ {HEDEF_T:.0f} MPa olan destekli aday: {len(uygun):,} (güvenlik payı ±{PAY_28:.2f} MPa)")
```

### 🧰 15.1 · Kullanıma Hazır Tahmin Fonksiyonu

```python
PAY_TABLOSU = meta["konformal_pay_MPa"]


def beton_tahmin_et(recete, yas_gun):
    """recete: {'cimento':…, 'curuf':…, 'kul':…, 'su':…, 'akiskanlastirici':…, 'iri_agrega':…, 'ince_agrega':…} (kg/m³)"""
    satir = pd.DataFrame([{**recete, "yas_gun": yas_gun}])[HAM_OZELLIKLER]
    nokta = float(yuklenen.predict(satir)[0])
    dilim = "≤ 7 gün" if yas_gun <= 7 else ("8–28 gün" if yas_gun <= 28 else "> 28 gün")
    pay = PAY_TABLOSU[dilim]
    return {"yaş (gün)": yas_gun, "tahmin (MPa)": round(nokta, 1), "alt sınır (MPa)": round(nokta - pay, 1), "üst sınır (MPa)": round(nokta + pay, 1)}


ornekler = {
    "Klasik çimentolu (cüruf/kül yok)": dict(cimento=350, curuf=0, kul=0, su=185, akiskanlastirici=0, iri_agrega=1000, ince_agrega=800),
    "Cüruflu + akışkanlaştırıcılı":     dict(cimento=300, curuf=120, kul=0, su=160, akiskanlastirici=9, iri_agrega=980, ince_agrega=780),
    "Uçucu küllü, düşük çimento":       dict(cimento=200, curuf=0, kul=120, su=175, akiskanlastirici=6, iri_agrega=990, ince_agrega=800),
}
tablo = pd.DataFrame([{"reçete": ad, **beton_tahmin_et(r, yas)} for ad, r in ornekler.items() for yas in (7, 28, 90)])
display(tablo)
rapor_kaydet(tablo, "ornek_recete_tahminleri", index=False)
```

### 📝 Mini Quiz

<details><summary><b>1.</b> Bu veride rastgele bölme neden iyimser sonuç veriyor? Hangi modeller daha çok etkileniyor?</summary>

Aynı reçetenin farklı yaşlardaki satırları hem eğitime hem teste düşüyor; model reçeteyi **tanıyıp** ezberliyor. **Ağaç toplulukları** (yüksek kapasite) çok etkileniyor, **doğrusal model** neredeyse hiç.
</details>

<details><summary><b>2.</b> Su/çimento oranı neden ham `su` ve `cimento` sütunlarından daha bilgilidir?</summary>

Dayanım suyun **çimentoya oranına** bağlıdır (Abrams yasası). Doğrusal bir model iki sütunun **çarpımsal** ilişkisini kendi bulamaz; oranı vermek bu ilişkiyi doğrudan sunar.
</details>

<details><summary><b>3.</b> Varsayılan Gradient Boosting neden varsayılan Extra Trees'in gerisinde kaldı? Ayar neyi değiştirdi?</summary>

Varsayılan 100 ağaç × 0,1 hız bu veri için **erken duruyor** (eğitim hatası hâlâ yüksek). Ayarla ağaç sayısı, hız, derinlik ve rastgelelik **birlikte** uyarlanıyor; sonuçlar bölüm 10'daki tablodadır.
</details>

<details><summary><b>4.</b> Taze katlarda yeniden ölçmek neden gerekli?</summary>

30 adaydan **en iyisini seçince**, o adayın CV skoru şansın da payıyla **iyimser** olur. Aramada kullanılmamış yeni katlarla ölçmek **seçim yanlılığını** ortaya çıkarır.
</details>

<details><summary><b>5.</b> `GradientBoostingRegressor(n_iter_no_change=…)` gruplu veride neden tuzaktır?</summary>

İçerideki doğrulama ayırması **rastgeledir**; aynı reçetenin satırları iki tarafa düşer, doğrulama hatası olduğundan iyi görünür ve model **fazla ağaç** ekler. Grup bilinçli CV ile tur sayısını seçmek daha dürüsttür.
</details>

<details><summary><b>6.</b> Konformal aralığın kapsama garantisi hangi varsayıma dayanır?</summary>

Kalibrasyon ve yeni örneklerin **birbirinin yerine geçebilir** (exchangeable) olmasına. Reçete grupları kalibrasyona ve teste **bölünmeden** dağıtıldığı için bu varsayım korunur; dağılım değişirse (yeni malzeme) garanti zayıflar.
</details>

<details><summary><b>7.</b> Neden "tahmin ≥ T" yerine "alt sınır ≥ T" kuralı kullanıyoruz?</summary>

Dayanımı **olduğundan yüksek tahmin etmek** güvenlik riskidir. Alt sınır kuralı, yanlış kabul oranını düşürür; bedel, yeterli bazı reçetelerin **kaçırılmasıdır** (maliyet).
</details>

<details><summary><b>8.</b> Reçete optimizasyonu neden "kazananın laneti"ne açıktır?</summary>

Model hatalı olduğu yerlerde **fazla iyimser** tahminler üretir; arama motoru tam da bu "iyimser hataları" seçer. Verinin desteklemediği bölgeye gidildikçe risk artar; bu yüzden destek kısıtı, güvenlik payı ve laboratuvar doğrulaması gerekir.
</details>
