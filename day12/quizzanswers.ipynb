```python
def myLoop(*args):
    for element in args:
        print(element / 2)
```

myLoop(3,2,1,5,3,4)

```python
# ==================== CEVAP - SORU 4 ====================
"""
📌 **DETAYLI AÇIKLAMA:**

⭐ ***args** → Bu özel bir Python parametresidir!

🔍 **ARGS KAVRAMI:**
   *args → "arguments" kelimesinin kısaltması
   ➡️ **Sınırsız sayıda argüman** almamızı sağlar
   ➡️ Argümanlar bir **TUPLE** olarak fonksiyona gelir
   ➡️ For döngüsü ile iterate edilebilir

📊 **FONKSİYON ÇAĞRISI ANALİZİ:**
   myLoop(3,2,1,5,3,4)
   
🔄 **DÖNGÜ ÇIKTI ADıM ADIM:**
   1. element = 3 → 3/2 = 1.5
   2. element = 2 → 2/2 = 1.0
   3. element = 1 → 1/2 = 0.5
   4. element = 5 → 5/2 = 2.5
   5. element = 3 → 3/2 = 1.5
   6. element = 4 → 4/2 = 2.0

✅ **SONUÇ:** Her sayı sırayla 2'ye bölünüp yazdırılır
"""
# Test edelim:
def myLoop(*args):
    for element in args:
        print(element / 2)

print("🎯 ÇIKTI:")
myLoop(3,2,1,5,3,4)

print("\n" + "="*50)
print("📝 ÖZET: 1.5, 1.0, 0.5, 2.5, 1.5, 2.0 (sırayla)")
print("="*50)
```

```python
def myFunc(num):
    return num ** 3
```

```python
myList = [2,3,4,5,6]
```

```python
# ==================== CEVAP - SORU 5 - YÖNTEM 1 ====================
"""
📌 **DETAYLI AÇIKLAMA:**

🎯 **GÖREV:** Her sayının küpünü (3. kuvvetini) hesaplayarak yeni liste oluştur

🔢 **MATEMATİK:**
   num ** 3 → num × num × num
   2³ = 8, 3³ = 27, 4³ = 64, 5³ = 125, 6³ = 216

💡 **ÇÖZÜM YÖNTEMLERİ:**

✅ **YÖNTEM 1: List Comprehension** (En Pythonic ve Performanslı!)
   [myFunc(num) for num in myList]
   
   ➡️ Her elemanı tek satırda fonksiyona tabi tut
   ➡️ Hızlı ve okunabilir
   ➡️ Profesyonel Python kodu!
"""

def myFunc(num):
    return num ** 3

myList = [2,3,4,5,6]

# 🌟 YÖNTEM 1: List Comprehension (ÖNERİLEN)
newList1 = [myFunc(num) for num in myList]
print("🎯 List Comprehension Sonucu:")
print(f"Orijinal Liste: {myList}")
print(f"Küpleri Alınmış: {newList1}")
print(f"\n✨ Sonuç: {newList1}")
```

```python
# ==================== CEVAP - SORU 5 - YÖNTEM 2 ====================
"""
✅ **YÖNTEM 2: map() Fonksiyonu** (Fonksiyonel Programlama Yaklaşımı)
   
🔍 **map() NEDİR?**
   ➡️ Bir fonksiyonu, iterable'ın her elemanına uygular
   ➡️ map object döner (liste yapmak için list() kullanılır)
   ➡️ Memory efficient (hafıza dostu)
   
📊 **KULLANIM:**
   map(fonksiyon, iterable)
   list(map(myFunc, myList))
"""

def myFunc(num):
    return num ** 3

myList = [2,3,4,5,6]

# 🌟 YÖNTEM 2: map() Fonksiyonu
newList2 = list(map(myFunc, myList))
print("🎯 map() Fonksiyonu Sonucu:")
print(f"Orijinal Liste: {myList}")
print(f"Küpleri Alınmış: {newList2}")

print("\n" + "="*60)
print("📚 **BONUS BİLGİ:** Lambda ile de kullanılabilir:")
newList3 = list(map(lambda x: x**3, myList))
print(f"Lambda ile: {newList3}")
print("="*60)
```

```python
#6) Aşağıdaki string dizisinde, içinde sadece XYZ geçen barkodları gösterecek yeni bir liste oluşturunuz
```

```python
barkodDizisi = ["ABC231","SA3123XYZ","XYZA123Q","QRE1231KJ","X112QGL"]
```

```python
# ==================== CEVAP - SORU 6 - YÖNTEM 1 ====================
"""
📌 **DETAYLI AÇIKLAMA:**

🎯 **GÖREV:** İçinde "XYZ" geçen barkodları filtrele

🔍 **STRING ARAMA YÖNTEMLERİ:**

✅ **YÖNTEM 1: List Comprehension + in Operatörü** (En Basit!)
   "XYZ" in string → String içinde XYZ varsa True döner
   
📊 **ANALİZ:**
   "ABC231"      → ❌ XYZ YOK
   "SA3123XYZ"   → ✅ XYZ VAR (sonda)
   "XYZA123Q"    → ✅ XYZ VAR (başta)
   "QRE1231KJ"   → ❌ XYZ YOK
   "X112QGL"     → ❌ XYZ YOK

💡 **ÖNEMLİ:** "in" operatörü case-sensitive (büyük/küçük harf duyarlı)
"""

barkodDizisi = ["ABC231","SA3123XYZ","XYZA123Q","QRE1231KJ","X112QGL"]

# 🌟 YÖNTEM 1: List Comprehension (ÖNERİLEN - En Okunabilir)
xyzBarkodlar1 = [barkod for barkod in barkodDizisi if "XYZ" in barkod]

print("🎯 List Comprehension Sonucu:")
print(f"Tüm Barkodlar: {barkodDizisi}")
print(f"XYZ İçerenler: {xyzBarkodlar1}")
print(f"\n✨ Sonuç: {len(xyzBarkodlar1)} adet barkod bulundu!")
```

```python
# ==================== CEVAP - SORU 6 - YÖNTEM 2 ====================
"""
✅ **YÖNTEM 2: filter() Fonksiyonu** (Fonksiyonel Programlama)

🔍 **filter() NEDİR?**
   ➡️ Bir koşulu sağlayan elemanları filtreler
   ➡️ filter(fonksiyon, iterable)
   ➡️ Fonksiyon True/False döndürmelidir
   
📊 **KULLANIM:** Lambda veya normal fonksiyon ile
"""

barkodDizisi = ["ABC231","SA3123XYZ","XYZA123Q","QRE1231KJ","X112QGL"]

# 🌟 YÖNTEM 2: filter() + lambda
xyzBarkodlar2 = list(filter(lambda barkod: "XYZ" in barkod, barkodDizisi))

print("🎯 filter() Fonksiyonu Sonucu:")
print(f"XYZ İçeren Barkodlar: {xyzBarkodlar2}")

print("\n" + "="*60)
print("📚 **ALTERNATİF:** Klasik For Döngüsü ile:")
xyzBarkodlar3 = []
for barkod in barkodDizisi:
    if "XYZ" in barkod:
        xyzBarkodlar3.append(barkod)
print(f"For Döngüsü ile: {xyzBarkodlar3}")
print("="*60)

print("\n💡 **PROFESYONEL TAVSİYE:** List comprehension tercih edilir!")
```

```python
myVar = "Atil Samancioglu"

def ornekFonksiyon():
    myVar = "Atil"
    
    def digerFonksiyon():
        print(myVar)
    
    digerFonksiyon()
```

```python
#ornekFonksiyon()
```

```python
# ==================== CEVAP - SORU 7 ====================
"""
📌 **DETAYLI AÇIKLAMA:**

🔍 **SCOPE (KAPSAM) KAVRAMI - Çok Önemli!**

Bu soru **LEGB Kuralı**nı test ediyor:
   🔤 L - Local (Yerel)
   🔤 E - Enclosing (Kapsayan)
   🔤 G - Global (Küresel)
   🔤 B - Built-in (Yerleşik)

📊 **KODUN DETAYLI ANALİZİ:**

1️⃣ **Global Scope (En dış):**
   myVar = "Atil Samancioglu"  ← Global değişken

2️⃣ **ornekFonksiyon içinde (Local Scope):**
   myVar = "Atil"  ← Bu fonksiyonun LOCAL değişkeni

3️⃣ **digerFonksiyon içinde:**
   print(myVar) çağrılıyor
   
⚠️ **KRİTİK NOKTA:**
   ➡️ digerFonksiyon kendi local scope'unda myVar bulamıyor
   ➡️ Bir üst scope'a bakıyor (ENCLOSING)
   ➡️ ornekFonksiyon'un myVar'ını buluyor: "Atil"
   ➡️ Global myVar'a hiç bakmıyor!

✅ **CEVAP: "Atil"**

💡 **ÖNEMLİ:** İç fonksiyon, kendinden önceki dış fonksiyonun değişkenine erişebilir!
"""

myVar = "Atil Samancioglu"  # Global

def ornekFonksiyon():
    myVar = "Atil"  # Enclosing (Kapsayan)
    
    def digerFonksiyon():
        print(myVar)  # Enclosing scope'taki "Atil" yazdırır
    
    digerFonksiyon()

print("🎯 ÇIKTI:")
ornekFonksiyon()  # Çıktı: Atil

print("\n" + "="*60)
print("📝 CEVAP: 'Atil' yazdırılır (Enclosing scope)")
print("="*60)
```

```python
class Kedi():
        
    def __init__(self,isim,yas=5):
        self.isim = isim
        self.yas = yas
        
    def yasiCarp(self):
        return self.yas * 3
```

```python
kedim = Kedi("Tonton")
```

```python
#kedim.yasiCarp()
```

```python
# ==================== CEVAP - SORU 8 ====================
"""
📌 **DETAYLI AÇIKLAMA:**

🎯 **OOP (Object-Oriented Programming) - Nesne Yönelimli Programlama**

🔍 **SINIF YAPISI ANALİZİ:**

**__init__ Metodu (Constructor - Yapıcı):**
   ➡️ Nesne oluşturulduğunda otomatik çalışır
   ➡️ self.isim ve self.yas attribute'ları (özellikleri) tanımlanır
   ➡️ yas parametresi **DEFAULT DEĞERE** sahip: yas=5

**yasiCarp Metodu:**
   ➡️ self.yas değerini 3 ile çarpar
   ➡️ **return** ile sonucu döndürür

📊 **NESNE OLUŞTURMA VE ÇAĞIRMA:**
   kedim = Kedi("Tonton")
   
   ⚠️ **DİKKAT:** Sadece isim verildi, yas verilmedi!
   ➡️ isim = "Tonton"
   ➡️ yas = 5 (default değer kullanıldı)

📈 **HESAPLAMA:**
   kedim.yasiCarp() → return self.yas * 3
   → return 5 * 3
   → return 15

✅ **CEVAP: 15**

💡 **KEY POINT:** Default parametreler, sınıflarda da çok kullanışlıdır!
"""

class Kedi():
    def __init__(self, isim, yas=5):
        self.isim = isim
        self.yas = yas
    
    def yasiCarp(self):
        return self.yas * 3

kedim = Kedi("Tonton")  # yas belirtilmedi → yas=5 (default)

print("🎯 ÇIKTI:")
sonuc = kedim.yasiCarp()
print(f"Kedinin yaşı: {kedim.yas}")
print(f"Yaş × 3 = {sonuc}")

print("\n" + "="*60)
print("📝 CEVAP: 15")
print("="*60)
```

```python
class Ogrenci():
    
    def __init__(self,isim,sinavNotu):
        self.isim = isim
        self.__sinavNotu = sinavNotu
    
    def notuGoster(self):
        print(f"{self.isim} sınav notu: {self.__sinavNotu}")
```

```python
ogrenci = Ogrenci("Mehmet",85)
```

```python
ogrenci.__sinavNotu = 75
```

```python
#ogrenci.notuGoster()
```

```python
# ==================== CEVAP - SORU 9 ====================
"""
📌 **DETAYLI AÇIKLAMA:**

🔒 **ENCAPSULATION (KAPSÜLLEME) - OOP'nin Temel Prensibi!**

🔍 **NAME MANGLING (İsim Bozma) KAVRAMI:**

**Private Attribute (Özel Özellik):**
   self.__sinavNotu  ← Başında **çift alt çizgi (__)** var!
   
⚠️ **KRİTİK NOKTA - Name Mangling:**
   ➡️ Python, __sinavNotu ismini **otomatik olarak değiştirir**
   ➡️ Gerçek ismi: _Ogrenci__sinavNotu olur
   ➡️ Dışarıdan direkt erişimi **ENGELLER** (encapsulation)

📊 **KOD ANALİZİ ADIM ADIM:**

1️⃣ **Nesne Oluşturma:**
   ogrenci = Ogrenci("Mehmet", 85)
   ➡️ Gerçek attribute: _Ogrenci__sinavNotu = 85

2️⃣ **Dışarıdan Değiştirme Denemesi:**
   ogrenci.__sinavNotu = 75
   ⚠️ **BU YANLIŞ!** Yeni bir public attribute yaratır!
   ➡️ ogrenci.__sinavNotu = 75 (yeni, farklı bir değişken!)
   ➡️ _Ogrenci__sinavNotu = 85 (eski, değişmedi!)

3️⃣ **notuGoster() Çağrısı:**
   ➡️ self.__sinavNotu kullanır
   ➡️ _Ogrenci__sinavNotu'na bakar (Name Mangling)
   ➡️ Değer: 85 (hiç değişmedi!)

✅ **CEVAP: "Mehmet sınav notu: 85"**

💡 **ÖĞRENME NOKTASI:** Private değişkenlere dışarıdan erişim çalışmaz!
"""

class Ogrenci():
    def __init__(self, isim, sinavNotu):
        self.isim = isim
        self.__sinavNotu = sinavNotu  # Private attribute
    
    def notuGoster(self):
        print(f"{self.isim} sınav notu: {self.__sinavNotu}")

ogrenci = Ogrenci("Mehmet", 85)
ogrenci.__sinavNotu = 75  # YENİ bir public attribute yaratır, private'ı değiştirmez!

print("🎯 ÇIKTI:")
ogrenci.notuGoster()  # 85 yazdırır (private değer değişmedi!)

print("\n" + "="*60)
print("🔍 DETAYLI ANALİZ:")
print(f"Public __sinavNotu: {ogrenci.__sinavNotu}")  # 75 (yeni yaratılan)
print(f"Private (gerçek): {ogrenci._Ogrenci__sinavNotu}")  # 85 (orijinal)
print("="*60)

print("\n📝 CEVAP: 'Mehmet sınav notu: 85'")
print("💡 Private değişken değişmedi, yeni bir public değişken yaratıldı!")
```

```python
# aynı zamanda büyük projelerde bize yapısal olarak fayda sağlayabilecek OOP prensibinin adı nedir?
```

```python
# ==================== CEVAP - SORU 10 ====================
"""
📌 **DETAYLI AÇIKLAMA:**

🎯 **SORUNUN SORDUĞU KAVRAM:**
   ✅ Soyut sınıflar (Abstract Classes)
   ✅ Soyut methodlar (Abstract Methods)
   ✅ Yapısal planlama
   ✅ Büyük projelerde organizasyon

🔍 **CEVAP: ABSTRACTION (SOYUTLAMA)**

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

📚 **ABSTRACTION NEDİR?**

**Tanım:** OOP'nin 4 temel prensibinden biri olan Abstraction,
karmaşık sistemleri basitleştirerek sadece **gerekli özellikleri**
göstermeyi ve gereksiz detayları gizlemeyi sağlar.

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🏗️ **ABSTRACTION'IN ÖZELLİKLERİ:**

✅ **Abstract Base Class (ABC):** Direkt kullanılamaz, sadece miras alınır
✅ **Abstract Methods:** Alt sınıflar bu methodları **MUTLAKA** implement eder
✅ **Blueprint (Şablon):** Diğer sınıflar için yapısal rehber sağlar
✅ **Consistency (Tutarlılık):** Tüm alt sınıflar aynı yapıyı takip eder
✅ **Flexibility (Esneklik):** Her alt sınıf kendi implementasyonunu yapar

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

💡 **NEDEN ÖNEMLİDİR?**

🔸 **Büyük Projelerde:** Kod organizasyonunu kolaylaştırır
🔸 **Takım Çalışmasında:** Herkes aynı yapıyı takip eder
🔸 **Hata Önleme:** Eksik implementasyon hata verir
🔸 **Bakım Kolaylığı:** Kod daha anlaşılır ve sürdürülebilir

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🎯 **PYTHON'DA KULLANIMI:**

Python'da abstraction için **abc** modülü kullanılır:
   • ABC (Abstract Base Class)
   • @abstractmethod decorator
"""

from abc import ABC, abstractmethod

# 🌟 ÖRNEK: Abstract Class Tanımı
class Hayvan(ABC):  # ABC'den miras alınır
    
    @abstractmethod  # Bu method mutlaka implement edilmeli
    def ses_cikar(self):
        pass
    
    @abstractmethod
    def hareket_et(self):
        pass

# ✅ Concrete Class (Somut Sınıf) - Abstract class'ı implement eder
class Kedi(Hayvan):
    def ses_cikar(self):
        return "Miyav!"
    
    def hareket_et(self):
        return "Kedi yürüyor..."

class Kopek(Hayvan):
    def ses_cikar(self):
        return "Hav hav!"
    
    def hareket_et(self):
        return "Köpek koşuyor..."

# Test edelim
kedi = Kedi()
kopek = Kopek()

print("🐱 KEDİ:")
print(f"   Ses: {kedi.ses_cikar()}")
print(f"   Hareket: {kedi.hareket_et()}")

print("\n🐶 KÖPEK:")
print(f"   Ses: {kopek.ses_cikar()}")
print(f"   Hareket: {kopek.hareket_et()}")

print("\n" + "="*60)
print("📝 **CEVAP: ABSTRACTION (Soyutlama)**")
print("="*60)

print("\n🎓 **OOP'NİN 4 TEMEL PRENSİBİ:**")
print("   1️⃣ Encapsulation (Kapsülleme)")
print("   2️⃣ Inheritance (Kalıtım)")
print("   3️⃣ Polymorphism (Çok Biçimlilik)")
print("   4️⃣ Abstraction (Soyutlama) ← SORUDA SORULAN!")
print("="*60)
```
