````python
# Rastgelelik için seed belirleme (tekrarlanabilirlik için)
np.random.seed(42)
random.seed(42)
Faker.seed(42)

# Veri seti boyutu
n_records = 34563

# Başlangıç tarihi: 2 yıllık veri (2024-2026)
start_date = datetime(2024, 1, 1)
end_date = datetime(2026, 2, 7)

print("🔄 Kapıdağ Glamping CRM veri seti oluşturuluyor...")
print(f"📊 Toplam kayıt sayısı: {n_records:,}")

# Tarih ve saat üretimi
def random_date(start, end):
    delta = end - start
    random_days = random.randint(0, delta.days)
    random_hour = random.randint(10, 22)  # Check-in saatleri 10:00-22:00 arası
    random_minute = random.randint(0, 59)
    return start + timedelta(days=random_days, hours=random_hour, minutes=random_minute)

# Tarihler
dates = [random_date(start_date, end_date) for _ in range(n_records)]
dates.sort()  # Kronolojik sıralama

# Sezon belirleme fonksiyonu
def get_season(date):
    month = date.month
    if month in [3, 4, 5]: return 'İlkbahar'
    elif month in [6, 7, 8]: return 'Yaz'
    elif month in [9, 10, 11]: return 'Sonbahar'
    else: return 'Kış'

# Müşteri tipleri ve ağırlıkları (gerçekçi dağılım)
musteri_tipleri = ['Çift', 'Aile', 'Arkadaş Grubu', 'Tek Kişi']
musteri_weights = [0.45, 0.30, 0.20, 0.05]  # Çiftler en popüler

# Çadır tipleri ve fiyat aralıkları
cadir_tipleri = ['Lüks', 'Royal', 'Standart', 'Ekonomik']
cadir_weights = [0.25, 0.35, 0.30, 0.10]
cadir_fiyat_base = {'Lüks': 3500, 'Royal': 2800, 'Standart': 2000, 'Ekonomik': 1500}

# Ödeme yöntemleri
odeme_yontemleri = ['Kredi Kartı', 'Nakit', 'Havale', 'Online Ödeme']
odeme_weights = [0.50, 0.15, 0.10, 0.25]

# Yaş grupları
yas_gruplari = ['18-25', '26-35', '36-45', '46-60', '60+']
yas_weights = [0.15, 0.35, 0.30, 0.15, 0.05]

# Veri setini oluştur
data = {
    'rezervasyon_id': [f'RZV{str(i+1).zfill(5)}' for i in range(n_records)],
    'tarih_saat': dates,
    'musteri_tipi': random.choices(musteri_tipleri, weights=musteri_weights, k=n_records),
    'cadir_tipi': random.choices(cadir_tipleri, weights=cadir_weights, k=n_records),
    'yas_grubu': random.choices(yas_gruplari, weights=yas_weights, k=n_records),
    'odeme_yontemi': random.choices(odeme_yontemleri, weights=odeme_weights, k=n_records),
}

# DataFrame oluştur
df = pd.DataFrame(data)

# Sezon ekle
df['sezon'] = df['tarih_saat'].apply(get_season)

# Konaklama gece sayısı (müşteri tipine göre değişken)
def konaklama_gece(musteri_tipi):
    if musteri_tipi == 'Tek Kişi': return np.random.choice([1, 2], p=[0.7, 0.3])
    elif musteri_tipi == 'Çift': return np.random.choice([2, 3, 4], p=[0.5, 0.3, 0.2])
    elif musteri_tipi == 'Aile': return np.random.choice([3, 4, 5, 7], p=[0.3, 0.4, 0.2, 0.1])
    else: return np.random.choice([2, 3, 4], p=[0.4, 0.4, 0.2])

df['konaklama_gece'] = df['musteri_tipi'].apply(konaklama_gece)

# Toplam tutar hesaplama (çadır tipi + gece sayısı + sezon etkisi)
def hesapla_tutar(row):
    base = cadir_fiyat_base[row['cadir_tipi']]
    gece = row['konaklama_gece']
    
    # Sezon çarpanı
    sezon_carpan = {'Yaz': 1.3, 'İlkbahar': 1.1, 'Sonbahar': 1.0, 'Kış': 0.85}
    
    # Toplam hesaplama
    toplam = base * gece * sezon_carpan[row['sezon']]
    
    # Rastgele varyasyon (%±10)
    toplam *= np.random.uniform(0.9, 1.1)
    
    return round(toplam, 2)

df['toplam_tutar'] = df.apply(hesapla_tutar, axis=1)

# Check-in süresi (dakika) - gerçekçi dağılım
# Yoğun saatlerde daha uzun sürer
def checkin_suresi(tarih_saat):
    hour = tarih_saat.hour
    if 14 <= hour <= 17:  # Yoğun saatler
        return int(np.random.gamma(shape=3, scale=5) + 10)  # 10-40 dk arası
    else:  # Sakin saatler
        return int(np.random.gamma(shape=2, scale=3) + 5)  # 5-20 dk arası

df['checkin_sure_dk'] = df['tarih_saat'].apply(checkin_suresi)

# Memnuniyet puanı (1.0-5.0)
# Lüks çadırlar daha yüksek memnuniyet, uzun check-in düşük memnuniyet
def memnuniyet_puani(row):
    # Base puan: çadır tipine göre
    cadir_base = {'Lüks': 4.5, 'Royal': 4.2, 'Standart': 3.8, 'Ekonomik': 3.5}
    base = cadir_base[row['cadir_tipi']]
    
    # Check-in süresi etkisi (uzun süre = düşük memnuniyet)
    if row['checkin_sure_dk'] > 30:
        base -= 0.3
    
    # Rastgele varyasyon
    puan = base + np.random.normal(0, 0.3)
    
    return round(np.clip(puan, 1.0, 5.0), 1)

df['memnuniyet_puan'] = df.apply(memnuniyet_puani, axis=1)

# Ekstra hizmetler (Boolean)
# Lüks ve Royal çadırlarda daha fazla ekstra hizmet alınır
def ekstra_hizmet_olasiligi(cadir_tipi, hizmet_tipi):
    if hizmet_tipi == 'kahvalti':
        prob = {'Lüks': 0.9, 'Royal': 0.85, 'Standart': 0.70, 'Ekonomik': 0.50}
    elif hizmet_tipi == 'somine':
        prob = {'Lüks': 0.7, 'Royal': 0.6, 'Standart': 0.40, 'Ekonomik': 0.20}
    else:  # aktivite
        prob = {'Lüks': 0.5, 'Royal': 0.45, 'Standart': 0.30, 'Ekonomik': 0.15}
    return prob[cadir_tipi]

df['ekstra_kahvalti'] = df['cadir_tipi'].apply(
    lambda x: np.random.random() < ekstra_hizmet_olasiligi(x, 'kahvalti')
)
df['ekstra_somine'] = df['cadir_tipi'].apply(
    lambda x: np.random.random() < ekstra_hizmet_olasiligi(x, 'somine')
)
df['ekstra_aktivite'] = df['cadir_tipi'].apply(
    lambda x: np.random.random() < ekstra_hizmet_olasiligi(x, 'aktivite')
)

# Önceki ziyaret sayısı (müşteri sadakati)
df['onceki_ziyaret'] = np.random.choice([0, 1, 2, 3, 4, 5, 6, 7, 8], 
                                        p=[0.40, 0.25, 0.15, 0.08, 0.05, 0.03, 0.02, 0.01, 0.01],
                                        size=n_records)

# Ek harcama (konaklama dışında yapılan harcama)
# Lüks müşteriler daha fazla ek harcama yapar
def ek_harcama_hesapla(row):
    cadir_carpan = {'Lüks': 800, 'Royal': 600, 'Standart': 400, 'Ekonomik': 200}
    base = cadir_carpan[row['cadir_tipi']]
    gece = row['konaklama_gece']
    
    # Ekstra hizmetler varsa ek harcama artar
    ekstra_carpan = 1.0
    if row['ekstra_kahvalti']: ekstra_carpan += 0.2
    if row['ekstra_somine']: ekstra_carpan += 0.15
    if row['ekstra_aktivite']: ekstra_carpan += 0.25
    
    harcama = base * np.sqrt(gece) * ekstra_carpan * np.random.uniform(0.7, 1.3)
    return round(harcama, 2)

df['ek_harcama'] = df.apply(ek_harcama_hesapla, axis=1)

print("\n✅ Veri seti başarıyla oluşturuldu!")
print(f"📊 Boyut: {df.shape[0]:,} satır × {df.shape[1]} sütun")
print(f"📅 Tarih aralığı: {df['tarih_saat'].min()} - {df['tarih_saat'].max()}")
````

**İş Problemi:** Kapıdağ Glamping'de check-in süreleri uzadığında müşteri memnuniyeti düşüyor. Hangi müşteri tipi ve çadır tipinde problem yaşanıyor?

### 📊 Caption 4'ten Öğrendiklerimiz
> "Birkaç Scatter Plot'u aynı graf üzerinde gösterebilirsek, işlem süreleri sorunlu işlemlere dair tespitleri rahatlıkla görebiliriz."

**Çözüm:** Farklı segmentleri aynı grafikte görselleştirerek **karşılaştırmalı analiz** yapalım.

````python
# ADIM 4: Çoklu Scatter Plot - İşlem Süresi Analizi
app4 = JupyterDash(__name__)

# Sorunlu işlemleri belirle (checkin > 20 dk VE memnuniyet < 7)
df['sorunlu_islem'] = ((df['checkin_sure_dk'] > 20) & (df['memnuniyet_puan'] < 7))

app4.layout = html.Div([
    html.H1("🚨 Check-in Süresi ve Memnuniyet Analizi", 
            style={'textAlign': 'center', 'color': '#C0392B', 'marginBottom': 30}),
    
    html.Div([
        html.Div([
            html.Label("Çadır Tipi:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='cadir-dropdown-4',
                options=[{'label': 'Tümü', 'value': 'Tümü'}] + 
                        [{'label': tip, 'value': tip} for tip in df['cadir_tipi'].unique()],
                value='Tümü',
                style={'width': '200px'}
            )
        ], style={'display': 'inline-block', 'marginRight': 30}),
        
        html.Div([
            html.Label("Sorunlu İşlemleri Vurgula:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.RadioItems(
                id='sorun-radio-4',
                options=[
                    {'label': ' Hayır', 'value': 'hayir'},
                    {'label': ' Evet', 'value': 'evet'}
                ],
                value='hayir',
                inline=True
            )
        ], style={'display': 'inline-block'})
    ], style={'marginBottom': 30, 'textAlign': 'center'}),
    
    dcc.Graph(id='checkin-grafik-4')
])

@app4.callback(
    Output('checkin-grafik-4', 'figure'),
    [Input('cadir-dropdown-4', 'value'),
     Input('sorun-radio-4', 'value')]
)
def update_checkin_graph(secilen_cadir, sorun_vurgu):
    # Veriyi filtrele
    if secilen_cadir == 'Tümü':
        filtered = df.copy()
    else:
        filtered = df[df['cadir_tipi'] == secilen_cadir]
    
    # Sorunlu işlemlere göre renklendirme
    if sorun_vurgu == 'evet':
        filtered['renk'] = filtered['sorunlu_islem'].map({
            True: '🚨 Sorunlu İşlem',
            False: '✅ Normal İşlem'
        })
        color_col = 'renk'
        color_map = {
            '🚨 Sorunlu İşlem': '#E74C3C',
            '✅ Normal İşlem': '#27AE60'
        }
    else:
        color_col = 'musteri_tipi'
        color_map = {
            'Çift': '#3498DB',
            'Aile': '#E67E22',
            'Arkadaş Grubu': '#9B59B6',
            'Tek Kişi': '#1ABC9C'
        }
    
    # Scatter plot oluştur
    fig = px.scatter(
        filtered,
        x='checkin_sure_dk',
        y='memnuniyet_puan',
        color=color_col,
        size='toplam_tutar',
        hover_data=['rezervasyon_id', 'cadir_tipi', 'konaklama_gece'],
        title=f"Check-in Süresi vs Memnuniyet ({secilen_cadir} - {len(filtered):,} Kayıt)",
        color_discrete_map=color_map,
        opacity=0.7
    )
    
    # Sorun eşik çizgileri ekle
    if sorun_vurgu == 'evet':
        fig.add_hline(y=7, line_dash="dash", line_color="red", 
                      annotation_text="Kritik Memnuniyet Eşiği")
        fig.add_vline(x=20, line_dash="dash", line_color="orange", 
                      annotation_text="Kritik Süre Eşiği")
    
    fig.update_layout(
        xaxis_title="Check-in Süresi (Dakika)",
        yaxis_title="Memnuniyet Puanı (1-10)",
        hovermode='closest',
        template='plotly_white',
        height=600,
        font=dict(size=12)
    )
    
    return fig

print("✅ ADIM 4 Dash uygulaması hazır! (Check-in Analizi)")
````

## 📦 **ADIM 5: Box Plot ve Violin Plot - Segment Dağılım Analizi**

**İş Problemi:** Kapıdağ Glamping'de hangi müşteri segmentinin harcama davranışları daha homojen, hangisi daha heterojen?

### 📊 Caption 5'ten Öğrendiklerimiz
> "Box Plot ve Violin Plot, veri dağılımlarını karşılaştırmak ve outlier'ları tespit etmek için idealdir."

**Analiz Hedefi:**
- **Box Plot:** Çeyreklikler, medyan, outlier'ları görmek
- **Violin Plot:** Dağılımın yoğunluğunu (density) görmek
- **Karşılaştırma:** Hangi segmentin davranışı daha tahmin edilebilir?

````python
# ADIM 5: Box Plot ve Violin Plot Analizi
app5 = JupyterDash(__name__)

app5.layout = html.Div([
    html.H1("📦 Segment Bazlı Harcama Dağılımı", 
            style={'textAlign': 'center', 'color': '#8E44AD', 'marginBottom': 30}),
    
    html.Div([
        html.Div([
            html.Label("Grafik Tipi:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.RadioItems(
                id='grafik-tipi-5',
                options=[
                    {'label': ' Box Plot', 'value': 'box'},
                    {'label': ' Violin Plot', 'value': 'violin'}
                ],
                value='box',
                inline=True,
                style={'marginRight': 30}
            )
        ], style={'display': 'inline-block', 'marginRight': 40}),
        
        html.Div([
            html.Label("Analiz Değişkeni:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='degisken-dropdown-5',
                options=[
                    {'label': 'Toplam Harcama', 'value': 'toplam_tutar'},
                    {'label': 'Ek Harcama', 'value': 'ek_harcama'},
                    {'label': 'Konaklama Süresi', 'value': 'konaklama_gece'},
                    {'label': 'Memnuniyet Puanı', 'value': 'memnuniyet_puan'}
                ],
                value='toplam_tutar',
                style={'width': '250px'}
            )
        ], style={'display': 'inline-block', 'marginRight': 40}),
        
        html.Div([
            html.Label("Segment Tipi:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='segment-dropdown-5',
                options=[
                    {'label': 'Müşteri Tipi', 'value': 'musteri_tipi'},
                    {'label': 'Çadır Tipi', 'value': 'cadir_tipi'},
                    {'label': 'Sezon', 'value': 'sezon'}
                ],
                value='musteri_tipi',
                style={'width': '200px'}
            )
        ], style={'display': 'inline-block'})
    ], style={'marginBottom': 30, 'textAlign': 'center'}),
    
    dcc.Graph(id='dagilim-grafik-5'),
    
    # İstatistiksel özet
    html.Div(id='istatistik-ozet-5', 
             style={'marginTop': 20, 'padding': 20, 'backgroundColor': '#F8F9FA', 
                    'borderRadius': 10, 'fontSize': 14})
])

@app5.callback(
    [Output('dagilim-grafik-5', 'figure'),
     Output('istatistik-ozet-5', 'children')],
    [Input('grafik-tipi-5', 'value'),
     Input('degisken-dropdown-5', 'value'),
     Input('segment-dropdown-5', 'value')]
)
def update_distribution(grafik_tipi, degisken, segment):
    # Grafik oluştur
    if grafik_tipi == 'box':
        fig = px.box(
            df,
            x=segment,
            y=degisken,
            color=segment,
            points='outliers',  # Sadece outlier'ları göster
            title=f"Box Plot: {degisken.replace('_', ' ').title()} - {segment.replace('_', ' ').title()}"
        )
    else:  # violin
        fig = px.violin(
            df,
            x=segment,
            y=degisken,
            color=segment,
            box=True,  # İçine box plot da ekle
            points='all',  # Tüm noktaları göster
            title=f"Violin Plot: {degisken.replace('_', ' ').title()} - {segment.replace('_', ' ').title()}"
        )
    
    fig.update_layout(
        xaxis_title=segment.replace('_', ' ').title(),
        yaxis_title=degisken.replace('_', ' ').title(),
        template='plotly_white',
        height=550,
        showlegend=False
    )
    
    # İstatistiksel özet hesapla
    stats = df.groupby(segment)[degisken].agg(['mean', 'median', 'std', 'min', 'max']).round(2)
    
    # Özet tablosu oluştur
    ozet_rows = []
    for idx, row in stats.iterrows():
        ozet_rows.append(
            html.Div([
                html.B(f"{idx}: ", style={'color': '#2C3E50'}),
                f"Ortalama={row['mean']}, Medyan={row['median']}, "
                f"Std.Sapma={row['std']}, Min={row['min']}, Max={row['max']}"
            ], style={'marginBottom': 8})
        )
    
    ozet_div = html.Div([
        html.H4("📊 İstatistiksel Özet", style={'color': '#34495E', 'marginBottom': 15}),
        *ozet_rows
    ])
    
    return fig, ozet_div

print("✅ ADIM 5 Dash uygulaması hazır! (Box/Violin Plot)")
````

### 📊 Caption 6'dan Öğrendiklerimiz
> "Histogram'lar frekans dağılımlarını gösterir. Facet Grid ile birden fazla histogram'ı alt panellerde karşılaştırabiliriz."

**Analiz Hedefi:**
- Her segment için ayrı histogram
- Normal dağılıma uygunluk kontrolü
- Segment bazlı davranış farklılıkları

````python
# ADIM 6: Histogram ve Facet Grid
app6 = JupyterDash(__name__)

app6.layout = html.Div([
    html.H1("📊 Çok Boyutlu Histogram Analizi", 
            style={'textAlign': 'center', 'color': '#16A085', 'marginBottom': 30}),
    
    html.Div([
        html.Div([
            html.Label("Analiz Değişkeni:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='histogram-degisken-6',
                options=[
                    {'label': 'Toplam Harcama (TL)', 'value': 'toplam_tutar'},
                    {'label': 'Ek Harcama (TL)', 'value': 'ek_harcama'},
                    {'label': 'Konaklama Süresi (Gece)', 'value': 'konaklama_gece'},
                    {'label': 'Check-in Süresi (dk)', 'value': 'checkin_sure_dk'}
                ],
                value='toplam_tutar',
                style={'width': '280px'}
            )
        ], style={'display': 'inline-block', 'marginRight': 40}),
        
        html.Div([
            html.Label("Facet (Alt Panel):", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='facet-dropdown-6',
                options=[
                    {'label': 'Yok (Tek Grafik)', 'value': 'yok'},
                    {'label': 'Müşteri Tipi', 'value': 'musteri_tipi'},
                    {'label': 'Sezon', 'value': 'sezon'},
                    {'label': 'Çadır Tipi', 'value': 'cadir_tipi'}
                ],
                value='musteri_tipi',
                style={'width': '250px'}
            )
        ], style={'display': 'inline-block', 'marginRight': 40}),
        
        html.Div([
            html.Label("Bin Sayısı:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Slider(
                id='bin-slider-6',
                min=10,
                max=100,
                step=10,
                value=30,
                marks={i: str(i) for i in range(10, 101, 20)},
                tooltip={"placement": "bottom", "always_visible": True}
            )
        ], style={'display': 'inline-block', 'width': '300px'})
    ], style={'marginBottom': 40, 'textAlign': 'center'}),
    
    dcc.Graph(id='histogram-grafik-6')
])

@app6.callback(
    Output('histogram-grafik-6', 'figure'),
    [Input('histogram-degisken-6', 'value'),
     Input('facet-dropdown-6', 'value'),
     Input('bin-slider-6', 'value')]
)
def update_histogram(degisken, facet, bins):
    if facet == 'yok':
        # Tek histogram
        fig = px.histogram(
            df,
            x=degisken,
            nbins=bins,
            marginal='box',  # Üstte box plot ekle
            title=f"Histogram: {degisken.replace('_', ' ').title()}",
            color_discrete_sequence=['#3498DB']
        )
    else:
        # Facet grid ile çoklu histogram
        fig = px.histogram(
            df,
            x=degisken,
            nbins=bins,
            facet_col=facet,
            facet_col_wrap=2,  # 2 sütunlu grid
            color=facet,
            title=f"Histogram: {degisken.replace('_', ' ').title()} - {facet.replace('_', ' ').title()} Bazlı"
        )
        
        # Facet başlıklarını düzenle
        fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))
    
    fig.update_layout(
        xaxis_title=degisken.replace('_', ' ').title(),
        yaxis_title="Frekans (Adet)",
        template='plotly_white',
        height=600 if facet == 'yok' else 800,
        showlegend=False
    )
    
    return fig

print("✅ ADIM 6 Dash uygulaması hazır! (Histogram ve Facet Grid)")
````

````python
# ADIM 7: Time Series Line Plot
app7 = JupyterDash(__name__)

# Zaman bazlı agregasyonlar hazırla
df['tarih'] = pd.to_datetime(df['tarih_saat']).dt.date
df['yil_ay'] = pd.to_datetime(df['tarih_saat']).dt.to_period('M').astype(str)

app7.layout = html.Div([
    html.H1("📈 Zaman Serisi Analizi ve Darboğaz Tespiti", 
            style={'textAlign': 'center', 'color': '#2980B9', 'marginBottom': 30}),
    
    html.Div([
        html.Div([
            html.Label("Zaman Periyodu:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.RadioItems(
                id='periyod-radio-7',
                options=[
                    {'label': ' Günlük', 'value': 'gunluk'},
                    {'label': ' Aylık', 'value': 'aylik'}
                ],
                value='aylik',
                inline=True,
                style={'marginRight': 40}
            )
        ], style={'display': 'inline-block', 'marginRight': 40}),
        
        html.Div([
            html.Label("Analiz Metriği:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='metrik-dropdown-7',
                options=[
                    {'label': 'Rezervasyon Sayısı', 'value': 'count'},
                    {'label': 'Ortalama Harcama (TL)', 'value': 'toplam_tutar'},
                    {'label': 'Ortalama Memnuniyet', 'value': 'memnuniyet_puan'},
                    {'label': 'Ortalama Check-in Süresi (dk)', 'value': 'checkin_sure_dk'}
                ],
                value='count',
                style={'width': '300px'}
            )
        ], style={'display': 'inline-block', 'marginRight': 40}),
        
        html.Div([
            html.Label("Segment Karşılaştırması:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='segment-dropdown-7',
                options=[
                    {'label': 'Yok (Toplam)', 'value': 'yok'},
                    {'label': 'Müşteri Tipi', 'value': 'musteri_tipi'},
                    {'label': 'Çadır Tipi', 'value': 'cadir_tipi'},
                    {'label': 'Sezon', 'value': 'sezon'}
                ],
                value='yok',
                style={'width': '250px'}
            )
        ], style={'display': 'inline-block'})
    ], style={'marginBottom': 30, 'textAlign': 'center'}),
    
    dcc.Graph(id='timeseries-grafik-7'),
    
    # Trend yorumu
    html.Div(id='trend-yorum-7', 
             style={'marginTop': 20, 'padding': 20, 'backgroundColor': '#EBF5FB', 
                    'borderRadius': 10, 'fontSize': 14, 'borderLeft': '5px solid #2980B9'})
])

@app7.callback(
    [Output('timeseries-grafik-7', 'figure'),
     Output('trend-yorum-7', 'children')],
    [Input('periyod-radio-7', 'value'),
     Input('metrik-dropdown-7', 'value'),
     Input('segment-dropdown-7', 'value')]
)
def update_timeseries(periyod, metrik, segment):
    # Periyoda göre gruplama
    if periyod == 'gunluk':
        group_col = 'tarih'
        df_sorted = df.groupby('tarih')
    else:
        group_col = 'yil_ay'
        df_sorted = df.groupby('yil_ay')
    
    # Metriğe göre agregasyon
    if metrik == 'count':
        if segment == 'yok':
            ts_data = df_sorted.size().reset_index(name='deger')
        else:
            ts_data = df.groupby([group_col, segment]).size().reset_index(name='deger')
        y_label = "Rezervasyon Sayısı"
    else:
        if segment == 'yok':
            ts_data = df_sorted[metrik].mean().reset_index(name='deger')
        else:
            ts_data = df.groupby([group_col, segment])[metrik].mean().reset_index(name='deger')
        y_label = f"Ortalama {metrik.replace('_', ' ').title()}"
    
    # Grafik oluştur
    if segment == 'yok':
        fig = px.line(
            ts_data,
            x=group_col,
            y='deger',
            markers=True,
            title=f"{periyod.title()} {y_label}",
            line_shape='spline'  # Yumuşak eğri
        )
        fig.update_traces(line=dict(color='#2980B9', width=3), marker=dict(size=8))
    else:
        fig = px.line(
            ts_data,
            x=group_col,
            y='deger',
            color=segment,
            markers=True,
            title=f"{periyod.title()} {y_label} - {segment.replace('_', ' ').title()} Bazlı",
            line_shape='spline'
        )
    
    fig.update_layout(
        xaxis_title="Tarih",
        yaxis_title=y_label,
        hovermode='x unified',
        template='plotly_white',
        height=550,
        font=dict(size=12)
    )
    
    # Trend yorumu oluştur
    if segment == 'yok':
        max_idx = ts_data['deger'].idxmax()
        min_idx = ts_data['deger'].idxmin()
        max_date = ts_data.loc[max_idx, group_col]
        min_date = ts_data.loc[min_idx, group_col]
        max_val = ts_data.loc[max_idx, 'deger']
        min_val = ts_data.loc[min_idx, 'deger']
        
        yorum = html.Div([
            html.H4("📊 Otomatik Trend Analizi", style={'color': '#2980B9', 'marginBottom': 10}),
            html.P([
                html.B("🔴 En Yüksek Nokta: ", style={'color': '#E74C3C'}),
                f"{max_date} tarihinde {max_val:.2f}"
            ], style={'marginBottom': 5}),
            html.P([
                html.B("🔵 En Düşük Nokta: ", style={'color': '#3498DB'}),
                f"{min_date} tarihinde {min_val:.2f}"
            ], style={'marginBottom': 5}),
            html.P([
                html.B("📈 Varyasyon: ", style={'color': '#27AE60'}),
                f"{((max_val - min_val) / min_val * 100):.1f}% fark var (darboğaz potansiyeli)"
            ])
        ])
    else:
        yorum = html.Div([
            html.H4("📊 Segment Karşılaştırması", style={'color': '#2980B9'}),
            html.P("Farklı segmentlerin zaman içindeki davranış değişimlerini karşılaştırın.")
        ])
    
    return fig, yorum

print("✅ ADIM 7 Dash uygulaması hazır! (Time Series Analizi)")
````

#### 🔍 Zaman Serisi Analizi Teknikleri

**1. Tarih Veri Manipülasyonu**
```python
df['tarih'] = pd.to_datetime(df['tarih_saat']).dt.date
df['yil_ay'] = pd.to_datetime(df['tarih_saat']).dt.to_period('M').astype(str)
```
- `pd.to_datetime()`: String'i tarih objesine çevirir
- `.dt.date`: Sadece tarih kısmını alır (saat olmadan)
- `.dt.to_period('M')`: Aylık periyod objesi oluşturur

**2. Dinamik Gruplama**
```python
if periyod == 'gunluk':
    group_col = 'tarih'
else:
    group_col = 'yil_ay'
```
- Kullanıcı seçimine göre farklı gruplama seviyesi

**3. line_shape='spline'**
- Noktalar arası **yumuşak eğri** çizer
- Trend takibi için daha okunaklı
- Alternatif: `'linear'` (düz çizgi)

**4. hovermode='x unified'**
- Mouse aynı X değerinde olduğunda **tüm serileri** gösterir
- Segment karşılaştırması için ideal

**5. Otomatik Trend Analizi**
```python
max_idx = ts_data['deger'].idxmax()
max_val = ts_data.loc[max_idx, 'deger']
```
- `.idxmax()`: En yüksek değerin index'i
- Otomatik insight üretimi

````python
# ADIM 8: Bar Charts
app8 = JupyterDash(__name__)

app8.layout = html.Div([
    html.H1("📊 Kategori Karşılaştırması - Bar Charts", 
            style={'textAlign': 'center', 'color': '#D35400', 'marginBottom': 30}),
    
    html.Div([
        html.Div([
            html.Label("Bar Modu:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.RadioItems(
                id='bar-mode-8',
                options=[
                    {'label': ' Grouped (Yan Yana)', 'value': 'group'},
                    {'label': ' Stacked (İstiflenmiş)', 'value': 'stack'}
                ],
                value='group',
                inline=True
            )
        ], style={'display': 'inline-block', 'marginRight': 40}),
        
        html.Div([
            html.Label("X Ekseni:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='x-dropdown-8',
                options=[
                    {'label': 'Müşteri Tipi', 'value': 'musteri_tipi'},
                    {'label': 'Çadır Tipi', 'value': 'cadir_tipi'},
                    {'label': 'Sezon', 'value': 'sezon'},
                    {'label': 'Ödeme Yöntemi', 'value': 'odeme_yontemi'}
                ],
                value='cadir_tipi',
                style={'width': '220px'}
            )
        ], style={'display': 'inline-block', 'marginRight': 40}),
        
        html.Div([
            html.Label("Renk (Grup):", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='color-dropdown-8',
                options=[
                    {'label': 'Yok', 'value': 'yok'},
                    {'label': 'Müşteri Tipi', 'value': 'musteri_tipi'},
                    {'label': 'Sezon', 'value': 'sezon'}
                ],
                value='musteri_tipi',
                style={'width': '200px'}
            )
        ], style={'display': 'inline-block'})
    ], style={'marginBottom': 30, 'textAlign': 'center'}),
    
    dcc.Graph(id='bar-grafik-8')
])

@app8.callback(
    Output('bar-grafik-8', 'figure'),
    [Input('bar-mode-8', 'value'),
     Input('x-dropdown-8', 'value'),
     Input('color-dropdown-8', 'value')]
)
def update_bar(bar_mode, x_col, color_col):
    # Veriyi grupla
    if color_col == 'yok':
        grouped = df.groupby(x_col).agg({
            'toplam_tutar': 'sum',
            'rezervasyon_id': 'count'
        }).reset_index()
        grouped.columns = [x_col, 'Toplam Gelir', 'Rezervasyon Sayısı']
        
        fig = px.bar(
            grouped,
            x=x_col,
            y='Toplam Gelir',
            text='Rezervasyon Sayısı',
            title=f"{x_col.replace('_', ' ').title()} Bazlı Gelir Analizi"
        )
    else:
        grouped = df.groupby([x_col, color_col]).agg({
            'toplam_tutar': 'sum'
        }).reset_index()
        
        fig = px.bar(
            grouped,
            x=x_col,
            y='toplam_tutar',
            color=color_col,
            barmode=bar_mode,
            title=f"{x_col.replace('_', ' ').title()} vs {color_col.replace('_', ' ').title()}"
        )
    
    fig.update_layout(
        xaxis_title=x_col.replace('_', ' ').title(),
        yaxis_title="Toplam Gelir (TL)",
        template='plotly_white',
        height=550,
        font=dict(size=12)
    )
    
    fig.update_traces(texttemplate='%{text}', textposition='outside')
    
    return fig

print("✅ ADIM 8 Dash uygulaması hazır! (Bar Charts)")
````

````python
# ADIM 9: Heatmap
app9 = JupyterDash(__name__)

app9.layout = html.Div([
    html.H1("🔥 Heatmap - Korelasyon ve Pivot Analizi", 
            style={'textAlign': 'center', 'color': '#C0392B', 'marginBottom': 30}),
    
    html.Div([
        html.Label("Heatmap Tipi:", style={'fontWeight': 'bold', 'marginRight': 10}),
        dcc.RadioItems(
            id='heatmap-tip-9',
            options=[
                {'label': ' Korelasyon Matrisi', 'value': 'korelasyon'},
                {'label': ' Pivot Tablo', 'value': 'pivot'}
            ],
            value='korelasyon',
            inline=True
        )
    ], style={'marginBottom': 20, 'textAlign': 'center'}),
    
    html.Div(id='pivot-controls-9', children=[
        html.Div([
            html.Label("Satır:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='pivot-row-9',
                options=[
                    {'label': 'Çadır Tipi', 'value': 'cadir_tipi'},
                    {'label': 'Müşteri Tipi', 'value': 'musteri_tipi'},
                    {'label': 'Sezon', 'value': 'sezon'}
                ],
                value='cadir_tipi',
                style={'width': '200px'}
            )
        ], style={'display': 'inline-block', 'marginRight': 30}),
        
        html.Div([
            html.Label("Sütun:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='pivot-col-9',
                options=[
                    {'label': 'Sezon', 'value': 'sezon'},
                    {'label': 'Müşteri Tipi', 'value': 'musteri_tipi'},
                    {'label': 'Ödeme Yöntemi', 'value': 'odeme_yontemi'}
                ],
                value='sezon',
                style={'width': '200px'}
            )
        ], style={'display': 'inline-block', 'marginRight': 30}),
        
        html.Div([
            html.Label("Değer:", style={'fontWeight': 'bold', 'marginRight': 10}),
            dcc.Dropdown(
                id='pivot-value-9',
                options=[
                    {'label': 'Toplam Harcama', 'value': 'toplam_tutar'},
                    {'label': 'Ortalama Memnuniyet', 'value': 'memnuniyet_puan'},
                    {'label': 'Rezervasyon Sayısı', 'value': 'count'}
                ],
                value='toplam_tutar',
                style={'width': '230px'}
            )
        ], style={'display': 'inline-block'})
    ], style={'marginBottom': 30, 'textAlign': 'center'}),
    
    dcc.Graph(id='heatmap-grafik-9')
])

@app9.callback(
    [Output('heatmap-grafik-9', 'figure'),
     Output('pivot-controls-9', 'style')],
    [Input('heatmap-tip-9', 'value'),
     Input('pivot-row-9', 'value'),
     Input('pivot-col-9', 'value'),
     Input('pivot-value-9', 'value')]
)
def update_heatmap(tip, row, col, value):
    if tip == 'korelasyon':
        # Korelasyon matrisi
        numeric_cols = ['toplam_tutar', 'ek_harcama', 'konaklama_gece', 
                        'memnuniyet_puan', 'checkin_sure_dk', 'onceki_ziyaret']
        corr_matrix = df[numeric_cols].corr()
        
        fig = px.imshow(
            corr_matrix,
            text_auto='.2f',
            aspect='auto',
            color_continuous_scale='RdBu_r',
            zmin=-1, zmax=1,
            title="Sayısal Değişkenler Korelasyon Matrisi"
        )
        
        pivot_style = {'display': 'none'}
    else:
        # Pivot tablo
        if value == 'count':
            pivot = df.groupby([row, col]).size().reset_index(name='deger')
            pivot_table = pivot.pivot(index=row, columns=col, values='deger').fillna(0)
            title_text = f"Rezervasyon Sayısı: {row.replace('_', ' ').title()} vs {col.replace('_', ' ').title()}"
        else:
            pivot = df.groupby([row, col])[value].sum().reset_index()
            pivot_table = pivot.pivot(index=row, columns=col, values=value).fillna(0)
            title_text = f"{value.replace('_', ' ').title()}: {row.replace('_', ' ').title()} vs {col.replace('_', ' ').title()}"
        
        fig = px.imshow(
            pivot_table,
            text_auto='.0f',
            aspect='auto',
            color_continuous_scale='YlOrRd',
            title=title_text
        )
        
        pivot_style = {'marginBottom': 30, 'textAlign': 'center'}
    
    fig.update_layout(
        template='plotly_white',
        height=550,
        font=dict(size=11)
    )
    
    return fig, pivot_style

print("✅ ADIM 9 Dash uygulaması hazır! (Heatmap)")
````
