### 🧮 2.1 · Matematik Ne Diyor? Çoğunluk Oyunun Doğruluğu

Her biri **yalnızca %60 doğru** olan ve birbirinden **bağımsız** hata yapan $n$ modelimiz olsun. Çoğunluk oyunun doğru çıkma olasılığı **binom dağılımıyla** hesaplanır:

$$P(\text{topluluk doğru}) = \sum_{k=\lfloor n/2 \rfloor + 1}^{n} \binom{n}{k}\, p^k\, (1-p)^{n-k}$$

Hem formülü hem de **"modeller birbirine benzedikçe ne olur?"** sorusunu simülasyonla görelim 👇

````python
def topluluk_dogrulugu(n_model, p):
    """p doğrulukta, bağımsız n_model modelin çoğunluk oyuyla doğru karar verme olasılığı."""
    esik = n_model // 2 + 1
    return sum(comb(n_model, k) * p**k * (1 - p) ** (n_model - k) for k in range(esik, n_model + 1))


print(f"1 model   (%60 doğru)  → %{topluluk_dogrulugu(1, .6) * 100:.1f}")
print(f"7 model   (%60 doğru)  → %{topluluk_dogrulugu(7, .6) * 100:.1f}")
print(f"25 model  (%60 doğru)  → %{topluluk_dogrulugu(25, .6) * 100:.1f}")
print(f"101 model (%60 doğru)  → %{topluluk_dogrulugu(101, .6) * 100:.1f}")

# ── Modeller birbirini kopyalarsa? ──────────────────────────
rng = np.random.default_rng(RANDOM_STATE)


def korelasyonlu_simulasyon(k_model, p, rho, n_ornek=20_000):
    """rho: her modelin ortak bir 'ana görüşü' kopyalama olasılığı (0 = tam bağımsız, 1 = hepsi aynı)."""
    ortak = rng.random(n_ornek) < p                         # ortak görüş doğru mu?
    bagimsiz = rng.random((n_ornek, k_model)) < p           # her modelin kendi görüşü
    kopya = rng.random((n_ornek, k_model)) < rho            # kim ortak görüşü kopyaladı?
    dogru = np.where(kopya, ortak[:, None], bagimsiz)
    return (dogru.sum(axis=1) > k_model / 2).mean()


rholar = np.linspace(0, 1, 11)
kopya_dogruluk = [korelasyonlu_simulasyon(25, 0.60, r) for r in rholar]

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
n_dizisi = np.arange(1, 102, 2)
for p, renk in [(0.55, "#B0B7C3"), (0.60, "#6C5CE7"), (0.70, "#00798C")]:
    axes[0].plot(n_dizisi, [topluluk_dogrulugu(n, p) for n in n_dizisi], lw=2.8, color=renk, label=f"Tek model doğruluğu = %{int(p*100)}")
    axes[0].axhline(p, color=renk, ls=":", lw=1)
axes[0].yaxis.set_major_formatter(PercentFormatter(1))
axes[0].set(xlabel="Topluluktaki model sayısı (n)", ylabel="Topluluk doğruluğu",
            title="Bağımsız modeller: sayı arttıkça doğruluk artar")
axes[0].legend(frameon=False, loc="lower right")

axes[1].plot(rholar, kopya_dogruluk, "o-", lw=2.8, color="#D1495B")
axes[1].axhline(0.60, color=VURGU, ls=":", lw=1.2)
axes[1].annotate("Tek model = %60", xy=(0.95, 0.605), xytext=(0.62, 0.66), arrowprops=dict(arrowstyle="->", color=VURGU))
axes[1].yaxis.set_major_formatter(PercentFormatter(1))
axes[1].set(xlabel="Modellerin birbirine benzerliği (0 = tamamen farklı, 1 = birebir aynı)", ylabel="25 modelli topluluğun doğruluğu",
            title="Modeller birbirini kopyaladıkça kazanç kaybolur")
fig.tight_layout()
kaydet(fig, "01_kalabalik_bilgeligi")
plt.show()
````

### 🪜 Bagging'in Temel Adımları

| Adım | Ne yapılır? |
|:-:|:--|
| 1️⃣ | Tüm veri setinden **bootstrap örneklemesiyle** (iadeli çekme) alt veri setleri oluşturulur |
| 2️⃣ | Her alt veri seti için **bir model** eğitilir (karar ağacı, lojistik regresyon vb.) |
| 3️⃣ | Tahmin sırasında **tüm modellerin sonuçları birleştirilir** |

⚡ **Paralellik:** Modeller birbirinden bağımsız olduğu için **aynı anda** eğitilebilir; bu, eğitim sürecini hızlandırır. (`n_jobs=-1`)

### 🎲 3.1 · Bootstrap Nedir? Elle Görelim

**Bootstrap**, $n$ satırlık veriden **iadeli** olarak yine $n$ satır çekmektir. İade olduğu için bazı satırlar **birden çok kez** seçilir, bazıları **hiç seçilmez**. Hiç seçilmeyen satırlara **OOB (out-of-bag)** denir; bu satırlar o modelin "görmediği" veridir ve ileride bedava bir doğrulama seti olacak.

10 satırlık küçük bir veriden 3 ağaç için bootstrap örneklemi çekelim: tablodaki sayı, satırın o ağacın örnekleminde **kaç kez** yer aldığını gösterir.

````python
rng = np.random.default_rng(7)
n = 10
satir_no = np.arange(1, n + 1)
ornekler = {f"Ağaç {i + 1} · D{'′' * (i + 1)}": rng.choice(satir_no, size=n, replace=True) for i in range(3)}

bootstrap_tablosu = pd.DataFrame(
    {ad: [(idx == s).sum() for s in satir_no] for ad, idx in ornekler.items()},
    index=[f"satır {s}" for s in satir_no])


def renklendir(v):
    if v == 0:
        return "background-color:#FDE2E4;color:#C1121F;font-weight:bold"       # OOB: hiç seçilmedi
    if v == 1:
        return "background-color:#D8F3DC"                                      # bir kez seçildi
    return "background-color:#FFD6A5;font-weight:bold"                         # birden çok kez seçildi


display(bootstrap_tablosu.style.map(renklendir))
oob_oran = (bootstrap_tablosu == 0).mean().round(2).to_dict()
print("OOB (hiç seçilmeyen) satır oranı:", oob_oran)
````

🟥 **Kırmızı = OOB** (o ağaç bu satırı hiç görmedi) · 🟩 **Yeşil** = bir kez · 🟧 **Turuncu** = birden çok kez seçilmiş.

Peki büyük veride **ortalama kaç satır** her ağacın örnekleminde yer alır? Matematik net bir cevap veriyor:

$$P(\text{bir satır hiç seçilmez}) = \left(1 - \frac{1}{n}\right)^{n} \xrightarrow{\;n \to \infty\;} \frac{1}{e} \approx 0{,}368$$

````python
def benzersiz_oran(n, tekrar=200):
    """n satırlık veriden bootstrap çekildiğinde örneklemde yer alan benzersiz satır oranı."""
    return np.mean([len(np.unique(rng.integers(0, n, n))) / n for _ in range(tekrar)])


boyutlar = [5, 10, 20, 50, 100, 500, 1_000, 5_000, 20_000]
simulasyon = [benzersiz_oran(n) for n in boyutlar]
kuramsal = [1 - (1 - 1 / n) ** n for n in boyutlar]

fig, ax = plt.subplots(figsize=(10, 4.6))
ax.plot(boyutlar, kuramsal, "-", lw=3, color="#6C5CE7", label="Kuramsal: 1 − (1 − 1/n)ⁿ")
ax.plot(boyutlar, simulasyon, "o", ms=9, color="#00798C", label="Simülasyon (200 tekrar)")
ax.axhline(1 - 1 / np.e, color="#D1495B", ls="--", lw=1.6)
ax.annotate("Sınır: 1 − 1/e ≈ %63,2\n(OOB ≈ %36,8)", xy=(2_000, 0.632), xytext=(150, 0.80), arrowprops=dict(arrowstyle="->", color="#D1495B"))
ax.set_xscale("log")
ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.set(xlabel="Veri setindeki satır sayısı (n, log ölçek)", ylabel="Örneklemde yer alan benzersiz satır oranı",
       title="Bootstrap: her ağaç verinin yaklaşık üçte ikisini görür")
ax.legend(frameon=False)
kaydet(fig, "02_bootstrap_orani")
plt.show()
````

### 🌓 3.2 · Bagging'i Sıfırdan Kuralım

Şimdi PDF'deki 3 adımı kodlayalım: **(1)** bootstrap örnekleri çek, **(2)** her biri için bir karar ağacı eğit, **(3)** tahminleri çoğunluk oyuyla birleştir. Görselleştirme için ikiboyutlu, gürültülü bir "iki ay" (`make_moons`) veri seti kullanıyoruz.

````python
Xm, ym = make_moons(n_samples=600, noise=0.35, random_state=RANDOM_STATE)
Xm_egitim, Xm_test, ym_egitim, ym_test = train_test_split(Xm, ym, test_size=0.3, stratify=ym, random_state=RANDOM_STATE)

N_AGAC = 25
rng = np.random.default_rng(RANDOM_STATE)
agaclar = []
for i in range(N_AGAC):
    idx = rng.integers(0, len(Xm_egitim), len(Xm_egitim))                      # 1) bootstrap örneklemi
    agaclar.append(DecisionTreeClassifier(random_state=i).fit(Xm_egitim[idx], ym_egitim[idx]))  # 2) model eğit

oylar = np.array([a.predict(Xm_test) for a in agaclar])                        # (25 ağaç × test örnekleri)
bagging_tahmin = (oylar.mean(axis=0) > 0.5).astype(int)                        # 3) çoğunluk oyu

tek_agac = DecisionTreeClassifier(random_state=RANDOM_STATE).fit(Xm_egitim, ym_egitim)
bireysel_dogruluk = np.array([accuracy_score(ym_test, o) for o in oylar])
sk_bagging = BaggingClassifier(DecisionTreeClassifier(), n_estimators=N_AGAC, random_state=RANDOM_STATE).fit(Xm_egitim, ym_egitim)

print(f"Tek ağaç (tüm veri)                    : {tek_agac.score(Xm_test, ym_test):.3f}")
print(f"25 bootstrap ağacının ortalama doğruluğu: {bireysel_dogruluk.mean():.3f}  (en kötü {bireysel_dogruluk.min():.3f} · en iyi {bireysel_dogruluk.max():.3f})")
print(f"Elle Bagging (25 ağaç, çoğunluk oyu)   : {accuracy_score(ym_test, bagging_tahmin):.3f}")
print(f"sklearn BaggingClassifier (25 ağaç)    : {sk_bagging.score(Xm_test, ym_test):.3f}")
````

### 🪜 Boosting'in Temel Adımları

| Adım | Ne yapılır? |
|:-:|:--|
| 1️⃣ | İlk zayıf model **tüm veri seti** üzerinde eğitilir |
| 2️⃣ | Sonraki modeller, önceki modelin **yanlış tahmin ettiği** veri noktalarına **ağırlık verir** |
| 3️⃣ | Süreç ardışık devam eder; en sonda tüm çıktılar birleştirilerek **güçlü bir model** oluşur |

Diyagramdaki $\varepsilon_i$ modelin **hata oranı**, $\alpha_i$ ise modelin **söz hakkıdır**. Birleştirme formülü:

$$H(x) = \operatorname{sign}\!\Big(\alpha_1 h_1(x) + \alpha_2 h_2(x) + \alpha_3 h_3(x)\Big), \qquad \alpha_t = \tfrac12 \ln\frac{1-\varepsilon_t}{\varepsilon_t}$$

Hata oranı küçük olan model **büyük $\alpha$** alır: iyi uzmanın oyu daha çok sayılır. **Örnek algoritmalar:** AdaBoost · Gradient Boosting · XGBoost.

### 🔬 4.1 · AdaBoost'u Elle Kuralım (3 Tur)

Zayıf öğrenici olarak **tek soruluk karar kütüğü** (*decision stump*, `max_depth=1`) kullanıyoruz. Her turdan sonra yanlış bilinen noktaların ağırlığı $e^{\alpha}$ kadar büyür, doğru bilinenler küçülür. Grafikte **nokta büyüklüğü = örnek ağırlığı**.

````python
Xb, yb = make_moons(n_samples=120, noise=0.28, random_state=3)
yb_pm = np.where(yb == 1, 1, -1)                                  # AdaBoost etiketleri: -1 / +1


def adaboost_elle(X, y_pm, tur):
    """Ayrık AdaBoost: her turun ağırlıklarını, hata oranını (ε) ve model ağırlığını (α) kaydeder."""
    w = np.full(len(y_pm), 1 / len(y_pm))
    kayit = []
    for _ in range(tur):
        kutuk = DecisionTreeClassifier(max_depth=1, random_state=RANDOM_STATE).fit(X, y_pm, sample_weight=w)
        h = kutuk.predict(X)
        eps = w[h != y_pm].sum()
        alfa = 0.5 * np.log((1 - eps) / eps)
        kayit.append({"model": kutuk, "w": w.copy(), "h": h, "eps": eps, "alfa": alfa})
        w = w * np.exp(-alfa * y_pm * h)                           # yanlışların ağırlığı ↑, doğruların ↓
        w /= w.sum()
    return kayit


tur_kaydi = adaboost_elle(Xb, yb_pm, tur=3)

fig, axes = plt.subplots(1, 4, figsize=(20, 4.8))
for t, (ax, k) in enumerate(zip(axes[:3], tur_kaydi), start=1):
    xx, yy = np.meshgrid(np.arange(-1.6, 2.7, 0.03), np.arange(-1.1, 1.7, 0.03))
    zz = (k["model"].predict(np.c_[xx.ravel(), yy.ravel()]) == 1).astype(int).reshape(xx.shape)
    ax.contourf(xx, yy, zz, levels=[-0.5, 0.5, 1.5], colors=[SINIF_RENK[0], SINIF_RENK[1]], alpha=0.25)
    yanlis = k["h"] != yb_pm
    ax.scatter(Xb[:, 0], Xb[:, 1], s=k["w"] * len(yb) * 55, c=[SINIF_RENK[int(s)] for s in yb], edgecolor="white", linewidth=0.8)
    ax.scatter(Xb[yanlis, 0], Xb[yanlis, 1], s=k["w"][yanlis] * len(yb) * 55 + 70, facecolor="none", edgecolor="#D1495B", linewidth=2)
    ax.set(title=f"Tur {t}:  ε = {k['eps']:.2f}  ·  α = {k['alfa']:.2f}", xticks=[], yticks=[])
    ax.spines[["left", "bottom"]].set_visible(False)

skor = lambda Z: sum(k["alfa"] * k["model"].predict(Z) for k in tur_kaydi)
birlesik_dogruluk = accuracy_score(yb_pm, np.sign(skor(Xb)))
sinir_ciz(axes[3], lambda Z: (skor(Z) > 0).astype(int), Xb, yb, r"Birleşik model $\mathrm{sign}(\sum \alpha_t h_t)$" + f"\ndoğruluk {birlesik_dogruluk:.2f}")
fig.suptitle("AdaBoost'un ilk 3 turu: kırmızı halka = o turda yanlış bilinen nokta, nokta büyüklüğü = ağırlık", fontsize=15, fontweight="bold", y=1.04)
kaydet(fig, "04_adaboost_turlari")
plt.show()

pd.DataFrame([{"tur": t, "hata oranı ε": round(k["eps"], 3), "model ağırlığı α": round(k["alfa"], 3),
               "yanlış bilinen nokta": int((k["h"] != yb_pm).sum()), "en büyük örnek ağırlığı": round(k["w"].max(), 4)}
              for t, k in enumerate(tur_kaydi, 1)]).set_index("tur")
````

### 🎞️ 4.2 · Boosting İş Başında (Animasyon)

AdaBoost'un **30 tur** boyunca sınırı nasıl şekillendirdiğini ve **nokta ağırlıklarının** nasıl kaydığını izleyelim. (Zayıf öğrenici: derinliği 2 olan ağaçlar.)

````python
def gif_kaydet(kare_ciz, n_kare, yol, figsize=(5.6, 4.6), fps=3, dpi=80):
    """kare_ciz(ax, i) fonksiyonuyla her kareyi çizip animasyonlu GIF olarak kaydeder."""
    fig, ax = plt.subplots(figsize=figsize)
    anim = FuncAnimation(fig, lambda i: kare_ciz(ax, i), frames=n_kare)
    anim.save(yol, writer=PillowWriter(fps=fps), dpi=dpi)
    plt.close(fig)
    print(f"💾 Animasyon → {Path(yol).as_posix()} ({Path(yol).stat().st_size / 1024:.0f} KB)")


Xg, yg = make_moons(n_samples=300, noise=0.30, random_state=11)
yg_pm = np.where(yg == 1, 1, -1)
w = np.full(len(yg), 1 / len(yg))
ada_turlar = []
for _ in range(30):
    m = DecisionTreeClassifier(max_depth=2, random_state=RANDOM_STATE).fit(Xg, yg_pm, sample_weight=w)
    h = m.predict(Xg)
    eps = np.clip(w[h != yg_pm].sum(), 1e-10, 1 - 1e-10)
    alfa = 0.5 * np.log((1 - eps) / eps)
    ada_turlar.append({"model": m, "alfa": alfa, "w": w.copy()})
    w = w * np.exp(-alfa * yg_pm * h)
    w /= w.sum()

izgara_x, izgara_y = np.meshgrid(np.arange(-1.8, 2.9, 0.04), np.arange(-1.3, 1.9, 0.04))
izgara = np.c_[izgara_x.ravel(), izgara_y.ravel()]
tur_skorlari = np.cumsum([k["alfa"] * k["model"].predict(izgara) for k in ada_turlar], axis=0)   # (tur, nokta)
egitim_skorlari = np.cumsum([k["alfa"] * k["model"].predict(Xg) for k in ada_turlar], axis=0)


def boosting_karesi(ax, i):
    ax.clear()
    ax.contourf(izgara_x, izgara_y, (tur_skorlari[i] > 0).reshape(izgara_x.shape), levels=[-0.5, 0.5, 1.5],
                colors=[SINIF_RENK[0], SINIF_RENK[1]], alpha=0.28)
    ax.scatter(Xg[:, 0], Xg[:, 1], s=ada_turlar[i]["w"] * len(yg) * 22, c=[SINIF_RENK[int(s)] for s in yg],
               edgecolor="white", linewidth=0.5)
    dogru = accuracy_score(yg_pm, np.sign(egitim_skorlari[i]))
    ax.set_title(f"AdaBoost · {i + 1}. tur · eğitim doğruluğu {dogru:.2f}", fontsize=11)
    ax.set(xticks=[], yticks=[])
    ax.spines[["left", "bottom"]].set_visible(False)


BOOST_GIF = FIG_DIR / "05_boosting_animasyonu.gif"
gif_kaydet(boosting_karesi, 30, BOOST_GIF, fps=3)
Image(filename=BOOST_GIF, width=520)
````

### ⚔️ 4.3 · Bagging mi, Boosting mi?

| | 🎒 Bagging | 🚀 Boosting |
|:--|:--|:--|
| **Eğitim şekli** | **Paralel** — modeller birbirinden bağımsız | **Ardışık** — her model öncekini bekler |
| **Hedef** | **Varyansı** azaltmak | **Bias'ı** azaltmak |
| **Öğrenici türü** | Genellikle **güçlü** öğreniciler (derin ağaçlar) | **Zayıf** öğreniciler (sığ ağaçlar) |
| **Veri kullanımı** | Her model farklı bir **bootstrap örnekleminde** | Her model **ağırlıkları güncellenmiş** verinin tamamında |
| **Birleştirme** | Eşit ağırlıklı oy / ortalama | **Ağırlıklı** oy ($\alpha_t$) |
| **Aşırı öğrenme riski** | Düşük | Daha yüksek (gürültülü veride dikkat) |
| **Örnekler** | **Random Forest** | AdaBoost · Gradient Boosting · XGBoost |

**Random Forest**, **bagging** yaklaşımını temel alan ve birçok **karar ağacının** birleşiminden oluşan güçlü bir topluluk yöntemidir. Hem **sınıflandırma** hem **regresyon** için kullanılır. Bagging'deki fikir aynıdır; Random Forest bu prensibi uygularken **tüm modellerini karar ağacı** olarak seçer ve bir **rastgelelik** daha ekler: **özellik örneklemesi**.

### 🌲 Random Forest'ın Yapısı

| Bileşen | Açıklama |
|:--|:--|
| 📦 **Veri seti** | $D$ boyutunda toplam veri, $m$ tane özellik ($f_1, f_2, \dots, f_m$) |
| 🌳 **Base learners** | Tüm modeller **karar ağacıdır**: $DT_1, DT_2, DT_3, \dots$ |
| 🧾 **Row sampling** | $D$'den rastgele satırlar seçilir ($D' < D$), **iadeli** (bootstrap) |
| 🎲 **Feature sampling** | Rastgele özellikler seçilir ($F' \subset F$) |

Her karar ağacına **farklı veri ve özellik kombinasyonları** verilir. Replacement (iade) sayesinde bazı satırlar kesişse bile ağaçlara verilen verilerin hepsi aynı olmaz; her ağaç **farklı örneklemleri** inceler.

### 🗳️ Tahmin Aşaması

- **Sınıflandırma:** Her ağaç kendi tahminini yapar. Nihai karar **çoğunluk oylamasıyla** belirlenir: $\;\hat{y} = \operatorname{mode}(\hat y_1, \hat y_2, \dots, \hat y_n)$
- **Regresyon:** Tahminler sürekli sayı olduğundan **ortalama** alınır: $\;\hat{y} = \frac1n \sum_{i=1}^{n} \hat y_i$

### ❓ Neden Random Forest?

Varsayılan parametrelerle kullanılan **tek bir karar ağacı** aşırı öğrenme riski taşır: eğitim başarımı çok yüksektir (*low bias*), test başarımı düşer (*high variance*). Random Forest, çok sayıda ağacın oyunu ortalayarak **varyansı düşürür** ve **daha genelleştirilebilir** sonuçlar verir.

**PDF ile `sklearn` arasındaki küçük fark:** Notlarda özellik örneklemesi **ağaç başına** anlatılır (şemadaki gibi). `scikit-learn` ise özellik alt kümesini **her bölünmede (split) yeniden** seçer (`max_features`). Sonuç aynıdır: ağaçlar birbirinden daha farklı olur. Ayrıca `sklearn` nihai kararı "sert oy" yerine **ağaç olasılıklarının ortalamasıyla** verir. Hemen doğrulayalım 👇

````python
orman = RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE).fit(Xm_egitim, ym_egitim)

# 1) Ormanın olasılığı = ağaçların olasılıklarının ortalaması mı?
olasilik_ortalamasi = np.mean([a.predict_proba(Xm_test) for a in orman.estimators_], axis=0)
print("predict_proba == ağaç olasılıklarının ortalaması:", np.allclose(olasilik_ortalamasi, orman.predict_proba(Xm_test)))

# 2) PDF'teki "sert oy" (mode) ile sklearn'ün kararı ne kadar örtüşüyor?
sert_oy = (np.mean([a.predict(Xm_test) for a in orman.estimators_], axis=0) > 0.5).astype(int)
print(f"Çoğunluk oyu (mode) ile orman.predict() uyumu  : %{(sert_oy == orman.predict(Xm_test)).mean() * 100:.1f}")

# 3) Ağaçlar gerçekten birbirinden farklı mı?
agac_dogruluklari = np.array([a.score(Xm_test, ym_test) for a in orman.estimators_])
print(f"Tek ağaçların test doğruluğu: ort. {agac_dogruluklari.mean():.3f} (min {agac_dogruluklari.min():.3f} · max {agac_dogruluklari.max():.3f})")
print(f"Orman (100 ağaç) test doğruluğu: {orman.score(Xm_test, ym_test):.3f}")
````

### 📉 5.1 · Varyans Gerçekten Düşüyor mu?

Eğitim verisinden **30 kez farklı %60'lık alt örnek** çekip her seferinde hem **tek bir ağacı** hem de **50 ağaçlık bir ormanı** eğitelim; test doğruluklarının dağılımına bakalım. *Varyans = veri biraz değişince sonucun ne kadar oynadığı.*

````python
rng = np.random.default_rng(RANDOM_STATE)
tek_sonuc, orman_sonuc = [], []
for tekrar in range(30):
    idx = rng.choice(len(Xm_egitim), size=int(0.6 * len(Xm_egitim)), replace=False)
    tek_sonuc.append(DecisionTreeClassifier(random_state=tekrar).fit(Xm_egitim[idx], ym_egitim[idx]).score(Xm_test, ym_test))
    orman_sonuc.append(RandomForestClassifier(n_estimators=50, random_state=tekrar, n_jobs=-1)
                       .fit(Xm_egitim[idx], ym_egitim[idx]).score(Xm_test, ym_test))

varyans_tablosu = pd.DataFrame({"Tek karar ağacı": tek_sonuc, "Random Forest (50 ağaç)": orman_sonuc})

fig, ax = plt.subplots(figsize=(9, 4.8))
sns.boxplot(data=varyans_tablosu, palette=[RENK["agac"], RENK["orman"]], width=0.5, ax=ax)
sns.stripplot(data=varyans_tablosu, color=VURGU, alpha=0.55, size=5, ax=ax)
ax.set(ylabel="Test doğruluğu", title="Farklı eğitim örneklerinde sonuç ne kadar oynuyor?")
kaydet(fig, "06_varyans_karsilastirmasi")
plt.show()

display(varyans_tablosu.agg(["mean", "std", "min", "max"]).round(3).rename(index={"mean": "ortalama", "std": "std. sapma"}))
````

### 🎞️ 5.2 · Orman Büyüyor (Animasyon)

Ormana **ağaç ekledikçe** karar sınırı nasıl pürüzsüzleşiyor ve test doğruluğu nasıl oturuyor? Soldaki ilk kare tek ağaç, sonraki her kare bir ağaç daha ekler (toplam 40).

````python
GIF_AGAC = 40
orman_gif = RandomForestClassifier(n_estimators=GIF_AGAC, random_state=RANDOM_STATE).fit(Xm_egitim, ym_egitim)
gx, gy = np.meshgrid(np.arange(-1.8, 2.9, 0.04), np.arange(-1.3, 1.9, 0.04))
grid_noktalari = np.c_[gx.ravel(), gy.ravel()]
grid_olasilik = np.array([a.predict_proba(grid_noktalari)[:, 1] for a in orman_gif.estimators_])          # (ağaç, nokta)
test_olasilik = np.array([a.predict_proba(Xm_test)[:, 1] for a in orman_gif.estimators_])
kumulatif_grid = np.cumsum(grid_olasilik, axis=0) / np.arange(1, GIF_AGAC + 1)[:, None]                   # ilk k ağacın ortalaması
kumulatif_test = np.cumsum(test_olasilik, axis=0) / np.arange(1, GIF_AGAC + 1)[:, None]


def orman_karesi(ax, i):
    ax.clear()
    ax.contourf(gx, gy, (kumulatif_grid[i] > 0.5).reshape(gx.shape), levels=[-0.5, 0.5, 1.5],
                colors=[SINIF_RENK[0], SINIF_RENK[1]], alpha=0.28)
    ax.scatter(Xm_test[:, 0], Xm_test[:, 1], c=[SINIF_RENK[int(s)] for s in ym_test], edgecolor="white", s=28, linewidth=0.6)
    dogruluk = accuracy_score(ym_test, (kumulatif_test[i] > 0.5).astype(int))
    ax.set_title(f"Random Forest · {i + 1} ağaç · test doğruluğu {dogruluk:.3f}", fontsize=11)
    ax.set(xticks=[], yticks=[])
    ax.spines[["left", "bottom"]].set_visible(False)


ORMAN_GIF = FIG_DIR / "07_orman_buyuyor.gif"
gif_kaydet(orman_karesi, GIF_AGAC, ORMAN_GIF, fps=4)
Image(filename=ORMAN_GIF, width=520)
````

### 🎛️ 5.3 · Ormanın Başlıca Ayarları (Hiperparametreler)

### 🔠 9.2 · Hangi Sütuna Ne Yapıyoruz?

| Sütun grubu | Sütunlar | İşlem | Neden? |
|:--|:--|:--|:--|
| 🔢 **Sayısal** | `age`, `fnlwgt`, `education_num`, `capital_gain`, `capital_loss`, `hours_per_week` | Olduğu gibi (*passthrough*) | Ağaçlar **ölçekten bağımsız**; ölçekleme gerekmez |
| 🏷️ **Az kategorili** | `workclass`, `marital_status`, `occupation`, `relationship`, `race`, `sex` | Mod ile doldur → **One-Hot** | Sırasız kategoriler; 2–14 değer, sütun sayısı makul |
| 🗺️ **Çok kategorili** | `native_country` (41 ülke) | Mod ile doldur → **Target Encoding** | One-hot 41 sütun üretirdi; hedef ortalaması tek sütunda toplar |

<table><tr><td>

<img src="assets/etiket_dikkat.svg" height="34" alt="Dikkat">

**Target Encoding'de sızıntı tuzağı:** Referans notebook ülke → ">50K oranı" haritasını **tüm eğitim setinden** hesaplıyor. Bu hesap pipeline dışında yapılırsa, çapraz doğrulamada her katmanın **doğrulama kısmı** haritanın içine girmiş olur (hedef bilgisi sızar).

`sklearn`'ün `TargetEncoder`'ı bunu iki şekilde çözer: **(1)** `Pipeline` içinde yalnızca eğitim kısmına uydurulur, **(2)** `fit_transform` sırasında eğitim satırları **çapraz uydurma (cross-fitting)** ile kodlanır; yani bir satırın kodu **kendi etiketini kullanmaz**. Üstelik `smooth=20` ayarıyla küçük gruplar (ör. 1–10 kişilik ülkeler) genel ortalamaya **doğru çekilir (smoothing)**: her ülke, kendi verisinin yanında 20 "sanal" ortalama gözlemle desteklenir.

</td></tr></table>

````python
KATEGORIK = X.select_dtypes("object").columns.tolist()
SAYISAL = [c for c in X.columns if c not in KATEGORIK]
YUKSEK_KARDINALITE = ["native_country"]
print("Kategorik:", KATEGORIK)
print("Sayısal  :", SAYISAL)


def on_isleme(ozellikler=None):
    """Sızıntısız ön işleme: eksik doldur + one-hot / target encoding + sayısalları aynen geçir."""
    ozellikler = list(X.columns) if ozellikler is None else list(ozellikler)
    dusuk = [c for c in KATEGORIK if c in ozellikler and c not in YUKSEK_KARDINALITE]
    yuksek = [c for c in YUKSEK_KARDINALITE if c in ozellikler]
    sayisal = [c for c in SAYISAL if c in ozellikler]
    adimlar = []
    if dusuk:
        adimlar.append(("az_kategorili", Pipeline([("doldur", SimpleImputer(strategy="most_frequent")),
                                                   ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))]), dusuk))
    if yuksek:
        adimlar.append(("cok_kategorili", Pipeline([("doldur", SimpleImputer(strategy="most_frequent")),
                                                    ("hedef", TargetEncoder(smooth=20.0, random_state=RANDOM_STATE))]), yuksek))
    if sayisal:
        adimlar.append(("sayisal", "passthrough", sayisal))
    return ColumnTransformer(adimlar, verbose_feature_names_out=False).set_output(transform="pandas")


def orman_pipeline(ozellikler=None, olcekle=False, **orman_parametreleri):
    """Ön işleme (+ isteğe bağlı RobustScaler) + RandomForestClassifier tek nesnede."""
    adimlar = [("on_isleme", on_isleme(ozellikler))]
    if olcekle:
        adimlar.append(("olcek", RobustScaler().set_output(transform="pandas")))
    parametreler = {"random_state": RANDOM_STATE, "n_jobs": -1, **orman_parametreleri}
    adimlar.append(("model", RandomForestClassifier(**parametreler)))
    return Pipeline(adimlar)


def agac_pipeline(ozellikler=None, **agac_parametreleri):
    """Karşılaştırma için: aynı ön işleme + tek bir karar ağacı."""
    return Pipeline([("on_isleme", on_isleme(ozellikler)),
                     ("model", DecisionTreeClassifier(random_state=RANDOM_STATE, **agac_parametreleri))])


orman_pipeline(n_estimators=100)
````

````python
SONUCLAR = {}


def degerlendir(ad, model):
    """Modeli eğitir, test setinde ölçer ve SONUCLAR sözlüğüne kaydeder."""
    t0 = time.time()
    model.fit(X_egitim, y_egitim)
    sure = time.time() - t0
    tahmin, olasilik = model.predict(X_test), model.predict_proba(X_test)[:, 1]
    SONUCLAR[ad] = {
        "eğitim doğruluğu": accuracy_score(y_egitim, model.predict(X_egitim)),
        "test doğruluğu": accuracy_score(y_test, tahmin),
        "dengeli doğruluk": balanced_accuracy_score(y_test, tahmin),
        "kesinlik (>50K)": precision_score(y_test, tahmin, zero_division=0),
        "duyarlılık (>50K)": recall_score(y_test, tahmin),
        "F1 (>50K)": f1_score(y_test, tahmin),
        "ROC-AUC": roc_auc_score(y_test, olasilik),
        "PR-AUC": average_precision_score(y_test, olasilik),
        "eğitim süresi (sn)": sure,
    }
    return model


taban = degerlendir("Taban (hep ≤50K)", Pipeline([("model", DummyClassifier(strategy="most_frequent"))]))
tek_agac = degerlendir("Tek karar ağacı", agac_pipeline())
orman10 = degerlendir("Orman · 10 ağaç", orman_pipeline(n_estimators=10))
orman100 = degerlendir("Orman · 100 ağaç", orman_pipeline(n_estimators=100))

karsilastirma = pd.DataFrame(SONUCLAR).T.round(4)
karsilastirma
````

Ormana ağaç eklemek ücretsiz değil: **süre** artar. Peki kaçıncı ağaçtan sonra kazanç anlamsızlaşır? Cevabı bulmanın bir yolu test setine bakmaktır, ama **ucuz ve etik** bir yolu daha var: **OOB (out-of-bag) skoru**.

> Bölüm 3'te gördük: her ağaç satırların ≈ %63'ünü görür, ≈ %37'si **OOB** kalır. Bir satır için yalnızca **onu görmemiş ağaçların** oyunu toplarsak, o satıra ait **dürüst bir tahmin** elde ederiz. Tüm eğitim satırları için bunu yapıp doğruluğu ölçersek **test setine dokunmadan** genelleme hatasını tahmin etmiş oluruz.

`RandomForestClassifier(oob_score=True)` bunu otomatik yapar. Aşağıda `warm_start=True` ile ormanı **adım adım büyütüp** her adımda OOB ve test skoruna bakıyoruz (3 farklı tohumla).

````python
warnings.filterwarnings("ignore", message="Some inputs do not have OOB scores")      # 5 ağaçlı ormanda bazı satırlar hiç OOB kalmaz: beklenen uyarı

onisleme = on_isleme()
Xt_egitim = onisleme.fit_transform(X_egitim, y_egitim)        # eğitim satırları çapraz uydurmayla kodlanır
Xt_test = onisleme.transform(X_test)

AGAC_ADIMLARI = [5, 10, 20, 30, 50, 75, 100, 150, 200, 300, 500]
kayitlar = []
for tohum in (1, 2, 3):
    orman_adim = RandomForestClassifier(n_estimators=1, warm_start=True, oob_score=True, n_jobs=-1, random_state=tohum)
    gecen = 0.0
    for n in AGAC_ADIMLARI:
        t0 = time.time()
        orman_adim.set_params(n_estimators=n).fit(Xt_egitim, y_egitim)
        gecen += time.time() - t0
        kayitlar.append({"tohum": tohum, "ağaç": n, "OOB doğruluğu": orman_adim.oob_score_,
                         "test doğruluğu": orman_adim.score(Xt_test, y_test),
                         "test ROC-AUC": roc_auc_score(y_test, orman_adim.predict_proba(Xt_test)[:, 1]), "toplam süre (sn)": gecen})

agac_egrisi = pd.DataFrame(kayitlar)
agac_egrisi.groupby("ağaç").mean().drop(columns="tohum").round(4).to_csv(RAPOR_DIR / "agac_sayisi_egrisi.csv")
agac_egrisi.groupby("ağaç").mean().drop(columns="tohum").round(4)
````

````python
ort = agac_egrisi.groupby("ağaç").mean()
std = agac_egrisi.groupby("ağaç").std()

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
for sutun, renk, ad in [("OOB doğruluğu", "#6C5CE7", "OOB doğruluğu (eğitim içinden)"), ("test doğruluğu", "#00798C", "Test doğruluğu")]:
    axes[0].plot(ort.index, ort[sutun], "o-", color=renk, lw=2.6, label=ad)
    axes[0].fill_between(ort.index, ort[sutun] - std[sutun], ort[sutun] + std[sutun], color=renk, alpha=0.15)
axes[0].set_xscale("log")
axes[0].set_xticks(AGAC_ADIMLARI, [str(n) for n in AGAC_ADIMLARI])
axes[0].xaxis.set_minor_locator(plt.NullLocator())
axes[0].set(xlabel="Ağaç sayısı (n_estimators, log ölçek)", ylabel="Doğruluk", title="OOB skoru, test skorunu takip ediyor")
axes[0].legend(frameon=False, loc="lower right")

axes[1].plot(ort.index, ort["test ROC-AUC"], "o-", color="#66A182", lw=2.6, label="Test ROC-AUC")
axes[1].set_xscale("log")
axes[1].set_xticks(AGAC_ADIMLARI, [str(n) for n in AGAC_ADIMLARI])
axes[1].xaxis.set_minor_locator(plt.NullLocator())
ikinci = axes[1].twinx()
ikinci.plot(ort.index, ort["toplam süre (sn)"], "s--", color="#D1495B", lw=2, label="Toplam eğitim süresi")
ikinci.set(ylabel="Toplam eğitim süresi (sn)")
ikinci.grid(False)
ikinci.spines["right"].set_visible(True)
axes[1].set(xlabel="Ağaç sayısı (log ölçek)", ylabel="ROC-AUC", title="Kazanç düzleşirken süre artmaya devam ediyor")
h1, l1 = axes[1].get_legend_handles_labels()
h2, l2 = ikinci.get_legend_handles_labels()
axes[1].legend(h1 + h2, l1 + l2, frameon=False, loc="lower right")
fig.tight_layout()
kaydet(fig, "14_agac_sayisi_ve_oob")
plt.show()
````

Modelden bağımsız ve <b>daha güvenilir</b>. Özelliğin gerçek katkısını ölçer.<br><br></td></tr>
</table>

````python
orman_son = orman_pipeline(n_estimators=300).fit(X_egitim, y_egitim)
donusmus_adlar = orman_son["on_isleme"].get_feature_names_out()


def ham_ozellik(ad):
    """One-hot sütun adını (ör. 'marital_status_Divorced') asıl özelliğe geri eşler."""
    for o in sorted(X.columns, key=len, reverse=True):
        if ad == o or ad.startswith(o + "_"):
            return o


mdi = pd.Series(orman_son["model"].feature_importances_, index=donusmus_adlar).groupby(ham_ozellik).sum()
perm = permutation_importance(orman_son, X_test, y_test, scoring="roc_auc", n_repeats=10, random_state=RANDOM_STATE, n_jobs=1)
onem = (pd.DataFrame({"gini_onemi": mdi, "permutasyon_onemi": pd.Series(perm.importances_mean, index=X.columns),
                      "permutasyon_std": pd.Series(perm.importances_std, index=X.columns)})
          .sort_values("permutasyon_onemi", ascending=False))
onem.round(4).to_csv(RAPOR_DIR / "ozellik_onemleri.csv")

fig, axes = plt.subplots(1, 2, figsize=(15, 5.6))
s1 = onem.sort_values("gini_onemi")
b = axes[0].barh([OZELLIK_TR[i] for i in s1.index], s1["gini_onemi"], color="#6C5CE7")
axes[0].bar_label(b, labels=[f"{v:.3f}" for v in s1["gini_onemi"]], padding=4, fontsize=8.5)
axes[0].set(title="Gini önemi (MDI)", xlabel="Toplam safsızlık azalması payı")
s2 = onem.sort_values("permutasyon_onemi")
b = axes[1].barh([OZELLIK_TR[i] for i in s2.index], s2["permutasyon_onemi"], xerr=s2["permutasyon_std"], color="#00798C", capsize=3)
axes[1].bar_label(b, labels=[f"{v:.3f}" for v in s2["permutasyon_onemi"]], padding=14, fontsize=8.5)
axes[1].set(title="Permütasyon önemi (test seti)", xlabel="ROC-AUC'deki ortalama düşüş")
fig.tight_layout()
kaydet(fig, "15_ozellik_onemi")
plt.show()
onem.round(4).rename(index=OZELLIK_TR)
````

### 🧮 13.1 · Karmaşıklık Matrisi: Dört Çeşit Sonuç

| | Model: "yüksek" | Model: "düşük" |
|:--|:-:|:-:|
| **Gerçek: yüksek gelirli** | ✅ **TP** — doğru yakalanan aday | ❌ **FN** — kaçırılan fırsat |
| **Gerçek: düşük gelirli** | ⚠️ **FP** — boşa yapılan teklif | ✅ **TN** — doğru elenen |

````python
def rapor_tablosu(y_gercek, y_tahmin):
    """classification_report çıktısını düzenli bir DataFrame'e çevirir."""
    rapor = classification_report(y_gercek, y_tahmin, target_names=[SINIF_ADI[0], SINIF_ADI[1]], output_dict=True, zero_division=0)
    return pd.DataFrame(rapor).T.round(3)


def karmasiklik_ciz(y_gercek, y_tahmin, baslik, dosya_adi):
    """Karmaşıklık matrisini adet ve satır yüzdesi olarak yan yana çizer."""
    etiketler = [SINIF_ADI[0], SINIF_ADI[1]]
    adet = confusion_matrix(y_gercek, y_tahmin)
    oran = adet / adet.sum(axis=1, keepdims=True)
    fig, axes = plt.subplots(1, 2, figsize=(11, 4.4))
    ortak = dict(cbar=False, xticklabels=etiketler, yticklabels=etiketler, linewidths=2, linecolor="white", annot_kws={"size": 14})
    sns.heatmap(adet, annot=True, fmt="d", cmap="Blues", ax=axes[0], **ortak)
    sns.heatmap(oran, annot=True, fmt=".0%", cmap="Purples", vmin=0, vmax=1, ax=axes[1], **ortak)
    axes[0].set_title("Adet")
    axes[1].set_title("Satır yüzdesi (= sınıf duyarlılığı)")
    for ax in axes:
        ax.set(xlabel="Tahmin edilen", ylabel="Gerçek")
        ax.tick_params(axis="y", rotation=0)
    axes[1].set(ylabel="", yticklabels=[])
    fig.subplots_adjust(wspace=0.08)
    fig.suptitle(baslik, fontsize=15, fontweight="bold", y=1.04)
    kaydet(fig, dosya_adi)
    plt.show()


y_tahmin100 = orman100.predict(X_test)
karmasiklik_ciz(y_test, y_tahmin100, "Orman (100 ağaç, varsayılan) — karmaşıklık matrisi", "17_orman100_karmasiklik")
tn, fp, fn, tp = confusion_matrix(y_test, y_tahmin100).ravel()
print(f"TP = {tp:,} · FN = {fn:,} · FP = {fp:,} · TN = {tn:,}")
rapor100 = rapor_tablosu(y_test, y_tahmin100)
rapor100.to_csv(RAPOR_DIR / "orman100_siniflandirma_raporu.csv")
rapor100
````

### 🎚️ 13.3 · Eşik Neden Önemli? Kesinlik ve Duyarlılık Takası

`predict()` varsayılan olarak olasılığı **0,5**'te keser. Ama iş hedefi değişince eşik de değişmeli: eşiği **düşürürsek** daha çok kişiyi "yüksek" ilan ederiz (duyarlılık ↑, kesinlik ↓), **yükseltirsek** tersi olur.

### 💸 15.3 · İş Değeri: Eşiği Kampanya Ekonomisine Göre Seçmek

Model yalnızca bir **olasılık** üretir; kimi arayacağımıza **eşik** karar verir. Eşiği "F1 en yüksek olsun" diye değil, **paraya** bakarak seçelim.

<table><tr><td>

<img src="assets/etiket_ybs.svg" height="34" alt="YBS bağlantısı">

**Varsayımsal maliyet modeli** (kendi kurumunuzun rakamlarıyla değiştirin):

- 📞 Bir adayla temasın maliyeti: **20 $** (arama, teklif, personel)
- 💳 Temas edilen aday **gerçekten yüksek gelirliyse** beklenen brüt kazanç: **120 $** (kart ücreti + kullanım payı × kabul olasılığı)
- Net kazanç = TP × **(120 − 20)** − FP × **20**. Başa baş kesinlik = 20 / 120 ≈ **%17**: bir aday **%17'den fazla olasılıkla** yüksek gelirliyse aramaya değer.

</td></tr></table>

**Dürüstlük kuralı:** Eşiği test setinde aramak testi "kopya"ya çevirir. Bu yüzden eşiği **eğitim verisinin çapraz doğrulama tahminlerinden** (*out-of-fold*) seçip **test setinde yalnızca bir kez** uyguluyoruz.

````python
TEMAS_MALIYETI, BASARI_KAZANCI = 20, 120
NET_KAZANC = BASARI_KAZANCI - TEMAS_MALIYETI


def kampanya_kazanci(y_gercek, olasilik, esik, n_baz=1000):
    """Eşiğin üstündeki herkese teklif sunulursa 'n_baz' aday başına beklenen net kazanç ($)."""
    secilen = np.asarray(olasilik) >= esik
    tp = (secilen & (np.asarray(y_gercek) == 1)).sum()
    fp = (secilen & (np.asarray(y_gercek) == 0)).sum()
    return (tp * NET_KAZANC - fp * TEMAS_MALIYETI) / len(y_gercek) * n_baz


# Eşik seçimi: yalnızca EĞİTİM verisinin çapraz doğrulama (out-of-fold) olasılıkları
en_iyi_parametreler = {k.replace("model__", ""): v for k, v in arama.best_params_.items()}
oof = cross_val_predict(orman_pipeline(**en_iyi_parametreler), X_egitim, y_egitim, cv=cv5, method="predict_proba")[:, 1]
esik_dizisi = np.linspace(0.02, 0.9, 89)
oof_kazanc = np.array([kampanya_kazanci(y_egitim, oof, e) for e in esik_dizisi])
OPTIMUM_ESIK = float(esik_dizisi[oof_kazanc.argmax()])

# Test setinde tek seferlik uygulama
p_final = final_model.predict_proba(X_test)[:, 1]
test_kazanc = np.array([kampanya_kazanci(y_test, p_final, e) for e in esik_dizisi])

fig, ax = plt.subplots(figsize=(11.5, 5))
ax.plot(esik_dizisi, oof_kazanc, lw=2.8, color="#6C5CE7", label="Eğitim (out-of-fold) — eşiği buradan seçtik")
ax.plot(esik_dizisi, test_kazanc, lw=2.8, color="#00798C", label="Test seti — doğrulama")
ax.axvline(OPTIMUM_ESIK, color="#D1495B", ls="--", lw=1.6)
ax.axvline(0.5, color=VURGU, ls=":", lw=1.5)
ax.annotate(f"seçilen eşik = {OPTIMUM_ESIK:.2f}", xy=(OPTIMUM_ESIK, 8_000), xytext=(OPTIMUM_ESIK + 0.03, 6_500), color="#D1495B", fontweight="bold")
ax.annotate("varsayılan eşik 0,5", xy=(0.5, 8_000), xytext=(0.52, 6_500), color=VURGU, fontweight="bold")
ax.set(xlabel="Karar eşiği (bu olasılığın üstündekilere teklif sunulur)", ylabel="1.000 aday başına net kazanç ($)", title="Kampanya kazancı eşiğe göre nasıl değişiyor?")
ax.legend(frameon=False, loc="lower left")
kaydet(fig, "23_kampanya_kazanci")
plt.show()

senaryolar = pd.DataFrame({
    "teklif sunulan aday oranı": [0.0, 1.0, (p_final >= 0.5).mean(), (p_final >= OPTIMUM_ESIK).mean()],
    "1.000 aday başına net kazanç ($)": [0.0, kampanya_kazanci(y_test, np.ones(len(y_test)), 0.5), kampanya_kazanci(y_test, p_final, 0.5), kampanya_kazanci(y_test, p_final, OPTIMUM_ESIK)],
}, index=["Kimseye teklif sunma", "Herkese teklif sun", "Model · varsayılan eşik 0,50", f"Model · seçilen eşik {OPTIMUM_ESIK:.2f}"]).round(2)
senaryolar.to_csv(RAPOR_DIR / "kampanya_senaryolari.csv")
senaryolar
````

````python
def grup_denetimi(sutun):
    satirlar = []
    p_seri = pd.Series(p_final, index=X_test.index)
    for deger, idx in X_test.groupby(sutun).groups.items():
        if len(idx) < 200:
            continue
        yg, tahmin = y_test.loc[idx], p_seri.loc[idx] >= OPTIMUM_ESIK
        satirlar.append({"grup": f"{OZELLIK_TR[sutun]}: {deger}", "kişi": len(idx), "gerçek >50K oranı": yg.mean(), "teklif alma oranı": tahmin.mean(),
                         "yakalama (recall)": recall_score(yg, tahmin), "kesinlik": precision_score(yg, tahmin, zero_division=0)})
    return pd.DataFrame(satirlar).set_index("grup")


adalet = pd.concat([grup_denetimi("sex"), grup_denetimi("race")]).round(3)
adalet.to_csv(RAPOR_DIR / "adalet_denetimi.csv")

fig, ax = plt.subplots(figsize=(12, 4.8))
x = np.arange(len(adalet))
for i, (sutun, renk) in enumerate(zip(["gerçek >50K oranı", "teklif alma oranı", "yakalama (recall)"], ["#B0B7C3", "#6C5CE7", "#00798C"])):
    k = ax.bar(x + (i - 1) * 0.27, adalet[sutun], 0.27, label=sutun, color=renk, edgecolor="white")
    ax.bar_label(k, fmt="%.2f", fontsize=8, padding=2)
ax.set_xticks(x, [g.replace(": ", ":\n") for g in adalet.index])
ax.set(ylim=(0, 1.1), title=f"Gruplara göre gerçek oran, teklif alma ve yakalama (eşik {OPTIMUM_ESIK:.2f})")
ax.legend(ncol=3, frameon=False, loc="upper right")
kaydet(fig, "25_adalet_denetimi")
plt.show()
adalet
````
