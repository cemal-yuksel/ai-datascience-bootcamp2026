### 📊 **Senaryo**

Elinde basit bir sayı listesi var: `[1, 2, 3, 4, 5]`  
Bu sayıları bir çizgi grafiği olarak görselleştireceğiz.

**YBS Karşılığı:** Günlük satış miktarları, haftalık müşteri sayısı gibi sıralı veriler.

## 🎯 **Örnek 2: X ve Y Değerlerini Belirleme**

<div style="background: linear-gradient(135deg, #FFF3E0 0%, #FFE0B2 100%); padding: 15px; border-radius: 8px;">

### 📊 **Senaryo: Yaş - Kilo İlişkisi**

Bir sporcunun farklı yaşlardaki kilosu:
- Yaş: 20, 22, 25, 28, 30
- Kilo: 70, 72, 75, 80, 82

**İş Sorusu:** Yaş arttıkça kilo nasıl değişiyor?

</div>

````python
# Veri hazırlama: X ve Y eksenleri
yas = [20, 22, 25, 28, 30]
kilo = [70, 72, 75, 80, 82]

# Grafik çizimi: plt.plot(x, y)
plt.plot(yas, kilo)

# Başlık ve etiketler ekleme
plt.title('Yas - Kilo Iliskisi')  # Başlık
plt.xlabel('Yas')  # X ekseni etiketi
plt.ylabel('Kilo (kg)')  # Y ekseni etiketi

# Grafik ızgarası (grid) ekleme - okunabilirliği artırır
plt.grid(True, alpha=0.3)  # alpha: şeffaflık (0-1 arası)

# Grafiği göster
plt.show()

print("✅ Yaş-Kilo ilişkisi görselleştirildi!")
````

<div style="background: linear-gradient(135deg, #FCE4EC 0%, #F8BBD0 100%); padding: 15px; border-radius: 8px; margin-top: 10px;">

### 🔍 Analiz Notları

1. **Trend:** Yaş arttıkça kilo artıyor (pozitif ilişki)
2. **Artış Hızı:** Doğrusal görünüyor
3. **YBS Yorumu:** Bu trend, sporcunun kas kütlesi artışını gösterebilir

**Tabloya bakarak bu trendi görmek zordur, grafik hızlı içgörü sağlar!**

## ❌ **Örnek 3: Yaygın Hata - Liste Uzunluklarının Farklı Olması**

<div style="background: linear-gradient(135deg, #FFEBEE 0%, #FFCDD2 100%); padding: 15px; border-radius: 8px; border-left: 5px solid #C62828;">

### ⚠️ **Dikkat: Veri Bütünlüğü Kontrolü**

En sık yapılan hatalardan biri: **X ve Y listelerinin farklı uzunlukta olması**

**YBS'de karşılık:** Sipariş tablosu ile ödeme tablosunun satır sayısı uyuşmuyorsa rapor bozulur.

</div>

````python
# YANLIŞ: X ve Y farklı uzunlukta
yas_yanlis = [20, 22, 25, 28, 30]  # 5 eleman
kilo_yanlis = [70, 72, 75]  # 3 eleman - EKSİK VERİ!

# Bu kod HATA verecek!
try:
    plt.plot(yas_yanlis, kilo_yanlis)
    plt.show()
except ValueError as e:
    print("❌ HATA YAKALANDI!")
    print(f"Hata mesajı: {e}")
    print("\n🔍 Sorun: X ve Y listelerinin uzunluğu farklı!")
    print(f"   - Yaş listesi uzunluğu: {len(yas_yanlis)}")
    print(f"   - Kilo listesi uzunluğu: {len(kilo_yanlis)}")
    print("\n💡 Çözüm: Her X değeri için bir Y değeri olmalı!")
````

## 🎯 **Örnek 6: Doğrusal Büyüme vs Üssel Büyüme**

<div style="background: linear-gradient(135deg, #FFF3E0 0%, #FFE0B2 100%); padding: 15px; border-radius: 8px;">

### 📊 **Senaryo: İki Farklı Büyüme Modeli**

**İş Sorusu:** 
- Bir şirketin satışları doğrusal mı artıyor (Y = X)?
- Yoksa hızlanarak mı artıyor (Y = X³)?

**YBS Yorumu:**
- **Doğrusal artış**: Tahmin edilebilir, planlama kolay
- **Üssel artış**: Hızlanan büyüme, acil kapasite artışı gerekir

Bu farkı görmek, **kriz sinyalini** veya **fırsat penceresini** yakalamak için kritiktir!

</div>

````python
# Veri hazırlama: Doğrusal ve üssel büyüme
x = np.linspace(0, 10, 50)  # 0-10 arası 50 nokta

# Doğrusal büyüme: Y = X
y_dogrusal = x

# Üssel büyüme: Y = X³ (küp)
y_ussel = x ** 3

# İki grafiği çizelim
plt.figure(figsize=(10, 5))  # Figür boyutu (genişlik, yükseklik)

# Doğrusal büyüme
plt.plot(x, y_dogrusal, color='#2196F3', linewidth=2.5, label='Dogrusal Buyume (Y=X)', marker='o', markevery=5, markersize=6)

# Üssel büyüme
plt.plot(x, y_ussel, color='#FF5722', linewidth=2.5, label='Ussel Buyume (Y=X³)', marker='s', markevery=5, markersize=6)

# Başlık ve etiketler
plt.title('Dogrusal vs Ussel Buyume Karsilastirmasi', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Zaman (Ay)', fontsize=12)
plt.ylabel('Satis Miktari', fontsize=12)

# Legend (açıklama kutusu) ekle
plt.legend(loc='upper left', fontsize=10, framealpha=0.9, shadow=True)

# Izgara
plt.grid(True, alpha=0.3, linestyle='--', linewidth=0.7)

# Arka plan rengini hafif gri yap
plt.gca().set_facecolor('#FAFAFA')

plt.tight_layout()  # Grafik elemanlarını düzenli yerleştirir
plt.show()

print("✅ Doğrusal ve üssel büyüme karşılaştırması tamamlandı!")
````

## 🎯 **Örnek 7: Yatay (1 Satır, 2 Sütun) Subplot**

<div style="background: linear-gradient(135deg, #E0F2F1 0%, #B2DFDB 100%); padding: 15px; border-radius: 8px;">

### 📊 **Senaryo: İki Farklı Fonksiyon Karşılaştırması**

- **Sol Panel**: Sinüs fonksiyonu (dalgalı hareket)
- **Sağ Panel**: Kosinüs fonksiyonu (sinüsün kaydırılmış hali)

**YBS Karşılığı:** 
- Sol: Bu ay satışlar
- Sağ: Geçen ay satışlar

</div>

````python
# Veri hazırlama
x = np.linspace(0, 2*np.pi, 100)  # 0'dan 2π'ye kadar 100 nokta
y_sin = np.sin(x)  # Sinüs fonksiyonu
y_cos = np.cos(x)  # Kosinüs fonksiyonu

# Figure ve subplot oluşturma
# plt.subplots(satır_sayısı, sütun_sayısı, figsize=(genişlik, yükseklik))
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# SOL PANEL: Sinüs grafiği
axes[0].plot(x, y_sin, color='#2196F3', linewidth=2.5, label='sin(x)')
axes[0].set_title('Sinus Fonksiyonu', fontsize=13, fontweight='bold')
axes[0].set_xlabel('X Degeri (Radyan)', fontsize=11)
axes[0].set_ylabel('Y Degeri', fontsize=11)
axes[0].grid(True, alpha=0.3, linestyle='--')
axes[0].legend(loc='upper right')
axes[0].set_facecolor('#FAFAFA')

# SAĞ PANEL: Kosinüs grafiği
axes[1].plot(x, y_cos, color='#FF5722', linewidth=2.5, label='cos(x)')
axes[1].set_title('Cosinus Fonksiyonu', fontsize=13, fontweight='bold')
axes[1].set_xlabel('X Degeri (Radyan)', fontsize=11)
axes[1].set_ylabel('Y Degeri', fontsize=11)
axes[1].grid(True, alpha=0.3, linestyle='--')
axes[1].legend(loc='upper right')
axes[1].set_facecolor('#FAFAFA')

# Genel başlık (tüm figure için)
fig.suptitle('Trigonometrik Fonksiyonlar Karsilastirmasi', fontsize=16, fontweight='bold', y=1.02)

plt.tight_layout()
plt.show()

print("✅ Yatay subplot (1x2) tamamlandı!")
````

## 🎯 **Örnek 8: Dikey (2 Satır, 1 Sütun) Subplot**

<div style="background: linear-gradient(135deg, #F3E5F5 0%, #E1BEE7 100%); padding: 15px; border-radius: 8px;">

### 📊 **Senaryo: Üst-Alt Dashboard Düzeni**

**YBS Gerçek Hayat Örneği:**
- **Üst Panel**: Günlük satış trendi (detay)
- **Alt Panel**: Aylık satış ortalaması (özet)

Bu düzen, **detay ve özet** bilgiyi aynı anda sunmak için idealdir!

</div>

````python
# Veri hazırlama: 30 günlük satış verisi (simülasyon)
np.random.seed(42)  # Aynı sonuçları almak için seed
gunler = np.arange(1, 31)  # 1'den 30'a kadar günler

# Günlük satış: Trend + rastgele gürültü
satis_gunluk = 100 + 2*gunler + np.random.normal(0, 10, 30)

# Haftalık ortalama (her 7 günü grupla)
haftalar = ['Hafta 1', 'Hafta 2', 'Hafta 3', 'Hafta 4']
satis_haftalik = [
    satis_gunluk[:7].mean(),
    satis_gunluk[7:14].mean(),
    satis_gunluk[14:21].mean(),
    satis_gunluk[21:28].mean()
]

# Figure ve subplot oluşturma: 2 satır, 1 sütun
fig, axes = plt.subplots(2, 1, figsize=(12, 9))

# ÜST PANEL: Günlük satış detayı
axes[0].plot(gunler, satis_gunluk, color='#4CAF50', linewidth=2, marker='o', markersize=6, markerfacecolor='#81C784', label='Gunluk Satis')
axes[0].axhline(y=satis_gunluk.mean(), color='#F44336', linestyle='--', linewidth=2, label=f'Ortalama: {satis_gunluk.mean():.1f}')
axes[0].set_title('Gunluk Satis Detayi (30 Gun)', fontsize=13, fontweight='bold', color='#1B5E20')
axes[0].set_xlabel('Gun', fontsize=11)
axes[0].set_ylabel('Satis Miktari (TL)', fontsize=11)
axes[0].grid(True, alpha=0.3, linestyle='--')
axes[0].legend(loc='upper left', fontsize=10)
axes[0].set_facecolor('#F1F8E9')

# ALT PANEL: Haftalık özet
axes[1].bar(haftalar, satis_haftalik, color=['#2196F3', '#FF9800', '#9C27B0', '#00BCD4'], edgecolor='black', linewidth=1.5)
axes[1].set_title('Haftalik Ortalama Satis (Ozet)', fontsize=13, fontweight='bold', color='#0D47A1')
axes[1].set_xlabel('Hafta', fontsize=11)
axes[1].set_ylabel('Ortalama Satis (TL)', fontsize=11)
axes[1].grid(True, alpha=0.3, linestyle='--', axis='y')
axes[1].set_facecolor('#E3F2FD')

# Her çubuğun üzerine değer yaz
for i, v in enumerate(satis_haftalik):
    axes[1].text(i, v + 2, f'{v:.1f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

# Genel başlık
fig.suptitle('Satis Performans Dashboard', fontsize=16, fontweight='bold', y=0.995)

plt.tight_layout()
plt.show()

print("✅ Dikey subplot (2x1) dashboard tamamlandı!")
print(f"📊 Aylık toplam satış: {satis_gunluk.sum():.2f} TL")
````

## 🎯 **Örnek 9: Grid (2x2) Subplot - Karmaşık Dashboard**

<div style="background: linear-gradient(135deg, #FFECB3 0%, #FFE082 100%); padding: 15px; border-radius: 8px;">

### 📊 **4 Panelli Dashboard**

**YBS Gerçek Senaryo:** Yönetim toplantısı sunumu

**4 Panel:**
1. **Sol Üst**: Satış trendi
2. **Sağ Üst**: Müşteri sayısı
3. **Sol Alt**: Ürün kategorisi dağılımı
4. **Sağ Alt**: Bölgesel performans

</div>

````python
# Veri hazırlama
np.random.seed(42)
aylar = ['Ocak', 'Subat', 'Mart', 'Nisan', 'Mayis', 'Haziran']

satis = [120, 135, 150, 145, 170, 190]
musteri_sayisi = [50, 58, 65, 63, 75, 85]
kategoriler = ['Elektronik', 'Giyim', 'Gida', 'Ev', 'Spor']
kategori_paylar = [35, 25, 20, 15, 5]
bolgeler = ['Istanbul', 'Ankara', 'Izmir', 'Antalya']
bolge_satis = [450, 320, 280, 220]

# 2x2 Grid oluşturma
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# PANEL 1 (Sol Üst): Satış Trendi - Çizgi Grafik
axes[0, 0].plot(aylar, satis, color='#4CAF50', linewidth=3, marker='o', markersize=10, markerfacecolor='#81C784')
axes[0, 0].fill_between(range(len(aylar)), satis, alpha=0.3, color='#4CAF50')  # Altını doldur
axes[0, 0].set_title('Aylik Satis Trendi', fontsize=13, fontweight='bold', color='#1B5E20', pad=10)
axes[0, 0].set_xlabel('Ay', fontsize=11)
axes[0, 0].set_ylabel('Satis (Bin TL)', fontsize=11)
axes[0, 0].grid(True, alpha=0.3, linestyle='--')
axes[0, 0].set_facecolor('#F1F8E9')

# PANEL 2 (Sağ Üst): Müşteri Sayısı - Alan Grafik
axes[0, 1].fill_between(range(len(aylar)), musteri_sayisi, color='#2196F3', alpha=0.6)
axes[0, 1].plot(aylar, musteri_sayisi, color='#0D47A1', linewidth=2.5, marker='s', markersize=8)
axes[0, 1].set_title('Aylik Musteri Sayisi', fontsize=13, fontweight='bold', color='#0D47A1', pad=10)
axes[0, 1].set_xlabel('Ay', fontsize=11)
axes[0, 1].set_ylabel('Musteri Sayisi', fontsize=11)
axes[0, 1].grid(True, alpha=0.3, linestyle='--')
axes[0, 1].set_facecolor('#E3F2FD')

# PANEL 3 (Sol Alt): Kategori Dağılımı - Pasta Grafik
renk_paleti = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8']
explode = (0.1, 0, 0, 0, 0)  # İlk dilimi biraz çıkar
axes[1, 0].pie(kategori_paylar, labels=kategoriler, autopct='%1.1f%%', colors=renk_paleti, 
               explode=explode, shadow=True, startangle=90, textprops={'fontsize': 10, 'weight': 'bold'})
axes[1, 0].set_title('Urun Kategorisi Dagilimi', fontsize=13, fontweight='bold', pad=10)

# PANEL 4 (Sağ Alt): Bölgesel Satış - Yatay Çubuk Grafik
bars = axes[1, 1].barh(bolgeler, bolge_satis, color=['#E91E63', '#9C27B0', '#3F51B5', '#00BCD4'], 
                        edgecolor='black', linewidth=1.5)
axes[1, 1].set_title('Bolgesel Satis Performansi', fontsize=13, fontweight='bold', color='#880E4F', pad=10)
axes[1, 1].set_xlabel('Satis (Bin TL)', fontsize=11)
axes[1, 1].set_ylabel('Bolge', fontsize=11)
axes[1, 1].grid(True, alpha=0.3, linestyle='--', axis='x')
axes[1, 1].set_facecolor('#FCE4EC')

# Değer etiketleri ekle
for i, (bar, value) in enumerate(zip(bars, bolge_satis)):
    axes[1, 1].text(value + 10, bar.get_y() + bar.get_height()/2, f'{value}', 
                    va='center', fontsize=10, fontweight='bold')

# Genel başlık
fig.suptitle('Sirket Performans Dashboard - 2026 H1', fontsize=17, fontweight='bold', y=0.995)

plt.tight_layout()
plt.show()

print("✅ 2x2 Grid dashboard tamamlandı!")
print(f"📊 Toplam satış: {sum(satis)} Bin TL")
print(f"👥 Toplam müşteri: {sum(musteri_sayisi)}")
````

## 🎯 **Örnek 10: Renk Seçenekleri - Üç Farklı Yöntem**

<div style="background: linear-gradient(135deg, #F3E5F5 0%, #E1BEE7 100%); padding: 15px; border-radius: 8px;">

### 🎨 Üç Renk Tanımlama Yöntemi

1. **İsim ile**: 'red', 'blue', 'green', 'orange'
2. **Hex Kod ile**: '#FF5722', '#4CAF50', '#2196F3'
3. **RGB/RGBA ile**: (1.0, 0.5, 0.0) veya (1.0, 0.5, 0.0, 0.7)

**YBS Önerisi:** Kurumsal projeler için **Hex kod** kullan (standart ve tutarlı)

</div>

````python
# Veri hazırlama
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.sin(x + 0.5)
y3 = np.sin(x + 1.0)

# Figure oluştur
plt.figure(figsize=(14, 5))

# 1. YÖNTEM: İsim ile renk
plt.subplot(1, 3, 1)
plt.plot(x, y1, color='red', linewidth=2.5, label='Kirmizi')
plt.plot(x, y2, color='blue', linewidth=2.5, label='Mavi')
plt.plot(x, y3, color='green', linewidth=2.5, label='Yesil')
plt.title('Renk: Isim ile', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)

# 2. YÖNTEM: Hex kod ile renk (ÖNERİLEN - Kurumsal)
plt.subplot(1, 3, 2)
plt.plot(x, y1, color='#E53935', linewidth=2.5, label='Kirmizi Hex')
plt.plot(x, y2, color='#1E88E5', linewidth=2.5, label='Mavi Hex')
plt.plot(x, y3, color='#43A047', linewidth=2.5, label='Yesil Hex')
plt.title('Renk: Hex Kod ile (Kurumsal)', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)

# 3. YÖNTEM: RGB ile renk
plt.subplot(1, 3, 3)
plt.plot(x, y1, color=(0.9, 0.2, 0.2), linewidth=2.5, label='RGB Kirmizi')
plt.plot(x, y2, color=(0.1, 0.5, 0.9), linewidth=2.5, label='RGB Mavi')
plt.plot(x, y3, color=(0.2, 0.6, 0.3), linewidth=2.5, label='RGB Yesil')
plt.title('Renk: RGB ile', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)

plt.suptitle('Uc Farkli Renk Tanimlama Yontemi', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("✅ Renk seçenekleri gösterildi!")
print("💡 Kurumsal projelerde Hex kod kullanımı önerilir.")
````

## 🎯 **Örnek 11: Çizgi Stilleri (Linestyle) - Plan vs Gerçekleşen**

<div style="background: linear-gradient(135deg, #E8F5E9 0%, #C8E6C9 100%); padding: 15px; border-radius: 8px;">

### 📊 **YBS Senaryosu**

**2026 Satış Hedefi vs Gerçekleşen**

- **Düz çizgi (-)**: Gerçekleşen satışlar
- **Kesik çizgi (--)**: Hedef/Plan
- **Noktalı çizgi (:)**: Tahmin/Projeksiyon
- **Çizgi-nokta (-.)**: Alternatif senaryo

Bu ayrım, **yöneticinin hangi çizginin ne olduğunu** anında anlamasını sağlar!

</div>

````python
# Veri hazırlama: 2026 ilk 6 ay satış verisi
aylar = ['Ocak', 'Subat', 'Mart', 'Nisan', 'Mayis', 'Haziran']
gerceklesen = [120, 135, 148, 155, 168, 180]
hedef = [130, 140, 150, 160, 170, 180]
tahmin = [115, 130, 145, 160, 175, 190]
alternatif = [125, 138, 152, 165, 178, 190]

# Grafik
plt.figure(figsize=(12, 6))

# Gerçekleşen - Düz çizgi (en belirgin)
plt.plot(aylar, gerceklesen, color='#4CAF50', linewidth=3, linestyle='-', 
         marker='o', markersize=10, label='Gerceklesen Satis', markerfacecolor='#81C784')

# Hedef - Kesik çizgi
plt.plot(aylar, hedef, color='#2196F3', linewidth=2.5, linestyle='--', 
         marker='s', markersize=8, label='Hedef/Plan', markerfacecolor='#64B5F6')

# Tahmin - Noktalı çizgi
plt.plot(aylar, tahmin, color='#FF9800', linewidth=2.5, linestyle=':', 
         marker='^', markersize=8, label='Tahmin', markerfacecolor='#FFB74D')

# Alternatif - Çizgi-nokta
plt.plot(aylar, alternatif, color='#9C27B0', linewidth=2.5, linestyle='-.', 
         marker='D', markersize=7, label='Alternatif Senaryo', markerfacecolor='#BA68C8')

# Başlık ve etiketler
plt.title('2026 H1 Satis Performansi: Plan vs Gerceklesen', 
          fontsize=15, fontweight='bold', pad=20)
plt.xlabel('Ay', fontsize=12, fontweight='bold')
plt.ylabel('Satis (Bin TL)', fontsize=12, fontweight='bold')

# Legend (sağ alt köşede)
plt.legend(loc='lower right', fontsize=11, framealpha=0.95, shadow=True, 
           fancybox=True, borderpad=1)

# Grid
plt.grid(True, alpha=0.4, linestyle='--', linewidth=0.7)

# Arka plan
plt.gca().set_facecolor('#FAFAFA')

# Y ekseni limitlerini ayarla
plt.ylim(100, 200)

plt.tight_layout()
plt.show()

print("✅ Çizgi stilleri ile plan-gerçekleşen karşılaştırması!")
print(f"🎯 Hedef tutturma oranı: {(gerceklesen[-1]/hedef[-1])*100:.1f}%")
````

## 🎯 **Örnek 12: Marker Stilleri - Veri Noktalarını Kanıtlamak**

<div style="background: linear-gradient(135deg, #E0F2F1 0%, #B2DFDB 100%); padding: 15px; border-radius: 8px;">

### 📊 **Marker Neden Önemli?**

**Çizgi sadece interpolasyon hissi verir.**  
**Marker ise "bu noktada gerçekten ölçüm var" der.**

**YBS Örneği:** Sensör verisi, anket sonucu, aylık kapanış rakamları - bunlar **gerçek ölçüm noktalarıdır**, marker ile gösterilmelidir!

</div>

````python
# Veri hazırlama
x = [1, 2, 3, 4, 5, 6, 7]
y_base = [2, 4, 3, 5, 4, 6, 5]

# Marker stilleri listesi
markers = ['o', 's', '^', 'D', '*', 'x', 'p', 'h', '+', 'v']
marker_names = ['Circle', 'Square', 'Triangle Up', 'Diamond', 'Star', 
                'X', 'Pentagon', 'Hexagon', 'Plus', 'Triangle Down']

# Figure oluştur
fig, axes = plt.subplots(2, 5, figsize=(16, 8))
axes = axes.flatten()

# Her marker için ayrı bir subplot
for i, (marker, name) in enumerate(zip(markers, marker_names)):
    y = [val + i*0.3 for val in y_base]  # Her marker için hafif kaydırma
    
    axes[i].plot(x, y, marker=marker, markersize=12, linewidth=2.5,
                color='#2196F3', markerfacecolor='#FF9800', 
                markeredgewidth=2, markeredgecolor='#E65100')
    
    axes[i].set_title(f"Marker: '{marker}' ({name})", fontsize=11, fontweight='bold')
    axes[i].grid(True, alpha=0.3, linestyle='--')
    axes[i].set_facecolor('#FAFAFA')
    axes[i].set_xlabel('X', fontsize=9)
    axes[i].set_ylabel('Y', fontsize=9)

plt.suptitle('Matplotlib Marker Stilleri Galerisi', fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("✅ 10 farklı marker stili gösterildi!")
print("💡 En yaygın kullanılanlar: 'o' (circle), 's' (square), '^' (triangle)")
````

## 🎯 **Örnek 13: Alpha (Şeffaflık) - Yoğun Veriyle Çalışma**

<div style="background: linear-gradient(135deg, #FFF3E0 0%, #FFE0B2 100%); padding: 15px; border-radius: 8px;">

### 📊 Senaryo: 1000 Müşterinin Yaş-Gelir Dağılımı

Çok sayıda nokta olduğunda **üst üste binme** problemi oluşur.  
**Alpha (şeffaflık)** parametresi bu problemi çözer!

**YBS Kullanımı:** Büyük veri setlerinde yoğunluk görünür olur.

</div>

````python
# Veri hazırlama: 1000 müşteri simülasyonu
np.random.seed(42)
yas = np.random.normal(35, 10, 1000)  # Ortalama 35, std 10
gelir = 30000 + 1200*yas + np.random.normal(0, 8000, 1000)  # Yaşa bağlı gelir

# İki grafik: Alpha = 1.0 vs Alpha = 0.3
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# SOLDA: Alpha = 1.0 (Opak - şeffaf değil)
axes[0].scatter(yas, gelir, color='#2196F3', s=50, alpha=1.0, edgecolors='none')
axes[0].set_title('Alpha = 1.0 (Seffaf Degil)\nYogunluk Gorunmuyor!', 
                  fontsize=13, fontweight='bold', color='#C62828')
axes[0].set_xlabel('Yas', fontsize=11)
axes[0].set_ylabel('Gelir (TL)', fontsize=11)
axes[0].grid(True, alpha=0.3, linestyle='--')
axes[0].set_facecolor('#FAFAFA')

# SAĞDA: Alpha = 0.3 (Şeffaf)
axes[1].scatter(yas, gelir, color='#4CAF50', s=50, alpha=0.3, edgecolors='none')
axes[1].set_title('Alpha = 0.3 (Seffaf)\nYogunluk Gorunuyor!', 
                  fontsize=13, fontweight='bold', color='#2E7D32')
axes[1].set_xlabel('Yas', fontsize=11)
axes[1].set_ylabel('Gelir (TL)', fontsize=11)
axes[1].grid(True, alpha=0.3, linestyle='--')
axes[1].set_facecolor('#FAFAFA')

plt.suptitle('Alpha (Seffaflik) Parametresinin Etkisi - 1000 Musteri', 
             fontsize=15, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()

print("✅ Alpha parametresi ile yoğunluk görselleştirildi!")
print(f"📊 Toplam müşteri sayısı: {len(yas)}")
print(f"📈 Ortalama yaş: {yas.mean():.1f}")
print(f"💰 Ortalama gelir: {gelir.mean():.0f} TL")
````

<div style="background: linear-gradient(135deg, #E1BEE7 0%, #CE93D8 100%); padding: 15px; border-radius: 8px; margin-top: 10px;">

### 💡 **Alpha Kullanım Kılavuzu**

**Alpha Değerleri:**
- **1.0**: Tamamen opak (şeffaf değil) - normal kullanım
- **0.7-0.9**: Hafif şeffaflık - orta yoğunluk
- **0.3-0.6**: Orta şeffaflık - yoğun veri için ideal
- **0.1-0.2**: Çok şeffaf - aşırı yoğun veri

### 🎯 YBS'de Kullanım

**Şeffaflık sayesinde:**
- ✅ Veri yoğunluğu görünür olur
- ✅ Kümelenme (clustering) gözle fark edilir
- ✅ Uç değerler (outliers) belirginleşir
- ✅ Grafik kalabalık görünmez

**Sağdaki grafikte fark ettiniz mi?**  
35 yaş civarında **yoğunluk koyu yeşil** görünüyor → Çoğu müşteri bu yaş aralığında!

## 1️⃣ **Scatter Plot (Dağılım Grafiği)**

### 📊 **Ne Zaman Kullanılır?**

<div style="background: linear-gradient(135deg, #E8F5E9 0%, #C8E6C9 100%); padding: 15px; border-radius: 8px;">

**Amaç:** İki sayısal değişken arasındaki ilişkiyi görmek

**YBS Soruları:**
- "Boy arttıkça kilo da artıyor mu?"
- "Deneyim arttıkça maaş artıyor mu?"
- "Çalışan sayısı ile satış arasında ilişki var mı?"

**Ne Görürüz:**
- ✅ Pozitif/Negatif korelasyon
- ✅ Kümelenme (clustering)
- ✅ Uç değerler (outliers)
- ✅ Doğrusal/Doğrusal olmayan ilişki

</div>

````python
# Örnek Veri: Çalışan Deneyimi vs Maaşı
np.random.seed(42)
deneyim_yil = np.random.uniform(0, 15, 100)  # 0-15 yıl arası deneyim
maas = 30000 + 3500 * deneyim_yil + np.random.normal(0, 8000, 100)  # Deneyime bağlı maaş

# Scatter Plot
plt.figure(figsize=(10, 6))
plt.scatter(deneyim_yil, maas, s=80, alpha=0.6, c='#2196F3', edgecolors='#0D47A1', linewidth=1.5)

# Trend çizgisi ekleyelim (polyfit ile)
z = np.polyfit(deneyim_yil, maas, 1)
p = np.poly1d(z)
plt.plot(deneyim_yil, p(deneyim_yil), "r--", linewidth=2.5, label=f'Trend: y={z[0]:.0f}x+{z[1]:.0f}')

# Başlık ve etiketler
plt.title('Calisma Deneyimi vs Maas Iliskisi\nInsan Kaynaklari Analizi', 
          fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Deneyim (Yil)', fontsize=12, fontweight='bold')
plt.ylabel('Maas (TL)', fontsize=12, fontweight='bold')

# Legend ve grid
plt.legend(loc='lower right', fontsize=11, framealpha=0.9)
plt.grid(True, alpha=0.3, linestyle='--')
plt.gca().set_facecolor('#FAFAFA')

plt.tight_layout()
plt.show()

# İstatistikler
corr = np.corrcoef(deneyim_yil, maas)[0, 1]
print(f"✅ Korelasyon katsayısı: {corr:.3f}")
print(f"📊 Her yıl deneyim başına ortalama maaş artışı: {z[0]:.0f} TL")
print("💡 YBS Yorumu: Güçlü pozitif ilişki var → Deneyim ile maaş artıyor")
````

## 2️⃣ **Histogram (Dağılım Frekans Grafiği)**

### 📊 **Ne Zaman Kullanılır?**

<div style="background: linear-gradient(135deg, #FFF3E0 0%, #FFE0B2 100%); padding: 15px; border-radius: 8px;">

**Amaç:** Tek bir sayısal değişkenin dağılımını görmek

**YBS Soruları:**
- "Çalışanların yaş dağılımı nasıl?"
- "Satışlar hangi aralıkta yoğunlaşıyor?"
- "Müşteri gelir dağılımı normal mi, çarpık mı?"

**Ne Görürüz:**
- ✅ Tek tepe mi, çift tepe mi? (unimodal/bimodal)
- ✅ Normal dağılım mı?
- ✅ Sağa/Sola çarpıklık var mı?
- ✅ Hangi aralıkta yoğunluk var?

</div>

````python
# Örnek Veri: Müşteri Yaş Dağılımı
np.random.seed(42)
musteri_yas = np.random.normal(35, 12, 500)  # Ortalama 35, std 12

# Histogram
plt.figure(figsize=(12, 6))

# Alt subplot 1: Histogram
plt.subplot(1, 2, 1)
n, bins, patches = plt.hist(musteri_yas, bins=25, color='#4CAF50', alpha=0.7, 
                             edgecolor='#1B5E20', linewidth=1.5)

# Ortalama çizgisi
mean_yas = musteri_yas.mean()
plt.axvline(mean_yas, color='#E53935', linestyle='--', linewidth=3, 
            label=f'Ortalama: {mean_yas:.1f}')

plt.title('Musteri Yas Dagilimi (Histogram)', fontsize=13, fontweight='bold')
plt.xlabel('Yas', fontsize=11)
plt.ylabel('Frekans (Musteri Sayisi)', fontsize=11)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3, axis='y')
plt.gca().set_facecolor('#FAFAFA')

# Alt subplot 2: Histogram + KDE (Yoğunluk eğrisi)
plt.subplot(1, 2, 2)
plt.hist(musteri_yas, bins=25, color='#2196F3', alpha=0.5, 
         edgecolor='#0D47A1', linewidth=1.5, density=True, label='Histogram')

# KDE (Kernel Density Estimation - Yoğunluk eğrisi)
from scipy.stats import gaussian_kde
kde = gaussian_kde(musteri_yas)
x_range = np.linspace(musteri_yas.min(), musteri_yas.max(), 100)
plt.plot(x_range, kde(x_range), color='#FF5722', linewidth=3, label='Yogunluk Egrisi (KDE)')

plt.axvline(mean_yas, color='#E53935', linestyle='--', linewidth=3, 
            label=f'Ortalama: {mean_yas:.1f}')

plt.title('Musteri Yas Dagilimi + Yogunluk Egrisi', fontsize=13, fontweight='bold')
plt.xlabel('Yas', fontsize=11)
plt.ylabel('Yogunluk', fontsize=11)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3, axis='y')
plt.gca().set_facecolor('#FAFAFA')

plt.suptitle('Histogram Analizi: Dagilim Sekli ve Yogunluk', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# İstatistikler
print(f"✅ Ortalama yaş: {musteri_yas.mean():.1f}")
print(f"✅ Medyan yaş: {np.median(musteri_yas):.1f}")
print(f"✅ Standart sapma: {musteri_yas.std():.1f}")
print(f"📊 Çoğunluk: {mean_yas - musteri_yas.std():.1f} - {mean_yas + musteri_yas.std():.1f} yaş arası")
print("💡 YBS Yorumu: Normal dağılım → Tek tip müşteri profili")
````

## 3️⃣ **Box Plot (Kutu Grafiği)**

### 📊 **Ne Zaman Kullanılır?**

<div style="background: linear-gradient(135deg, #F3E5F5 0%, #E1BEE7 100%); padding: 15px; border-radius: 8px;">

**Amaç:** Gruplar arası karşılaştırma ve uç değer tespiti

**YBS Soruları:**
- "Departmanlar arası maaş farkı var mı?"
- "Hangi branşta kilo dağılımı daha geniş?"
- "Uç değerler (outliers) hangi grupta?"

**Ne Görürüz:**
- ✅ Medyan (orta çizgi)
- ✅ Q1 ve Q3 (kutu alt/üst)
- ✅ Min-Max (bıyıklar)
- ✅ Uç değerler (noktalar)

**Box Plot Anatomisi:**
```
      |---- Whisker (Bıyık) - Min
      |
    ┌─┴─┐
    │   │  Q3 (Üst Çeyrek)
    ├───┤  Medyan
    │   │  Q1 (Alt Çeyrek)
    └─┬─┘
      |
      |---- Whisker (Bıyık) - Max
      •      Outlier (Uç Değer)
```

</div>

````python
# Örnek Veri: Departman Bazlı Maaş Dağılımı
np.random.seed(42)
it_maas = np.random.normal(65000, 15000, 100)
pazarlama_maas = np.random.normal(55000, 12000, 100)
finans_maas = np.random.normal(70000, 18000, 100)
insan_kay_maas = np.random.normal(50000, 10000, 100)

veriler = [it_maas, pazarlama_maas, finans_maas, insan_kay_maas]
departmanlar = ['IT', 'Pazarlama', 'Finans', 'Insan Kay.']

# Box Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# SOL: Dikey Box Plot
bp1 = axes[0].boxplot(veriler, labels=departmanlar, patch_artist=True,
                       notch=True, showmeans=True,
                       boxprops=dict(facecolor='#BBDEFB', linewidth=2, edgecolor='#1976D2'),
                       whiskerprops=dict(linewidth=2, color='#0D47A1'),
                       capprops=dict(linewidth=2, color='#0D47A1'),
                       medianprops=dict(linewidth=3, color='#E53935'),
                       meanprops=dict(marker='D', markerfacecolor='#FF9800', markersize=8))

axes[0].set_title('Departman Bazli Maas Dagilimi (Dikey)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Departman', fontsize=11, fontweight='bold')
axes[0].set_ylabel('Maas (TL)', fontsize=11, fontweight='bold')
axes[0].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[0].set_facecolor('#FAFAFA')

# Ortalama çizgisi ekle
genel_ortalama = np.mean([np.mean(v) for v in veriler])
axes[0].axhline(genel_ortalama, color='#4CAF50', linestyle='--', linewidth=2, 
                label=f'Genel Ort: {genel_ortalama:.0f}')
axes[0].legend(fontsize=10)

# SAĞ: Yatay Box Plot
bp2 = axes[1].boxplot(veriler, labels=departmanlar, patch_artist=True, vert=False,
                       notch=True, showmeans=True,
                       boxprops=dict(facecolor='#C8E6C9', linewidth=2, edgecolor='#388E3C'),
                       whiskerprops=dict(linewidth=2, color='#1B5E20'),
                       capprops=dict(linewidth=2, color='#1B5E20'),
                       medianprops=dict(linewidth=3, color='#E53935'),
                       meanprops=dict(marker='D', markerfacecolor='#FF9800', markersize=8))

axes[1].set_title('Departman Bazli Maas Dagilimi (Yatay)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Maas (TL)', fontsize=11, fontweight='bold')
axes[1].set_ylabel('Departman', fontsize=11, fontweight='bold')
axes[1].grid(True, alpha=0.3, axis='x', linestyle='--')
axes[1].set_facecolor('#FAFAFA')
axes[1].axvline(genel_ortalama, color='#4CAF50', linestyle='--', linewidth=2, 
                label=f'Genel Ort: {genel_ortalama:.0f}')
axes[1].legend(fontsize=10)

plt.suptitle('Box Plot Analizi: Departmanlar Arasi Maas Karsilastirmasi', 
             fontsize=15, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()

# İstatistikler
print("📊 Departman Bazlı Maaş İstatistikleri:\n")
for i, dept in enumerate(departmanlar):
    print(f"{dept:12} → Medyan: {np.median(veriler[i]):>7.0f} TL | Ortalama: {np.mean(veriler[i]):>7.0f} TL | Std: {np.std(veriler[i]):>7.0f}")
print("\n💡 YBS Yorumu: Finans departmanı en yüksek, İnsan Kaynakları en düşük ortalamaya sahip")
````

### 🔑 **Box Plot Elemanları**

- **🔴 Kırmızı Çizgi**: Medyan (ortanca değer)
- **🟠 Turuncu Elmas**: Ortalama (mean)
- **🔵 Kutu**: Q1-Q3 arası (verinin %50'si)
- **┃ Bıyıklar**: Min-Max (1.5*IQR kuralı)
- **• Noktalar**: Uç değerler (outliers)
- **Notch (Çentik)**: Güven aralığı

## 4️⃣ **Bar Chart (Çubuk Grafik)**

### 📊 **Ne Zaman Kullanılır?**

<div style="background: linear-gradient(135deg, #E0F2F1 0%, #B2DFDB 100%); padding: 15px; border-radius: 8px;">

**Amaç:** Kategoriler arası sayısal karşılaştırma

**YBS Soruları:**
- "Hangi ürün kategorisi en çok satıyor?"
- "Bölgelere göre müşteri sayısı nasıl?"
- "Yıllık bazda hangi departman en çok çalışan aldı?"

**Ne Görürüz:**
- ✅ En yüksek/En düşük kategori
- ✅ Kategoriler arası fark
- ✅ Grup karşılaştırması (Grouped Bar)
- ✅ Yığılmalı karşılaştırma (Stacked Bar)

</div>

````python
# Örnek Veri: Ürün Kategorilerine Göre Satış
kategoriler = ['Elektronik', 'Giyim', 'Gida', 'Ev', 'Spor', 'Kitap']
satis_2025 = [450, 320, 280, 220, 180, 150]
satis_2026 = [520, 340, 310, 250, 210, 170]

x = np.arange(len(kategoriler))
width = 0.35

fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# 1. BASİT BAR CHART
bars1 = axes[0, 0].bar(kategoriler, satis_2026, color='#4CAF50', alpha=0.8, 
                       edgecolor='#1B5E20', linewidth=2)
axes[0, 0].set_title('2026 Kategori Bazli Satis (Basit Bar)', fontsize=13, fontweight='bold')
axes[0, 0].set_ylabel('Satis (Bin TL)', fontsize=11)
axes[0, 0].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[0, 0].set_facecolor('#FAFAFA')

# Değer etiketleri
for bar in bars1:
    height = bar.get_height()
    axes[0, 0].text(bar.get_x() + bar.get_width()/2., height,
                    f'{int(height)}', ha='center', va='bottom', fontsize=10, fontweight='bold')

# 2. GROUPED BAR CHART (Yan Yana)
bars2a = axes[0, 1].bar(x - width/2, satis_2025, width, label='2025', 
                        color='#FF9800', alpha=0.8, edgecolor='#E65100', linewidth=1.5)
bars2b = axes[0, 1].bar(x + width/2, satis_2026, width, label='2026', 
                        color='#2196F3', alpha=0.8, edgecolor='#0D47A1', linewidth=1.5)
axes[0, 1].set_title('Yillik Karsilastirma (Grouped Bar)', fontsize=13, fontweight='bold')
axes[0, 1].set_ylabel('Satis (Bin TL)', fontsize=11)
axes[0, 1].set_xticks(x)
axes[0, 1].set_xticklabels(kategoriler, rotation=45, ha='right')
axes[0, 1].legend(fontsize=11)
axes[0, 1].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[0, 1].set_facecolor('#FAFAFA')

# 3. STACKED BAR CHART (Üst Üste)
axes[1, 0].bar(kategoriler, satis_2025, label='2025', color='#FF9800', 
               alpha=0.8, edgecolor='#E65100', linewidth=1.5)
axes[1, 0].bar(kategoriler, satis_2026, bottom=satis_2025, label='2026', 
               color='#2196F3', alpha=0.8, edgecolor='#0D47A1', linewidth=1.5)
axes[1, 0].set_title('Yillik Toplam (Stacked Bar)', fontsize=13, fontweight='bold')
axes[1, 0].set_ylabel('Toplam Satis (Bin TL)', fontsize=11)
axes[1, 0].legend(fontsize=11)
axes[1, 0].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[1, 0].set_facecolor('#FAFAFA')
plt.setp(axes[1, 0].xaxis.get_majorticklabels(), rotation=45, ha='right')

# 4. YATAY BAR CHART
y_pos = np.arange(len(kategoriler))
bars4 = axes[1, 1].barh(y_pos, satis_2026, color=['#E53935', '#1E88E5', '#43A047', 
                                                    '#FB8C00', '#8E24AA', '#00ACC1'],
                        alpha=0.8, edgecolor='black', linewidth=1.5)
axes[1, 1].set_yticks(y_pos)
axes[1, 1].set_yticklabels(kategoriler)
axes[1, 1].set_title('2026 Satis Siralaması (Yatay Bar)', fontsize=13, fontweight='bold')
axes[1, 1].set_xlabel('Satis (Bin TL)', fontsize=11)
axes[1, 1].grid(True, alpha=0.3, axis='x', linestyle='--')
axes[1, 1].set_facecolor('#FAFAFA')

# Değer etiketleri
for i, bar in enumerate(bars4):
    width_val = bar.get_width()
    axes[1, 1].text(width_val + 10, bar.get_y() + bar.get_height()/2.,
                    f'{int(width_val)}', ha='left', va='center', fontsize=10, fontweight='bold')

plt.suptitle('Bar Chart Cesitleri: Kategori Bazli Satis Analizi', 
             fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

# Analiz
print("📊 Kategori Performans Analizi:\n")
for i, kat in enumerate(kategoriler):
    artis = ((satis_2026[i] - satis_2025[i]) / satis_2025[i]) * 100
    print(f"{kat:12} → 2025: {satis_2025[i]:3} | 2026: {satis_2026[i]:3} | Artış: %{artis:5.1f}")

print(f"\n🏆 En çok satan: {kategoriler[np.argmax(satis_2026)]}")
print(f"📉 En az satan: {kategoriler[np.argmin(satis_2026)]}")
print("💡 YBS Yorumu: Elektronik lider, fakat Kitap kategorisinde büyüme potansiyeli var")
````

## 7️⃣ **Heatmap (Korelasyon Haritası)**

### 📊 **Ne Zaman Kullanılır?**

<div style="background: linear-gradient(135deg, #FFF9C4 0%, #FFF59D 100%); padding: 15px; border-radius: 8px;">

**Amaç:** Çoklu değişken korelasyonunu görmek

**YBS Soruları:**
- "Hangi değişkenler birbiriyle ilişkili?"
- "Boy-Kilo-Yaş arasında korelasyon var mı?"
- "Multicollinearity (çoklu doğrusal bağlantı) var mı?"

**Ne Görürüz:**
- ✅ +1: Tam pozitif korelasyon
- ✅  0: İlişki yok
- ✅ -1: Tam negatif korelasyon
- ✅ Renk yoğunluğu = İlişki gücü

</div>

````python
# Örnek Veri: Sporcu Özellikleri
np.random.seed(42)
n = 200

boy = np.random.normal(175, 10, n)
kilo = 0.6 * boy + np.random.normal(0, 5, n)  # Boy ile pozitif ilişki
yas = np.random.uniform(18, 35, n)
antrenman_saat = np.random.normal(25, 8, n)
performans = 0.3 * antrenman_saat - 0.2 * yas + np.random.normal(0, 3, n)

# DataFrame oluştur
veri_dict = {
    'Boy (cm)': boy,
    'Kilo (kg)': kilo,
    'Yas': yas,
    'Antrenman/Hafta': antrenman_saat,
    'Performans': performans
}

# Korelasyon matrisini hesapla
degiskenler = list(veri_dict.keys())
korelasyon_matrisi = np.zeros((len(degiskenler), len(degiskenler)))

for i, var1 in enumerate(degiskenler):
    for j, var2 in enumerate(degiskenler):
        korelasyon_matrisi[i, j] = np.corrcoef(veri_dict[var1], veri_dict[var2])[0, 1]

# Heatmap çiz
plt.figure(figsize=(10, 8))
im = plt.imshow(korelasyon_matrisi, cmap='RdYlGn', aspect='auto', vmin=-1, vmax=1)

# Eksen etiketleri
plt.xticks(np.arange(len(degiskenler)), degiskenler, rotation=45, ha='right', fontsize=11)
plt.yticks(np.arange(len(degiskenler)), degiskenler, fontsize=11)

# Hücrelere korelasyon değerlerini yaz
for i in range(len(degiskenler)):
    for j in range(len(degiskenler)):
        value = korelasyon_matrisi[i, j]
        # Renk kontrastı için metin rengi seç
        text_color = 'white' if abs(value) > 0.5 else 'black'
        plt.text(j, i, f'{value:.2f}', ha='center', va='center',
                fontsize=12, fontweight='bold', color=text_color)

# Colorbar
cbar = plt.colorbar(im, fraction=0.046, pad=0.04)
cbar.set_label('Korelasyon Katsayisi', fontsize=11, fontweight='bold')
cbar.ax.tick_params(labelsize=10)

# Başlık
plt.title('Heatmap: Sporcu Ozellikleri Korelasyon Matrisi\nRenk: Yesil=Pozitif, Kirmizi=Negatif', 
          fontsize=14, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()

# En yüksek korelasyonları bul (köşegen hariç)
print("📊 En Yüksek Korelasyonlar (|r| > 0.3):\n")
for i in range(len(degiskenler)):
    for j in range(i+1, len(degiskenler)):
        r = korelasyon_matrisi[i, j]
        if abs(r) > 0.3:
            yon = "Pozitif" if r > 0 else "Negatif"
            print(f"✅ {degiskenler[i]:20} ↔ {degiskenler[j]:20} → r={r:.3f} ({yon})")

print("\n💡 YBS Yorumu:")
print("   - Boy-Kilo arasında kuvvetli pozitif ilişki (beklenen)")
print("   - Antrenman-Performans arasında pozitif ilişki (iyi haber!)")
print("   - Yaş-Performans arasında negatif ilişki (dikkat!)")
````

````python
# Örnek Veri: Cinsiyet Bazlı Boy Dağılımı
np.random.seed(42)
erkek_boy = np.random.normal(178, 8, 150)
kadin_boy = np.random.normal(165, 7, 150)

veriler = [erkek_boy, kadin_boy]
etiketler = ['Erkek', 'Kadin']

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# SOL: Violin Plot
vp = axes[0].violinplot(veriler, positions=[1, 2], showmeans=True, showmedians=True, 
                        widths=0.7)

# Violin renklerini özelleştir
colors = ['#42A5F5', '#EC407A']
for i, pc in enumerate(vp['bodies']):
    pc.set_facecolor(colors[i])
    pc.set_alpha(0.7)
    pc.set_edgecolor('black')
    pc.set_linewidth(2)

# Diğer elemanların renklerini ayarla
vp['cbars'].set_edgecolor('black')
vp['cmaxes'].set_edgecolor('black')
vp['cmins'].set_edgecolor('black')
vp['cmedians'].set_edgecolor('#E53935')
vp['cmedians'].set_linewidth(3)
vp['cmeans'].set_edgecolor('#FF9800')
vp['cmeans'].set_linewidth(2)

axes[0].set_title('Cinsiyet Bazli Boy Dagilimi (Violin Plot)', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Boy (cm)', fontsize=11, fontweight='bold')
axes[0].set_xticks([1, 2])
axes[0].set_xticklabels(etiketler, fontsize=12, fontweight='bold')
axes[0].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[0].set_facecolor('#FAFAFA')

# Legend ekle
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='#E53935', lw=3, label='Medyan'),
    Line2D([0], [0], color='#FF9800', lw=2, label='Ortalama'),
]
axes[0].legend(handles=legend_elements, loc='upper right', fontsize=10)

# SAĞ: Violin + Box Plot Karşılaştırması
# Violin
vp2 = axes[1].violinplot(veriler, positions=[0.8, 1.8], showmeans=False, showmedians=False, 
                         widths=0.5)
for i, pc in enumerate(vp2['bodies']):
    pc.set_facecolor(colors[i])
    pc.set_alpha(0.4)
    pc.set_edgecolor('none')

# Box Plot (üstüne)
bp = axes[1].boxplot(veriler, positions=[1, 2], widths=0.2, patch_artist=True,
                     boxprops=dict(facecolor='white', linewidth=2, edgecolor='black'),
                     whiskerprops=dict(linewidth=2, color='black'),
                     capprops=dict(linewidth=2, color='black'),
                     medianprops=dict(linewidth=3, color='#E53935'),
                     showfliers=False)

axes[1].set_title('Violin + Box Plot Kombinasyonu', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Boy (cm)', fontsize=11, fontweight='bold')
axes[1].set_xticks([1, 2])
axes[1].set_xticklabels(etiketler, fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[1].set_facecolor('#FAFAFA')

plt.suptitle('Violin Plot: Dagilim Sekli + Merkezi Egilim', 
             fontsize=15, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()

# İstatistikler
print("📊 Cinsiyet Bazlı Boy İstatistikleri:\n")
for i, etiket in enumerate(etiketler):
    data = veriler[i]
    print(f"{etiket:6} → Ortalama: {data.mean():.1f} cm | Medyan: {np.median(data):.1f} cm | Std: {data.std():.1f} cm")

fark = erkek_boy.mean() - kadin_boy.mean()
print(f"\n✅ Ortalama fark: {fark:.1f} cm")
print("💡 YBS Yorumu: Her iki cinsiyet de normal dağılım gösteriyor, erkekler ortalama 13 cm daha uzun")
````

````python
# Örnek Veri: Kategori Bazlı Aylık Satış Payı
np.random.seed(42)
aylar = np.arange(1, 13)
ay_isimleri = ['Oca', 'Sub', 'Mar', 'Nis', 'May', 'Haz',
               'Tem', 'Agu', 'Eyl', 'Eki', 'Kas', 'Ara']

# Üç ürün kategorisi
elektronik = 30 + 3 * aylar + np.random.normal(0, 3, 12)
giyim = 25 + 1.5 * aylar + np.random.normal(0, 3, 12)
gida = 20 + 0.5 * aylar + np.random.normal(0, 2, 12)

fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# ÜST: Stacked Area Plot
axes[0].fill_between(aylar, 0, elektronik, 
                      label='Elektronik', alpha=0.7, color='#42A5F5', 
                      edgecolor='#0D47A1', linewidth=2)
axes[0].fill_between(aylar, elektronik, elektronik + giyim, 
                      label='Giyim', alpha=0.7, color='#66BB6A', 
                      edgecolor='#1B5E20', linewidth=2)
axes[0].fill_between(aylar, elektronik + giyim, elektronik + giyim + gida, 
                      label='Gida', alpha=0.7, color='#FFA726', 
                      edgecolor='#E65100', linewidth=2)

axes[0].set_title('Yigilmali Alan Grafik: Kategori Bazli Aylik Satis', 
                  fontsize=13, fontweight='bold')
axes[0].set_xlabel('Ay', fontsize=11, fontweight='bold')
axes[0].set_ylabel('Satis (Bin TL)', fontsize=11, fontweight='bold')
axes[0].set_xticks(aylar)
axes[0].set_xticklabels(ay_isimleri)
axes[0].legend(loc='upper left', fontsize=11, framealpha=0.95)
axes[0].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[0].set_facecolor('#FAFAFA')

# ALTI: Yüzde Stacked Area Plot
toplam = elektronik + giyim + gida
elektronik_yuzde = (elektronik / toplam) * 100
giyim_yuzde = (giyim / toplam) * 100
gida_yuzde = (gida / toplam) * 100

axes[1].fill_between(aylar, 0, elektronik_yuzde, 
                      label='Elektronik', alpha=0.7, color='#42A5F5', 
                      edgecolor='#0D47A1', linewidth=2)
axes[1].fill_between(aylar, elektronik_yuzde, elektronik_yuzde + giyim_yuzde, 
                      label='Giyim', alpha=0.7, color='#66BB6A', 
                      edgecolor='#1B5E20', linewidth=2)
axes[1].fill_between(aylar, elektronik_yuzde + giyim_yuzde, 100, 
                      label='Gida', alpha=0.7, color='#FFA726', 
                      edgecolor='#E65100', linewidth=2)

axes[1].set_title('Yuzde Bazli Alan Grafik: Pazar Payi Degisimi', 
                  fontsize=13, fontweight='bold')
axes[1].set_xlabel('Ay', fontsize=11, fontweight='bold')
axes[1].set_ylabel('Pazar Payi (%)', fontsize=11, fontweight='bold')
axes[1].set_xticks(aylar)
axes[1].set_xticklabels(ay_isimleri)
axes[1].set_ylim(0, 100)
axes[1].legend(loc='upper left', fontsize=11, framealpha=0.95)
axes[1].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[1].set_facecolor('#FAFAFA')

plt.suptitle('Area Plot: Kategori Bazli Zaman Serisi Analizi', 
             fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

# İstatistikler
print("📊 Kategori Bazlı Yıllık Analiz:\n")
print(f"Elektronik → Toplam: {elektronik.sum():.0f} Bin TL | Ortalama Pay: %{elektronik_yuzde.mean():.1f}")
print(f"Giyim      → Toplam: {giyim.sum():.0f} Bin TL | Ortalama Pay: %{giyim_yuzde.mean():.1f}")
print(f"Gida       → Toplam: {gida.sum():.0f} Bin TL | Ortalama Pay: %{gida_yuzde.mean():.1f}")
print(f"\n🎯 Toplam Satış: {toplam.sum():.0f} Bin TL")
print("\n💡 YBS Yorumu: Elektronik baskın kategori, pazar payı stabil")
````

## 🔟 **Stem Plot (Dal Grafik)**

### 📊 **Ne Zaman Kullanılır?**

<div style="background: linear-gradient(135deg, #FFEBEE 0%, #FFCDD2 100%); padding: 15px; border-radius: 8px;">

**Amaç:** Ayrık (discrete) veri noktalarını görmek

**YBS Soruları:**
- "Günlük hata sayıları nasıl?"
- "Haftalık toplantı sayısı nasıl seyrediyor?"
- "Event'ler hangi günlerde yoğunlaşıyor?"

**Ne Görürüz:**
- ✅ Her veri noktası net görülür
- ✅ Ani değişimler
- ✅ Sıfır çizgisinden mesafe
- ✅ Ayrık olaylar

**💡 Fark:** Line plot yerine ayrık olaylar için ideal!

</div>

````python
# Örnek Veri: Günlük Sistem Hata Sayısı (20 iş günü)
np.random.seed(42)
gunler = np.arange(1, 21)
hata_sayisi = np.random.poisson(3, 20)  # Poisson dağılımı (ayrık olaylar için)
hata_sayisi[7] = 12  # 8. günde anormal yüksek hata
hata_sayisi[14] = 11  # 15. günde anormal yüksek hata

fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# ÜST: Stem Plot
markerline, stemlines, baseline = axes[0].stem(gunler, hata_sayisi, 
                                                linefmt='C3-', markerfmt='C3o', basefmt='k-')

# Stem çizgilerini kalınlaştır
plt.setp(stemlines, linewidth=2.5)
plt.setp(markerline, markersize=10, markeredgecolor='black', markeredgewidth=1.5)
plt.setp(baseline, linewidth=2, color='black')

# Kritik hataları vurgula (>8)
kritik_gunler = gunler[hata_sayisi > 8]
kritik_hatalar = hata_sayisi[hata_sayisi > 8]
axes[0].scatter(kritik_gunler, kritik_hatalar, s=400, 
                color='#E53935', marker='*', zorder=5, 
                edgecolors='black', linewidth=2, label='Kritik Hata (>8)')

# Ortalama çizgisi
ortalama = hata_sayisi.mean()
axes[0].axhline(ortalama, color='#4CAF50', linestyle='--', linewidth=3, 
                label=f'Ortalama: {ortalama:.1f}')

# Eşik çizgisi
esik = 8
axes[0].axhline(esik, color='#FF9800', linestyle=':', linewidth=3, 
                label=f'Esik Deger: {esik}')

axes[0].set_title('Gunluk Sistem Hata Sayisi (Stem Plot)', 
                  fontsize=13, fontweight='bold')
axes[0].set_xlabel('Is Gunu', fontsize=11, fontweight='bold')
axes[0].set_ylabel('Hata Sayisi', fontsize=11, fontweight='bold')
axes[0].legend(loc='upper left', fontsize=11, framealpha=0.95)
axes[0].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[0].set_facecolor('#FAFAFA')
axes[0].set_xlim(0, 21)

# ALT: Stem Plot vs Line Plot Karşılaştırması
# Stem
markerline2, stemlines2, baseline2 = axes[1].stem(gunler, hata_sayisi, 
                                                   linefmt='C0-', markerfmt='C0o', basefmt='k-',
                                                   label='Stem Plot (Ayrik)')
plt.setp(stemlines2, linewidth=2, alpha=0.7)
plt.setp(markerline2, markersize=8)

# Line Plot (üstüne)
axes[1].plot(gunler, hata_sayisi, 'r--', linewidth=2.5, alpha=0.6, 
             marker='s', markersize=6, label='Line Plot (Surekli)')

axes[1].axhline(ortalama, color='#4CAF50', linestyle='--', linewidth=2, alpha=0.5)

axes[1].set_title('Stem Plot vs Line Plot Karsilastirmasi', 
                  fontsize=13, fontweight='bold')
axes[1].set_xlabel('Is Gunu', fontsize=11, fontweight='bold')
axes[1].set_ylabel('Hata Sayisi', fontsize=11, fontweight='bold')
axes[1].legend(loc='upper left', fontsize=11, framealpha=0.95)
axes[1].grid(True, alpha=0.3, axis='y', linestyle='--')
axes[1].set_facecolor('#FAFAFA')
axes[1].set_xlim(0, 21)

plt.suptitle('Stem Plot: Ayrik Olaylar ve Event Gorsellestirme', 
             fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

# İstatistikler
print("📊 Hata Analizi (20 İş Günü):\n")
print(f"✅ Toplam hata: {hata_sayisi.sum()}")
print(f"✅ Ortalama: {hata_sayisi.mean():.2f} hata/gün")
print(f"✅ En az: {hata_sayisi.min()} hata (Gün {gunler[np.argmin(hata_sayisi)]})")
print(f"✅ En çok: {hata_sayisi.max()} hata (Gün {gunler[np.argmax(hata_sayisi)]})")
print(f"✅ Eşiği aşan gün sayısı: {len(kritik_gunler)}")
print(f"✅ Kritik günler: {list(kritik_gunler)}")
print("\n💡 YBS Yorumu: 8. ve 15. günlerde anormal yüksek hata → Kök neden analizi gerekli!")
````

````python
# Olimpiyat Verisini yükleyelim
# Gerçek Kaggle dataseti: 120 years of Olympic History

url = 'https://raw.githubusercontent.com/rgriff23/Olympic_history/master/data/athlete_events.csv'

print("📥 Veri yükleniyor...")
try:
    df_olimpiyat = pd.read_csv(url)
    print("✅ Veri başarıyla yüklendi!")
except Exception as e:
    print(f"❌ Hata: {e}")
    print("⚠️ Alternatif: Manuel veri oluşturuyorum...")
    
    # Alternatif: Gerçekçi sentetik veri
    np.random.seed(42)
    n_sporcu = 5000
    
    sporlar = ['Basketbol', 'Voleybol', 'Yuzme', 'Atletizm', 'Futbol', 
               'Hentbol', 'Jimnastik', 'Halter', 'Gures', 'Kayak']
    cinsiyetler = ['Erkek', 'Kadin']
    
    df_olimpiyat = pd.DataFrame({
        'Name': [f'Sporcu_{i}' for i in range(n_sporcu)],
        'Sex': np.random.choice(cinsiyetler, n_sporcu, p=[0.55, 0.45]),
        'Age': np.random.randint(15, 40, n_sporcu),
        'Height': np.random.normal(175, 12, n_sporcu),
        'Weight': np.random.normal(72, 15, n_sporcu),
        'Sport': np.random.choice(sporlar, n_sporcu),
        'Medal': np.random.choice(['Gold', 'Silver', 'Bronze', np.nan], n_sporcu, p=[0.05, 0.08, 0.12, 0.75])
    })
    
    # Cinsiyet bazlı boy-kilo ayarlaması
    df_olimpiyat.loc[df_olimpiyat['Sex'] == 'Erkek', 'Height'] += 10
    df_olimpiyat.loc[df_olimpiyat['Sex'] == 'Erkek', 'Weight'] += 12
    df_olimpiyat.loc[df_olimpiyat['Sex'] == 'Kadin', 'Height'] -= 5
    
    # Spor dalına göre profil ayarlaması
    df_olimpiyat.loc[df_olimpiyat['Sport'] == 'Basketbol', 'Height'] += 15
    df_olimpiyat.loc[df_olimpiyat['Sport'] == 'Jimnastik', 'Height'] -= 10
    df_olimpiyat.loc[df_olimpiyat['Sport'] == 'Halter', 'Weight'] += 20
    
    print("✅ Sentetik veri oluşturuldu!")

# İlk bakış
print(f"\n📊 Veri Boyutu: {df_olimpiyat.shape[0]:,} satır x {df_olimpiyat.shape[1]} sütun")
print("\n📋 İlk 5 satır:")
df_olimpiyat.head()
````

## 📊 **ANALİZ 1: Boy-Kilo İlişkisi (Scatter Plot)**

### 🎯 **İş Sorusu: "Sporcularımızın boy-kilo profili sağlıklı mı? Cinsiyet farkı var mı?"**

````python
# Boy-Kilo Scatter Plot (Cinsiyet Bazlı)
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# SOL: Tüm Veri
for cinsiyet, color, marker in [('M', '#2196F3', 'o'), ('F', '#EC407A', '^')]:
    if 'M' in df_temiz['Sex'].values or 'F' in df_temiz['Sex'].values:
        # Gerçek veri için M/F
        mask = df_temiz['Sex'] == cinsiyet
        label = 'Erkek' if cinsiyet == 'M' else 'Kadin'
    else:
        # Sentetik veri için Erkek/Kadin
        cinsiyet_tr = 'Erkek' if cinsiyet == 'M' else 'Kadin'
        mask = df_temiz['Sex'] == cinsiyet_tr
        label = cinsiyet_tr
    
    if mask.sum() > 0:
        axes[0].scatter(df_temiz[mask]['Height'], 
                       df_temiz[mask]['Weight'],
                       s=20, alpha=0.4, c=color, marker=marker, 
                       edgecolors='none', label=label)

axes[0].set_title('Boy-Kilo Iliskisi (Cinsiyet Bazli)\nTum Sporcular', 
                  fontsize=14, fontweight='bold', pad=15)
axes[0].set_xlabel('Boy (cm)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Kilo (kg)', fontsize=12, fontweight='bold')
axes[0].legend(fontsize=11, loc='upper left', framealpha=0.9)
axes[0].grid(True, alpha=0.3, linestyle='--')
axes[0].set_facecolor('#FAFAFA')

# Korelasyon hesapla
corr = df_temiz[['Height', 'Weight']].corr().iloc[0, 1]
axes[0].text(0.98, 0.02, f'Korelasyon: r={corr:.3f}', 
            transform=axes[0].transAxes, fontsize=11, 
            ha='right', va='bottom', 
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

# SAĞ: Sample (daha net görünüm için)
# Her cinsiyetten 500 örnek alalım
sample_size = min(500, len(df_temiz) // 2)

for cinsiyet, color, marker in [('M', '#2196F3', 'o'), ('F', '#EC407A', '^')]:
    if 'M' in df_temiz['Sex'].values or 'F' in df_temiz['Sex'].values:
        mask = df_temiz['Sex'] == cinsiyet
        label = 'Erkek' if cinsiyet == 'M' else 'Kadin'
    else:
        cinsiyet_tr = 'Erkek' if cinsiyet == 'M' else 'Kadin'
        mask = df_temiz['Sex'] == cinsiyet_tr
        label = cinsiyet_tr
    
    if mask.sum() > 0:
        sample = df_temiz[mask].sample(n=min(sample_size, mask.sum()), random_state=42)
        axes[1].scatter(sample['Height'], sample['Weight'],
                       s=80, alpha=0.6, c=color, marker=marker, 
                       edgecolors='black', linewidth=0.5, label=label)

# Trend çizgileri ekle
for cinsiyet, color in [('M', '#2196F3'), ('F', '#EC407A')]:
    if 'M' in df_temiz['Sex'].values or 'F' in df_temiz['Sex'].values:
        mask = df_temiz['Sex'] == cinsiyet
    else:
        cinsiyet_tr = 'Erkek' if cinsiyet == 'M' else 'Kadin'
        mask = df_temiz['Sex'] == cinsiyet_tr
    
    if mask.sum() > 0:
        x = df_temiz[mask]['Height'].values
        y = df_temiz[mask]['Weight'].values
        z = np.polyfit(x, y, 1)
        p = np.poly1d(z)
        x_line = np.linspace(x.min(), x.max(), 100)
        axes[1].plot(x_line, p(x_line), '--', color=color, linewidth=2.5, alpha=0.7)

axes[1].set_title(f'Boy-Kilo Iliskisi (Sample: {sample_size} x 2)\nDetayli Gorunum', 
                  fontsize=14, fontweight='bold', pad=15)
axes[1].set_xlabel('Boy (cm)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Kilo (kg)', fontsize=12, fontweight='bold')
axes[1].legend(fontsize=11, loc='upper left', framealpha=0.9)
axes[1].grid(True, alpha=0.3, linestyle='--')
axes[1].set_facecolor('#FAFAFA')

plt.suptitle('Sporcu Fiziksel Profil Analizi: Boy-Kilo Korelasyonu', 
             fontsize=16, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()

# İstatistikler
print("📊 Boy-Kilo Analizi:\n")
for cinsiyet in df_temiz['Sex'].unique():
    data = df_temiz[df_temiz['Sex'] == cinsiyet]
    corr_c = data[['Height', 'Weight']].corr().iloc[0, 1]
    print(f"{cinsiyet:6} → Ortalama Boy: {data['Height'].mean():.1f} cm | "
          f"Ortalama Kilo: {data['Weight'].mean():.1f} kg | "
          f"Korelasyon: {corr_c:.3f}")

print("\n💡 YBS Yorumu:")
print("   ✅ Boy-Kilo arasında güçlü pozitif ilişki var")
print("   ✅ Erkek sporcular ortalama 12-15 cm daha uzun ve 10-15 kg daha ağır")
print("   ✅ Her iki cinsiyet de sağlıklı vücut profili gösteriyor")
````

## 📊 **ANALİZ 2: Yaş Dağılımı (Histogram + Box Plot)**

### 🎯 **İş Sorusu: "Sporcularımızın yaş profili nasıl? Hangi yaş grubunda yoğunlaşıyoruz?"**

````python
# Yaş Dağılımı Analizi
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# ÜST SOL: Genel Histogram
axes[0, 0].hist(df_temiz['Age'], bins=30, color='#4CAF50', alpha=0.7, 
                edgecolor='#1B5E20', linewidth=1.5)

ortalama_yas = df_temiz['Age'].mean()
medyan_yas = df_temiz['Age'].median()

axes[0, 0].axvline(ortalama_yas, color='#E53935', linestyle='--', linewidth=3, 
                   label=f'Ortalama: {ortalama_yas:.1f}')
axes[0, 0].axvline(medyan_yas, color='#FF9800', linestyle=':', linewidth=3, 
                   label=f'Medyan: {medyan_yas:.1f}')

axes[0, 0].set_title('Tum Sporcularin Yas Dagilimi', fontsize=13, fontweight='bold')
axes[0, 0].set_xlabel('Yas', fontsize=11)
axes[0, 0].set_ylabel('Sporcu Sayisi', fontsize=11)
axes[0, 0].legend(fontsize=10)
axes[0, 0].grid(True, alpha=0.3, axis='y')
axes[0, 0].set_facecolor('#FAFAFA')

# ÜST SAĞ: Cinsiyet Bazlı Histogram
for cinsiyet, color, label in [('M', '#2196F3', 'Erkek'), ('F', '#EC407A', 'Kadin')]:
    if 'M' in df_temiz['Sex'].values or 'F' in df_temiz['Sex'].values:
        mask = df_temiz['Sex'] == cinsiyet
    else:
        cinsiyet_tr = 'Erkek' if cinsiyet == 'M' else 'Kadin'
        mask = df_temiz['Sex'] == cinsiyet_tr
    
    if mask.sum() > 0:
        axes[0, 1].hist(df_temiz[mask]['Age'], bins=25, alpha=0.6, 
                       color=color, label=label, edgecolor='black', linewidth=1)

axes[0, 1].set_title('Cinsiyet Bazli Yas Dagilimi', fontsize=13, fontweight='bold')
axes[0, 1].set_xlabel('Yas', fontsize=11)
axes[0, 1].set_ylabel('Sporcu Sayisi', fontsize=11)
axes[0, 1].legend(fontsize=10)
axes[0, 1].grid(True, alpha=0.3, axis='y')
axes[0, 1].set_facecolor('#FAFAFA')

# ALT SOL: Box Plot (Cinsiyet Bazlı)
yas_data = []
labels_yas = []
for cinsiyet in df_temiz['Sex'].unique():
    yas_data.append(df_temiz[df_temiz['Sex'] == cinsiyet]['Age'].values)
    labels_yas.append(cinsiyet)

bp = axes[1, 0].boxplot(yas_data, labels=labels_yas, patch_artist=True, 
                        notch=True, showmeans=True,
                        boxprops=dict(facecolor='#BBDEFB', linewidth=2, edgecolor='#1976D2'),
                        whiskerprops=dict(linewidth=2, color='#0D47A1'),
                        capprops=dict(linewidth=2, color='#0D47A1'),
                        medianprops=dict(linewidth=3, color='#E53935'),
                        meanprops=dict(marker='D', markerfacecolor='#FF9800', markersize=8))

axes[1, 0].set_title('Cinsiyet Bazli Yas Karsilastirmasi (Box Plot)', 
                     fontsize=13, fontweight='bold')
axes[1, 0].set_ylabel('Yas', fontsize=11)
axes[1, 0].grid(True, alpha=0.3, axis='y')
axes[1, 0].set_facecolor('#FAFAFA')
axes[1, 0].axhline(ortalama_yas, color='#4CAF50', linestyle='--', linewidth=2, 
                   alpha=0.5, label=f'Genel Ort: {ortalama_yas:.1f}')
axes[1, 0].legend(fontsize=9)

# ALT SAĞ: Yaş Kategorileri
yas_kategorileri = pd.cut(df_temiz['Age'], 
                          bins=[0, 20, 25, 30, 35, 100],
                          labels=['15-20', '21-25', '26-30', '31-35', '36+'])

yas_kategori_counts = yas_kategorileri.value_counts().sort_index()

bars = axes[1, 1].bar(range(len(yas_kategori_counts)), yas_kategori_counts.values,
                      color=['#FF9800', '#4CAF50', '#2196F3', '#9C27B0', '#F44336'],
                      alpha=0.8, edgecolor='black', linewidth=2)

axes[1, 1].set_xticks(range(len(yas_kategori_counts)))
axes[1, 1].set_xticklabels(yas_kategori_counts.index, fontsize=11)
axes[1, 1].set_title('Yas Kategorileri Dagilimi', fontsize=13, fontweight='bold')
axes[1, 1].set_xlabel('Yas Grubu', fontsize=11)
axes[1, 1].set_ylabel('Sporcu Sayisi', fontsize=11)
axes[1, 1].grid(True, alpha=0.3, axis='y')
axes[1, 1].set_facecolor('#FAFAFA')

# Değer etiketleri
for i, bar in enumerate(bars):
    height = bar.get_height()
    yuzde = (height / len(df_temiz)) * 100
    axes[1, 1].text(bar.get_x() + bar.get_width()/2., height,
                   f'{int(height)}\n(%{yuzde:.1f})',
                   ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.suptitle('Sporcu Yas Profili Analizi: Dagilim ve Karsilastirma', 
             fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

# İstatistikler
print("📊 Yaş Analizi:\n")
print(f"✅ Genel Ortalama: {ortalama_yas:.1f} yaş")
print(f"✅ Medyan: {medyan_yas:.0f} yaş")
print(f"✅ En genç: {df_temiz['Age'].min():.0f} yaş")
print(f"✅ En yaşlı: {df_temiz['Age'].max():.0f} yaş")
print(f"✅ Standart sapma: {df_temiz['Age'].std():.1f}")

print("\n📊 Cinsiyet Bazlı:")
for cinsiyet in df_temiz['Sex'].unique():
    data = df_temiz[df_temiz['Sex'] == cinsiyet]['Age']
    print(f"{cinsiyet:6} → Ortalama: {data.mean():.1f} | Medyan: {data.median():.0f}")

print("\n📊 Yaş Kategori Dağılımı:")
for kat, sayi in yas_kategori_counts.items():
    yuzde = (sayi / len(df_temiz)) * 100
    print(f"{kat:6} → {sayi:,} sporcu (%{yuzde:.1f})")

print("\n💡 YBS Yorumu:")
print("   ✅ Sporcuların çoğunluğu 21-30 yaş aralığında (prime age)")
print("   ✅ Normal dağılım gösteriyor → Dengeli yaş yapısı")
print("   ✅ 36+ yaş grubunda da aktif sporcular var")
````

## 📊 **ANALİZ 4: Madalya Analizi (Pie + Bar)**

### 🎯 **İş Sorusu: "Madalya dağılımı nasıl? Hangi sporlar daha başarılı?"**

````python
# Madalya verisi olanları filtrele
df_madalyali = df_temiz[df_temiz['Medal'].notna()].copy()

print(f"🏅 Madalya Analizi:")
print(f"   Toplam sporcu: {len(df_temiz):,}")
print(f"   Madalya kazanan: {len(df_madalyali):,}")
print(f"   Başarı oranı: %{(len(df_madalyali)/len(df_temiz)*100):.1f}")

fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# ÜST SOL: Madalya Türü Pie Chart
madalya_counts = df_madalyali['Medal'].value_counts()
colors_medal = {'Gold': '#FFD700', 'Silver': '#C0C0C0', 'Bronze': '#CD7F32'}
colors = [colors_medal.get(m, '#90CAF9') for m in madalya_counts.index]
explode = [0.05 if m == 'Gold' else 0 for m in madalya_counts.index]

axes[0, 0].pie(madalya_counts.values, labels=madalya_counts.index, autopct='%1.1f%%',
              colors=colors, startangle=90, explode=explode,
              textprops={'fontsize': 12, 'fontweight': 'bold'},
              wedgeprops={'edgecolor': 'black', 'linewidth': 2})
axes[0, 0].set_title('Madalya Turu Dagilimi', fontsize=13, fontweight='bold')

# ÜST SAĞ: Top 10 Spor - Madalya Sayısı
top_10_madalya_spor = df_madalyali['Sport'].value_counts().head(10)
bars_medal = axes[0, 1].barh(range(len(top_10_madalya_spor)), top_10_madalya_spor.values,
                             color=plt.cm.Spectral(np.linspace(0.2, 0.8, 10)),
                             alpha=0.8, edgecolor='black', linewidth=2)

axes[0, 1].set_yticks(range(len(top_10_madalya_spor)))
axes[0, 1].set_yticklabels(top_10_madalya_spor.index, fontsize=10)
axes[0, 1].set_xlabel('Madalya Sayisi', fontsize=11, fontweight='bold')
axes[0, 1].set_title('En Basarili 10 Spor Dali (Madalya Sayisi)', 
                     fontsize=13, fontweight='bold')
axes[0, 1].grid(True, alpha=0.3, axis='x')
axes[0, 1].set_facecolor('#FAFAFA')

# Değer etiketleri
for i, (bar, value) in enumerate(zip(bars_medal, top_10_madalya_spor.values)):
    axes[0, 1].text(value + 10, bar.get_y() + bar.get_height()/2.,
                   f'{int(value)}', ha='left', va='center', 
                   fontsize=10, fontweight='bold')

# ALT SOL: Madalya Türü - Stacked Bar
spor_madalya_pivot = pd.crosstab(df_madalyali['Sport'], df_madalyali['Medal'])
top_5_sporlar = df_madalyali['Sport'].value_counts().head(5).index
spor_madalya_top5 = spor_madalya_pivot.loc[top_5_sporlar]

x_pos = np.arange(len(top_5_sporlar))
bottom_vals = np.zeros(len(top_5_sporlar))

for medal, color in [('Gold', '#FFD700'), ('Silver', '#C0C0C0'), ('Bronze', '#CD7F32')]:
    if medal in spor_madalya_top5.columns:
        values = spor_madalya_top5[medal].values
        axes[1, 0].bar(x_pos, values, bottom=bottom_vals, label=medal,
                      color=color, alpha=0.9, edgecolor='black', linewidth=1.5)
        bottom_vals += values

axes[1, 0].set_xticks(x_pos)
axes[1, 0].set_xticklabels(top_5_sporlar, rotation=45, ha='right', fontsize=10)
axes[1, 0].set_ylabel('Madalya Sayisi', fontsize=11, fontweight='bold')
axes[1, 0].set_title('Top 5 Spor: Madalya Turlerine Gore Dagilim', 
                     fontsize=13, fontweight='bold')
axes[1, 0].legend(fontsize=10, loc='upper right')
axes[1, 0].grid(True, alpha=0.3, axis='y')
axes[1, 0].set_facecolor('#FAFAFA')

# ALT SAĞ: Cinsiyet Bazlı Madalya
cinsiyet_madalya = pd.crosstab(df_madalyali['Sex'], df_madalyali['Medal'])

x_cinsiyet = np.arange(len(cinsiyet_madalya.index))
width_cinsiyet = 0.25

for i, (medal, color) in enumerate([('Gold', '#FFD700'), ('Silver', '#C0C0C0'), ('Bronze', '#CD7F32')]):
    if medal in cinsiyet_madalya.columns:
        axes[1, 1].bar(x_cinsiyet + i*width_cinsiyet, cinsiyet_madalya[medal].values,
                      width_cinsiyet, label=medal, color=color, 
                      alpha=0.9, edgecolor='black', linewidth=1.5)

axes[1, 1].set_xticks(x_cinsiyet + width_cinsiyet)
axes[1, 1].set_xticklabels(cinsiyet_madalya.index, fontsize=11, fontweight='bold')
axes[1, 1].set_ylabel('Madalya Sayisi', fontsize=11, fontweight='bold')
axes[1, 1].set_title('Cinsiyet Bazli Madalya Dagilimi', fontsize=13, fontweight='bold')
axes[1, 1].legend(fontsize=10, loc='upper right')
axes[1, 1].grid(True, alpha=0.3, axis='y')
axes[1, 1].set_facecolor('#FAFAFA')

plt.suptitle('Madalya Analizi: Basari Profili ve Dagilim', 
             fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

# İstatistikler
print("\n📊 Madalya İstatistikleri:\n")
print(f"{'Madalya Türü':15} {'Sayı':>10} {'Yüzde':>10}")
print("-" * 40)
for medal, count in madalya_counts.items():
    yuzde = (count / len(df_madalyali)) * 100
    print(f"{medal:15} {count:>10,} {yuzde:>9.1f}%")

print(f"\n🏅 En başarılı spor: {top_10_madalya_spor.index[0]} ({top_10_madalya_spor.values[0]:,} madalya)")

print("\n💡 YBS Yorumu:")
print("   ✅ Altın madalya oranı dengeli")
print("   ✅ Bazı sporlar belirgin şekilde daha başarılı")
print("   ✅ Cinsiyet bazlı madalya dağılımı dengeli")
````
