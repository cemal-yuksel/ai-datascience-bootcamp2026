````python
# ============================================================
# ⚙️ ADIM 4: VERİ ÖN İŞLEME
# ============================================================

# ── 4.1 Özellik / Hedef Ayırma ──────────────────────────────
# X → Bağımsız değişkenler (özellikler / features)
# y → Bağımlı değişken (hedef / target)
X = df.drop(columns=['high_risk_flag'])   # Özellikleri al
y = df['high_risk_flag']                  # Hedefi al

print("=" * 55)
print("✂️  ÖZELLİK / HEDEF AYIRMA")
print("=" * 55)
print(f"  X (özellikler) boyutu: {X.shape}")
print(f"  y (hedef) boyutu     : {y.shape}")
print(f"  Özellik sütunları    : {list(X.columns)}")
print()

# ── 4.2 Train-Test Bölme ─────────────────────────────────────
# Stratify=y → Eğitim ve test setlerinde sınıf dağılımı korunur
# Bu özellikle dengesiz sınıflarda çok önemlidir!
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,          # %20 test, %80 eğitim
    random_state=RANDOM_STATE,
    stratify=y              # Sınıf dengesi korunuyor
)

print("=" * 55)
print("🔀 TRAIN-TEST BÖLME SONUÇLARI")
print("=" * 55)
print(f"  Eğitim seti boyutu  : {X_train.shape[0]:4d} kayıt (%{X_train.shape[0]/len(df)*100:.0f})")
print(f"  Test seti boyutu    : {X_test.shape[0]:4d} kayıt (%{X_test.shape[0]/len(df)*100:.0f})")
print()
print("  Eğitim seti sınıf dağılımı:")
train_dist = y_train.value_counts(normalize=True) * 100
print(f"    0 (Düşük Risk) : %{train_dist.get(0, 0):.1f}")
print(f"    1 (Yüksek Risk): %{train_dist.get(1, 0):.1f}")
print()
print("  Test seti sınıf dağılımı (stratify sayesinde benzer):")
test_dist = y_test.value_counts(normalize=True) * 100
print(f"    0 (Düşük Risk) : %{test_dist.get(0, 0):.1f}")
print(f"    1 (Yüksek Risk): %{test_dist.get(1, 0):.1f}")
print()

# ── 4.3 Ölçeklendirme (StandardScaler) ───────────────────────
# StandardScaler: Her özelliği ortalaması 0, std sapması 1 olacak şekilde dönüştürür
# Formül: z = (x - μ) / σ
# ÖNEMLI: Scaler sadece X_train üzerinde fit edilir!
# X_test'e sadece transform uygulanır (veri sızıntısı önlenir)

scaler = StandardScaler()

# Fit → Eğitim verisinden ortalama ve std hesapla
# Transform → Bu istatistikleri kullanarak dönüştür
X_train_sc = scaler.fit_transform(X_train)
X_test_sc  = scaler.transform(X_test)    # Sadece transform! (fit değil!)

print("=" * 55)
print("📏 STANDARTLAŞTIRMA (StandardScaler)")
print("=" * 55)
print("  Özellik Ortalamaları (Eğitim):")
for col, mean in zip(X.columns, scaler.mean_):
    print(f"    {col:35s}: {mean:+.4f}")
print()
print("  Özellik Std Sapmaları (Eğitim):")
for col, std in zip(X.columns, scaler.scale_):
    print(f"    {col:35s}: {std:.4f}")
print()
print("  ✅ X_train_sc ve X_test_sc hazır!")
print("  ✅ Veri sızıntısı (data leakage) önlendi!")
print()

# ── 4.4 Ölçeklendirme Öncesi/Sonrası Karşılaştırma ──────────
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
fig.patch.set_facecolor('#fafafa')

for i, col in enumerate(X.columns):
    # Ham veri (üst satır)
    axes[0, i].hist(X_train.iloc[:, i], bins=30, color='#c4b5fd',
                    edgecolor='white', alpha=0.8)
    axes[0, i].set_title(f'Ham: {col}', fontweight='bold', fontsize=11)
    axes[0, i].set_facecolor('#f8fafc')

    # Ölçeklendirilmiş veri (alt satır)
    axes[1, i].hist(X_train_sc[:, i], bins=30, color='#6ee7b7',
                    edgecolor='white', alpha=0.8)
    axes[1, i].set_title(f'Ölçeklendirilmiş: {col}', fontweight='bold', fontsize=11)
    axes[1, i].axvline(0, color='#dc2626', linestyle='--', linewidth=1.5, label='Ortalama=0')
    axes[1, i].set_facecolor('#f8fafc')
    axes[1, i].legend(fontsize=9)

plt.suptitle('⚙️ StandardScaler Öncesi vs Sonrası Dağılım',
             fontsize=14, fontweight='bold', y=1.01, color='#1a0030')
plt.tight_layout()
plt.show()
print("✅ Veri ön işleme tamamlandı!")
````

**❓ Baseline Model Neden Gereklidir?**

Makine öğrenmesi modelinin başarılı olup olmadığını anlayabilmek için bir **referans noktasına** ihtiyaç duyarız.

**Analoji:** Bir şirketin yeni işe aldığı bir analist, mevcut pratikler kadar iyi çalışabilmeli. Baseline model "en basit, en aptal tahmin stratejisi"nin performansını gösterir.

- Eğer modelimiz baseline'dan **daha iyiyse** → Model öğrenme yapmıştır ✅
- Eğer modelimiz baseline'a **eşit veya kötüyse** → Modelde bir sorun var ❌

</div>


````python
# ============================================================
# 🎯 ADIM 5: BASELINE MODEL
# ============================================================
# DummyClassifier → Herhangi bir öğrenme yapmadan tahmin eden model
# 3 farklı strateji deneyeceğiz:
#   1. most_frequent → Her zaman en sık görülen sınıfı tahmin et
#   2. stratified    → Sınıf olasılıklarına göre rastgele tahmin et
#   3. uniform       → Eşit olasılıkla rastgele tahmin et

# Sonuçları saklamak için liste
baseline_sonuclar = []

baseline_stratejileri = {
    'Most Frequent': DummyClassifier(strategy='most_frequent', random_state=RANDOM_STATE),
    'Stratified'   : DummyClassifier(strategy='stratified',    random_state=RANDOM_STATE),
    'Uniform'      : DummyClassifier(strategy='uniform',       random_state=RANDOM_STATE),
}

print("=" * 65)
print("🎯 BASELINE MODEL SONUÇLARI")
print("=" * 65)
print(f"{'Strateji':<20} {'Accuracy':>10} {'Precision':>10} {'Recall':>10} {'F1':>10}")
print("-" * 65)

for isim, model in baseline_stratejileri.items():
    model.fit(X_train_sc, y_train)    # Eğitim (sadece istatistik hesaplar)
    y_pred = model.predict(X_test_sc) # Tahmin

    acc  = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec  = recall_score(y_test, y_pred, zero_division=0)
    f1   = f1_score(y_test, y_pred, zero_division=0)

    baseline_sonuclar.append({
        'Model': f'Baseline ({isim})',
        'Accuracy': acc, 'Precision': prec,
        'Recall': rec, 'F1': f1
    })
    print(f"{isim:<20} {acc:>10.4f} {prec:>10.4f} {rec:>10.4f} {f1:>10.4f}")

print("-" * 65)
print()

# En iyi baseline accuracy'yi sakla (diğer modeller bunu geçmeli!)
BASELINE_ACC = max([s['Accuracy'] for s in baseline_sonuclar])
print(f"🏁 Geçilmesi gereken baseline accuracy: {BASELINE_ACC:.4f} ({BASELINE_ACC*100:.1f}%)")
print()
print("💡 YBS Notu: 'Sınıflandırma diyorum' deyip sınıfı")
print("   tahmin edemeksizin bu kadar başarı elde edebilirsiniz.")
print("   KNN modelimiz bunu kesinlikle geçmeliydi!")
````

````python
# ============================================================
# 🤖 ADIM 6: KNN CLASSIFICATION MODELİ (K=5 Başlangıç)
# ============================================================

# ── 6.1 Model Kurulum & Eğitim ───────────────────────────────
# K=5 ile KNN modeli - yaygın başlangıç noktası
knn_model = KNeighborsClassifier(
    n_neighbors=5,        # K değeri → kaç komşuya bakılacak
    metric='euclidean',   # Öklid mesafesi
    weights='uniform',    # Tüm komşulara eşit ağırlık
    n_jobs=-1             # Tüm CPU çekirdeklerini kullan
)

# Ölçeklendirilmiş eğitim verisiyle eğit
knn_model.fit(X_train_sc, y_train)

# Test verisi üzerinde tahmin
y_pred_knn = knn_model.predict(X_test_sc)
y_prob_knn = knn_model.predict_proba(X_test_sc)[:, 1]  # ROC-AUC için olasılıklar

# ── 6.2 Performans Metrikleri ────────────────────────────────
acc_knn  = accuracy_score(y_test, y_pred_knn)
prec_knn = precision_score(y_test, y_pred_knn)
rec_knn  = recall_score(y_test, y_pred_knn)
f1_knn   = f1_score(y_test, y_pred_knn)
roc_knn  = roc_auc_score(y_test, y_prob_knn)

print("=" * 60)
print("🤖 KNN MODEL SONUÇLARI (K=5)")
print("=" * 60)
print(f"  Accuracy  : {acc_knn:.4f}  ({acc_knn*100:.2f}%)")
print(f"  Precision : {prec_knn:.4f}  — Yüksek risk dediğimizin {prec_knn*100:.1f}%'i gerçekten yüksek")
print(f"  Recall    : {rec_knn:.4f}  — Gerçek vakaların {rec_knn*100:.1f}%'ini yakaladık")
print(f"  F1-Score  : {f1_knn:.4f}  — Precision & Recall dengesi")
print(f"  ROC-AUC   : {roc_knn:.4f}  — Ayrım gücü")
print()
print(f"  📈 Baseline ({BASELINE_ACC:.4f}) → KNN ({acc_knn:.4f})")
if acc_knn > BASELINE_ACC:
    artis = (acc_knn - BASELINE_ACC) * 100
    print(f"  ✅ Baseline'ı %{artis:.2f} puan geçti! Model öğrendi.")
else:
    print("  ❌ Baseline geçilemedi! Hiperparametre ayarı gerekli.")
print()

# ── 6.3 Sınıflandırma Raporu ─────────────────────────────────
print("📋 SINIFLANDIRMA RAPORU (Classification Report):")
print("=" * 60)
print(classification_report(
    y_test, y_pred_knn,
    target_names=['Düşük Risk (0)', 'Yüksek Risk (1)']
))

# ── 6.4 Görselleştirme: Confusion Matrix + ROC Curve ─────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.patch.set_facecolor('#fafafa')

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred_knn)
sns.heatmap(cm, annot=True, fmt='d', ax=axes[0],
            cmap='Purples',
            xticklabels=['Düşük Risk', 'Yüksek Risk'],
            yticklabels=['Düşük Risk', 'Yüksek Risk'],
            linewidths=2, linecolor='white',
            annot_kws={'size': 16, 'weight': 'bold'})
axes[0].set_title('Confusion Matrix — KNN (K=5)', fontweight='bold', fontsize=13, pad=15)
axes[0].set_xlabel('Tahmin Edilen Sınıf', fontsize=11)
axes[0].set_ylabel('Gerçek Sınıf', fontsize=11)

# Açıklama etiketleri
axes[0].text(0.5, -0.15,
             f"TP={cm[1,1]} | TN={cm[0,0]} | FP={cm[0,1]} | FN={cm[1,0]}",
             transform=axes[0].transAxes, ha='center', fontsize=10, color='#374151')

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_prob_knn)
axes[1].plot(fpr, tpr, color='#7c3aed', linewidth=2.5,
             label=f'KNN (AUC = {roc_knn:.3f})')
axes[1].plot([0, 1], [0, 1], 'k--', linewidth=1.5, alpha=0.5, label='Rastgele Tahmin')
axes[1].fill_between(fpr, tpr, alpha=0.15, color='#7c3aed')
axes[1].set_xlabel('False Positive Rate (1 - Özgüllük)', fontsize=11)
axes[1].set_ylabel('True Positive Rate (Duyarlılık)', fontsize=11)
axes[1].set_title('ROC Eğrisi — KNN (K=5)', fontweight='bold', fontsize=13, pad=15)
axes[1].legend(fontsize=11)
axes[1].set_facecolor('#f8fafc')
axes[1].grid(True, alpha=0.3)

plt.suptitle('KNN Model Değerlendirmesi', fontsize=15, fontweight='bold',
             y=1.02, color='#1a0030')
plt.tight_layout()
plt.show()
````

<div style="background:#fce7f3;border-left:5px solid #ec4899;border-radius:10px;padding:18px 22px;margin:12px 0;">

**🎯 Hedef:** Varsayılan K=5 yerine **optimum K değerini** bulmak.

**Yöntem 1 — Elbow (Dirsek) Yöntemi:** Farklı K değerleri için doğruluğu çizeriz, "dirsek" noktası optimal K'dır.

**Yöntem 2 — GridSearchCV:** Sistematik olarak tüm parametre kombinasyonlarını dener, çapraz doğrulama ile en iyiyi seçer.

</div>


````python
# ============================================================
# 🔧 ADIM 7: KNN HİPERPARAMETRE OPTİMİZASYONU
# ============================================================

# ── 7.1 Elbow Yöntemi: En İyi K ─────────────────────────────
# K=1'den K=31'e kadar her K için 5-katlı çapraz doğrulama yapıyoruz
k_aralik   = range(1, 32, 2)   # Tek sayılar: 1,3,5,...,31
train_accs = []
test_accs  = []
cv_accs    = []

print("K değerleri test ediliyor...")
for k in k_aralik:
    knn_k = KNeighborsClassifier(n_neighbors=k, n_jobs=-1)
    knn_k.fit(X_train_sc, y_train)

    train_accs.append(accuracy_score(y_train, knn_k.predict(X_train_sc)))
    test_accs.append(accuracy_score(y_test, knn_k.predict(X_test_sc)))

    # 5-katlı çapraz doğrulama (daha sağlıklı tahmin)
    cv_score = cross_val_score(knn_k, X_train_sc, y_train,
                               cv=5, scoring='accuracy', n_jobs=-1)
    cv_accs.append(cv_score.mean())

# En iyi K değerini bul
best_k_idx = np.argmax(cv_accs)
best_k     = list(k_aralik)[best_k_idx]
print(f"✅ Elbow yöntemine göre en iyi K: {best_k}  (CV Accuracy: {cv_accs[best_k_idx]:.4f})")
print()

# ── 7.2 Elbow Grafiği ────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
fig.patch.set_facecolor('#fafafa')

# Sol: K vs Accuracy (Elbow)
axes[0].plot(k_aralik, train_accs, 'o-', color='#c4b5fd',
             linewidth=2, markersize=5, label='Eğitim')
axes[0].plot(k_aralik, cv_accs, 's-', color='#7c3aed',
             linewidth=2.5, markersize=6, label='CV (5-fold)')
axes[0].plot(k_aralik, test_accs, '^-', color='#ec4899',
             linewidth=2, markersize=5, label='Test')
axes[0].axvline(best_k, color='#dc2626', linestyle='--',
                linewidth=2, label=f'Optimal K={best_k}')
axes[0].annotate(f'K={best_k}\n{cv_accs[best_k_idx]:.3f}',
                 xy=(best_k, cv_accs[best_k_idx]),
                 xytext=(best_k + 2, cv_accs[best_k_idx] - 0.02),
                 fontsize=10, fontweight='bold', color='#dc2626',
                 arrowprops=dict(arrowstyle='->', color='#dc2626', lw=1.5))
axes[0].set_xlabel('K Değeri (Komşu Sayısı)', fontsize=11)
axes[0].set_ylabel('Accuracy', fontsize=11)
axes[0].set_title('Elbow Yöntemi: K vs Accuracy', fontweight='bold', fontsize=13)
axes[0].legend(fontsize=10)
axes[0].set_facecolor('#f8fafc')
axes[0].grid(True, alpha=0.3)

# Sağ: Overfitting analizi (train vs test farkı)
fark = [t - cv for t, cv in zip(train_accs, cv_accs)]
renk = ['#dc2626' if f > 0.03 else '#059669' for f in fark]
bars = axes[1].bar(k_aralik, fark, color=renk, edgecolor='white', alpha=0.8)
axes[1].axhline(0.03, color='#f59e0b', linestyle='--', linewidth=2,
                label='Overfitting Eşiği (0.03)')
axes[1].set_xlabel('K Değeri', fontsize=11)
axes[1].set_ylabel('Eğitim - CV Accuracy Farkı', fontsize=11)
axes[1].set_title('Overfitting Analizi', fontweight='bold', fontsize=13)
axes[1].legend(fontsize=10)
axes[1].set_facecolor('#f8fafc')

plt.suptitle('KNN Hiperparametre Analizi', fontsize=15, fontweight='bold',
             y=1.02, color='#1a0030')
plt.tight_layout()
plt.show()

# ── 7.3 GridSearchCV ile Kapsamlı Optimizasyon ──────────────
print("=" * 60)
print("🔍 GridSearchCV ile Kapsamlı Arama...")
print("=" * 60)

# Denenecek parametre ızgarası
param_grid = {
    'n_neighbors': [3, 5, 7, 9, 11, 15],
    'weights'    : ['uniform', 'distance'],
    'metric'     : ['euclidean', 'manhattan'],
}

grid_search = GridSearchCV(
    KNeighborsClassifier(n_jobs=-1),
    param_grid,
    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE),
    scoring='f1',         # F1 skoru optimize ediyoruz
    n_jobs=-1,
    verbose=0
)

grid_search.fit(X_train_sc, y_train)

print(f"  ✅ En iyi parametreler: {grid_search.best_params_}")
print(f"  ✅ En iyi CV F1-Skoru : {grid_search.best_score_:.4f}")
print()

# En iyi modelle tahmin
knn_best     = grid_search.best_estimator_
y_pred_best  = knn_best.predict(X_test_sc)
y_prob_best  = knn_best.predict_proba(X_test_sc)[:, 1]

acc_best  = accuracy_score(y_test, y_pred_best)
f1_best   = f1_score(y_test, y_pred_best)
roc_best  = roc_auc_score(y_test, y_prob_best)

print("=" * 60)
print("📊 OPTİMİZE KNN KARŞILAŞTIRMASI")
print("=" * 60)
print(f"{'Metrik':<15} {'KNN (K=5)':>12} {'Optimize KNN':>14} {'Fark':>8}")
print("-" * 60)
print(f"{'Accuracy':<15} {acc_knn:>12.4f} {acc_best:>14.4f} {(acc_best-acc_knn)*100:>+7.2f}%")
print(f"{'F1-Score':<15} {f1_knn:>12.4f} {f1_best:>14.4f} {(f1_best-f1_knn)*100:>+7.2f}%")
print(f"{'ROC-AUC':<15} {roc_knn:>12.4f} {roc_best:>14.4f} {(roc_best-roc_knn)*100:>+7.2f}%")
print("-" * 60)

# Sonuçları ana listeye ekle
baseline_sonuclar.append({
    'Model': 'KNN (K=5)',
    'Accuracy': acc_knn, 'Precision': prec_knn,
    'Recall': rec_knn, 'F1': f1_knn
})
baseline_sonuclar.append({
    'Model': f"KNN Optimized {grid_search.best_params_}",
    'Accuracy': acc_best,
    'Precision': precision_score(y_test, y_pred_best),
    'Recall': recall_score(y_test, y_pred_best),
    'F1': f1_best
})
````

````python
# ============================================================
# ⚖️ ADIM 8: 12 SINIFLANDIRMA ALGORİTMASI KARŞILAŞTIRMASI
# ============================================================
# Her algoritmanın accuracy, precision, recall, f1 ve eğitim
# süresini kaydediyoruz. Gerçek dünya projelerinde bu tablo
# proje yöneticisine sunum için hazırlanan raporu oluşturur.

# ── Algoritma Sözlüğü ────────────────────────────────────────
modeller = {
    '① Logistic Regression' : LogisticRegression(random_state=RANDOM_STATE, max_iter=1000),
    '② KNN (Optimized)'     : grid_search.best_estimator_,
    '③ SVM'                 : SVC(probability=True, random_state=RANDOM_STATE),
    '④ Decision Tree'       : DecisionTreeClassifier(random_state=RANDOM_STATE),
    '⑤ Random Forest'       : RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1),
    '⑥ Gradient Boosting'   : GradientBoostingClassifier(n_estimators=100, random_state=RANDOM_STATE),
    '⑨ Naive Bayes'         : GaussianNB(),
    '⑩ MLP (Neural Net)'    : MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=500,
                                              random_state=RANDOM_STATE),
    '⑪ AdaBoost'            : AdaBoostClassifier(n_estimators=100, random_state=RANDOM_STATE),
    '⑫ Extra Trees'         : ExtraTreesClassifier(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1),
}

# XGBoost ve LightGBM varsa ekle
if XGBOOST_AVAILABLE:
    modeller['⑦ XGBoost'] = XGBClassifier(
        n_estimators=100, random_state=RANDOM_STATE,
        use_label_encoder=False, eval_metric='logloss', n_jobs=-1, verbosity=0)
if LGBM_AVAILABLE:
    modeller['⑧ LightGBM'] = LGBMClassifier(
        n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1, verbose=-1)

# ── Eğitim Döngüsü ───────────────────────────────────────────
print("=" * 75)
print("⚙️  MODELLERİ EĞİTİYORUZ...")
print("=" * 75)
print(f"{'Model Adı':<28} {'Acc':>6} {'Prec':>6} {'Rec':>6} {'F1':>6} {'AUC':>6} {'Süre(s)':>8}")
print("-" * 75)

tum_sonuclar = []  # Tüm sonuçları buraya toplayacağız

for model_adi, model in modeller.items():
    baslangic = time.time()
    model.fit(X_train_sc, y_train)
    sure = time.time() - baslangic

    y_pred = model.predict(X_test_sc)

    # Olasılık tahminleri (ROC-AUC için)
    if hasattr(model, 'predict_proba'):
        y_prob = model.predict_proba(X_test_sc)[:, 1]
    else:
        y_prob = model.decision_function(X_test_sc)

    acc  = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec  = recall_score(y_test, y_pred, zero_division=0)
    f1   = f1_score(y_test, y_pred, zero_division=0)
    auc  = roc_auc_score(y_test, y_prob)

    tum_sonuclar.append({
        'Model'    : model_adi,
        'Accuracy' : acc,
        'Precision': prec,
        'Recall'   : rec,
        'F1-Score' : f1,
        'ROC-AUC'  : auc,
        'Süre (s)' : round(sure, 3)
    })

    print(f"{model_adi:<28} {acc:>6.4f} {prec:>6.4f} {rec:>6.4f} {f1:>6.4f} {auc:>6.4f} {sure:>8.3f}")

print("-" * 75)
print()

# ── Sonuçları DataFrame'e çevir ──────────────────────────────
df_sonuclar = pd.DataFrame(tum_sonuclar).sort_values('F1-Score', ascending=False).reset_index(drop=True)
df_sonuclar.index += 1  # 1'den başlat

print("🏆 SONUÇ TABLOSU (F1-Score'a göre sıralı):")
print("=" * 75)
display(df_sonuclar.style
    .background_gradient(subset=['Accuracy','Precision','Recall','F1-Score','ROC-AUC'],
                         cmap='Greens', vmin=0.5, vmax=1.0)
    .background_gradient(subset=['Süre (s)'], cmap='Reds_r')
    .format({
        'Accuracy': '{:.4f}', 'Precision': '{:.4f}',
        'Recall': '{:.4f}', 'F1-Score': '{:.4f}',
        'ROC-AUC': '{:.4f}', 'Süre (s)': '{:.3f}'
    })
    .set_properties(**{'text-align': 'center'})
    .highlight_max(subset=['Accuracy','F1-Score','ROC-AUC'],
                   color='#bbf7d0', axis=0)
    .highlight_min(subset=['Süre (s)'], color='#bfdbfe', axis=0)
)
````
