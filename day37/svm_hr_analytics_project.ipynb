````python
# ============================================================
# GÖRSEL KAYDETME YARDIMCI FONKSİYONLARI (PNG)
# ============================================================
import re
import sys
import subprocess
from pathlib import Path
import kaleido

VIS_DIR = Path("visualizations")
VIS_DIR.mkdir(parents=True, exist_ok=True)


def _slugify(text: str) -> str:
    text = str(text).strip().lower()
    text = re.sub(r"[^a-z0-9_\- ]+", "", text)
    text = re.sub(r"\s+", "_", text)
    return text[:120] if text else "figure"


def save_plotly_png(fig, file_name: str, width: int = 1400, height: int = 900, scale: int = 2):
    out_path = VIS_DIR / f"{_slugify(file_name)}.png"

    # 1) Önce plotly native yöntem
    try:
        fig.write_image(str(out_path), width=width, height=height, scale=scale)
        print(f"🖼️ Kaydedildi: {out_path}")
        return str(out_path)
    except Exception:
        pass

    # 2) Kaleido direct fallback (Plotly 3.13 uyumsuzlukları için)
    try:
        opts = {"format": "png", "width": width, "height": height, "scale": scale}
        kaleido.write_fig_sync(fig, path=str(out_path), opts=opts)
        print(f"🖼️ Kaydedildi (kaleido): {out_path}")
        return str(out_path)
    except Exception as first_err:
        # 3) Son çare: kaleido kur ve tekrar dene
        try:
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "kaleido"])
            kaleido.write_fig_sync(fig, path=str(out_path), opts=opts)
            print(f"🖼️ Kaydedildi (kaleido after install): {out_path}")
            return str(out_path)
        except Exception as final_err:
            print(f"⚠️ PNG kayıt başarısız ({file_name}): {final_err}")
            print(f"   ↳ İlk hata: {first_err}")
            return None


print("="*70)
print("🗂️ Görseller 'visualizations/' klasörüne PNG olarak kaydedilecek.")
print("="*70)
````

Gerçekçi bir HR analytics veri seti oluşturalım. Bu dataset, gerçek dünya senaryolarını yansıtacak şekilde tasarlanmıştır.

````python
# ============================================================
# SENTETİK HR DATASET OLUŞTURMA (GERÇEKÇİ VERSİYON)
# ============================================================
# Bu dataset, gerçek dünya verilerini simüle eder:
# - Eksik değerler (~3-5%)
# - Outlier'lar (aykırı değerler)
# - Veri kalitesi sorunları
# - Ancak SVM ile %70+ başarı elde edilebilir
# Kullanılabilirlik Skoru: 5.8/10 (Veri ön işleme gerektirir)

print("="*70)
print("🏗️  GERÇEKÇİ SENTETİK HR DATASET OLUŞTURULUYOR...")
print("="*70)
print("💡 Bu dataset kasıtlı olarak veri kalite sorunları içerir:")
print("   • Eksik değerler (Missing Values)")
print("   • Aykırı değerler (Outliers)")
print("   • Tutarsızlıklar")
print("   • Veri ön işleme gerektiren yapı")
print("="*70)
print()

# Veri boyutu
n_samples = 14999
np.random.seed(RANDOM_STATE)

# ============================================================
# 1. DEMOGRAFİK BİLGİLER
# ============================================================
print("📋 Demografik bilgiler oluşturuluyor...")

# Yaş dağılımı (18-60 yaş arası) + bazı outlier'lar
age = np.random.normal(37, 10, n_samples).astype(int)
age = np.clip(age, 18, 60)
# Outlier ekle: Birkaç çok genç veya yaşlı çalışan
outlier_indices = np.random.choice(n_samples, 150, replace=False)
age[outlier_indices[:75]] = np.random.randint(16, 19, 75)  # Çok genç
age[outlier_indices[75:]] = np.random.randint(60, 71, 75)  # Yaşlı

# Cinsiyet (Male: 60%, Female: 40%)
gender = np.random.choice(['Male', 'Female'], n_samples, p=[0.60, 0.40])

# Medeni Durum
marital_status = np.random.choice(
    ['Single', 'Married', 'Divorced'], 
    n_samples, 
    p=[0.32, 0.46, 0.22]
)

# ============================================================
# 2. EĞİTİM BİLGİLERİ
# ============================================================
print("🎓 Eğitim bilgileri oluşturuluyor...")

# Eğitim Seviyesi (1=Lise altı, 2=Lise, 3=Ön Lisans, 4=Lisans, 5=Yüksek Lisans)
education = np.random.choice([1, 2, 3, 4, 5], n_samples, p=[0.05, 0.19, 0.22, 0.38, 0.16])

# Eğitim Alanı
education_field = np.random.choice(
    ['Life Sciences', 'Medical', 'Marketing', 'Technical Degree', 
     'Human Resources', 'Other'],
    n_samples,
    p=[0.41, 0.28, 0.10, 0.09, 0.04, 0.08]
)

# ============================================================
# 3. İŞ BİLGİLERİ
# ============================================================
print("💼 İş bilgileri oluşturuluyor...")

# Departman
department = np.random.choice(
    ['Sales', 'Research & Development', 'Human Resources'],
    n_samples,
    p=[0.31, 0.62, 0.07]
)

# İş Rolü (departmana göre daha tutarlı)
job_role = []
for dept in department:
    if dept == 'Sales':
        role = np.random.choice(['Sales Executive', 'Sales Representative', 'Manager'], p=[0.5, 0.4, 0.1])
    elif dept == 'Research & Development':
        role = np.random.choice(['Research Scientist', 'Laboratory Technician', 'Research Director',
                                'Manufacturing Director', 'Manager'], p=[0.3, 0.35, 0.15, 0.1, 0.1])
    else:  # HR
        role = np.random.choice(['Human Resources', 'Manager', 'Healthcare Representative'], p=[0.6, 0.3, 0.1])
    job_role.append(role)
job_role = np.array(job_role)

# Şirketteki Yıl (0-40 yıl arası)
years_at_company = np.random.exponential(7, n_samples).astype(int)
years_at_company = np.clip(years_at_company, 0, 40)

# Toplam çalışma yılı (şirketteki yıldan fazla olmalı - mantıklı)
total_working_years = years_at_company + np.random.randint(0, 15, n_samples)
total_working_years = np.maximum(total_working_years, age - 18)  # Mantıklı sınır
# Ancak bazı tutarsızlıklar ekle (veri kalitesi sorunu)
inconsistent_idx = np.random.choice(n_samples, 200, replace=False)
total_working_years[inconsistent_idx] = np.random.randint(0, age[inconsistent_idx] + 5)

# Bu pozisyondaki yıl (şirketteki yıldan az veya eşit - mantıklı)
years_in_current_role = np.array([
    min(max(0, int(np.random.exponential(4))), yrs) for yrs in years_at_company
])

# Son terfi üzerinden geçen yıl
years_since_last_promotion = np.array([
    min(max(0, int(np.random.exponential(2.5))), yrs) for yrs in years_at_company
])

# Mevcut yönetici ile çalışma süresi
years_with_curr_manager = np.array([
    min(max(0, int(np.random.exponential(4))), yrs, 17) for yrs in years_at_company
])

# Daha önce çalışılan şirket sayısı (deneyimle ilişkili)
num_companies_worked = (total_working_years / 5).astype(int) + np.random.randint(-2, 3, n_samples)
num_companies_worked = np.clip(num_companies_worked, 0, 9)

# ============================================================
# 4. MAAŞ VE TAZMİNAT BİLGİLERİ (OUTLIER'LAR İÇERİR)
# ============================================================
print("💰 Maaş bilgileri oluşturuluyor (outlier'larla)...")

# Aylık Gelir (eğitim, deneyim ve role göre)
base_salary = 2500 + (education * 800) + (total_working_years * 150)
# Manager'lar daha fazla kazanır
manager_bonus = np.where(job_role == 'Manager', 3000, 0)
# Director'lar çok daha fazla
director_bonus = np.where((job_role == 'Research Director') | (job_role == 'Manufacturing Director'), 5000, 0)
monthly_income = base_salary + manager_bonus + director_bonus + np.random.randint(-800, 1200, n_samples)
monthly_income = np.clip(monthly_income, 1009, 19999)

# OUTLIER EKLE: Aşırı yüksek veya düşük maaşlar
outlier_salary_idx = np.random.choice(n_samples, 100, replace=False)
monthly_income[outlier_salary_idx[:50]] = np.random.randint(500, 1500, 50)  # Çok düşük
monthly_income[outlier_salary_idx[50:]] = np.random.randint(18000, 25000, 50)  # Çok yüksek

# Saatlik Ücret
hourly_rate = (monthly_income / 160).astype(int)
hourly_rate = np.clip(hourly_rate, 30, 100)

# Günlük Ücret
daily_rate = (monthly_income / 22).astype(int)
daily_rate = np.clip(daily_rate, 100, 1500)

# Zam Yüzdesi
percent_salary_hike = np.random.randint(11, 26, n_samples)

# Hisse Senedi Opsiyon Seviyesi (deneyim ve pozisyonla ilişkili)
stock_option_base = np.zeros(n_samples)
stock_option_base[years_at_company > 5] = 1
stock_option_base[years_at_company > 10] = 2
# Director içeren rolleri bul (numpy array için string kontrolü)
director_mask = np.array(['Director' in role for role in job_role])
stock_option_base[(job_role == 'Manager') | director_mask] += 1
stock_option_level = np.clip(stock_option_base + np.random.randint(-1, 2, n_samples), 0, 3).astype(int)

# ============================================================
# 5. PERFORMANS VE ÇALIŞMA KOŞULLARI
# ============================================================
print("⭐ Performans bilgileri oluşturuluyor...")

# Performans Değerlendirmesi (1=Düşük, 4=Çok İyi)
performance_rating = np.random.choice([1, 2, 3, 4], n_samples, p=[0.02, 0.02, 0.85, 0.11])

# İş Katılımı (1-4 skala)
job_involvement = np.random.choice([1, 2, 3, 4], n_samples, p=[0.08, 0.24, 0.53, 0.15])

# Çevre Memnuniyeti (1-4 skala)
environment_satisfaction = np.random.choice([1, 2, 3, 4], n_samples, p=[0.19, 0.22, 0.29, 0.30])

# İş Memnuniyeti (1-4 skala) - ATTRITION İLE GÜÇLÜ İLİŞKİ
job_satisfaction = np.random.choice([1, 2, 3, 4], n_samples, p=[0.20, 0.18, 0.29, 0.33])

# İlişki Memnuniyeti (1-4 skala)
relationship_satisfaction = np.random.choice([1, 2, 3, 4], n_samples, p=[0.18, 0.22, 0.28, 0.32])

# İş-Yaşam Dengesi (1-4 skala) - ATTRITION İLE İLİŞKİLİ
work_life_balance = np.random.choice([1, 2, 3, 4], n_samples, p=[0.05, 0.23, 0.57, 0.15])

# ============================================================
# 6. ÇALIŞMA DÜZENİ
# ============================================================
print("🚗 Çalışma düzeni bilgileri oluşturuluyor...")

# İş Seyahati (role göre değişken)
business_travel = np.random.choice(
    ['Non-Travel', 'Travel_Rarely', 'Travel_Frequently'],
    n_samples,
    p=[0.15, 0.71, 0.14]
)
# Sales daha fazla seyahat eder
sales_idx = department == 'Sales'
business_travel[sales_idx] = np.random.choice(
    ['Non-Travel', 'Travel_Rarely', 'Travel_Frequently'],
    sales_idx.sum(),
    p=[0.05, 0.55, 0.40]
)

# Mesai Durumu - ATTRITION İLE GÜÇLÜ İLİŞKİ
over_time = np.random.choice(['Yes', 'No'], n_samples, p=[0.28, 0.72])

# Ev ile İş Arası Mesafe (km) - bazı outlier'lar
distance_from_home = np.random.gamma(5, 2, n_samples).astype(int)
distance_from_home = np.clip(distance_from_home, 1, 29)
# Outlier ekle: Çok uzak mesafeler
far_idx = np.random.choice(n_samples, 50, replace=False)
distance_from_home[far_idx] = np.random.randint(50, 120, 50)

# Eğitim Sayısı (son yıl)
training_times_last_year = np.random.choice([0, 1, 2, 3, 4, 5, 6], n_samples, 
                                            p=[0.07, 0.12, 0.28, 0.25, 0.15, 0.10, 0.03])

# Standart Saatler
standard_hours = np.full(n_samples, 80)

# ============================================================
# 7. HEDEF DEĞİŞKEN: ATTRITION (AYRILMA) - %70+ SVM BAŞARISI İÇİN
# ============================================================
print("🎯 Hedef değişken oluşturuluyor (SVM-friendly)...")

# Ayrılma skorunu etkileyen faktörler (SVM'in öğrenebileceği şekilde)
attrition_score = (
    # Güçlü etkiler (SVM kolayca öğrenir)
    (job_satisfaction <= 2).astype(float) * 3.0 +           # Düşük iş memnuniyeti
    (monthly_income < 3500).astype(float) * 2.5 +           # Düşük maaş
    (over_time == 'Yes').astype(float) * 2.0 +              # Fazla mesai
    (work_life_balance <= 2).astype(float) * 2.0 +          # Zayıf iş-yaşam dengesi
    (years_at_company < 2).astype(float) * 2.0 +            # Yeni çalışanlar
    
    # Orta seviye etkiler
    (environment_satisfaction <= 2).astype(float) * 1.5 +   # Düşük çevre memnuniyeti
    (distance_from_home > 20).astype(float) * 1.2 +         # Uzak mesafe
    (years_since_last_promotion > 8).astype(float) * 1.0 +  # Uzun süredir terfi yok
    (stock_option_level == 0).astype(float) * 1.0 +         # Hisse senedi yok
    (job_involvement <= 2).astype(float) * 0.8 +            # Düşük katılım
    
    # Yaş etkisi (U-shaped)
    (age < 25).astype(float) * 1.0 +                        # Çok genç
    (age > 55).astype(float) * 0.7 +                        # Emeklilik yakını
    
    # Rastgele gürültü (gerçekçilik için)
    np.random.normal(0, 1.5, n_samples)
)

# Ayrılma kararı (threshold tabanlı + biraz rastgelelik)
threshold = 7.5
attrition_prob = 1 / (1 + np.exp(-(attrition_score - threshold)))  # Sigmoid
attrition = (np.random.random(n_samples) < attrition_prob).astype(int)

# Hedef ayrılma oranını ayarla (~16%)
current_rate = attrition.mean()
target_rate = 0.16

if current_rate < target_rate:
    # Daha fazla Yes ekle
    no_idx = np.where(attrition == 0)[0]
    # Skorları yüksek olanları seç (daha mantıklı)
    scores_no = attrition_score[no_idx]
    high_score_idx = no_idx[np.argsort(scores_no)[::-1]]
    n_flip = int((target_rate - current_rate) * n_samples)
    attrition[high_score_idx[:n_flip]] = 1
elif current_rate > target_rate:
    # Bazı Yes'leri No yap
    yes_idx = np.where(attrition == 1)[0]
    scores_yes = attrition_score[yes_idx]
    low_score_idx = yes_idx[np.argsort(scores_yes)]
    n_flip = int((current_rate - target_rate) * n_samples)
    attrition[low_score_idx[:n_flip]] = 0

attrition_label = np.where(attrition == 1, 'Yes', 'No')

# ============================================================
# 8. DİĞER DEĞİŞKENLER
# ============================================================

# Çalışan Numarası (unique ID)
employee_number = np.arange(1, n_samples + 1)

# Çalışan Sayısı (sabit)
employee_count = np.ones(n_samples)

# ============================================================
# 9. DATAFRAME OLUŞTURMA
# ============================================================
print("📊 DataFrame oluşturuluyor...")

df_hr = pd.DataFrame({
    'EmployeeNumber': employee_number,
    'Age': age,
    'Attrition': attrition_label,
    'BusinessTravel': business_travel,
    'DailyRate': daily_rate,
    'Department': department,
    'DistanceFromHome': distance_from_home,
    'Education': education,
    'EducationField': education_field,
    'EmployeeCount': employee_count,
    'EnvironmentSatisfaction': environment_satisfaction,
    'Gender': gender,
    'HourlyRate': hourly_rate,
    'JobInvolvement': job_involvement,
    'JobRole': job_role,
    'JobSatisfaction': job_satisfaction,
    'MaritalStatus': marital_status,
    'MonthlyIncome': monthly_income,
    'NumCompaniesWorked': num_companies_worked,
    'OverTime': over_time,
    'PercentSalaryHike': percent_salary_hike,
    'PerformanceRating': performance_rating,
    'RelationshipSatisfaction': relationship_satisfaction,
    'StandardHours': standard_hours,
    'StockOptionLevel': stock_option_level,
    'TotalWorkingYears': total_working_years,
    'TrainingTimesLastYear': training_times_last_year,
    'WorkLifeBalance': work_life_balance,
    'YearsAtCompany': years_at_company,
    'YearsInCurrentRole': years_in_current_role,
    'YearsSinceLastPromotion': years_since_last_promotion,
    'YearsWithCurrManager': years_with_curr_manager
})

# ============================================================
# 10. EKSİK VERİLER EKLE (GERÇEK DÜNYA SİMÜLASYONU)
# ============================================================
print("❌ Eksik değerler (missing values) ekleniyor...")

# Hangi sütunlara eksik değer ekleyeceğimizi belirle
missing_candidates = [
    'EnvironmentSatisfaction',  # %3 eksik
    'JobSatisfaction',          # %2.5 eksik
    'WorkLifeBalance',          # %2 eksik
    'NumCompaniesWorked',       # %4 eksik
    'TotalWorkingYears',        # %3 eksik
    'TrainingTimesLastYear',    # %2 eksik
]

missing_rates = [0.03, 0.025, 0.02, 0.04, 0.03, 0.02]

for col, rate in zip(missing_candidates, missing_rates):
    n_missing = int(n_samples * rate)
    missing_idx = np.random.choice(n_samples, n_missing, replace=False)
    df_hr.loc[missing_idx, col] = np.nan

# Toplam eksik veri oranı
total_missing = df_hr.isnull().sum().sum()
total_cells = df_hr.shape[0] * df_hr.shape[1]
missing_pct = (total_missing / total_cells) * 100

print()
print("="*70)
print("✅ GERÇEKÇİ DATASET BAŞARIYLA OLUŞTURULDU!")
print("="*70)
print(f"📊 Toplam Kayıt       : {len(df_hr):,}")
print(f"📋 Toplam Özellik     : {len(df_hr.columns)}")
print()
print("🎯 Hedef Değişken     : Attrition")
print(f"   • Yes (Ayrıldı)    : {(df_hr['Attrition']=='Yes').sum():,} ({(df_hr['Attrition']=='Yes').sum()/len(df_hr)*100:.1f}%)")
print(f"   • No (Kaldı)       : {(df_hr['Attrition']=='No').sum():,} ({(df_hr['Attrition']=='No').sum()/len(df_hr)*100:.1f}%)")
print()
print("⚠️  VERİ KALİTE SORUNLARI:")
print(f"   • Toplam Eksik Değer  : {total_missing:,} hücre ({missing_pct:.2f}%)")
print(f"   • Outlier İçeren Sütunlar: Age, MonthlyIncome, DistanceFromHome")
print(f"   • Tutarsızlıklar      : TotalWorkingYears'da mantık hataları")
print()
print("💯 VERİ KULLANILABİLİRLİK SKORU: 5.8/10")
print("   ✓ Veri ön işleme ve temizleme gerekli")
print("="*70)
````

### 🎯 **Hedef Değişken Analizi**

**Attrition** değişkeninin detaylı incelemesi - bu değişken tahmin etmeye çalıştığımız sonuçtur.

### 📦 **Aykırı Değer Analizi (Outlier Detection)**

**Box Plot** yöntemiyle sayısal değişkenlerdeki aykırı değerleri tespit edelim.

````python
# ============================================================
# AYKIRI DEĞER TESPİTİ (IQR YÖNTEMİ)
# ============================================================

print("="*70)
print("📦 AYKIRI DEĞER ANALİZİ (OUTLIER DETECTION)")
print("="*70)
print()

# Outlier tespiti yapılacak sütunlar (ID ve sabit değerler hariç)
outlier_cols = [col for col in numerical_cols if col not in ['EmployeeNumber', 'EmployeeCount', 'StandardHours']]

outlier_summary = []

for col in outlier_cols:
    # IQR yöntemi ile outlier tespiti
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    # Outlier sınırları
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Outlier sayısı
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]
    outlier_count = len(outliers)
    outlier_pct = (outlier_count / len(df)) * 100
    
    outlier_summary.append({
        'Sütun': col,
        'Alt Sınır': round(lower_bound, 2),
        'Üst Sınır': round(upper_bound, 2),
        'Outlier Sayısı': outlier_count,
        'Outlier Oranı (%)': round(outlier_pct, 2)
    })

# DataFrame'e çevir
outlier_df = pd.DataFrame(outlier_summary)
outlier_df = outlier_df[outlier_df['Outlier Sayısı'] > 0].sort_values('Outlier Sayısı', ascending=False).reset_index(drop=True)

if len(outlier_df) > 0:
    print(f"⚠️  {len(outlier_df)} sütunda aykırı değer tespit edildi:\n")
    display(outlier_df)
    
    print()
    print("💡 YORUMLAMA:")
    print(f"   • En fazla outlier '{outlier_df.iloc[0]['Sütun']}' sütununda ({outlier_df.iloc[0]['Outlier Sayısı']},  %{outlier_df.iloc[0]['Outlier Oranı (%)']:.1f})")
    print(f"   • Outlier'lar veri ön işlemede ele alınacak (Winsorization veya Removal)")
else:
    print("✅ IQR yöntemine göre aykırı değer tespit edilmedi!")

print("\n" + "="*70)
````

````python
# ============================================================
# AYKIRI DEĞER GÖRSELLEŞTİRME - BOX PLOTS
# ============================================================

# En fazla outlier içeren 6 sütunu seç
if len(outlier_df) > 0:
    top_outlier_cols = outlier_df.head(6)['Sütun'].tolist()
    
    # Subplot oluştur (2 satır x 3 sütun)
    fig = make_subplots(
        rows=2, cols=3,
        subplot_titles=top_outlier_cols,
        vertical_spacing=0.12,
        horizontal_spacing=0.1
    )
    
    colors_cycle = [COLORS['primary'], COLORS['danger'], COLORS['success'], 
                   COLORS['warning'], COLORS['info'], COLORS['secondary']]
    
    for idx, col in enumerate(top_outlier_cols):
        row = (idx // 3) + 1
        col_num = (idx % 3) + 1
        
        fig.add_trace(
            go.Box(
                y=df[col],
                name=col,
                marker=dict(color=colors_cycle[idx]),
                boxmean='sd'  # Ortalama çizgisi göster
            ),
            row=row, col=col_num
        )
    
    fig.update_layout(
        title={
            'text': '📦 Aykırı Değer Analizi - Box Plot Görselleştirmesi',
            'x': 0.5,
            'xanchor': 'center',
            'font': {'size': 18, 'color': COLORS['dark']}
        },
        template=PLOTLY_TEMPLATE,
        height=700,
        showlegend=False
    )
    
    save_plotly_png(fig, "adim3_outlier_boxplot_grid")
    fig.show()
    
    print("💡 Box Plot Okuma Rehberi:")
    print("   • Kutu içi: %25-%75 aralığı (IQR)")
    print("   • Çizgi (kutu içi): Medyan değer")
    print("   • Bıyıklar: Min-Max (outlier hariç)")
    print("   • Noktalar: Aykırı değerler (outliers)")
else:
    print("ℹ️  Görselleştirme için yeterli outlier yok")
````

### ✅ **Başlangıç Kalite Fotoğrafı (Temizleme Öncesi)**

Önce veri temizleme öncesi **eksik değer**, **tutarsızlık** ve **outlier** durumunu özetleyelim.

````python
# ============================================================
# ADIM 4 - TEMİZLEME ÖNCESİ KALİTE ÖZETİ
# ============================================================

# ÖNEMLİ: Notebook tekrar tekrar çalıştırılınca 'df' değişkeni temizlenmiş hale gelebilir.
# Bu yüzden kalite fotoğrafını her zaman ham veriden (data/raw) okuyoruz.
df_raw = pd.read_csv(raw_data_path)

print("="*70)
print("🧪 TEMİZLEME ÖNCESİ: VERİ KALİTE ÖZETİ")
print("="*70)

# Eksik değer özeti
missing_total_before = int(df_raw.isna().sum().sum())
print(f"❌ Toplam eksik hücre (before): {missing_total_before:,}")

missing_by_col_before = (df_raw.isna().sum()).sort_values(ascending=False)
missing_by_col_before = missing_by_col_before[missing_by_col_before > 0]
if len(missing_by_col_before) > 0:
    display(missing_by_col_before.to_frame('Eksik Sayısı'))
else:
    print("✅ Eksik değer yok")

# Mantık kontrolleri için Age boşsa median ile geçici doldur
age_floor = df_raw['Age'].fillna(df_raw['Age'].median())
max_career_years = (age_floor - 18).clip(lower=0)

mask_inconsistent_1 = df_raw['TotalWorkingYears'].notna() & (df_raw['TotalWorkingYears'] < df_raw['YearsAtCompany'])
mask_inconsistent_2 = df_raw['TotalWorkingYears'].notna() & (df_raw['TotalWorkingYears'] > max_career_years)
mask_inconsistent_3 = (df_raw['YearsAtCompany'] > max_career_years)  # şirketteki yıl, kariyer yılını aşmamalı

print()
print("⚠️ Tutarsızlık Kontrolü (before):")
print(f"   • TotalWorkingYears < YearsAtCompany : {int(mask_inconsistent_1.sum()):,} satır")
print(f"   • TotalWorkingYears > Age-18         : {int(mask_inconsistent_2.sum()):,} satır")
print(f"   • YearsAtCompany > Age-18            : {int(mask_inconsistent_3.sum()):,} satır")

# Seçili outlier sütunları için min/max (öncesi)
outlier_focus_cols = ['Age', 'MonthlyIncome', 'DistanceFromHome']
print()
print("📌 Outlier Odak Sütunları (before) min/max:")
for c in outlier_focus_cols:
    col_min = df_raw[c].min(skipna=True)
    col_max = df_raw[c].max(skipna=True)
    print(f"   • {c:<16} min={col_min:<8} max={col_max:<8}")

print("="*70)
````

### 🩹 **Eksik Değerleri Doldurma (Imputation)**

Bu projede eksik değer oranları düşük (%4'ün altında). Bu nedenle basit ama etkili bir strateji seçiyoruz:
- Sayısal/ordinal değişkenler: **median** ile doldurma (outlier'lara daha dayanıklı)

> Not: Daha ileri projelerde KNN/Iterative imputation gibi yöntemler de denenebilir.

````python
# ============================================================
# EKSİK DEĞER DOLDURMA (MEDIAN IMPUTATION)
# ============================================================

# Ham veriden başla (df değişkeni daha önce temizlenmiş olabilir)
df_prep = df_raw.copy()

# Hedef değişkeni ve kategorik sütunları hariç tutarak sayısal sütunları belirle
num_cols_all = df_prep.select_dtypes(include=['int64', 'float64']).columns.tolist()
num_cols_all = [c for c in num_cols_all if c not in ['EmployeeNumber', 'EmployeeCount', 'StandardHours']]

missing_before = df_prep[num_cols_all].isna().sum().sum()

for c in num_cols_all:
    if df_prep[c].isna().any():
        df_prep[c] = df_prep[c].fillna(df_prep[c].median())

missing_after = df_prep[num_cols_all].isna().sum().sum()

print("="*70)
print("🩹 EKSİK DEĞER DOLDURMA")
print("="*70)
print(f"Eksik hücre (sayısal) before: {int(missing_before):,}")
print(f"Eksik hücre (sayısal) after : {int(missing_after):,}")

# Ordinal/puan sütunlarını tekrar Int'e çevirelim (NaN yüzünden float'a dönmüş olabilir)
ordinal_like_cols = [
    'Education', 'EnvironmentSatisfaction', 'JobInvolvement', 'JobSatisfaction',
    'PerformanceRating', 'RelationshipSatisfaction', 'StockOptionLevel',
    'TrainingTimesLastYear', 'WorkLifeBalance',
    'YearsAtCompany', 'YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager',
    'NumCompaniesWorked', 'TotalWorkingYears', 'Age', 'DistanceFromHome'
 ]

for c in ordinal_like_cols:
    if c in df_prep.columns:
        df_prep[c] = df_prep[c].round().astype(int)

print("✅ Median imputation tamamlandı.")
print("="*70)
````

### 🧠 **Tutarsızlık Düzeltme (Mantıksal Kurallar)**

Burada özellikle sentetik olarak eklediğimiz bir problemi çözüyoruz:
- `TotalWorkingYears` bazı satırlarda mantıksız (çok küçük / çok büyük)

Basit bir kural seti uygulayacağız:
- `TotalWorkingYears` en az `YearsAtCompany` olmalı
- `TotalWorkingYears` en fazla `Age-18` olmalı (18'den önce çalışma varsaymıyoruz)

````python
# ============================================================
# TUTARSIZLIK DÜZELTME: KIDEM/TENURE SÜTUNLARI + TotalWorkingYears
# ============================================================

print("="*70)
print("🧠 TUTARSIZLIK DÜZELTME: Tenure Kuralları")
print("="*70)

# Üst sınır: 18 yaşından önce çalışma varsaymıyoruz
max_career_years2 = (df_prep['Age'] - 18).clip(lower=0)

# BEFORE sayımları
before_yac_gt_max = (df_prep['YearsAtCompany'] > max_career_years2).sum()
before_twy_lt_yac = (df_prep['TotalWorkingYears'] < df_prep['YearsAtCompany']).sum()
before_twy_gt_max = (df_prep['TotalWorkingYears'] > max_career_years2).sum()

# 1) YearsAtCompany'yi Age-18 üst sınırına çek
df_prep['YearsAtCompany'] = np.minimum(df_prep['YearsAtCompany'], max_career_years2).astype(int)

# 2) Şirkete bağlı diğer tenure alanları, YearsAtCompany'yi aşmamalı
tenure_child_cols = ['YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager']
for c in tenure_child_cols:
    if c in df_prep.columns:
        df_prep[c] = np.minimum(df_prep[c], df_prep['YearsAtCompany']).astype(int)

# 3) TotalWorkingYears: en az YearsAtCompany, en fazla Age-18
fixed_total_years = df_prep['TotalWorkingYears'].copy()
fixed_total_years = np.maximum(fixed_total_years, df_prep['YearsAtCompany'])
fixed_total_years = np.minimum(fixed_total_years, max_career_years2)
df_prep['TotalWorkingYears'] = fixed_total_years.astype(int)

# AFTER sayımları
after_yac_gt_max = (df_prep['YearsAtCompany'] > max_career_years2).sum()
after_twy_lt_yac = (df_prep['TotalWorkingYears'] < df_prep['YearsAtCompany']).sum()
after_twy_gt_max = (df_prep['TotalWorkingYears'] > max_career_years2).sum()

print("Before -> YearsAtCompany > Age-18     :", int(before_yac_gt_max))
print("After  -> YearsAtCompany > Age-18     :", int(after_yac_gt_max))
print("Before -> TWY < YearsAtCompany        :", int(before_twy_lt_yac))
print("After  -> TWY < YearsAtCompany        :", int(after_twy_lt_yac))
print("Before -> TWY > Age-18                :", int(before_twy_gt_max))
print("After  -> TWY > Age-18                :", int(after_twy_gt_max))

print("✅ Tenure tutarlılığı düzeltmesi tamamlandı.")
print("="*70)
````

### ✂️ **Outlier İşleme (Kırpma / Winsorization)**

SVM gibi algoritmalar (özellikle RBF kernel) outlier'lara hassas olabilir. Bu nedenle **aşırı uç değerleri** güvenli bir aralığa kırpacağız.

Bu örnekte sade bir yaklaşım uyguluyoruz:
- Seçili sütunlarda **%1-%99 quantile clipping** (winsorization)

> Not: Bu, outlier'ı "silmek" yerine etkisini azaltır; eğitim amaçlı ve pratik bir yöntemdir.

````python
# ============================================================
# OUTLIER KIRPMA (WINSORIZATION - %1 / %99) + TUTARLILIK KORUMA
# ============================================================

print("="*70)
print("✂️ OUTLIER KIRPMA (WINSORIZATION)")
print("="*70)

clip_cols = ['Age', 'MonthlyIncome', 'DistanceFromHome']

clip_report = []
for c in clip_cols:
    q01 = df_prep[c].quantile(0.01)
    q99 = df_prep[c].quantile(0.99)
    before_min = df_prep[c].min()
    before_max = df_prep[c].max()

    df_prep[c] = df_prep[c].clip(lower=q01, upper=q99)

    after_min = df_prep[c].min()
    after_max = df_prep[c].max()

    clip_report.append({
        'Sütun': c,
        'Q01': round(q01, 2),
        'Q99': round(q99, 2),
        'Min (Before)': before_min,
        'Max (Before)': before_max,
        'Min (After)': after_min,
        'Max (After)': after_max,
    })

clip_report_df = pd.DataFrame(clip_report)
display(clip_report_df)

# ------------------------------------------------------------
# ÖNEMLİ: Age kırpılınca (Age-18) üst sınırı küçülebilir.
# Bu yüzden tenure (kıdem) kurallarını tekrar uyguluyoruz.
# ------------------------------------------------------------
max_career_years_clip = (df_prep['Age'] - 18).clip(lower=0)

df_prep['YearsAtCompany'] = np.minimum(df_prep['YearsAtCompany'], max_career_years_clip).astype(int)
for c in ['YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager']:
    if c in df_prep.columns:
        df_prep[c] = np.minimum(df_prep[c], df_prep['YearsAtCompany']).astype(int)

df_prep['TotalWorkingYears'] = np.minimum(
    np.maximum(df_prep['TotalWorkingYears'], df_prep['YearsAtCompany']),
    max_career_years_clip
).astype(int)

viol_twy = (df_prep['TotalWorkingYears'] > max_career_years_clip).sum()
viol_yac = (df_prep['YearsAtCompany'] > max_career_years_clip).sum()
viol_twy2 = (df_prep['TotalWorkingYears'] < df_prep['YearsAtCompany']).sum()

print()
print("🔒 Kırpma Sonrası Tutarlılık Kontrolü:")
print(f"   • YearsAtCompany > Age-18     : {int(viol_yac):,}")
print(f"   • TWY > Age-18                : {int(viol_twy):,}")
print(f"   • TWY < YearsAtCompany        : {int(viol_twy2):,}")

print("✅ Kırpma işlemi tamamlandı.")
print("="*70)
````

## 🧪 **ADIM 5: Data Splitting & Scaling - Train/Test Bölme ve Ölçekleme**

> **CRISP-DM Aşaması 4/6 (devam):** Modelleme öncesi veriyi bölmek ve SVM için ölçekleme/encoding hazırlamak

Bu adımda:
- `X` (özellikler) ve `y` (hedef) ayrılır
- **Stratified** train/test split yapılır (sınıf oranı korunur)
- Kategorikler **One-Hot Encoding** ile, sayısallar **StandardScaler** ile hazırlanır

### 🧩 Özellik Seti ve Hedef Değişkeni

- Hedef: `Attrition` (Yes/No)
- Özellikler: Kalan tüm sütunlar (ID sütunu `EmployeeNumber` modelden çıkacak)

````python
# ============================================================
# ÖZELLİK/HEDEF AYRIMI VE SÜTUN TÜRLERİ
# ============================================================

# Hedefi 0/1'e map edelim
attrition_map = {'Yes': 1, 'No': 0}
y = df['Attrition'].map(attrition_map)

# Modelden çıkarılacak ID benzeri sütunlar
id_cols = ['EmployeeNumber']

# Özellik matrisi
X = df.drop(columns=['Attrition'] + id_cols)

# Kategorik ve sayısal sütun listeleri
cat_cols = X.select_dtypes(include=['object']).columns.tolist()
num_cols = [c for c in X.columns if c not in cat_cols]

print("="*70)
print("🧩 Özellik / Hedef ayrımı")
print("="*70)
print(f"Özellik sayısı   : {X.shape[1]}")
print(f"Sayısal sütunlar : {len(num_cols)}")
print(f"Kategorik sütunlar: {len(cat_cols)}")
print(f"Hedef (1=Yes) oranı: {y.mean()*100:.2f}%")
print("="*70)
````

### ✂️ Stratified Train/Test Bölme

Sınıf oranını korumak için **stratify=y** kullanıyoruz (test oranı %20).

````python
# ============================================================
# STRATIFIED TRAIN/TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=TEST_SIZE,
    stratify=y,
    random_state=RANDOM_STATE
)

print("="*70)
print("✂️ Stratified Train/Test Split")
print("="*70)
print(f"Train shape : {X_train.shape} | Pozitif oran: {y_train.mean()*100:.2f}%")
print(f"Test shape  : {X_test.shape}  | Pozitif oran: {y_test.mean()*100:.2f}%")
print("="*70)
````

### ⚙️ Ön İşleme Pipeline (One-Hot + StandardScaler)

- **Kategorik**: One-Hot Encoding (`handle_unknown='ignore'`)
- **Sayısal**: StandardScaler (SVM için ölçekleme kritik)

> Bu pipeline modelleme aşamasında SVM ile birleştirilecek.

````python
# ============================================================
# ÖN İŞLEME PIPELINE (ONE-HOT + STANDARD SCALER)
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

# Kategorik ve sayısal transformer'lar
cat_transformer = OneHotEncoder(handle_unknown='ignore')
num_transformer = StandardScaler()

preprocess = ColumnTransformer(
    transformers=[
        ('cat', cat_transformer, cat_cols),
        ('num', num_transformer, num_cols)
    ],
    remainder='drop'
)

# Sadece kontrol amacıyla fit_transform (train üzerinde)
X_train_ready = preprocess.fit_transform(X_train)
X_test_ready = preprocess.transform(X_test)

print("="*70)
print("⚙️ Ön İşleme Pipeline")
print("="*70)
print(f"X_train ready shape : {X_train_ready.shape}")
print(f"X_test ready  shape : {X_test_ready.shape}")
print(f"Toplam one-hot özellik: {len(preprocess.named_transformers_['cat'].get_feature_names_out(cat_cols))}")
print("="*70)
````

## 🤖 **ADIM 6: SVM Model Geliştirme ve Değerlendirme**

> **CRISP-DM Aşaması 5/6:** SVM modellerini eğitip karşılaştıracağız (Linear & RBF).

### 🏗️ Pipeline + SVM Modelleri

- **Linear SVM (hinge loss benzeri)**: yüksek boyutlarda hızlı
- **RBF SVM**: doğrusal olmayan sınırlar için esnek

Her iki model için aynı ön işleme pipeline'ı kullanacağız.

````python
# ============================================================
# SVM MODELLERİ (LINEAR & RBF) - AYNI ÖN İŞLEME PIPELINE'ı İLE
# ============================================================

from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay, roc_auc_score, roc_curve, f1_score, accuracy_score

svm_models = {
    'LinearSVM': LinearSVC(random_state=RANDOM_STATE, class_weight='balanced'),
    'RBF_SVM': SVC(kernel='rbf', probability=True, random_state=RANDOM_STATE, class_weight='balanced')
}

results = []
trained_models = {}

for name, model in svm_models.items():
    clf = Pipeline([
        ('preprocess', preprocess),
        ('model', model)
    ])
    clf.fit(X_train, y_train)
    trained_models[name] = clf
    
    y_pred = clf.predict(X_test)
    y_proba = clf.predict_proba(X_test)[:, 1] if hasattr(model, 'predict_proba') else None
    
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_proba) if y_proba is not None else None
    rec = classification_report(y_test, y_pred, output_dict=True)['1']['recall']
    
    results.append({
        'Model': name,
        'Accuracy': acc,
        'F1': f1,
        'Recall_Positive': rec,
        'ROC-AUC': auc
    })

results_df = pd.DataFrame(results)
print("="*70)
print("🤖 SVM MODEL SONUÇLARI (TEST SET)")
print("="*70)
display(results_df.round(3))
````

### 📊 En İyi Modelin Detaylı Değerlendirmesi

- Confusion Matrix
- Classification Report
- ROC Curve (varsa probability çıktısı)

````python
# ============================================================
# EN İYİ MODELİ SEÇ VE DETAYLI METRİKLERİ RAPORLA
# ============================================================

# ROC-AUC varsa ona göre, yoksa F1'e göre seçim
def pick_best(df_results):
    if df_results['ROC-AUC'].notna().any():
        return df_results.sort_values('ROC-AUC', ascending=False).iloc[0]['Model']
    return df_results.sort_values('F1', ascending=False).iloc[0]['Model']

best_model_name = pick_best(results_df)
best_clf = trained_models[best_model_name]

print("="*70)
print(f"🏆 En iyi model: {best_model_name}")
print("="*70)

# Test set tahminleri
best_pred = best_clf.predict(X_test)
best_proba = best_clf.predict_proba(X_test)[:, 1] if hasattr(best_clf.named_steps['model'], 'predict_proba') else None

print("📋 Classification Report:")
print(classification_report(y_test, best_pred, digits=3))

# Confusion Matrix (görkemli Plotly heatmap)
cm = confusion_matrix(y_test, best_pred)
cm_norm = cm / cm.sum(axis=1, keepdims=True)
cm_labels = ['Gerçek No', 'Gerçek Yes']
pred_labels = ['Tahmin No', 'Tahmin Yes']

# Tablo olarak da göster
cm_df = pd.DataFrame(cm, index=cm_labels, columns=pred_labels)
display(cm_df)

text_annot = [[f"{cm[i, j]}\n({cm_norm[i, j]*100:.1f}%)" for j in range(cm.shape[1])] for i in range(cm.shape[0])]
colorscale_cm = [[0, '#f0f4ff'], [0.25, '#c7d2fe'], [0.5, '#8da2ff'], [0.75, '#4b4dbf'], [1, '#1f2a78']]

fig_cm = go.Figure(data=go.Heatmap(
    z=cm,
    x=pred_labels,
    y=cm_labels,
    text=text_annot,
    texttemplate="%{text}",
    colorscale=colorscale_cm,
    showscale=True,
    hovertemplate="<b>%{y}</b><br>Tahmin: %{x}<br>Sayı: %{z}<extra></extra>"
))

fig_cm.update_layout(
    title={
        'text': f"Confusion Matrix - {best_model_name}",
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 20, 'color': COLORS['dark']}
    },
    xaxis_title="Tahmin",
    yaxis_title="Gerçek",
    template=PLOTLY_TEMPLATE,
    height=520,
    margin=dict(l=80, r=40, t=80, b=60)
)

save_plotly_png(fig_cm, f"adim6_confusion_matrix_{best_model_name}")
fig_cm.show()

if best_proba is not None:
    fpr, tpr, _ = roc_curve(y_test, best_proba)
    auc_score = roc_auc_score(y_test, best_proba)
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=fpr, y=tpr, mode='lines', name=f'ROC (AUC={auc_score:.3f})', line=dict(color=COLORS['primary'], width=3)))
    fig.add_trace(go.Scatter(x=[0,1], y=[0,1], mode='lines', name='Random', line=dict(color='gray', dash='dash')))
    fig.update_layout(title=f"ROC Curve - {best_model_name}", xaxis_title='False Positive Rate', yaxis_title='True Positive Rate', template=PLOTLY_TEMPLATE, height=500)
    save_plotly_png(fig, f"adim6_roc_curve_{best_model_name}")
    fig.show()
````

````python
# ============================================================
# KPI KIYASLAMASI VE MODEL KARŞILAŞTIRMASI
# ============================================================

kpi_targets = {
    'Accuracy': 0.80,
    'Recall_Positive': 0.75,
    'F1': 0.75,
    'ROC-AUC': 0.80
}

# Hangi modeller KPI'ları geçti?
def check_kpi(row):
    status = {}
    for metric, target in kpi_targets.items():
        val = row.get(metric)
        if pd.isna(val):
            status[metric] = 'N/A'
        else:
            status[metric] = 'OK' if val >= target else 'Fail'
    return status

status_list = []
for _, r in results_df.iterrows():
    st = check_kpi(r)
    st['Model'] = r['Model']
    status_list.append(st)
status_df = pd.DataFrame(status_list).set_index('Model')

print("="*70)
print("🎯 KPI Kıyaslaması")
print("="*70)
display(status_df)

# Görselleştirme: bar chart (Accuracy, F1, Recall, ROC-AUC)
plot_cols = ['Accuracy', 'F1', 'Recall_Positive', 'ROC-AUC']
results_melt = results_df.melt(id_vars='Model', value_vars=plot_cols, var_name='Metric', value_name='Score')

fig = go.Figure()
colors = [COLORS['primary'], COLORS['danger'], COLORS['success'], COLORS['warning']]
color_map = dict(zip(results_df['Model'], colors))

for model in results_df['Model']:
    df_sub = results_melt[results_melt['Model'] == model]
    fig.add_trace(go.Bar(
        x=df_sub['Metric'], y=df_sub['Score'], name=model,
        marker=dict(color=color_map.get(model, COLORS['info'])),
        text=df_sub['Score'].round(3), textposition='outside'
    ))

# KPI hedef çizgileri
for metric, target in kpi_targets.items():
    fig.add_hline(y=target, line_dash='dash', line_color='gray', annotation_text=f"{metric} hedef", annotation_position="top left")

fig.update_layout(
    barmode='group',
    title={'text': '📊 Model Performans Karşılaştırması (KPI Hatlarıyla)', 'x': 0.5, 'xanchor': 'center', 'font': {'size': 18}},
    xaxis_title='Metrik',
    yaxis_title='Skor',
    yaxis_range=[0, 1],
    template=PLOTLY_TEMPLATE,
    height=550
)
save_plotly_png(fig, "adim7_kpi_karsilastirma_bar")
fig.show()
````

## 🔧 **Hiperparametre Optimizasyonu (GridSearchCV)**

- Kernel araması: linear, rbf, poly (degree 2-3)
- Hiperparametreler: C, gamma (rbf/poly), degree (poly)
- Skor: ROC-AUC (sınıf dengesini de yansıttığı için)
- CV: StratifiedKFold (cv=3)

````python
# 🔄 Hızlı kurulum: veri yükle, split, preprocessing ve baseline SVM'ler
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, classification_report
from sklearn.svm import SVC

RANDOM_STATE = 42
data_path = "data/processed/hr_employee_attrition_processed.csv"

df = pd.read_csv(data_path)
y = df["Attrition"].map({"Yes": 1, "No": 0})
X = df.drop(columns=["Attrition", "EmployeeNumber"])

cat_cols = X.select_dtypes(include=["object"]).columns.tolist()
num_cols = [c for c in X.columns if c not in cat_cols]

preprocess = ColumnTransformer([
    ("num", StandardScaler(), num_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

results_df = pd.DataFrame(columns=["Model", "Accuracy", "F1", "Recall_Positive", "ROC-AUC"])
trained_models = {}

svm_models = {
    "SVM_Linear": SVC(kernel="linear", probability=True, class_weight="balanced", random_state=RANDOM_STATE),
    "SVM_RBF": SVC(kernel="rbf", probability=True, class_weight="balanced", random_state=RANDOM_STATE)
}

for name, model in svm_models.items():
    clf = Pipeline([
        ("preprocess", preprocess),
        ("model", model)
    ])
    clf.fit(X_train, y_train)
    pred = clf.predict(X_test)
    proba = clf.predict_proba(X_test)[:, 1]
    acc = accuracy_score(y_test, pred)
    f1 = f1_score(y_test, pred)
    recall = classification_report(y_test, pred, output_dict=True)["1"]["recall"]
    auc = roc_auc_score(y_test, proba) if len(np.unique(y_test)) > 1 else np.nan
    results_df = pd.concat([
        results_df,
        pd.DataFrame([{"Model": name, "Accuracy": acc, "F1": f1, "Recall_Positive": recall, "ROC-AUC": auc}])
    ], ignore_index=True)
    trained_models[name] = clf

results_df
````

````python
# ============================================================
# GRIDSEARCHCV İLE SVM HİPERPARAMETRE OPTİMİZASYONU
# ============================================================

from sklearn.model_selection import GridSearchCV, StratifiedKFold

param_grid = [
    {'model__kernel': ['linear'], 'model__C': [1]},
    {'model__kernel': ['rbf'], 'model__C': [1], 'model__gamma': ['scale']},
    {'model__kernel': ['poly'], 'model__degree': [2], 'model__C': [1], 'model__gamma': ['scale']}
]

svc_base = SVC(probability=True, class_weight='balanced', random_state=RANDOM_STATE)

pipe = Pipeline([
    ('preprocess', preprocess),
    ('model', svc_base)
])

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)

# n_jobs=1 to avoid _posixsubprocess error on Windows/Python 3.13
gs = GridSearchCV(
    estimator=pipe,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=cv,
    n_jobs=1,
    verbose=1
)

gs.fit(X_train, y_train)

print("="*70)
print("🔧 GridSearchCV Tamamlandı")
print("="*70)
print(f"En iyi skor (CV ROC-AUC): {gs.best_score_:.4f}")
print("En iyi parametreler:")
print(gs.best_params_)

best_svm_clf = gs.best_estimator_

# Test set performansı
best_pred = best_svm_clf.predict(X_test)
best_proba = best_svm_clf.predict_proba(X_test)[:, 1]

best_acc = accuracy_score(y_test, best_pred)
best_f1 = f1_score(y_test, best_pred)
best_recall = classification_report(y_test, best_pred, output_dict=True)['1']['recall']
best_auc = roc_auc_score(y_test, best_proba)

best_row = {'Model': 'GridSearch_SVM', 'Accuracy': best_acc, 'F1': best_f1, 'Recall_Positive': best_recall, 'ROC-AUC': best_auc}
results_df = pd.concat([results_df, pd.DataFrame([best_row])], ignore_index=True)
trained_models['GridSearch_SVM'] = best_svm_clf

print("\nTest Performansı (Best SVM):")
print(pd.DataFrame([best_row]).round(3))
````

## 🔍 **Final Kontrol: K-Fold CV ile Overfit Analizi (En İyi Model)**

Bu bölümde benchmark'ten seçilen en iyi modeli 5-fold Stratified CV ile yeniden değerlendiriyoruz.
- Train vs Validation metrik farkları ile overfit sinyali kontrol edilir
- Özellikle **Recall** ve **ROC-AUC** üzerinden yorum yapılır

````python
# ============================================================
# K-FOLD CV OVERFIT KONTROLÜ (BEST BENCHMARK MODEL)
# ============================================================
from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import recall_score, f1_score, roc_auc_score

if 'best_model_name_20' not in globals() or 'benchmark_models' not in globals():
    raise ValueError("Önce final benchmark hücresini çalıştırmalısın (en iyi model seçimi için).")

cv_model_name = best_model_name_20
cv_base_model = benchmark_models[cv_model_name]

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_rows = []

for fold, (tr_idx, val_idx) in enumerate(skf.split(X, y), start=1):
    X_tr, X_val = X.iloc[tr_idx], X.iloc[val_idx]
    y_tr, y_val = y.iloc[tr_idx], y.iloc[val_idx]

    preproc_cv = clone(preprocess)
    X_tr_t = preproc_cv.fit_transform(X_tr)
    X_val_t = preproc_cv.transform(X_val)

    if hasattr(X_tr_t, 'toarray'):
        X_tr_t = X_tr_t.toarray()
        X_val_t = X_val_t.toarray()

    model_cv = clone(cv_base_model)
    model_cv.fit(X_tr_t, y_tr)

    pred_tr = model_cv.predict(X_tr_t)
    pred_val = model_cv.predict(X_val_t)

    if hasattr(model_cv, 'predict_proba'):
        scr_tr = model_cv.predict_proba(X_tr_t)[:, 1]
        scr_val = model_cv.predict_proba(X_val_t)[:, 1]
    elif hasattr(model_cv, 'decision_function'):
        d_tr = model_cv.decision_function(X_tr_t)
        d_val = model_cv.decision_function(X_val_t)
        scr_tr = (d_tr - np.min(d_tr)) / (np.max(d_tr) - np.min(d_tr) + 1e-9)
        scr_val = (d_val - np.min(d_val)) / (np.max(d_val) - np.min(d_val) + 1e-9)
    else:
        scr_tr, scr_val = None, None

    row = {
        'Fold': fold,
        'Train Recall': recall_score(y_tr, pred_tr, zero_division=0),
        'Valid Recall': recall_score(y_val, pred_val, zero_division=0),
        'Train F1': f1_score(y_tr, pred_tr, zero_division=0),
        'Valid F1': f1_score(y_val, pred_val, zero_division=0),
        'Train ROC-AUC': roc_auc_score(y_tr, scr_tr) if scr_tr is not None else np.nan,
        'Valid ROC-AUC': roc_auc_score(y_val, scr_val) if scr_val is not None else np.nan,
    }
    cv_rows.append(row)

cv_df = pd.DataFrame(cv_rows)

summary = pd.DataFrame({
    'Metric': ['Recall', 'F1', 'ROC-AUC'],
    'Train Mean': [cv_df['Train Recall'].mean(), cv_df['Train F1'].mean(), cv_df['Train ROC-AUC'].mean()],
    'Valid Mean': [cv_df['Valid Recall'].mean(), cv_df['Valid F1'].mean(), cv_df['Valid ROC-AUC'].mean()],
    'Gap (Train-Valid)': [
        cv_df['Train Recall'].mean() - cv_df['Valid Recall'].mean(),
        cv_df['Train F1'].mean() - cv_df['Valid F1'].mean(),
        cv_df['Train ROC-AUC'].mean() - cv_df['Valid ROC-AUC'].mean()
    ],
    'Valid Std': [cv_df['Valid Recall'].std(), cv_df['Valid F1'].std(), cv_df['Valid ROC-AUC'].std()]
})

print("="*90)
print(f"🔍 K-Fold Overfit Kontrolü | Model: {cv_model_name}")
print("="*90)
display(cv_df.round(4))
display(summary.round(4))

recall_gap = summary.loc[summary['Metric'] == 'Recall', 'Gap (Train-Valid)'].values[0]
auc_gap = summary.loc[summary['Metric'] == 'ROC-AUC', 'Gap (Train-Valid)'].values[0]
valid_recall_std = summary.loc[summary['Metric'] == 'Recall', 'Valid Std'].values[0]

overfit_flag = (recall_gap > 0.06) or (auc_gap > 0.04)
instability_flag = (valid_recall_std > 0.05)

print("🧠 Yorum:")
if overfit_flag:
    print(f"   ⚠️ Overfit sinyali var (Recall gap={recall_gap:.4f}, AUC gap={auc_gap:.4f}).")
else:
    print(f"   ✅ Belirgin overfit sinyali yok (Recall gap={recall_gap:.4f}, AUC gap={auc_gap:.4f}).")

if instability_flag:
    print(f"   ⚠️ Fold'lar arası recall oynaklığı yüksek (std={valid_recall_std:.4f}).")
else:
    print(f"   ✅ Recall fold stabilitesi kabul edilebilir (std={valid_recall_std:.4f}).")

print("="*90)
````

## 🎯 **Recall Odaklı İyileştirme: Cost-Sensitive Tuning + Threshold Optimizasyonu**

Bu bölümde:
- Pozitif sınıfa (Attrition=Yes) daha fazla ağırlık vererek **cost-sensitive tuning** yapıyoruz
- Ardından karar eşiğini optimize ederek recall/F2 dengesini iyileştiriyoruz

````python
# ============================================================
# RECALL ODAKLI TUNING: COST-SENSITIVE + THRESHOLD SEARCH
# ============================================================
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score

if 'best_model_name_20' not in globals() or 'benchmark_models' not in globals():
    raise ValueError("Önce final benchmark hücresini çalıştırmalısın (best model değişkenleri için).")

# Cost-sensitive tuning'i test set'e sızdırmadan train içinden valid ayırarak yap
X_tr_cs, X_val_cs, y_tr_cs, y_val_cs = train_test_split(
    X_train, y_train, test_size=0.2, random_state=RANDOM_STATE, stratify=y_train
)

preproc_cs = clone(preprocess)
X_tr_cs_t = preproc_cs.fit_transform(X_tr_cs)
X_val_cs_t = preproc_cs.transform(X_val_cs)

if hasattr(X_tr_cs_t, 'toarray'):
    X_tr_cs_t = X_tr_cs_t.toarray()
    X_val_cs_t = X_val_cs_t.toarray()

base_model_cs = benchmark_models[best_model_name_20]
weight_grid = [1.0, 1.5, 2.0, 3.0, 4.0, 5.0]
cs_rows = []

for pos_w in weight_grid:
    model_try = clone(base_model_cs)

    train_weights = np.where(y_tr_cs.values == 1, pos_w, 1.0)

    if hasattr(model_try, 'get_params') and ('class_weight' in model_try.get_params()):
        model_try.set_params(class_weight={0: 1.0, 1: pos_w})
        model_try.fit(X_tr_cs_t, y_tr_cs)
    else:
        try:
            model_try.fit(X_tr_cs_t, y_tr_cs, sample_weight=train_weights)
        except TypeError:
            model_try.fit(X_tr_cs_t, y_tr_cs)

    val_pred = model_try.predict(X_val_cs_t)

    if hasattr(model_try, 'predict_proba'):
        val_score = model_try.predict_proba(X_val_cs_t)[:, 1]
    elif hasattr(model_try, 'decision_function'):
        d = model_try.decision_function(X_val_cs_t)
        val_score = (d - np.min(d)) / (np.max(d) - np.min(d) + 1e-9)
    else:
        val_score = None

    rec = recall_score(y_val_cs, val_pred, zero_division=0)
    pre = precision_score(y_val_cs, val_pred, zero_division=0)
    f1v = f1_score(y_val_cs, val_pred, zero_division=0)
    aucv = roc_auc_score(y_val_cs, val_score) if val_score is not None else np.nan

    objective = (0.70 * rec) + (0.20 * f1v) + (0.10 * pre)

    cs_rows.append({
        'Positive Weight': pos_w,
        'Recall': rec,
        'Precision': pre,
        'F1': f1v,
        'ROC-AUC': aucv,
        'Objective': objective
    })

cs_df = pd.DataFrame(cs_rows).sort_values(['Objective', 'Recall', 'F1'], ascending=False).reset_index(drop=True)
best_pos_weight = float(cs_df.iloc[0]['Positive Weight'])

print("="*90)
print(f"🎯 Cost-Sensitive Tuning | Model: {best_model_name_20}")
print("="*90)
display(cs_df.round(4))
print(f"✅ Seçilen pozitif sınıf ağırlığı: {best_pos_weight}")

# 1) Eşik optimizasyonu: validation set üzerinde (test leakage yok)
model_for_threshold = clone(base_model_cs)
if hasattr(model_for_threshold, 'get_params') and ('class_weight' in model_for_threshold.get_params()):
    model_for_threshold.set_params(class_weight={0: 1.0, 1: best_pos_weight})
    model_for_threshold.fit(X_tr_cs_t, y_tr_cs)
else:
    tr_w = np.where(y_tr_cs.values == 1, best_pos_weight, 1.0)
    try:
        model_for_threshold.fit(X_tr_cs_t, y_tr_cs, sample_weight=tr_w)
    except TypeError:
        model_for_threshold.fit(X_tr_cs_t, y_tr_cs)

if hasattr(model_for_threshold, 'predict_proba'):
    val_score_best = model_for_threshold.predict_proba(X_val_cs_t)[:, 1]
elif hasattr(model_for_threshold, 'decision_function'):
    dval = model_for_threshold.decision_function(X_val_cs_t)
    val_score_best = (dval - np.min(dval)) / (np.max(dval) - np.min(dval) + 1e-9)
else:
    val_score_best = None

if val_score_best is None:
    best_threshold = 0.5
    th_df = None
    print("⚠️ Model skor üretemediği için threshold optimizasyonu atlandı (0.50 kullanılacak).")
else:
    thresholds = np.linspace(0.05, 0.95, 181)
    th_rows = []
    for th in thresholds:
        pred_th = (val_score_best >= th).astype(int)
        rec_th = recall_score(y_val_cs, pred_th, zero_division=0)
        pre_th = precision_score(y_val_cs, pred_th, zero_division=0)
        f1_th = f1_score(y_val_cs, pred_th, zero_division=0)
        f2_th = (5 * pre_th * rec_th) / (4 * pre_th + rec_th + 1e-9)  # beta=2
        th_rows.append({
            'Threshold': th,
            'Recall': rec_th,
            'Precision': pre_th,
            'F1': f1_th,
            'F2': f2_th
        })

    th_df = pd.DataFrame(th_rows)
    precision_floor = 0.25
    th_candidate = th_df[th_df['Precision'] >= precision_floor].copy()
    if th_candidate.empty:
        th_candidate = th_df.copy()

    best_idx = th_candidate.sort_values(['F2', 'Recall', 'F1'], ascending=False).index[0]
    best_threshold = float(th_df.loc[best_idx, 'Threshold'])

    print(f"\n🎛️ Validation'dan seçilen eşik (recall odaklı): {best_threshold:.3f}")
    print(f"   (Precision floor: {precision_floor})")

# 2) Seçilen weight + threshold ile full-train fit, test evaluate
preproc_final_tuned = clone(preprocess)
X_train_tuned = preproc_final_tuned.fit_transform(X_train)
X_test_tuned = preproc_final_tuned.transform(X_test)

if hasattr(X_train_tuned, 'toarray'):
    X_train_tuned = X_train_tuned.toarray()
    X_test_tuned = X_test_tuned.toarray()

model_tuned = clone(base_model_cs)
if hasattr(model_tuned, 'get_params') and ('class_weight' in model_tuned.get_params()):
    model_tuned.set_params(class_weight={0: 1.0, 1: best_pos_weight})
    model_tuned.fit(X_train_tuned, y_train)
else:
    train_w_full = np.where(y_train.values == 1, best_pos_weight, 1.0)
    try:
        model_tuned.fit(X_train_tuned, y_train, sample_weight=train_w_full)
    except TypeError:
        model_tuned.fit(X_train_tuned, y_train)

if hasattr(model_tuned, 'predict_proba'):
    test_score_tuned = model_tuned.predict_proba(X_test_tuned)[:, 1]
elif hasattr(model_tuned, 'decision_function'):
    dtest = model_tuned.decision_function(X_test_tuned)
    test_score_tuned = (dtest - np.min(dtest)) / (np.max(dtest) - np.min(dtest) + 1e-9)
else:
    test_score_tuned = None

pred_05 = model_tuned.predict(X_test_tuned)
rec_05 = recall_score(y_test, pred_05, zero_division=0)
pre_05 = precision_score(y_test, pred_05, zero_division=0)
f1_05 = f1_score(y_test, pred_05, zero_division=0)
auc_05 = roc_auc_score(y_test, test_score_tuned) if test_score_tuned is not None else np.nan

if test_score_tuned is None:
    pred_best_th = pred_05
else:
    pred_best_th = (test_score_tuned >= best_threshold).astype(int)

rec_bt = recall_score(y_test, pred_best_th, zero_division=0)
pre_bt = precision_score(y_test, pred_best_th, zero_division=0)
f1_bt = f1_score(y_test, pred_best_th, zero_division=0)

print("\n📌 Test Performansı (Cost-Sensitive, threshold=0.50):")
print(f"Recall={rec_05:.4f} | Precision={pre_05:.4f} | F1={f1_05:.4f} | ROC-AUC={'-' if np.isnan(auc_05) else f'{auc_05:.4f}'}")

print("\n✅ Test Performansı (Cost-Sensitive + Validation-Optimized Threshold):")
print(f"Recall={rec_bt:.4f} | Precision={pre_bt:.4f} | F1={f1_bt:.4f}")

# Eşik eğrisi (validation üstünden)
if th_df is not None:
    fig_th = go.Figure()
    fig_th.add_trace(go.Scatter(x=th_df['Threshold'], y=th_df['Recall'], mode='lines', name='Recall (val)', line=dict(color=COLORS['danger'], width=3)))
    fig_th.add_trace(go.Scatter(x=th_df['Threshold'], y=th_df['Precision'], mode='lines', name='Precision (val)', line=dict(color=COLORS['primary'], width=2)))
    fig_th.add_trace(go.Scatter(x=th_df['Threshold'], y=th_df['F2'], mode='lines', name='F2 (val)', line=dict(color=COLORS['success'], width=2)))
    fig_th.add_vline(x=best_threshold, line_dash='dash', line_color='black', annotation_text=f"Best th={best_threshold:.3f}", annotation_position='top right')

    fig_th.update_layout(
        title=f"Validation Threshold Eğrisi | {best_model_name_20}",
        xaxis_title='Threshold',
        yaxis_title='Skor',
        template=PLOTLY_TEMPLATE,
        height=520
    )
    save_plotly_png(fig_th, f"final_threshold_curve_validation_{best_model_name_20}")
    fig_th.show()

compare_df = pd.DataFrame([
    {'Setup': 'Cost-Sensitive (th=0.50)', 'Recall': rec_05, 'Precision': pre_05, 'F1': f1_05, 'ROC-AUC': auc_05},
    {'Setup': f'Cost-Sensitive (th={best_threshold:.3f})', 'Recall': rec_bt, 'Precision': pre_bt, 'F1': f1_bt, 'ROC-AUC': auc_05}
])

print("\n📊 Karşılaştırma:")
display(compare_df.round(4))
print("="*90)
````

## 🧾 **Final Değerlendirme: Son Modelin Confusion Matrix ve Metrikleri**

Bu bölümde en son seçilen kurulum için:
- Confusion Matrix (adet + yüzde)
- Accuracy, Precision, Recall, F1, ROC-AUC
- Classification Report
tek noktadan raporlanır.

````python
# ============================================================
# FINAL RAPOR: SON MODEL (TUNED) - CM + METRİKLER
# ============================================================
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report

# Gerekli objeler yoksa bilgilendir
required_final_vars = ['y_test']
missing_final_vars = [v for v in required_final_vars if v not in globals()]
if missing_final_vars:
    raise ValueError(f"Gerekli değişken(ler) eksik: {missing_final_vars}")

# Tahmin vektörü: mümkünse mevcut pred_best_th kullan
if 'pred_best_th' in globals():
    y_pred_final_tuned = pred_best_th.copy()
else:
    if 'model_tuned' not in globals() or 'X_test_tuned' not in globals() or 'best_threshold' not in globals():
        raise ValueError("pred_best_th bulunamadı. Önce recall tuning hücresini çalıştırmalısın.")

    if hasattr(model_tuned, 'predict_proba'):
        score_final = model_tuned.predict_proba(X_test_tuned)[:, 1]
    elif hasattr(model_tuned, 'decision_function'):
        dscore = model_tuned.decision_function(X_test_tuned)
        score_final = (dscore - np.min(dscore)) / (np.max(dscore) - np.min(dscore) + 1e-9)
    else:
        score_final = None

    if score_final is None:
        y_pred_final_tuned = model_tuned.predict(X_test_tuned)
    else:
        y_pred_final_tuned = (score_final >= best_threshold).astype(int)

# Skor/olasılık vektörü (AUC için)
if 'test_score_tuned' in globals() and test_score_tuned is not None:
    score_eval = test_score_tuned
else:
    if 'model_tuned' in globals() and 'X_test_tuned' in globals():
        if hasattr(model_tuned, 'predict_proba'):
            score_eval = model_tuned.predict_proba(X_test_tuned)[:, 1]
        elif hasattr(model_tuned, 'decision_function'):
            dscore2 = model_tuned.decision_function(X_test_tuned)
            score_eval = (dscore2 - np.min(dscore2)) / (np.max(dscore2) - np.min(dscore2) + 1e-9)
        else:
            score_eval = None
    else:
        score_eval = None

# Metrikler
acc_final_tuned = accuracy_score(y_test, y_pred_final_tuned)
pre_final_tuned = precision_score(y_test, y_pred_final_tuned, zero_division=0)
rec_final_tuned = recall_score(y_test, y_pred_final_tuned, zero_division=0)
f1_final_tuned = f1_score(y_test, y_pred_final_tuned, zero_division=0)
auc_final_tuned = roc_auc_score(y_test, score_eval) if score_eval is not None else np.nan

final_metrics_df = pd.DataFrame([{
    'Model': f"{best_model_name_20} (Cost-Sensitive + th={best_threshold:.3f})" if 'best_model_name_20' in globals() and 'best_threshold' in globals() else 'Final Tuned Model',
    'Accuracy': acc_final_tuned,
    'Precision': pre_final_tuned,
    'Recall': rec_final_tuned,
    'F1': f1_final_tuned,
    'ROC-AUC': auc_final_tuned
}])

print("="*95)
print("🧾 FINAL METRİK ÖZETİ")
print("="*95)
display(final_metrics_df.round(4))

print("\n📋 Classification Report:")
print(classification_report(y_test, y_pred_final_tuned, digits=4))

# Confusion Matrix (adet + yüzde)
cm_final = confusion_matrix(y_test, y_pred_final_tuned)
cm_final_norm = cm_final / cm_final.sum(axis=1, keepdims=True)
cm_final_labels_y = ['Gerçek No', 'Gerçek Yes']
cm_final_labels_x = ['Tahmin No', 'Tahmin Yes']

cm_final_df = pd.DataFrame(cm_final, index=cm_final_labels_y, columns=cm_final_labels_x)
print("\n🔢 Confusion Matrix Değerleri (adet):")
display(cm_final_df)

cm_text = [[f"{cm_final[i, j]}\n({cm_final_norm[i, j]*100:.1f}%)" for j in range(cm_final.shape[1])] for i in range(cm_final.shape[0])]

fig_cm_final = go.Figure(data=go.Heatmap(
    z=cm_final,
    x=cm_final_labels_x,
    y=cm_final_labels_y,
    text=cm_text,
    texttemplate="%{text}",
    colorscale=[[0, '#eef2ff'], [0.25, '#c7d2fe'], [0.5, '#8da2ff'], [0.75, '#4b4dbf'], [1, '#1f2a78']],
    showscale=True,
    hovertemplate='<b>%{y}</b><br>Tahmin: %{x}<br>Adet: %{z}<extra></extra>'
))

final_model_name = best_model_name_20 if 'best_model_name_20' in globals() else 'FinalModel'
final_th_text = f"th={best_threshold:.3f}" if 'best_threshold' in globals() else "th=default"

fig_cm_final.update_layout(
    title=f"Final Confusion Matrix - {final_model_name} ({final_th_text})",
    xaxis_title='Tahmin',
    yaxis_title='Gerçek',
    template=PLOTLY_TEMPLATE,
    height=540
)

save_plotly_png(fig_cm_final, f"final_cm_tuned_{final_model_name}_{final_th_text.replace('=', '_').replace('.', '_')}")
fig_cm_final.show()

print("="*95)
````
