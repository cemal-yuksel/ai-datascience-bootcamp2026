## 📚 Kütüphanelerin Yüklenmesi

**YBS Perspektifi:** Bir projeye başlarken doğru araçları seçmek kritiktir. Her kütüphane farklı bir amaca hizmet eder:

| 🔧 Kütüphane | 💼 İş Amaçları | 🎯 Kullanım Alanı |
|-------------|---------------|------------------|
| **Pandas** | Veri manipülasyonu ve analizi | Excel'in kod versiyonu - veri tablolarıyla çalışma |

````python
# Temel veri işleme ve analiz kütüphaneleri
import pandas as pd
import numpy as np

# Görselleştirme kütüphaneleri
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Sistem ve yardımcı kütüphaneler
import warnings
import os
from datetime import datetime

# Uyarıları gizle (temiz çıktı için)
warnings.filterwarnings('ignore')

# Görselleştirme ayarları
plt.style.use('seaborn-v0_8-darkgrid')  # Matplotlib için güzel bir stil
sns.set_palette("husl")  # Seaborn için renkli palet
pd.set_option('display.max_columns', None)  # Tüm sütunları göster
pd.set_option('display.max_rows', 100)  # Maksimum 100 satır göster
pd.set_option('display.float_format', lambda x: '%.2f' % x)  # Ondalık formatı

print("✅ Tüm kütüphaneler başarıyla yüklendi!")
print(f"📊 Pandas versiyonu: {pd.__version__}")
print(f"🔢 NumPy versiyonu: {np.__version__}")
print(f"📈 Matplotlib versiyonu: {plt.matplotlib.__version__}")
````

## 🌐 Titanic Dataset'ini İndirme

<div style='background: linear-gradient(135deg, #c2e9fb 0%, #a1c4fd 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #2196F3; margin: 15px 0;'>
    <h4 style='color: #1565c0; margin-top: 0;'>📌 Veri Kaynağı Notu</h4>
    <p style='color: #1976d2; margin-bottom: 0;'>
        <strong>Kaggle</strong>, dünyanın en büyük veri bilimi ve makine öğrenmesi topluluğudur. 
        Titanic dataset'i, Kaggle'ın en ünlü "başlangıç" yarışmasıdır ve 
        <strong>891 yolcunun</strong> verilerini içerir.
    </p>
</div>

````python
# Titanic dataset'ini indirme ve kaydetme
import urllib.request

# GitHub'dan güvenilir Titanic dataset URL'i
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

# Dosya adı
dosya_adi = "titanic.csv"

# Eğer dosya yoksa indir
if not os.path.exists(dosya_adi):
    print("⏳ Titanic dataset'i indiriliyor...")
    try:
        urllib.request.urlretrieve(url, dosya_adi)
        print(f"✅ Dataset başarıyla indirildi ve '{dosya_adi}' olarak kaydedildi!")
    except Exception as e:
        print(f"❌ İndirme hatası: {e}")
else:
    print(f"✅ '{dosya_adi}' zaten mevcut!")

# Dosya bilgilerini göster
dosya_boyutu = os.path.getsize(dosya_adi) / 1024  # KB cinsinden
print(f"\n📁 Dosya Bilgileri:")
print(f"   └─ Boyut: {dosya_boyutu:.2f} KB")
print(f"   └─ Konum: {os.path.abspath(dosya_adi)}")
````

## 📂 Veriyi Pandas DataFrame'e Yükleme

**DataFrame Nedir?** 
- Excel'deki tablo gibi satır ve sütunlardan oluşan 2 boyutlu veri yapısı
- Pandas'ın kalbi, tüm veri analizi işlemleri burada gerçekleşir
- Her sütun farklı veri tipinde olabilir (sayısal, metin, tarih vb.)

````python
# CSV dosyasını pandas DataFrame'e yükleme
df = pd.read_csv(dosya_adi)

print("🎉 Veri başarıyla yüklendi!")
print("\n" + "="*70)
print("📊 VERİ SETİ HAKKINDA İLK BİLGİLER")
print("="*70)
print(f"\n📏 Boyut         : {df.shape[0]} satır × {df.shape[1]} sütun")
print(f"💾 Bellek Kullanımı : {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
print(f"🔢 Toplam Hücre  : {df.shape[0] * df.shape[1]:,} adet")
print("\n" + "="*70)
````

## 🎨 Veri Tiplerinin Görsel Dağılımı

Veri setimizde ne tür değişkenler var? Bunları görselleştirerek daha iyi anlayalım!

````python
# Veri tiplerinin dağılımını görselleştirme
dtype_counts = df.dtypes.value_counts()

# Plotly ile interaktif pie chart
fig = go.Figure(data=[go.Pie(
    labels=[f'{dtype}' for dtype in dtype_counts.index],
    values=dtype_counts.values,
    hole=0.4,
    marker=dict(colors=['#FFB6C1', '#B0E0E6', '#FFD700', '#98FB98']),
    textinfo='label+value+percent',
    textfont=dict(size=14, color='#000000'),
    hovertemplate='<b>%{label}</b><br>Sayı: %{value}<br>Yüzde: %{percent}<extra></extra>'
)])

fig.update_layout(
    title=dict(
        text='Veri Tiplerinin Dağılımı',
        font=dict(size=20, color='#2c3e50', family='Arial Black')
    ),
    height=500,
    showlegend=True,
    paper_bgcolor='#fafafa',
    plot_bgcolor='#fafafa'
)

fig.show()

print("\n📋 VERİ TİPİ DETAYLARI:")
print("="*70)
for dtype, count in dtype_counts.items():
    print(f"   {dtype}: {count} sütun")
````

## 🔄 Özellik Mühendisliği: Yeni Değişkenler Oluşturma

Şimdi teoride tartıştığımız yeni değişkenleri gerçekten oluşturalım!

````python
# Orijinal veri setini korumak için kopya oluştur
df_engineered = df.copy()

print("🔧 ÖZELLİK MÜHENDİSLİĞİ BAŞLIYOR...")
print("="*80)

# 1. Aile Büyüklüğü
df_engineered['Family_Size'] = df_engineered['SibSp'] + df_engineered['Parch'] + 1
print("\n✅ 1. Family_Size (Aile Büyüklüğü) oluşturuldu")
print(f"   Minimum: {df_engineered['Family_Size'].min()}, Maksimum: {df_engineered['Family_Size'].max()}")

# 2. Yalnız mı?
df_engineered['Is_Alone'] = (df_engineered['Family_Size'] == 1).astype(int)
print("\n✅ 2. Is_Alone (Yalnız mı?) oluşturuldu")
print(f"   Yalnız yolcular: {df_engineered['Is_Alone'].sum()} kişi ({df_engineered['Is_Alone'].sum()/len(df_engineered)*100:.1f}%)")

# 3. Unvan çıkarımı
df_engineered['Title'] = df_engineered['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)
# Nadir unvanları 'Other' olarak grupla
title_mapping = {
    'Mr': 'Mr', 'Miss': 'Miss', 'Mrs': 'Mrs', 'Master': 'Master',
    'Dr': 'Rare', 'Rev': 'Rare', 'Col': 'Rare', 'Major': 'Rare',
    'Mlle': 'Miss', 'Countess': 'Rare', 'Ms': 'Miss', 'Lady': 'Rare',
    'Jonkheer': 'Rare', 'Don': 'Rare', 'Dona': 'Rare', 'Mme': 'Mrs',
    'Capt': 'Rare', 'Sir': 'Rare'
}
df_engineered['Title'] = df_engineered['Title'].map(title_mapping)
df_engineered['Title'].fillna('Other', inplace=True)
print("\n✅ 3. Title (Unvan) oluşturuldu")
print(f"   Unvanlar: {df_engineered['Title'].unique()}")

# 4. Kabin katı (ilk harf)
df_engineered['Deck'] = df_engineered['Cabin'].str[0]
df_engineered['Deck'].fillna('Unknown', inplace=True)
print("\n✅ 4. Deck (Kabin Katı) oluşturuldu")
print(f"   Katlar: {sorted(df_engineered['Deck'].unique())}")

# 5. Yaş grubu
def categorize_age(age):
    if pd.isna(age):
        return 'Unknown'
    elif age < 13:
        return 'Çocuk'
    elif age < 19:
        return 'Genç'
    elif age < 60:
        return 'Yetişkin'
    else:
        return 'Yaşlı'

df_engineered['Age_Group'] = df_engineered['Age'].apply(categorize_age)
print("\n✅ 5. Age_Group (Yaş Grubu) oluşturuldu")
print(f"   Gruplar: {df_engineered['Age_Group'].unique()}")

# 6. Ücret grubu
df_engineered['Fare_Group'] = pd.qcut(df_engineered['Fare'], q=4, labels=['Düşük', 'Orta-Alt', 'Orta-Üst', 'Yüksek'], duplicates='drop')
print("\n✅ 6. Fare_Group (Ücret Grubu) oluşturuldu")
print(f"   Gruplar: {df_engineered['Fare_Group'].unique()}")

print("\n" + "="*80)
print("🎉 ÖZELLİK MÜHENDİSLİĞİ TAMAMLANDI!")
print(f"\n📊 Yeni Veri Seti Boyutu: {df_engineered.shape[0]} satır × {df_engineered.shape[1]} sütun")
print(f"📈 Eklenen Değişken Sayısı: {df_engineered.shape[1] - df.shape[1]}")
print("="*80)
````

## 💡 Eksik Veri Nedir ve Neden Önemlidir?

<div style='background: linear-gradient(135deg, #ffecd2 0%, #fcb69f 100%); padding: 20px; border-radius: 10px; margin: 20px 0; box-shadow: 0 5px 15px rgba(0,0,0,0.2);'>
    
<h3 style='color: #5d4037; margin-top: 0;'>🎯 YBS Perspektifi: Eksik Veri Yönetimi</h3>

**Eksik veri**, veri setinde bazı gözlemlerde değerlerin bulunmaması durumudur. Gerçek dünya verilerinde kaçınılmazdır.

### 📊 Eksik Veri Türleri

**1. 🔀 MCAR (Missing Completely At Random - Tamamen Rastgele Eksik)**
- Eksiklik tamamen tesadüfidir
- Başka değişkenlerle ilişkisi yoktur
- **Örnek:** Anket formunda bazı sorular yanlışlıkla atlanmış

**2. 📉 MAR (Missing At Random - Rastgele Eksik)**
- Eksiklik gözlemlenen diğer değişkenlerle ilişkilidir
- **Örnek:** Yaşlı yolcular yaş bilgisini daha az paylaşmış olabilir

**3. 🎯 MNAR (Missing Not At Random - Rastgele Olmayan Eksik)**
- Eksiklik, eksik olan değerin kendisiyle ilişkilidir
- **Örnek:** Yüksek gelirli kişiler gelir bilgisini paylaşmak istememiş

### 💼 İş Etkisi

| ⚠️ Risk | 📊 Etki | 🛠️ Çözüm |
|--------|--------|---------|
| **Veri Kaybı** | Analiz gücü azalır | Doğru doldurma stratejisi |
| **Yanlılık (Bias)** | Sonuçlar çarpık olabilir | Eksik veri paternini anla |
| **Model Performansı** | Tahmin doğruluğu düşer | Uygun imputation yöntemi |

</div>

## 🔍 Eksik Veri Tespiti

İlk olarak, veri setimizdeki eksik değerleri tespit edelim.

````python
# Eksik veri analizi
missing_data = pd.DataFrame({
    'Değişken': df.columns,
    'Eksik Sayı': df.isnull().sum().values,
    'Eksik Yüzde': (df.isnull().sum() / len(df) * 100).round(2).values,
    'Veri Tipi': df.dtypes.values,
    'Toplam Değer': len(df)
})

# Sadece eksik verisi olanları filtrele
missing_data = missing_data[missing_data['Eksik Sayı'] > 0].sort_values('Eksik Yüzde', ascending=False)

print("🕳️ EKSİK VERİ ANALİZİ")
print("="*90)
print(f"\n📊 Toplam Değişken Sayısı: {len(df.columns)}")
print(f"⚠️  Eksik Verisi Olan Değişken: {len(missing_data)}")
print(f"✅ Tam Verisi Olan Değişken: {len(df.columns) - len(missing_data)}")
print("\n" + "="*90)
print("📋 EKSİK VERİ DETAYLARI")
print("="*90)
print(missing_data.to_string(index=False))
print("="*90)
````

## 📊 Eksik Veri Etkisi: Hayatta Kalma Analizi

Eksik verisi olan ve olmayan yolcuların hayatta kalma oranları farklı mı?

````python
# Eksik veri ile hayatta kalma arasındaki ilişki
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# 1. Age eksikliği ve hayatta kalma
age_missing = df['Age'].isnull()
age_survival = pd.crosstab(age_missing, df['Survived'], normalize='index') * 100

age_survival.plot(kind='bar', ax=axes[0], color=['#e74c3c', '#2ecc71'], width=0.7)
axes[0].set_title('Yaş Bilgisi ve Hayatta Kalma', fontsize=14, fontweight='bold', color='#2c3e50')
axes[0].set_xlabel('Yaş Bilgisi Eksik mi?', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Yolcu Yüzdesi (%)', fontsize=12, fontweight='bold')
axes[0].set_xticklabels(['Mevcut', 'Eksik'], rotation=0, fontsize=11)
axes[0].legend(['Hayatta Kalmadı', 'Hayatta Kaldı'], loc='upper right', fontsize=10)
axes[0].grid(axis='y', alpha=0.3)

# 2. Cabin eksikliği ve hayatta kalma
cabin_missing = df['Cabin'].isnull()
cabin_survival = pd.crosstab(cabin_missing, df['Survived'], normalize='index') * 100

cabin_survival.plot(kind='bar', ax=axes[1], color=['#e74c3c', '#2ecc71'], width=0.7)
axes[1].set_title('Kabin Bilgisi ve Hayatta Kalma', fontsize=14, fontweight='bold', color='#2c3e50')
axes[1].set_xlabel('Kabin Bilgisi Eksik mi?', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Yolcu Yüzdesi (%)', fontsize=12, fontweight='bold')
axes[1].set_xticklabels(['Mevcut', 'Eksik'], rotation=0, fontsize=11)
axes[1].legend(['Hayatta Kalmadı', 'Hayatta Kaldı'], loc='upper right', fontsize=10)
axes[1].grid(axis='y', alpha=0.3)

# 3. Toplam eksik veri sayısı ve hayatta kalma
total_missing = df.isnull().sum(axis=1)
df_temp = pd.DataFrame({'Total_Missing': total_missing, 'Survived': df['Survived']})
missing_survival = df_temp.groupby('Total_Missing')['Survived'].mean() * 100

axes[2].bar(missing_survival.index, missing_survival.values, color='#3498db', edgecolor='#2c3e50', linewidth=1.5)
axes[2].set_title('Toplam Eksik Veri ve Hayatta Kalma', fontsize=14, fontweight='bold', color='#2c3e50')
axes[2].set_xlabel('Eksik Veri Sayısı', fontsize=12, fontweight='bold')
axes[2].set_ylabel('Hayatta Kalma Oranı (%)', fontsize=12, fontweight='bold')
axes[2].grid(axis='y', alpha=0.3)
axes[2].axhline(y=df['Survived'].mean() * 100, color='#e74c3c', linestyle='--', linewidth=2, label='Genel Ortalama')
axes[2].legend(fontsize=10)

plt.tight_layout()
plt.show()

# İstatistikler
print("\n📊 EKSİK VERİ ve HAYATTA KALMA İLİŞKİSİ")
print("="*70)
print(f"\n1️⃣  Yaş Bilgisi:")
print(f"   • Yaş mevcut olanlar: {df[~age_missing]['Survived'].mean()*100:.1f}% hayatta kaldı")
print(f"   • Yaş eksik olanlar: {df[age_missing]['Survived'].mean()*100:.1f}% hayatta kaldı")

print(f"\n2️⃣  Kabin Bilgisi:")
print(f"   • Kabin mevcut olanlar: {df[~cabin_missing]['Survived'].mean()*100:.1f}% hayatta kaldı")
print(f"   • Kabin eksik olanlar: {df[cabin_missing]['Survived'].mean()*100:.1f}% hayatta kaldı")

print(f"\n3️⃣  Genel Durum:")
print(f"   • Eksik veri yok (0): {df[total_missing == 0]['Survived'].mean()*100:.1f}% hayatta kaldı")
print(f"   • En az 1 eksik veri: {df[total_missing > 0]['Survived'].mean()*100:.1f}% hayatta kaldı")
print("="*70)
````

## 🎯 Univariate Analysis Nedir?

**Tekli değişken analizi**, tek bir değişkeni izole ederek inceleme yöntemidir. Diğer değişkenlerle ilişkisine bakmadan, sadece o değişkenin:

- 📊 **Dağılımını** (distribution)
- 📈 **Merkezi eğilimini** (mean, median, mode)
- 📉 **Yayılımını** (range, variance, std)
- 🎲 **Frekanslarını** (value counts)

analiz ederiz.

## 🎯 Hedef Değişken: Survived (Hayatta Kalma)

**En önemli değişkenimizle başlayalım!** Bu, tahmin etmeye çalıştığımız değişken.

````python
# Hayatta kalma dağılımı
survived_counts = df['Survived'].value_counts()
survived_percentage = df['Survived'].value_counts(normalize=True) * 100

# Görselleştirme
fig = make_subplots(
    rows=1, cols=2,
    specs=[[{"type": "pie"}, {"type": "bar"}]],
    subplot_titles=('Hayatta Kalma Oranı', 'Hayatta Kalma Sayıları')
)

# Pie chart
fig.add_trace(
    go.Pie(
        labels=['Hayatta Kalmadı', 'Hayatta Kaldı'],
        values=survived_counts.values,
        marker=dict(colors=['#e74c3c', '#2ecc71'], line=dict(color='#ffffff', width=3)),
        hole=0.4,
        textinfo='label+percent',
        textfont=dict(size=14, color='#000000', family='Arial Black'),
        hovertemplate='<b>%{label}</b><br>Sayı: %{value}<br>Yüzde: %{percent}<extra></extra>'
    ),
    row=1, col=1
)

# Bar chart
fig.add_trace(
    go.Bar(
        x=['Hayatta Kalmadı', 'Hayatta Kaldı'],
        y=survived_counts.values,
        marker=dict(color=['#e74c3c', '#2ecc71'], line=dict(color='#2c3e50', width=2)),
        text=survived_counts.values,
        textposition='outside',
        textfont=dict(size=14, color='#2c3e50', family='Arial Black'),
        hovertemplate='<b>%{x}</b><br>Sayı: %{y}<extra></extra>'
    ),
    row=1, col=2
)

fig.update_layout(
    title=dict(
        text='Titanic: Hayatta Kalma Analizi',
        font=dict(size=22, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    height=500,
    showlegend=False,
    paper_bgcolor='#fafafa'
)

fig.update_xaxes(title=dict(text='Durum', font=dict(size=12)), row=1, col=2)
fig.update_yaxes(title=dict(text='Yolcu Sayısı', font=dict(size=12)), row=1, col=2)

fig.show()

# İstatistikler
print("\n🎯 HAYATTA KALMA İSTATİSTİKLERİ")
print("="*70)
print(f"   💀 Hayatta Kalmadı: {survived_counts[0]} kişi ({survived_percentage[0]:.1f}%)")
print(f"   ✅ Hayatta Kaldı  : {survived_counts[1]} kişi ({survived_percentage[1]:.1f}%)")
print(f"\n   📊 Hayatta Kalma Oranı: {survived_percentage[1]:.2f}%")
print(f"   📉 Ölüm Oranı: {survived_percentage[0]:.2f}%")
print("="*70)
````

## 🎫 Bilet Sınıfı: Pclass

**Sosyoekonomik statü göstergesi** - 1. sınıf (lüks), 2. sınıf (orta), 3. sınıf (ekonomi)

````python
# Pclass dağılımı
pclass_counts = df['Pclass'].value_counts().sort_index()
pclass_percentage = df['Pclass'].value_counts(normalize=True).sort_index() * 100

# Görselleştirme
fig = go.Figure()

fig.add_trace(go.Bar(
    x=['1. Sınıf', '2. Sınıf', '3. Sınıf'],
    y=pclass_counts.values,
    marker=dict(
        color=['#FFD700', '#C0C0C0', '#CD7F32'],  # Altın, Gümüş, Bronz
        line=dict(color='#2c3e50', width=2)
    ),
    text=[f'{count}<br>({pct:.1f}%)' for count, pct in zip(pclass_counts.values, pclass_percentage.values)],
    textposition='outside',
    textfont=dict(size=14, color='#2c3e50', family='Arial Black'),
    hovertemplate='<b>%{x}</b><br>Yolcu: %{y}<br>Yüzde: %{text}<extra></extra>'
))

fig.update_layout(
    title=dict(
        text='Yolcuların Bilet Sınıfı Dağılımı',
        font=dict(size=20, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    xaxis=dict(title=dict(text='Bilet Sınıfı', font=dict(size=13, color='#2c3e50'))),
    yaxis=dict(title=dict(text='Yolcu Sayısı', font=dict(size=13, color='#2c3e50'))),
    height=500,
    paper_bgcolor='#fafafa',
    plot_bgcolor='#ffffff',
    showlegend=False
)

fig.show()

# İstatistikler
print("\n🎫 BİLET SINIFI İSTATİSTİKLERİ")
print("="*70)
for pclass in [1, 2, 3]:
    count = pclass_counts[pclass]
    pct = pclass_percentage[pclass]
    print(f"   {pclass}. Sınıf: {count:3d} yolcu ({pct:5.1f}%)")
print(f"\n   📊 En kalabalık: 3. Sınıf ({pclass_counts[3]} kişi)")
print(f"   📊 Modal değer (Mod): {df['Pclass'].mode()[0]}. Sınıf")
print("="*70)
````

### 👶 Yaş: Age

**Sürekli değişken** - Yolcuların yaşları (0.42 ile 80 arasında)

````python
# Yaş istatistikleri (NaN hariç)
age_data = df['Age'].dropna()

# Görselleştirme
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Yaş Dağılımı - Histogram', 'Yaş Kutu Grafiği (Box Plot)',
                    'Yaş Yoğunluk Grafiği (KDE)', 'Yaş İstatistikleri'),
    specs=[[{"type": "histogram"}, {"type": "box"}],
           [{"type": "scatter"}, {"type": "table"}]],
    vertical_spacing=0.12,
    horizontal_spacing=0.1
)

# 1. Histogram
fig.add_trace(
    go.Histogram(
        x=age_data,
        nbinsx=30,
        marker=dict(color='#3498db', line=dict(color='#2c3e50', width=1)),
        name='Histogram',
        hovertemplate='Yaş: %{x}<br>Frekans: %{y}<extra></extra>'
    ),
    row=1, col=1
)

# 2. Box Plot
fig.add_trace(
    go.Box(
        y=age_data,
        marker=dict(color='#e74c3c'),
        name='Box Plot',
        boxmean='sd',  # Ortalama ve standart sapma göster
        hovertemplate='Yaş: %{y}<extra></extra>'
    ),
    row=1, col=2
)

# 3. KDE (Yoğunluk) - Matplotlib ile hesaplayıp Plotly'ye aktarıyoruz
from scipy import stats
density = stats.gaussian_kde(age_data)
xs = np.linspace(age_data.min(), age_data.max(), 200)
fig.add_trace(
    go.Scatter(
        x=xs,
        y=density(xs),
        mode='lines',
        fill='tozeroy',
        line=dict(color='#2ecc71', width=3),
        name='KDE',
        hovertemplate='Yaş: %{x:.1f}<br>Yoğunluk: %{y:.4f}<extra></extra>'
    ),
    row=2, col=1
)

# 4. İstatistik tablosu
stats_data = [
    ['Ortalama (Mean)', f'{age_data.mean():.2f}'],
    ['Medyan', f'{age_data.median():.2f}'],
    ['Mod', f'{age_data.mode()[0]:.2f}'],
    ['Std Sapma', f'{age_data.std():.2f}'],
    ['Minimum', f'{age_data.min():.2f}'],
    ['Maksimum', f'{age_data.max():.2f}'],
    ['Q1 (25%)', f'{age_data.quantile(0.25):.2f}'],
    ['Q3 (75%)', f'{age_data.quantile(0.75):.2f}'],
    ['IQR', f'{age_data.quantile(0.75) - age_data.quantile(0.25):.2f}']
]

fig.add_trace(
    go.Table(
        header=dict(
            values=['<b>Metrik</b>', '<b>Değer</b>'],
            fill_color='#3498db',
            align='left',
            font=dict(color='white', size=13, family='Arial Black')
        ),
        cells=dict(
            values=list(zip(*stats_data)),
            fill_color=['#ecf0f1', '#ffffff'],
            align='left',
            font=dict(size=12)
        )
    ),
    row=2, col=2
)

fig.update_layout(
    title=dict(
        text='Yaş Değişkeni: Kapsamlı Analiz',
        font=dict(size=20, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    height=800,
    showlegend=False,
    paper_bgcolor='#fafafa'
)

fig.update_xaxes(title=dict(text='Yaş', font=dict(size=11)), row=1, col=1)
fig.update_yaxes(title=dict(text='Frekans', font=dict(size=11)), row=1, col=1)
fig.update_yaxes(title=dict(text='Yaş', font=dict(size=11)), row=1, col=2)
fig.update_xaxes(title=dict(text='Yaş', font=dict(size=11)), row=2, col=1)
fig.update_yaxes(title=dict(text='Yoğunluk', font=dict(size=11)), row=2, col=1)

fig.show()

# Detaylı istatistikler
print("\n👶 YAŞ İSTATİSTİKLERİ")
print("="*70)
print(f"   📊 Toplam Veri: {len(age_data)} (Eksik: {df['Age'].isnull().sum()})")
print(f"   📈 Ortalama: {age_data.mean():.2f} yaş")
print(f"   📊 Medyan: {age_data.median():.2f} yaş")
print(f"   📌 Mod: {age_data.mode()[0]:.2f} yaş")
print(f"   📉 Standart Sapma: {age_data.std():.2f}")
print(f"   🔽 Minimum: {age_data.min():.2f} yaş")
print(f"   🔼 Maksimum: {age_data.max():.2f} yaş")
print(f"   📦 IQR (Q3-Q1): {age_data.quantile(0.75) - age_data.quantile(0.25):.2f}")
print("="*70)
````

### 💰 Bilet Ücreti: Fare

**Sürekli değişken** - Yolcuların ödediği bilet ücretleri (£0 ile £512 arasında)

````python
# Fare istatistikleri
fare_data = df['Fare']

# Görselleştirme - 2x2 grid
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Bilet Ücreti (Fare) Analizi', fontsize=20, fontweight='bold', color='#2c3e50', y=0.995)

# 1. Histogram
axes[0, 0].hist(fare_data, bins=50, color='#9b59b6', edgecolor='#2c3e50', alpha=0.7)
axes[0, 0].axvline(fare_data.mean(), color='#e74c3c', linestyle='--', linewidth=2, label=f'Ortalama: £{fare_data.mean():.2f}')
axes[0, 0].axvline(fare_data.median(), color='#2ecc71', linestyle='--', linewidth=2, label=f'Medyan: £{fare_data.median():.2f}')
axes[0, 0].set_xlabel('Bilet Ücreti (£)', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Frekans', fontsize=12, fontweight='bold')
axes[0, 0].set_title('Histogram', fontsize=14, fontweight='bold', color='#2c3e50')
axes[0, 0].legend(fontsize=10)
axes[0, 0].grid(axis='y', alpha=0.3)

# 2. Box Plot
box = axes[0, 1].boxplot(fare_data, vert=True, patch_artist=True)
box['boxes'][0].set_facecolor('#f39c12')
box['boxes'][0].set_edgecolor('#2c3e50')
box['boxes'][0].set_linewidth(2)
for element in ['whiskers', 'fliers', 'means', 'medians', 'caps']:
    plt.setp(box[element], color='#2c3e50', linewidth=2)
axes[0, 1].set_ylabel('Bilet Ücreti (£)', fontsize=12, fontweight='bold')
axes[0, 1].set_title('Box Plot (Aykırı Değerler Görünür)', fontsize=14, fontweight='bold', color='#2c3e50')
axes[0, 1].grid(axis='y', alpha=0.3)

# 3. Log-scale Histogram (Çarpık dağılım için)
axes[1, 0].hist(fare_data[fare_data > 0], bins=50, color='#1abc9c', edgecolor='#2c3e50', alpha=0.7)
axes[1, 0].set_xlabel('Bilet Ücreti (£) - Log Ölçek', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Frekans', fontsize=12, fontweight='bold')
axes[1, 0].set_title('Histogram (Log Ölçekli)', fontsize=14, fontweight='bold', color='#2c3e50')
axes[1, 0].set_xscale('log')
axes[1, 0].grid(axis='y', alpha=0.3)

# 4. İstatistikler
axes[1, 1].axis('off')
stats_text = f"""
📊 BİLET ÜCRETİ İSTATİSTİKLERİ

💰 Ortalama (Mean): £{fare_data.mean():.2f}
📊 Medyan: £{fare_data.median():.2f}
📌 Mod: £{fare_data.mode()[0]:.2f}
📉 Std Sapma: £{fare_data.std():.2f}

🔽 Minimum: £{fare_data.min():.2f}
🔼 Maksimum: £{fare_data.max():.2f}

📦 Q1 (25%): £{fare_data.quantile(0.25):.2f}
📦 Q2 (50%): £{fare_data.quantile(0.50):.2f}
📦 Q3 (75%): £{fare_data.quantile(0.75):.2f}

📏 IQR: £{fare_data.quantile(0.75) - fare_data.quantile(0.25):.2f}
📊 Çarpıklık (Skewness): {fare_data.skew():.2f}
"""
axes[1, 1].text(0.1, 0.5, stats_text, fontsize=12, verticalalignment='center',
                bbox=dict(boxstyle='round', facecolor='#ecf0f1', edgecolor='#2c3e50', linewidth=2),
                family='monospace')

plt.tight_layout()
plt.show()

print("\n💰 BİLET ÜCRETİ DETAYLI ANALİZ")
print("="*70)
print(f"   📊 Ortalama: £{fare_data.mean():.2f}")
print(f"   📊 Medyan: £{fare_data.median():.2f}")
print(f"   📌 Dağılım Çarpıklığı: {fare_data.skew():.2f} (Pozitif = Sağa çarpık)")
print(f"   ⚠️  Aykırı Değer Sayısı: {len(fare_data[fare_data > fare_data.quantile(0.75) + 1.5 * (fare_data.quantile(0.75) - fare_data.quantile(0.25))])}")
print("="*70)
````

### 💡 Karar Destek Sistemi İçgörüleri

**Univariate analysis bize hangi soruları cevaplıyor?**

1. **Veri dağılımı nasıl?** → Histogram, box plot ile görselleştirdik
2. **Merkezi eğilim nedir?** → Mean, median, mode hesapladık
3. **Değişkenlik ne kadar?** → Std, range, IQR inceledik
4. **Aykırı değerler var mı?** → Box plot ile tespit ettik
5. **Eksik veriler nerede?** → Missing data analizi yaptık

**Bir sonraki adımda neler yapacağız?**

- 📊 Değişkenler arası **ilişkileri** keşfedeceğiz (Bivariate Analysis)
- 🎯 Hayatta kalma ile diğer değişkenlerin **bağlantılarını** bulacağız
- 📈 **Cross-tabulation** ve **grup analizleri** yapacağız

## 🎯 Bivariate Analysis Nedir?

**İkili değişken analizi**, iki değişken arasındaki ilişkiyi inceler. Titanic veri setinde en önemli soru:

> **"Hangi faktörler hayatta kalmayı etkiliyor?"**

Bu soruyu cevaplamak için **Survived** değişkenini diğer değişkenlerle karşılaştıracağız.

### 📊 Kullanacağımız Teknikler:

- **Cross-tabulation** (Çapraz tablo)
- **Grouped bar charts** (Gruplu çubuk grafikler)
- **Stacked bar charts** (Yığılmış çubuk grafikler)
- **Violin plots** (Keman grafikleri)
- **Box plots by group** (Gruplara göre kutu grafikleri)

## 🎫 Survived vs Pclass: Sosyal Sınıf ve Hayatta Kalma

**Hipotez:** Üst sınıf yolcuların hayatta kalma şansı daha yüksektir (can simitleri ve botlara öncelikli erişim).

````python
# Cross-tabulation: Pclass vs Survived
ct_pclass = pd.crosstab(df['Pclass'], df['Survived'], margins=True)
ct_pclass_pct = pd.crosstab(df['Pclass'], df['Survived'], normalize='index') * 100

# Görselleştirme
fig = make_subplots(
    rows=1, cols=3,
    subplot_titles=('Sınıfa Göre Hayatta Kalma Sayıları', 
                    'Sınıfa Göre Hayatta Kalma Oranları (%)',
                    'Yığılmış Yüzde Grafik'),
    specs=[[{"type": "bar"}, {"type": "bar"}, {"type": "bar"}]],
    horizontal_spacing=0.1
)

# 1. Sayılar - Grouped Bar
for survived, color, name in [(0, '#e74c3c', 'Ölü'), (1, '#2ecc71', 'Hayatta')]:
    counts = [ct_pclass.loc[i, survived] for i in [1, 2, 3]]
    fig.add_trace(
        go.Bar(
            x=['1. Sınıf', '2. Sınıf', '3. Sınıf'],
            y=counts,
            name=name,
            marker=dict(color=color, line=dict(color='#2c3e50', width=2)),
            text=counts,
            textposition='outside',
            legendgroup='group1',
            showlegend=True
        ),
        row=1, col=1
    )

# 2. Yüzdeler - Grouped Bar
for survived, color, name in [(0, '#e74c3c', 'Ölü'), (1, '#2ecc71', 'Hayatta')]:
    percentages = [ct_pclass_pct.loc[i, survived] for i in [1, 2, 3]]
    fig.add_trace(
        go.Bar(
            x=['1. Sınıf', '2. Sınıf', '3. Sınıf'],
            y=percentages,
            name=name,
            marker=dict(color=color, line=dict(color='#2c3e50', width=2)),
            text=[f'{p:.1f}%' for p in percentages],
            textposition='outside',
            legendgroup='group2',
            showlegend=False
        ),
        row=1, col=2
    )

# 3. Stacked 100% Bar
for survived, color, name in [(0, '#e74c3c', 'Ölü'), (1, '#2ecc71', 'Hayatta')]:
    percentages = [ct_pclass_pct.loc[i, survived] for i in [1, 2, 3]]
    fig.add_trace(
        go.Bar(
            x=['1. Sınıf', '2. Sınıf', '3. Sınıf'],
            y=percentages,
            name=name,
            marker=dict(color=color, line=dict(color='#2c3e50', width=1)),
            text=[f'{p:.1f}%' for p in percentages],
            textposition='inside',
            legendgroup='group3',
            showlegend=False
        ),
        row=1, col=3
    )

fig.update_layout(
    title=dict(
        text='Bilet Sınıfı ve Hayatta Kalma İlişkisi',
        font=dict(size=20, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    height=500,
    barmode='group',
    paper_bgcolor='#fafafa',
    legend=dict(x=0.01, y=0.99, bgcolor='rgba(255,255,255,0.8)', bordercolor='#2c3e50', borderwidth=1)
)

fig.update_yaxes(title=dict(text='Sayı', font=dict(size=11)), row=1, col=1)
fig.update_yaxes(title=dict(text='Yüzde (%)', font=dict(size=11)), row=1, col=2)
fig.update_yaxes(title=dict(text='Yüzde (%)', font=dict(size=11)), row=1, col=3)
fig.update_xaxes(title=dict(text='Bilet Sınıfı', font=dict(size=11)), row=1, col=1)
fig.update_xaxes(title=dict(text='Bilet Sınıfı', font=dict(size=11)), row=1, col=2)
fig.update_xaxes(title=dict(text='Bilet Sınıfı', font=dict(size=11)), row=1, col=3)

# Stacked için barmode değiştir
fig.update_layout(barmode='stack')

fig.show()

# İstatistikler
print("\n🎫 PCLASS vs SURVIVED - ÇAPRAZ TABLO")
print("="*70)
print("\n📊 SAYILAR:")
print(ct_pclass)
print("\n📊 YÜZDELER (Satır bazında):")
print(ct_pclass_pct.round(2))
print("\n💡 ÖNEMLİ BULGULAR:")
print(f"   ✅ 1. Sınıf Hayatta Kalma: {ct_pclass_pct.loc[1, 1]:.1f}%")
print(f"   ✅ 2. Sınıf Hayatta Kalma: {ct_pclass_pct.loc[2, 1]:.1f}%")
print(f"   ✅ 3. Sınıf Hayatta Kalma: {ct_pclass_pct.loc[3, 1]:.1f}%")
print(f"\n   📈 1. ile 3. sınıf arasında {ct_pclass_pct.loc[1, 1] - ct_pclass_pct.loc[3, 1]:.1f}% fark var!")
print("="*70)
````

## 👫 Survived vs Sex: Cinsiyet ve Hayatta Kalma

**Hipotez:** "Kadınlar ve çocuklar önce" politikası nedeniyle kadınların hayatta kalma oranı çok daha yüksektir.

````python
# Cross-tabulation: Sex vs Survived
ct_sex = pd.crosstab(df['Sex'], df['Survived'], margins=True)
ct_sex_pct = pd.crosstab(df['Sex'], df['Survived'], normalize='index') * 100

# Görselleştirme
fig = make_subplots(
    rows=1, cols=2,
    specs=[[{"type": "bar"}, {"type": "pie"}]],
    subplot_titles=('Cinsiyete Göre Hayatta Kalma Oranları', 'Hayatta Kalanların Cinsiyet Dağılımı')
)

# 1. Grouped Bar (Yüzdeler)
for survived, color, name in [(0, '#e74c3c', 'Ölü'), (1, '#2ecc71', 'Hayatta')]:
    percentages = [ct_sex_pct.loc['female', survived], ct_sex_pct.loc['male', survived]]
    fig.add_trace(
        go.Bar(
            x=['Kadın', 'Erkek'],
            y=percentages,
            name=name,
            marker=dict(color=color, line=dict(color='#2c3e50', width=2)),
            text=[f'{p:.1f}%' for p in percentages],
            textposition='outside',
            textfont=dict(size=14, color='#2c3e50', family='Arial Black')
        ),
        row=1, col=1
    )

# 2. Pie chart - Sadece hayatta kalanlar
survived_by_sex = df[df['Survived'] == 1]['Sex'].value_counts()
fig.add_trace(
    go.Pie(
        labels=['Kadın', 'Erkek'],
        values=[survived_by_sex['female'], survived_by_sex['male']],
        marker=dict(colors=['#e91e63', '#3498db'], line=dict(color='#ffffff', width=3)),
        textinfo='label+percent+value',
        textfont=dict(size=13, color='#ffffff', family='Arial Black'),
        hole=0.4,
        hovertemplate='<b>%{label}</b><br>Sayı: %{value}<br>Yüzde: %{percent}<extra></extra>'
    ),
    row=1, col=2
)

fig.update_layout(
    title=dict(
        text='Cinsiyet ve Hayatta Kalma İlişkisi',
        font=dict(size=20, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    height=500,
    barmode='group',
    showlegend=True,
    paper_bgcolor='#fafafa',
    legend=dict(x=0.01, y=0.99, bgcolor='rgba(255,255,255,0.8)', bordercolor='#2c3e50', borderwidth=1)
)

fig.update_yaxes(title=dict(text='Hayatta Kalma Oranı (%)', font=dict(size=12)), row=1, col=1)
fig.update_xaxes(title=dict(text='Cinsiyet', font=dict(size=12)), row=1, col=1)

fig.show()

# İstatistikler
print("\n👫 SEX vs SURVIVED - ÇAPRAZ TABLO")
print("="*70)
print("\n📊 SAYILAR:")
print(ct_sex)
print("\n📊 YÜZDELER (Satır bazında):")
print(ct_sex_pct.round(2))
print("\n💡 ÖNEMLİ BULGULAR:")
print(f"   👩 Kadın Hayatta Kalma: {ct_sex_pct.loc['female', 1]:.1f}%")
print(f"   👨 Erkek Hayatta Kalma: {ct_sex_pct.loc['male', 1]:.1f}%")
print(f"\n   📈 Kadınların hayatta kalma şansı erkeklerden {ct_sex_pct.loc['female', 1] / ct_sex_pct.loc['male', 1]:.2f}x daha fazla!")
print(f"   📊 Fark: {ct_sex_pct.loc['female', 1] - ct_sex_pct.loc['male', 1]:.1f} puan")
print("="*70)
````

## 🎭 Survived vs Sex vs Pclass: Üçlü Kombinasyon

**Çok boyutlu analiz:** Cinsiyet ve sınıfın birlikte etkisi nedir?

````python
# 3 değişkenli çapraz tablo
survival_by_sex_class = df.groupby(['Sex', 'Pclass'])['Survived'].agg(['sum', 'count', 'mean']).reset_index()
survival_by_sex_class.columns = ['Sex', 'Pclass', 'Survived_Count', 'Total_Count', 'Survival_Rate']
survival_by_sex_class['Survival_Rate'] = survival_by_sex_class['Survival_Rate'] * 100

# Görselleştirme - Heat map style
fig = go.Figure()

# Kadınlar için
female_data = survival_by_sex_class[survival_by_sex_class['Sex'] == 'female']
fig.add_trace(
    go.Bar(
        x=['1. Sınıf', '2. Sınıf', '3. Sınıf'],
        y=female_data['Survival_Rate'].values,
        name='Kadın',
        marker=dict(color='#e91e63', line=dict(color='#2c3e50', width=2)),
        text=[f'{v:.1f}%' for v in female_data['Survival_Rate'].values],
        textposition='outside',
        textfont=dict(size=13, family='Arial Black'),
        hovertemplate='<b>Kadın - %{x}</b><br>Hayatta Kalma: %{y:.1f}%<extra></extra>'
    )
)

# Erkekler için
male_data = survival_by_sex_class[survival_by_sex_class['Sex'] == 'male']
fig.add_trace(
    go.Bar(
        x=['1. Sınıf', '2. Sınıf', '3. Sınıf'],
        y=male_data['Survival_Rate'].values,
        name='Erkek',
        marker=dict(color='#3498db', line=dict(color='#2c3e50', width=2)),
        text=[f'{v:.1f}%' for v in male_data['Survival_Rate'].values],
        textposition='outside',
        textfont=dict(size=13, family='Arial Black'),
        hovertemplate='<b>Erkek - %{x}</b><br>Hayatta Kalma: %{y:.1f}%<extra></extra>'
    )
)

fig.update_layout(
    title=dict(
        text='Cinsiyet ve Sınıfa Göre Hayatta Kalma Oranları',
        font=dict(size=20, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    xaxis=dict(title=dict(text='Bilet Sınıfı', font=dict(size=13, color='#2c3e50'))),
    yaxis=dict(title=dict(text='Hayatta Kalma Oranı (%)', font=dict(size=13, color='#2c3e50'))),
    barmode='group',
    height=550,
    paper_bgcolor='#fafafa',
    plot_bgcolor='#ffffff',
    legend=dict(x=0.85, y=0.99, bgcolor='rgba(255,255,255,0.9)', bordercolor='#2c3e50', borderwidth=2)
)

fig.show()

# Detaylı tablo
print("\n🎭 SEX × PCLASS × SURVIVED - DETAYLI ANALİZ")
print("="*80)
print(survival_by_sex_class.to_string(index=False))
print("\n💡 KRİTİK BULGULAR:")
print(f"   🏆 En yüksek hayatta kalma: Kadın + 1. Sınıf ({female_data[female_data['Pclass']==1]['Survival_Rate'].values[0]:.1f}%)")
print(f"   ⚠️  En düşük hayatta kalma: Erkek + 3. Sınıf ({male_data[male_data['Pclass']==3]['Survival_Rate'].values[0]:.1f}%)")
print(f"\n   📊 Fark: {female_data[female_data['Pclass']==1]['Survival_Rate'].values[0] - male_data[male_data['Pclass']==3]['Survival_Rate'].values[0]:.1f} puan!")
print("="*80)
````

## 👶 Survived vs Age: Yaş ve Hayatta Kalma

**Sürekli değişken analizi** - Yaşın hayatta kalmaya etkisini violin plot ve box plot ile görelim.

````python
# Yaş ve hayatta kalma (NaN'ları çıkar)
age_survived = df[df['Age'].notna()].copy()

# İstatistikler
age_stats_by_survived = age_survived.groupby('Survived')['Age'].describe()

# Görselleştirme
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Yaş Dağılımı - Violin Plot', 'Yaş Dağılımı - Box Plot'),
    specs=[[{"type": "violin"}, {"type": "box"}]]
)

# 1. Violin Plot
for survived, color, name in [(0, '#e74c3c', 'Ölü'), (1, '#2ecc71', 'Hayatta')]:
    ages = age_survived[age_survived['Survived'] == survived]['Age']
    fig.add_trace(
        go.Violin(
            y=ages,
            name=name,
            box_visible=True,
            meanline_visible=True,
            fillcolor=color,
            line_color='#2c3e50',
            opacity=0.7,
            x0=name
        ),
        row=1, col=1
    )

# 2. Box Plot
for survived, color, name in [(0, '#e74c3c', 'Ölü'), (1, '#2ecc71', 'Hayatta')]:
    ages = age_survived[age_survived['Survived'] == survived]['Age']
    fig.add_trace(
        go.Box(
            y=ages,
            name=name,
            marker=dict(color=color),
            boxmean='sd',
            x0=name
        ),
        row=1, col=2
    )

fig.update_layout(
    title=dict(
        text='Yaş ve Hayatta Kalma İlişkisi',
        font=dict(size=20, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    height=550,
    showlegend=False,
    paper_bgcolor='#fafafa'
)

fig.update_yaxes(title=dict(text='Yaş', font=dict(size=12)), row=1, col=1)
fig.update_yaxes(title=dict(text='Yaş', font=dict(size=12)), row=1, col=2)

fig.show()

# İstatistikler
print("\n👶 AGE vs SURVIVED - İSTATİSTİKSEL ANALİZ")
print("="*80)
print(age_stats_by_survived.round(2))
print("\n💡 ÖNEMLİ BULGULAR:")
print(f"   📊 Ölenler Ortalama Yaş: {age_stats_by_survived.loc[0, 'mean']:.2f}")
print(f"   📊 Hayatta Kalanlar Ortalama Yaş: {age_stats_by_survived.loc[1, 'mean']:.2f}")
print(f"   📈 Fark: {age_stats_by_survived.loc[0, 'mean'] - age_stats_by_survived.loc[1, 'mean']:.2f} yaş")
print(f"\n   🧒 Çocuklar (<18 yaş):")
children = age_survived[age_survived['Age'] < 18]
print(f"      • Toplam: {len(children)} kişi")
print(f"      • Hayatta Kalma: {(children['Survived'].sum() / len(children) * 100):.1f}%")
print(f"\n   👨 Yetişkinler (>=18 yaş):")
adults = age_survived[age_survived['Age'] >= 18]
print(f"      • Toplam: {len(adults)} kişi")
print(f"      • Hayatta Kalma: {(adults['Survived'].sum() / len(adults) * 100):.1f}%")
print("="*80)
````

## 💰 Survived vs Fare: Bilet Ücreti ve Hayatta Kalma

**Ekonomik faktör analizi** - Daha pahalı bilet alanlar daha fazla hayatta kaldı mı?

````python
# Fare istatistikleri
fare_stats_by_survived = df.groupby('Survived')['Fare'].describe()

# Görselleştirme - Matplotlib ile daha detaylı
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Bilet Ücreti (Fare) ve Hayatta Kalma İlişkisi', fontsize=20, fontweight='bold', color='#2c3e50', y=0.995)

# 1. Histogram (Log scale)
for survived, color, label in [(0, '#e74c3c', 'Ölü'), (1, '#2ecc71', 'Hayatta')]:
    fare_data = df[df['Survived'] == survived]['Fare']
    fare_data_log = fare_data[fare_data > 0]  # Log için 0'ları çıkar
    axes[0, 0].hist(fare_data_log, bins=30, alpha=0.6, color=color, edgecolor='#2c3e50', label=label)
axes[0, 0].set_xlabel('Bilet Ücreti (£) - Log Ölçek', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Frekans', fontsize=12, fontweight='bold')
axes[0, 0].set_title('Histogram (Log Ölçekli)', fontsize=14, fontweight='bold', color='#2c3e50')
axes[0, 0].set_xscale('log')
axes[0, 0].legend(fontsize=11)
axes[0, 0].grid(axis='y', alpha=0.3)

# 2. Box Plot
box_data = [df[df['Survived'] == 0]['Fare'], df[df['Survived'] == 1]['Fare']]
box = axes[0, 1].boxplot(box_data, labels=['Ölü', 'Hayatta'], patch_artist=True)
for patch, color in zip(box['boxes'], ['#e74c3c', '#2ecc71']):
    patch.set_facecolor(color)
    patch.set_edgecolor('#2c3e50')
    patch.set_linewidth(2)
axes[0, 1].set_ylabel('Bilet Ücreti (£)', fontsize=12, fontweight='bold')
axes[0, 1].set_title('Box Plot', fontsize=14, fontweight='bold', color='#2c3e50')
axes[0, 1].grid(axis='y', alpha=0.3)

# 3. Fare gruplarına göre hayatta kalma
fare_bins = [0, 10, 20, 50, 100, 520]
fare_labels = ['£0-10', '£10-20', '£20-50', '£50-100', '£100+']
df_temp = df.copy()
df_temp['Fare_Group'] = pd.cut(df_temp['Fare'], bins=fare_bins, labels=fare_labels, include_lowest=True)
fare_group_survival = df_temp.groupby('Fare_Group', observed=True)['Survived'].agg(['sum', 'count', 'mean']).reset_index()
fare_group_survival['Survival_Rate'] = fare_group_survival['mean'] * 100

axes[1, 0].bar(range(len(fare_labels)), fare_group_survival['Survival_Rate'], 
               color=['#e74c3c', '#e67e22', '#f39c12', '#2ecc71', '#27ae60'],
               edgecolor='#2c3e50', linewidth=2)
axes[1, 0].set_xticks(range(len(fare_labels)))
axes[1, 0].set_xticklabels(fare_labels, rotation=0, fontsize=11)
axes[1, 0].set_xlabel('Ücret Grubu', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Hayatta Kalma Oranı (%)', fontsize=12, fontweight='bold')
axes[1, 0].set_title('Ücret Grubuna Göre Hayatta Kalma', fontsize=14, fontweight='bold', color='#2c3e50')
for i, v in enumerate(fare_group_survival['Survival_Rate']):
    axes[1, 0].text(i, v + 2, f'{v:.1f}%', ha='center', fontsize=11, fontweight='bold')
axes[1, 0].grid(axis='y', alpha=0.3)

# 4. İstatistik tablosu
axes[1, 1].axis('off')
stats_text = f"""
📊 BİLET ÜCRETİ İSTATİSTİKLERİ

💀 ÖLENLER:
   Ortalama: £{fare_stats_by_survived.loc[0, 'mean']:.2f}
   Medyan  : £{fare_stats_by_survived.loc[0, '50%']:.2f}
   Std     : £{fare_stats_by_survived.loc[0, 'std']:.2f}

✅ HAYATTA KALANLAR:
   Ortalama: £{fare_stats_by_survived.loc[1, 'mean']:.2f}
   Medyan  : £{fare_stats_by_survived.loc[1, '50%']:.2f}
   Std     : £{fare_stats_by_survived.loc[1, 'std']:.2f}

📈 FARK:
   Ortalama: £{fare_stats_by_survived.loc[1, 'mean'] - fare_stats_by_survived.loc[0, 'mean']:.2f}
   Medyan  : £{fare_stats_by_survived.loc[1, '50%'] - fare_stats_by_survived.loc[0, '50%']:.2f}
"""
axes[1, 1].text(0.1, 0.5, stats_text, fontsize=11, verticalalignment='center',
                bbox=dict(boxstyle='round', facecolor='#ecf0f1', edgecolor='#2c3e50', linewidth=2),
                family='monospace')

plt.tight_layout()
plt.show()

# Detaylı çıktı
print("\n💰 FARE vs SURVIVED - DETAYLI ANALİZ")
print("="*80)
print("\n📊 GRUP BAZINDA HAYATTA KALMA:")
print(fare_group_survival.to_string(index=False))
print("\n💡 ÖNEMLİ BULGULAR:")
print(f"   📈 En düşük ücret grubu (£0-10) hayatta kalma: {fare_group_survival.iloc[0]['Survival_Rate']:.1f}%")
print(f"   📈 En yüksek ücret grubu (£100+) hayatta kalma: {fare_group_survival.iloc[-1]['Survival_Rate']:.1f}%")
print(f"   🎯 Fark: {fare_group_survival.iloc[-1]['Survival_Rate'] - fare_group_survival.iloc[0]['Survival_Rate']:.1f} puan")
print("="*80)
````

## ⚓ Survived vs Embarked: Biniş Limanı ve Hayatta Kalma

**Coğrafi faktör** - Hangi limandan binen yolcuların şansı daha yüksekti?

````python
# Embarked analizi (NaN'ları çıkar)
embarked_survived = df[df['Embarked'].notna()].copy()

# Cross-tabulation
ct_embarked = pd.crosstab(embarked_survived['Embarked'], embarked_survived['Survived'], margins=True)
ct_embarked_pct = pd.crosstab(embarked_survived['Embarked'], embarked_survived['Survived'], normalize='index') * 100

# Liman isimleri
port_names = {'C': 'Cherbourg', 'Q': 'Queenstown', 'S': 'Southampton'}

# Görselleştirme
fig = go.Figure()

# Hayatta kalma oranları
for survived, color, name in [(0, '#e74c3c', 'Ölü'), (1, '#2ecc71', 'Hayatta')]:
    percentages = [ct_embarked_pct.loc[port, survived] for port in ['C', 'Q', 'S']]
    fig.add_trace(
        go.Bar(
            x=[f'{port_names[p]}<br>({p})' for p in ['C', 'Q', 'S']],
            y=percentages,
            name=name,
            marker=dict(color=color, line=dict(color='#2c3e50', width=2)),
            text=[f'{p:.1f}%' for p in percentages],
            textposition='inside',
            textfont=dict(size=13, color='#ffffff', family='Arial Black')
        )
    )

fig.update_layout(
    title=dict(
        text='Biniş Limanına Göre Hayatta Kalma Oranları',
        font=dict(size=20, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    xaxis=dict(title=dict(text='Biniş Limanı', font=dict(size=13, color='#2c3e50'))),
    yaxis=dict(title=dict(text='Yüzde (%)', font=dict(size=13, color='#2c3e50'))),
    barmode='stack',
    height=550,
    paper_bgcolor='#fafafa',
    plot_bgcolor='#ffffff',
    legend=dict(x=0.85, y=0.95, bgcolor='rgba(255,255,255,0.9)', bordercolor='#2c3e50', borderwidth=2)
)

fig.show()

# İstatistikler
print("\n⚓ EMBARKED vs SURVIVED - ÇAPRAZ TABLO")
print("="*80)
print("\n📊 SAYILAR:")
print(ct_embarked)
print("\n📊 YÜZDELER (Satır bazında):")
print(ct_embarked_pct.round(2))
print("\n💡 ÖNEMLİ BULGULAR:")
for port in ['C', 'Q', 'S']:
    print(f"   {port_names[port]:15s} ({port}): {ct_embarked_pct.loc[port, 1]:.1f}% hayatta kalma")
print("\n   📌 En yüksek: Cherbourg (Fransa)")
print("   📌 En düşük: Southampton (İngiltere)")
print("\n   🔍 Neden? Cherbourg'dan 1. sınıf yolcular daha fazla binmiş olabilir!")
print("="*80)

# Embarked ve Pclass ilişkisi
embarked_class = embarked_survived.groupby(['Embarked', 'Pclass']).size().unstack(fill_value=0)
embarked_class_pct = embarked_class.div(embarked_class.sum(axis=1), axis=0) * 100
print("\n📊 BİNİŞ LİMANI ve SINIF İLİŞKİSİ:")
print(embarked_class_pct.round(1))
print("\n   ✅ Cherbourg'dan binen yolcuların büyük kısmı 1. sınıf!")
print("="*80)
````

## 🎯 Multivariate Analysis Nedir?

**Çoklu değişken analizi**, üç veya daha fazla değişkeni aynı anda inceleyerek:

- 📊 **Korelasyonları** (değişkenler arası doğrusal ilişkiler)

## 📊 Korelasyon Matrisi: Sayısal Değişkenler

**Korelasyon**, iki değişken arasındaki doğrusal ilişkinin gücünü ve yönünü ölçer (-1 ile +1 arasında).

- **+1**: Mükemmel pozitif korelasyon
- **0**: İlişki yok
- **-1**: Mükemmel negatif korelasyon

````python
# Sayısal değişkenler için korelasyon matrisi
numerical_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
correlation_matrix = df[numerical_cols].corr()

# Görselleştirme
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Korelasyon Matrisi: Sayısal Değişkenler', fontsize=20, fontweight='bold', color='#2c3e50', y=1.02)

# 1. Tam korelasyon matrisi
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.3f',
    cmap='RdYlGn',
    center=0,
    square=True,
    linewidths=2,
    cbar_kws={'label': 'Korelasyon Katsayısı'},
    ax=axes[0],
    vmin=-1,
    vmax=1
)
axes[0].set_title('Korelasyon Matrisi (Tüm Değişkenler)', fontsize=14, fontweight='bold', pad=10)

# 2. Sadece Survived ile korelasyonlar
survived_corr = correlation_matrix['Survived'].sort_values(ascending=False)
colors = ['#2ecc71' if x > 0 else '#e74c3c' for x in survived_corr.values]
axes[1].barh(range(len(survived_corr)), survived_corr.values, color=colors, edgecolor='#2c3e50', linewidth=2)
axes[1].set_yticks(range(len(survived_corr)))
axes[1].set_yticklabels(survived_corr.index, fontsize=12, fontweight='bold')
axes[1].set_xlabel('Korelasyon Katsayısı', fontsize=12, fontweight='bold')
axes[1].set_title('Survived ile Korelasyonlar', fontsize=14, fontweight='bold', pad=10)
axes[1].axvline(0, color='#2c3e50', linewidth=2, linestyle='--')
axes[1].grid(axis='x', alpha=0.3)

# Değerleri etiketle
for i, v in enumerate(survived_corr.values):
    axes[1].text(v + 0.02 if v > 0 else v - 0.02, i, f'{v:.3f}', 
                 va='center', ha='left' if v > 0 else 'right', 
                 fontsize=11, fontweight='bold', color='#2c3e50')

plt.tight_layout()
plt.show()

# Detaylı çıktı
print("\n📊 KORELASYON MATRİSİ ANALİZİ")
print("="*80)
print("\n🎯 Survived ile Korelasyonlar (Sıralı):")
print(survived_corr.to_string())

print("\n💡 ÖNEMLİ BULGULAR:")
print(f"   📈 En güçlü pozitif korelasyon: Fare ({survived_corr['Fare']:.3f})")
print(f"   📉 En güçlü negatif korelasyon: Pclass ({survived_corr['Pclass']:.3f})")
print(f"\n   ⚠️  Pclass negatif → Alt sınıf = düşük hayatta kalma")
print(f"   ✅ Fare pozitif → Yüksek ücret = yüksek hayatta kalma")

print("\n🔗 Değişkenler Arası İlginç İlişkiler:")
print(f"   • Pclass ↔ Fare: {correlation_matrix.loc['Pclass', 'Fare']:.3f} (Güçlü negatif - doğru!)")
print(f"   • SibSp ↔ Parch: {correlation_matrix.loc['SibSp', 'Parch']:.3f} (Aile büyüklüğü ilişkisi)")
print(f"   • Age ↔ Pclass: {correlation_matrix.loc['Age', 'Pclass']:.3f} (Zayıf pozitif)")
print("="*80)
````

## 🎨 Pairplot: Tüm Değişken Çiftlerini Görselleştirme

**Pairplot**, her değişken çiftini scatter plot ile, her değişkeni ise histogram ile gösterir. **Hayatta kalma durumuna göre renklendireceğiz.**

````python
# Pairplot için veri hazırlama (Sadece sayısal değişkenler + Survived)
pairplot_data = df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].dropna()
pairplot_data['Survived'] = pairplot_data['Survived'].map({0: 'Ölü', 1: 'Hayatta'})

# Pairplot
plt.figure(figsize=(2, 2))  # Küçük figure (pairplot kendi boyutunu ayarlar)
pairplot_fig = sns.pairplot(
    pairplot_data,
    hue='Survived',
    palette={'Ölü': '#e74c3c', 'Hayatta': '#2ecc71'},
    diag_kind='kde',  # Diagonal'da KDE
    plot_kws={'alpha': 0.6, 'edgecolor': 'none', 's': 30},
    diag_kws={'linewidth': 2.5, 'alpha': 0.7}
)

# Başlık ekle
pairplot_fig.fig.suptitle('Pairplot: Tüm Değişken Çiftleri (Hayatta Kalma Durumuna Göre)', 
                          fontsize=18, fontweight='bold', color='#2c3e50', y=1.01)

# Legend konumunu düzenle
plt.legend(title='Durum', bbox_to_anchor=(1.05, 1), loc='upper left', 
           fontsize=11, title_fontsize=12, frameon=True, fancybox=True, shadow=True)

plt.tight_layout()
plt.show()

print("\n🎨 PAIRPLOT ANALİZİ")
print("="*80)
print("\n💡 PAIRPLOT'TAN GÖZLEMLER:")
print("\n   📊 Diagonal (Köşegen):")
print("      • Tek değişkenin dağılımı (KDE - yoğunluk grafiği)")
print("      • Yeşil (Hayatta) vs Kırmızı (Ölü) karşılaştırması")
print("\n   📊 Scatter Plots (Dağılım Grafikleri):")
print("      • İki değişken arasındaki ilişki")
print("      • Ayrışma varsa → güçlü ayırt edici özellik")
print("\n   🔍 Önemli Gözlemler:")
print("      • Fare vs Pclass: Net ayrışma (negatif korelasyon)")
print("      • Age dağılımında örtüşme var (zayıf ayırt edici)")
print("      • SibSp ve Parch: Çoğu yolcu 0 değerinde (yalnız)")
print("="*80)
````

## 🎭 Kategorik + Sayısal: Çok Boyutlu Segmentasyon

**Sex, Pclass, Age gruplarını birleştirerek en detaylı segmentasyon analizini yapalım.**

````python
# Yaş grupları oluştur
df_segment = df.copy()
df_segment['Age_Category'] = pd.cut(
    df_segment['Age'], 
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Çocuk (0-12)', 'Genç (12-18)', 'Yetişkin (18-35)', 'Orta Yaş (35-60)', 'Yaşlı (60+)'],
    include_lowest=True
)

# Çok boyutlu segmentasyon
segment_analysis = df_segment.groupby(['Sex', 'Pclass', 'Age_Category'], observed=True)['Survived'].agg([
    ('Total', 'count'),
    ('Survived', 'sum'),
    ('Survival_Rate', 'mean')
]).reset_index()
segment_analysis['Survival_Rate'] = segment_analysis['Survival_Rate'] * 100
segment_analysis = segment_analysis.sort_values('Survival_Rate', ascending=False)

# Top 10 ve Bottom 10
top_10 = segment_analysis.head(10)
bottom_10 = segment_analysis.tail(10)

# Görselleştirme
fig, axes = plt.subplots(2, 1, figsize=(15, 12))
fig.suptitle('Çok Boyutlu Segmentasyon: Cinsiyet × Sınıf × Yaş Grubu', 
             fontsize=20, fontweight='bold', color='#2c3e50', y=0.995)

# 1. Top 10
top_labels = [f"{row['Sex'][:1].upper()}/{row['Pclass']}/{row['Age_Category']}" 
              for _, row in top_10.iterrows()]
axes[0].barh(range(len(top_10)), top_10['Survival_Rate'], 
             color='#2ecc71', edgecolor='#2c3e50', linewidth=2)
axes[0].set_yticks(range(len(top_10)))
axes[0].set_yticklabels(top_labels, fontsize=10)
axes[0].set_xlabel('Hayatta Kalma Oranı (%)', fontsize=12, fontweight='bold')
axes[0].set_title('🏆 En Yüksek Hayatta Kalma Oranları (Top 10)', 
                  fontsize=14, fontweight='bold', color='#27ae60', pad=10)
axes[0].set_xlim(0, 105)
for i, v in enumerate(top_10['Survival_Rate'].values):
    axes[0].text(v + 2, i, f'{v:.1f}% (n={top_10.iloc[i]["Total"]})', 
                 va='center', fontsize=10, fontweight='bold')
axes[0].grid(axis='x', alpha=0.3)

# 2. Bottom 10
bottom_labels = [f"{row['Sex'][:1].upper()}/{row['Pclass']}/{row['Age_Category']}" 
                 for _, row in bottom_10.iterrows()]
axes[1].barh(range(len(bottom_10)), bottom_10['Survival_Rate'], 
             color='#e74c3c', edgecolor='#2c3e50', linewidth=2)
axes[1].set_yticks(range(len(bottom_10)))
axes[1].set_yticklabels(bottom_labels, fontsize=10)
axes[1].set_xlabel('Hayatta Kalma Oranı (%)', fontsize=12, fontweight='bold')
axes[1].set_title('⚠️ En Düşük Hayatta Kalma Oranları (Bottom 10)', 
                  fontsize=14, fontweight='bold', color='#c0392b', pad=10)
axes[1].set_xlim(0, 105)
for i, v in enumerate(bottom_10['Survival_Rate'].values):
    axes[1].text(v + 2, i, f'{v:.1f}% (n={bottom_10.iloc[i]["Total"]})', 
                 va='center', fontsize=10, fontweight='bold')
axes[1].grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()

# Detaylı tablo
print("\n🎭 ÇOKLU BOYUTLU SEGMENTASYON ANALİZİ")
print("="*100)
print("\n🏆 EN YÜKSEK HAYATTA KALMA ORANLARI (TOP 10):")
print(top_10.to_string(index=False))
print("\n⚠️ EN DÜŞÜK HAYATTA KALMA ORANLARI (BOTTOM 10):")
print(bottom_10.to_string(index=False))

print("\n💡 KRİTİK BULGULAR:")
print(f"   🥇 En iyi segment: {top_10.iloc[0]['Sex']} / {top_10.iloc[0]['Pclass']}. Sınıf / {top_10.iloc[0]['Age_Category']}")
print(f"      Hayatta Kalma: {top_10.iloc[0]['Survival_Rate']:.1f}% ({top_10.iloc[0]['Survived']}/{top_10.iloc[0]['Total']} kişi)")
print(f"\n   💀 En kötü segment: {bottom_10.iloc[-1]['Sex']} / {bottom_10.iloc[-1]['Pclass']}. Sınıf / {bottom_10.iloc[-1]['Age_Category']}")
print(f"      Hayatta Kalma: {bottom_10.iloc[-1]['Survival_Rate']:.1f}% ({bottom_10.iloc[-1]['Survived']}/{bottom_10.iloc[-1]['Total']} kişi)")
print("\n   📊 Fark: {:.1f} puan!".format(top_10.iloc[0]['Survival_Rate'] - bottom_10.iloc[-1]['Survival_Rate']))
print("="*100)
````

## 🌟 Feature Importance: Hangi Değişken Ne Kadar Önemli?

Basit bir **Decision Tree** modeli ile değişken önemlerini (feature importance) bulalım.

````python
# Veri hazırlama
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Encoding için veri kopyası
df_model = df[['Survived', 'Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']].dropna().copy()

# Kategorik değişkenleri encode et
le_sex = LabelEncoder()
le_embarked = LabelEncoder()
df_model['Sex_encoded'] = le_sex.fit_transform(df_model['Sex'])
df_model['Embarked_encoded'] = le_embarked.fit_transform(df_model['Embarked'])

# Feature'lar ve target
features = ['Pclass', 'Sex_encoded', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked_encoded']
X = df_model[features]
y = df_model['Survived']

# Decision Tree modeli
dt_model = DecisionTreeClassifier(max_depth=5, random_state=42)
dt_model.fit(X, y)

# Feature importance
feature_importance = pd.DataFrame({
    'Feature': ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked'],
    'Importance': dt_model.feature_importances_
}).sort_values('Importance', ascending=False)

# Görselleştirme
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
fig.suptitle('Feature Importance: Değişken Önem Analizi', 
             fontsize=20, fontweight='bold', color='#2c3e50', y=1.00)

# 1. Bar chart
colors = plt.cm.RdYlGn(np.linspace(0.3, 0.9, len(feature_importance)))
axes[0].barh(range(len(feature_importance)), feature_importance['Importance'], 
             color=colors, edgecolor='#2c3e50', linewidth=2)
axes[0].set_yticks(range(len(feature_importance)))
axes[0].set_yticklabels(feature_importance['Feature'], fontsize=12, fontweight='bold')
axes[0].set_xlabel('Önem Skoru (Importance Score)', fontsize=12, fontweight='bold')
axes[0].set_title('Değişken Önem Sıralaması', fontsize=14, fontweight='bold', pad=10)
for i, v in enumerate(feature_importance['Importance'].values):
    axes[0].text(v + 0.005, i, f'{v:.4f}', va='center', fontsize=11, fontweight='bold')
axes[0].grid(axis='x', alpha=0.3)

# 2. Pie chart
axes[1].pie(
    feature_importance['Importance'],
    labels=feature_importance['Feature'],
    autopct='%1.1f%%',
    colors=colors,
    startangle=90,
    textprops={'fontsize': 11, 'fontweight': 'bold'},
    wedgeprops={'edgecolor': '#2c3e50', 'linewidth': 2}
)
axes[1].set_title('Değişken Önem Dağılımı (%)', fontsize=14, fontweight='bold', pad=10)

plt.tight_layout()
plt.show()

# Detaylı çıktı
print("\n🌟 FEATURE IMPORTANCE ANALİZİ")
print("="*80)
print("\n📊 DEĞIŞKEN ÖNEM SIRALAMASI:")
print(feature_importance.to_string(index=False))

print("\n💡 YORUMLAMA:")
print(f"   🥇 En önemli: {feature_importance.iloc[0]['Feature']} ({feature_importance.iloc[0]['Importance']:.4f})")
print(f"   🥈 İkinci: {feature_importance.iloc[1]['Feature']} ({feature_importance.iloc[1]['Importance']:.4f})")
print(f"   🥉 Üçüncü: {feature_importance.iloc[2]['Feature']} ({feature_importance.iloc[2]['Importance']:.4f})")

print("\n   📈 İlk 3 değişken toplam önem: {:.1f}%".format(
    feature_importance.head(3)['Importance'].sum() * 100
))

print("\n⚠️ NOT: Bu skorlar Decision Tree'nin bu veriye göre hesapladığı önemlerdir.")
print("   Gerçek önem, farklı modellerde farklı olabilir!")
print("="*80)
````

- ✅ **Correlation Analysis:** Korelasyon matrisi ve heatmap
- ✅ **Segmentation:** Çok boyutlu müşteri segmentasyonu
- ✅ **Feature Engineering:** Yeni değişken oluşturma

**3. Görselleştirme Türleri:**
- ✅ **Bar Chart, Pie Chart:** Kategorik dağılımlar
- ✅ **Histogram, KDE:** Sayısal dağılımlar
- ✅ **Box Plot, Violin Plot:** Merkezi eğilim ve yayılım
- ✅ **Heatmap:** Korelasyon ve ilişkiler
- ✅ **Pairplot:** Çok değişkenli görselleştirme
- ✅ **Mermaid Diagrams:** Süreç ve karar akışları

</div>
