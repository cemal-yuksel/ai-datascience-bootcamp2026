### 🏥 2.1 · Senaryo

Bir **aile sağlığı merkezi** (veya bir sağlık yönetimi birimi) sınırlı sayıda **ileri tetkik** (ör. şeker yükleme testi) kontenjanına sahip. Herkesi aynı hafta çağıramıyor. Elindeki basit ölçümlerle (glukoz, kilo, yaş, hamilelik sayısı…) **kimlerin diyabet riskinin yüksek olduğunu** önceden tahmin edip tarama sırasını belirlemek istiyor. Bu bir **YBS (Yönetim Bilişim Sistemleri) kapasite planlama** problemidir: model hastayı teşhis etmez, **kimin önce çağrılacağını** söyler.

**Veri:** Pima Indians Diabetes; ABD Ulusal Diyabet ve Sindirim ve Böbrek Hastalıkları Enstitüsü'nün topladığı, **21 yaş ve üzeri Pima kökenli kadınlara** ait 768 kayıt (Smith vd., 1988). Hedef değişken `Outcome`: 5 yıl içinde diyabet gelişti mi? (1 = evet, 0 = hayır).

### 🔁 3.1 · Tekrarlanan Satır ve "Eksiklik Bilgi Taşıyor mu?"

Önce tekrar eden satır var mı bakalım; sonra sıfırları **gerçek eksik değere** (`NaN`) çevirip eksik olmanın hedefle ilişkisini inceleyelim:

````python
print("Tekrarlanan satır sayısı:", ham.duplicated().sum())

veri = ham.copy()
veri[SIFIR_SUTUN] = veri[SIFIR_SUTUN].replace(0, np.nan)             # sıfır → NaN (öğrenilen bir bilgi değil, sabit bir KURAL)

eksiklik = pd.DataFrame({
    "eksik sayısı": veri[SIFIR_SUTUN].isna().sum(),
    "eksikken diyabet oranı": [veri.loc[veri[c].isna(), HEDEF].mean() for c in SIFIR_SUTUN],
    "ölçülmüşken diyabet oranı": [veri.loc[veri[c].notna(), HEDEF].mean() for c in SIFIR_SUTUN],
}).round(3)
eksiklik.index = [TR[c] for c in SIFIR_SUTUN]
eksiklik
````

### ✂️ 5.1 · Eğitim / Test Ayrımı

Veriyi **%80 eğitim, %20 test** olarak bölüyoruz. Referans notebook'tan farkımız: `stratify=y` ile **her iki parçada da diyabet oranı aynı kalır** (%34.9). Test setine **en sona kadar dokunmayacağız**.

````python
X = veri.drop(columns=HEDEF)
y = veri[HEDEF]
X_egitim, X_test, y_egitim, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE)

bolme = pd.DataFrame({"satır": [len(y_egitim), len(y_test)], "diyabet oranı": [y_egitim.mean(), y_test.mean()]},
                     index=["Eğitim", "Test"]).round(3)
bolme
````

### 🛡️ 5.2 · Sızıntısız Pipeline

Eksik değerleri **medyanla** dolduracağız. Medyan, **sadece eğitim verisinden** öğrenilmeli; test verisinden (veya çapraz doğrulamanın doğrulama katından) öğrenirsek model **geleceği önceden görmüş** olur: buna **veri sızıntısı** (*data leakage*) denir. `Pipeline`, "doldur → modelle" zincirini tek nesne yapar ve çapraz doğrulamada her katta medyanı **yeniden** yalnızca o katın eğitim kısmından hesaplar.

````python
def pipeline_kur(**ada_parametreleri):
    """Medyan doldurma + AdaBoost (varsayılan zayıf öğrenici: derinliği 1 olan karar kütüğü)."""
    return Pipeline([
        ("doldur", SimpleImputer(strategy="median")),
        ("ada", AdaBoostClassifier(random_state=RANDOM_STATE, **ada_parametreleri)),
    ])


ilk_model = pipeline_kur().fit(X_egitim, y_egitim)
medyanlar = pd.Series(ilk_model.named_steps["doldur"].statistics_, index=[TR[c] for c in X.columns], name="eğitimden öğrenilen medyan")
medyanlar.round(1).to_frame()
````

### 🧪 5.3 · Deney: AdaBoost Ölçeklemeye İhtiyaç Duyar mı?

Referans notebook'taki not doğru: karar ağaçları **eşik** soruları sorar ("glukoz > 127 mi?"), sayıların ölçeği önemli değildir. Mesafe ve gradyan kullanan modellerde (KNN, lojistik regresyon) ise ölçekleme şarttır. Deneyle görelim:

````python
cv5 = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
olcek_sonuc = {}
for ad, adimlar in {
    "Ölçekleme YOK": [("doldur", SimpleImputer(strategy="median")), ("ada", AdaBoostClassifier(random_state=RANDOM_STATE))],
    "StandardScaler VAR": [("doldur", SimpleImputer(strategy="median")), ("olcek", StandardScaler()), ("ada", AdaBoostClassifier(random_state=RANDOM_STATE))],
}.items():
    cv = cross_validate(Pipeline(adimlar), X_egitim, y_egitim, cv=cv5, scoring=["accuracy", "roc_auc"])
    olcek_sonuc[ad] = {"doğruluk": cv["test_accuracy"].mean(), "ROC-AUC": cv["test_roc_auc"].mean()}
pd.DataFrame(olcek_sonuc).T.round(4)
````

Bir modelin iyi olup olmadığını **kıyaslayarak** anlarız. Önce üç basit referans (çıta), sonra varsayılan ayarlı AdaBoost:

| Model | Fikir |
|:--|:--|
| 🪧 **Taban** | Herkese çoğunluk sınıfını söyler ("diyabet yok") |
| 🌱 **Tek kütük** | Tek bir soru soran karar ağacı (AdaBoost'un zayıf öğrenicisi) |
| 🌳 **Tek ağaç (derinlik 3)** | Biraz daha güçlü tek ağaç |
| 🚀 **AdaBoost (varsayılan)** | 50 kütük, `learning_rate=1.0` |

````python
def degerlendir(ad, model, X_, y_):
    """Bir modelin tüm temel metriklerini tek satırda döndürür (diyabet = pozitif sınıf)."""
    tahmin = model.predict(X_)
    skor = model.predict_proba(X_)[:, 1]
    return {"model": ad, "doğruluk": accuracy_score(y_, tahmin), "dengeli doğruluk": balanced_accuracy_score(y_, tahmin),
            "duyarlılık (recall)": recall_score(y_, tahmin, zero_division=0), "kesinlik (precision)": precision_score(y_, tahmin, zero_division=0),
            "F1": f1_score(y_, tahmin, zero_division=0), "ROC-AUC": roc_auc_score(y_, skor), "PR-AUC": average_precision_score(y_, skor)}


def basit_pipeline(model):
    return Pipeline([("doldur", SimpleImputer(strategy="median")), ("model", model)])


taban = basit_pipeline(DummyClassifier(strategy="most_frequent")).fit(X_egitim, y_egitim)
tek_kutuk = basit_pipeline(DecisionTreeClassifier(max_depth=1, random_state=RANDOM_STATE)).fit(X_egitim, y_egitim)
tek_agac = basit_pipeline(DecisionTreeClassifier(max_depth=3, random_state=RANDOM_STATE)).fit(X_egitim, y_egitim)
ada_ilk = pipeline_kur().fit(X_egitim, y_egitim)

ilk_modeller = {"Taban (hep 'yok')": taban, "Tek kütük": tek_kutuk, "Tek ağaç (derinlik 3)": tek_agac, "AdaBoost (varsayılan)": ada_ilk}
ilk_tablo = pd.DataFrame([degerlendir(ad, m, X_test, y_test) for ad, m in ilk_modeller.items()]).set_index("model")
ilk_tablo.round(3)
````

### 🔁 6.1 · Daha Sağlam Ölçüm: Tekrarlı Çapraz Doğrulama

Tek bir bölmenin şansına güvenmek yerine **eğitim setinde** 5 katlı çapraz doğrulamayı 3 kez tekrarlıyoruz (15 değerlendirme). Test setine yine dokunmuyoruz.

````python
cv_tekrar = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=RANDOM_STATE)
cv_sonuclari = {}
for ad, m in {"Taban (hep 'yok')": basit_pipeline(DummyClassifier(strategy="most_frequent")),
              "Tek kütük": basit_pipeline(DecisionTreeClassifier(max_depth=1, random_state=RANDOM_STATE)),
              "Tek ağaç (derinlik 3)": basit_pipeline(DecisionTreeClassifier(max_depth=3, random_state=RANDOM_STATE)),
              "AdaBoost (varsayılan)": pipeline_kur()}.items():
    cv = cross_validate(m, X_egitim, y_egitim, cv=cv_tekrar, scoring=["accuracy", "recall", "roc_auc"], n_jobs=-1)
    cv_sonuclari[ad] = {"doğruluk": f"{cv['test_accuracy'].mean():.3f} ± {cv['test_accuracy'].std():.3f}",
                        "duyarlılık": f"{cv['test_recall'].mean():.3f} ± {cv['test_recall'].std():.3f}",
                        "ROC-AUC": f"{cv['test_roc_auc'].mean():.3f} ± {cv['test_roc_auc'].std():.3f}"}
pd.DataFrame(cv_sonuclari).T
````

Kavram dersinde AdaBoost'u elle hesaplamıştık. Şimdi eğittiğimiz modelin **içine bakalım**: her kütük hangi soruyu soruyor, kaç söz hakkı ($\alpha$) almış?

> 📝 `scikit-learn` söz hakkını $\alpha = \text{öğrenme hızı}\times\ln\frac{1-\varepsilon}{\varepsilon}$ olarak tutar. Kavram notebook'undaki $\tfrac12\ln(\cdot)$ formülünün **2 katıdır**; bu ölçek farkı kararı değiştirmez (Hastie vd., 2009).

````python
ada_model = ada_ilk.named_steps["ada"]
ozellik_adlari = list(X.columns)


def kutuk_kurali(kutuk):
    """Bir kütüğün sorduğu soruyu ve iki koldaki tahmini okunur metne çevirir."""
    agac = kutuk.tree_
    ozellik, esik = ozellik_adlari[agac.feature[0]], agac.threshold[0]
    sol, sag = int(np.argmax(agac.value[1])), int(np.argmax(agac.value[2]))        # sol: "≤ eşik" kolu, sağ: "> eşik" kolu
    ad = {0: "Diyabet yok", 1: "Diyabet var"}
    return ozellik, esik, f"{TR[ozellik]} ≤ {esik:.1f} ise → {ad[sol]}, değilse → {ad[sag]}"


kutuk_tablosu = pd.DataFrame({
    "tur": np.arange(1, len(ada_model.estimators_) + 1),
    "özellik": [kutuk_kurali(k)[0] for k in ada_model.estimators_],
    "eşik": [kutuk_kurali(k)[1] for k in ada_model.estimators_],
    "kütüğün kuralı": [kutuk_kurali(k)[2] for k in ada_model.estimators_],
    "hata ε": ada_model.estimator_errors_,
    "söz hakkı α": ada_model.estimator_weights_,
}).set_index("tur")

print("İlk 10 tur:")
kutuk_tablosu.head(10).round(3)
````

Hangi özellik kaç kez soru sormuş ve toplam ne kadar söz hakkı biriktirmiş? 👇

````python
ozet_ozellik = (kutuk_tablosu.groupby("özellik").agg(kütük_sayısı=("söz hakkı α", "size"), toplam_söz_hakkı=("söz hakkı α", "sum"))
                .sort_values("toplam_söz_hakkı", ascending=False))
ozet_ozellik["pay %"] = (100 * ozet_ozellik["toplam_söz_hakkı"] / ozet_ozellik["toplam_söz_hakkı"].sum()).round(1)

fig, axes = plt.subplots(1, 2, figsize=(15.5, 4.8), gridspec_kw={"width_ratios": [1.5, 1]})
ozellik_renk = dict(zip(X.columns, sns.color_palette("Set2", len(X.columns))))
axes[0].bar(kutuk_tablosu.index, kutuk_tablosu["söz hakkı α"], color=[ozellik_renk[o] for o in kutuk_tablosu["özellik"]], width=0.85)
axes[0].set(xlabel="Tur", ylabel="Söz hakkı α", title="Her turun söz hakkı (renk = sorulan özellik)")
handles = [plt.Rectangle((0, 0), 1, 1, color=ozellik_renk[o]) for o in X.columns]
axes[0].legend(handles, [TR[o] for o in X.columns], ncol=2, fontsize=8.5, loc="upper right")

sirali = ozet_ozellik.sort_values("toplam_söz_hakkı")
axes[1].barh([TR[o] for o in sirali.index], sirali["toplam_söz_hakkı"], color=[ozellik_renk[o] for o in sirali.index])
for i, (v, p) in enumerate(zip(sirali["toplam_söz_hakkı"], sirali["pay %"])):
    axes[1].text(v + 0.3, i, f"%{p}", va="center", fontsize=10)
axes[1].set(xlabel="Toplam söz hakkı", title="Hangi özellik toplamda daha çok konuştu?")
axes[1].grid(False)
fig.tight_layout()
kaydet(fig, "proje_06_kutuk_soz_haklari")
plt.show()
ozet_ozellik.round(2)
````

### 🎞️ 7.1 · Bir Hastanın Karar Yolculuğu (Animasyon)

Test setinden **gerçekten diyabeti olan ama kıl payı yakalanan** bir kişiyi seçelim. Her turda bir kütük oy verir; çubuklar o oyun söz hakkıyla çarpılmış halidir (yeşil = "risk var", kırmızı = "risk yok"), çizgi **oyların toplamıdır**. Toplam sıfırın üstüne çıkarsa karar "diyabet var" olur:

````python
X_test_dolu = ada_ilk.named_steps["doldur"].transform(X_test)                 # medyanla doldurulmuş test verisi
alfa_vek = ada_model.estimator_weights_
oylar = np.column_stack([2 * k.predict(X_test_dolu) - 1 for k in ada_model.estimators_])   # her kütük: +1 (diyabet) / -1 (yok)
skor_toplam = (oylar * alfa_vek).sum(axis=1)                                  # F(x) = Σ α·h(x)

# kendi oy toplamımız scikit-learn'ün kararıyla aynı mı?
ayni_mi = np.array_equal((skor_toplam > 0).astype(int), ada_ilk.predict(X_test))
print(f"Elle topladığımız ağırlıklı oyun kararı, sklearn.predict ile aynı mı?  {'EVET ✅' if ayni_mi else 'HAYIR ❌'}")

# seçilen hasta: gerçekten diyabetli, doğru yakalanmış ve skoru sıfıra en yakın olan
dogru_pozitif = np.where((y_test.to_numpy() == 1) & (skor_toplam > 0))[0]
hasta = dogru_pozitif[np.argmin(skor_toplam[dogru_pozitif])]
hasta_oy = oylar[hasta] * alfa_vek
kumulatif = np.cumsum(hasta_oy)
ozellik_dizisi = kutuk_tablosu["özellik"].to_numpy()

hasta_bilgi = X_test.iloc[hasta].rename(index=TR).round(1)
print(f"\nSeçilen hasta (test setindeki {hasta}. satır): gerçek durum = diyabet VAR · nihai skor F(x) = {kumulatif[-1]:+.2f}")
hasta_bilgi.to_frame("değer (eksikse NaN)").T
````

````python
def yolculuk_karesi(fig, i):
    fig.clf()
    ax = fig.add_subplot(111)
    tur = np.arange(1, i + 2)
    ax.bar(tur, hasta_oy[: i + 1], color=[RENK["evet"] if v > 0 else RENK["hayir"] for v in hasta_oy[: i + 1]], width=0.8, alpha=0.75)
    ax.plot(tur, kumulatif[: i + 1], color=RENK["koyu"], lw=3, marker="o", ms=4, label="oyların toplamı F(x)")
    ax.axhline(0, color="black", lw=1)
    ax.set(xlim=(0, len(hasta_oy) + 1), ylim=(min(hasta_oy.min(), kumulatif.min()) - 0.4, max(hasta_oy.max(), kumulatif.max()) + 0.5),
           xlabel="Kaçıncı kütük oy veriyor?", ylabel="Söz hakkıyla ağırlıklı oy")
    karar = "DİYABET VAR (risk)" if kumulatif[i] > 0 else "Diyabet yok"
    ax.set_title(f"{i + 1}. kütük sonrası toplam: {kumulatif[i]:+.2f}  →  karar: {karar}", fontsize=12)
    ax.legend(loc="lower right")
    ax.grid(alpha=0.25)
    fig.tight_layout()


YOLCULUK_GIF = FIG_DIR / "proje_01_hasta_yolculugu.gif"
gif_kaydet(yolculuk_karesi, len(hasta_oy), YOLCULUK_GIF, figsize=(8.4, 4.2), fps=6)
Image(filename=YOLCULUK_GIF, width=680)
````

AdaBoost'un iki ana düğmesi var:

- 🔢 **`n_estimators`**: kaç kütük (tur) eğitilecek?
- 🐢 **`learning_rate`** (öğrenme hızı): her kütüğün oyunu **ne kadar kısmak** gerektiği. Küçük hız → daha temkinli, daha çok tur gerekir.

Bunları **test setine bakmadan**, yalnızca eğitim setindeki 5 katlı çapraz doğrulamayla inceleyelim. (Her katta medyan doldurma **yeniden** o katın eğitim kısmından öğrenilir.)

````python
MAX_TUR = 300
ORANLAR = [0.05, 0.25, 1.0]


def asamali_cv_auc(ogrenme_hizi, max_tur=MAX_TUR):
    """Her turda 5 katlı çapraz doğrulama ROC-AUC'sini (aşama aşama) hesaplar. Sızıntısızdır: doldurma her katta yeniden öğrenilir."""
    egri_e = np.zeros(max_tur); egri_v = np.zeros(max_tur)
    for tr, va in cv5.split(X_egitim, y_egitim):
        dolgu = SimpleImputer(strategy="median").fit(X_egitim.iloc[tr])
        Xtr, Xva = dolgu.transform(X_egitim.iloc[tr]), dolgu.transform(X_egitim.iloc[va])
        m = AdaBoostClassifier(n_estimators=max_tur, learning_rate=ogrenme_hizi, random_state=RANDOM_STATE).fit(Xtr, y_egitim.iloc[tr])
        egri_e += [roc_auc_score(y_egitim.iloc[tr], d) for d in m.staged_decision_function(Xtr)]
        egri_v += [roc_auc_score(y_egitim.iloc[va], d) for d in m.staged_decision_function(Xva)]
    return egri_e / 5, egri_v / 5


cv_egrileri = {lr: asamali_cv_auc(lr) for lr in ORANLAR}

fig, ax = plt.subplots(figsize=(10, 5))
for (lr, (e, v)), renk in zip(cv_egrileri.items(), [RENK["mavi"], RENK["evet"], RENK["hayir"]]):
    ax.plot(range(1, MAX_TUR + 1), v, color=renk, lw=2.5, label=f"öğrenme hızı {lr}  (doğrulama)")
    ax.plot(range(1, MAX_TUR + 1), e, color=renk, lw=1.2, ls="--", alpha=0.6)
    en_iyi = int(np.argmax(v)) + 1
    ax.scatter([en_iyi], [v[en_iyi - 1]], color=renk, edgecolor="black", s=90, zorder=5)
    ax.annotate(f"{en_iyi}. tur\n{v[en_iyi - 1]:.3f}", (en_iyi, v[en_iyi - 1]), textcoords="offset points", xytext=(6, -26), fontsize=9, color=renk)
ax.set(xlabel="Tur (kütük) sayısı", ylabel="ROC-AUC", title="Kaç tur yeter? (düz çizgi = doğrulama, kesikli = eğitim)", ylim=(0.78, 1.0))
ax.legend(loc="lower right")
kaydet(fig, "proje_07_tur_sayisi_ogrenme_hizi")
plt.show()

pd.DataFrame({f"hız {lr}": {"en iyi tur": int(np.argmax(v)) + 1, "doğrulama AUC": round(v.max(), 4), "300. turda AUC": round(v[-1], 4), "300. turda eğitim AUC": round(e[-1], 4)}
              for lr, (e, v) in cv_egrileri.items()})
````

````python
adaylar = {
    "Taban": basit_pipeline(DummyClassifier(strategy="most_frequent")),
    "Lojistik regresyon": Pipeline([("doldur", SimpleImputer(strategy="median")), ("olcek", StandardScaler()), ("model", LogisticRegression(max_iter=1000))]),
    "KNN (k=15)": Pipeline([("doldur", SimpleImputer(strategy="median")), ("olcek", StandardScaler()), ("model", KNeighborsClassifier(n_neighbors=15))]),
    "Karar ağacı (derinlik 3)": basit_pipeline(DecisionTreeClassifier(max_depth=3, random_state=RANDOM_STATE)),
    "Random Forest (300)": basit_pipeline(RandomForestClassifier(n_estimators=300, min_samples_leaf=3, random_state=RANDOM_STATE, n_jobs=-1)),
    "Gradient Boosting": basit_pipeline(GradientBoostingClassifier(random_state=RANDOM_STATE)),
    "AdaBoost (varsayılan)": pipeline_kur(),
    "AdaBoost (ayarlı)": ada_ayarli,
}

cv_tablo, auc_dagilim = [], {}
for ad, m in adaylar.items():
    cv = cross_validate(m, X_egitim, y_egitim, cv=cv_tekrar, scoring=["roc_auc", "accuracy", "recall", "f1"], n_jobs=-1)
    auc_dagilim[ad] = cv["test_roc_auc"]
    cv_tablo.append({"model": ad, "CV ROC-AUC": cv["test_roc_auc"].mean(), "AUC std": cv["test_roc_auc"].std(),
                     "CV doğruluk": cv["test_accuracy"].mean(), "CV duyarlılık": cv["test_recall"].mean(), "CV F1": cv["test_f1"].mean()})
cv_tablo = pd.DataFrame(cv_tablo).set_index("model").sort_values("CV ROC-AUC", ascending=False)

fig, ax = plt.subplots(figsize=(10, 5.2))
sira = list(cv_tablo.index)[::-1]
kutu = ax.boxplot([auc_dagilim[a] for a in sira], orientation="horizontal", tick_labels=sira, patch_artist=True, widths=0.6)
for patch, ad in zip(kutu["boxes"], sira):
    patch.set(facecolor=RENK["altin"] if "AdaBoost" in ad else RENK["gri"], alpha=0.85)
ax.set(xlabel="ROC-AUC (15 değerlendirme)", title="Rakip modellerin çapraz doğrulama ROC-AUC dağılımı (AdaBoost = altın)")
ax.grid(axis="y", visible=False)
kaydet(fig, "proje_13_model_yarisi")
plt.show()
cv_tablo.round(3)
````

### 💸 13.2 · Hata Maliyetine Göre Eşik Seçimi

Aşağıdaki maliyetler **örnektir (varsayım)**; gerçek kurumda sağlık ekonomisi uzmanlarıyla belirlenir:

- ⚠️ **Yanlış alarm (FP):** gereksiz ileri tetkik çağrısı → **1 birim**
- ❌ **Kaçırılan hasta (FN):** geç teşhis → **5 birim**

Eşiği **eğitim verisindeki 5 katlı tahminlerle** (test setine bakmadan) en düşük toplam maliyeti verecek şekilde seçiyoruz. Teorik bir kontrol de var: kalibre olasılıklarda maliyet-optimal eşik $\;t^* = \frac{C_{FP}}{C_{FP}+C_{FN}} = \frac{1}{6}\approx 0.17$ olmalıdır.

````python
MALIYET_FP, MALIYET_FN = 1, 5

oof_kal = cross_val_predict(CalibratedClassifierCV(ada_ayarli, method="sigmoid", cv=5), X_egitim, y_egitim, cv=cv5, method="predict_proba")[:, 1]
adaylar_esik = np.linspace(0.05, 0.8, 76)


def toplam_maliyet(y_, p_, t):
    tahmin = (p_ >= t).astype(int)
    fp = ((tahmin == 1) & (y_ == 0)).sum(); fn = ((tahmin == 0) & (y_ == 1)).sum()
    return MALIYET_FP * fp + MALIYET_FN * fn


maliyet_egitim = np.array([toplam_maliyet(y_egitim.to_numpy(), oof_kal, t) for t in adaylar_esik]) / len(y_egitim)
ESIK = float(adaylar_esik[np.argmin(maliyet_egitim)])
teorik = MALIYET_FP / (MALIYET_FP + MALIYET_FN)
print(f"Eğitim (OOF) verisinde maliyet-optimal eşik: {ESIK:.2f}   ·   teorik değer: {teorik:.2f}   ·   varsayılan 0.5")

fig, ax = plt.subplots(figsize=(9.5, 4.6))
ax.plot(adaylar_esik, maliyet_egitim, color=RENK["mor"], lw=3)
ax.axvline(ESIK, color=RENK["evet"], lw=2, ls="--", label=f"seçilen eşik {ESIK:.2f}")
ax.axvline(0.5, color=RENK["gri"], lw=2, ls="--", label="varsayılan 0.5")
ax.scatter([ESIK], [maliyet_egitim.min()], color=RENK["altin"], edgecolor="black", s=110, zorder=5)
ax.set(xlabel="Karar eşiği (kalibre olasılık)", ylabel="Kişi başı beklenen maliyet (birim)", title="Maliyet duyarlı eşik: 0.5 yerine düşük bir eşik daha ucuz")
ax.legend()
kaydet(fig, "proje_15_maliyet_esigi")
plt.show()
````

### 🚦 13.3 · Risk Basamakları

Kalibre olasılığı, sağlık personelinin kullanabileceği üç basamağa çeviriyoruz. Sınırlar: **Düşük** (< maliyet eşiği), **Orta** (eşik – 0.50), **Yüksek** (≥ 0.50). Basamakların gerçekten **farklı risk düzeylerini** ayırıp ayırmadığını test setinde gözlenen diyabet oranlarıyla doğruluyoruz:

````python
def risk_basamagi(p):
    return pd.cut(p, bins=[-0.01, ESIK, 0.5, 1.01], labels=["Düşük", "Orta", "Yüksek"])


basamak = pd.DataFrame({"basamak": risk_basamagi(p_kal), "gercek": y_test.to_numpy()})
basamak_tablo = (basamak.groupby("basamak", observed=True)
                 .agg(kişi=("gercek", "size"), gerçek_diyabetli=("gercek", "sum"), gözlenen_oran=("gercek", "mean")).reset_index())
basamak_tablo["kişi %"] = (100 * basamak_tablo["kişi"] / len(basamak)).round(1)
basamak_tablo["gözlenen_oran"] = basamak_tablo["gözlenen_oran"].round(3)
basamak_tablo.to_csv(RAPOR_DIR / "risk_basamaklari.csv", index=False)

fig, ax = plt.subplots(figsize=(8.5, 4.4))
renk_b = {"Düşük": RENK["evet"], "Orta": RENK["altin"], "Yüksek": RENK["hayir"]}
cubuk = ax.bar(basamak_tablo["basamak"].astype(str), basamak_tablo["gözlenen_oran"], color=[renk_b[b] for b in basamak_tablo["basamak"].astype(str)], width=0.55)
for c, (_, s) in zip(cubuk, basamak_tablo.iterrows()):
    ax.text(c.get_x() + c.get_width() / 2, c.get_height() + 0.015, f"%{s['gözlenen_oran'] * 100:.0f}\n({int(s['kişi'])} kişi)", ha="center", fontsize=10.5, fontweight="bold")
ax.axhline(y_test.mean(), color="black", ls="--", lw=1, label=f"genel oran %{y_test.mean() * 100:.0f}")
ax.set(ylim=(0, 1.05), ylabel="Gerçekte diyabetli oranı", title="Risk basamakları gerçek riski ayırıyor mu?")
ax.yaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1.0))
ax.legend(loc="upper left")
ax.grid(False)
kaydet(fig, "proje_17_risk_basamaklari")
plt.show()
basamak_tablo
````

### 🔎 13.4 · "Neden?" — Tek Bir Kişinin Kararı

Bir hasta "Yüksek risk" listesine girdiğinde sağlık görevlisi **gerekçe** isteyecektir. AdaBoost kütüklerden oluştuğu için gerekçe üretmek kolaydır: her özelliğin **kütük oylarına katkısını** toplarız (yeşil = riski artırır, kırmızı = azaltır). Bu, SHAP gibi bir yöntem değil, **modelin kendi oy tablosunun** özetidir.

````python
ada_son = ada_ayarli.named_steps["ada"]
dolgu_son = ada_ayarli.named_steps["doldur"]
kutuklar_son = ada_son.estimators_
ozellik_idx = np.array([k.tree_.feature[0] for k in kutuklar_son])


def neden_aciklama(X_satir, baslik, ax):
    """Bir kişi için özellik bazında oy katkısını (Σ α·oy) çizer."""
    xd = dolgu_son.transform(X_satir)
    oy = np.array([2 * k.predict(xd)[0] - 1 for k in kutuklar_son]) * ada_son.estimator_weights_
    katki = pd.Series({TR[c]: oy[ozellik_idx == i].sum() for i, c in enumerate(X.columns)}).sort_values()
    ax.barh(katki.index, katki.values, color=[RENK["evet"] if v > 0 else RENK["hayir"] for v in katki.values])
    ax.axvline(0, color="black", lw=1)
    ax.set_title(baslik, fontsize=11.5)
    ax.grid(False)
    return katki


yuksek = int(np.argmax(p_kal)); sinirda = int(np.argmin(np.abs(p_kal - 0.5)))
fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
for ax, idx, ad in zip(axes, [yuksek, sinirda], ["En yüksek riskli kişi", "Sınırda kalan kişi"]):
    neden_aciklama(X_test.iloc[[idx]], f"{ad}  ·  kalibre olasılık {p_kal[idx]:.2f}  ·  gerçek: {'diyabet VAR' if y_test.iloc[idx] else 'diyabet yok'}", ax)
    ax.set_xlabel("Oy katkısı (Σ söz hakkı × oy):  → riski artırır")
fig.tight_layout()
kaydet(fig, "proje_18_hasta_aciklamasi")
plt.show()

pd.concat([X_test.iloc[[yuksek]].rename(columns=TR).assign(**{"risk": "en yüksek"}), X_test.iloc[[sinirda]].rename(columns=TR).assign(**{"risk": "sınırda"})]).round(1)
````

### 🧑‍⚕️ 14.2 · Yeni Bir Kişi İçin Tahmin Fonksiyonu

Kullanıcılar eksik ölçümü çoğu zaman **0** yazar. Fonksiyon, eğitimdeki **aynı kuralı** (0 → eksik) uygular ve model dosyasını yükleyip risk basamağını döndürür:

````python
def risk_tahmin(**olcumler):
    """Bir kişinin ölçümlerinden (eksikler 0 veya None olabilir) kalibre olasılığı ve risk basamağını üretir."""
    meta_ = json.loads((MODEL_DIR / "meta.json").read_text(encoding="utf-8"))
    model_ = joblib.load(MODEL_DIR / "adaboost_diyabet.joblib")
    satir = pd.DataFrame([olcumler], columns=meta_["ozellik_sirasi"]).astype(float)
    satir[meta_["sifir_eksik_sayilacak_sutunlar"]] = satir[meta_["sifir_eksik_sayilacak_sutunlar"]].replace(0, np.nan)   # eğitimdeki kuralla aynı
    p = float(model_.predict_proba(satir)[0, 1])
    basamak_ = "Yüksek" if p >= meta_["risk_sinirlari"]["orta_yuksek"] else "Orta" if p >= meta_["risk_sinirlari"]["dusuk_orta"] else "Düşük"
    return {"olasılık": round(p, 3), "risk basamağı": basamak_}


ornekler = {
    "Genç, düşük glukoz, normal kilo": dict(Pregnancies=1, Glucose=88, BloodPressure=66, SkinThickness=22, Insulin=0, BMI=24.5, DiabetesPedigreeFunction=0.25, Age=24),
    "Orta yaş, sınırda glukoz, fazla kilolu": dict(Pregnancies=4, Glucose=125, BloodPressure=74, SkinThickness=0, Insulin=0, BMI=33.0, DiabetesPedigreeFunction=0.55, Age=38),
    "Yüksek glukoz, yüksek BMI, ileri yaş": dict(Pregnancies=7, Glucose=185, BloodPressure=82, SkinThickness=35, Insulin=0, BMI=38.4, DiabetesPedigreeFunction=0.80, Age=52),
}
pd.DataFrame({ad: risk_tahmin(**o) for ad, o in ornekler.items()}).T
````

### 📝 Mini Quiz

<details><summary><b>1.</b> Bu veride "0" değerlerine neden dikkat ettik? Hangi sütunlarda sıfır geçerli, hangilerinde değil?</summary>

Glukoz, kan basıncı, deri kıvrımı, insülin ve BMI'da 0 **fiziksel olarak imkânsızdır**, yani eksik veridir. **Hamilelik sayısında** 0 geçerlidir. Sıfırları bilgi sanmak modeli yanıltır.
</details>

<details><summary><b>2.</b> Medyanı bölmeden önce tüm veriden hesaplasak ne olurdu? Pipeline bunu nasıl çözer?</summary>

Test verisinin bilgisi eğitime **sızar** (veri sızıntısı); skorlar iyimser çıkar. `Pipeline` + çapraz doğrulamada medyan her katta **yalnızca o katın eğitim kısmından** öğrenilir.
</details>

<details><summary><b>3.</b> AdaBoost ölçeklemeye neden ihtiyaç duymaz?</summary>

Zayıf öğrenicileri karar kütükleridir; "değer > eşik mi?" diye sorarlar. Sayıların ölçeği eşiği değiştirir ama **sıralamayı** değiştirmez, dolayısıyla sonuç aynıdır (Bölüm 5.3 deneyi).
</details>

<details><summary><b>4.</b> Öğrenme hızını 0.05'ten 1.0'a çıkarmanın etkisi ne olur?</summary>

Küçük hız her kütüğün etkisini **kısar**: model aynı noktaya daha yavaş yaklaşır. Büyük hız (1.0) hızlı ilerler ama eğitim–doğrulama makası en çok açılan, **aşırı öğrenmeye en yatkın** seçenektir (Bölüm 8). Bu veride üç hızın zirve skoru benzerdir; fark eğrinin **şeklinde**dir. Çok küçük hız (0.001) ise 200 turda bile yetersiz kalır (Bölüm 11).
</details>

<details><summary><b>5.</b> `predict_proba` çıktısını doğrudan "%55 diyabet olasılığı" diye yorumlayabilir miyiz?</summary>

Hayır. AdaBoost'un olasılıkları 0.5 etrafında **sıkışmış, kalibre olmayan** skorlardır. Sıralama anlamlıdır ama değer gerçek olasılık değildir; gerçek olasılık için `CalibratedClassifierCV` kullanılır.
</details>

<details><summary><b>6.</b> Yanlış alarm 1 birim, kaçırılan hasta 5 birim ise teorik optimum eşik nedir? Neden 0.5 değil?</summary>

$t^* = 1/(1+5)\approx 0.17$. Kaçırmak 5 kat pahalı olduğundan **daha düşük skorlu kişileri de** çağırmak ucuzdur; bu yüzden eşik 0.5'ten aşağı iner.
</details>

<details><summary><b>7.</b> Ayarlamadan sonra test doğruluğu 1-2 puan arttıysa "ayar işe yaradı" diyebilir miyiz?</summary>

Dikkatle. 154 kişilik test setinde doğruluğun standart hatası ≈ 3.5 puandır; 1-2 puanlık fark **şans** olabilir. CV skorlarındaki fark ve standart sapmalar da birlikte değerlendirilmelidir.
</details>
