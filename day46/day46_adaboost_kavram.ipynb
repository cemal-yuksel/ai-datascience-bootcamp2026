### 🌱 2.1 · Zayıf Öğrenici Nedir?

**Zayıf öğrenici**, rastgele tahminden *biraz* iyi olan basit bir modeldir. AdaBoost'ta bu genellikle **karar kütüğüdür** (*decision stump*): derinliği **1** olan, yani **tek bir soru soran** karar ağacı.

| | 🌱 Zayıf öğrenici (kütük) | 🏆 Güçlü model (AdaBoost) |
|:--|:--|:--|
| **Kaç soru sorar?** | Yalnızca 1 | Yüzlerce kütüğün soruları |
| **Tek başına doğruluğu** | Yazı-turadan biraz iyi | Çok daha yüksek |
| **Yorumlaması** | Çok kolay | Orta |
| **Eğitim** | Saniyenin altında | Hızlı (kütükler küçük) |

Aşağıda iki özelliği olan küçük bir veri için tek bir kütüğün **ne kadar basit** bir şey olduğunu görelim:

````python
Xb, yb = make_moons(n_samples=200, noise=0.25, random_state=RANDOM_STATE)   # iki "yarım ay" şeklinde veri
yb_pm = np.where(yb == 1, 1, -1)                                             # etiketler: -1 / +1


def sinir_ciz(ax, tahmin_fn, X, y, baslik, adim=0.03):
    """Bir modelin karar sınırını (arka plan rengi) ve noktaları çizer."""
    xx, yy = np.meshgrid(np.arange(-1.7, 2.8, adim), np.arange(-1.2, 1.8, adim))
    zz = tahmin_fn(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, zz, levels=[-1.5, 0, 1.5], colors=[SINIF_RENK[-1], SINIF_RENK[1]], alpha=0.25)
    ax.scatter(X[:, 0], X[:, 1], c=[SINIF_RENK[int(s)] for s in y], edgecolor="white", linewidth=0.7, s=40)
    ax.set(title=baslik, xticks=[], yticks=[])
    ax.grid(False)
    ax.spines[["left", "bottom"]].set_visible(False)


kutuk = DecisionTreeClassifier(max_depth=1, random_state=RANDOM_STATE).fit(Xb, yb_pm)
ada = AdaBoostClassifier(n_estimators=50, random_state=RANDOM_STATE).fit(Xb, yb_pm)     # varsayılan zayıf öğrenici: kütük

fig, axes = plt.subplots(1, 2, figsize=(11, 4.4))
sinir_ciz(axes[0], kutuk.predict, Xb, yb_pm, f"Tek kütük: tek düz çizgi\ndoğruluk {kutuk.score(Xb, yb_pm):.2f}")
sinir_ciz(axes[1], ada.predict, Xb, yb_pm, f"50 kütüğün ağırlıklı oyu\ndoğruluk {ada.score(Xb, yb_pm):.2f}")
fig.suptitle("Aynı basit kütükler: tek başına zayıf, birlikte güçlü", fontsize=14, fontweight="bold", y=1.03)
kaydet(fig, "kavram_01_kutuk_ve_adaboost")
plt.show()
````

### 🪜 3.1 · Beş Adımın Sade Anlatımı

| Adım | Ne yapılır? | Gündelik benzetme |
|:-:|:--|:--|
| 1️⃣ | **Herkese eşit ağırlık** verilir: $n$ satır varsa her satır $1/n$ | Sınıftaki herkese eşit söz hakkı |
| 2️⃣ | Bu ağırlıklarla **ilk zayıf model** (kütük) eğitilir | İlk öğrenci tahtaya kalkar |
| 3️⃣ | **Yanlış** bildiği satırların ağırlığı **artar**, doğrularınki **azalır** | "Şu soruyu kaçırdık, bir dahaki sefere daha çok çalışalım" |
| 4️⃣ | Ağırlıklar **toplamı 1 olacak şekilde** yeniden ölçeklenir (normalleştirme) | Pastanın toplamı hep %100 |
| 5️⃣ | Belirlenen **tur sayısı** boyunca tekrarlanır; sonunda tüm modeller **ağırlıklı oy** kullanır | Kurul toplanır, güvenilir üyelerin oyu daha ağır sayılır |

Nihai karar, PDF notumuzdaki gösterimle şöyledir:

$$F(x) = \sum_{m=1}^{M} \alpha_m \, h_m(x) \qquad\Longrightarrow\qquad \text{karar} = \operatorname{işaret}\big(F(x)\big)$$

- $h_m(x)$: $m$. zayıf öğrenicinin tahmini ($+1$ = Evet, $-1$ = Hayır)
- $\alpha_m$: o modelin **söz hakkı** (güvenilirlik ağırlığı)
- $M$: toplam model (tur) sayısı

Her kütüğün oyu aynı ağırlıkta sayılmaz. **Hata oranı** ($\varepsilon$, epsilon) küçük olan model **büyük söz hakkı**, hata oranı yüksek olan model **küçük söz hakkı** alır:

$$\alpha = \frac{1}{2}\,\ln\!\left(\frac{1-\varepsilon}{\varepsilon}\right)$$

Burada $\varepsilon$ = *yanlış bilinen satırların ağırlıkları toplamı* (ağırlıklar toplamı 1 iken).

````python
eps = np.linspace(0.01, 0.99, 400)
alfa = 0.5 * np.log((1 - eps) / eps)

fig, ax = plt.subplots(figsize=(9.5, 4.8))
ax.plot(eps, alfa, color=RENK["mor"], lw=3)
ax.axhline(0, color="black", lw=1)
ax.fill_between(eps, alfa, 0, where=eps < 0.5, color=RENK["evet"], alpha=0.15, label="ε < 0.5  →  söz hakkı pozitif")
ax.fill_between(eps, alfa, 0, where=eps > 0.5, color=RENK["hayir"], alpha=0.15, label="ε > 0.5  →  söz hakkı negatif (modeli ters çevirmek daha iyi)")

# okunaklı birkaç örnek nokta
for e, not_, yon in [(0.05, "çok güvenilir", (12, 6)), (0.143, "PDF örneğimiz\nε = 1/7", (14, 8)), (0.25, "idare eder", (14, 8)), (0.5, "yazı-tura kadar\n→ söz hakkı = 0", (10, 18))]:
    a = 0.5 * np.log((1 - e) / e)
    ax.scatter([e], [a], color=RENK["altin"], edgecolor=RENK["koyu"], s=90, zorder=5)
    ax.annotate(f"ε={e:.3g}  →  α={a:.2f}\n{not_}", (e, a), textcoords="offset points", xytext=yon, fontsize=9.5)
ax.set(xlabel="Hata oranı  ε", ylabel="Söz hakkı  α", ylim=(-2.5, 2.8), title="Hata azaldıkça söz hakkı hızla artar")
ax.legend(loc="lower left", fontsize=9)
kaydet(fig, "kavram_02_soz_hakki")
plt.show()

tablo = pd.DataFrame({"hata oranı ε": [0.05, 0.10, 0.143, 0.25, 0.40, 0.50, 0.60]})
tablo["söz hakkı α"] = (0.5 * np.log((1 - tablo["hata oranı ε"]) / tablo["hata oranı ε"])).round(3)
tablo["yorum"] = ["çok güvenilir", "güvenilir", "PDF örneği (1 hata / 7 satır)", "idare eder", "zayıf ama işe yarar", "yazı-tura: oyu sayılmaz", "yanlış yönde! (negatif)"]
tablo.set_index("hata oranı ε")
````

````python
basvuru = pd.DataFrame({
    "ID":        [1, 2, 3, 4, 5, 6, 7],
    "GPA":       ["<3.0", "<3.0", "<3.0", ">3.0", ">3.0", ">3.0", "<3.0"],
    "Mülakat":   ["Düşük", "Yüksek", "Yüksek", "Düşük", "Yüksek", "Normal", "Normal"],
    "Onay":      ["Hayır", "Evet", "Evet", "Hayır", "Evet", "Evet", "Hayır"],
}).set_index("ID")

# Bilgisayar için sayıya çevirelim: GPA>3 → 1, Mülakat: Düşük=0 / Normal=1 / Yüksek=2, Onay: Evet=+1 / Hayır=-1
Xb7 = pd.DataFrame({"gpa_yuksek": (basvuru["GPA"] == ">3.0").astype(int),
                    "mulakat": basvuru["Mülakat"].map({"Düşük": 0, "Normal": 1, "Yüksek": 2})})
y7 = basvuru["Onay"].map({"Evet": 1, "Hayır": -1}).to_numpy()
n7 = len(y7)

basvuru
````

### 🔍 5.1 · İlk Kütük Hangi Soruyu Sormalı?

İlk turda iki aday soru var: **"GPA < 3 mü?"** ve **"Mülakat = Yüksek mi?"** Hangisi daha iyi ayırıyor? Cevaplar ne kadar **"saf"** (tek renk) kalırsa soru o kadar iyidir. Bunu **Gini safsızlığı** ile ölçeriz (0 = tamamen saf, 0.5 = tam karışık). Teknik ayrıntı için endişelenmeyin: **küçük sayı = iyi soru**.

````python
def gini(evet, hayir):
    """Bir kutudaki Evet/Hayır sayısından safsızlık: 0 = saf, 0.5 = tam karışık."""
    t = evet + hayir
    return 0.0 if t == 0 else 1 - (evet / t) ** 2 - (hayir / t) ** 2


def soru_degerlendir(ad, evet_maske):
    sol, sag = evet_maske, ~evet_maske                     # soruya "evet" diyenler / "hayır" diyenler
    satir = {"soru": ad}
    for yon, m in (("EVET kolu", sol), ("HAYIR kolu", sag)):
        e, h = int((y7[m] == 1).sum()), int((y7[m] == -1).sum())
        satir[yon] = f"{e}E / {h}H"
    toplam = (sol.sum() * gini((y7[sol] == 1).sum(), (y7[sol] == -1).sum()) +
              sag.sum() * gini((y7[sag] == 1).sum(), (y7[sag] == -1).sum())) / n7
    satir["ağırlıklı safsızlık"] = round(toplam, 3)
    return satir


adaylar = pd.DataFrame([
    soru_degerlendir("GPA < 3.0 mü?", (Xb7["gpa_yuksek"] == 0).to_numpy()),
    soru_degerlendir("Mülakat = Yüksek mi?", (Xb7["mulakat"] == 2).to_numpy()),
]).set_index("soru")
adaylar
````

**Mülakat = Yüksek mi?** sorusu kazanır: *Evet* kolunda 3 kişinin **üçü de** onaylanmış (saf!), *Hayır* kolunda ise 4 kişiden 3'ü reddedilmiş. GPA sorusu ise iki koldan da karışık sonuç veriyor. Bu seçimi bilgisayara yaptırırsak aynı sonucu bulur ve kütüğü bir **ağaç diyagramı** olarak çizebiliriz:

### 🧮 6.1 · Hata Oranı ve Söz Hakkı

Başta **7 satırın hepsi eşit ağırlıkta**: $w_i = 1/7 \approx 0.143$. Yanlış bilinen tek satır olduğundan:

$$\varepsilon_1 = \frac{1}{7} \approx 0.143 \qquad \alpha_1 = \tfrac12 \ln\!\Big(\tfrac{1-0.143}{0.143}\Big) \approx 0.896$$

````python
w = np.full(n7, 1 / n7)                                       # Adım 1: herkese eşit ağırlık
eps1 = w[yanlis1].sum()                                       # hata oranı = yanlışların ağırlık toplamı
alfa1 = 0.5 * np.log((1 - eps1) / eps1)                       # söz hakkı
print(f"ε₁ = {eps1:.4f}   α₁ = {alfa1:.4f}")
````

### ⚖️ 6.2 · Ağırlıklar Nasıl Güncellenir?

Yanlışların ağırlığı $e^{\alpha}$ ile **çarpılır** (büyür), doğruların ağırlığı $e^{-\alpha}$ ile **çarpılır** (küçülür). Sonra toplam 1 olsun diye hepsi toplama bölünür (**normalleştirme**):

$$w_i' = w_i \cdot e^{\mp\alpha} \qquad\qquad w_i'' = \frac{w_i'}{\sum_j w_j'}$$

````python
carpan = np.where(yanlis1, np.exp(alfa1), np.exp(-alfa1))      # yanlış: e^α (büyür) · doğru: e^-α (küçülür)
w_ham = w * carpan                                            # Adım 3: güncelle
w_yeni = w_ham / w_ham.sum()                                  # Adım 4: normalleştir (toplam = 1)

agirlik_tablosu = pd.DataFrame({
    "doğru mu?": np.where(yanlis1, "❌", "✅"),
    "eski ağırlık w": w.round(3),
    "çarpan e^(∓α)": carpan.round(3),
    "güncel ağırlık w'": w_ham.round(3),
    "normalleşmiş w''": w_yeni.round(3),
}, index=basvuru.index)
print(f"Güncel ağırlıkların toplamı: {w_ham.sum():.3f}  →  1'e bölünce normalleşmiş toplam: {w_yeni.sum():.0f}")
agirlik_tablosu
````

### 🗳️ 8.1 · Nihai Karar: Ağırlıklı Oylama

İki kütüğümüz var: $\alpha_1 \approx 0.896$ (Mülakat kütüğü) ve $\alpha_2 \approx 0.549$ (GPA kütüğü). Yeni bir aday gelince **ikisi de oy verir**, oylar söz hakkıyla çarpılıp toplanır:

$$F(x) = \alpha_1 h_1(x) + \alpha_2 h_2(x) \qquad\Longrightarrow\qquad \text{karar} = \operatorname{işaret}(F(x))$$

<p align="center">
  <img src="assets/diyagram_oylama.svg" width="100%" alt="Ağırlıklı oylama">
</p>

PDF'teki iki test adayını hesaplayalım:

````python
def kutuk1_oyu(gpa, mulakat):  return 1 if mulakat == "Yüksek" else -1        # Mülakat kütüğü
def kutuk2_oyu(gpa, mulakat):  return 1 if gpa > 3.0 else -1                 # GPA kütüğü


adaylar = [("Aday A", 3.2, "Yüksek"), ("Aday B", 2.8, "Yüksek")]
satirlar = []
for ad, gpa, mul in adaylar:
    o1, o2 = kutuk1_oyu(gpa, mul), kutuk2_oyu(gpa, mul)
    F = alfa1 * o1 + alfa2 * o2
    satirlar.append({"aday": ad, "GPA": gpa, "Mülakat": mul, "oy₁ (Mülakat)": o1, "oy₂ (GPA)": o2,
                     "α₁·oy₁": round(alfa1 * o1, 3), "α₂·oy₂": round(alfa2 * o2, 3), "F(x)": round(F, 3), "karar": "Evet ✅" if F > 0 else "Hayır ❌"})
oy_tablosu = pd.DataFrame(satirlar).set_index("aday")

fig, axes = plt.subplots(1, 2, figsize=(10.5, 3.9), sharey=True)
for ax, (ad, gpa, mul), satir in zip(axes, adaylar, satirlar):
    ust_t, alt_t = 0.0, 0.0                                   # pozitif oylar yukarı, negatifler aşağı yığılır
    for etiket, v in zip(["Mülakat kütüğü", "GPA kütüğü"], [satir["α₁·oy₁"], satir["α₂·oy₂"]]):
        taban = ust_t if v > 0 else alt_t
        ax.bar([0], [v], bottom=[taban], color=RENK["evet"] if v > 0 else RENK["hayir"], width=0.5, edgecolor="white")
        ax.text(0.3, taban + v / 2, f"{etiket}\n{v:+.2f}", va="center", fontsize=9.5)
        if v > 0: ust_t += v
        else: alt_t += v
    ax.axhline(0, color="black", lw=1)
    ax.axhline(satir["F(x)"], color=RENK["altin"], lw=2.5, ls="--")
    ax.set(title=f"{ad}: GPA {gpa}, Mülakat {mul}\nF(x) = {satir['F(x)']:+.2f}  →  {satir['karar'].split()[0]}", xticks=[], xlim=(-0.5, 1.2), ylim=(-0.9, 1.7))
    ax.grid(False)
axes[0].set_ylabel("Oyların toplamı")
fig.suptitle("Ağırlıklı oylama: güvenilir kütük (α₁) daha ağır basar", fontsize=13, fontweight="bold", y=1.04)
kaydet(fig, "kavram_06_agirlikli_oy")
plt.show()
oy_tablosu
````

</p>

Beş adımı yaklaşık **15 satırda** kodlayabiliriz. Zayıf öğrenici olarak `DecisionTreeClassifier(max_depth=1)` kullanıyor ve ağırlıkları `sample_weight` ile veriyoruz.

````python
def adaboost_sifirdan(X, y_pm, tur):
    """Ayrık AdaBoost: her turun kütüğünü, hata oranını (ε), söz hakkını (α) ve ağırlıklarını kaydeder."""
    w = np.full(len(y_pm), 1 / len(y_pm))                                   # Adım 1: eşit ağırlık
    kayit = []
    for _ in range(tur):
        kutuk = DecisionTreeClassifier(max_depth=1, random_state=RANDOM_STATE).fit(X, y_pm, sample_weight=w)   # Adım 2
        h = kutuk.predict(X)
        eps = np.clip(w[h != y_pm].sum(), 1e-10, 1 - 1e-10)                  # hata oranı
        alfa = 0.5 * np.log((1 - eps) / eps)                                 # söz hakkı
        kayit.append({"model": kutuk, "w": w.copy(), "h": h, "eps": eps, "alfa": alfa})
        w = w * np.exp(-alfa * y_pm * h)                                     # Adım 3: yanlışlar ↑, doğrular ↓
        w /= w.sum()                                                         # Adım 4: normalleştir
    return kayit                                                             # Adım 5: döngü `tur` kez tekrarlandı


def adaboost_skor(kayit, X):
    """F(x) = Σ α·h(x)"""
    return sum(k["alfa"] * k["model"].predict(X) for k in kayit)


# Önce 7 satırlık örneğimize uygulayalım (3 tur)
kayit7 = adaboost_sifirdan(Xb7.to_numpy(), y7, tur=3)
ozet7 = pd.DataFrame([{"tur": t, "ε": round(k["eps"], 3), "α": round(k["alfa"], 3), "yanlış bilinen ID": list(basvuru.index[k["h"] != y7])}
                      for t, k in enumerate(kayit7, 1)]).set_index("tur")
print(f"3 turluk AdaBoost'un 7 satırdaki doğruluğu: {accuracy_score(y7, np.sign(adaboost_skor(kayit7, Xb7.to_numpy()))):.2f}")
ozet7
````

**Üç tur sonunda 7 satırın hepsi doğru!** Tur 1'de ID 6 yanlıştı → Tur 2'de ağırlığı fırladı, doğru bilindi → Tur 3'te diğer zor satırlar düzeltildi. Şimdi aynı kodu daha gerçekçi, kıvrımlı bir veriye uygulayıp `scikit-learn` ile karşılaştıralım.

### 🎞️ 9.2 · Boosting İş Başında (Animasyon)

Sol panel: o ana kadar eğitilmiş kütüklerin **birleşik karar sınırı** ve **nokta ağırlıkları** (büyük nokta = zor satır). Sağ panel: tur sayısına göre eğitim ve test doğruluğu.

````python
# Gösterim için küçük bir veri ve 40 tur; her turda yalnızca KÜTÜK (derinlik 1) kullanıyoruz
Xg, yg = make_moons(n_samples=300, noise=0.20, random_state=5)
yg_pm = np.where(yg == 1, 1, -1)
Xg_e, Xg_t, yg_e, yg_t = train_test_split(Xg, yg_pm, test_size=0.3, random_state=RANDOM_STATE, stratify=yg_pm)

w = np.full(len(yg_e), 1 / len(yg_e))
turlar = []
for _ in range(40):
    m = DecisionTreeClassifier(max_depth=1, random_state=RANDOM_STATE).fit(Xg_e, yg_e, sample_weight=w)
    h = m.predict(Xg_e)
    eps = np.clip(w[h != yg_e].sum(), 1e-10, 1 - 1e-10)
    alfa = 0.5 * np.log((1 - eps) / eps)
    turlar.append({"model": m, "alfa": alfa, "w": w.copy()})
    w = w * np.exp(-alfa * yg_e * h)
    w /= w.sum()

gx, gy = np.meshgrid(np.arange(-1.8, 2.9, 0.04), np.arange(-1.3, 1.9, 0.04))
izgara = np.c_[gx.ravel(), gy.ravel()]
sk_izgara = np.cumsum([k["alfa"] * k["model"].predict(izgara) for k in turlar], axis=0)
sk_egitim = np.cumsum([k["alfa"] * k["model"].predict(Xg_e) for k in turlar], axis=0)
sk_test = np.cumsum([k["alfa"] * k["model"].predict(Xg_t) for k in turlar], axis=0)
dog_e = [accuracy_score(yg_e, np.sign(s)) for s in sk_egitim]
dog_t = [accuracy_score(yg_t, np.sign(s)) for s in sk_test]


def boosting_karesi(fig, i):
    fig.clf()
    ax1, ax2 = fig.subplots(1, 2, gridspec_kw={"width_ratios": [1.15, 1]})
    ax1.contourf(gx, gy, (sk_izgara[i] > 0).reshape(gx.shape), levels=[-0.5, 0.5, 1.5], colors=[SINIF_RENK[-1], SINIF_RENK[1]], alpha=0.28)
    ax1.scatter(Xg_e[:, 0], Xg_e[:, 1], s=turlar[i]["w"] * len(yg_e) * 24, c=[SINIF_RENK[int(s)] for s in yg_e], edgecolor="white", linewidth=0.5)
    ax1.set(title=f"{i + 1}. tur sonunda sınır", xticks=[], yticks=[])
    ax1.grid(False)
    ax1.spines[["left", "bottom"]].set_visible(False)
    ax2.plot(range(1, i + 2), dog_e[: i + 1], color=RENK["mavi"], lw=2.5, label="eğitim")
    ax2.plot(range(1, i + 2), dog_t[: i + 1], color=RENK["hayir"], lw=2.5, label="test")
    ax2.set(xlim=(1, 40), ylim=(0.6, 1.02), xlabel="Tur sayısı", ylabel="Doğruluk", title="Doğruluk tırmanıyor")
    ax2.legend(loc="lower right")
    fig.tight_layout()


BOOST_GIF = FIG_DIR / "kavram_03_boosting_animasyonu.gif"
gif_kaydet(boosting_karesi, 40, BOOST_GIF, figsize=(9.2, 4.1), fps=4, dpi=75)
Image(filename=BOOST_GIF, width=720)
````

AdaBoost yalnızca **sınıflandırma** (Evet/Hayır) için değil, **sayı tahmini** (maaş, fiyat, süre) için de kullanılır. PDF notundaki fikir şudur: tahminlerle gerçekler arasındaki farkı (**artık**, *residual*) her turda yeni bir küçük modele öğretmek.

| | 🏷️ Sınıflandırma | 📈 Regresyon |
|:--|:--|:--|
| **Tahmin** | Sınıf etiketi (Evet/Hayır) | Sayı (maaş, fiyat…) |
| **Her turda öğrenilen** | Yanlış bilinen **satırlar** (ağırlıklar) | Kalan hata: **artık** $r_i = y_i - F(x_i)$ |
| **Güncelleme** | Ağırlıklı oy | $F_m(x) = F_{m-1}(x) + \alpha_m h_m(x)$ |

### 🧑‍💼 10.1 · PDF Örneği: Deneyim → Maaş

````python
maas = pd.DataFrame({"Deneyim (yıl)": [1, 2, 3, 4, 5, 6], "Maaş (bin $)": [40, 45, 50, 58, 60, 65]}, index=pd.RangeIndex(1, 7, name="ID"))
xr = maas[["Deneyim (yıl)"]].to_numpy()
yr = maas["Maaş (bin $)"].to_numpy(dtype=float)

F0 = np.full(len(yr), yr.mean())                              # İlk model: ortalama maaş
maas.assign(**{"F₀(x) = ortalama": F0, "artık r₁ = Y − F₀": yr - F0})
````

PDF'teki gibi ilk model **ortalama maaştır** (53). Artıklar $-13, -8, -3, +5, +7, +12$. Şimdi her turda bir kütük bu artıkları öğrenir ve modele eklenir ($\alpha=1$ alıyoruz). *PDF'te ilk kütük örnek olsun diye "−10 / +8" değerleriyle gösterilmişti; biz değerleri ve eşiği bilgisayara hesaplatıyoruz (ilk turda eşik 3.5 yıl, değerler −8 / +5.25).*

````python
TUR_R = 4
Fler = [F0.copy()]
artiklar = [yr - F0]
esikler = []
for _ in range(TUR_R):
    kutuk_r = DecisionTreeRegressor(max_depth=1, random_state=RANDOM_STATE).fit(xr, artiklar[-1])   # h_m: artıkları tahmin et
    esikler.append(kutuk_r.tree_.threshold[0])
    Fler.append(Fler[-1] + 1.0 * kutuk_r.predict(xr))                                               # F_m = F_{m-1} + α·h_m (α = 1)
    artiklar.append(yr - Fler[-1])

ozet_r = pd.DataFrame({
    "tur": range(0, TUR_R + 1),
    "kütük sorusu": ["(başlangıç: ortalama)"] + [f"Deneyim ≤ {e:.1f} yıl mı?" for e in esikler],
    "ortalama |artık| (bin $)": [round(np.abs(a).mean(), 2) for a in artiklar],
}).set_index("tur")
ozet_r
````

### 🎞️ 10.2 · Artıkları Avlayan Model (Animasyon)

````python
xs = np.linspace(0.6, 6.4, 300).reshape(-1, 1)
# animasyon için aynı turlarla modeli yeniden kuralım (yoğun x ekseninde eğriyi çizebilmek için)
F_yogun = [np.full(len(xs), yr.mean())]
r_i = yr - F0
for _ in range(TUR_R):
    k = DecisionTreeRegressor(max_depth=1, random_state=RANDOM_STATE).fit(xr, r_i)
    F_yogun.append(F_yogun[-1] + k.predict(xs))
    r_i = r_i - k.predict(xr)


def regresyon_karesi(fig, i):
    fig.clf()
    ax = fig.add_subplot(111)
    ax.plot(xs, F_yogun[i], color=RENK["mor"], lw=3, label=f"Model F{i}(x)")
    ax.scatter(xr, yr, color=RENK["altin"], edgecolor=RENK["koyu"], s=90, zorder=5, label="Gerçek maaş")
    for x_, y_, f_ in zip(xr.ravel(), yr, Fler[i]):
        ax.plot([x_, x_], [y_, f_], color=RENK["hayir"], lw=2, alpha=0.8)
    ax.set(xlabel="Deneyim (yıl)", ylabel="Maaş (bin $)", ylim=(35, 70), title=f"Tur {i}: ortalama |artık| = {np.abs(artiklar[i]).mean():.2f}  (kırmızı çizgi = hata)")
    ax.legend(loc="upper left")
    fig.tight_layout()


REG_GIF = FIG_DIR / "kavram_04_regresyon_animasyonu.gif"
gif_kaydet(regresyon_karesi, TUR_R + 1, REG_GIF, figsize=(7.2, 4.2), fps=1, dpi=80)
Image(filename=REG_GIF, width=600)
````

### ⚖️ Artılar ve Eksiler

| ✅ Artılar | ⚠️ Eksiler |
|:--|:--|
| Basit ve **hızlı** zayıf öğrenicilerle çalışır | **Gürültüye / hatalı etikete** duyarlı (zor satırlara kilitlenebilir) |

### 📝 Mini Quiz

<details><summary><b>1.</b> "Zayıf öğrenici" ne demektir? AdaBoost'ta hangi model kullanılır?</summary>

Rastgele tahminden **biraz iyi** olan basit modeldir. AdaBoost'ta genellikle **karar kütüğü** (derinliği 1 olan, tek soru soran karar ağacı) kullanılır.
</details>

<details><summary><b>2.</b> Bir turda yanlış bilinen satırların ağırlığına ne olur? Normalleştirme neden yapılır?</summary>

Ağırlıkları $e^{\alpha}$ ile çarpılarak **artar**, doğrularınki $e^{-\alpha}$ ile çarpılarak **azalır**. Normalleştirme ağırlıkların toplamını yeniden 1 yapar; böylece bir sonraki turda ağırlıklar **olasılık** gibi (ve rulet dilimleri gibi) kullanılabilir.
</details>

<details><summary><b>3.</b> Bir modelin hata oranı ε = 0.5 ise söz hakkı α kaç olur? Bu ne anlama gelir?</summary>

$\alpha = \tfrac12\ln(1)=0$. Model **yazı-tura atmaktan farksızdır**, bu yüzden oyu hiç sayılmaz.
</details>

<details><summary><b>4.</b> Örneğimizde ε₁ = 1/7 için α₁ kaçtır? Normalleşmiş yeni ağırlıkta yanlış bilinen satırın payı kaçtır?</summary>

$\alpha_1 = \tfrac12\ln(6)\approx 0.896$. Yanlış bilinen satırın normalleşmiş ağırlığı tam **0.5** olur (doğru bilinen 6 satır kalan 0.5'i paylaşır, her biri 1/12 ≈ 0.083).
</details>

<details><summary><b>5.</b> GPA = 2.8, Mülakat = Yüksek olan aday için kütükler farklı oy verirse (Mülakat: Evet, GPA: Hayır) nihai karar neye göre belirlenir?</summary>

**Söz hakkı** büyük olan kütüğe göre: $F=0.896\cdot(+1)+0.549\cdot(-1)\approx+0.35>0$ → **Evet**. Oylar eşit değil, güvenilirliğe göre tartılır.
</details>

<details><summary><b>6.</b> AdaBoost gürültülü veride neden sorun yaşayabilir?</summary>

Hatalı etiketli satırlar sürekli yanlış bilindiği için ağırlıkları tur tur **büyür**; model bu "sorunlu" satırlara odaklanıp genel örüntüyü kaçırabilir. Çözüm: daha az tur, daha küçük öğrenme hızı, veri temizliği.
</details>
