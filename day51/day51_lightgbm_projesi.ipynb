```python
# ── Kütüphaneler ─────────────────────────────────────────────
import hashlib
import json
import os
import platform
import subprocess
import sys
import time
import warnings
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
import seaborn as sns
import sklearn
import xgboost
import lightgbm
import catboost
from IPython.display import Image, Markdown, display
from joblib import Parallel, delayed
from matplotlib.animation import FuncAnimation, PillowWriter
from scipy.stats import loguniform, randint, uniform

from sklearn.base import clone
from sklearn.calibration import CalibratedClassifierCV, calibration_curve
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import (AdaBoostClassifier, BaggingClassifier, ExtraTreesClassifier, GradientBoostingClassifier,
                              HistGradientBoostingClassifier, RandomForestClassifier, StackingClassifier, VotingClassifier)
from sklearn.feature_selection import mutual_info_classif
from sklearn.gaussian_process import GaussianProcessClassifier
from sklearn.inspection import permutation_importance
from sklearn.linear_model import LogisticRegression, Perceptron, RidgeClassifier, SGDClassifier
from sklearn.metrics import (accuracy_score, average_precision_score, balanced_accuracy_score, brier_score_loss,
                             classification_report, confusion_matrix, f1_score, log_loss, matthews_corrcoef,
                             precision_recall_curve, precision_score, recall_score, roc_auc_score, roc_curve)
from sklearn.model_selection import (GroupKFold, RandomizedSearchCV, RepeatedStratifiedKFold, StratifiedGroupKFold,
                                     StratifiedKFold, cross_val_predict, train_test_split)
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC, LinearSVC
from sklearn.tree import DecisionTreeClassifier, ExtraTreeClassifier
from xgboost import XGBClassifier, XGBRFClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

from boston_yardimci import (ACIKLAMA, ESIK_BIN_DOLAR, ILCE_SUTUNLARI, OZELLIK_A, OZELLIK_B, SINIF_ADLARI,
                             BostonOzellikleri, ilce_kimligi, veri_oku)   # pipeline içindeki yardımcılar (ayrı dosyada)

warnings.filterwarnings("ignore")
os.environ["PYTHONWARNINGS"] = "ignore"                    # paralel işçi süreçlerde de uyarıları sustur

# ── Sabitler ve klasörler ───────────────────────────────────
RANDOM_STATE = 42
N_ISCI = 4                                                  # paralel çalışacak süreç sayısı
VERI_YOLU = Path("23-boston.csv")
FIG_DIR, RAPOR_DIR, MODEL_DIR = Path("outputs/figures"), Path("outputs/reports"), Path("outputs/model")
for klasor in (FIG_DIR, RAPOR_DIR, MODEL_DIR):
    klasor.mkdir(parents=True, exist_ok=True)

# ── Görsel tema (notebook boyunca aynı renkler) ──────────────
RENK = {"altin": "#EDAE49", "mor": "#6C5CE7", "pembe": "#D1495B", "mavi": "#2B6CB0",
        "yesil": "#2E9E6B", "turuncu": "#E8833A", "gri": "#8A94A6", "koyu": "#2E4057", "cyan": "#0E9AA7"}
SINIF_RENK = {0: RENK["mavi"], 1: RENK["altin"]}            # standart: mavi · premium: altın
AILE_RENK = {"Taban": "#B8C0CC", "Doğrusal": RENK["altin"], "Olasılıksal": RENK["cyan"], "Komşuluk ve çekirdek": RENK["pembe"],
             "Sinir ağı": RENK["mavi"], "Ağaç ve orman": RENK["yesil"], "Boosting": RENK["turuncu"], "Topluluk": RENK["mor"]}

plt.rcParams.update({
    "figure.dpi": 100, "savefig.dpi": 150, "savefig.bbox": "tight",
    "axes.titleweight": "bold", "axes.titlesize": 13, "axes.grid": True, "grid.alpha": 0.25,
    "axes.spines.top": False, "axes.spines.right": False,
})
sns.set_theme(style="whitegrid", rc={"axes.spines.top": False, "axes.spines.right": False, "grid.alpha": 0.25})


def kaydet(fig, ad):
    """Grafiği outputs/figures altına PNG olarak kaydeder."""
    yol = FIG_DIR / f"{ad}.png"
    fig.savefig(yol)
    print(f"💾 Kaydedildi → {yol.as_posix()}")


def rapor_kaydet(df, ad, index=True):
    """Tabloyu outputs/reports altına CSV olarak kaydeder."""
    yol = RAPOR_DIR / f"{ad}.csv"
    df.to_csv(yol, index=index, encoding="utf-8")
    print(f"💾 Rapor → {yol.as_posix()}  ({len(df)} satır)")


def gif_kaydet(kare_ciz, n_kare, yol, figsize=(7, 4), fps=2, dpi=70):
    """kare_ciz(fig, i) fonksiyonuyla her kareyi çizip animasyonlu GIF olarak kaydeder."""
    fig = plt.figure(figsize=figsize)
    anim = FuncAnimation(fig, lambda i: kare_ciz(fig, i), frames=n_kare)
    anim.save(yol, writer=PillowWriter(fps=fps), dpi=dpi)
    plt.close(fig)
    print(f"💾 Animasyon → {Path(yol).as_posix()} ({Path(yol).stat().st_size / 1024:.0f} KB)")


def metrikler(y, tahmin, skor=None, olasilik=None):
    """İkili sınıflandırma ölçütleri. Sıra: gerçek, tahmin (sınıf), isteğe bağlı skor (sıralama için) ve olasılık."""
    sonuc = {"Doğruluk": accuracy_score(y, tahmin), "Dengeli doğruluk": balanced_accuracy_score(y, tahmin),
             "F1 (premium)": f1_score(y, tahmin), "Kesinlik": precision_score(y, tahmin, zero_division=0),
             "Duyarlılık": recall_score(y, tahmin), "MCC": matthews_corrcoef(y, tahmin)}
    if skor is not None:
        sonuc["ROC-AUC"] = roc_auc_score(y, skor)
        sonuc["AP (PR-AUC)"] = average_precision_score(y, skor)
    if olasilik is not None:
        sonuc["Log-loss"] = log_loss(y, olasilik, labels=[0, 1])
        sonuc["Brier"] = brier_score_loss(y, olasilik)
    return sonuc


def skor_al(model, X):
    """Modelin 'premium' skoru: olasılık varsa o, yoksa karar fonksiyonu (sıralama ölçütleri için yeterli)."""
    if hasattr(model, "predict_proba"):
        return model.predict_proba(X)[:, 1]
    return model.decision_function(X)


print(f"Python {platform.python_version()} · scikit-learn {sklearn.__version__} · pandas {pd.__version__} · numpy {np.__version__}")
print(f"LightGBM {lightgbm.__version__} · XGBoost {xgboost.__version__} · CatBoost {catboost.__version__} · SciPy {scipy.__version__}")
```

### 🏘️ 2.1 · İş Problemi

Bir **gayrimenkul yatırım fonunun** analisti olduğunuzu düşünün. Şehirde yüzlerce mahalle var; hepsini tek tek gezmek mümkün değil. Elinizde her mahalle için **ucuza ölçülebilen** bilgiler duruyor: suç oranı, hava kirliliği, okul kalitesi, merkeze uzaklık, emlak vergisi… Soru şudur:

> **Yalnızca bu çevre ölçümlerine bakarak** bir mahallenin **premium** (konut medyan değeri **25 bin $ ve üstü**) olup olmadığını ne kadar güvenle söyleyebiliriz? Ve **sınırlı inceleme bütçesini** hangi mahallelere harcamalıyız?

İyi bir model:

- 🔍 İnceleme ekibini yalnızca **umut vadeden mahallelere** yönlendirir (az yanlış alarm),
- 💎 Gerçekten değerli (premium) mahalleleri **kaçırmamaya** çalışır,
- 🧭 Hangi çevre özelliklerinin değeri **neden** yükselttiğini açıklayabilir.

### 📚 2.2 · Veri Seti

**Boston konut verisi** (Harrison ve Rubinfeld, 1978): 1970 nüfus sayımına dayalı **506 mahalle (sayım bölgesi)**. Dosyada **başlık satırı yoktur**; sütun adlarını veri setinin özgün tanımından biz veriyoruz.

| Sütun | Anlamı |
|:--|:--|
| `CRIM` | Kişi başı **suç oranı** |
| `ZN` | 25 bin ft²'den büyük parsellere ayrılmış konut arazisi payı (%) |
| `INDUS` | Perakende dışı **sanayi** arazisi payı (%) |
| `CHAS` | Charles Nehri kıyısında mı? (1/0) |
| `NOX` | **Azot oksit** derişimi (hava kirliliği) |
| `RM` | Konut başına ortalama **oda sayısı** |
| `AGE` | 1940'tan önce yapılmış konut payı (%) |
| `DIS` | Beş iş merkezine ağırlıklı **uzaklık** |
| `RAD` | Çevre yoluna **erişim endeksi** (1–24) |
| `TAX` | Emlak vergisi oranı (10 bin $ başına) |
| `PTRATIO` | **Öğrenci/öğretmen** oranı |
| `B` | Irksal bileşenli bir endeks → **bu projede kullanılmaz** (bölüm 3) |
| `LSTAT` | **Düşük statülü** nüfus payı (%) |
| **`MEDV`** | Konut medyan değeri (bin $). **Hedefimiz bundan türetilir** |

Hedef: **`premium = (MEDV ≥ 25)`**. Yani bir regresyon verisini, bir **karar sorusuna** çeviriyoruz: *"Bu mahalle yatırım için premium mu?"*

```python
ham = veri_oku(VERI_YOLU)                                    # başlıksız, boşlukla ayrılmış dosya → sütun adları yardımcı dosyadan
display(ham.head())
print(f"Satır × sütun: {ham.shape}  ·  eksik değer: {int(ham.isna().sum().sum())}  ·  tekrar eden satır: {int(ham.duplicated().sum())}")
```

### 📏 2.3 · Başarı Ölçütleri

Sınıflar **dengesiz** (premium ≈ %26): "hiç premium deme" bile **%74 doğruluk** verir. Bu yüzden birkaç ölçütü birlikte izleriz:

| Ölçüt | Anlamı |
|:--|:--|
| **Doğruluk** | Doğru tahmin oranı (dengesiz veride tek başına yanıltıcı) |
| **Dengeli doğruluk** | İki sınıfın duyarlılık ortalaması: her sınıfa eşit ağırlık |
| **F1 (premium)** (*ana ölçütümüz*) | Kesinlik ve duyarlılığın uyumlu ortalaması; premium sınıfına odaklanır |
| **Kesinlik / Duyarlılık** | "Premium dediklerimizin kaçı gerçekten premium?" / "Gerçek premiumların kaçını bulduk?" |
| **MCC** | Dengesiz veride güvenilir, tek sayılık özet (−1…+1) |
| **ROC-AUC, AP** | Eşikten bağımsız sıralama kalitesi (AP = PR eğrisinin altındaki alan) |
| **Log-loss, Brier** | Olasılıkların kalitesi (finalistlerde) |

Referans noktası olarak **hep "standart" diyen** bir "taban model" kullanacağız: hiçbir model bundan kötü olmamalı.

```python
y_tum = (ham["MEDV"] >= ESIK_BIN_DOLAR).astype(int).to_numpy()       # hedef: 1 = premium mahalle
sayim = pd.Series(y_tum).map(dict(enumerate(SINIF_ADLARI))).value_counts()
print(sayim.to_string())
print(f"\nPremium oranı: %{y_tum.mean() * 100:.1f}  →  'hep standart de' tabanının doğruluğu: %{(1 - y_tum.mean()) * 100:.1f}")
```

### 🧬 3.1 · Dikkat Çeken Sütunlar

Tabloya bakınca birkaç şey gözümüze çarpıyor:

- **Başlık yok, eksik değer yok, tekrar yok.** Temiz görünüyor; ama temiz görünmek yeterli değil.
- **`ZN`** satırların büyük bölümünde **sıfır** (büyük parselli konut yok) → ağaçlar için sorun değil, doğrusal modellerde dikkat gerekir.
- **`CHAS`** yalnızca 0/1, **`RAD`** yalnızca **9 farklı değer** alıyor → bunlar aslında *kategori* gibi davranıyor. (Bölüm 10'da `RAD`'ı LightGBM'e kategorik olarak da vereceğiz.)
- **`MEDV` 50'de kırpılmış:** 16 mahalle "50 ve üstü" yerine tam 50 yazılmış. Hedefi ≥ 25 eşiğiyle kestiğimiz için bu kırpma sınıf etiketini **değiştirmez**; ama regresyon yapsaydık sorun olurdu.
- **Aynı ilçenin mahalleleri aynı değerleri paylaşıyor:** `INDUS`, `RAD`, `TAX`, `PTRATIO`, `ZN` gibi sütunlar **kasaba düzeyinde** ölçülmüş; dolayısıyla komşu mahalleler bu sütunlarda **birebir aynı** (aşağıdaki hücre).

```python
ham["ilce"] = ilce_kimligi(ham)                              # kasaba kimliği yok → ortak sütunlardan *vekil* grup üretiyoruz
grup_boy = ham["ilce"].value_counts()
print(f"Vekil ilçe sayısı: {ham['ilce'].nunique()}  ·  en kalabalık 5 grup: {grup_boy.head(5).tolist()}  ·  tek mahalleli grup: {(grup_boy == 1).sum()}")
dev = ham[ham["ilce"] == grup_boy.index[0]]
print(f"\nEn büyük grup: {len(dev)} mahalle (verinin %{len(dev) / len(ham) * 100:.0f}'i) → RAD={dev['RAD'].unique()}, TAX={dev['TAX'].unique()}, INDUS={dev['INDUS'].unique()}")
print(f"Bu grupta premium oranı: %{(dev['MEDV'] >= ESIK_BIN_DOLAR).mean() * 100:.1f}   ·   diğerlerinde: %{(ham.loc[ham['ilce'] != grup_boy.index[0], 'MEDV'] >= ESIK_BIN_DOLAR).mean() * 100:.1f}")
```

### 🧪 5.1 · Deney: Hedefin Kendisini Özellik Yaparsak?

`MEDV` sütununu özellik olarak bırakırsak ne olur? Bu **apaçık** bir hata gibi görünür, ama gerçek projelerde "hedefin bir kopyası" çoğu zaman *farklı bir adla* veriye sızar (örn. "fiyat/m²" ve "toplam fiyat" birlikte verilir). Önce rastgele katlı çapraz doğrulamada, sonra grup bazlı doğrulamada bakalım.

```python
def fold_degerlendir(model, X, y, egitim_idx, dogrulama_idx):
    """Tek katı eğitir ve doğrulama metriklerini döndürür (paralel çalıştırmak için ayrı işlev)."""
    m = clone(model).fit(X.iloc[egitim_idx], y[egitim_idx])
    tahmin = m.predict(X.iloc[dogrulama_idx])
    sk = skor_al(m, X.iloc[dogrulama_idx])
    olas = m.predict_proba(X.iloc[dogrulama_idx])[:, 1] if hasattr(m, "predict_proba") else None
    return metrikler(y[dogrulama_idx], tahmin, sk, olas)


def cv_ozet(model, X, y, katlar, n_jobs=N_ISCI):
    """Verilen katlarda ortalama metrikleri hesaplar (katlar: (eğitim_idx, doğrulama_idx) listesi)."""
    sonuc = Parallel(n_jobs=n_jobs)(delayed(fold_degerlendir)(model, X, y, e, d) for e, d in katlar)
    return pd.DataFrame(sonuc)


KATLAR_RASTGELE = list(RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=RANDOM_STATE).split(X_egitim, y_egitim))
lgbm_basit = LGBMClassifier(n_jobs=1, random_state=RANDOM_STATE, verbosity=-1)

X_sizinti = X_egitim.copy()
X_sizinti["MEDV"] = ham.loc[idx_egitim, "MEDV"].to_numpy()        # hedefin kendisi!
satir = []
for ad, X_ in [("Doğru: MEDV yok", X_egitim), ("Sızıntı: MEDV var", X_sizinti)]:
    r = cv_ozet(lgbm_basit, X_, y_egitim, KATLAR_RASTGELE)
    satir.append({"deney": ad, "F1": r["F1 (premium)"].mean(), "ROC-AUC": r["ROC-AUC"].mean(), "Doğruluk": r["Doğruluk"].mean()})
sizinti_hedef = pd.DataFrame(satir).set_index("deney")
display(sizinti_hedef.round(4))
```

Bu skor **gerçek bir başarı değil**; üretimde `MEDV` bilinmeyecek. Kural: *hedefi üreten ya da hedefle aynı ölçümden türeyen her sütun, tahmin anında elde yoksa özellik olamaz.*

### 🧪 5.2 · Deney: Komşu Mahalleler Skoru Şişiriyor mu?

Rastgele katlamada aynı ilçenin mahalleleri **hem eğitime hem doğrulamaya** düşer. Model, ilçeye özgü sütunları (`TAX`, `INDUS`, `PTRATIO`…) bir **ilçe parmak izi** olarak ezberleyip komşunun etiketini "tahmin eder". Gerçek kullanımda ise **hiç görmediğimiz bir ilçeyi** değerlendirmek isteriz. Bunu sınamak için doğrulamayı **ilçe gruplarına göre** kuruyoruz: bir ilçe ya tamamen eğitimde ya tamamen doğrulamadadır (`StratifiedGroupKFold`).

```python
KATLAR_GRUP = []
for tohum in range(3):                                              # 3 tekrar × 5 kat = 15 grup bazlı kat
    KATLAR_GRUP += list(StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=tohum).split(X_egitim, y_egitim, g_egitim))

kisa_modeller = {
    "Lojistik regresyon": Pipeline([("olcek", StandardScaler()), ("model", LogisticRegression(max_iter=2000))]),
    "Rastgele orman": RandomForestClassifier(n_estimators=300, n_jobs=1, random_state=RANDOM_STATE),
    "LightGBM": lgbm_basit,
    "XGBoost": XGBClassifier(n_estimators=200, n_jobs=1, random_state=RANDOM_STATE, verbosity=0),
}
satir = []
for ad, m in kisa_modeller.items():
    r1 = cv_ozet(m, X_egitim, y_egitim, KATLAR_RASTGELE)
    r2 = cv_ozet(m, X_egitim, y_egitim, KATLAR_GRUP)
    satir.append({"model": ad, "F1 · rastgele": r1["F1 (premium)"].mean(), "F1 · ilçe bazlı": r2["F1 (premium)"].mean(),
                  "AUC · rastgele": r1["ROC-AUC"].mean(), "AUC · ilçe bazlı": r2["ROC-AUC"].mean()})
komsu = pd.DataFrame(satir).set_index("model")
komsu["F1 kaybı"] = komsu["F1 · rastgele"] - komsu["F1 · ilçe bazlı"]
komsu["AUC kaybı"] = komsu["AUC · rastgele"] - komsu["AUC · ilçe bazlı"]
display(komsu.round(3))
rapor_kaydet(komsu, "ilce_sizintisi_deneyi")

fig, eksen = plt.subplots(1, 3, figsize=(17, 4.4))
sizinti_hedef["F1"].plot.bar(ax=eksen[0], color=[RENK["yesil"], RENK["pembe"]], rot=0)
eksen[0].set(title="Hedef sızıntısı", ylabel="F1 (premium)", ylim=(0, 1.08))
for i, v in enumerate(sizinti_hedef["F1"]):
    eksen[0].text(i, v + 0.015, f"{v:.3f}", ha="center", fontweight="bold")
komsu[["F1 · rastgele", "F1 · ilçe bazlı"]].plot.bar(ax=eksen[1], color=[RENK["mavi"], RENK["pembe"]], rot=15)
eksen[1].set(title="Komşu mahalle sızıntısı: F1", ylabel="F1 (premium)", ylim=(0, 1.15))
eksen[1].legend(loc="upper center", ncol=2, fontsize=8)
komsu[["AUC · rastgele", "AUC · ilçe bazlı"]].plot.bar(ax=eksen[2], color=[RENK["mavi"], RENK["pembe"]], rot=15)
eksen[2].set(title="Komşu mahalle sızıntısı: ROC-AUC", ylabel="ROC-AUC", ylim=(0.5, 1.05))
eksen[2].legend(loc="upper center", ncol=2, fontsize=8)
plt.tight_layout()
kaydet(fig, "05_sizinti_deneyleri")
plt.show()
```

Her modeli **aynı kalıba** dökeriz: `Pipeline = özellik üretimi → ölçekleme → model`. Böylece dönüşümler **her katın kendi eğitim kısmında** öğrenilir ve doğrulama kısmına uygulanır: **sızıntı olmaz**. Burada özellik üretimi satır bazlı olduğu için eğitimden bir şey öğrenmez; ölçekleyici ise öğrenir.

```python
def pipe_kur(model, olcekle=False, mod="ham", sutunlar=None):
    """Özellik üretimi → (isteğe bağlı) ölçekleme → model. Ölçekleyici her katın eğitim kısmında öğrenilir."""
    return Pipeline([("ozellik", BostonOzellikleri(mod=mod)), ("olcek", StandardScaler() if olcekle else "passthrough"), ("model", model)])


def yeni_lgbm(**kw):
    """Bu projedeki LightGBM varsayılanı: tek iş parçacığı (paralellik dış döngüde), sessiz."""
    return LGBMClassifier(**{"n_jobs": 1, "random_state": RANDOM_STATE, "verbosity": -1, **kw})


print(pipe_kur(yeni_lgbm(), olcekle=True, mod="oran"))
```

```python
def deney(ad, model, olcekle, mod="ham", X=None):
    X = X_egitim if X is None else X
    r = cv_ozet(pipe_kur(model, olcekle, mod), X, y_egitim, KATLAR_RASTGELE)
    return {"deney": ad, "F1": r["F1 (premium)"].mean(), "Duyarlılık": r["Duyarlılık"].mean(), "Kesinlik": r["Kesinlik"].mean(),
            "ROC-AUC": r["ROC-AUC"].mean(), "AP": r["AP (PR-AUC)"].mean()}


X_egitim_B = ham.iloc[idx_egitim][OZELLIK_A + ["B"]].reset_index(drop=True)          # yalnızca B'nin bedelini ölçmek için
satir = []
for model_adi, mk, olcek in [("LightGBM", lambda **kw: yeni_lgbm(**kw), False), ("Lojistik regresyon", lambda **kw: LogisticRegression(max_iter=3000, **kw), True),
                             ("SVM (RBF)", lambda **kw: SVC(**kw), True), ("kNN (k=15)", lambda **kw: KNeighborsClassifier(n_neighbors=15), True)]:
    satir.append({**deney(f"{model_adi} · ham", mk(), olcek), "model": model_adi, "grup": "1 · özellikler"})
    satir.append({**deney(f"{model_adi} · +oranlar", mk(), olcek, mod="oran"), "model": model_adi, "grup": "1 · özellikler"})
    satir.append({**deney(f"{model_adi} · ölçeksiz", mk(), False), "model": model_adi, "grup": "2 · ölçek"})
    satir.append({**deney(f"{model_adi} · ölçekli", mk(), True), "model": model_adi, "grup": "2 · ölçek"})
satir.append({**deney("LightGBM · ağırlıksız", yeni_lgbm(), False), "model": "LightGBM", "grup": "3 · sınıf ağırlığı"})
satir.append({**deney("LightGBM · balanced", yeni_lgbm(class_weight="balanced"), False), "model": "LightGBM", "grup": "3 · sınıf ağırlığı"})
satir.append({**deney("Lojistik · ağırlıksız", LogisticRegression(max_iter=3000), True), "model": "Lojistik regresyon", "grup": "3 · sınıf ağırlığı"})
satir.append({**deney("Lojistik · balanced", LogisticRegression(max_iter=3000, class_weight="balanced"), True), "model": "Lojistik regresyon", "grup": "3 · sınıf ağırlığı"})
satir.append({**deney("LightGBM · B yok", yeni_lgbm(), False), "model": "LightGBM", "grup": "4 · B sütunu"})
satir.append({**deney("LightGBM · B var", yeni_lgbm(), False, X=X_egitim_B), "model": "LightGBM", "grup": "4 · B sütunu"})
hazirlik = pd.DataFrame(satir).set_index("deney")
display(hazirlik[["grup", "F1", "Duyarlılık", "Kesinlik", "ROC-AUC", "AP"]].round(4))
rapor_kaydet(hazirlik, "deney_ozellik_olcek_agirlik")
```

### 🏁 7.1 · Yarışçılar

**29 algoritma + 1 taban** sekiz aileden: taban, doğrusal, olasılıksal, komşuluk ve çekirdek, sinir ağı, ağaç ve orman, boosting, topluluk. Ölçek isteyen modeller (`olcekle=True`) `StandardScaler` ile birlikte pipeline'a girer; ağaç tabanlı modeller ham değerle çalışır. Hepsi **aynı 15 katta** (5 kat × 3 tekrar) yarışır; böylece sıralama adil olur.

```python
def model_listesi():
    """(ad, aile, model, ölçekle) listesi. Paralellik dış döngüde olduğu için her model tek iş parçacığıyla çalışır."""
    rs = RANDOM_STATE
    return [
        ("Taban (hep standart)", "Taban", DummyClassifier(strategy="most_frequent"), False),
        # ── Doğrusal
        ("Lojistik regresyon", "Doğrusal", LogisticRegression(max_iter=3000), True),
        ("Ridge sınıflandırıcı", "Doğrusal", RidgeClassifier(random_state=rs), True),
        ("SGD (log-loss)", "Doğrusal", SGDClassifier(loss="log_loss", random_state=rs), True),
        ("Perceptron", "Doğrusal", Perceptron(random_state=rs), True),
        ("Doğrusal SVM", "Doğrusal", LinearSVC(max_iter=20000, random_state=rs), True),
        ("LDA", "Doğrusal", LinearDiscriminantAnalysis(), True),
        # ── Olasılıksal
        ("Gauss Naive Bayes", "Olasılıksal", GaussianNB(), False),
        ("QDA", "Olasılıksal", QuadraticDiscriminantAnalysis(reg_param=0.1), True),
        ("Gauss süreci", "Olasılıksal", GaussianProcessClassifier(random_state=rs), True),
        # ── Komşuluk ve çekirdek
        ("kNN (k=15)", "Komşuluk ve çekirdek", KNeighborsClassifier(n_neighbors=15), True),
        ("SVM (RBF)", "Komşuluk ve çekirdek", SVC(probability=True, random_state=rs), True),
        ("SVM (polinom)", "Komşuluk ve çekirdek", SVC(kernel="poly", degree=3, probability=True, random_state=rs), True),
        # ── Sinir ağı
        ("Sinir ağı (MLP)", "Sinir ağı", MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=2000, random_state=rs), True),
        # ── Ağaç ve orman
        ("Karar ağacı", "Ağaç ve orman", DecisionTreeClassifier(max_depth=5, random_state=rs), False),
        ("Ekstra ağaç (tek)", "Ağaç ve orman", ExtraTreeClassifier(max_depth=5, random_state=rs), False),
        ("Rastgele orman", "Ağaç ve orman", RandomForestClassifier(n_estimators=300, n_jobs=1, random_state=rs), False),
        ("Ekstra ağaçlar", "Ağaç ve orman", ExtraTreesClassifier(n_estimators=300, n_jobs=1, random_state=rs), False),
        ("Bagging (ağaç)", "Ağaç ve orman", BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=100, random_state=rs), False),
        # ── Boosting
        ("AdaBoost", "Boosting", AdaBoostClassifier(n_estimators=200, random_state=rs), False),
        ("Gradient Boosting", "Boosting", GradientBoostingClassifier(random_state=rs), False),
        ("HistGradientBoosting", "Boosting", HistGradientBoostingClassifier(random_state=rs), False),
        ("XGBoost", "Boosting", XGBClassifier(n_estimators=200, n_jobs=1, random_state=rs, verbosity=0), False),
        ("XGBoost (rastgele orman)", "Boosting", XGBRFClassifier(n_estimators=200, n_jobs=1, random_state=rs, verbosity=0), False),
        ("LightGBM", "Boosting", yeni_lgbm(), False),
        ("LightGBM · DART", "Boosting", yeni_lgbm(boosting_type="dart"), False),
        ("LightGBM · GOSS", "Boosting", yeni_lgbm(data_sample_strategy="goss"), False),
        ("CatBoost", "Boosting", CatBoostClassifier(iterations=300, thread_count=1, random_seed=rs, verbose=0, allow_writing_files=False), False),
        # ── Topluluk
        ("Oylama (RF + LGBM + SVM)", "Topluluk", VotingClassifier([("rf", RandomForestClassifier(n_estimators=200, random_state=rs)), ("lgb", yeni_lgbm()),
                                                                  ("svm", Pipeline([("o", StandardScaler()), ("s", SVC(probability=True, random_state=rs))]))], voting="soft"), False),
        ("Yığınlama (RF + LGBM + kNN → LR)", "Topluluk", StackingClassifier([("rf", RandomForestClassifier(n_estimators=200, random_state=rs)), ("lgb", yeni_lgbm()),
                                                                          ("knn", Pipeline([("o", StandardScaler()), ("k", KNeighborsClassifier(15))]))],
                                                                         final_estimator=LogisticRegression(max_iter=2000), cv=5), False),
    ]


MODELLER = model_listesi()
aile_say = pd.Series([a for _, a, _, _ in MODELLER]).value_counts()
print(f"Toplam {len(MODELLER)} yarışçı = 1 taban + {len(MODELLER) - 1} algoritma")
print(aile_say.to_string())
```

### 🏎️ 7.2 · Yarışı Koşalım (Senaryo A: Tüm Çevre Ölçümleri)

```python
def yaris_kos(ad, aile, model, olcekle, sutunlar, X, y, katlar, oof_tut=True):
    """Bir modeli tüm katlarda eğitir: kat metrikleri, süre ve (ilk tekrar için) out-of-fold tahminleri döndürür."""
    kayit, oof_tahmin, oof_skor = [], np.full(len(y), -1), np.full(len(y), np.nan)
    for k, (e, d) in enumerate(katlar):
        t0 = time.perf_counter()
        m = pipe_kur(clone(model), olcekle).fit(X.iloc[e][sutunlar], y[e])
        sure = time.perf_counter() - t0
        tahmin, sk = m.predict(X.iloc[d][sutunlar]), skor_al(m, X.iloc[d][sutunlar])
        olas = m.predict_proba(X.iloc[d][sutunlar])[:, 1] if hasattr(m, "predict_proba") else None
        kayit.append({**metrikler(y[d], tahmin, sk, olas), "süre": sure})
        if oof_tut and k < 5:                                         # ilk tekrarın 5 katı → her satır bir kez tahmin edilir
            oof_tahmin[d], oof_skor[d] = tahmin, sk
    return ad, aile, pd.DataFrame(kayit), oof_tahmin, oof_skor


def yaris(sutunlar, X=None, y=None, katlar=None, modeller=None):
    X = X_egitim if X is None else X
    y = y_egitim if y is None else y
    katlar = KATLAR_RASTGELE if katlar is None else katlar
    modeller = MODELLER if modeller is None else modeller
    sonuc = Parallel(n_jobs=N_ISCI)(delayed(yaris_kos)(ad, aile, m, o, sutunlar, X, y, katlar) for ad, aile, m, o in modeller)
    satir, oof = [], {}
    for ad, aile, df, ot, os_ in sonuc:
        ort, sd = df.mean(), df.std()
        satir.append({"model": ad, "aile": aile, **{f"{k}": ort[k] for k in df.columns if k != "süre"}, "F1 sd": sd["F1 (premium)"], "süre (sn/kat)": ort["süre"]})
        oof[ad] = (ot, os_)
    tablo = pd.DataFrame(satir).set_index("model").sort_values("F1 (premium)", ascending=False)
    return tablo, oof


t0 = time.perf_counter()
tablo_A, oof_A = yaris(OZELLIK_A)
print(f"Yarış süresi: {time.perf_counter() - t0:.0f} sn")
cols_goster = ["aile", "F1 (premium)", "F1 sd", "Duyarlılık", "Kesinlik", "ROC-AUC", "AP (PR-AUC)", "MCC", "Log-loss", "süre (sn/kat)"]
display(tablo_A[cols_goster].round(3))
rapor_kaydet(tablo_A, "yaris_senaryo_A_cv")
```

```python
def test_kos(ad, aile, model, olcekle, sutunlar, X_e, y_e, X_t, y_t):
    """Modeli tüm eğitim setiyle eğitir, kilitli test setinde ölçer."""
    t0 = time.perf_counter()
    m = pipe_kur(clone(model), olcekle).fit(X_e[sutunlar], y_e)
    sure = time.perf_counter() - t0
    tahmin, sk = m.predict(X_t[sutunlar]), skor_al(m, X_t[sutunlar])
    olas = m.predict_proba(X_t[sutunlar])[:, 1] if hasattr(m, "predict_proba") else None
    return {"model": ad, "aile": aile, **metrikler(y_t, tahmin, sk, olas), "süre (sn)": sure}, tahmin, sk


def test_degerlendir(sutunlar, tablo, X_e=None, y_e=None, X_t=None, y_t=None):
    X_e = X_egitim if X_e is None else X_e
    y_e = y_egitim if y_e is None else y_e
    X_t = X_test if X_t is None else X_t
    y_t = y_test if y_t is None else y_t
    sonuc = Parallel(n_jobs=N_ISCI)(delayed(test_kos)(ad, aile, m, o, sutunlar, X_e, y_e, X_t, y_t) for ad, aile, m, o in MODELLER)
    test = pd.DataFrame([s[0] for s in sonuc]).set_index("model")
    tahminler = {s[0]["model"]: (s[1], s[2]) for s in sonuc}
    test["CV F1"] = tablo["F1 (premium)"]
    return test.loc[tablo.index], tahminler


test_A, tahmin_test_A = test_degerlendir(OZELLIK_A, tablo_A)
display(test_A[["aile", "CV F1", "F1 (premium)", "ROC-AUC", "AP (PR-AUC)", "MCC", "Duyarlılık", "Kesinlik"]].round(3).rename(columns={"F1 (premium)": "Test F1"}))
rapor_kaydet(test_A, "yaris_senaryo_A_test")
```

```python
test_g = test_A.drop(index="Taban (hep standart)")                      # taban (F1 = 0) grafiği ve korelasyonu bozmasın
rho = test_g[["CV F1", "F1 (premium)"]].corr(method="spearman").iloc[0, 1]
fig, eksen = plt.subplots(1, 2, figsize=(15, 5.6))
for ad, r in test_g.iterrows():
    eksen[0].scatter(r["CV F1"], r["F1 (premium)"], s=70, color=AILE_RENK[r["aile"]], edgecolor="white", zorder=3)
for ad in ["CatBoost", "XGBoost", "AdaBoost", "Rastgele orman", "LightGBM", "Lojistik regresyon"]:
    r = test_A.loc[ad]
    eksen[0].annotate(ad, (r["CV F1"], r["F1 (premium)"]), fontsize=7.5, xytext=(4, 4), textcoords="offset points")
lim = [0.5, 1]
eksen[0].plot(lim, lim, ls="--", color=RENK["gri"])
eksen[0].set(title=f"CV ↔ Test (Spearman ρ = {rho:.2f})", xlabel="CV F1 (eğitim içi)", ylabel="Test F1 (kilitli, 102 mahalle)", xlim=(0.62, 0.9), ylim=(0.62, 0.95))
eksen[0].legend(handles=[Patch(color=c, label=a) for a, c in AILE_RENK.items()], fontsize=7, loc="lower right")
# test süresi ↔ CV F1 (hız)
for ad, r in test_g.iterrows():
    eksen[1].scatter(r["süre (sn)"], r["F1 (premium)"], s=70, color=AILE_RENK[r["aile"]], edgecolor="white", zorder=3)
for ad in ["LightGBM", "XGBoost", "CatBoost", "Rastgele orman", "Gauss süreci", "Yığınlama (RF + LGBM + kNN → LR)", "Sinir ağı (MLP)"]:
    r = test_A.loc[ad]
    eksen[1].annotate(ad, (r["süre (sn)"], r["F1 (premium)"]), fontsize=7.5, xytext=(4, 4), textcoords="offset points")
eksen[1].set(xscale="log", title="Eğitim süresi ↔ Test F1", xlabel="Eğitim süresi (sn, log)", ylabel="Test F1")
plt.tight_layout()
kaydet(fig, "08_cv_test_ve_hiz")
plt.show()
```

### 🎲 8.2 · Fark Gerçek mi, Şans mı? (Bootstrap)

Eğitim kümesindeki **out-of-fold** tahminleri (her mahalle bir kez, kendisi eğitimde yokken tahmin edildi) **yeniden örnekleyerek** (2000 kez) iki model arasındaki F1 farkının güven aralığını hesaplıyoruz. Aralık **0'ı kesiyorsa** fark şanstan ayırt edilemez.

```python
def f1_hizli(y, p):
    """Bootstrap için hızlı F1 (premium)."""
    tp = np.sum((y == 1) & (p == 1)); fp = np.sum((y == 0) & (p == 1)); fn = np.sum((y == 1) & (p == 0))
    return 2 * tp / max(2 * tp + fp + fn, 1)


def bootstrap_fark(oof_a, oof_b, y, B=2000, seed=RANDOM_STATE):
    rng = np.random.default_rng(seed)
    n = len(y); fark = np.empty(B)
    for b in range(B):
        i = rng.integers(0, n, n)
        fark[b] = f1_hizli(y[i], oof_a[i]) - f1_hizli(y[i], oof_b[i])
    return fark.mean(), np.percentile(fark, 2.5), np.percentile(fark, 97.5)


lider = tablo_A.index[0]
satir = []
for rakip in tablo_A.index[1:]:
    ort, alt, ust = bootstrap_fark(oof_A[lider][0], oof_A[rakip][0], y_egitim)
    satir.append({"rakip": rakip, "F1 farkı (lider − rakip)": ort, "alt %2,5": alt, "üst %97,5": ust, "fark gerçek mi?": "evet" if alt > 0 else "ayırt edilemez"})
boot_A = pd.DataFrame(satir).set_index("rakip")
display(boot_A.round(3).head(14))
rapor_kaydet(boot_A, "bootstrap_senaryo_A")
print(f"Lider: {lider}  ·  Liderden anlamlı geride olan: {(boot_A['fark gerçek mi?'] == 'evet').sum()} / {len(boot_A)}  ·  ayırt edilemeyen: {(boot_A['fark gerçek mi?'] != 'evet').sum()}")

ilk = boot_A.head(14).iloc[::-1]
fig, ax = plt.subplots(figsize=(9, 5.4))
ax.errorbar(ilk["F1 farkı (lider − rakip)"], ilk.index, xerr=[ilk["F1 farkı (lider − rakip)"] - ilk["alt %2,5"], ilk["üst %97,5"] - ilk["F1 farkı (lider − rakip)"]],
            fmt="o", color=RENK["mor"], ecolor=RENK["gri"], capsize=3)
ax.axvline(0, color=RENK["pembe"], ls="--")
ax.set(title=f"Lider ({lider}) ile fark, %95 aralık", xlabel="F1 farkı (0 = şans)")
plt.tight_layout()
kaydet(fig, "09_bootstrap_fark")
plt.show()
```

LightGBM'in kalbi **yaprak yaprak büyüyen** (leaf-wise) ağaçlardır; bu hızlı ve güçlüdür ama **küçük veride kolay ezberler**. Burada yalnızca **404 eğitim satırımız** var. Bu yüzden düğmelerin etkisi bu bölümde özellikle öğreticidir. Her deney **15 katlı CV** ile ölçülür; yalnızca bir düğme (ya da iki) değişir.

```python
def lgbm_cv(X=None, **parametre):
    """Verilen LightGBM ayarını 15 katlı CV'de ölçer: ortalama F1, log-loss, AUC ve eğitim F1'i (ezber göstergesi)."""
    X = X_egitim if X is None else X
    r = cv_ozet(yeni_lgbm(**parametre), X, y_egitim, KATLAR_RASTGELE)
    return {"F1": r["F1 (premium)"].mean(), "Log-loss": r["Log-loss"].mean(), "ROC-AUC": r["ROC-AUC"].mean(), "AP": r["AP (PR-AUC)"].mean()}


def egri_cv(n=600, **parametre):
    """İlk 5 katın doğrulama ve eğitim log-loss eğrilerini (tur tur) ortalar."""
    def bir(e, d):
        m = yeni_lgbm(n_estimators=n, **parametre).fit(X_egitim.iloc[e], y_egitim[e], eval_X=(X_egitim.iloc[e], X_egitim.iloc[d]),
                                                       eval_y=(y_egitim[e], y_egitim[d]), eval_metric=["binary_logloss", "auc"])
        h = m.evals_result_                                           # valid_0 = eğitim kısmı, valid_1 = doğrulama kısmı
        return (np.array(h["valid_0"]["binary_logloss"]), np.array(h["valid_1"]["binary_logloss"]), np.array(h["valid_1"]["auc"]))
    sonuc = Parallel(n_jobs=N_ISCI)(delayed(bir)(e, d) for e, d in KATLAR_RASTGELE[:5])
    return [np.mean([s[i] for s in sonuc], axis=0) for i in range(3)]


print("Varsayılan LightGBM:", {k: round(v, 3) for k, v in lgbm_cv().items()})
print("Varsayılan değerler → num_leaves =", yeni_lgbm().get_params()["num_leaves"], "· min_child_samples =", yeni_lgbm().get_params()["min_child_samples"],
      "· learning_rate =", yeni_lgbm().get_params()["learning_rate"], "· n_estimators =", yeni_lgbm().get_params()["n_estimators"])
```

### 🍃 10.3 · LightGBM'in Özel Düğmesi: `num_leaves` × `max_depth`

Çoğu kütüphane ağacı **seviye seviye** büyütür; LightGBM ise **en çok hata azaltacak yaprağı** seçip onu böler (yaprak yaprak büyüme). Bu yüzden ağacın karmaşıklığını **`num_leaves`** (yaprak sayısı) belirler; `max_depth` ise ek bir emniyet kemeridir. Kural: `num_leaves` ≤ 2^`max_depth` olmalı, aksi hâlde derinlik sınırı devreye girer.

```python
yaprak_listesi, derinlik_listesi = [3, 4, 7, 15, 31, 63], [2, 3, 5, -1]
satir = []
for nl in yaprak_listesi:
    for md in derinlik_listesi:
        if md != -1 and nl > 2 ** md:                                  # derinlik sınırı yaprak sayısından önce devreye girer → atla
            continue
        satir.append({"num_leaves": nl, "max_depth": md, **lgbm_cv(num_leaves=nl, max_depth=md, n_estimators=200, learning_rate=0.05)})
yaprak = pd.DataFrame(satir)
rapor_kaydet(yaprak, "lgbm_yaprak_derinlik", index=False)

fig, eksen = plt.subplots(1, 2, figsize=(14, 4.6))
for ax, metrik, cmap in zip(eksen, ["F1", "Log-loss"], ["YlGn", "YlOrRd_r"]):
    p = yaprak.pivot(index="num_leaves", columns="max_depth", values=metrik)
    sns.heatmap(p, annot=True, fmt=".3f", cmap=cmap, ax=ax, cbar=False, linewidths=0.5)
    ax.set(title=f"{metrik} · yaprak × derinlik (boş = anlamsız birleşim)")
plt.tight_layout()
kaydet(fig, "15_lgbm_yaprak_derinlik")
plt.show()

en_f1 = yaprak.sort_values("F1", ascending=False).iloc[0]
print(f"En yüksek F1: num_leaves={int(en_f1['num_leaves'])}, max_depth={int(en_f1['max_depth'])} → {en_f1['F1']:.3f}  ·  varsayılan (31 yaprak, derinlik −1): {yaprak[(yaprak.num_leaves == 31) & (yaprak.max_depth == -1)]['F1'].iloc[0]:.3f}")
```

### 🛡️ 10.6 · Düzenleme Düğmeleri: λ, α ve `min_split_gain`

- **`reg_lambda` (λ, L2):** yaprak değerlerini küçültür → ağaçlar "temkinli" olur.
- **`reg_alpha` (α, L1):** küçük katkıları sıfıra iter.
- **`min_split_gain` (γ):** bir bölünmenin **en az ne kadar kazanç** getirmesi gerektiğini söyler; zayıf bölünmeler reddedilir.

```python
satir = []
for lam in [0, 1, 5, 20]:
    for alfa in [0, 0.1, 1, 5]:
        satir.append({"reg_lambda": lam, "reg_alpha": alfa, **lgbm_cv(reg_lambda=lam, reg_alpha=alfa, n_estimators=200, learning_rate=0.05)})
duz = pd.DataFrame(satir)
rapor_kaydet(duz, "lgbm_duzenleme_lambda_alpha", index=False)
satir = [{"min_split_gain": g, **lgbm_cv(min_split_gain=g, n_estimators=200, learning_rate=0.05)} for g in [0, 0.01, 0.05, 0.1, 0.3, 1.0]]
gam = pd.DataFrame(satir).set_index("min_split_gain")
rapor_kaydet(gam, "lgbm_min_split_gain")

fig, eksen = plt.subplots(1, 3, figsize=(17, 4.4))
sns.heatmap(duz.pivot(index="reg_lambda", columns="reg_alpha", values="F1"), annot=True, fmt=".3f", cmap="YlGn", ax=eksen[0], cbar=False, linewidths=0.5); eksen[0].set(title="F1 · λ × α")
sns.heatmap(duz.pivot(index="reg_lambda", columns="reg_alpha", values="Log-loss"), annot=True, fmt=".3f", cmap="YlOrRd_r", ax=eksen[1], cbar=False, linewidths=0.5); eksen[1].set(title="Log-loss · λ × α")
eksen[2].plot(gam.index.astype(str), gam["F1"], "o-", color=RENK["mor"], lw=2, label="F1"); eksen[2].set(title="min_split_gain (γ)", xlabel="γ", ylabel="F1")
ax2 = eksen[2].twinx(); ax2.plot(gam.index.astype(str), gam["Log-loss"], "s--", color=RENK["pembe"], label="log-loss"); ax2.set_ylabel("log-loss"); ax2.grid(False)
plt.tight_layout()
kaydet(fig, "18_lgbm_duzenleme")
plt.show()
```

Tek tek düğme çevirmek **fikir verir**; ama düğmeler birbirini etkiler. **Rastgele arama** (Bergstra ve Bengio, 2012), geniş bir uzayda rastgele ayarlar deneyip en iyisini bulur. Dürüst olmak için üç kural koyuyoruz: (1) arama **yalnızca eğitim kümesinde** yapılır, (2) amaç **log-loss**'tur (olasılık kalitesi; F1 eşik bağımlı ve kaba), (3) bulunan ayar, aramada **hiç kullanılmamış taze katlarda** yeniden sınanır; çünkü aramanın kendi skoru **iyimser olabilir** (seçilimin yanlılığı; bu veride ne kadar olduğunu göreceğiz).

```python
ARAMA_KATLARI = RepeatedStratifiedKFold(n_splits=5, n_repeats=2, random_state=11)          # 10 kat; bölüm 7'deki katlardan farklı tohum


def ara(ad, model, uzay, n_iter, tohum=RANDOM_STATE):
    """RandomizedSearchCV: log-loss'u en küçük yapan ayarı arar; sonuçları tablo olarak döndürür."""
    t0 = time.perf_counter()
    rs = RandomizedSearchCV(model, uzay, n_iter=n_iter, cv=ARAMA_KATLARI, scoring={"logloss": "neg_log_loss", "f1": "f1", "auc": "roc_auc"},
                            refit="logloss", n_jobs=N_ISCI, random_state=tohum)
    rs.fit(X_egitim, y_egitim)
    sure = time.perf_counter() - t0
    sonuc = pd.DataFrame(rs.cv_results_)
    tablo = pd.DataFrame({"log-loss": -sonuc["mean_test_logloss"], "F1": sonuc["mean_test_f1"], "AUC": sonuc["mean_test_auc"], "params": sonuc["params"]}).sort_values("log-loss")
    print(f"{ad}: {n_iter} aday × 10 kat = {n_iter * 10} eğitim · {sure:.0f} sn · en iyi log-loss {tablo['log-loss'].iloc[0]:.4f}")
    return rs, tablo


lgbm_uzay = {"n_estimators": randint(80, 700), "learning_rate": loguniform(0.01, 0.2), "num_leaves": randint(3, 40),
             "max_depth": [-1, 3, 4, 5, 6, 8], "min_child_samples": randint(3, 40), "subsample": uniform(0.5, 0.5), "subsample_freq": [1],
             "colsample_bytree": uniform(0.5, 0.5), "reg_alpha": loguniform(1e-3, 10), "reg_lambda": loguniform(1e-3, 30), "min_split_gain": uniform(0, 0.3)}
rs_lgbm, lgbm_arama = ara("LightGBM", yeni_lgbm(), lgbm_uzay, n_iter=120)
rapor_kaydet(lgbm_arama.drop(columns="params").assign(params=lgbm_arama["params"].astype(str)), "lgbm_arama_sonuclari")
print("\nEn iyi LightGBM ayarı:")
for k, v in rs_lgbm.best_params_.items():
    print(f"  {k:<18} = {v:.4g}" if isinstance(v, float) else f"  {k:<18} = {v}")
```

```python
xgb_uzay = {"n_estimators": randint(80, 600), "learning_rate": loguniform(0.01, 0.3), "max_depth": randint(2, 8), "min_child_weight": randint(1, 8),
            "subsample": uniform(0.5, 0.5), "colsample_bytree": uniform(0.5, 0.5), "reg_lambda": loguniform(1e-2, 30), "gamma": uniform(0, 0.5)}
rs_xgb, xgb_arama = ara("XGBoost", XGBClassifier(n_jobs=1, random_state=RANDOM_STATE, verbosity=0), xgb_uzay, n_iter=60)
cat_uzay = {"iterations": randint(150, 600), "learning_rate": loguniform(0.02, 0.3), "depth": randint(3, 8), "l2_leaf_reg": loguniform(1, 20)}
rs_cat, cat_arama = ara("CatBoost", CatBoostClassifier(thread_count=1, random_seed=RANDOM_STATE, verbose=0, allow_writing_files=False), cat_uzay, n_iter=30)
rapor_kaydet(xgb_arama.assign(params=xgb_arama["params"].astype(str)), "xgb_arama_sonuclari")
rapor_kaydet(cat_arama.assign(params=cat_arama["params"].astype(str)), "catboost_arama_sonuclari")
```

### 🔍 11.2 · Taze Katlarda Doğrulama ve Final Adayları

Arama bittiğinde "en iyi ayar" seçilimin ürünüdür. Dürüst karşılaştırma için **hiç kullanılmamış** yeni katlarda (3 tekrar × 5 kat, farklı tohum) varsayılan ve ayarlı modelleri yeniden ölçüyoruz. Ek olarak **elle seçilmiş, küçük-veri dostu** bir LightGBM ayarını da yarıştırıyoruz (az yaprak, yüksek `min_child_samples`).

```python
TAZE = list(RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=2024).split(X_egitim, y_egitim))
TAZE_ILCE = []
for tohum in (21, 22, 23):
    TAZE_ILCE += list(StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=tohum).split(X_egitim, y_egitim, g_egitim))

elle_lgbm = dict(n_estimators=300, learning_rate=0.05, num_leaves=8, min_child_samples=10, subsample=0.8, subsample_freq=1, colsample_bytree=0.8, reg_lambda=1.0)
adaylar = {
    "LightGBM · varsayılan": yeni_lgbm(),
    "LightGBM · elle (küçük veri)": yeni_lgbm(**elle_lgbm),
    "LightGBM · ayarlı": clone(rs_lgbm.best_estimator_),
    "XGBoost · varsayılan": XGBClassifier(n_estimators=200, n_jobs=1, random_state=RANDOM_STATE, verbosity=0),
    "XGBoost · ayarlı": clone(rs_xgb.best_estimator_),
    "CatBoost · varsayılan": CatBoostClassifier(iterations=300, thread_count=1, random_seed=RANDOM_STATE, verbose=0, allow_writing_files=False),
    "CatBoost · ayarlı": clone(rs_cat.best_estimator_),
    "Rastgele orman": RandomForestClassifier(n_estimators=300, n_jobs=1, random_state=RANDOM_STATE),
}
satir = []
for ad, m in adaylar.items():
    r = cv_ozet(m, X_egitim, y_egitim, TAZE)
    ri = cv_ozet(m, X_egitim, y_egitim, TAZE_ILCE)
    satir.append({"aday": ad, "F1": r["F1 (premium)"].mean(), "F1 sd": r["F1 (premium)"].std(), "Log-loss": r["Log-loss"].mean(), "ROC-AUC": r["ROC-AUC"].mean(),
                  "AP": r["AP (PR-AUC)"].mean(), "F1 · ilçe bazlı": ri["F1 (premium)"].mean(), "Log-loss · ilçe bazlı": ri["Log-loss"].mean()})
taze = pd.DataFrame(satir).set_index("aday").sort_values("Log-loss")
display(taze.round(4))
rapor_kaydet(taze, "final_adaylari_taze_katlar")

arama_iyimser = {"LightGBM": lgbm_arama["log-loss"].iloc[0], "XGBoost": xgb_arama["log-loss"].iloc[0], "CatBoost": cat_arama["log-loss"].iloc[0]}
print("\nAramanın kendi (iyimser) en iyi log-loss'u → taze katlardaki gerçeği:")
for ad, v in arama_iyimser.items():
    print(f"  {ad:<9} arama {v:.4f}  →  taze {taze.loc[f'{ad} · ayarlı', 'Log-loss']:.4f}   (varsayılan: {taze.loc[f'{ad} · varsayılan', 'Log-loss']:.4f})")

fig, eksen = plt.subplots(1, 3, figsize=(17, 4.8))
t = taze.sort_values("Log-loss", ascending=False)
renkler = [RENK["turuncu"] if "LightGBM" in a else (RENK["gri"] if "orman" in a else RENK["cyan"]) for a in t.index]
eksen[0].barh(t.index, t["Log-loss"], color=renkler); eksen[0].set(title="Taze katlarda log-loss (düşük iyi)", xlim=(t["Log-loss"].min() * 0.85, t["Log-loss"].max() * 1.04))
eksen[1].barh(t.index, t["F1"], xerr=t["F1 sd"], color=renkler, error_kw={"ecolor": "#555"}); eksen[1].set(title="Taze katlarda F1 (yüksek iyi)", xlim=(0.7, 0.92)); eksen[1].set_yticklabels([])
eksen[2].barh(t.index, t["F1 · ilçe bazlı"], color=renkler); eksen[2].set(title="İlçe bazlı taze katlarda F1", xlim=(0.6, 0.9)); eksen[2].set_yticklabels([])
plt.tight_layout()
kaydet(fig, "21_final_adaylari_taze_katlar")
plt.show()
```

Şimdi şunu soruyoruz: *model hangi sütunlara bakıyor?* Tek bir "önem" ölçüsü yoktur; **dört farklı bakış açısı** kullanıp birlikte okuruz: kaç kez bölündü (`split`), ne kadar kazanç getirdi (`gain`), karıştırınca skor ne kadar düştü (permütasyon) ve her tahmine **ne kadar katkı** yaptı (TreeSHAP).

```python
FINAL_FIT = pipe_kur(clone(FINAL_MODEL), olcekle=False).fit(X_egitim, y_egitim)       # final: yalnızca eğitim kümesiyle eğitildi
booster = FINAL_FIT.named_steps["model"].booster_
ozellik_adlari = list(X_egitim.columns)

onem = pd.DataFrame({"split (kaç kez bölündü)": booster.feature_importance("split"), "gain (toplam kazanç)": booster.feature_importance("gain")}, index=ozellik_adlari)
onem = onem / onem.sum()                                                                  # yüzde payına çevir

# Permütasyon önemi: test setinde sütunu karıştır → log-loss ne kadar artıyor? (30 tekrar)
perm = permutation_importance(FINAL_FIT, X_test, y_test, scoring="neg_log_loss", n_repeats=30, random_state=RANDOM_STATE, n_jobs=1)
onem["permütasyon (log-loss artışı)"] = perm.importances_mean
onem = onem.sort_values("gain (toplam kazanç)", ascending=False)
display(onem.round(4))
rapor_kaydet(onem, "ozellik_onemi_lightgbm")

fig, eksen = plt.subplots(1, 3, figsize=(17, 4.8), sharey=True)
for ax, kolon, renk in zip(eksen, onem.columns, [RENK["altin"], RENK["turuncu"], RENK["mor"]]):
    t = onem[kolon].iloc[::-1]
    ax.barh(t.index, t.values, color=renk, xerr=perm.importances_std[[ozellik_adlari.index(i) for i in t.index]] if kolon.startswith("perm") else None, error_kw={"ecolor": "#555"})
    ax.set(title=kolon)
plt.tight_layout()
kaydet(fig, "22_ozellik_onemi_turleri")
plt.show()
```

### 🧪 12.3 · Ablasyon: Bilgiyi Gerçekten Çıkarınca Ne Oluyor?

Önem puanları **tahmindir**; asıl test, sütunu **çıkarıp modeli yeniden eğitmek**tir (taze katlar, aynı LightGBM ayarı). Ayrıca **`B`'yi geri koymanın** getirisini ölçüyoruz: etik kararımızın **bedeli** nedir?

```python
def ablasyon_olc(ad, sutunlar, X=None):
    X = X_egitim if X is None else X
    r = cv_ozet(clone(FINAL_MODEL), X[sutunlar], y_egitim, TAZE)
    return {"deney": ad, "F1": r["F1 (premium)"].mean(), "Log-loss": r["Log-loss"].mean(), "ROC-AUC": r["ROC-AUC"].mean()}


satir = [ablasyon_olc("hepsi (senaryo A)", OZELLIK_A)]
for s in OZELLIK_A:
    satir.append(ablasyon_olc(f"− {s}", [c for c in OZELLIK_A if c != s]))
gruplar = {"− suç + sanayi + kirlilik (CRIM, INDUS, NOX)": ["CRIM", "INDUS", "NOX"], "− erişim (DIS, RAD, TAX)": ["DIS", "RAD", "TAX"],
           "− konut (RM, AGE)": ["RM", "AGE"], "− okul (PTRATIO)": ["PTRATIO"], "− statü (LSTAT)": ["LSTAT"]}
for ad, cikar in gruplar.items():
    satir.append(ablasyon_olc(ad, [c for c in OZELLIK_A if c not in cikar]))
satir.append(ablasyon_olc("+ B (etik dışı; yalnızca bedel ölçümü)", OZELLIK_A + ["B"], X=X_egitim.assign(B=ham.iloc[idx_egitim]["B"].to_numpy())))
abl = pd.DataFrame(satir).set_index("deney")
abl["ΔF1"] = abl["F1"] - abl.loc["hepsi (senaryo A)", "F1"]
abl["ΔLog-loss"] = abl["Log-loss"] - abl.loc["hepsi (senaryo A)", "Log-loss"]
display(abl.round(4))
rapor_kaydet(abl, "ablasyon")

t = abl.drop(index="hepsi (senaryo A)").sort_values("ΔLog-loss")
fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(t.index, t["ΔLog-loss"], color=[RENK["pembe"] if v > 0 else RENK["yesil"] for v in t["ΔLog-loss"]])
ax.axvline(0, color="k", lw=0.8)
ax.set(title="Ablasyon: sütunu çıkarınca log-loss ne kadar değişiyor?", xlabel="Δ log-loss (+ = model kötüleşti)")
plt.tight_layout()
kaydet(fig, "25_ablasyon")
plt.show()
```

İkincisi, "yepyeni bir ilçede çalışır mı?" sorusuna cevap verir.

```python
FINAL_FIT_G = pipe_kur(clone(FINAL_MODEL), olcekle=False).fit(X_tum.iloc[idx_g_egitim], y_tum[idx_g_egitim])      # grup görünümü için ayrı eğitim
X_g_test, y_g_test = X_tum.iloc[idx_g_test], y_tum[idx_g_test]

def final_degerlendir(model, X, y, esik=0.5):
    p = model.predict_proba(X)[:, 1]
    return p, metrikler(y, (p >= esik).astype(int), p, p)


p_test, m_test = final_degerlendir(FINAL_FIT, X_test, y_test)
p_gtest, m_gtest = final_degerlendir(FINAL_FIT_G, X_g_test, y_g_test)

# karşılaştırma için: aynı iki görünümde diğer finalistler
rakipler = {"LightGBM · varsayılan": adaylar["LightGBM · varsayılan"], "XGBoost · ayarlı": adaylar["XGBoost · ayarlı"], "CatBoost · varsayılan": adaylar["CatBoost · varsayılan"],
            "Rastgele orman": adaylar["Rastgele orman"], "Lojistik regresyon": Pipeline([("o", StandardScaler()), ("m", LogisticRegression(max_iter=3000))])}
satir = [{"model": f"{FINAL_AD} (final)", "görünüm": "rastgele test", **m_test}, {"model": f"{FINAL_AD} (final)", "görünüm": "ilçe bazlı test", **m_gtest}]
for ad, m in rakipler.items():
    _, a = final_degerlendir(clone(m).fit(X_egitim, y_egitim), X_test, y_test); satir.append({"model": ad, "görünüm": "rastgele test", **a})
    _, b = final_degerlendir(clone(m).fit(X_tum.iloc[idx_g_egitim], y_tum[idx_g_egitim]), X_g_test, y_g_test); satir.append({"model": ad, "görünüm": "ilçe bazlı test", **b})
final_tablo = pd.DataFrame(satir)
display(final_tablo.set_index(["görünüm", "model"]).sort_index(level=0).round(3))
rapor_kaydet(final_tablo, "final_model_karsilastirma_test", index=False)
rapor_kaydet(pd.DataFrame(classification_report(y_test, (p_test >= 0.5).astype(int), target_names=SINIF_ADLARI, output_dict=True)).T, "siniflandirma_raporu_test")
print(classification_report(y_test, (p_test >= 0.5).astype(int), target_names=SINIF_ADLARI, digits=3))
```

```python
def bootstrap_metrik(y, p, B=2000, seed=RANDOM_STATE):
    """Test metriklerinin %95 bootstrap aralığı (mahalleleri yeniden örnekleyerek)."""
    rng = np.random.default_rng(seed); n = len(y); sonuc = {"F1": [], "ROC-AUC": [], "Doğruluk": []}
    for _ in range(B):
        i = rng.integers(0, n, n)
        if y[i].min() == y[i].max(): continue
        sonuc["F1"].append(f1_score(y[i], p[i] >= 0.5)); sonuc["ROC-AUC"].append(roc_auc_score(y[i], p[i])); sonuc["Doğruluk"].append(accuracy_score(y[i], p[i] >= 0.5))
    return {k: (np.mean(v), np.percentile(v, 2.5), np.percentile(v, 97.5)) for k, v in sonuc.items()}


satir = []
for ad, y_, p_ in [("rastgele test (102)", y_test, p_test), ("ilçe bazlı test (151)", y_g_test, p_gtest)]:
    for metrik, (o, a, u) in bootstrap_metrik(y_, p_).items():
        satir.append({"görünüm": ad, "metrik": metrik, "ortalama": o, "alt %2,5": a, "üst %97,5": u})
boot_final = pd.DataFrame(satir)
display(boot_final.round(3))
rapor_kaydet(boot_final, "bootstrap_final_test", index=False)

fig, eksen = plt.subplots(1, 4, figsize=(19, 4.6))
for ax, (ad, y_, p_) in zip(eksen[:2], [("Rastgele test", y_test, p_test), ("İlçe bazlı test", y_g_test, p_gtest)]):
    cm = confusion_matrix(y_, (p_ >= 0.5).astype(int))
    sns.heatmap(cm, annot=np.array([[f"{v}\n(%{v / cm[i].sum() * 100:.0f})" for v in row] for i, row in enumerate(cm)]), fmt="", cmap="YlOrBr", cbar=False, ax=ax,
                xticklabels=SINIF_ADLARI, yticklabels=SINIF_ADLARI, annot_kws={"size": 13})
    ax.set(title=f"{ad}: karışıklık matrisi", xlabel="Tahmin", ylabel="Gerçek"); ax.grid(False)
for ad, y_, p_, renk in [("rastgele test", y_test, p_test, RENK["mavi"]), ("ilçe bazlı test", y_g_test, p_gtest, RENK["pembe"])]:
    f, t, _ = roc_curve(y_, p_); eksen[2].plot(f, t, lw=2.2, color=renk, label=f"{ad} (AUC {roc_auc_score(y_, p_):.3f})")
    pr, rc, _ = precision_recall_curve(y_, p_); eksen[3].plot(rc, pr, lw=2.2, color=renk, label=f"{ad} (AP {average_precision_score(y_, p_):.3f})")
eksen[2].plot([0, 1], [0, 1], "--", color=RENK["gri"]); eksen[2].set(title="ROC eğrisi", xlabel="Yanlış pozitif oranı", ylabel="Duyarlılık"); eksen[2].legend(loc="lower right")
eksen[3].set(title="Kesinlik–duyarlılık eğrisi", xlabel="Duyarlılık", ylabel="Kesinlik", ylim=(0, 1.02)); eksen[3].legend(loc="lower left")
plt.tight_layout()
kaydet(fig, "26_final_karisiklik_roc_pr")
plt.show()
```

### 🔬 13.2 · Segment Denetimi: Hangi Alt Grupta Daha Çok Yanılıyoruz?

Genel doğruluk, bazı alt grupların **kötü hizmet aldığını** gizleyebilir. Eğitim kümesinin **out-of-fold** tahminleriyle (404 mahalle; test setinden daha kalabalık) alt gruplarda doğruluğu ve **Wilson güven aralığını** (Wilson, 1927) hesaplıyoruz.

```python
oof_final = cross_val_predict(pipe_kur(clone(FINAL_MODEL), False), X_egitim, y_egitim, cv=StratifiedKFold(5, shuffle=True, random_state=RANDOM_STATE), method="predict_proba")[:, 1]
tahmin_oof = (oof_final >= 0.5).astype(int)
medv_egitim = ham.iloc[idx_egitim]["MEDV"].to_numpy()


def wilson(k, n, z=1.96):
    """Wilson güven aralığı (küçük gruplarda yüzdeler için güvenilir)."""
    if n == 0: return (np.nan, np.nan)
    p = k / n; d = 1 + z ** 2 / n; c = p + z ** 2 / (2 * n); h = z * np.sqrt(p * (1 - p) / n + z ** 2 / (4 * n ** 2))
    return ((c - h) / d, (c + h) / d)


segmentler = {
    "Nehir kıyısı": np.where(X_egitim["CHAS"] == 1, "kıyıda", "kıyıda değil"),
    "Şehir merkezi kümesi (RAD=24)": np.where(X_egitim["RAD"] == 24, "RAD = 24", "diğer"),
    "Oda sayısı (RM)": pd.cut(X_egitim["RM"], [0, 5.9, 6.6, 99], labels=["≤ 5,9", "5,9–6,6", "> 6,6"]).astype(str),
    "Suç oranı (CRIM)": pd.cut(X_egitim["CRIM"], [0, 0.1, 3, 999], labels=["düşük (≤0,1)", "orta", "yüksek (>3)"]).astype(str),
    "Gerçek değer bandı (MEDV)": pd.cut(medv_egitim, [0, 20, 22, 25, 28, 30, 100], labels=["< 20", "20–22", "22–25", "25–28", "28–30", "> 30"]).astype(str),
}
satir = []
for sg_ad, etiket in segmentler.items():
    for deger in pd.unique(etiket):
        m = etiket == deger; n = int(m.sum()); k = int((tahmin_oof[m] == y_egitim[m]).sum()); a, u = wilson(k, n)
        satir.append({"segment": sg_ad, "grup": deger, "n": n, "premium %": y_egitim[m].mean() * 100, "doğruluk": k / n, "alt %2,5": a, "üst %97,5": u})
seg = pd.DataFrame(satir)
display(seg.round(3))
rapor_kaydet(seg, "segment_denetimi", index=False)

fig, ax = plt.subplots(figsize=(10, 0.36 * len(seg) + 1.3))
y_ = np.arange(len(seg))[::-1]
ax.errorbar(seg["doğruluk"], y_, xerr=[seg["doğruluk"] - seg["alt %2,5"], seg["üst %97,5"] - seg["doğruluk"]], fmt="o", color=RENK["mor"], ecolor=RENK["gri"], capsize=3)
ax.axvline((tahmin_oof == y_egitim).mean(), color=RENK["pembe"], ls="--", label=f"genel doğruluk {(tahmin_oof == y_egitim).mean():.3f}")
ax.set_yticks(y_); ax.set_yticklabels([f"{s} · {g} (n={n})" for s, g, n in zip(seg["segment"], seg["grup"], seg["n"])], fontsize=8.5)
ax.set(title="Segment denetimi: doğruluk ve %95 Wilson aralığı", xlabel="doğruluk"); ax.legend(loc="lower left")
plt.tight_layout()
kaydet(fig, "27_segment_denetimi")
plt.show()
```

</p>

Bir sınıflandırıcı **olasılık** üretir; **karar** ise olasılığın üstüne bir **iş kuralıdır**. Bu bölümde üç soruyu cevaplıyoruz: (1) Olasılıklar **güvenilir** mi? (2) Sınırlı bütçeyle **hangi mahalleleri** incelemeliyiz? (3) Hata türlerinin **maliyeti** farklıysa **eşik** ne olmalı?

```python
def ece_top(y, olas, n_kutu=10):
    """Beklenen kalibrasyon hatası (ECE): kutular içinde |ortalama olasılık − gerçek oran|'ın ağırlıklı ortalaması."""
    kutu = np.minimum((olas * n_kutu).astype(int), n_kutu - 1)
    return sum(abs(olas[kutu == k].mean() - y[kutu == k].mean()) * (kutu == k).mean() for k in range(n_kutu) if (kutu == k).any())


kalibre_modeller = {
    "kalibre edilmemiş": pipe_kur(clone(FINAL_MODEL)),
    "sigmoid (Platt)": CalibratedClassifierCV(pipe_kur(clone(FINAL_MODEL)), method="sigmoid", cv=5),
    "isotonic": CalibratedClassifierCV(pipe_kur(clone(FINAL_MODEL)), method="isotonic", cv=5),
}
iclerde = StratifiedKFold(5, shuffle=True, random_state=99)
satir, egriler_k = [], {}
for ad, m in kalibre_modeller.items():
    p = cross_val_predict(m, X_egitim, y_egitim, cv=iclerde, method="predict_proba")[:, 1]
    egriler_k[ad] = p
    satir.append({"yöntem": ad, "Log-loss": log_loss(y_egitim, p), "Brier": brier_score_loss(y_egitim, p), "ECE": ece_top(y_egitim, p), "ROC-AUC": roc_auc_score(y_egitim, p)})
kal = pd.DataFrame(satir).set_index("yöntem")
display(kal.round(4))
rapor_kaydet(kal, "kalibrasyon")

fig, eksen = plt.subplots(1, 2, figsize=(14, 5))
for (ad, p), renk in zip(egriler_k.items(), [RENK["mavi"], RENK["altin"], RENK["pembe"]]):
    gercek, tahmin_ort = calibration_curve(y_egitim, p, n_bins=8, strategy="quantile")
    eksen[0].plot(tahmin_ort, gercek, "o-", color=renk, lw=2, label=f"{ad} (ECE {kal.loc[ad, 'ECE']:.3f})")
eksen[0].plot([0, 1], [0, 1], "--", color=RENK["gri"]); eksen[0].set(title="Güvenilirlik (kalibrasyon) eğrisi", xlabel="tahmin edilen olasılık", ylabel="gerçekleşen premium oranı"); eksen[0].legend()
eksen[1].hist(egriler_k["kalibre edilmemiş"][y_egitim == 0], bins=25, alpha=0.7, color=SINIF_RENK[0], label="gerçek standart")
eksen[1].hist(egriler_k["kalibre edilmemiş"][y_egitim == 1], bins=25, alpha=0.7, color=SINIF_RENK[1], label="gerçek premium")
eksen[1].set(title="Olasılık dağılımı (out-of-fold)", xlabel="P(premium)", ylabel="mahalle"); eksen[1].legend()
plt.tight_layout()
kaydet(fig, "28_kalibrasyon")
plt.show()
```

### 💶 14.3 · Maliyete Duyarlı Eşik

Fon için **premium bir mahalleyi kaçırmak**, yanlış alarm verip **boşuna incelemek**ten daha pahalı. Varsayım: kaçırılan premium **3 birim**, gereksiz inceleme **1 birim**. Olasılıklar kalibreyse teorik en iyi eşik `τ = C_FP / (C_FP + C_FN) = 0,25`'tir. Biz bunu **eğitimin out-of-fold olasılıklarında deneysel olarak** arıyoruz; test setinde **yalnızca doğrulama** yapıyoruz.

```python
C_KACIRMA, C_BOSA = 3.0, 1.0
esikler = np.linspace(0.02, 0.98, 97)
p_oof = egriler_k["kalibre edilmemiş"]


def toplam_maliyet(y, p, esik):
    tahmin = p >= esik
    return C_KACIRMA * np.sum((y == 1) & ~tahmin) + C_BOSA * np.sum((y == 0) & tahmin)


maliyet_oof = np.array([toplam_maliyet(y_egitim, p_oof, t) / len(y_egitim) for t in esikler])
TAU = float(esikler[np.argmin(maliyet_oof)])
teorik = C_BOSA / (C_BOSA + C_KACIRMA)

satir = []
for ad, y_, p_ in [("rastgele test", y_test, p_test), ("ilçe bazlı test", y_g_test, p_gtest)]:
    for kural, t in [("eşik 0,50", 0.5), (f"teorik τ = {teorik:.2f}", teorik), (f"deneysel τ = {TAU:.2f}", TAU)]:
        tah = p_ >= t
        satir.append({"görünüm": ad, "kural": kural, "toplam maliyet / mahalle": toplam_maliyet(y_, p_, t) / len(y_), "kaçırılan premium": int(((y_ == 1) & ~tah).sum()),
                      "gereksiz inceleme": int(((y_ == 0) & tah).sum()), "F1": f1_score(y_, tah)})
    satir.append({"görünüm": ad, "kural": "hepsini incele", "toplam maliyet / mahalle": toplam_maliyet(y_, p_, 0.0) / len(y_), "kaçırılan premium": 0, "gereksiz inceleme": int((y_ == 0).sum()), "F1": f1_score(y_, np.ones_like(y_))})
    satir.append({"görünüm": ad, "kural": "hiçbirini inceleme", "toplam maliyet / mahalle": toplam_maliyet(y_, p_, 1.01) / len(y_), "kaçırılan premium": int(y_.sum()), "gereksiz inceleme": 0, "F1": 0.0})
maliyet_tablo = pd.DataFrame(satir)
display(maliyet_tablo.round(3))
rapor_kaydet(maliyet_tablo, "maliyete_duyarli_karar", index=False)

fig, ax = plt.subplots(figsize=(9.5, 4.8))
ax.plot(esikler, maliyet_oof, lw=2.4, color=RENK["mor"], label="eğitim (out-of-fold)")
ax.plot(esikler, [toplam_maliyet(y_test, p_test, t) / len(y_test) for t in esikler], lw=1.8, color=RENK["gri"], label="rastgele test")
ax.axvline(0.5, color=RENK["mavi"], ls=":", label="0,5"); ax.axvline(TAU, color=RENK["yesil"], ls="--", label=f"deneysel τ = {TAU:.2f}"); ax.axvline(teorik, color=RENK["pembe"], ls="--", alpha=0.6, label=f"teorik τ = {teorik:.2f}")
ax.set(title="Eşik ↔ mahalle başına maliyet", xlabel="eşik τ", ylabel="maliyet / mahalle"); ax.legend()
kaydet(fig, "30_maliyet_esigi")
plt.show()
print(f"Seçilen eşik τ = {TAU:.2f}  (teorik {teorik:.2f})")
```

### ❓ Quiz (cevaplar açılır kutularda)

<details><summary><b>1.</b> `MEDV` neden özellik olarak kullanılamaz? Kullanılırsa ne olur?</summary>

Etiket `MEDV ≥ 25` kuralıyla **doğrudan** üretildi; modele vermek cevabı vermektir. Model tek bölünmeyle kuralı bulur, skor ≈ 1'e çıkar; ama üretimde `MEDV` bilinmediği için model işe yaramaz. İlke: hedefi üreten ya da tahmin anında **elde olmayan** her sütun özellik olamaz.
</details>

<details><summary><b>2.</b> İlçe bazlı doğrulamada skor neden düşüyor?</summary>

Aynı ilçenin mahalleleri `TAX`, `INDUS`, `PTRATIO` gibi sütunlarda **birebir aynıdır**; rastgele katlamada model bunları ilçe parmak izi olarak ezberleyip komşunun etiketini okur. İlçe bazlı katlamada ilçe **hiç görülmemiştir**, dolayısıyla skor gerçeğe yaklaşır.
</details>

<details><summary><b>3.</b> Üst sıradaki iki modelin F1 farkının bootstrap aralığı sıfırı kesiyorsa ne anlama gelir?</summary>

Fark **şanstan ayırt edilemez**. Hangisinin "gerçekten daha iyi" olduğunu söyleyemeyiz; sıralama bir tahmindir, kesinlik değil.
</details>

<details><summary><b>4.</b> LightGBM neden "yaprak yaprak" büyür ve bunun küçük veride riski nedir?</summary>

Her adımda **en çok hata azaltacak yaprağı** bölerek hızlı ve verimli öğrenir. Ancak yaprak sayısı serbestse ağaç **çok dengesiz ve derin** büyüyüp eğitim satırlarını ezberleyebilir; küçük veride `min_child_samples`, `num_leaves` ve `max_depth` bu yüzden kritik frenlerdir.
</details>

<details><summary><b>5.</b> Ayar neden log-loss'u çok, F1'i az iyileştirdi?</summary>

F1, olasılığı 0,5'te **keserek** hesaplanır: model çoğu mahallede zaten doğru tarafta olduğundan kesinti değişmez. Log-loss ise olasılığın **ne kadar emin** olduğuna bakar; frenler (λ, `min_split_gain`, alt örnekleme) aşırı özgüveni azaltır ve log-loss'u düşürür.
</details>

<details><summary><b>6.</b> Maliyete duyarlı eşik neden 0,5'ten küçük çıktı?</summary>

Kaçırılan premium (3 birim) yanlış alarmdan (1 birim) pahalı. Beklenen maliyeti en düşük yapan karar kuralı `P(premium) ≥ C_FP / (C_FP + C_FN) = 0,25` civarındadır. Biz eşiği **eğitimin out-of-fold olasılıklarında** seçtik, test setinde yalnızca doğruladık.
</details>

<details><summary><b>7.</b> TreeSHAP toplamsallık kontrolü ne söyler?</summary>

Bir mahalle için tüm özellik katkıları ve taban değer toplanırsa modelin **ham çıktısı (log-odds)** elde edilir. Kontrolde fark ≈ 10⁻¹⁴ çıktı: katkılar **tutarlı** bir açıklama sunuyor.
</details>
