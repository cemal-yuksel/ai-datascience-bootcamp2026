## 🛠️ Tabloları Oluşturma (CREATE TABLE)

Şimdi SQL komutlarıyla tablolarımızı oluşturacağız. **CREATE TABLE** komutu tablo yaratmak için kullanılır.

````python
# Eğer tablolar zaten varsa, önce silelim (temiz başlangıç için)
cursor.execute("DROP TABLE IF EXISTS order_items")
cursor.execute("DROP TABLE IF EXISTS orders")
cursor.execute("DROP TABLE IF EXISTS customers")
cursor.execute("DROP TABLE IF EXISTS products")

print("🗑️ Eski tablolar temizlendi (varsa)")
print("📋 Yeni tablolar oluşturuluyor...\n")
````

````python
# 1️⃣ CUSTOMERS (Müşteriler) Tablosu
cursor.execute("""
    CREATE TABLE customers (
        customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
        first_name TEXT NOT NULL,
        last_name TEXT NOT NULL,
        email TEXT UNIQUE NOT NULL,
        city TEXT NOT NULL,
        age INTEGER,
        registration_date DATE NOT NULL
    )
""")

print("✅ 1. CUSTOMERS tablosu oluşturuldu!")
print("   └─ Kolonlar: customer_id, first_name, last_name, email, city, age, registration_date")
````

````python
# 2️⃣ PRODUCTS (Ürünler) Tablosu
cursor.execute("""
    CREATE TABLE products (
        product_id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        category TEXT NOT NULL,
        price REAL NOT NULL,
        stock INTEGER NOT NULL DEFAULT 0
    )
""")

print("✅ 2. PRODUCTS tablosu oluşturuldu!")
print("   └─ Kolonlar: product_id, name, category, price, stock")
````

````python
# 3️⃣ ORDERS (Siparişler) Tablosu
cursor.execute("""
    CREATE TABLE orders (
        order_id INTEGER PRIMARY KEY AUTOINCREMENT,
        customer_id INTEGER NOT NULL,
        order_date DATE NOT NULL,
        status TEXT NOT NULL,
        total_amount REAL NOT NULL,
        FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
    )
""")

print("✅ 3. ORDERS tablosu oluşturuldu!")
print("   └─ Kolonlar: order_id, customer_id (FK), order_date, status, total_amount")
````

````python
# 4️⃣ ORDER_ITEMS (Sipariş Detayları) Tablosu
cursor.execute("""
    CREATE TABLE order_items (
        item_id INTEGER PRIMARY KEY AUTOINCREMENT,
        order_id INTEGER NOT NULL,
        product_id INTEGER NOT NULL,
        quantity INTEGER NOT NULL,
        unit_price REAL NOT NULL,
        FOREIGN KEY (order_id) REFERENCES orders(order_id),
        FOREIGN KEY (product_id) REFERENCES products(product_id)
    )
""")

print("✅ 4. ORDER_ITEMS tablosu oluşturuldu!")
print("   └─ Kolonlar: item_id, order_id (FK), product_id (FK), quantity, unit_price")

# Değişiklikleri kaydet
conn.commit()
print("\n💾 Tüm tablolar veritabanına kaydedildi!")
````

## 🌟 SELECT * : Tüm Kolonları Çekme

**Yıldız (*)** sembolü **"her şey"** demektir. Tablodaki **tüm kolonları** getirir.

### 📋 Örnek 1: Müşteri Tablosundaki Tüm Verileri Görüntüleme

````python
# SQL sorgusu: Customers tablosundan TÜM kolonları ve TÜM satırları çek
query = """
SELECT * 
FROM customers
"""

# Sorguyu çalıştır ve pandas DataFrame'e çevir
df_customers = pd.read_sql_query(query, conn)

# Başlık
print("=" * 80)
print("👥 TÜM MÜŞTERİ VERİLERİ (SELECT * FROM customers)")
print("=" * 80)

# DataFrame'i göster
display(df_customers)

# İstatistikler
print(f"\n📊 Toplam {len(df_customers)} müşteri kaydı bulundu.")
print(f"📋 Kolonlar: {', '.join(df_customers.columns)}")
````

## 🎚️ LIMIT: Satır Sayısını Sınırlama

**LIMIT**, sorgu sonucunda **maksimum kaç satır** getirileceğini belirtir. Büyük tablolarda **mutlaka** kullanılmalıdır!

### 📋 Örnek 2: İlk 5 Müşteriyi Görüntüleme

````python
# SQL sorgusu: Sadece ilk 5 müşteriyi getir
query = """
SELECT * 
FROM customers
LIMIT 5
"""

# Sorguyu çalıştır
df_top5 = pd.read_sql_query(query, conn)

# Başlık
print("=" * 80)
print("👥 İLK 5 MÜŞTERİ (SELECT * FROM customers LIMIT 5)")
print("=" * 80)

# DataFrame'i göster
display(df_top5)

print(f"\n✅ LIMIT 5 sayesinde sadece {len(df_top5)} satır geldi!")
print("💡 İpucu: Büyük tablolarda 'hızlı bakış' için LIMIT kullanın!")
````

### 📋 Örnek 3: Ürünler Tablosundan İlk 10 Kayıt

````python
# Ürün kataloğuna hızlı bakış
query = """
SELECT *
FROM products
LIMIT 10
"""

df_products = pd.read_sql_query(query, conn)

print("=" * 80)
print("📦 ÜRÜN KATALOĞU - İLK 10 ÜRÜN")
print("=" * 80)
display(df_products)

print(f"\n💰 Fiyat aralığı: {df_products['price'].min():,.0f} ₺ - {df_products['price'].max():,.0f} ₺")
print(f"📊 Kategoriler: {', '.join(df_products['category'].unique())}")
````

### 📋 Örnek 4: Sadece İsim ve Email Çekme

````python
# Sadece isim ve email kolonlarını seçiyoruz
query = """
SELECT first_name, last_name, email
FROM customers
LIMIT 10
"""

df_contact = pd.read_sql_query(query, conn)

print("=" * 80)
print("📧 MÜŞTERİ İLETİŞİM LİSTESİ (Sadece İsim ve Email)")
print("=" * 80)
display(df_contact)

print(f"\n✅ Sadece {len(df_contact.columns)} kolon çektik (customer_id, city, age gelmedi)")
print("💡 Gereksiz kolonları çekmemek, büyük tablolarda ciddi performans kazandırır!")
````

### 📋 Örnek 5: Ürün İsmi ve Fiyatları

````python
# Fiyat listesi oluşturma
query = """
SELECT name, price
FROM products
"""

df_price_list = pd.read_sql_query(query, conn)

print("=" * 80)
print("💰 FİYAT LİSTESİ (name, price)")
print("=" * 80)
display(df_price_list)

# En pahalı ve en ucuz ürün
en_pahali = df_price_list.loc[df_price_list['price'].idxmax()]
en_ucuz = df_price_list.loc[df_price_list['price'].idxmin()]

print(f"\n🔝 En pahalı: {en_pahali['name']} - {en_pahali['price']:,.0f} ₺")
print(f"💵 En ucuz: {en_ucuz['name']} - {en_ucuz['price']:,.0f} ₺")
````

## 📝 Eşitlik Filtreleme (=)

### 📋 Örnek 1: İstanbul'daki Müşteriler

````python
# Sadece İstanbul'daki müşterileri getir
query = """
SELECT first_name, last_name, email, city, age
FROM customers
WHERE city = 'İstanbul'
"""

df_istanbul = pd.read_sql_query(query, conn)

print("=" * 80)
print("🏙️ İSTANBUL'DAKİ MÜŞTERİLER (WHERE city = 'İstanbul')")
print("=" * 80)
display(df_istanbul)

print(f"\n✅ Toplam {len(df_istanbul)} İstanbul müşterisi bulundu.")
print(f"📊 Ortalama yaş: {df_istanbul['age'].mean():.1f}")
````

### 📋 Örnek 3: 30 Yaş Üstü Müşteriler

````python
# 30 yaş üstü müşterileri getir
query = """
SELECT first_name, last_name, age, city
FROM customers
WHERE age > 30
"""

df_over_30 = pd.read_sql_query(query, conn)

print("=" * 80)
print("👴 30 YAŞ ÜSTÜ MÜŞTERİLER (WHERE age > 30)")
print("=" * 80)
display(df_over_30)

print(f"\n✅ {len(df_over_30)} müşteri 30 yaşın üstünde.")
print(f"📊 Yaş aralığı: {df_over_30['age'].min()} - {df_over_30['age'].max()}")
print(f"📍 Şehir dağılımı:")
print(df_over_30['city'].value_counts().to_string())
````

## 🔤 LIKE: Metin Arama ve Pattern Matching

**LIKE** operatörü, metin kolonlarında **kısmi eşleşme** yapmak için kullanılır.

<div style="background-color: #E8F5E9; padding: 20px; border-radius: 10px; margin: 20px 0;">
    <h4 style="color: #2E7D32; margin-top: 0;">🎯 Wildcard (Joker) Karakterler</h4>
    <ul style="font-size: 15px; line-height: 1.7;">
        <li><strong>%</strong> → Herhangi bir karakter dizisi (0 veya daha fazla)</li>
        <li><strong>_</strong> → Tek bir karakter</li>
    </ul>
    <p style="font-size: 15px; margin: 10px 0;"><strong>Örnekler:</strong></p>
    <ul style="font-size: 14px; line-height: 1.7;">
        <li><code>'A%'</code> → A ile başlayan (Ahmet, Ali, Ayşe)</li>
        <li><code>'%gmail%'</code> → gmail içeren (ahmet@gmail.com)</li>
        <li><code>'%son'</code> → son ile biten (terson, wilson)</li>
        <li><code>'M_rat'</code> → M ile başlayıp rat ile biten 5 harfli (Murat)</li>
    </ul>
</div>

### 📋 Örnek 6: İsmi 'A' ile Başlayan Müşteriler

````python
# İsmi A harfi ile başlayan müşteriler
query = """
SELECT first_name, last_name, email, city
FROM customers
WHERE first_name LIKE 'A%'
"""

df_starts_a = pd.read_sql_query(query, conn)

print("=" * 80)
print("🅰️ İSMİ 'A' İLE BAŞLAYAN MÜŞTERİLER (WHERE first_name LIKE 'A%')")
print("=" * 80)
display(df_starts_a)

print(f"\n✅ {len(df_starts_a)} müşteri bulundu.")
print(f"📝 İsimler: {', '.join(df_starts_a['first_name'].unique())}")
````

### 📋 Örnek 7: Gmail Kullanan Müşteriler

````python
# Email adresi gmail olan müşteriler
query = """
SELECT first_name, last_name, email
FROM customers
WHERE email LIKE '%gmail%'
"""

df_gmail = pd.read_sql_query(query, conn)

print("=" * 80)
print("📧 GMAIL KULLANAN MÜŞTERİLER (WHERE email LIKE '%gmail%')")
print("=" * 80)
display(df_gmail)

print(f"\n✅ {len(df_gmail)} müşteri Gmail kullanıyor.")
print(f"📊 Toplam müşterinin %{(len(df_gmail)/15)*100:.1f}'ü Gmail tercih ediyor.")
````

### 📋 Örnek 8: İsminde 'Phone' Geçen Ürünler

````python
# Ürün isminde 'Phone' geçenler (iPhone, Galaxy Phone vs.)
query = """
SELECT name, category, price, stock
FROM products
WHERE name LIKE '%Phone%'
"""

df_phones_like = pd.read_sql_query(query, conn)

print("=" * 80)
print("📱 İSMİNDE 'PHONE' GEÇEN ÜRÜNLER (WHERE name LIKE '%Phone%')")
print("=" * 80)
display(df_phones_like)

if len(df_phones_like) > 0:
    print(f"\n✅ {len(df_phones_like)} ürün bulundu.")
else:
    print("\n⚠️ İsminde 'Phone' geçen ürün bulunamadı.")
    print("💡 İpucu: LIKE büyük-küçük harf duyarlıdır (case-sensitive)!")
````

## 📦 IN: Çoklu Değer Filtreleme

**IN** operatörü, bir kolonun **birden fazla değerden birine eşit olup olmadığını** kontrol eder.

<div style="background-color: #FFF3E0; padding: 20px; border-radius: 10px; margin: 20px 0;">
    <h4 style="color: #E65100; margin-top: 0;">💡 IN Kullanımı</h4>
    <p style="font-size: 15px; line-height: 1.7; margin-bottom: 0;">
        <strong>Uzun yol:</strong> <code>WHERE city = 'İstanbul' OR city = 'Ankara' OR city = 'İzmir'</code><br/>
        <strong>Kısa yol:</strong> <code>WHERE city IN ('İstanbul', 'Ankara', 'İzmir')</code>
    </p>
</div>

### 📋 Örnek 9: Büyük Şehirlerdeki Müşteriler

````python
# İstanbul, Ankara veya İzmir'deki müşteriler
query = """
SELECT first_name, last_name, city, age
FROM customers
WHERE city IN ('İstanbul', 'Ankara', 'İzmir')
"""

df_big_cities = pd.read_sql_query(query, conn)

print("=" * 80)
print("🏙️ BÜYÜK ŞEHİRLERDEKİ MÜŞTERİLER (WHERE city IN (...))")
print("=" * 80)
display(df_big_cities)

print(f"\n✅ {len(df_big_cities)} müşteri İstanbul, Ankara veya İzmir'de.")
print(f"📊 Şehir dağılımı:")
print(df_big_cities['city'].value_counts().to_string())
````

## 📏 BETWEEN: Aralık Belirleme

**BETWEEN** operatörü, bir değerin **belirli bir aralıkta olup olmadığını** kontrol eder (başlangıç ve bitiş değerleri **dahil**).

<div style="background-color: #E8EAF6; padding: 20px; border-radius: 10px; margin: 20px 0;">
    <h4 style="color: #3F51B5; margin-top: 0;">💡 BETWEEN Kullanımı</h4>
    <p style="font-size: 15px; line-height: 1.7; margin-bottom: 0;">
        <strong>Uzun yol:</strong> <code>WHERE age >= 25 AND age <= 35</code><br/>
        <strong>Kısa yol:</strong> <code>WHERE age BETWEEN 25 AND 35</code>
    </p>
</div>

### 📋 Örnek 11: 25-35 Yaş Aralığındaki Müşteriler

````python
# 25-35 yaş arası müşteriler (genç yetişkin segment)
query = """
SELECT first_name, last_name, age, city
FROM customers
WHERE age BETWEEN 25 AND 35
"""

df_age_range = pd.read_sql_query(query, conn)

print("=" * 80)
print("👥 25-35 YAŞ ARASI MÜŞTERİLER (WHERE age BETWEEN 25 AND 35)")
print("=" * 80)
display(df_age_range)

print(f"\n✅ {len(df_age_range)} müşteri bu yaş aralığında.")
print(f"📊 Ortalama yaş: {df_age_range['age'].mean():.1f}")
print(f"🎯 Hedef segment: Genç yetişkinler")
````

### 📋 Örnek 12: Orta Segment Fiyat Aralığı (10.000-30.000 TL)

````python
# 10.000 - 30.000 TL arası ürünler (orta segment)
query = """
SELECT name, category, price, stock
FROM products
WHERE price BETWEEN 10000 AND 30000
"""

df_mid_range = pd.read_sql_query(query, conn)

print("=" * 80)
print("💰 ORTA SEGMENT ÜRÜNLER (10.000-30.000 TL)")
print("=" * 80)
display(df_mid_range)

print(f"\n✅ {len(df_mid_range)} orta segment ürün mevcut.")
print(f"💵 Ortalama fiyat: {df_mid_range['price'].mean():,.0f} ₺")
print(f"📦 Toplam stok değeri: {(df_mid_range['price'] * df_mid_range['stock']).sum():,.0f} ₺")
````

### 📋 Örnek 7: Hedef Müşteri Segmenti (Karmaşık Filtre)

### 📋 Örnek 8: Satış Fırsatları (Karmaşık Ürün Filtresi)

**Aggregate (Toplam) Fonksiyonlar**, birden fazla satır üzerinde **hesaplama yaparak tek bir sonuç** döndürür.

### 📋 Örnek 1: Toplam Müşteri Sayısı (COUNT)

````python
# Veritabanındaki toplam müşteri sayısı
query = """
SELECT 
    COUNT(*) as toplam_musteri,
    COUNT(DISTINCT city) as sehir_sayisi,
    COUNT(DISTINCT email) as benzersiz_email
FROM customers
"""

df_counts = pd.read_sql_query(query, conn)

print("=" * 80)
print("📊 MÜŞTERİ İSTATİSTİKLERİ (COUNT)")
print("=" * 80)
display(df_counts)

print(f"\n✅ Sistemde {df_counts['toplam_musteri'][0]} müşteri kayıtlı")
print(f"🏙️ {df_counts['sehir_sayisi'][0]} farklı şehirde hizmet veriyoruz")
print(f"📧 {df_counts['benzersiz_email'][0]} benzersiz e-posta adresi mevcut")
````

### 📋 Örnek 2: Envanter Değeri (SUM)

````python
# Stoktaki toplam ürün değeri (SUM ile hesaplama)
query = """
SELECT 
    SUM(stock) as toplam_stok_adedi,
    SUM(price * stock) as toplam_envanter_degeri
FROM products
"""

df_inventory = pd.read_sql_query(query, conn)

print("=" * 80)
print("💰 ENVANTER DEĞERİ ANALİZİ (SUM)")
print("=" * 80)
display(df_inventory)

toplam_adet = df_inventory['toplam_stok_adedi'][0]
toplam_deger = df_inventory['toplam_envanter_degeri'][0]

print(f"\n📦 Toplam {toplam_adet:,} adet ürün stokta")
print(f"💵 Envanter değeri: {toplam_deger:,.0f} ₺")
print(f"📊 Ürün başına ortalama değer: {(toplam_deger/toplam_adet):,.0f} ₺")
````

### 📋 Örnek 3: Ortalama Değerler (AVG)

````python
# Ortalama müşteri yaşı ve ortalama ürün fiyatı
query_customers = """
SELECT AVG(age) as ortalama_yas
FROM customers
"""

query_products = """
SELECT AVG(price) as ortalama_fiyat
FROM products
"""

df_avg_age = pd.read_sql_query(query_customers, conn)
df_avg_price = pd.read_sql_query(query_products, conn)

print("=" * 80)
print("📊 ORTALAMA DEĞERLER ANALİZİ (AVG)")
print("=" * 80)
print(f"\n👥 Müşteri Profili:")
print(f"   Ortalama Yaş: {df_avg_age['ortalama_yas'][0]:.1f} yaş")

print(f"\n💰 Ürün Profili:")
print(f"   Ortalama Fiyat: {df_avg_price['ortalama_fiyat'][0]:,.0f} ₺")

# Yaş segmentasyonu
if df_avg_age['ortalama_yas'][0] < 30:
    print(f"   📊 Segment: Genç Müşteri Kitlesi")
elif df_avg_age['ortalama_yas'][0] < 40:
    print(f"   📊 Segment: Orta Yaş Grubu")
else:
    print(f"   📊 Segment: Olgun Müşteri Kitlesi")
````

### 📋 Örnek 4: Maksimum ve Minimum Değerler (MAX, MIN)

### 📋 Örnek 6: Şehir Bazlı Müşteri Dağılımı

````python
# Her şehirde kaç müşteri var?
query = """
SELECT 
    city as sehir,
    COUNT(*) as musteri_sayisi,
    AVG(age) as ortalama_yas
FROM customers
GROUP BY city
ORDER BY musteri_sayisi DESC
"""

df_city_distribution = pd.read_sql_query(query, conn)

print("=" * 80)
print("🏙️ ŞEHİR BAZLI MÜŞTERİ DAĞILIMI (GROUP BY)")
print("=" * 80)
display(df_city_distribution)

print(f"\n📊 İstatistikler:")
print(f"   • En fazla müşteri: {df_city_distribution.iloc[0]['sehir']} ({df_city_distribution.iloc[0]['musteri_sayisi']} kişi)")
print(f"   • Toplam şehir: {len(df_city_distribution)} şehir")
print(f"   • Ortalama müşteri/şehir: {df_city_distribution['musteri_sayisi'].mean():.1f} kişi")
````

### 📋 Örnek 7: Kategori Bazlı Ürün Analizi

````python
# Her kategoride kaç ürün var ve toplam değeri ne?
query = """
SELECT 
    category as kategori,
    COUNT(*) as urun_sayisi,
    SUM(stock) as toplam_stok,
    AVG(price) as ortalama_fiyat,
    MIN(price) as en_ucuz,
    MAX(price) as en_pahali,
    SUM(price * stock) as kategori_degeri
FROM products
GROUP BY category
ORDER BY kategori_degeri DESC
"""

df_category_analysis = pd.read_sql_query(query, conn)

print("=" * 80)
print("📦 KATEGORİ BAZLI ÜRÜN ANALİZİ (GROUP BY)")
print("=" * 80)
display(df_category_analysis)

print(f"\n📊 Kategori Performansı:")
for idx, row in df_category_analysis.iterrows():
    print(f"\n🏷️ {row['kategori']}:")
    print(f"   • Ürün Çeşidi: {row['urun_sayisi']} adet")
    print(f"   • Toplam Stok: {row['toplam_stok']} adet")
    print(f"   • Ortalama Fiyat: {row['ortalama_fiyat']:,.0f} ₺")
    print(f"   • Kategori Değeri: {row['kategori_degeri']:,.0f} ₺")
````

### 📋 Örnek 8: Şehir ve Yaş Segmenti Analizi

### 📋 Örnek 9: Yüksek Müşteri Sayısına Sahip Şehirler

````python
# Sadece 2'den fazla müşterisi olan şehirler
query = """
SELECT 
    city as sehir,
    COUNT(*) as musteri_sayisi,
    AVG(age) as ortalama_yas
FROM customers
GROUP BY city
HAVING COUNT(*) > 2
ORDER BY musteri_sayisi DESC
"""

df_high_customer_cities = pd.read_sql_query(query, conn)

print("=" * 80)
print("🏙️ YÜKSEK MÜŞTERİ POTANSİYELİNE SAHİP ŞEHİRLER (HAVING)")
print("   Kriterler: 2'den fazla müşteri")
print("=" * 80)
display(df_high_customer_cities)

print(f"\n✅ {len(df_high_customer_cities)} şehir bu kritere uyuyor")
print(f"💼 Bu şehirlere özel kampanyalar düzenlenebilir!")
````

### 📋 Örnek 10: Yüksek Değerli Ürün Kategorileri

````python
# Sadece toplam değeri 500.000 TL'den fazla olan kategoriler
query = """
SELECT 
    category as kategori,
    COUNT(*) as urun_sayisi,
    SUM(price * stock) as toplam_deger,
    AVG(price) as ortalama_fiyat
FROM products
GROUP BY category
HAVING SUM(price * stock) > 500000
ORDER BY toplam_deger DESC
"""

df_high_value_categories = pd.read_sql_query(query, conn)

print("=" * 80)
print("💎 YÜKSEK DEĞERLİ ÜRÜN KATEGORİLERİ (HAVING)")
print("   Kriterler: Toplam değer > 500.000 ₺")
print("=" * 80)
display(df_high_value_categories)

if len(df_high_value_categories) > 0:
    print(f"\n✅ {len(df_high_value_categories)} kategori bu kritere uyuyor")
    print(f"💰 Bu kategorilere özel odaklanma gerekli!")
    print(f"📊 Toplam değer: {df_high_value_categories['toplam_deger'].sum():,.0f} ₺")
else:
    print("\n⚠️ Kriterlere uyan kategori bulunamadı")
````

### 📋 Örnek 11: WHERE + GROUP BY + HAVING Kombinasyonu

````python
# Karmaşık analiz: 10.000 TL üzeri ürünlerin kategori analizi
# Sadece ortalama fiyatı 20.000 TL'den fazla olan kategoriler
query = """
SELECT 
    category as kategori,
    COUNT(*) as premium_urun_sayisi,
    AVG(price) as ortalama_fiyat,
    MIN(price) as en_ucuz,
    MAX(price) as en_pahali,
    SUM(stock) as toplam_stok
FROM products
WHERE price >= 10000
GROUP BY category
HAVING AVG(price) > 20000
ORDER BY ortalama_fiyat DESC
"""

df_premium_analysis = pd.read_sql_query(query, conn)

print("=" * 80)
print("💎 PREMİUM ÜRÜN KATEGORİLERİ - DETAYLI ANALİZ")
print("   WHERE: Fiyat >= 10.000 ₺")
print("   HAVING: Ortalama fiyat > 20.000 ₺")
print("=" * 80)
display(df_premium_analysis)

if len(df_premium_analysis) > 0:
    print(f"\n🎯 {len(df_premium_analysis)} premium kategori bulundu!")
    for idx, row in df_premium_analysis.iterrows():
        print(f"\n📦 {row['kategori']}:")
        print(f"   • Ürün Sayısı: {row['premium_urun_sayisi']}")
        print(f"   • Ortalama Fiyat: {row['ortalama_fiyat']:,.0f} ₺")
        print(f"   • Fiyat Aralığı: {row['en_ucuz']:,.0f} - {row['en_pahali']:,.0f} ₺")
else:
    print("\n💡 Premium kategoriler için asgari eşik yükseltilebilir!")
````

### 📋 Senaryo 1: Satış Performans Raporu (Management Dashboard)

````python
# YÖNETİCİ DASHBOARD: Sipariş Analizi
query = """
SELECT 
    o.order_id,
    c.first_name || ' ' || c.last_name as musteri,
    c.city,
    o.order_date,
    COUNT(oi.item_id) as urun_adedi,
    SUM(oi.quantity * oi.unit_price) as toplam_tutar,
    AVG(oi.unit_price) as ortalama_urun_fiyati
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
GROUP BY o.order_id, c.first_name, c.last_name, c.city, o.order_date
ORDER BY toplam_tutar DESC
LIMIT 10
"""

df_sales_report = pd.read_sql_query(query, conn)

print("=" * 80)
print("📊 SATIŞ PERFORMANS RAPORU - TOP 10 SİPARİŞ")
print("=" * 80)
display(df_sales_report)

toplam_ciro = df_sales_report['toplam_tutar'].sum()
print(f"\n💰 Top 10 Sipariş Cirosu: {toplam_ciro:,.0f} ₺")
print(f"📦 Ortalama Sepet Değeri: {df_sales_report['toplam_tutar'].mean():,.0f} ₺")
print(f"🎯 En Yüksek Sipariş: {df_sales_report.iloc[0]['musteri']} - {df_sales_report.iloc[0]['toplam_tutar']:,.0f} ₺")
````

### 📋 Senaryo 2: Müşteri Segmentasyonu (RFM Analizi)

````python
# MÜŞTERİ SEGMENTASYONU: Sipariş sayısı ve toplam harcama
query = """
SELECT 
    c.customer_id,
    c.first_name || ' ' || c.last_name as musteri,
    c.city,
    c.age,
    COUNT(DISTINCT o.order_id) as siparis_sayisi,
    SUM(oi.quantity * oi.unit_price) as toplam_harcama,
    AVG(oi.quantity * oi.unit_price) as ortalama_siparis_tutari,
    CASE 
        WHEN SUM(oi.quantity * oi.unit_price) > 100000 THEN 'VIP'
        WHEN SUM(oi.quantity * oi.unit_price) > 50000 THEN 'Premium'
        WHEN SUM(oi.quantity * oi.unit_price) > 20000 THEN 'Standart'
        ELSE 'Yeni Müşteri'
    END as musteri_segmenti
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
LEFT JOIN order_items oi ON o.order_id = oi.order_id
GROUP BY c.customer_id, c.first_name, c.last_name, c.city, c.age
ORDER BY toplam_harcama DESC
"""

df_customer_segments = pd.read_sql_query(query, conn)

print("=" * 80)
print("👥 MÜŞTERİ SEGMENTASYONU ANALİZİ")
print("=" * 80)
display(df_customer_segments)

print(f"\n📊 Segment Dağılımı:")
segment_counts = df_customer_segments['musteri_segmenti'].value_counts()
for segment, count in segment_counts.items():
    print(f"   • {segment}: {count} müşteri")

print(f"\n💰 En Değerli Müşteri: {df_customer_segments.iloc[0]['musteri']}")
print(f"   Harcama: {df_customer_segments.iloc[0]['toplam_harcama']:,.0f} ₺")
print(f"   Sipariş: {df_customer_segments.iloc[0]['siparis_sayisi']} adet")
````

### 📋 Senaryo 3: Ürün Satış Performansı (En Çok Satanlar)

````python
# EN ÇOK SATAN ÜRÜNLER ANALİZİ
query = """
SELECT 
    p.name as urun,
    p.category as kategori,
    p.price as liste_fiyati,
    COUNT(DISTINCT oi.order_id) as siparis_sayisi,
    SUM(oi.quantity) as satilan_adet,
    SUM(oi.quantity * oi.unit_price) as toplam_gelir,
    AVG(oi.unit_price) as ortalama_satis_fiyati,
    p.stock as guncel_stok
FROM products p
LEFT JOIN order_items oi ON p.product_id = oi.product_id
GROUP BY p.product_id, p.name, p.category, p.price, p.stock
HAVING SUM(oi.quantity) > 0
ORDER BY toplam_gelir DESC
LIMIT 10
"""

df_best_sellers = pd.read_sql_query(query, conn)

print("=" * 80)
print("🏆 EN ÇOK SATAN 10 ÜRÜN")
print("=" * 80)
display(df_best_sellers)

print(f"\n📊 Performans Özeti:")
print(f"   • Toplam Gelir: {df_best_sellers['toplam_gelir'].sum():,.0f} ₺")
print(f"   • Toplam Satış: {df_best_sellers['satilan_adet'].sum()} adet")
print(f"   • Ortalama Ürün Geliri: {df_best_sellers['toplam_gelir'].mean():,.0f} ₺")

print(f"\n🥇 1. Sıra: {df_best_sellers.iloc[0]['urun']}")
print(f"   Gelir: {df_best_sellers.iloc[0]['toplam_gelir']:,.0f} ₺")
print(f"   Satış: {df_best_sellers.iloc[0]['satilan_adet']} adet")
````

### 📋 Senaryo 4: Şehir Bazlı Satış Analizi (Coğrafi Performans)

````python
# COĞRAFİ SATIŞ PERFORMANSI
query = """
SELECT 
    c.city as sehir,
    COUNT(DISTINCT c.customer_id) as musteri_sayisi,
    COUNT(DISTINCT o.order_id) as siparis_sayisi,
    SUM(oi.quantity * oi.unit_price) as toplam_ciro,
    AVG(oi.quantity * oi.unit_price) as ortalama_siparis,
    ROUND(1.0 * COUNT(DISTINCT o.order_id) / COUNT(DISTINCT c.customer_id), 2) as musteri_basina_siparis
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
LEFT JOIN order_items oi ON o.order_id = oi.order_id
GROUP BY c.city
HAVING SUM(oi.quantity * oi.unit_price) > 0
ORDER BY toplam_ciro DESC
"""

df_city_performance = pd.read_sql_query(query, conn)

print("=" * 80)
print("🗺️ ŞEHİR BAZLI SATIŞ PERFORMANSI")
print("=" * 80)
display(df_city_performance)

print(f"\n📊 Coğrafi Performans Özeti:")
print(f"   • En Yüksek Ciro: {df_city_performance.iloc[0]['sehir']} - {df_city_performance.iloc[0]['toplam_ciro']:,.0f} ₺")
print(f"   • Toplam Ciro: {df_city_performance['toplam_ciro'].sum():,.0f} ₺")
print(f"   • En Aktif Şehir: {df_city_performance.iloc[0]['sehir']} ({df_city_performance.iloc[0]['siparis_sayisi']} sipariş)")

print(f"\n💡 Strateji Önerisi:")
top_city = df_city_performance.iloc[0]['sehir']
print(f"   • {top_city} en yüksek potansiyele sahip!")
print(f"   • Burada özel kampanyalar düzenlenebilir.")
````

### 📋 Senaryo 5: KPI Dashboard (Temel İş Metrikleri)

````python
# KPI DASHBOARD - Tüm Metrikler Bir Arada
print("=" * 80)
print("📊 YÖNETİCİ KPI DASHBOARD")
print("=" * 80)

# 1. Müşteri Metrikleri
query_customers = """
SELECT 
    COUNT(*) as toplam_musteri,
    COUNT(DISTINCT city) as sehir_sayisi,
    AVG(age) as ortalama_yas
FROM customers
"""
df_kpi_customers = pd.read_sql_query(query_customers, conn)

# 2. Ürün Metrikleri
query_products = """
SELECT 
    COUNT(*) as toplam_urun,
    COUNT(DISTINCT category) as kategori_sayisi,
    SUM(stock) as toplam_stok,
    SUM(price * stock) as envanter_degeri
FROM products
"""
df_kpi_products = pd.read_sql_query(query_products, conn)

# 3. Satış Metrikleri
query_sales = """
SELECT 
    COUNT(DISTINCT o.order_id) as toplam_siparis,
    SUM(oi.quantity) as toplam_satis_adedi,
    SUM(oi.quantity * oi.unit_price) as toplam_ciro,
    AVG(oi.quantity * oi.unit_price) as ortalama_siparis_tutari
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
"""
df_kpi_sales = pd.read_sql_query(query_sales, conn)

# KPI'ları göster
print("\n🔵 MÜŞTERİ KPI'LARI")
print("-" * 80)
print(f"   • Toplam Müşteri: {df_kpi_customers['toplam_musteri'][0]:,}")
print(f"   • Aktif Şehir: {df_kpi_customers['sehir_sayisi'][0]}")
print(f"   • Ortalama Yaş: {df_kpi_customers['ortalama_yas'][0]:.1f}")

print("\n🔵 ÜRÜN KPI'LARI")
print("-" * 80)
print(f"   • Ürün Çeşidi: {df_kpi_products['toplam_urun'][0]:,}")
print(f"   • Kategori Sayısı: {df_kpi_products['kategori_sayisi'][0]}")
print(f"   • Toplam Stok: {df_kpi_products['toplam_stok'][0]:,} adet")
print(f"   • Envanter Değeri: {df_kpi_products['envanter_degeri'][0]:,.0f} ₺")

print("\n🔵 SATIŞ KPI'LARI")
print("-" * 80)
print(f"   • Toplam Sipariş: {df_kpi_sales['toplam_siparis'][0]:,}")
print(f"   • Satılan Ürün: {df_kpi_sales['toplam_satis_adedi'][0]:,} adet")
print(f"   • Toplam Ciro: {df_kpi_sales['toplam_ciro'][0]:,.0f} ₺")
print(f"   • Ortalama Sepet: {df_kpi_sales['ortalama_siparis_tutari'][0]:,.0f} ₺")

# Hesaplanan Metrikler
conversion_rate = (df_kpi_sales['toplam_siparis'][0] / df_kpi_customers['toplam_musteri'][0]) * 100
print("\n🔵 HESAPLANAN METRİKLER")
print("-" * 80)
print(f"   • Dönüşüm Oranı: %{conversion_rate:.1f}")
print(f"   • Müşteri Başına Ortalama Ciro: {df_kpi_sales['toplam_ciro'][0] / df_kpi_customers['toplam_musteri'][0]:,.0f} ₺")
print(f"   • Stok Devir Hızı: {df_kpi_sales['toplam_satis_adedi'][0] / df_kpi_products['toplam_stok'][0]:.2%}")
````
