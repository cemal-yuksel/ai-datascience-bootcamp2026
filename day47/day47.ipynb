### 🔁 3.1 · Algoritma Adım Adım
$N$ örnek, $T$ tur, öğrenme hızı $\eta$ (`learning_rate`) olsun. Başlangıçta her örneğin ağırlığı $w_i = 1/N$.

Her **tur $t$** için:

1. **Ağırlıklı örnekleme:** $w_i$ olasılıklarıyla, iadeli $N$ örnek çek (yüksek ağırlıklı örnek, eğitim kümesinde **daha sık** görünür).
2. **Zayıf öğrenici:** bu örneklemde sığ bir regresyon ağacı $h_t(x)$ eğit (`sklearn` varsayılanı: `max_depth=3`).
3. **Hata:** tüm örnekler için $e_i = \lvert y_i - h_t(x_i) \rvert$; en büyük hata $D = \max_i e_i$ ile ölçekle: $\tilde{e}_i = e_i / D \in [0,1]$.
   Kayıp türü (`loss`):
   $$L_i = \tilde{e}_i \quad(\text{linear}) \qquad L_i = \tilde{e}_i^{\,2} \quad(\text{square}) \qquad L_i = 1 - e^{-\tilde{e}_i} \quad(\text{exponential})$$
4. **Ortalama kayıp ve güven katsayısı:** $\bar{L} = \sum_i w_i L_i$ ve $\beta_t = \dfrac{\bar{L}}{1-\bar{L}}$. Eğer $\bar{L} \ge 0{,}5$ ise model rastgeleden iyi değildir: **erken dur** (`n_estimators`'a ulaşılmasa bile).
5. **Ağırlık güncelle:** $w_i \leftarrow w_i \cdot \beta_t^{\,(1-L_i)\,\eta}$, sonra $\sum w_i = 1$ olacak şekilde normalize et. Hatası **büyük** olan örnek ($L_i$ yüksek) daha az küçülür; yani **göreli ağırlığı artar**.

Modelin **güven katsayısı** $\alpha_t = \eta \ln(1/\beta_t)$ olur: ortalama hatası küçük modelin sözü ağırdır.

**Tahmin** (ortalama değil!): $\hat{y}(x) = $ **ağırlıklı medyan** $\{ h_t(x) \}_{t=1}^{T}$, ağırlıklar $\alpha_t$.

### 📍 3.2 · Neden Ortalama Değil de Ağırlıklı Medyan?
Ortalama, **tek bir aykırı** ağacın tahminine duyarlıdır; medyan ise değildir. AdaBoost.R2 zor örneklere odaklandığı için bazı ağaçlar uç noktalara kayabilir; ağırlıklı medyan bu uçlara karşı **dayanıklı** bir birleştirme sağlar. Bedeli de vardır: tahminler, **ağaçların verdiği değerlerle sınırlı** kalır, yeni bir değer "üretilmez". (Bu özelliğin sonuçlarını Bölüm 12'de göreceğiz.)

<p align="center">
  <img src="assets/diyagram_agirlikli_medyan.svg" width="100%" alt="Ağırlıklı medyan ve ağırlıklı ortalama">
</p>

````python
def agirlikli_medyan(tahmin, agirlik):
    """Ağırlıkların toplamının yarısına ilk ulaşıldığı tahmin değeri (sklearn'in AdaBoostRegressor.predict kuralı)."""
    sira = np.argsort(tahmin)
    kdf = np.cumsum(np.asarray(agirlik)[sira])
    return np.asarray(tahmin)[sira][np.argmax(kdf >= 0.5 * kdf[-1])]


# Diyagramdaki beş ağaç: tahminler (×1 lakh) ve güven katsayıları
tahmin = np.array([1.2, 3.0, 4.4, 5.8, 9.2])
agirlik = np.array([1.0, 3.0, 2.2, 1.2, 0.5])
print(f"Ağırlıklı ORTALAMA : {np.average(tahmin, weights=agirlik):.2f} lakh")
print(f"Ağırlıklı MEDYAN   : {agirlikli_medyan(tahmin, agirlik):.2f} lakh")

# Son ağaç uç bir değere (92 lakh) kayarsa?
tahmin_aykiri = tahmin.copy(); tahmin_aykiri[-1] = 92.0
print(f"\n5. ağaç 9,2 → 92 lakh olursa:")
print(f"Ağırlıklı ORTALAMA : {np.average(tahmin_aykiri, weights=agirlik):.2f} lakh   (≈ {np.average(tahmin_aykiri, weights=agirlik) - np.average(tahmin, weights=agirlik):+.1f} kaydı)")
print(f"Ağırlıklı MEDYAN   : {agirlikli_medyan(tahmin_aykiri, agirlik):.2f} lakh   (değişmedi)")
````

Algoritmayı **kendimiz yazalım** ve `sklearn` ile karşılaştıralım. Veri, gürültülü ve **aykırı değerli** tek boyutlu bir problem: $y = \sin x + 0{,}35x + \varepsilon$; satırların %6'sı ±3 birim kaydırılmış.

````python
def f_gercek(x):
    return np.sin(x) + 0.35 * x


def oyuncak_veri(n=220, seed=3):
    r = np.random.default_rng(seed)
    x = np.sort(r.uniform(0, 10, n))
    y = f_gercek(x) + r.normal(0, 0.25, n)
    aykiri = r.choice(n, size=int(0.06 * n), replace=False)
    y[aykiri] += r.choice([-3.0, 3.0], size=aykiri.size)
    return x.reshape(-1, 1), y, aykiri


Xo, yo, aykiri_idx = oyuncak_veri()
Xg = np.linspace(0, 10, 400).reshape(-1, 1)             # değerlendirme ızgarası (gürültüsüz gerçek eğri bilinir)


def adaboost_r2_elle(X, y, T=50, derinlik=2, loss="linear", lr=1.0, seed=0):
    """Drucker (1997) AdaBoost.R2 — sklearn'in AdaBoostRegressor'ı ile aynı mantık."""
    rng = np.random.default_rng(seed)
    n = len(y)
    w = np.full(n, 1 / n)
    kayit = {"agaclar": [], "alfa": [], "L_ort": [], "w_baslangic": []}
    for t in range(T):
        idx = rng.choice(n, size=n, replace=True, p=w)                       # 1) ağırlıklı bootstrap
        agac = DecisionTreeRegressor(max_depth=derinlik, random_state=int(rng.integers(1e9))).fit(X[idx], y[idx])   # 2) zayıf öğrenici
        e = np.abs(y - agac.predict(X))
        D = e.max()
        kayit["w_baslangic"].append(w.copy())
        if D == 0:                                                           # kusursuz uyum: güven katsayısı 1, dur
            kayit["agaclar"].append(agac); kayit["alfa"].append(1.0); kayit["L_ort"].append(0.0)
            break
        e = e / D
        L = {"linear": e, "square": e ** 2, "exponential": 1 - np.exp(-e)}[loss]   # 3) kayıp
        L_ort = float(np.sum(w * L))                                         # 4) ortalama kayıp → β
        if L_ort >= 0.5:                                                     # rastgeleden kötü: bu ağacı alma, dur
            if t == 0:
                kayit["agaclar"].append(agac); kayit["alfa"].append(1.0); kayit["L_ort"].append(L_ort)
            break
        beta = L_ort / (1 - L_ort)
        kayit["agaclar"].append(agac); kayit["alfa"].append(lr * np.log(1 / beta)); kayit["L_ort"].append(L_ort)
        w = w * beta ** ((1 - L) * lr)                                       # 5) ağırlık güncelle
        w = w / w.sum()
    return kayit


def tahmin_elle(kayit, X, k=None, agac_tahminleri=None):
    """İlk k ağacın ağırlıklı medyanı."""
    k = k or len(kayit["agaclar"])
    P = agac_tahminleri[:, :k] if agac_tahminleri is not None else np.column_stack([a.predict(X) for a in kayit["agaclar"][:k]])
    a = np.asarray(kayit["alfa"][:k])
    sira = np.argsort(P, axis=1)
    Ps, kdf = np.take_along_axis(P, sira, axis=1), np.cumsum(a[sira], axis=1)
    return Ps[np.arange(len(P)), (kdf >= 0.5 * kdf[:, -1:]).argmax(axis=1)]


elle = adaboost_r2_elle(Xo, yo, T=100, derinlik=4, seed=0)
print(f"Elle kurulan model: {len(elle['agaclar'])} ağaç · ilk 5 güven katsayısı α = {np.round(elle['alfa'][:5], 2)} · ilk 5 ortalama kayıp L̄ = {np.round(elle['L_ort'][:5], 3)}")
````

### ✅ 4.1 · `sklearn` ile Doğrulama
İki şeyi sınıyoruz: **(1)** `sklearn`'in `predict`'i gerçekten ağırlıklı medyan mı? (Ağaçları ve `estimator_weights_` değerlerini alıp kendimiz hesaplıyoruz.) **(2)** Elle yazdığımız algoritma, `sklearn` ile benzer kalitede mi?

````python
sk = AdaBoostRegressor(DecisionTreeRegressor(max_depth=4), n_estimators=100, random_state=0).fit(Xo, yo)

# (1) sklearn tahmini == ağaçların ağırlıklı medyanı?
P = np.column_stack([a.predict(Xg) for a in sk.estimators_])
elle_sk = tahmin_elle({"agaclar": sk.estimators_, "alfa": list(sk.estimator_weights_)}, Xg, agac_tahminleri=P)
print("sklearn.predict == ağaçların ağırlıklı medyanı :", np.allclose(elle_sk, sk.predict(Xg)))
kul = sk.estimator_weights_ > 0                                  # erken durmada sondaki kullanılmayan ağaçların ağırlığı 0'dır
print("estimator_weights_ == learning_rate · ln((1−L̄)/L̄) :", np.allclose(sk.learning_rate * np.log((1 - sk.estimator_errors_[kul]) / sk.estimator_errors_[kul]), sk.estimator_weights_[kul]),
      f"(ilk beşi: {np.round(sk.estimator_weights_[:5], 2)})")

# (2) Gürültüsüz gerçek eğriye uzaklık
gercek = f_gercek(Xg.ravel())
tek_kutuk = DecisionTreeRegressor(max_depth=4, random_state=0).fit(Xo, yo)
print(f"\nGerçek eğriye MAE → tek ağaç (derinlik 4): {mean_absolute_error(gercek, tek_kutuk.predict(Xg)):.3f}"
      f" · elle AdaBoost.R2: {mean_absolute_error(gercek, tahmin_elle(elle, Xg)):.3f}"
      f" · sklearn AdaBoost: {mean_absolute_error(gercek, sk.predict(Xg)):.3f}")
````

### ✂️ 9.1 · Eğitim / Test Ayrımı
Regresyonda `stratify` doğrudan kullanılamaz; ama hedefi **fiyat onda birliklerine** bölüp onlara göre tabakalarsak, hem eğitim hem test kümesi **lüks dahil bütün fiyat aralıklarını** aynı oranda içerir.

````python
HEDEF = "selling_price"
X = veri.drop(columns=[HEDEF, "log_fiyat"])
y = veri[HEDEF]

fiyat_onlu = pd.qcut(y, 10, labels=False)
X_egitim, X_test, y_egitim, y_test = train_test_split(X, y, test_size=0.2, stratify=fiyat_onlu, random_state=RANDOM_STATE)

kontrol = pd.DataFrame({"eğitim": y_egitim.describe(percentiles=[.5, .9, .99]), "test": y_test.describe(percentiles=[.5, .9, .99])}).round(0)
display(kontrol)
print(f"Eğitim: {len(X_egitim):,} satır · Test: {len(X_test):,} satır · (test seti, final değerlendirmeye kadar kilitli)")
````

### 🔠 9.2 · Hangi Sütuna Ne Yapıyoruz?

| Sütun grubu | Dönüşüm | Neden? |
|:--|:--|:--|
| `vehicle_age`, `mileage`, `engine`, `max_power` | medyanla doldur (+ ölçekle*) | eksik yok ama güvenlik ağı; \*yalnızca ölçek-duyarlı modellerde |
| `seats` | **medyanla doldur** | 2 gizli eksik (`0`) |
| `km_driven` | **log1p** | sağa çarpık, 4 basamak ölçek |

| **`guc_litre`** *(yeni)* | `max_power / (engine/1000)` | motorun "spor" derecesi |
| `seller_type`, `fuel_type`, `transmission_type` | **OneHotEncoder** (`handle_unknown="ignore"`) | az sayıda sınıf |
| `brand` (32), `model` (120) | **`TargetEncoder`** | çok sınıflı; hedef ortalamasıyla sayıya çevirir, **çapraz uydurma** ile sızıntıyı önler |

<table><tr><td>

<img src="assets/etiket_kavram.svg" height="34" alt="Kavram">

**`TargetEncoder` nasıl sızıntıyı önler?** Her kategoriyi, o kategoriye ait **hedef ortalamasıyla** değiştirmek güçlüdür ama tehlikelidir: bir satırın kendi fiyatı, kendi özelliğinin kodlanmasına girerse model "cevabı" görmüş olur. `sklearn.preprocessing.TargetEncoder`, `fit_transform` sırasında veriyi katlara böler ve her satırı **kendisini görmeyen** katlardan hesaplanan ortalamayla kodlar; ayrıca **büzülme (`smooth`)** ile nadir kategorilerin ortalamasını genel ortalamaya doğru çeker. Referans notebook ise frekans kodlaması kullanmıştı: model adlarını *kaç kez göründüklerine* göre sayıya çevirmek **fiyat bilgisini taşımaz**; yalnızca popülerliği.

</td></tr></table>

````python
KATEGORIK_AZ = ["seller_type", "fuel_type", "transmission_type"]
KATEGORIK_COK = ["brand", "model"]
SAYISAL_TEMEL = ["vehicle_age", "km_driven", "mileage", "engine", "max_power", "seats"]
SAYISAL_YENI = ["km_yil", "guc_litre"]


def ozellik_uret(d):
    """Türetilmiş özellikler (satır bazlı, öğrenilecek bir şey yok → sızıntı riski yok)."""
    d = d.copy()
    d["km_yil"] = np.log1p(d["km_driven"] / (d["vehicle_age"] + 1))
    d["guc_litre"] = d["max_power"] / (d["engine"] / 1000)
    d["km_driven"] = np.log1p(d["km_driven"])
    return d


def log_km(d):
    d = d.copy()
    d["km_driven"] = np.log1p(d["km_driven"])
    return d


def ozellik_adlari(_, adlar):
    return np.r_[adlar, SAYISAL_YENI]


def on_isleme(olcekle=False, ozellik=True):
    """Ön işleme hattı. olcekle=True → sayısal ve kodlanmış sütunlar standartlaştırılır (doğrusal/mesafe tabanlı modeller için)."""
    sayisal = SAYISAL_TEMEL + (SAYISAL_YENI if ozellik else [])
    adim_sayisal = [("doldur", SimpleImputer(strategy="median"))] + ([("olcek", StandardScaler())] if olcekle else [])
    adim_kod = [("hedef_kodu", TargetEncoder(target_type="continuous", smooth="auto", random_state=RANDOM_STATE))] + ([("olcek", StandardScaler())] if olcekle else [])
    ct = ColumnTransformer([("sayisal", Pipeline(adim_sayisal), sayisal),
                            ("kategorik", OneHotEncoder(handle_unknown="ignore", sparse_output=False), KATEGORIK_AZ),
                            ("kodlu", Pipeline(adim_kod), KATEGORIK_COK)])
    if ozellik:
        ilk = ("ozellik", FunctionTransformer(ozellik_uret, feature_names_out=ozellik_adlari))
    else:
        ilk = ("log_km", FunctionTransformer(log_km, feature_names_out="one-to-one"))
    return Pipeline([ilk, ("ct", ct)])


def regresor_hatti(model, olcekle=False, log_hedef=True, ozellik=True):
    """Ön işleme + model; log_hedef=True ise model log1p(fiyat) üzerinde eğitilir, tahmin ₹'ye geri çevrilir."""
    hat = Pipeline([("hazirlik", on_isleme(olcekle, ozellik)), ("model", model)])
    return TransformedTargetRegressor(hat, func=np.log1p, inverse_func=np.expm1, check_inverse=False) if log_hedef else hat


hazirlik = on_isleme().fit(X_egitim, np.log1p(y_egitim))
Xt = pd.DataFrame(hazirlik.transform(X_egitim.head(8)), columns=[a.split("__")[-1] for a in hazirlik.get_feature_names_out()])
print(f"Modelin gördüğü tablo: {len(Xt.columns)} sütun")
Xt.round(2)
````

````python
cv5 = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)


def cv_skor(model, **kw):
    s = cross_validate(regresor_hatti(model, **kw), X_egitim, y_egitim, cv=cv5, scoring={"r2": "r2", "mae": "neg_mean_absolute_error"}, n_jobs=N_ISCI)
    return s["test_r2"].mean(), -s["test_mae"].mean()


adaylar = {"Ridge": lambda: Ridge(alpha=1.0), "KNN (k=10)": lambda: KNeighborsRegressor(10, weights="distance"),
           "Random Forest": lambda: RandomForestRegressor(100, n_jobs=1, random_state=RANDOM_STATE),
           "AdaBoost (derinlik 6)": lambda: AdaBoostRegressor(DecisionTreeRegressor(max_depth=6), n_estimators=100, random_state=RANDOM_STATE)}
olcek_sonuc = {(ad, olcekli): cv_skor(f(), olcekle=olcekli) for ad, f in adaylar.items() for olcekli in (False, True)}
olcek_tablo = pd.DataFrame([{"model": ad, "ölçekleme": "ölçekli" if o else "ölçeksiz", "CV R²": r2, "CV MAE (₹)": mae} for (ad, o), (r2, mae) in olcek_sonuc.items()])

fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
for ax, kolon, baslik in zip(axes, ["CV R²", "CV MAE (₹)"], ["CV R² (yüksek iyi)", "CV MAE ₹ (düşük iyi)"]):
    sns.barplot(data=olcek_tablo, x="model", y=kolon, hue="ölçekleme", palette=[RENK["gri"], RENK["orman"]], ax=ax)
    ax.set(title=baslik, xlabel="", ylabel="")
    ax.tick_params(axis="x", rotation=12)
    ax.legend(frameon=False, title="")
kaydet(fig, "11_olcekleme_deneyi")
plt.show()
olcek_tablo.to_csv(RAPOR_DIR / "olcekleme_deneyi.csv", index=False)
olcek_tablo.round(3)
````

### 🛡️ 9.6 · Deney: Hedef Kodlamada Sızıntı Gerçekten Fark Yaratıyor mu?
"Naif" yol: marka/model için **bütün eğitim verisinden** hedef ortalamasını hesapla, sonra çapraz doğrulama yap. Bu durumda her doğrulama satırının **kendi fiyatı** kodlamaya karışmıştır. Doğru yol: kodlamayı `Pipeline` içinde, katlar **içinde** öğrenmek.

````python
log_y_egitim = np.log1p(y_egitim)


def naif_tablo(df, ort_haritalari, genel):
    """Hedef ortalamasıyla kodlanmış tablo (haritalar dışarıdan verilir)."""
    d = ozellik_uret(df)
    for c in KATEGORIK_COK:
        d[c + "_kod"] = d[c].map(ort_haritalari[c]).fillna(genel)
    d = d.drop(columns=KATEGORIK_COK)
    return pd.get_dummies(d, columns=KATEGORIK_AZ, dtype=float).fillna(d["seats"].median())


haritalar = {c: log_y_egitim.groupby(X_egitim[c]).mean() for c in KATEGORIK_COK}     # TÜM eğitim verisinden (sızıntılı!)
Xn_egitim = naif_tablo(X_egitim, haritalar, log_y_egitim.mean())
Xn_test = naif_tablo(X_test, haritalar, log_y_egitim.mean()).reindex(columns=Xn_egitim.columns, fill_value=0.0)

rf = lambda: RandomForestRegressor(100, n_jobs=N_ISCI, random_state=RANDOM_STATE)
cv_naif = cross_validate(rf(), Xn_egitim, log_y_egitim, cv=cv5, scoring="r2")["test_score"].mean()
cv_dogru = cross_validate(regresor_hatti(rf(), log_hedef=False), X_egitim, log_y_egitim, cv=cv5, scoring="r2")["test_score"].mean()    # log fiyat üzerinde, naifle aynı ölçekte
test_naif = r2_score(np.log1p(y_test), rf().fit(Xn_egitim, log_y_egitim).predict(Xn_test))
test_dogru = r2_score(np.log1p(y_test), np.log1p(regresor_hatti(rf()).fit(X_egitim, y_egitim).predict(X_test)))

sizinti = pd.DataFrame({"naif (sızıntılı) kodlama": [cv_naif, test_naif], "pipeline içinde TargetEncoder": [cv_dogru, test_dogru]}, index=["CV R² (log fiyat)", "test R² (log fiyat)"])
sizinti.loc["CV – test farkı"] = sizinti.loc["CV R² (log fiyat)"] - sizinti.loc["test R² (log fiyat)"]
sizinti.to_csv(RAPOR_DIR / "sizinti_deneyi.csv", index_label="ölçüt")
display(sizinti.round(4))
````

### 🏟️ 10.1 · Yarışı Koşturalım
Her model için: **5 katlı CV** (katlar içinde eğitim → doğrulama; her satır tam bir kez doğrulama satırı olur → *out-of-fold* tahmin), sonra bütün eğitim kümesiyle yeniden eğitim ve **test** tahmini. Ölçüler: `MAE`, `RMSE`, `R²` (₹ cinsinden) ve `R²`, `MdAPE` (log/yüzde cinsinden).

````python
SATIRLAR, FOLD_SATIRLARI, OOF, TEST_TAHMIN, EGITILMIS = [], [], {}, {}, {}
log_y_egitim = np.log1p(y_egitim)


def yaris_kos(ad, aile, fabrika, olcek):
    oof = np.zeros(len(y_egitim))
    fold_mae, fold_r2l, egitim_mae, sure = [], [], [], []
    for k, (tr, va) in enumerate(cv5.split(X_egitim)):
        t0 = time.time()
        m = regresor_hatti(fabrika(), olcekle=olcek).fit(X_egitim.iloc[tr], y_egitim.iloc[tr])
        sure.append(time.time() - t0)
        p = m.predict(X_egitim.iloc[va])
        oof[va] = p
        fold_mae.append(mean_absolute_error(y_egitim.iloc[va], p))
        fold_r2l.append(r2_score(np.log1p(y_egitim.iloc[va]), np.log1p(p)))
        egitim_mae.append(mean_absolute_error(y_egitim.iloc[tr], m.predict(X_egitim.iloc[tr])))
        FOLD_SATIRLARI.append({"model": ad, "kat": k + 1, "MAE (₹)": fold_mae[-1], "R² (log fiyat)": fold_r2l[-1]})
    son = regresor_hatti(fabrika(), olcekle=olcek).fit(X_egitim, y_egitim)
    tahmin = son.predict(X_test)
    OOF[ad], TEST_TAHMIN[ad], EGITILMIS[ad] = oof, tahmin, son
    cv_m, test_m = metrikler(y_egitim, oof), metrikler(y_test, tahmin)
    SATIRLAR.append({"model": ad, "aile": aile, "eğitim süresi (sn)": float(np.mean(sure)),
                     "CV MAE (₹)": cv_m["MAE (₹)"], "CV MAE std": float(np.std(fold_mae)), "CV R²": cv_m["R²"], "CV R² (log fiyat)": cv_m["R² (log fiyat)"],
                     "CV MdAPE (%)": cv_m["MdAPE (%)"], "eğitim MAE (₹)": float(np.mean(egitim_mae)),
                     **{f"test {k}": v for k, v in test_m.items()}})


t_yaris = time.time()
for ad, (aile, fabrika, olcek) in {TABAN[0]: TABAN[1], **ZOO}.items():
    yaris_kos(ad, aile, fabrika, olcek)
    s = SATIRLAR[-1]
    print(f"{ad:<24} CV MAE ₹{s['CV MAE (₹)']:>9,.0f} · test MAE ₹{s['test MAE (₹)']:>9,.0f} · test R² (log) {s['test R² (log fiyat)']:.3f} · {s['eğitim süresi (sn)']:.1f} sn/kat")
print(f"\nToplam yarış süresi: {time.time() - t_yaris:.0f} sn")
````

````python
yaris = pd.DataFrame(SATIRLAR).set_index("model")
yaris["CV sırası"] = yaris["CV MAE (₹)"].rank().astype(int)
yaris["test sırası"] = yaris["test MAE (₹)"].rank().astype(int)
yaris.round(4).to_csv(RAPOR_DIR / "model_karsilastirma.csv")
pd.DataFrame(FOLD_SATIRLARI).round(4).to_csv(RAPOR_DIR / "cv_fold_skorlari.csv", index=False)

gorunum = yaris.sort_values("CV MAE (₹)")[["aile", "CV MAE (₹)", "CV R² (log fiyat)", "CV MdAPE (%)", "test MAE (₹)", "test R² (log fiyat)", "test MdAPE (%)", "test R²", "eğitim süresi (sn)", "CV sırası", "test sırası"]]
gorunum.round(3)
````

### 🎲 11.2 · Fark Gerçek mi, Şans mı?
Beş katlık CV ile birkaç yüz ₹'lik farklar **yorumlanamaz**; ama **aynı satırlar üzerinde** yapılan eşleştirilmiş karşılaştırma daha güçlüdür. En iyi modelin *out-of-fold* mutlak hatası ile diğer modellerinki arasındaki **ortalama farkı**, satırları yeniden örnekleyerek (**bootstrap**, 1.000 tekrar) güven aralığıyla gösteriyoruz. Aralık **0'ı kapsıyorsa** fark şanstan ayırt edilemez.

````python
en_iyi_cv = d["CV MAE (₹)"].idxmin()
rng = np.random.default_rng(RANDOM_STATE)
hata_ref = np.abs(OOF[en_iyi_cv] - y_egitim.to_numpy())
idx_boot = rng.integers(0, len(y_egitim), size=(1000, len(y_egitim)))
fark_satirlari = []
for ad in d.sort_values("CV MAE (₹)").index:
    if ad == en_iyi_cv:
        continue
    fark = np.abs(OOF[ad] - y_egitim.to_numpy()) - hata_ref                # >0 → rakip daha kötü
    boot = fark[idx_boot].mean(axis=1)
    fark_satirlari.append({"model": ad, "aile": d.loc[ad, "aile"], "ΔMAE (₹)": fark.mean(), "alt (%2,5)": np.percentile(boot, 2.5), "üst (%97,5)": np.percentile(boot, 97.5)})
fark_tablo = pd.DataFrame(fark_satirlari).set_index("model")
fark_tablo["fark anlamlı mı?"] = np.where(fark_tablo["alt (%2,5)"] > 0, "evet — rakip daha kötü", "ayırt edilemez")
fark_tablo.round(0).to_csv(RAPOR_DIR / "eslestirilmis_fark.csv")

fig, ax = plt.subplots(figsize=(11, 8))
f = fark_tablo.iloc[::-1]
ax.hlines(range(len(f)), f["alt (%2,5)"], f["üst (%97,5)"], color=[AILE_RENK[a] for a in f["aile"]], lw=4, alpha=0.9)
ax.scatter(f["ΔMAE (₹)"], range(len(f)), color=VURGU, s=28, zorder=3)
ax.axvline(0, color="black", lw=1)
ax.set_yticks(range(len(f)), f.index)
ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v / 1e3:,.0f}K"))
ax.set(xlabel=f"MAE farkı (₹): rakip − {en_iyi_cv}   (sağda = rakip daha kötü)", title=f"{en_iyi_cv} ile eşleştirilmiş karşılaştırma (%95 bootstrap aralığı)")
kaydet(fig, "16_eslestirilmis_fark")
plt.show()
print(f"En iyi (CV MAE): {en_iyi_cv} · ondan istatistiksel olarak ayırt edilemeyen model sayısı: {(fark_tablo['fark anlamlı mı?'] == 'ayırt edilemez').sum()}")
````

### 🗺️ 11.3 · Hata Nerede? Fiyat Dilimine Göre Yüzde Hata ve Yanlılık
Genel MAE hatanın **nerede** yoğunlaştığını gizler. Test kümesini tahmin edilen fiyata göre dört dilime bölüp her model için **medyan yüzde hata (MdAPE)** ve **yanlılık** (medyan işaretli yüzde hata; negatif = düşük tahmin) görelim.

````python
dilim = pd.qcut(y_test, 4, labels=["Ç1 ucuz", "Ç2 orta-alt", "Ç3 orta-üst", "Ç4 pahalı"])
mdape_tab, yanlilik_tab = {}, {}
for ad in yaris.index.drop(TABAN[0]):
    oran = (TEST_TAHMIN[ad] - y_test.to_numpy()) / y_test.to_numpy()
    mdape_tab[ad] = pd.Series(np.abs(oran) * 100).groupby(dilim.to_numpy(), observed=True).median()
    yanlilik_tab[ad] = pd.Series(oran * 100).groupby(dilim.to_numpy(), observed=True).median()
mdape_tab, yanlilik_tab = pd.DataFrame(mdape_tab).T, pd.DataFrame(yanlilik_tab).T
sira = yaris.drop(index=TABAN[0]).sort_values("test MAE (₹)").index
mdape_tab, yanlilik_tab = mdape_tab.loc[sira], yanlilik_tab.loc[sira]
pd.concat([mdape_tab.add_prefix("MdAPE (%) · "), yanlilik_tab.add_prefix("yanlılık (%) · ")], axis=1).round(2).to_csv(RAPOR_DIR / "fiyat_dilimine_gore_hata.csv", index_label="model")

fig, axes = plt.subplots(1, 2, figsize=(15, 8.4), sharey=True)
sns.heatmap(mdape_tab, annot=True, fmt=".1f", cmap="YlOrRd", cbar_kws={"label": "MdAPE (%)", "shrink": 0.7}, linewidths=0.4, ax=axes[0])
axes[0].set(title="Medyan mutlak yüzde hata (MdAPE %)", ylabel="")
sns.heatmap(yanlilik_tab, annot=True, fmt=".1f", cmap="RdBu", center=0, cbar_kws={"label": "yanlılık (%) · mavi = fazla, kırmızı = eksik tahmin", "shrink": 0.7}, linewidths=0.4, ax=axes[1])
axes[1].set(title="Yanlılık: medyan işaretli yüzde hata (%)", ylabel="")
for ax in axes:
    ax.tick_params(axis="x", rotation=15)
kaydet(fig, "17_fiyat_dilimine_gore_hata")
plt.show()
````

AdaBoost'un iki yapılandırmasının yarıştaki yeri sonuç tablosunda. Burada asıl soru **neden?** Üç şüphe var:

1. **Zayıf öğrenici çok zayıf:** varsayılan `max_depth=3`, 12 bin satır ve 20 özellik için yetersiz olabilir (**bias**).

3. **Birleştirme biçimi:** tahminler **ağaçların değerleriyle** sınırlı kalan ağırlıklı medyan; yumuşak bir ortalama değil.

Bunları `staged_predict`, `estimator_errors_` ve `estimator_weights_` ile **tek tek** sınayalım. (Bu bölümde ön işleme hattını bir kez uyduruyoruz ve AdaBoost'u doğrudan dönüştürülmüş matris üzerinde eğitiyoruz; böylece tur tur tahmin alabiliriz.)

### 📈 12.1 · Tur Sayısı ve Zayıf Öğrenici Derinliği

````python
hz = on_isleme()
ly_egitim, ly_test = np.log1p(y_egitim), np.log1p(y_test)
Xt_egitim = hz.fit_transform(X_egitim, ly_egitim)                       # TargetEncoder: çapraz uydurmalı
Xt_test = hz.transform(X_test)
ozellik_adlari_ = [a.split("__")[-1] for a in hz.get_feature_names_out()]
print(f"Dönüşmüş eğitim matrisi: {Xt_egitim.shape} · test: {Xt_test.shape}")


def asamali_egri(model):
    """Her turdaki test ve eğitim MAE (₹) değerleri."""
    model.fit(Xt_egitim, ly_egitim)
    test = [mean_absolute_error(y_test, np.expm1(p)) for p in model.staged_predict(Xt_test)]
    egit = [mean_absolute_error(y_egitim, np.expm1(p)) for p in model.staged_predict(Xt_egitim)]
    return model, np.array(egit), np.array(test)


T_MAX = 100 * 3
egriler = {}
for derinlik in (1, 3, 6, 10):
    egriler[derinlik] = asamali_egri(AdaBoostRegressor(DecisionTreeRegressor(max_depth=derinlik), n_estimators=T_MAX, random_state=RANDOM_STATE))
    m = egriler[derinlik][0]
    print(f"derinlik {derinlik:>2}: kullanılan ağaç {(m.estimator_weights_ > 0).sum():>3}/{T_MAX} · son tur test MAE ₹{egriler[derinlik][2][-1]:,.0f} · en iyi tur {egriler[derinlik][2].argmin() + 1} (₹{egriler[derinlik][2].min():,.0f})")

gb_ref = GradientBoostingRegressor(n_estimators=T_MAX, random_state=RANDOM_STATE).fit(Xt_egitim, ly_egitim)         # karşılaştırma için gradyan artırma
gb_egri = np.array([mean_absolute_error(y_test, np.expm1(p)) for p in gb_ref.staged_predict(Xt_test)])

fig, axes = plt.subplots(1, 2, figsize=(16, 5.4))
renk_d = {1: "#D1495B", 3: "#EDAE49", 6: "#66A182", 10: "#00798C"}
for derinlik, (m, egit, test) in egriler.items():
    axes[0].plot(np.arange(1, len(test) + 1), test, color=renk_d[derinlik], lw=2.3, label=f"AdaBoost · derinlik {derinlik}")
axes[0].plot(np.arange(1, len(gb_egri) + 1), gb_egri, color=VURGU, ls="--", lw=1.8, label="GradientBoosting (derinlik 3)")
axes[0].set(xlabel="tur sayısı", ylabel="test MAE (₹)", title="Tur sayısına göre test hatası", ylim=(0, None))
axes[0].yaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v / 1e3:,.0f}K"))
axes[0].legend(frameon=False)
m3, egit3, test3 = egriler[3]
axes[1].plot(np.arange(1, len(test3) + 1), test3, color=renk_d[3], lw=2.3, label="test")
axes[1].plot(np.arange(1, len(egit3) + 1), egit3, color=renk_d[3], ls="--", lw=2, label="eğitim")
axes[1].set(xlabel="tur sayısı", ylabel="MAE (₹)", title="Derinlik 3: eğitim ≈ test → yetersiz öğrenme", ylim=(0, None))
axes[1].yaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v / 1e3:,.0f}K"))
axes[1].legend(frameon=False)
kaydet(fig, "19_asamali_egriler")
plt.show()
asamali = pd.DataFrame({**{f"AdaBoost derinlik {k}": pd.Series(v[2]) for k, v in egriler.items()}, "GradientBoosting": pd.Series(gb_egri)})
asamali.index = asamali.index + 1
asamali.loc[[i for i in (1, 5, 10, 25, 50, 100, 200, T_MAX) if i <= T_MAX]].round(0).rename_axis("tur").to_csv(RAPOR_DIR / "asamali_test_mae.csv")
````

### 🔀 14.1 · Özellik Önemi: Ağaç İçi (MDI) mi, Permütasyon mu?
**MDI** (*mean decrease in impurity*): ağaçların bölünmelerde sağladığı hata azalması; **eğitim** verisinden hesaplanır ve çok seviyeli sürekli özellikleri kayırır. **Permütasyon önemi:** test kümesinde tek bir sütunu **karıştırıp** hatanın ne kadar arttığına bakar; modelden bağımsızdır ve gerçek performansa dayanır. İkisini yan yana koyup sıralamanın tutarlı olup olmadığına bakıyoruz.

````python
FINAL_AD = ayar["CV MAE sonrası (₹)"].idxmin()                    # SEÇİM KURALI: çapraz doğrulama MAE'si en düşük ayarlı model (test kümesine bakılmaz)
ada_ayarli, final_model = ARAMA["AdaBoost"].best_estimator_, ARAMA[FINAL_AD].best_estimator_
print(f"Çapraz doğrulamada en iyi ayarlı model: {FINAL_AD}  (CV MAE ₹{ayar.loc[FINAL_AD, 'CV MAE sonrası (₹)']:,.0f})")

# — MDI: ayarlı AdaBoost'un feature_importances_ değerlerini özgün sütunlara topla
ada_ic = ada_ayarli.regressor_.named_steps["model"]
hz_ada = ada_ayarli.regressor_.named_steps["hazirlik"]
adlar_ada = [a.split("__")[-1] for a in hz_ada.get_feature_names_out()]


def ozgun(ad):
    for g in KATEGORIK_AZ:
        if ad.startswith(g):
            return g
    return ad


mdi = pd.Series(ada_ic.feature_importances_, index=adlar_ada).groupby(ozgun).sum().sort_values(ascending=False)

# — Permütasyon önemi (test kümesi, özgün sütunlar, ölçüt: MAE artışı)
def perm(model):
    r = permutation_importance(model, X_test, y_test, scoring="neg_mean_absolute_error", n_repeats=10, random_state=RANDOM_STATE, n_jobs=N_ISCI)
    return pd.DataFrame({"artış": r.importances_mean, "std": r.importances_std}, index=X_test.columns).sort_values("artış", ascending=False)


perm_ada, perm_final = perm(ada_ayarli), perm(final_model)
onem = pd.concat({"AdaBoost MDI (pay)": mdi, "AdaBoost permütasyon (₹)": perm_ada["artış"], f"{FINAL_AD} permütasyon (₹)": perm_final["artış"]}, axis=1)
onem.round(4).to_csv(RAPOR_DIR / "ozellik_onemleri.csv", index_label="özellik")

fig, axes = plt.subplots(1, 3, figsize=(19, 5.4))
mdi_s = mdi.sort_values()
axes[0].barh(mdi_s.index, mdi_s.values * 100, color=RENK["ada"], edgecolor="white")
axes[0].set(title="AdaBoost · MDI (eğitim)", xlabel="önem payı (%)")
for ax, tablo, baslik, renk in ((axes[1], perm_ada, "AdaBoost · permütasyon (test)", RENK["ada_koyu"]), (axes[2], perm_final, f"{FINAL_AD} · permütasyon (test)", RENK["orman"])):
    t = tablo.sort_values("artış")
    ax.barh(t.index, t["artış"], xerr=t["std"], color=renk, edgecolor="white", error_kw={"lw": 1})
    ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v / 1e3:,.0f}K"))
    ax.set(title=baslik, xlabel="MAE artışı (₹), sütun karıştırılınca")
kaydet(fig, "26_ozellik_onemi")
plt.show()
rho_onem = stats.spearmanr(onem.dropna()["AdaBoost MDI (pay)"], onem.dropna()["AdaBoost permütasyon (₹)"]).statistic
print(f"AdaBoost için MDI ↔ permütasyon sıra korelasyonu (Spearman): {rho_onem:.2f}")
````

### 🌐 15.3 · Belirsizlik: Tek Fiyat Değil, Fiyat Bandı
Satıcıya "₹6,2 lakh" demek yerine **"₹5,5–7,1 lakh arası"** demek hem dürüst hem faydalıdır. Modelimiz nokta tahmini verir; bandı **ayrık (split) konformal** mantıkla kuruyoruz:

1. Eğitim kümesinde **katlar-dışı (out-of-fold)** tahminleri al; model o satırı hiç görmemiştir.
2. Her satır için **log oranı** $z_i = \ln(y_i / \hat{y}_i)$'nin %5 ve %95 niceliklerini bul.
3. Yeni ilan için bant: $[\hat{y}\,e^{q_{5}},\ \hat{y}\,e^{q_{95}}]$ → **%90 kapsama** hedefi.

**Neden log oran?** Hata fiyatla orantılı büyüyor; sabit ₹ genişliğinde bir bant ucuz araçta fazla geniş, pahalıda fazla dar olurdu. **Tek bir global bant** yerine tahmin edilen fiyat üçte birliklerine göre **ayrı bantlar** (Mondrian) kuruyor ve kapsamayı karşılaştırıyoruz.

````python
oof_final = cross_val_predict(clone(final_model), X_egitim, y_egitim, cv=cv5, n_jobs=N_ISCI)
z_oof = np.log(y_egitim.to_numpy() / oof_final)
ALFA = 0.10
q_global = np.quantile(z_oof, [ALFA / 2, 1 - ALFA / 2])

kesim = np.quantile(oof_final, [1 / 3, 2 / 3])                                  # tahmin edilen fiyat üçte birlikleri
bin_oof, bin_test = np.digitize(oof_final, kesim), np.digitize(tahmin_test, kesim)
q_bin = {b: np.quantile(z_oof[bin_oof == b], [ALFA / 2, 1 - ALFA / 2]) for b in range(3)}

alt_g, ust_g = tahmin_test * np.exp(q_global[0]), tahmin_test * np.exp(q_global[1])
alt_b = tahmin_test * np.exp(np.array([q_bin[b][0] for b in bin_test]))
ust_b = tahmin_test * np.exp(np.array([q_bin[b][1] for b in bin_test]))
y_np = y_test.to_numpy()
ici = lambda a, u: (y_np >= a) & (y_np <= u)

bin_ad = {0: "ucuz üçte biri", 1: "orta üçte biri", 2: "pahalı üçte biri"}
satirlar = [{"yöntem": "global bant", "dilim": "tümü", "kapsama (%)": ici(alt_g, ust_g).mean() * 100, "medyan genişlik (% tahmin)": np.median((ust_g - alt_g) / tahmin_test) * 100},
            {"yöntem": "dilim bazlı (Mondrian)", "dilim": "tümü", "kapsama (%)": ici(alt_b, ust_b).mean() * 100, "medyan genişlik (% tahmin)": np.median((ust_b - alt_b) / tahmin_test) * 100}]
for b in range(3):
    m = bin_test == b
    satirlar.append({"yöntem": "global bant", "dilim": bin_ad[b], "kapsama (%)": ici(alt_g, ust_g)[m].mean() * 100, "medyan genişlik (% tahmin)": np.median(((ust_g - alt_g) / tahmin_test)[m]) * 100})
    satirlar.append({"yöntem": "dilim bazlı (Mondrian)", "dilim": bin_ad[b], "kapsama (%)": ici(alt_b, ust_b)[m].mean() * 100, "medyan genişlik (% tahmin)": np.median(((ust_b - alt_b) / tahmin_test)[m]) * 100})
aralik = pd.DataFrame(satirlar)
aralik.round(2).to_csv(RAPOR_DIR / "tahmin_araligi.csv", index=False)
display(aralik.round(1))

fig, axes = plt.subplots(1, 2, figsize=(17, 5.4), gridspec_kw={"width_ratios": [1, 1.6]})
g = aralik[aralik["dilim"] != "tümü"]
sns.barplot(data=g, x="dilim", y="kapsama (%)", hue="yöntem", palette=[RENK["gri"], RENK["orman"]], ax=axes[0])
axes[0].axhline(90, color=RENK["agac"], ls="--", lw=1.4, label="hedef %90")
axes[0].set(ylim=(70, 100), title="Kapsama: bant gerçek fiyatı yakalıyor mu?", xlabel="", ylabel="kapsama (%)")
axes[0].legend(frameon=False, fontsize=9, loc="upper center", bbox_to_anchor=(0.5, -0.12), ncol=3)

ornek = np.random.default_rng(RANDOM_STATE).choice(len(y_np), 45, replace=False)
ornek = ornek[np.argsort(tahmin_test[ornek])]
disinda = ~ici(alt_b, ust_b)[ornek]
x_ = np.arange(len(ornek))
axes[1].vlines(x_, alt_b[ornek], ust_b[ornek], color=RENK["orman"], lw=5, alpha=0.5)
axes[1].scatter(x_[~disinda], y_np[ornek][~disinda], color=VURGU, s=26, zorder=3, label="gerçek fiyat (bant içinde)")
axes[1].scatter(x_[disinda], y_np[ornek][disinda], color=RENK["agac"], s=46, marker="X", zorder=3, label="gerçek fiyat (bant dışında)")
axes[1].set(yscale="log", title="45 rastgele test ilanı: %90 fiyat bandı (tahmine göre sıralı)", xlabel="ilanlar (ucuzdan pahalıya)", ylabel="₹ lakh (log)")
axes[1].set_yticks([1e5, 3e5, 1e6, 3e6, 1e7], [f"{t / 1e5:g}" for t in [1e5, 3e5, 1e6, 3e6, 1e7]]); axes[1].minorticks_off()
axes[1].legend(frameon=False, loc="upper left")
kaydet(fig, "30_tahmin_araligi")
plt.show()
````

### 💸 15.4 · İş Değeri: Hata **Simetrik Değilse** Önerilen Fiyat da Simetrik Olmamalı
Fiyat önerisinde iki hata türü vardır ve **maliyetleri aynı değildir**:

- **Eksik fiyat (satıcı fazla ucuza koyar):** satıcı **para kaybeder** (kaçan gelir).
- **Fazla fiyat (satıcı pahalı koyar):** araç **satılmaz**; bekleme, indirim ve platformdan soğuma maliyeti doğar.

Birim maliyetler $c_u$ (eksik) ve $c_o$ (fazla) ise, beklenen maliyeti en aza indiren önerilen fiyat $m \cdot \hat{y}$ için $m$, gerçek/tahmin oranı $y/\hat{y}$'nin $\tau = \frac{c_u}{c_u + c_o}$ **niceliğidir** (gazete bayii / *newsvendor* problemi). $c_o > c_u$ ise $\tau < 0{,}5$: önerilen fiyat tahminin **biraz altına** çekilir.

<table><tr><td>

<img src="assets/etiket_dikkat.svg" height="34" alt="Dikkat">

**Varsayım uyarısı:** Aşağıdaki maliyet oranları **örnektir**; veri setinde gerçek bekleme/indirim maliyeti yok. Gerçek bir platformda bunlar satış süresi, komisyon ve müşteri kaybı verisinden tahmin edilmelidir. Ölçüt: model tahmininin yüzdesi olarak **araç başına ortalama maliyet**.

````python
oran_oof = y_egitim.to_numpy() / oof_final                                   # gerçek / tahmin (katlar-dışı)
oran_test = y_np / tahmin_test
senaryolar = {"Simetrik (cᵤ = cₒ)": (1.0, 1.0), "Bekleme pahalı (cₒ = 2·cᵤ)": (1.0, 2.0), "Bekleme çok pahalı (cₒ = 4·cᵤ)": (1.0, 4.0), "Kaçan gelir pahalı (cᵤ = 2·cₒ)": (2.0, 1.0)}


def maliyet(m, cu, co, oran=oran_test):
    """Önerilen fiyat = m · tahmin. Maliyet, model tahmininin yüzdesi olarak araç başına ortalama (pinball / asimetrik kayıp)."""
    return np.mean(cu * np.maximum(oran - m, 0) + co * np.maximum(m - oran, 0)) * 100


sonuc_senaryo = []
for ad, (cu, co) in senaryolar.items():
    tau = cu / (cu + co)
    m_opt = float(np.quantile(oran_oof, tau))
    sonuc_senaryo.append({"senaryo": ad, "τ = cᵤ/(cᵤ+cₒ)": tau, "önerilen çarpan m*": m_opt, "maliyet · m=1 (%)": maliyet(1.0, cu, co), "maliyet · m* (%)": maliyet(m_opt, cu, co)})
senaryo = pd.DataFrame(sonuc_senaryo).set_index("senaryo")
senaryo["tasarruf (%)"] = (1 - senaryo["maliyet · m* (%)"] / senaryo["maliyet · m=1 (%)"]) * 100
senaryo.round(3).to_csv(RAPOR_DIR / "fiyat_politikasi_senaryolari.csv")
display(senaryo.round(3))

ms = np.linspace(0.75, 1.25, 200)
fig, ax = plt.subplots(figsize=(11.5, 5.2))
for (ad, (cu, co)), renk in zip(list(senaryolar.items())[:3], [RENK["mavi"], RENK["ada"], RENK["agac"]]):
    egri = [maliyet(m, cu, co) for m in ms]
    ax.plot(ms, egri, color=renk, lw=2.4, label=ad)
    m_opt = senaryo.loc[ad, "önerilen çarpan m*"]
    ax.scatter([m_opt], [maliyet(m_opt, cu, co)], color=renk, s=110, edgecolor="black", zorder=4)
ax.axvline(1.0, color=VURGU, ls=":", lw=1.4)
ax.text(1.003, ax.get_ylim()[1] * 0.97, "m = 1 (tahmini aynen öner)", fontsize=9, color=VURGU, va="top")
ax.set(xlabel="önerilen fiyat / model tahmini  (m)", ylabel="araç başına ortalama maliyet (tahminin %'si)", title="Asimetrik maliyet altında en iyi fiyat çarpanı (● = en iyi m*)")
ax.legend(frameon=False)
kaydet(fig, "31_fiyat_politikasi")
plt.show()
````

- **Simetrik maliyette** (cᵤ = cₒ) en iyi çarpan **m\* ≈ 1,00**: tahmini aynen önermek zaten optimal; kazanç yok.
- **Bekleme pahalıysa** (cₒ = 2·cᵤ) m\* = **0,941**: tahminin ≈ %6 altında fiyat önermek ortalama maliyeti (tahminin yüzdesi olarak) %18,2'den %16,7'e (**%8 azalma**) indiriyor. Bekleme **çok pahalıysa** (cₒ = 4·cᵤ) m\* = **0,887**, tasarruf **%28**. Kaçan gelir daha pahalıysa (cᵤ = 2·cₒ) çarpan **1,061**: tahminin biraz **üstünde** fiyat önermek mantıklı.
- Ders: "en iyi tahmin" ile "en iyi **karar**" aynı şey değil. Model aynı kalırken yalnızca maliyet yapısını hesaba katan **bir çarpan**, bu varsayımlarla, ayarlamanın getirdiği %4'lük model iyileşmesinden çok daha fazla maliyet düşürüyor. Gerçek bir platformda cᵤ ve cₒ satış süresi, komisyon ve müşteri kaybı verisinden **tahmin edilmeli**; buradaki sayılar örnektir.

### ⚖️ 15.5 · Segment Denetimi: Hata Gruplar Arasında Tutarlı mı?
Genel MAE iyi görünürken **belirli bir segmentte sistematik sapma** olabilir. Bu, hem **finansal** (ucuz araç satıcısına sürekli fazla fiyat önermek) hem **itibar** riskidir. Her segmentte üç şeye bakıyoruz: **yanlılık** (medyan işaretli yüzde hata), **MdAPE** ve **%90 bandının kapsaması**. Yeterince gözlem (n ≥ 30) olan segmentlerde, yanlılığın mutlak değeri **%3'ü** aşarsa ya da kapsama **%85'in** altına inerse bayrak koyuyoruz.

````python
marka_medyan = y_egitim.groupby(X_egitim["brand"]).median()
marka_kademe = pd.cut(marka_medyan, [0, 6e5, 1.5e6, np.inf], labels=["ekonomi (<6 lakh)", "orta (6–15 lakh)", "premium (>15 lakh)"])
seg = X_test.assign(kademe=X_test["brand"].map(marka_kademe).astype(object).fillna("orta (6–15 lakh)"),
                    yas_grubu=pd.cut(X_test["vehicle_age"], [-1, 2, 5, 9, 40], labels=["0–2 yıl", "3–5 yıl", "6–9 yıl", "10+ yıl"]).astype(str),
                    gorece=gorece, kapsadi=ici(alt_b, ust_b), mutlak=np.abs(gorece))
segmentler = ["kademe", "yas_grubu", "fuel_type", "transmission_type", "seller_type"]
denetim = []
for s in segmentler:
    for deger, g in seg.groupby(s):
        if len(g) >= 30:
            denetim.append({"segment": f"{s} = {deger}", "n": len(g), "yanlılık (%)": float(np.median(g["gorece"]) * 100), "MdAPE (%)": float(np.median(g["mutlak"]) * 100), "kapsama (%)": float(g["kapsadi"].mean() * 100)})
denetim = pd.DataFrame(denetim).set_index("segment")
denetim["bayrak"] = np.where((denetim["yanlılık (%)"].abs() > 3) | (denetim["kapsama (%)"] < 85), "⚠️", "")
denetim.round(2).to_csv(RAPOR_DIR / "segment_denetimi.csv")

fig, axes = plt.subplots(1, 3, figsize=(19, 6.2), sharey=True)
d_ = denetim.iloc[::-1]
axes[0].barh(d_.index, d_["yanlılık (%)"], color=[RENK["agac"] if b else RENK["mavi"] for b in d_["bayrak"] != ""], edgecolor="white")
axes[0].axvline(0, color="black", lw=1); axes[0].axvspan(-3, 3, color="green", alpha=0.08)
axes[0].set(title="Yanlılık (%) · + fazla, − eksik · kırmızı = bayraklı segment", xlabel="medyan işaretli göreli hata (%)")
axes[1].barh(d_.index, d_["MdAPE (%)"], color=RENK["ada"], edgecolor="white")
axes[1].set(title="Medyan mutlak yüzde hata (%)", xlabel="MdAPE (%)")
axes[2].barh(d_.index, d_["kapsama (%)"], color=RENK["orman"], edgecolor="white")
axes[2].axvline(90, color=RENK["agac"], ls="--", lw=1.4)
axes[2].set(title="%90 bandının kapsaması (hedef çizgisi)", xlabel="kapsama (%)", xlim=(60, 100))
for ax, k, f in ((axes[0], "yanlılık (%)", "{:+.1f}"), (axes[1], "MdAPE (%)", "{:.1f}"), (axes[2], "kapsama (%)", "{:.0f}")):
    for i, v in enumerate(d_[k]):
        ax.text(v, i, " " + f.format(v), va="center", fontsize=8.5)
kaydet(fig, "32_segment_denetimi")
plt.show()
display(denetim.round(1))
````

### 🔎 15.6 · "Neden Bu Fiyat?" Beş Örnek İlan
Beş **varsayımsal ilanı** (gerçek veri satırı değil; ama marka/model/özellikleri gerçek bir ilana benziyor) iki modelle fiyatlıyoruz. Yanlarına **%90 bandı** ve iki **"ya şöyle olsaydı?"** (*what-if*) senaryosu ekliyoruz. **AdaBoost'un oy dağılımı** ise yöntemin içini gösteriyor: her ağaç bir fiyat öneriyor, ağırlıklı medyan karar veriyor.

````python
ilanlar = pd.DataFrame([
    ("Şehir içi küçük araba", "Maruti", "Alto", 6, 45_000, "Individual", "Petrol", "Manual", 22.7, 796, 47.3, 5),
    ("Aile SUV'u", "Hyundai", "Creta", 3, 30_000, "Dealer", "Diesel", "Manual", 21.4, 1493, 113.4, 5),
    ("Eski 7 kişilik", "Toyota", "Innova", 8, 110_000, "Individual", "Diesel", "Manual", 13.0, 2494, 100.6, 7),
    ("Otomatik sedan", "Honda", "City", 4, 40_000, "Dealer", "Petrol", "Automatic", 17.4, 1497, 117.6, 5),
    ("Lüks sedan", "Mercedes-Benz", "E-Class", 5, 45_000, "Dealer", "Diesel", "Automatic", 13.0, 1950, 191.3, 5),
], columns=["ilan", "brand", "model", "vehicle_age", "km_driven", "seller_type", "fuel_type", "transmission_type", "mileage", "engine", "max_power", "seats"]).set_index("ilan")
assert all(((X_egitim["brand"] == r.brand) & (X_egitim["model"] == r.model)).any() for r in ilanlar.itertuples()), "ilan, eğitimde olmayan bir marka/model içeriyor"

p_final, p_ada = final_model.predict(ilanlar), ada_ayarli.predict(ilanlar)
b_ = np.digitize(p_final, kesim)
alt_i = p_final * np.exp([q_bin[b][0] for b in b_]); ust_i = p_final * np.exp([q_bin[b][1] for b in b_])
bir_yil_sonra = final_model.predict(ilanlar.assign(vehicle_age=ilanlar["vehicle_age"] + 1))
yirmibin_km = final_model.predict(ilanlar.assign(km_driven=ilanlar["km_driven"] + 20_000))
ornek = pd.DataFrame({"tahmin · final (₹)": p_final, "bant alt (₹)": alt_i, "bant üst (₹)": ust_i, "tahmin · AdaBoost (₹)": p_ada,
                      "1 yıl daha yaşlı (%)": (bir_yil_sonra / p_final - 1) * 100, "+20.000 km (%)": (yirmibin_km / p_final - 1) * 100}, index=ilanlar.index)
ornek.round(1).to_csv(RAPOR_DIR / "ornek_ilan_tahminleri.csv")
display(ornek.round(1))

# AdaBoost'un ağaç oyları: her ağacın önerdiği fiyat ve ağırlıkları
Xt_ilan = hz_ada.transform(ilanlar)
oylar = np.column_stack([np.expm1(e.predict(Xt_ilan)) for e in ada_ic.estimators_])
oy_w = ada_ic.estimator_weights_

fig, axes = plt.subplots(1, 5, figsize=(21, 4.0))
for ax, (ad, satir), p_med in zip(axes, ilanlar.iterrows(), p_ada):
    i = list(ilanlar.index).index(ad)
    ax.hist(oylar[i] / 1e5, bins=22, weights=oy_w, color=RENK["ada"], edgecolor="white")
    ax.axvline(p_med / 1e5, color=RENK["agac"], lw=2, label="ağırlıklı medyan")
    ax.axvline(p_final[i] / 1e5, color=RENK["orman"], lw=2, ls="--", label=f"{FINAL_AD}")
    ax.set(title=f"{ad}\n{satir['brand']} {satir['model']}", xlabel="ağacın önerdiği fiyat (₹ lakh)")
axes[0].set_ylabel("oy ağırlığı (αₜ toplamı)")
axes[0].legend(frameon=False, fontsize=8)
fig.suptitle("AdaBoost'un oylaması: 'zayıf' ağaçlar neden farklı fiyatlar öneriyor?", fontsize=15, fontweight="bold", y=1.08)
kaydet(fig, "33_adaboost_oy_dagilimlari")
plt.show()
````

### 📝 Mini Quiz

<details><summary><b>1.</b> AdaBoost.R2'de son tahmin neden ağırlıklı <i>ortalama</i> değil de ağırlıklı <i>medyan</i>? Bunun bir bedeli var mı?</summary>

Medyan, tek bir aykırı ağacın tahmininden **etkilenmez**; zor örneklere odaklanan boosting'de bazı ağaçlar uç değerlere kayabileceğinden dayanıklıdır. Bedel: tahminler **ağaçların verdiği değerlerle sınırlıdır** ve çıktı **basamaklıdır**; yumuşak bir ortalama kadar ince ayrım yapamaz.
</details>

<details><summary><b>2.</b> Sınıflandırmadaki <code>AdaBoostClassifier</code>'da zayıf öğrenici karar kütüğüdür; <code>AdaBoostRegressor</code> varsayılanı neden derinlik 3? Kütükle ne olur?</summary>

Regresyonda kütük, kayıp $\bar{L}$'yi hızla **0,5'e** taşıyabilir ve algoritma **erken durur** (oyuncak veride birkaç ağaçla). Yeterince ifade gücü olan bir zayıf öğrenici gerekir; bu yüzden varsayılan 3'tür. Ama gerçek veride 3 bile **az** kalabilir: ayarlı derinlik çok daha büyük çıktı.
</details>

<details><summary><b>3.</b> Bir Ferrari'nin test kümesinde olması R²'yi neden bu kadar değiştirdi? Hangi ölçüleri tercih ederiz?</summary>

R² ve RMSE **karesel** hata kullanır; ₹ cinsinden tek bir devasa hata toplamı domine eder. **MAE, MdAPE ve log fiyat R²** uç değerlere daha dayanıklıdır; birlikte raporlanmalıdır.
</details>

<details><summary><b>4.</b> Hedef kodlamada (target encoding) sızıntı nasıl oluşur ve <code>TargetEncoder</code> bunu nasıl önler?</summary>

Kategori ortalamasını **tüm eğitim satırlarından** hesaplamak, her satırın kendi hedefini kendi özelliğine karıştırır. `TargetEncoder.fit_transform` veriyi katlara böler, her satırı **kendisini görmeyen** katlardan hesaplanan ortalamayla kodlar ve `smooth` ile nadir kategorileri genel ortalamaya çeker.
</details>

<details><summary><b>5.</b> Ağaç tabanlı modeller ölçeklemeye neden ihtiyaç duymaz, KNN ve SVR neden duyar?</summary>

Ağaçlar yalnızca değerlerin **sıralamasına** bakar (bölünme eşikleri); KNN ve SVR **mesafe/çekirdek** hesaplar, dolayısıyla büyük ölçekli bir özellik mesafeyi domine eder.
</details>

<details><summary><b>6.</b> Konformal bantta <code>log(y/ŷ)</code> nicelikleri neden ₹ cinsinden hataya tercih edildi?</summary>

Fiyat modellerinde hata fiyatla **orantılı** büyür; sabit ₹ genişliği ucuz araçta aşırı geniş, pahalıda aşırı dar olur. Log oran **göreli** hatayı ölçer ve tek bir nicelik bütün fiyat aralıklarında anlamlı kalır; dilim bazlı niceliklerle kapsama daha dengeli olur.
</details>

<details><summary><b>7.</b> Eşleştirilmiş bootstrap farkı bize neyi gösterir ve bunu 5 katlı CV ortalamalarını kıyaslamaktan ne ayırır?</summary>

Aynı satırlar üzerinde **model farkının** belirsizliğini verir: aralık 0'ı kapsıyorsa fark şanstan ayırt edilemez. Yalnızca ortalamalara bakmak, varyansı görmezden gelir ve küçük farkları yanlış yorumlatır.
</details>
