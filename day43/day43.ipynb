| ⚡ **Hızlı:** Eğitim ve tahmin çok çabuktur | 🎯 **Açgözlü (greedy):** Her adımda yerel en iyiyi seçer, global optimumu garanti etmez |

Ağaç her düğümde tüm özellikleri ve olası eşikleri dener; **çocuk düğümleri en saf hâle getiren** bölmeyi seçer. Saflığı ölçmek için iki popüler ölçüt vardır ($p_k$: düğümdeki $k$ sınıfının oranı):

$$\text{Gini}(D) = 1 - \sum_{k=1}^{K} p_k^2 \qquad\qquad H(D) = -\sum_{k=1}^{K} p_k \log_2 p_k$$

Bir bölmenin kalitesi ise **bilgi kazancı** ile ölçülür: ebeveynin safsızlığından çocukların ağırlıklı safsızlığı çıkarılır.

$$IG = I(\text{ebeveyn}) - \sum_{j \,\in\, \{\text{sol},\ \text{sağ}\}} \frac{n_j}{n}\, I(\text{çocuk}_j)$$

<table><tr><td>

<img src="assets/etiket_kavram.svg" height="34" alt="Kavram">

**Elle örnek:** Kredi verisinin kökünde 6 *Onay* ve 8 *Red* var:

**Gini = 1 − (6/14)² − (8/14)² ≈ 0.490**

Tamamen saf bir düğümde (hepsi aynı sınıf) Gini = **0** olur; iki sınıflı bir düğümde en kötü durum %50–%50 karışımdır (Gini = 0.5).

</td></tr></table>

````python
def gini(y):
    p = pd.Series(y).value_counts(normalize=True).to_numpy()
    return 1 - np.sum(p ** 2)


def entropi(y):
    p = pd.Series(y).value_counts(normalize=True).to_numpy()
    return -np.sum(p * np.log2(p))


print(f"Kök düğüm → Gini: {gini(y_kredi):.3f} · Entropi: {entropi(y_kredi):.3f}")

# İki sınıflı bir düğümde p (pozitif sınıf oranı) değiştikçe safsızlık nasıl değişir?
p = np.linspace(0.001, 0.999, 300)
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(p, 1 - p**2 - (1 - p)**2, lw=3, color="#6C5CE7", label="Gini")
ax.plot(p, -(p * np.log2(p) + (1 - p) * np.log2(1 - p)), lw=3, color="#00798C", label="Entropi")
ax.plot(p, 1 - np.maximum(p, 1 - p), lw=2, ls="--", color="#999999", label="Sınıflandırma hatası")
ax.axvline(0.5, color=VURGU, ls=":", lw=1)
ax.annotate("En karışık düğüm\n(%50 – %50)", xy=(0.5, 1.0), xytext=(0.64, 0.86),
            arrowprops=dict(arrowstyle="->", color=VURGU))
ax.set(xlabel="Düğümdeki pozitif sınıf oranı (p)", ylabel="Safsızlık",
       title="İki sınıflı bir düğümde safsızlık ölçüleri")
ax.legend(frameon=False)
kaydet(fig, "02_safsizlik_olculeri")
plt.show()
````

### 🔬 sklearn'ün Yaptığını Elle Yapalım

Her özellik için tüm eşikleri deneyip bilgi kazancını en yükselteni bulalım ve sonucu ağacın gerçek kök bölmesiyle karşılaştıralım.

````python
def en_iyi_bolme(x, y, olcut=gini):
    """Sayısal bir özellik için bilgi kazancını en yükselten eşiği döndürür."""
    degerler = np.sort(x.unique())
    adaylar = []
    for esik in (degerler[:-1] + degerler[1:]) / 2:            # ardışık değerlerin orta noktaları
        sol, sag = y[x <= esik], y[x > esik]
        agirlikli = len(sol) / len(y) * olcut(sol) + len(sag) / len(y) * olcut(sag)
        adaylar.append((esik, olcut(y) - agirlikli))
    return max(adaylar, key=lambda t: t[1])


satirlar = []
for olcut_adi, olcut in [("Gini", gini), ("Entropi", entropi)]:
    for ozellik in X_kredi.columns:
        esik, kazanc = en_iyi_bolme(X_kredi[ozellik], y_kredi, olcut)
        satirlar.append({"ölçüt": olcut_adi, "özellik": ozellik,
                         "en iyi eşik": esik, "bilgi kazancı": round(kazanc, 4)})
display(pd.DataFrame(satirlar))

kok = kredi_agaci.tree_
print(f"🌳 sklearn'ün seçtiği kök bölme: {X_kredi.columns[kok.feature[0]]} <= {kok.threshold[0]:.2f}")
````

<table>

Ders boyunca farklı ayarlarla çok sayıda ağaç kuracağımız için küçük bir **"fabrika" fonksiyonu** yazalım:

````python
def agac_pipeline(**agac_parametreleri):
    """Sıralı kodlayıcı + karar ağacından oluşan sızıntısız bir pipeline döndürür."""
    return Pipeline([
        ("kodlayici", OrdinalEncoder(categories=[KATEGORI_SIRASI[o] for o in OZELLIKLER])
                        .set_output(transform="pandas")),
        ("agac", DecisionTreeClassifier(random_state=RANDOM_STATE, **agac_parametreleri)),
    ])


agac_pipeline(max_depth=3)
````

### 🎨 Ağacı Çizelim

Görselleştirmeyi tüm ders boyunca tutarlı kılmak için düğümleri **sınıf renklerimizle** boyayan bir yardımcı fonksiyon yazıyoruz: iç düğümler açık, yapraklar koyu tonda.

<table><tr><td>

<img src="assets/etiket_not.svg" height="34" alt="Not">

**Kutuları okuma rehberi**

- Eşikler **kodlanmış** değerlerdir: `Güvenlik <= 0.5` → *yalnızca low*, `Kişi kapasitesi <= 0.5` → *yalnızca 2 kişilik*. Sol ok (**True**) koşulun sağlandığı daldır.
- `samples` düğüme düşen eğitim örneklerinin yüzdesi, `value` sınıf oranlarıdır. `value` sırası sklearn'ün **alfabetik** sınıf sırasıdır: `[acc, good, unacc, vgood]`.

</td></tr></table>

````python
def agaci_ciz(pipeline, baslik, dosya_adi, max_depth=None, figsize=(18, 8), fontsize=10):
    """Ağacı Türkçe özellik adları ve sınıf renkleriyle çizer, PNG olarak kaydeder."""
    agac = pipeline["agac"]
    fig, ax = plt.subplots(figsize=figsize)
    kutular = plot_tree(agac, feature_names=[OZELLIK_TR[o] for o in OZELLIKLER],
                        class_names=list(agac.classes_), filled=True, rounded=True,
                        impurity=False, proportion=True, precision=2,
                        max_depth=max_depth, fontsize=fontsize, ax=ax)
    for kutu in kutular:
        metin = kutu.get_text()
        if "class = " in metin:
            sinif = metin.rsplit("class = ", 1)[1].strip()
            oran = 0.75 if "<=" not in metin else 0.3            # yaprak koyu, iç düğüm açık
            renk = oran * np.array(to_rgb(SINIF_RENK[sinif])) + (1 - oran) * np.ones(3)
            kutu.get_bbox_patch().set(facecolor=renk, edgecolor=VURGU)
    ax.set_title(baslik, fontsize=15, fontweight="bold")
    kaydet(fig, dosya_adi)
    plt.show()


agaci_ciz(sig_agac, "İlk model — derinlik 3 karar ağacı", "05_ilk_model_agaci", figsize=(18, 7))
````

<table><tr><td>

<img src="assets/etiket_ybs.svg" height="34" alt="YBS bağlantısı">

Kodlanmış eşikleri okumak yöneticiler için zor. Ağacın her yaprağını **kategori adlarıyla** yazılmış **"EĞER … İSE …"** kurallarına çeviren bir fonksiyon yazalım.

Bu, karar ağaçlarının YBS'deki en büyük süper gücüdür: **model = okunabilir iş kuralları.**

</td></tr></table>

````python
def okunur_kurallar(pipeline):
    """Her yaprağı kategori adlarıyla 'EĞER … İSE …' kuralına çevirir."""
    agac, siniflar = pipeline["agac"].tree_, pipeline["agac"].classes_
    kurallar = []

    def gez(dugum, izinli):
        if agac.children_left[dugum] == -1:                       # yaprak düğüm
            oranlar = agac.value[dugum][0] / agac.value[dugum][0].sum()
            kosullar = [f"{OZELLIK_TR[o]} ∈ {{{', '.join(d)}}}" for o, d in izinli.items()
                        if len(d) < len(KATEGORI_SIRASI[o])]      # kısıtlanmamış özellikleri yazma
            kurallar.append({"EĞER": " VE ".join(kosullar) or "her durumda",
                             "İSE": siniflar[oranlar.argmax()],
                             "saflık": round(oranlar.max(), 3),
                             "örnek sayısı": int(agac.n_node_samples[dugum])})
            return
        ozellik = OZELLIKLER[agac.feature[dugum]]
        k = int(np.floor(agac.threshold[dugum])) + 1               # kod <= eşik  →  ilk k kategori
        sira = KATEGORI_SIRASI[ozellik]
        gez(agac.children_left[dugum],  {**izinli, ozellik: [d for d in izinli[ozellik] if sira.index(d) < k]})
        gez(agac.children_right[dugum], {**izinli, ozellik: [d for d in izinli[ozellik] if sira.index(d) >= k]})

    gez(0, {o: list(KATEGORI_SIRASI[o]) for o in OZELLIKLER})
    return pd.DataFrame(kurallar).sort_values("örnek sayısı", ascending=False, ignore_index=True)


with pd.option_context("display.max_colwidth", None):
    display(okunur_kurallar(sig_agac))
````

### 📏 Hangi Metrik Hangi Soruyu Cevaplar?

| | Metrik | Cevapladığı soru | Ne zaman önemli? |
|:-:|:--|:--|:--|
| 🎯 | **Doğruluk** (Accuracy) | Tüm tahminlerin yüzde kaçı doğru? | Sınıflar dengeliyse |
| 🔎 | **Kesinlik** (Precision) | "İyi" dediğim araçların kaçı gerçekten iyi? | Yanlış alarm pahalıysa (gereksiz satın alma) |
| 🪝 | **Duyarlılık** (Recall) | Gerçekten iyi araçların kaçını yakaladım? | Fırsat kaçırmak pahalıysa |
| ⚖️ | **F1** | Kesinlik ile duyarlılığın harmonik ortalaması | İkisini dengelemek gerektiğinde |
| ⭐ | **Makro F1** | Sınıf F1'lerinin basit ortalaması | Azınlık sınıfları da eşit önemliyse |
| 🧮 | **Dengeli doğruluk** | Sınıf duyarlılıklarının ortalaması | Dengesiz veride doğruluk yerine |

````python
def rapor_tablosu(y_gercek, y_tahmin):
    """classification_report çıktısını düzenli bir DataFrame'e çevirir."""
    rapor = classification_report(y_gercek, y_tahmin, labels=SINIF_SIRASI,
                                  output_dict=True, zero_division=0)
    return pd.DataFrame(rapor).T.round(3)


def karmasiklik_ciz(y_gercek, y_tahmin, baslik, dosya_adi):
    """Karmaşıklık matrisini adet ve satır yüzdesi olarak yan yana çizer."""
    etiketler = [SINIF_TR[s] for s in SINIF_SIRASI]
    adet = confusion_matrix(y_gercek, y_tahmin, labels=SINIF_SIRASI)
    oran = adet / adet.sum(axis=1, keepdims=True)
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    ortak = dict(cbar=False, xticklabels=etiketler, yticklabels=etiketler,
                 linewidths=2, linecolor="white", annot_kws={"size": 12})
    sns.heatmap(adet, annot=True, fmt="d", cmap="Blues", ax=axes[0], **ortak)
    sns.heatmap(oran, annot=True, fmt=".0%", cmap="Purples", vmin=0, vmax=1, ax=axes[1], **ortak)
    axes[0].set_title("Adet")
    axes[1].set_title("Satır yüzdesi (= sınıf duyarlılığı)")
    for ax in axes:
        ax.set(xlabel="Tahmin edilen", ylabel="Gerçek")
        ax.tick_params(axis="y", rotation=0)
    axes[1].set(ylabel="", yticklabels=[])                      # etiketler soldaki panelle ortak
    fig.subplots_adjust(wspace=0.08)
    fig.suptitle(baslik, fontsize=15, fontweight="bold", y=1.03)
    kaydet(fig, dosya_adi)
    plt.show()


y_pred_sig = sig_agac.predict(X_test)
ilk_rapor = rapor_tablosu(y_test, y_pred_sig)
ilk_rapor.to_csv(RAPOR_DIR / "ilk_model_siniflandirma_raporu.csv")
ilk_rapor
````

Eğitim doğruluğu **%100**; ağaç eğitim verisini **ezberledi**. Bu veri setinde test skoru yine de yüksek, çünkü etiketler bir uzman modelinden **gürültüsüz** üretilmiş. Gerçek kurumsal veride ise yanlış girilmiş kayıtlar, tutarsız kararlar ve ölçüm hataları vardır.

<table><tr><td>

<img src="assets/etiket_deney.svg" height="34" alt="Deney">

Eğitim etiketlerinin **%15'ini rastgele bozup** gerçek hayatı simüle edelim. Derinlik arttıkça **eğitim** ve **çapraz doğrulama** skorlarının nasıl değiştiğini izleyelim.

</td></tr></table>

````python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
DERINLIKLER = range(1, 16)


def derinlik_egrisi(X_, y_):
    """Her derinlik için 5 katlı CV ile eğitim ve doğrulama makro F1 skorlarını hesaplar."""
    satirlar = []
    for d in DERINLIKLER:
        s = cross_validate(agac_pipeline(max_depth=d), X_, y_, cv=cv,
                           scoring="f1_macro", return_train_score=True)
        satirlar.append({"derinlik": d, "eğitim": s["train_score"].mean(),
                         "doğrulama": s["test_score"].mean(), "std": s["test_score"].std()})
    return pd.DataFrame(satirlar)


rng = np.random.default_rng(RANDOM_STATE)
y_gurultulu = y_train.copy()
bozulacak = rng.choice(len(y_train), size=int(0.15 * len(y_train)), replace=False)
y_gurultulu.iloc[bozulacak] = rng.choice(SINIF_SIRASI, size=len(bozulacak))

egriler = {"Temiz etiketler": derinlik_egrisi(X_train, y_train),
           "%15 etiket gürültüsü (gerçek hayat)": derinlik_egrisi(X_train, y_gurultulu)}

fig, axes = plt.subplots(1, 2, figsize=(15, 5), sharey=True)
for ax, (baslik, e) in zip(axes, egriler.items()):
    ax.plot(e["derinlik"], e["eğitim"], "o-", color="#6C5CE7", lw=2.5, label="Eğitim")
    ax.plot(e["derinlik"], e["doğrulama"], "o-", color="#00798C", lw=2.5, label="Doğrulama (5-kat CV)")
    ax.fill_between(e["derinlik"], e["doğrulama"] - e["std"], e["doğrulama"] + e["std"],
                    color="#00798C", alpha=0.15)
    en_iyi = e.loc[e["doğrulama"].idxmax()]
    ax.axvline(en_iyi["derinlik"], color=VURGU, ls="--", lw=1)
    ax.annotate(f"en iyi derinlik = {int(en_iyi['derinlik'])}",
                xy=(en_iyi["derinlik"], en_iyi["doğrulama"]),
                xytext=(en_iyi["derinlik"] + 0.6, 0.35), arrowprops=dict(arrowstyle="->", color=VURGU))
    ax.set(title=baslik, xlabel="max_depth", ylabel="Makro F1", xticks=list(DERINLIKLER))
    ax.legend(loc="lower right", frameon=False)
fig.suptitle("Derinlik — performans ilişkisi", fontsize=16, fontweight="bold", y=1.02)
kaydet(fig, "07_derinlik_analizi")
plt.show()
````

YBS açısından asıl soru şu: **Performanstan çok az ödün vererek ağacı ne kadar sadeleştirebiliriz?**

Daha az kural = daha kolay **denetim**, daha kolay **açıklama**, daha az **bakım yükü**.

</td></tr></table>

````python
X_train_kod = agac_pipeline()["kodlayici"].fit_transform(X_train)
yol = DecisionTreeClassifier(random_state=RANDOM_STATE).cost_complexity_pruning_path(X_train_kod, y_train)
alfalar = np.unique(yol.ccp_alphas[:-1])                    # sonuncusu yalnızca kökü bırakır
alfalar = alfalar[np.linspace(0, len(alfalar) - 1, min(40, len(alfalar))).astype(int)]

ccp = pd.DataFrame([{
    "ccp_alpha": a,
    "yaprak": agac_pipeline(ccp_alpha=a).fit(X_train, y_train)["agac"].get_n_leaves(),
    "cv_f1": (s := cross_val_score(agac_pipeline(ccp_alpha=a), X_train, y_train, cv=cv, scoring="f1_macro")).mean(),
    "std": s.std(),
} for a in alfalar])

# "1 standart hata kuralı": en iyi skora istatistiksel olarak denk olan EN SADE ağaç
en_iyi = ccp.loc[ccp["cv_f1"].idxmax()]
sade = ccp[ccp["cv_f1"] >= en_iyi["cv_f1"] - en_iyi["std"]].sort_values("yaprak").iloc[0]

egri = ccp.groupby("yaprak", as_index=False).agg(cv_f1=("cv_f1", "max"), std=("std", "first"))
fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(egri["yaprak"], egri["cv_f1"], "o-", color="#00798C", lw=2.5)
ax.fill_between(egri["yaprak"], egri["cv_f1"] - egri["std"], egri["cv_f1"] + egri["std"],
                color="#00798C", alpha=0.15)
ax.axhspan(en_iyi["cv_f1"] - en_iyi["std"], en_iyi["cv_f1"], color="#EDAE49", alpha=0.25,
           label="En iyi skorun 1 std. sapma bandı")
for nokta, etiket, renk, kaydir in [(en_iyi, "En yüksek skor", "#6C5CE7", (-30, -0.22)),
                                    (sade, "Sade seçim", "#D1495B", (-30, -0.12))]:
    ax.scatter(nokta["yaprak"], nokta["cv_f1"], s=160, color=renk, zorder=5)
    ax.annotate(f"{etiket}\n{int(nokta['yaprak'])} yaprak · F1={nokta['cv_f1']:.3f}",
                xy=(nokta["yaprak"], nokta["cv_f1"]),
                xytext=(nokta["yaprak"] + kaydir[0], nokta["cv_f1"] + kaydir[1]),
                arrowprops=dict(arrowstyle="->", color=renk), color=renk, fontweight="bold")
ax.set(xlabel="Yaprak sayısı (= iş kuralı sayısı)", ylabel="CV makro F1",
       title="Maliyet-karmaşıklık budaması: sadelik ↔ performans")
ax.legend(loc="lower right", frameon=False)
kaydet(fig, "08_ccp_budama")
plt.show()

print(f"En yüksek skor : {en_iyi['cv_f1']:.3f} → {int(en_iyi['yaprak'])} yaprak (ccp_alpha={en_iyi['ccp_alpha']:.5f})")
print(f"Sade seçim     : {sade['cv_f1']:.3f} → {int(sade['yaprak'])} yaprak (ccp_alpha={sade['ccp_alpha']:.5f})")
print(f"⇒ Kural sayısı %{(1 - sade['yaprak'] / en_iyi['yaprak']) * 100:.0f} azalırken skor yalnızca "
      f"{en_iyi['cv_f1'] - sade['cv_f1']:.3f} düşüyor.")
````

### 🔎 "Neden?" — Tek Bir Kararın Gerekçesi

Komite üyesi soruyor: *"Bu aracı neden reddettiniz / önerdiniz?"* Karar ağacında cevap, aracın kökten yaprağa izlediği **yoldur**:

````python
def karari_acikla(pipeline, arac):
    """Bir aracın kökten yaprağa izlediği yolu iş diliyle yazdırır."""
    girdi = arac[OZELLIKLER].to_frame().T
    agac = pipeline["agac"]
    dugumler = agac.decision_path(pipeline["kodlayici"].transform(girdi)).indices

    print(f"🚗 {arac['arac_id']} — {arac['aciklama']}")
    for adim, dugum in enumerate(dugumler[:-1], 1):
        ozellik = OZELLIKLER[agac.tree_.feature[dugum]]
        sira = KATEGORI_SIRASI[ozellik]
        k = int(np.floor(agac.tree_.threshold[dugum])) + 1
        grup = sira[:k] if sira.index(arac[ozellik]) < k else sira[k:]
        print(f"   {adim}. {OZELLIK_TR[ozellik]:<16} = {arac[ozellik]:<6} → {{{', '.join(grup)}}} dalı")
    tahmin = pipeline.predict(girdi)[0]
    print(f"   ⇒ Karar: {SINIF_TR[tahmin]} ({tahmin}) — {AKSIYON[tahmin]}\n")


for arac_id in ["ARC-01", "ARC-06", "ARC-04"]:
    karari_acikla(model, aday.set_index("arac_id", drop=False).loc[arac_id])
````

### ✍️ Mini Quiz

<details>
<summary><b>1️⃣ Bir düğümdeki tüm örnekler aynı sınıftaysa Gini değeri kaçtır?</b></summary>

**0.** Tek sınıf varsa $p = 1$ ve $1 - 1^2 = 0$. Düğüm tamamen saftır ve yaprak olur.
</details>

<details>
<summary><b>2️⃣ Karar ağaçları için neden <code>StandardScaler</code> gerekmez?</b></summary>

Ağaç yalnızca "değer ≤ eşik mi?" diye sorar. Ölçekleme gibi **sırayı koruyan** dönüşümler bölmelerin sonucunu değiştirmez.
</details>

<details>
<summary><b>3️⃣ Eğitim doğruluğu 1.00, test doğruluğu 0.78. Teşhisiniz ve iki çözüm öneriniz?</b></summary>

**Aşırı öğrenme.** Çözümler: `max_depth` / `min_samples_leaf` ile ön budama, `ccp_alpha` ile son budama, daha fazla veri ya da topluluk yöntemleri (Random Forest).
</details>

<details>
<summary><b>4️⃣ Bu derste modeli neden doğruluk yerine makro F1 ile optimize ettik?</b></summary>

Veri dengesiz (~%70 *unacc*). Doğruluk, azınlık sınıflarını (*good*, *vgood*) hiç bilemeyen bir modeli bile iyi gösterebilir. Makro F1 her sınıfa eşit ağırlık verir.
</details>

<details>
<summary><b>5️⃣ <code>ccp_alpha</code> artırıldığında ağaca ne olur?</b></summary>

Her ek yaprak için ceza artar, zayıf dallar budanır; ağaç **küçülür**, daha sade ve genellikle daha genellenebilir hâle gelir. Çok büyük değerler ise ağacı tek bir köke indirger (yetersiz öğrenme).
</details>
