### **Kod Karşılaştırması: Aynı Grafik, İki Yöntem**

````python
# Önce gerekli kütüphaneleri import edelim
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Basit bir veri seti oluşturalım: Kampüs yemekhanesi günlük satış verileri
gunler = ['Pazartesi', 'Salı', 'Çarşamba', 'Perşembe', 'Cuma']
ogrenci_sayisi = [450, 520, 480, 510, 380]
personel_sayisi = [80, 85, 78, 90, 65]

# DataFrame oluştur
df_yemekhane = pd.DataFrame({
    'Gun': gunler,
    'Ogrenci': ogrenci_sayisi,
    'Personel': personel_sayisi
})

print("Kampüs Yemekhanesi Haftalık Veri:")
print(df_yemekhane)
````

````python
# MATPLOTLIB ile çizim (Daha fazla kod, manuel ayarlar)
fig, ax = plt.subplots(figsize=(10, 6))

# X eksenini sayısal hale getir
x = np.arange(len(gunler))
width = 0.35  # Bar genişliği

# İki ayrı bar grubu çiz
bars1 = ax.bar(x - width/2, ogrenci_sayisi, width, label='Öğrenci', color='#42A5F5')
bars2 = ax.bar(x + width/2, personel_sayisi, width, label='Personel', color='#66BB6A')

# Eksen ayarları
ax.set_xlabel('Günler', fontsize=12)
ax.set_ylabel('Kişi Sayısı', fontsize=12)
ax.set_title('Haftalık Yemekhane Kullanımı (Matplotlib)', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(gunler)
ax.legend()
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("👆 Matplotlib: Daha fazla kod satırı, her detay elle ayarlandı")
````

````python
# SEABORN ile çizim (Daha az kod, otomatik estetik)

# Veriyi "long format"a dönüştür (Seaborn bu formatı sever)
df_long = df_yemekhane.melt(id_vars='Gun', var_name='Kategori', value_name='Kisi_Sayisi')

# Tek satırda grafik çiz
plt.figure(figsize=(10, 6))
sns.barplot(data=df_long, x='Gun', y='Kisi_Sayisi', hue='Kategori', palette='Set2')
plt.title('Haftalık Yemekhane Kullanımı (Seaborn)', fontsize=14, fontweight='bold')
plt.xlabel('Günler', fontsize=12)
plt.ylabel('Kişi Sayısı', fontsize=12)
plt.tight_layout()
plt.show()

print("👆 Seaborn: Daha az kod, daha temiz görünüm, otomatik renkler")
````

- **toplam_hesap**: Ödenen tutar (TL)
- **bahsis**: Bırakılan bahşiş (TL)
- **kisi_sayisi**: Kaç kişilik masa
- **odeme_yontemi**: Nakit, Kart, veya Kampüs Kartı
- **memnuniyet**: 1-5 arası memnuniyet skoru

````python
# Rastgeleliği sabitlemek için seed ayarla (tekrarlanabilirlik için)
np.random.seed(42)

# 500 işlem kaydı oluşturalım
kayit_sayisi = 500

# Tarih aralığı: Son 3 ay (Kasım 2025 - Ocak 2026)
baslangic_tarihi = datetime(2025, 11, 1)
tarihler = [baslangic_tarihi + timedelta(days=np.random.randint(0, 90)) for _ in range(kayit_sayisi)]

# Haftanın günü
gunler = [tarih.strftime('%A') for tarih in tarihler]
gun_tr = {
    'Monday': 'Pazartesi',
    'Tuesday': 'Salı',
    'Wednesday': 'Çarşamba',
    'Thursday': 'Perşembe',
    'Friday': 'Cuma',
    'Saturday': 'Cumartesi',
    'Sunday': 'Pazar'
}
gunler_tr = [gun_tr[gun] for gun in gunler]

# Saat dilimi (Hafta içi daha çok yoğunluk, hafta sonu daha az)
saat_dilimi = []
for gun in gunler_tr:
    if gun in ['Cumartesi', 'Pazar']:
        # Hafta sonu daha çok öğle
        saat_dilimi.append(np.random.choice(['Öğle', 'Akşam'], p=[0.7, 0.3]))
    else:
        # Hafta içi dengeli
        saat_dilimi.append(np.random.choice(['Öğle', 'Akşam'], p=[0.55, 0.45]))

# Müşteri tipi (Öğrenci %75, Personel %25)
musteri_tipi = np.random.choice(['Öğrenci', 'Personel'], size=kayit_sayisi, p=[0.75, 0.25])

# Kişi sayısı (1-6 arası)
kisi_sayisi = np.random.choice([1, 2, 3, 4, 5, 6], size=kayit_sayisi, p=[0.25, 0.35, 0.20, 0.12, 0.05, 0.03])

# Toplam hesap (Müşteri tipine, kişi sayısına ve saat dilimine göre değişir)
toplam_hesap = []
for i in range(kayit_sayisi):
    # Kişi başı ortalama harcama
    if musteri_tipi[i] == 'Öğrenci':
        kisi_basi = np.random.uniform(35, 60)  # Öğrenci daha düşük bütçe
    else:
        kisi_basi = np.random.uniform(55, 95)  # Personel daha yüksek bütçe
    
    # Akşam yemeği genelde daha pahalı
    if saat_dilimi[i] == 'Akşam':
        kisi_basi *= 1.2
    
    toplam = kisi_basi * kisi_sayisi[i]
    toplam_hesap.append(round(toplam, 2))

# Bahşiş (Toplam hesabın %0-%15'i arası, bazı müşteriler hiç vermez)
bahsis = []
for hesap in toplam_hesap:
    # %30 ihtimal hiç bahşiş yok
    if np.random.random() < 0.30:
        bahsis.append(0)
    else:
        # %5-15 arası bahşiş
        oran = np.random.uniform(0.05, 0.15)
        bahsis.append(round(hesap * oran, 2))

# Ödeme yöntemi
odeme_yontemi = np.random.choice(
    ['Kampüs Kartı', 'Kredi Kartı', 'Nakit'], 
    size=kayit_sayisi, 
    p=[0.60, 0.30, 0.10]  # Kampüs kartı en yaygın
)

# Memnuniyet skoru (1-5 arası, genelde yüksek)
memnuniyet = np.random.choice([1, 2, 3, 4, 5], size=kayit_sayisi, p=[0.02, 0.08, 0.20, 0.45, 0.25])

# DataFrame oluştur
df_yemekhane_detayli = pd.DataFrame({
    'tarih': tarihler,
    'gun': gunler_tr,
    'saat_dilimi': saat_dilimi,
    'musteri_tipi': musteri_tipi,
    'kisi_sayisi': kisi_sayisi,
    'toplam_hesap': toplam_hesap,
    'bahsis': bahsis,
    'odeme_yontemi': odeme_yontemi,
    'memnuniyet': memnuniyet
})

# Tarihe göre sırala
df_yemekhane_detayli = df_yemekhane_detayli.sort_values('tarih').reset_index(drop=True)

# Veri setinin özeti
print("=" * 60)
print("🍽️  KAMPÜS YEMEKHANESİ VERİ SETİ OLUŞTURULDU")
print("=" * 60)
print(f"📊 Toplam Kayıt Sayısı: {len(df_yemekhane_detayli)}")
print(f"📅 Tarih Aralığı: {df_yemekhane_detayli['tarih'].min().strftime('%d/%m/%Y')} - {df_yemekhane_detayli['tarih'].max().strftime('%d/%m/%Y')}")
print(f"💰 Toplam Gelir: {df_yemekhane_detayli['toplam_hesap'].sum():,.2f} TL")
print(f"💵 Toplam Bahşiş: {df_yemekhane_detayli['bahsis'].sum():,.2f} TL")
print(f"👥 Ortalama Kişi Sayısı: {df_yemekhane_detayli['kisi_sayisi'].mean():.2f}")
print(f"⭐ Ortalama Memnuniyet: {df_yemekhane_detayli['memnuniyet'].mean():.2f} / 5.00")
print("\n" + "=" * 60)

# İlk 10 kayıt
print("\n📋 İlk 10 Kayıt:")
df_yemekhane_detayli.head(10)
````

## 🏋️ **<strong>Kampüs Spor Kulübü Veri Seti Oluşturma</strong>**

        <strong>Durum:</strong> Kampüs spor kulübü yönetimi, üyelik paketlerini ve hizmet kalitesini gözden geçirmek istiyor. Sizden mevcut 300 üyenin verilerini analiz etmeniz istendi.<br><br>
        <strong>Göreviniz:</strong> Hangi üyelik paketi daha popüler? Yaş, cinsiyet ve boy-kilo dağılımları nasıl? Aktivite tercihleri nedir?

### **Veri Seti Yapısı**

- **uye_id**: Üye numarası
- **cinsiyet**: Kadın veya Erkek
- **yas**: Üyenin yaşı
- **boy**: Boy (cm)
- **kilo**: Kilo (kg)
- **uyelik_paketi**: Bronze, Silver, Gold
- **aylik_ucret**: Aylık ödenen ücret (TL)
- **aktivite**: Fitness, Yüzme, Pilates, veya Yoga
- **haftalik_giris**: Haftada kaç kez kulübü ziyaret ediyor

````python
# Spor kulübü verisi için seed
np.random.seed(123)

uye_sayisi = 300

# Üye ID'leri
uye_id = [f"UYE{str(i).zfill(4)}" for i in range(1, uye_sayisi + 1)]

# Cinsiyet (Dengeli dağılım)
cinsiyet = np.random.choice(['Kadın', 'Erkek'], size=uye_sayisi, p=[0.48, 0.52])

# Yaş (18-55 arası, genç ağırlıklı)
# Olasılıkları normalize et (toplamı 1 olmalı)
p_yas = [0.05]*5 + [0.10]*10 + [0.08]*15 + [0.03]*8
p_yas = np.array(p_yas) / np.sum(p_yas)
yas = np.random.choice(range(18, 56), size=uye_sayisi, p=p_yas)

# Boy ve Kilo (Cinsiyete göre farklı dağılımlar)
boy = []
kilo = []
for c in cinsiyet:
    if c == 'Kadın':
        # Kadınlar için ortalama 160-175 cm
        b = np.random.normal(167, 6)
        boy.append(int(np.clip(b, 150, 185)))
        # Kilo boy ile ilişkili ama değişken
        k = np.random.normal(62, 8)
        kilo.append(int(np.clip(k, 45, 90)))
    else:
        # Erkekler için ortalama 170-185 cm
        b = np.random.normal(177, 7)
        boy.append(int(np.clip(b, 160, 200)))
        # Kilo boy ile ilişkili ama değişken
        k = np.random.normal(78, 10)
        kilo.append(int(np.clip(k, 55, 110)))

# Üyelik paketi (Bronze en popüler)
uyelik_paketi = np.random.choice(['Bronze', 'Silver', 'Gold'], size=uye_sayisi, p=[0.55, 0.30, 0.15])

# Aylık ücret (Pakete göre değişir)
aylik_ucret = []
for paket in uyelik_paketi:
    if paket == 'Bronze':
        ucret = np.random.uniform(200, 250)
    elif paket == 'Silver':
        ucret = np.random.uniform(300, 400)
    else:  # Gold
        ucret = np.random.uniform(500, 650)
    aylik_ucret.append(round(ucret, 2))

# Aktivite tercihi
aktivite = np.random.choice(
    ['Fitness', 'Yüzme', 'Pilates', 'Yoga'], 
    size=uye_sayisi, 
    p=[0.50, 0.25, 0.15, 0.10]
)

# Haftalık giriş sayısı (0-7 arası, genelde 2-4)
haftalik_giris = np.random.choice([0, 1, 2, 3, 4, 5, 6, 7], size=uye_sayisi, p=[0.02, 0.08, 0.25, 0.30, 0.20, 0.10, 0.03, 0.02])

# DataFrame oluştur
df_spor = pd.DataFrame({
    'uye_id': uye_id,
    'cinsiyet': cinsiyet,
    'yas': yas,
    'boy': boy,
    'kilo': kilo,
    'uyelik_paketi': uyelik_paketi,
    'aylik_ucret': aylik_ucret,
    'aktivite': aktivite,
    'haftalik_giris': haftalik_giris
})

# Veri setinin özeti
print("=" * 60)
print("🏋️  KAMPÜS SPOR KULÜBÜ VERİ SETİ OLUŞTURULDU")
print("=" * 60)
print(f"📊 Toplam Üye Sayısı: {len(df_spor)}")
print(f"👩 Kadın Üye: {(df_spor['cinsiyet'] == 'Kadın').sum()} ({(df_spor['cinsiyet'] == 'Kadın').sum()/len(df_spor)*100:.1f}%)")
print(f"👨 Erkek Üye: {(df_spor['cinsiyet'] == 'Erkek').sum()} ({(df_spor['cinsiyet'] == 'Erkek').sum()/len(df_spor)*100:.1f}%)")
print(f"💰 Ortalama Aylık Ücret: {df_spor['aylik_ucret'].mean():.2f} TL")
print(f"🏃 Ortalama Haftalık Giriş: {df_spor['haftalik_giris'].mean():.2f}")
print(f"📈 En Popüler Paket: {df_spor['uyelik_paketi'].mode()[0]}")
print(f"💪 En Popüler Aktivite: {df_spor['aktivite'].mode()[0]}")
print("\n" + "=" * 60)

# İlk 10 kayıt
print("\n📋 İlk 10 Kayıt:")
df_spor.head(10)
````

        Kampüs yemekhanesi verilerinden <strong>müşteri tipine göre ortalama harcama</strong> grafiğini 5 farklı stille çizeceğiz. Hangi stilin hangi durumda daha etkili olduğunu gözlemleyeceğiz.
    </p>
</div>

````python
# Yemekhane verisi ile stil karşılaştırması
import matplotlib.pyplot as plt
import seaborn as sns

# 5 farklı stil
stiller = ['whitegrid', 'darkgrid', 'white', 'ticks', 'dark']

# 5 alt grafik oluştur (2 satır, 3 sütun)
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Seaborn Stil Karşılaştırması: Yemekhane Müşteri Harcamaları', 
             fontsize=16, fontweight='bold', y=0.995)

# Her stil için grafik çiz
for idx, stil in enumerate(stiller):
    # Alt grafik konumunu hesapla
    row = idx // 3
    col = idx % 3
    ax = axes[row, col]
    
    # Stili uygula
    sns.set_style(stil)
    
    # Barplot çiz
    sns.barplot(
        data=df_yemekhane_detayli, 
        x='musteri_tipi', 
        y='toplam_hesap', 
        estimator='mean',
        palette='Set2',
        ax=ax
    )
    
    # Başlık ve etiketler
    ax.set_title(f'Stil: {stil}', fontsize=14, fontweight='bold', pad=10)
    ax.set_xlabel('Müşteri Tipi', fontsize=11)
    ax.set_ylabel('Ortalama Harcama (TL)', fontsize=11)
    ax.tick_params(labelsize=10)

# Son boş alt grafiği gizle
axes[1, 2].axis('off')

plt.tight_layout()
plt.show()

# Varsayılan stile geri dön
sns.set_style('whitegrid')

print("✅ Stil karşılaştırması tamamlandı!")
print("\n💡 İpucu: Hangi stil sizce en profesyonel görünüyor?")
````

## 🎨 **set_theme() ile Kapsamlı Tema Yönetimi**

<div style="background: linear-gradient(135deg, #FCE4EC 0%, #F8BBD0 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #E91E63; margin: 20px 0;">
    <h4 style="color: #880E4F; margin-top: 0;">🌐 Tema vs Stil Farkı</h4>
    <p style="color: #AD1457; font-size: 15px; line-height: 1.8;">
        <strong>Stil</strong> sadece arka plan ve ızgarayı etkiler.<br>
        <strong>Tema</strong> ise stil + renk paleti + font boyutları + çizgi kalınlıkları gibi tüm görsel parametreleri kapsar.
    </p>
    <p style="color: #AD1457; font-size: 15px; line-height: 1.8;">
        <code>set_theme()</code> fonksiyonu, bir projede tutarlı görsellik sağlamak için en pratik yöntemdir.
    </p>
</div>

````python
# set_theme() kullanımı: Kampüs spor verisi ile örnek
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Sol grafik: Varsayılan tema
sns.set_theme()  # Default tema
sns.boxplot(
    data=df_spor, 
    x='uyelik_paketi', 
    y='aylik_ucret',
    palette='pastel',
    ax=axes[0]
)
axes[0].set_title('Varsayılan Tema', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Üyelik Paketi', fontsize=12)
axes[0].set_ylabel('Aylık Ücret (TL)', fontsize=12)

# Sağ grafik: Özelleştirilmiş tema
sns.set_theme(
    style='ticks',              # Stil = ticks
    palette='muted',            # Renk paleti = muted (pastel tonlar)
    font_scale=1.2,             # Font büyütme oranı
    rc={
        'axes.labelsize': 13,   # Eksen etiketi boyutu
        'xtick.labelsize': 11,  # X ekseni tick boyutu
        'ytick.labelsize': 11,  # Y ekseni tick boyutu
        'legend.fontsize': 11,  # Legend font boyutu
        'figure.figsize': (8, 6)  # Varsayılan figür boyutu
    }
)

sns.boxplot(
    data=df_spor, 
    x='uyelik_paketi', 
    y='aylik_ucret',
    palette='muted',
    ax=axes[1]
)
axes[1].set_title('Özelleştirilmiş Tema (ticks + muted)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Üyelik Paketi', fontsize=12)
axes[1].set_ylabel('Aylık Ücret (TL)', fontsize=12)

plt.tight_layout()
plt.show()

# Varsayılan temaya geri dön
sns.set_theme()

print("✅ Tema karşılaştırması tamamlandı!")
print("\n💡 YBS İpucu: Proje başında tema belirleyin, sonra tüm grafikler otomatik uyumlu olur.")
````

    <p style="color: #6A1B9A; font-size: 15px; line-height: 1.8;">
        <code>style</code> parametresi, noktaların şeklini kategoriye göre değiştirir (daire, kare, üçgen, vb.). Renk körlüğü olan kişiler için alternatif bir ayırt etme yöntemidir.

    <p style="color: #6A1B9A; font-size: 15px; line-height: 1.8;">

## 🎻 **Uygulama 4: Violin Plot - Detaylı Dağılım Analizi**

<div style="background: linear-gradient(135deg, #FFF9C4 0%, #FFF59D 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #FBC02D; margin: 20px 0;">
    <h4 style="color: #F57F17; margin-top: 0;">🎻 Violin Plot vs Box Plot</h4>
    <p style="color: #F57F17; font-size: 15px; line-height: 1.8;">
        <strong>Violin plot</strong>, box plot'un gelişmiş versiyonudur. Sadece çeyrekler arası aralığı değil, <strong>tam dağılım yoğunluğunu (density)</strong> gösterir. Geniş bölge = o değerde çok veri, dar bölge = az veri.
    </p>
</div>

````python
# Violin Plot: Saat dilimine göre hesap dağılımı
plt.figure(figsize=(14, 6))

sns.violinplot(
    data=df_yemekhane_detayli,
    x='saat_dilimi',
    y='toplam_hesap',
    hue='musteri_tipi',
    split=True,           # İki grubu birleştir (space-efficient)
    palette='muted',
    inner='quartile'      # Çeyrekleri içeride göster
)

# Başlık ve etiketler
plt.title('Saat Dilimi ve Müşteri Tipine Göre Hesap Dağılımı (Violin Plot)', fontsize=14, fontweight='bold')
plt.xlabel('Saat Dilimi', fontsize=12)
plt.ylabel('Toplam Hesap (TL)', fontsize=12)
plt.legend(title='Müşteri Tipi', fontsize=11)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print("✅ Violin plot oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Öğle yemeği en yoğun saat dilimi (violin'in geniş olan kısmı)")
print("   • Sabah kahvaltısı daha homojen (dar violin)")
print("   • Split=True: İki müşteri tipini yan yana gösterir")
````

## 🔵 **Uygulama 5: Strip Plot ve Swarm Plot - Tüm Veri Noktaları**

<div style="background: linear-gradient(135deg, #E3F2FD 0%, #BBDEFB 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #2196F3; margin: 20px 0;">
    <h4 style="color: #0D47A1; margin-top: 0;">🔵 Strip vs Swarm Plot</h4>
    <ul style="color: #1565C0; font-size: 15px; line-height: 1.8;">
        <li><strong>Strip Plot:</strong> Tüm veri noktalarını gösterir, ama üst üste binebilir</li>
        <li><strong>Swarm Plot:</strong> Noktaları üst üste bindirmeden gösterir (arı sürüsü gibi)</li>
        <li><strong>Küçük veri setleri için</strong> idealdir (500+ nokta için yavaş olabilir)</li>
    </ul>
</div>

````python
# Strip Plot ve Swarm Plot karşılaştırması
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Sol: Strip Plot
sns.stripplot(
    data=df_yemekhane_detayli[df_yemekhane_detayli['gun'] == 'Pazartesi'],
    x='saat_dilimi',
    y='toplam_hesap',
    hue='musteri_tipi',
    palette='Set1',
    alpha=0.6,
    jitter=True,         # Noktaları yatayda biraz kaydır
    ax=axes[0]
)
axes[0].set_title('Strip Plot (Noktalar Üst Üste Binebilir)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Saat Dilimi', fontsize=11)
axes[0].set_ylabel('Toplam Hesap (TL)', fontsize=11)
axes[0].legend(title='Müşteri Tipi', fontsize=9)
axes[0].grid(axis='y', alpha=0.3)

# Sağ: Swarm Plot
sns.swarmplot(
    data=df_yemekhane_detayli[df_yemekhane_detayli['gun'] == 'Pazartesi'],
    x='saat_dilimi',
    y='toplam_hesap',
    hue='musteri_tipi',
    palette='Set1',
    alpha=0.7,
    ax=axes[1]
)
axes[1].set_title('Swarm Plot (Noktalar Üst Üste Binmez)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Saat Dilimi', fontsize=11)
axes[1].set_ylabel('Toplam Hesap (TL)', fontsize=11)
axes[1].legend(title='Müşteri Tipi', fontsize=9)
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("✅ Strip plot ve Swarm plot oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Strip Plot: Hızlı, ancak yoğun noktalarda karmaşık")
print("   • Swarm Plot: Daha net, ancak büyük veri setlerinde yavaş")
print("   • Sadece Pazartesi günü gösterildi (hız için)")
````

## 🎨 **Uygulama 6: Catplot ile Facet Grid - Çok Boyutlu Analiz**

<div style="background: linear-gradient(135deg, #E8F5E9 0%, #C8E6C9 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #4CAF50; margin: 20px 0;">

    <p style="color: #2E7D32; font-size: 15px; line-height: 1.8;">
        <code>catplot()</code> fonksiyonunun en güçlü özelliği, <strong>col</strong> ve <strong>row</strong> parametreleri ile <strong>alt grafik ızgarası (facet grid)</strong> oluşturabilmesidir. Böylece birden fazla kategorinin grafiklerini yan yana görebilirsiniz.

## 🔥 **Uygulama 1: Heatmap - Yemekhane Korelasyon Matrisi**

<div style="background: linear-gradient(135deg, #E0F7FA 0%, #B2EBF2 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #00ACC1; margin: 20px 0;">

````python
# Sadece sayısal sütunları seç ve korelasyon hesapla
numeric_cols = df_yemekhane_detayli.select_dtypes(include=['float64', 'int64']).columns
corr_matrix = df_yemekhane_detayli[numeric_cols].corr()

# Heatmap çiz
plt.figure(figsize=(10, 8))

sns.heatmap(
    corr_matrix,
    annot=True,           # Hücrelere değerleri yaz
    fmt='.2f',            # 2 ondalık basamak
    cmap='coolwarm',      # Renk paleti (mavi=negatif, kırmızı=pozitif)
    center=0,             # 0'ı ortala
    square=True,          # Kare hücreler
    linewidths=1,         # Izgara çizgileri
    cbar_kws={'shrink': 0.8}
)

plt.title('Yemekhane Verileri: Korelasyon Matrisi', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("✅ Heatmap oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • toplam_hesap ile bahsis arasında GÜÇLÜ POZİTİF korelasyon (0.90+)")
print("   • toplam_hesap ile kisi_sayisi arasında ORTA-GÜÇLÜ POZİTİF korelasyon")
print("   • Kırmızı = pozitif, Mavi = negatif, Beyaz = ilişki yok")
````

## 🎨 **Uygulama 2: Özelleştirilmiş Heatmap - Maske ile**

<div style="background: linear-gradient(135deg, #FCE4EC 0%, #F8BBD0 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #E91E63; margin: 20px 0;">
    <h4 style="color: #880E4F; margin-top: 0;">🎭 Gelişmiş Heatmap Teknikleri</h4>
    <p style="color: #AD1457; font-size: 15px; line-height: 1.8;">
        Korelasyon matrisi simetriktir (üst ve alt üçgen aynı). <strong>Mask kullanarak sadece alt üçgeni</strong> gösterebilir, daha temiz bir görünüm elde edebiliriz.
    </p>
</div>

````python
# Sadece alt üçgen için maske oluştur
import numpy as np

mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Özelleştirilmiş heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    corr_matrix,
    mask=mask,            # Üst üçgeni maskele
    annot=True,
    fmt='.2f',
    cmap='RdYlGn',        # Kırmızı-Sarı-Yeşil gradient
    center=0,
    square=True,
    linewidths=1.5,
    cbar_kws={'shrink': 0.8, 'label': 'Korelasyon Katsayısı'}
)

plt.title('Yemekhane Korelasyon Matrisi (Alt Üçgen)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("✅ Maskeli heatmap oluşturuldu!")
print("\n💡 YBS İpucu: Sadece alt üçgen gösterildiğinde grafik daha az karmaşık görünür.")
print("   Sunum ve raporlarda tercih edilir.")
````

        <code>sns.regplot()</code>, scatterplot'a <strong>regresyon çizgisi (trend çizgisi)</strong> ekler. Bu, korelasyonun yönünü ve gücünü görsel olarak daha net gösterir.

## 📈 **Uygulama 2: Histogram + KDE - İkisi Birlikte**

<div style="background: linear-gradient(135deg, #FCE4EC 0%, #F8BBD0 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #E91E63; margin: 20px 0;">
    <h4 style="color: #880E4F; margin-top: 0;">📈 KDE (Kernel Density Estimation) Nedir?</h4>
    <p style="color: #AD1457; font-size: 15px; line-height: 1.8;">
        KDE, histogram'ın <strong>yumuşatılmış (smoothed) versiyonudur</strong>. Barlar yerine sürekli bir eğri çizer. Dağılımın genel şeklini görmek için idealdir.
    </p>
</div>

````python
# Histogram + KDE birlikte
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Sol: Histogram + KDE
sns.histplot(
    data=df_yemekhane_detayli,
    x='toplam_hesap',
    bins=25,
    kde=True,             # KDE eğrisi ekle
    color='#66BB6A',
    edgecolor='black',
    ax=axes[0]
)
axes[0].set_title('Histogram + KDE', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Toplam Hesap (TL)', fontsize=11)
axes[0].set_ylabel('Frekans / Yoğunluk', fontsize=11)
axes[0].grid(axis='y', alpha=0.3)

# Sağ: Sadece KDE
sns.kdeplot(
    data=df_yemekhane_detayli,
    x='toplam_hesap',
    fill=True,
    color='#FF7043',
    alpha=0.6,
    ax=axes[1]
)
axes[1].set_title('Sadece KDE (Yoğunluk Eğrisi)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Toplam Hesap (TL)', fontsize=11)
axes[1].set_ylabel('Yoğunluk', fontsize=11)
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("✅ Histogram + KDE oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • KDE, dağılımın genel şeklini net gösterir")
print("   • Histogram: Kesin sayılar, KDE: Genel trend")
print("   • İkisini birlikte kullanmak en iyi pratik")
````

## 🎭 **Uygulama 3: Çoklu Dağılım Karşılaştırması - Hue ile**

<div style="background: linear-gradient(135deg, #F3E5F5 0%, #E1BEE7 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #9C27B0; margin: 20px 0;">
    <h4 style="color: #4A148C; margin-top: 0;">🎭 Grupları Karşılaştırmak</h4>
    <p style="color: #6A1B9A; font-size: 15px; line-height: 1.8;">
        <code>hue</code> parametresi ile farklı grupların dağılımlarını aynı grafikte karşılaştırabiliriz. Örneğin: Öğrenci vs Personel hesap dağılımları.
    </p>
</div>

````python
# Müşteri tipine göre hesap dağılımları
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Sol: Histogram ile karşılaştırma
sns.histplot(
    data=df_yemekhane_detayli,
    x='toplam_hesap',
    hue='musteri_tipi',
    bins=20,
    kde=True,
    palette='Set1',
    alpha=0.6,
    ax=axes[0]
)
axes[0].set_title('Müşteri Tipine Göre Hesap Dağılımı (Histogram)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Toplam Hesap (TL)', fontsize=11)
axes[0].set_ylabel('Frekans', fontsize=11)
axes[0].legend(title='Müşteri Tipi', fontsize=10)
axes[0].grid(axis='y', alpha=0.3)

# Sağ: Sadece KDE ile karşılaştırma
sns.kdeplot(
    data=df_yemekhane_detayli,
    x='toplam_hesap',
    hue='musteri_tipi',
    fill=True,
    palette='viridis',
    alpha=0.5,
    ax=axes[1]
)
axes[1].set_title('Müşteri Tipine Göre Hesap Dağılımı (KDE)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Toplam Hesap (TL)', fontsize=11)
axes[1].set_ylabel('Yoğunluk', fontsize=11)
axes[1].legend(title='Müşteri Tipi', fontsize=10)
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("✅ Çoklu dağılım karşılaştırması oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Personel hesapları ortalama olarak daha yüksek")
print("   • Öğrenci dağılımı daha dar (düşük varyans)")
print("   • İki grubun tepe noktaları (peaks) farklı")
````

## 📉 **Uygulama 4: ECDF - Kümülatif Dağılım**

<div style="background: linear-gradient(135deg, #FFF9C4 0%, #FFF59D 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #FBC02D; margin: 20px 0;">
    <h4 style="color: #F57F17; margin-top: 0;">📉 ECDF (Empirical Cumulative Distribution Function)</h4>
    <p style="color: #F57F17; font-size: 15px; line-height: 1.8;">
        ECDF, <strong>"Bu değerin altında yüzde kaç veri var?"</strong> sorusuna cevap verir. Y ekseni 0'dan 1'e (0% - 100%) gider. <strong>Medyan, çeyrekler arası kolayca okunur</strong>.
    </p>
</div>

````python
# ECDF: Spor kulübü yaş dağılımı
plt.figure(figsize=(12, 6))

sns.ecdfplot(
    data=df_spor,
    x='yas',
    hue='cinsiyet',
    palette='Set2',
    linewidth=2.5
)

# Referans çizgileri ekle (medyan %50, Q1 %25, Q3 %75)
plt.axhline(y=0.5, color='red', linestyle='--', linewidth=1, alpha=0.7, label='Medyan (%50)')
plt.axhline(y=0.25, color='orange', linestyle='--', linewidth=1, alpha=0.7, label='Q1 (%25)')
plt.axhline(y=0.75, color='purple', linestyle='--', linewidth=1, alpha=0.7, label='Q3 (%75)')

plt.title('Spor Kulübü Üye Yaşları: Kümülatif Dağılım (ECDF)', fontsize=14, fontweight='bold')
plt.xlabel('Yaş', fontsize=12)
plt.ylabel('Kümülatif Oran (0-1)', fontsize=12)
plt.legend(fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("✅ ECDF oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • %50 çizgisi = medyan yaş (grafikteki kesişim noktası)")
print("   • Kadın üyeler erkeklerden biraz daha genç (eğri sola kaymış)")
print("   • 30 yaşın altında %70'ten fazla üye var")
````

## 🎨 **Uygulama 5: Displot - Facet Grid ile Dağılımlar**

<div style="background: linear-gradient(135deg, #E3F2FD 0%, #BBDEFB 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #2196F3; margin: 20px 0;">
    <h4 style="color: #0D47A1; margin-top: 0;">🎨 Displot - Figure-Level Fonksiyon</h4>
    <p style="color: #1565C0; font-size: 15px; line-height: 1.8;">
        <code>sns.displot()</code>, histogram/KDE/ECDF'yi <strong>facet grid ile birleştirir</strong>. <code>kind</code> parametresi ile grafik türünü, <code>col/row</code> ile alt grafikleri ayarlayabilirsiniz.
    </p>
</div>

````python
# Displot: Günlere göre hesap dağılımları
g = sns.displot(
    data=df_yemekhane_detayli,
    x='toplam_hesap',
    hue='musteri_tipi',
    col='gun',           # Her gün için ayrı grafik
    kind='kde',          # KDE grafikleri
    fill=True,
    palette='husl',
    height=3,
    aspect=1,
    facet_kws={'sharex': True, 'sharey': True}
)

g.fig.suptitle('Günlere Göre Hesap Dağılımları (KDE)', fontsize=14, fontweight='bold', y=1.02)
g.set_axis_labels('Toplam Hesap (TL)', 'Yoğunluk', fontsize=10)
g.set_titles('{col_name}', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

print("✅ Displot ile facet grid oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Her günün dağılım şekli farklı")
print("   • Cuma günü dağılım daha geniş (varyans yüksek)")
print("   • Facet grid, çoklu karşılaştırma için mükemmel")
````

## 📍 **Uygulama 6: Rug Plot - Veri Noktalarını Göster**

<div style="background: linear-gradient(135deg, #E8F5E9 0%, #C8E6C9 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #4CAF50; margin: 20px 0;">
    <h4 style="color: #1B5E20; margin-top: 0;">📍 Rug Plot Nedir?</h4>
    <p style="color: #2E7D32; font-size: 15px; line-height: 1.8;">
        Rug plot, <strong>her veri noktasının konumunu x ekseninde küçük bir çizgi ile gösterir</strong>. KDE veya histogram ile birlikte kullanıldığında, verilerin gerçek konumlarını görselleştirir.
    </p>
</div>

````python
# KDE + Rug Plot birlikte
plt.figure(figsize=(14, 6))

# KDE çiz
sns.kdeplot(
    data=df_spor,
    x='boy',
    hue='cinsiyet',
    fill=True,
    palette='Set1',
    alpha=0.5
)

# Rug plot ekle
sns.rugplot(
    data=df_spor,
    x='boy',
    hue='cinsiyet',
    palette='Set1',
    height=0.05,
    alpha=0.8
)

plt.title('Spor Kulübü Üye Boyları: KDE + Rug Plot', fontsize=14, fontweight='bold')
plt.xlabel('Boy (cm)', fontsize=12)
plt.ylabel('Yoğunluk', fontsize=12)
plt.legend(title='Cinsiyet', fontsize=11)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("✅ KDE + Rug plot oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Rug plot, veri noktalarının yoğunluk gösterir")
print("   • Alt kısımdaki çizgiler = gerçek ölçümler")
print("   • KDE'nin hangi bölgelerin gerçek veriden geldiğini gösterir")
````

## 📈 **Uygulama 1: Lineplot - Günlük Toplam Ciro**

## 📊 **Uygulama 2: Hareketli Ortalama (Rolling Average)**

<div style="background: linear-gradient(135deg, #FCE4EC 0%, #F8BBD0 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #E91E63; margin: 20px 0;">
    <h4 style="color: #880E4F; margin-top: 0;">📊 Hareketli Ortalama Nedir?</h4>
    <p style="color: #AD1457; font-size: 15px; line-height: 1.8;">
        <strong>Hareketli ortalama (rolling/moving average)</strong>, gürültüyü azaltarak <strong>genel trendi</strong> görmemizi sağlar. Son N günün ortalaması alınır ve zaman içinde kaydırılır.
    </p>
    <p style="color: #AD1457; font-size: 15px; line-height: 1.8;">
        <strong>YBS Kullanımı:</strong> Günlük dalgalanmaları filtreleyerek asıl eğilimi (yükseliş/düşüş) net gösterir.
    </p>
</div>

````python
# 3 günlük hareketli ortalama ekle
gunluk_ciro['MA_3'] = gunluk_ciro['Toplam_Ciro'].rolling(window=3).mean()

# Lineplot: Gerçek değer vs Hareketli ortalama
plt.figure(figsize=(14, 6))

# Gerçek değerler
plt.plot(gunluk_ciro['Tarih'], gunluk_ciro['Toplam_Ciro'], 
         marker='o', linewidth=1.5, markersize=6, alpha=0.6, 
         color='#90CAF9', label='Gerçek Ciro')

# 3 günlük hareketli ortalama
plt.plot(gunluk_ciro['Tarih'], gunluk_ciro['MA_3'], 
         linewidth=3, color='#D32F2F', label='3 Günlük Hareketli Ortalama')

plt.title('Günlük Ciro vs Hareketli Ortalama', fontsize=14, fontweight='bold')
plt.xlabel('Tarih', fontsize=12)
plt.ylabel('Toplam Ciro (TL)', fontsize=12)
plt.legend(fontsize=11)
plt.xticks(rotation=45)
plt.grid(alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()

print("✅ Hareketli ortalama oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Kırmızı çizgi = Trend (gürültüsüz)")
print("   • Mavi çizgi = Günlük dalgalanmalar")
print("   • Hareketli ortalama, genel yönü netleştirir")
````

## 🔢 **Uygulama 5: Kümülatif Toplam (Cumulative Sum)**

<div style="background: linear-gradient(135deg, #E3F2FD 0%, #BBDEFB 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #2196F3; margin: 20px 0;">
    <h4 style="color: #0D47A1; margin-top: 0;">🔢 Kümülatif Toplam Nedir?</h4>
    <p style="color: #1565C0; font-size: 15px; line-height: 1.8;">
        Kümülatif toplam (cumsum), <strong>zaman içinde toplam biriken değeri</strong> gösterir. "Şimdiye kadar toplam ne kadar kazandık?" sorusuna cevap verir.
    </p>
    <p style="color: #1565C0; font-size: 15px; line-height: 1.8;">
        <strong>YBS Kullanımı:</strong> Yıllık hedef takibi, satış performansı izleme.
    </p>
</div>

````python
# Kümülatif toplam ciro ekle
gunluk_ciro['Kumulative_Ciro'] = gunluk_ciro['Toplam_Ciro'].cumsum()

# Lineplot: Kümülatif toplam
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Sol: Günlük ciro
axes[0].plot(gunluk_ciro['Tarih'], gunluk_ciro['Toplam_Ciro'], 
             marker='o', linewidth=2, markersize=6, color='#42A5F5')
axes[0].set_title('Günlük Ciro', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Tarih', fontsize=11)
axes[0].set_ylabel('Günlük Ciro (TL)', fontsize=11)
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(alpha=0.3, linestyle='--')

# Sağ: Kümülatif toplam
axes[1].plot(gunluk_ciro['Tarih'], gunluk_ciro['Kumulative_Ciro'], 
             linewidth=3, color='#43A047')
axes[1].set_title('Kümülatif Toplam Ciro', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Tarih', fontsize=11)
axes[1].set_ylabel('Kümülatif Ciro (TL)', fontsize=11)
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()

print("✅ Kümülatif toplam oluşturuldu!")
print("\n💡 YBS İçgörü:")
print(f"   • Toplam ciro: {gunluk_ciro['Kumulative_Ciro'].iloc[-1]:,.2f} TL")
print("   • Sağdaki grafik sürekli yükseliyor (kümülatif)")
print("   • Soldaki grafik günlük dalgalanmaları gösteriyor")
````

<div style="background: linear-gradient(135deg, #E0F7FA 0%, #B2EBF2 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #00ACC1; margin: 20px 0;">
    <h4 style="color: #006064; margin-top: 0;">🎯 İş Sorusu</h4>
    <p style="color: #00838F; font-size: 15px; line-height: 1.8;">
        Her <strong>saat diliminde</strong> ve <strong>günlerde</strong> toplam hesap dağılımı nasıl? Öğle yemeği her gün benzer mi?
    </p>
</div>

````python
# FacetGrid: Saat dilimine göre günlere ayrılmış histogram
g = sns.FacetGrid(
    df_yemekhane_detayli,
    col='saat_dilimi',
    hue='musteri_tipi',
    col_wrap=3,  # 3 sütunda göster
    height=4,
    aspect=1.2,
    palette='Set2'
)

g.map(sns.histplot, 'toplam_hesap', bins=20, alpha=0.7)
g.add_legend(title='Müşteri Tipi')
g.set_axis_labels('Toplam Hesap (TL)', 'Frekans')
g.set_titles(col_template='{col_name}')
g.fig.suptitle('Saat Dilimlerine Göre Hesap Dağılımı', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("✅ FacetGrid oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Her panel = 1 saat dilimi")
print("   • Hue = Müşteri tipi (renk ayırımı)")
print("   • Öğle yemeğinde en geniş dağılım")
````

## 🔥 **Uygulama 2: PairGrid - Özelleştirilebilir Çoklu İlişki**

<div style="background: linear-gradient(135deg, #FCE4EC 0%, #F8BBD0 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #E91E63; margin: 20px 0;">
    <h4 style="color: #880E4F; margin-top: 0;">🔥 PairGrid vs Pairplot</h4>
    <p style="color: #AD1457; font-size: 15px; line-height: 1.8;">
        <strong>Pairplot</strong> = Hızlı, otomatik çoklu ilişki<br/>
        <strong>PairGrid</strong> = Tam kontrol, her pozisyona (köşegen, üst, alt) farklı grafik atayabilirsin!
    </p>
</div>

````python
# PairGrid: Köşegen, üst ve alt üçgenlere farklı grafikler
pairplot_vars = ['boy', 'kilo', 'yas']
df_spor_sample = df_spor[pairplot_vars].copy()

g = sns.PairGrid(df_spor_sample, height=2.5, aspect=1.1)

# Köşegen: KDE (dağılım)
g.map_diag(sns.kdeplot, fill=True, color='#42A5F5', alpha=0.6)

# Alt üçgen: Scatter (nokta)
g.map_lower(sns.scatterplot, color='#66BB6A', alpha=0.5, s=40)

# Üst üçgen: KDE contour (yoğunluk çizgileri)
g.map_upper(sns.kdeplot, color='#EF5350', alpha=0.6)

g.fig.suptitle('PairGrid: Özelleştirilmiş Çoklu İlişki Analizi', fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

print("✅ PairGrid oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Köşegen = Tek değişken dağılımı (KDE)")
print("   • Alt üçgen = Scatter (detaylı noktalar)")
print("   • Üst üçgen = KDE contour (yoğunluk haritası)")
print("   • 3 değişken için 9 farklı görselleştirme!")
````

## 💎 **Uygulama 3: JointGrid - Merkez + Marjinal İlişki**

<div style="background: linear-gradient(135deg, #F3E5F5 0%, #E1BEE7 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #9C27B0; margin: 20px 0;">
    <h4 style="color: #4A148C; margin-top: 0;">💎 JointGrid Nedir?</h4>
    <p style="color: #6A1B9A; font-size: 15px; line-height: 1.8;">
        <strong>JointGrid</strong>, merkezdeki ana ilişki grafiğine ek olarak <strong>marjinal dağılımları</strong> (x ve y eksenlerinde) gösterir.
    </p>
    <p style="color: #6A1B9A; font-size: 15px; line-height: 1.8;">
        <strong>YBS Kullanımı:</strong> İki metriği (satış-maliyet, hız-kalite) hem ilişkisel hem de dağılımsal olarak analiz ederiz.
    </p>
</div>

````python
# JointGrid: Toplam hesap vs Kişi sayısı
g = sns.JointGrid(
    data=df_yemekhane_detayli,
    x='kisi_sayisi',
    y='toplam_hesap',
    height=8
)

# Merkez: Scatter + regresyon
g.plot_joint(sns.scatterplot, alpha=0.5, color='#7E57C2', s=50)
g.plot_joint(sns.regplot, scatter=False, color='#D32F2F', line_kws={'linewidth': 2.5})

# Marjinal (x ekseni): Histogram
g.plot_marginals(sns.histplot, kde=True, color='#66BB6A', bins=20, alpha=0.6)

g.set_axis_labels('Kişi Sayısı', 'Toplam Hesap (TL)', fontsize=12)
g.fig.suptitle('JointGrid: Kişi Sayısı - Toplam Hesap İlişkisi', 
               fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

print("✅ JointGrid oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • Merkez = İlişki (scatter + regresyon çizgisi)")
print("   • Üst = Kişi sayısı dağılımı")
print("   • Sağ = Toplam hesap dağılımı")
print("   • Hem ilişkiyi hem dağılımı aynı anda görüyoruz!")
````

## 📊 **Uygulama 4: Professional Dashboard - Çoklu Grafik Kombinasyonu**

<div style="background: linear-gradient(135deg, #FFF9C4 0%, #FFF59D 100%); padding: 20px; border-radius: 10px; border-left: 5px solid #FBC02D; margin: 20px 0;">
    <h4 style="color: #F57F17; margin-top: 0;">📊 Dashboard Yaklaşımı</h4>
    <p style="color: #F57F17; font-size: 15px; line-height: 1.8;">
        Gerçek YBS raporlarında <strong>tek bir dashboard'da birden fazla grafik</strong> gösteririz. Matplotlib'in <code>subplot</code> sistemi ile Seaborn'u kombine ederiz.
    </p>
</div>

````python
# Professional Dashboard: 6 panel
fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 2, hspace=0.3, wspace=0.3)

# Panel 1: Günlük ciro trendi
ax1 = fig.add_subplot(gs[0, 0])
sns.lineplot(data=gunluk_ciro, x='Tarih', y='Toplam_Ciro', marker='o', linewidth=2, color='#1976D2', ax=ax1)
ax1.set_title('1. Günlük Ciro Trendi', fontsize=11, fontweight='bold')
ax1.set_xlabel('Tarih', fontsize=9)
ax1.set_ylabel('Ciro (TL)', fontsize=9)
ax1.tick_params(axis='x', rotation=45, labelsize=8)
ax1.grid(alpha=0.3)

# Panel 2: Müşteri tipi dağılımı
ax2 = fig.add_subplot(gs[0, 1])
sns.countplot(data=df_yemekhane_detayli, x='musteri_tipi', palette='Set2', ax=ax2)
ax2.set_title('2. Müşteri Tipi Dağılımı', fontsize=11, fontweight='bold')
ax2.set_xlabel('Müşteri Tipi', fontsize=9)
ax2.set_ylabel('İşlem Sayısı', fontsize=9)

# Panel 3: Saat dilimi box plot
ax3 = fig.add_subplot(gs[1, 0])
sns.boxplot(data=df_yemekhane_detayli, x='saat_dilimi', y='toplam_hesap', palette='husl', ax=ax3)
ax3.set_title('3. Saat Dilimi - Hesap Dağılımı', fontsize=11, fontweight='bold')
ax3.set_xlabel('Saat Dilimi', fontsize=9)
ax3.set_ylabel('Toplam Hesap (TL)', fontsize=9)
ax3.tick_params(axis='x', rotation=45, labelsize=8)

# Panel 4: Korelasyon heatmap (spor)
ax4 = fig.add_subplot(gs[1, 1])
corr_spor = df_spor[['boy', 'kilo', 'yas', 'aylik_ucret']].corr()
sns.heatmap(corr_spor, annot=True, fmt='.2f', cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={'shrink': 0.8}, ax=ax4)
ax4.set_title('4. Korelasyon Matrisi (Spor Kulübü)', fontsize=11, fontweight='bold')

# Panel 5: Histogram - Toplam hesap dağılımı
ax5 = fig.add_subplot(gs[2, 0])
sns.histplot(data=df_yemekhane_detayli, x='toplam_hesap', bins=30, kde=True, 
             color='#43A047', alpha=0.6, ax=ax5)
ax5.set_title('5. Toplam Hesap Dağılımı', fontsize=11, fontweight='bold')
ax5.set_xlabel('Toplam Hesap (TL)', fontsize=9)
ax5.set_ylabel('Frekans', fontsize=9)

# Panel 6: Scatter - Boy vs Kilo
ax6 = fig.add_subplot(gs[2, 1])
sns.scatterplot(data=df_spor, x='boy', y='kilo', hue='cinsiyet', 
                palette='Set1', s=60, alpha=0.7, ax=ax6)
ax6.set_title('6. Boy - Kilo İlişkisi (Cinsiyet)', fontsize=11, fontweight='bold')
ax6.set_xlabel('Boy (cm)', fontsize=9)
ax6.set_ylabel('Kilo (kg)', fontsize=9)
ax6.legend(title='Cinsiyet', fontsize=8)

fig.suptitle('YBS Yemek Hizmeti & Spor Kulübü - Executive Dashboard', 
             fontsize=15, fontweight='bold', y=0.995)
plt.show()

print("✅ Professional dashboard oluşturuldu!")
print("\n💡 YBS İçgörü:")
print("   • 6 farklı grafik = Kapsamlı analiz")
print("   • Tek sayfada tüm KPI'lar")
print("   • Yönetim sunumu için hazır!")
````

- **Ne Yapar?** Kullanıcıların üzerine gelince detay gösteren, zoom yapılabilen grafikler
- **Neden Öğrenmeliyim?** Web tabanlı dashboard'lar, interaktif raporlar
- **Başlangıç:** `pip install plotly` → `import plotly.express as px`

### **2️⃣ Matplotlib İleri Seviye**
- **Ne Yapar?** Seaborn'un altında çalışan kütüphane, tam kontrol
- **Neden Öğrenmeliyim?** Özel tasarımlar, scientific publication grafikleri
- **Öğren:** `plt.subplots()`, `ax.annotate()`, custom legends

### **3️⃣ Tableau / Power BI**
- **Ne Yapar?** Kod yazmadan drag-drop interface ile görselleştirme
- **Neden Öğrenmeliyim?** YBS'de çok kullanılır, business intelligence raporları
- **Avantaj:** Seaborn bilgisi, Tableau mantığını anlamanızı kolaylaştırır

### **4️⃣ Dash - Python Web Dashboard**
- **Ne Yapar?** Python ile interaktif web uygulamaları
- **Neden Öğrenmeliyim?** Real-time monitoring, executive dashboards
- **Kombinasyon:** Plotly + Dash = Güçlü çözüm

### **5️⃣ İstatistiksel Modelleme (Statsmodels)**
- **Ne Yapar?** Regresyon, zaman serisi forecasting, hipotez testleri
- **Neden Öğrenmeliyim?** Görselleştirme + istatistik = Tam analiz
