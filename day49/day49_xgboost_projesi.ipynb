````python
# ── Kütüphaneler ─────────────────────────────────────────────
import json
import platform
import time
import warnings
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
import seaborn as sns
import sklearn
import xgboost
import lightgbm
import catboost
from IPython.display import Image, Markdown, display
from joblib import Parallel, delayed
from matplotlib.animation import FuncAnimation, PillowWriter
from scipy.stats import loguniform, randint, uniform

from sklearn.base import clone
from sklearn.calibration import CalibratedClassifierCV
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import (AdaBoostClassifier, BaggingClassifier, ExtraTreesClassifier, GradientBoostingClassifier,
                              HistGradientBoostingClassifier, RandomForestClassifier, StackingClassifier, VotingClassifier)
from sklearn.feature_selection import mutual_info_classif
from sklearn.inspection import permutation_importance
from sklearn.linear_model import LogisticRegression, RidgeClassifier, SGDClassifier
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, brier_score_loss, classification_report, confusion_matrix,
                             f1_score, log_loss, matthews_corrcoef, precision_recall_curve, precision_score, recall_score,
                             roc_auc_score, roc_curve)
from sklearn.model_selection import (GroupKFold, RandomizedSearchCV, RepeatedStratifiedKFold, StratifiedKFold, cross_val_predict,
                                     train_test_split)
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import FunctionTransformer, LabelEncoder, StandardScaler, label_binarize
from sklearn.svm import SVC, LinearSVC
from sklearn.tree import DecisionTreeClassifier, ExtraTreeClassifier
from xgboost import XGBClassifier, XGBRFClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

from gok_yardimci import FOTOMETRI, renk_indeksleri        # pipeline içindeki özellik mühendisliği (ayrı dosyada)

warnings.filterwarnings("ignore")

# ── Sabitler ve klasörler ───────────────────────────────────
RANDOM_STATE = 42
N_ISCI = 4                                                  # paralel çalışacak süreç sayısı
VERI_YOLU = Path("20-digitalskysurvey.csv")
FIG_DIR, RAPOR_DIR, MODEL_DIR = Path("outputs/figures"), Path("outputs/reports"), Path("outputs/model")
for klasor in (FIG_DIR, RAPOR_DIR, MODEL_DIR):
    klasor.mkdir(parents=True, exist_ok=True)

# ── Görsel tema (notebook boyunca aynı renkler) ──────────────
RENK = {"altin": "#EDAE49", "mor": "#6C5CE7", "pembe": "#D1495B", "mavi": "#2B6CB0",
        "yesil": "#2E9E6B", "turuncu": "#E8833A", "gri": "#8A94A6", "koyu": "#2E4057", "cyan": "#0E9AA7"}
SINIF_RENK = {"GALAXY": RENK["mavi"], "QSO": RENK["pembe"], "STAR": RENK["altin"]}      # galaksi: mavi · kuasar: pembe · yıldız: altın
AILE_RENK = {"Taban": "#B8C0CC", "Doğrusal": RENK["altin"], "Olasılıksal": RENK["cyan"], "Komşuluk ve çekirdek": RENK["pembe"],
             "Sinir ağı": RENK["mavi"], "Ağaç ve orman": RENK["yesil"], "Boosting": RENK["turuncu"], "Topluluk": RENK["mor"]}

plt.rcParams.update({
    "figure.dpi": 100, "savefig.dpi": 150, "savefig.bbox": "tight",
    "axes.titleweight": "bold", "axes.titlesize": 13, "axes.grid": True, "grid.alpha": 0.25,
    "axes.spines.top": False, "axes.spines.right": False,
})
sns.set_theme(style="whitegrid", rc={"axes.spines.top": False, "axes.spines.right": False, "grid.alpha": 0.25})


def kaydet(fig, ad):
    """Grafiği outputs/figures altına PNG olarak kaydeder."""
    yol = FIG_DIR / f"{ad}.png"
    fig.savefig(yol)
    print(f"💾 Kaydedildi → {yol.as_posix()}")


def rapor_kaydet(df, ad, index=True):
    """Tabloyu outputs/reports altına CSV olarak kaydeder."""
    yol = RAPOR_DIR / f"{ad}.csv"
    df.to_csv(yol, index=index, encoding="utf-8")
    print(f"💾 Rapor → {yol.as_posix()}  ({len(df)} satır)")


def gif_kaydet(kare_ciz, n_kare, yol, figsize=(7, 4), fps=2, dpi=70):
    """kare_ciz(fig, i) fonksiyonuyla her kareyi çizip animasyonlu GIF olarak kaydeder."""
    fig = plt.figure(figsize=figsize)
    anim = FuncAnimation(fig, lambda i: kare_ciz(fig, i), frames=n_kare)
    anim.save(yol, writer=PillowWriter(fps=fps), dpi=dpi)
    plt.close(fig)
    print(f"💾 Animasyon → {Path(yol).as_posix()} ({Path(yol).stat().st_size / 1024:.0f} KB)")


def makro_f1_hizli(y, p, k=3):
    """Hızlı makro-F1 (bootstrap için): karışıklık matrisini bincount ile kurar."""
    cm = np.bincount(y * k + p, minlength=k * k).reshape(k, k)
    tp = np.diag(cm).astype(float)
    prec = tp / np.maximum(cm.sum(0), 1)
    rec = tp / np.maximum(cm.sum(1), 1)
    return np.mean(2 * prec * rec / np.maximum(prec + rec, 1e-12))


def metrikler(y, tahmin, olasilik=None):
    """Çok sınıflı sınıflandırma ölçütleri. Argüman sırası: önce gerçek, sonra tahmin (sınıf), isteğe bağlı olasılık."""
    sonuc = {"Doğruluk": accuracy_score(y, tahmin), "Dengeli doğruluk": balanced_accuracy_score(y, tahmin),
             "Makro-F1": f1_score(y, tahmin, average="macro"), "MCC": matthews_corrcoef(y, tahmin),
             "QSO geri çağırma": recall_score(y, tahmin, labels=[1], average=None)[0]}
    if olasilik is not None:
        sonuc["Log-loss"] = log_loss(y, olasilik, labels=[0, 1, 2])
        sonuc["AUC (OvR)"] = roc_auc_score(y, olasilik, multi_class="ovr")
    return sonuc


print(f"Python {platform.python_version()} · scikit-learn {sklearn.__version__} · pandas {pd.__version__} · numpy {np.__version__}")
print(f"XGBoost {xgboost.__version__} · LightGBM {lightgbm.__version__} · CatBoost {catboost.__version__} · SciPy {scipy.__version__}")
````

### 🔭 2.1 · İş Problemi

Gökbilimciler her gece **milyonlarca** gök cismini fotoğraflar. Her cismin **spektrumunu** (ışığı ayrıntılı bileşenlerine ayıran ölçüm) almak çok pahalı ve yavaştır; teleskop zamanı kıttır. Oysa her cismin beş renkteki **parlaklığı** (fotometri) ucuza elde edilir. Soru şudur:

> **Yalnızca ışık ölçümlerine bakarak** bir cismin **yıldız mı, galaksi mi, kuasar mı** olduğunu ne kadar güvenle söyleyebiliriz? Ve spektroskopiyi **hangi nesneler için** harcamalıyız?

İyi bir model:

- 🔭 **Spektroskopi bütçesini** yalnızca belirsiz / nadir ve değerli nesnelere ayırır,
- 🗂️ Spektrumu olmayan **milyonlarca nesneyi** önceden sınıflandırır (benzer yaklaşımlar SDSS'te yüz milyonlarca nesne için uygulanmıştır; Clarke vd., 2020),
- 💫 **Nadir ama bilimsel olarak değerli** kuasarları kaçırmamaya çalışır.

### 📏 2.3 · Başarı Ölçütleri

### ✂️ 5.1 · Test Setini Kilitlemek

Verinin **%20'sini** (2.000 nesne) en baştan ayırıyoruz. Bölme **sınıf oranlarını korur** (*stratified*). Test seti **tüm seçimler bittikten sonra** raporlama için kullanılır; model seçimi ve ayar **yalnızca eğitim kümesinde 5 katlı çapraz doğrulamayla** yapılır.

````python
idx_egitim, idx_test = train_test_split(np.arange(len(veri)), test_size=0.20, stratify=y_tum, random_state=RANDOM_STATE)
X_egitim, X_test = veri.iloc[idx_egitim].reset_index(drop=True), veri.iloc[idx_test].reset_index(drop=True)
y_egitim, y_test = y_tum[idx_egitim], y_tum[idx_test]
plaka_egitim = X_egitim["plate"].to_numpy()

print(f"Eğitim: {len(X_egitim):,} · Test: {len(X_test):,}")
print("Sınıf oranları (eğitim):", np.round(np.bincount(y_egitim) / len(y_egitim), 3), "· (test):", np.round(np.bincount(y_test) / len(y_test), 3))

# Yarışta ve ayarda kullanılacak ana çapraz doğrulama katları (liste → tüm modeller AYNI katları görür)
CV = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
KATLAR = list(CV.split(X_egitim, y_egitim))
print("Kat boyutları (doğrulama):", [len(v) for _, v in KATLAR])
````

### 🧪 5.2 · Deney: Kimlik Sütunları ve Konum Skoru Şişiriyor mu?

Karar defterinde `plate`, `mjd`, `fiberid` (gözlem kimlikleri) ile `ra`, `dec` (konum) sütunlarını çıkarmıştık. Bu kararı **ölçerek** sınayalım. İki doğrulama düzenini karşılaştırıyoruz:

- **Rastgele katlar** (nesneler karışık)
- **Plaka bazlı katlar** (`GroupKFold`): bir plakanın **tüm nesneleri aynı katta**; model hiç görmediği plakalarda sınanır

Aynı XGBoost, beş farklı özellik kümesi:

````python
FOTO9 = FOTOMETRI + ["u_g", "g_r", "r_i", "i_z"]
kumeler = {
    "Fotometri": FOTOMETRI,
    "Fotometri + konum (ra, dec)": FOTOMETRI + ["ra", "dec"],
    "Fotometri + gözlem kimliği (plate, mjd, fiberid)": FOTOMETRI + ["plate", "mjd", "fiberid"],
    "Fotometri + redshift": FOTOMETRI + ["redshift"],
    "Hepsi": FOTOMETRI + ["redshift", "ra", "dec", "plate", "mjd", "fiberid"],
}
hizli_xgb = XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.1, n_jobs=1, random_state=RANDOM_STATE)
plaka_katlari = list(GroupKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE).split(X_egitim, y_egitim, groups=plaka_egitim))


def kume_degerlendir(ad, kolonlar):
    sonuc = {"özellik kümesi": ad}
    for duzen, katlar in [("rastgele", KATLAR), ("plaka bazlı", plaka_katlari)]:
        tahmin = cross_val_predict(make_pipeline(FunctionTransformer(renk_indeksleri), clone(hizli_xgb)), X_egitim[kolonlar], y_egitim, cv=katlar, n_jobs=1)
        sonuc[f"doğruluk · {duzen}"] = accuracy_score(y_egitim, tahmin)
        sonuc[f"makro-F1 · {duzen}"] = f1_score(y_egitim, tahmin, average="macro")
    return sonuc


kisayol = pd.DataFrame(Parallel(n_jobs=N_ISCI)(delayed(kume_degerlendir)(ad, k) for ad, k in kumeler.items()))
kisayol["fark (plaka−rastgele) · doğruluk"] = kisayol["doğruluk · plaka bazlı"] - kisayol["doğruluk · rastgele"]
display(kisayol.round(4))
rapor_kaydet(kisayol, "kisayol_deneyi", index=False)

fig, ax = plt.subplots(figsize=(12, 4.4))
konum = np.arange(len(kisayol))
ax.bar(konum - 0.2, kisayol["doğruluk · rastgele"], width=0.4, color=RENK["mavi"], label="rastgele katlar")
ax.bar(konum + 0.2, kisayol["doğruluk · plaka bazlı"], width=0.4, color=RENK["turuncu"], label="plaka bazlı katlar")
ax.set_xticks(konum)
ax.set_xticklabels([a.replace(" + ", "\n+ ") for a in kisayol["özellik kümesi"]], fontsize=8.5)
ax.set(ylim=(0.94, 1.0), ylabel="doğruluk", title="Plaka bazlı bölmede skor düşmüyor: gözlem kimliği bir 'sızıntı' kanıtı vermiyor")
ax.legend()
kaydet(fig, "05_kisayol_deneyi")
plt.show()
````

### 🧱 6.1 · Pipeline: Neyi Nerede Öğreniyoruz?

Tüm modeller aynı kalıpta kurulur:

```
ham girdi ──▶ renk indeksleri ekle ──▶ (ölçekle: yalnızca ölçeğe duyarlı modellerde) ──▶ model
```

- **Renk indeksleri** satır bazlıdır (komşu iki sütunun farkı): eğitim verisinden **hiçbir şey öğrenmez**, dolayısıyla sızıntı üretmez.
- **Ölçekleme** (`StandardScaler`) ortalama ve standart sapmayı **yalnızca eğitim katından** öğrenir; `Pipeline` bunu her katta otomatik doğru yapar.
- **Ağaç tabanlı modeller ölçekten etkilenmez**; doğrusal, komşuluk, çekirdek ve sinir ağı modelleri etkilenir.

````python
def pipe_kur(model, olcekle=False):
    """Renk indeksi → (isteğe bağlı ölçek) → model."""
    return Pipeline([("renk", FunctionTransformer(renk_indeksleri)),
                     ("olcek", StandardScaler() if olcekle else "passthrough"),
                     ("model", model)])


def cv_tahmin(pipe, X, y, katlar=KATLAR, yontem="predict", n_jobs=1):
    """Katlı çapraz doğrulama ile 'katın dışında kalan' (OOF) tahminler."""
    return cross_val_predict(clone(pipe), X, y, cv=katlar, method=yontem, n_jobs=n_jobs)


print(pipe_kur(XGBClassifier(n_estimators=100), olcekle=False))
````

### 🧪 6.2 · Deneyler: Hangi Hazırlık İşe Yarıyor?

Dört küçük soruyu **veriye** soruyoruz (Senaryo B, yalnızca fotometri; aynı katlar):

1. **Renk indeksi** eklemek gerçekten yardım ediyor mu? (XGBoost ve lojistik regresyon)
2. **Ölçekleme** modele göre fark yaratıyor mu?
3. Dengesiz sınıf için **sınıf ağırlığı** (kuasarı daha çok önemsemek) makro-F1'i ve kuasar geri çağırmasını değiştiriyor mu?

````python
class HamOzellik(FunctionTransformer):
    """Renk indeksi eklemeden yalnızca ham sütunları geçiren dönüştürücü (deney için)."""


ham_donusum = FunctionTransformer(lambda X: pd.DataFrame(X))
satir = []


def deney(ad, pipe, agirlik=False):
    ek = {"model__sample_weight": None}
    tahmin = np.empty(len(y_egitim), dtype=int)
    for tr, va in KATLAR:
        p = clone(pipe)
        if agirlik:
            w = len(tr) / (3 * np.bincount(y_egitim[tr]))[y_egitim[tr]]          # "balanced" ağırlık: nadir sınıfa yüksek
            p.fit(X_egitim.iloc[tr][SENARYO_B], y_egitim[tr], model__sample_weight=w)
        else:
            p.fit(X_egitim.iloc[tr][SENARYO_B], y_egitim[tr])
        tahmin[va] = p.predict(X_egitim.iloc[va][SENARYO_B])
    m = metrikler(y_egitim, tahmin)
    return {"deney": ad, "doğruluk": m["Doğruluk"], "makro-F1": m["Makro-F1"], "QSO geri çağırma": m["QSO geri çağırma"]}


xgb_k = lambda: XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.1, n_jobs=4, random_state=RANDOM_STATE)
lr_k = lambda: LogisticRegression(max_iter=3000)

satir.append(deney("XGBoost · ham magnitüd", Pipeline([("renk", ham_donusum), ("model", xgb_k())])))
satir.append(deney("XGBoost · magnitüd + renk indeksi", pipe_kur(xgb_k())))
satir.append(deney("Lojistik · ham magnitüd · ölçeksiz", Pipeline([("renk", ham_donusum), ("model", lr_k())])))
satir.append(deney("Lojistik · ham magnitüd · ölçekli", Pipeline([("renk", ham_donusum), ("olcek", StandardScaler()), ("model", lr_k())])))
satir.append(deney("Lojistik · + renk · ölçekli", pipe_kur(lr_k(), olcekle=True)))
satir.append(deney("XGBoost · + renk · ölçekli", pipe_kur(xgb_k(), olcekle=True)))
satir.append(deney("XGBoost · + renk · sınıf ağırlıklı", pipe_kur(xgb_k()), agirlik=True))
deneyler = pd.DataFrame(satir)
display(deneyler.round(4))
rapor_kaydet(deneyler, "deney_ozellik_olcek_agirlik", index=False)

fig, ax = plt.subplots(figsize=(11, 4.4))
yy = np.arange(len(deneyler))[::-1]
ax.barh(yy + 0.2, deneyler["makro-F1"], height=0.38, color=RENK["mor"], label="makro-F1")
ax.barh(yy - 0.2, deneyler["QSO geri çağırma"], height=0.38, color=RENK["pembe"], label="QSO geri çağırma")
ax.set_yticks(yy)
ax.set_yticklabels(deneyler["deney"], fontsize=9)
ax.set(xlim=(0.7, 1.0), title="Hazırlık deneyleri (5 katlı CV, Senaryo B: yalnız fotometri)")
ax.legend(loc="lower right")
kaydet(fig, "06_hazirlik_deneyleri")
plt.show()
````

### 🏎️ 7.2 · Yarışı Koşalım (Senaryo A: Fotometri + Redshift)

Her model **5 katlı çapraz doğrulama** ile değerlendirilir; her katta modelin **hiç görmediği** nesneler için tahmin üretilir ve 8.000 eğitim nesnesinin tamamı için bir **katın-dışı (OOF) tahmin** elde edilir. Modeller **dört süreçte paralel** koşar (her model kendi içinde tek iş parçacığı kullanır, süreler karşılaştırılabilir kalır).

````python
def yaris_kos(ad, aile, model, olcekle, kolonlar):
    """Tek modeli 5 katlı CV ile koşar; OOF tahmin ve süre döndürür."""
    t0 = time.perf_counter()
    oof = np.asarray(cross_val_predict(pipe_kur(clone(model), olcekle), X_egitim[kolonlar], y_egitim, cv=KATLAR, n_jobs=1)).ravel().astype(int)   # CatBoost (n,1) döndürür
    return ad, oof, time.perf_counter() - t0


def yaris(kolonlar):
    t0 = time.perf_counter()
    sonuc = Parallel(n_jobs=N_ISCI)(delayed(yaris_kos)(ad, a, m, o, kolonlar) for ad, (a, m, o) in MODELLER.items())
    oofs = {ad: oof for ad, oof, _ in sonuc}
    sureler = {ad: s for ad, _, s in sonuc}
    satirlar = []
    for ad, oof in oofs.items():
        kat_dog = [accuracy_score(y_egitim[va], oof[va]) for _, va in KATLAR]
        m = metrikler(y_egitim, oof)
        satirlar.append({"aile": MODELLER[ad][0], "algoritma": ad, "doğruluk": m["Doğruluk"], "kat std": np.std(kat_dog), "dengeli doğruluk": m["Dengeli doğruluk"],
                         "makro-F1": m["Makro-F1"], "MCC": m["MCC"], "QSO geri çağırma": m["QSO geri çağırma"], "CV süresi (sn)": sureler[ad]})
    tablo = pd.DataFrame(satirlar).sort_values("makro-F1", ascending=False).reset_index(drop=True)
    tablo.insert(0, "sıra", np.arange(1, len(tablo) + 1))
    print(f"✔ Yarış toplam {time.perf_counter() - t0:.0f} sn sürdü")
    return tablo, oofs


tablo_A, oof_A = yaris(SENARYO_A)
display(tablo_A.round(4))
rapor_kaydet(tablo_A, "yaris_senaryo_A_cv", index=False)
````

### 🔬 8.1 · CV ↔ Test: Tutarlı mı?

Tüm modelleri **tüm eğitim kümesiyle** yeniden eğitip hiç dokunulmamış **test setinde** ölçelim. CV ve test birbirine yakınsa, CV'nin tahminleri güvenilirdir.

````python
def test_kos(ad, aile, model, olcekle, kolonlar):
    t0 = time.perf_counter()
    p = pipe_kur(clone(model), olcekle).fit(X_egitim[kolonlar], y_egitim)
    egitim_s = time.perf_counter() - t0
    t1 = time.perf_counter()
    tahmin = np.asarray(p.predict(X_test[kolonlar])).ravel().astype(int)
    return ad, tahmin, egitim_s, time.perf_counter() - t1


def test_degerlendir(kolonlar, tablo_cv):
    sonuc = Parallel(n_jobs=N_ISCI)(delayed(test_kos)(ad, a, m, o, kolonlar) for ad, (a, m, o) in MODELLER.items())
    satirlar, tahminler = [], {}
    for ad, tahmin, e_s, t_s in sonuc:
        m = metrikler(y_test, tahmin)
        tahminler[ad] = tahmin
        satirlar.append({"algoritma": ad, "test doğruluk": m["Doğruluk"], "test makro-F1": m["Makro-F1"], "test QSO geri çağırma": m["QSO geri çağırma"],
                         "eğitim süresi (sn)": e_s, "tahmin süresi (sn)": t_s})
    t = pd.DataFrame(satirlar).merge(tablo_cv[["algoritma", "aile", "sıra", "doğruluk", "makro-F1"]].rename(columns={"doğruluk": "CV doğruluk", "makro-F1": "CV makro-F1"}), on="algoritma")
    t["fark (test − CV) makro-F1"] = t["test makro-F1"] - t["CV makro-F1"]
    return t.sort_values("sıra").reset_index(drop=True), tahminler


test_A, tahmin_test_A = test_degerlendir(SENARYO_A, tablo_A)
display(test_A[["sıra", "aile", "algoritma", "CV doğruluk", "test doğruluk", "CV makro-F1", "test makro-F1", "fark (test − CV) makro-F1", "eğitim süresi (sn)", "tahmin süresi (sn)"]].round(4))
rapor_kaydet(test_A, "yaris_senaryo_A_test", index=False)

fig, eksen = plt.subplots(1, 2, figsize=(15, 5.4))
for ax, (a, b, bas) in zip(eksen, [("CV makro-F1", "test makro-F1", "CV ↔ test makro-F1"), ("eğitim süresi (sn)", "test makro-F1", "Hız ↔ başarı")]):
    for _, r in test_A.iterrows():
        ax.scatter(r[a], r[b], s=90, color=AILE_RENK[r["aile"]], edgecolor="white", zorder=3)
        if r["sıra"] <= 8 or r["algoritma"].startswith("XGBoost ·"):
            ax.annotate(r["algoritma"].split(" (")[0], (r[a], r[b]), fontsize=7.5, xytext=(4, 4), textcoords="offset points")
    ax.set(title=bas, xlabel=a, ylabel=b)
sinir = [test_A["CV makro-F1"].min() - 0.01, 1.0]
eksen[0].plot(sinir, sinir, "--", color=RENK["gri"])
eksen[1].set_xscale("log")
fig.tight_layout()
kaydet(fig, "08_cv_test_ve_hiz")
plt.show()
````

### 🎲 8.2 · Fark Gerçek mi, Şans mı? (Bootstrap)

8.000 nesnenin OOF tahmini üzerinde **nesneleri yeniden örnekleyerek** (bootstrap, 2.000 tekrar) en iyi birkaç modelin **varsayılan XGBoost'tan farkını** ve **%95 güven aralığını** hesaplıyoruz. Aralık **0'ı içeriyorsa** fark şanstan ayırt edilemez.

````python
def bootstrap_fark(oof_a, oof_b, B=2000, seed=RANDOM_STATE):
    """a − b farkı için doğruluk ve makro-F1 bootstrap aralıkları (eşleştirilmiş)."""
    rng = np.random.default_rng(seed)
    n = len(y_egitim)
    farklar = np.empty((B, 2))
    for k in range(B):
        i = rng.integers(0, n, n)
        farklar[k] = ((oof_a[i] == y_egitim[i]).mean() - (oof_b[i] == y_egitim[i]).mean(),
                      makro_f1_hizli(y_egitim[i], oof_a[i]) - makro_f1_hizli(y_egitim[i], oof_b[i]))
    return np.percentile(farklar, [2.5, 97.5], axis=0), farklar.mean(0)


referans = "XGBoost · varsayılan"
satirlar = []
for ad in tablo_A.head(8)["algoritma"]:
    if ad == referans:
        continue
    (alt, ust), ort = bootstrap_fark(oof_A[ad], oof_A[referans])
    satirlar.append({"model": ad, "Δ doğruluk": ort[0], "Δ doğruluk %95 alt": alt[0], "Δ doğruluk %95 üst": ust[0],
                     "Δ makro-F1": ort[1], "Δ makro-F1 alt": alt[1], "Δ makro-F1 üst": ust[1],
                     "ayırt edilebilir mi?": "evet" if (alt[1] > 0 or ust[1] < 0) else "hayır (şans)"})
bs_A = pd.DataFrame(satirlar)
display(bs_A.round(4))
rapor_kaydet(bs_A, "bootstrap_senaryo_A", index=False)

fig, ax = plt.subplots(figsize=(10, 4.2))
for i, r in bs_A.iterrows():
    ax.plot([r["Δ makro-F1 alt"], r["Δ makro-F1 üst"]], [i, i], color=RENK["mor"], lw=3)
    ax.scatter(r["Δ makro-F1"], i, s=90, color=RENK["altin"], edgecolor="black", zorder=3)
ax.axvline(0, color="black", lw=1)
ax.set_yticks(range(len(bs_A)))
ax.set_yticklabels(bs_A["model"], fontsize=9)
ax.set(title=f"'{referans}' ile farkın %95 aralığı (makro-F1) · 0'ı kesen aralık = şans", xlabel="Δ makro-F1")
kaydet(fig, "09_bootstrap_fark")
plt.show()
````

Bu bölümde **ana senaryoda (B: yalnız fotometri)** XGBoost'un düğmelerini **tek tek** çevirip etkisini görüyoruz. Ayar için eğitim kümesini **%80 / %20** olarak ikiye bölüyoruz (`egitim_i` / `dogrulama_i`); **test seti hâlâ kilitli**.

````python
idx_e, idx_d = train_test_split(np.arange(len(X_egitim)), test_size=0.2, stratify=y_egitim, random_state=RANDOM_STATE)
Xe = renk_indeksleri(X_egitim.iloc[idx_e][SENARYO_B]); ye = y_egitim[idx_e]
Xd = renk_indeksleri(X_egitim.iloc[idx_d][SENARYO_B]); yd = y_egitim[idx_d]
print(f"İç eğitim: {len(Xe):,} · iç doğrulama: {len(Xd):,} · özellikler: {list(Xe.columns)}")

TEMEL = dict(objective="multi:softprob", n_jobs=N_ISCI, random_state=RANDOM_STATE, eval_metric=["mlogloss", "merror"])


def xgb_egit(**ek):
    """Verilen ayarla XGBoost'u iç eğitimde eğitir; eğitim ve doğrulama eğrilerini kaydeder."""
    m = XGBClassifier(**{**TEMEL, **ek})
    m.fit(Xe, ye, eval_set=[(Xe, ye), (Xd, yd)], verbose=False)
    return m


def ozet(m, ad, **ek):
    gec = m.evals_result()
    kayip = np.array(gec["validation_1"]["mlogloss"])
    best = int(kayip.argmin())
    p = m.predict(Xd)
    return {"ayar": ad, **ek, "en iyi tur": best + 1, "en düşük doğrulama log-loss": kayip.min(), "son tur doğrulama log-loss": kayip[-1],
            "doğrulama doğruluğu": accuracy_score(yd, p), "doğrulama makro-F1": f1_score(yd, p, average="macro")}
````

### 🪜 10.3 · Derinlik × Çocuk Ağırlığı

`max_depth` ağacın ne kadar **karmaşık** soru sorabileceğini, `min_child_weight` ise bir yaprağın **en az ne kadar bilgi** (Hessian toplamı = *cover*) taşıması gerektiğini belirler. Kavram dersindeki **cover** budaması bu parametredir.

````python
satir_d = []
for d in [2, 3, 4, 6, 8, 10]:
    for mcw in [1, 5, 20]:
        m = xgb_egit(n_estimators=250, learning_rate=0.1, max_depth=d, min_child_weight=mcw)
        satir_d.append(ozet(m, f"derinlik {d}, mcw {mcw}", derinlik=d, min_child_weight=mcw))
derin = pd.DataFrame(satir_d)
rapor_kaydet(derin, "xgb_derinlik_cocuk_agirligi", index=False)

fig, eksen = plt.subplots(1, 2, figsize=(14, 4.2))
for ax, kol, bas in zip(eksen, ["en düşük doğrulama log-loss", "doğrulama makro-F1"], ["En düşük doğrulama log-loss (düşük = iyi)", "Doğrulama makro-F1 (yüksek = iyi)"]):
    p = derin.pivot(index="derinlik", columns="min_child_weight", values=kol)
    sns.heatmap(p, annot=True, fmt=".3f", cmap="viridis_r" if "log" in kol else "viridis", ax=ax, cbar=False)
    ax.set(title=bas, xlabel="min_child_weight", ylabel="max_depth")
fig.tight_layout()
kaydet(fig, "16_xgb_derinlik_cocuk_agirligi")
plt.show()
````

### 🎚️ 11.1 · Rastgele Arama (XGBoost ve LightGBM)

Referans notebook'taki gibi ızgara araması yapabilirdik; ama 9 düğmeli bir uzayda ızgara **çok pahalıdır**. **`RandomizedSearchCV`** her denemede düğmeleri **dağılımlardan** çeker (Bergstra ve Bengio, 2012) ve aynı bütçeyle uzayı çok daha iyi tarar. **5 katlı CV**, ölçüt **makro-F1**; arama yalnızca **eğitim kümesinde** yapılır.

````python
xgb_uzay = {"model__n_estimators": randint(150, 900), "model__learning_rate": loguniform(0.02, 0.3), "model__max_depth": randint(2, 9),
            "model__min_child_weight": loguniform(0.5, 10), "model__subsample": uniform(0.6, 0.4), "model__colsample_bytree": uniform(0.5, 0.5),
            "model__reg_lambda": loguniform(0.1, 30), "model__reg_alpha": loguniform(1e-3, 5), "model__gamma": uniform(0, 1.5)}
lgbm_uzay = {"model__n_estimators": randint(150, 900), "model__learning_rate": loguniform(0.02, 0.3), "model__num_leaves": randint(8, 80),
             "model__min_child_samples": randint(5, 60), "model__subsample": uniform(0.6, 0.4), "model__subsample_freq": [1],
             "model__colsample_bytree": uniform(0.5, 0.5), "model__reg_lambda": loguniform(0.1, 30), "model__reg_alpha": loguniform(1e-3, 5)}

N_DENEME = 40
t0 = time.perf_counter()
arama_xgb = RandomizedSearchCV(pipe_kur(XGBClassifier(n_jobs=1, random_state=RANDOM_STATE)), xgb_uzay, n_iter=N_DENEME, scoring="f1_macro", cv=KATLAR,
                               n_jobs=N_ISCI, random_state=RANDOM_STATE, refit=False).fit(X_egitim[SENARYO_B], y_egitim)
print(f"XGBoost araması: {time.perf_counter() - t0:.0f} sn")
t0 = time.perf_counter()
arama_lgbm = RandomizedSearchCV(pipe_kur(LGBMClassifier(n_jobs=1, verbose=-1, random_state=RANDOM_STATE)), lgbm_uzay, n_iter=N_DENEME, scoring="f1_macro", cv=KATLAR,
                                n_jobs=N_ISCI, random_state=RANDOM_STATE, refit=False).fit(X_egitim[SENARYO_B], y_egitim)
print(f"LightGBM araması: {time.perf_counter() - t0:.0f} sn")


def arama_tablosu(arama):
    t = pd.DataFrame(arama.cv_results_)
    t = t[[c for c in t.columns if c.startswith("param_") or c in ("mean_test_score", "std_test_score", "rank_test_score", "mean_fit_time")]]
    t.columns = [c.replace("param_model__", "") for c in t.columns]
    return t.sort_values("rank_test_score").reset_index(drop=True)


xgb_arama_t, lgbm_arama_t = arama_tablosu(arama_xgb), arama_tablosu(arama_lgbm)
display(xgb_arama_t.head(8).round(4))
rapor_kaydet(xgb_arama_t, "xgb_arama_sonuclari", index=False)
rapor_kaydet(lgbm_arama_t, "lgbm_arama_sonuclari", index=False)
````

### 🔍 11.2 · Taze Katlarda Doğrulama ve Final Adayları

Aramada **en iyi çıkan** ayar, aynı katlarda seçildiği için **şanslı** olabilir (kazananın laneti). Bu yüzden her aramanın **ilk 5 adayını** ve varsayılanları **hiç kullanılmamış taze katlarda** (`RepeatedStratifiedKFold`, 3 tekrar × 5 kat) yeniden ölçüp **taze skora** göre seçiyoruz.

````python
TAZE = list(RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=2024).split(X_egitim, y_egitim))


def aday_kur(tur, satir):
    p = {c: satir[c] for c in satir.index if c not in ("mean_test_score", "std_test_score", "rank_test_score", "mean_fit_time")}
    tamsayi = {"n_estimators", "max_depth", "num_leaves", "min_child_samples", "subsample_freq"}
    p = {k: (int(v) if k in tamsayi else float(v)) for k, v in p.items()}
    if tur == "xgb":
        return pipe_kur(XGBClassifier(n_jobs=1, random_state=RANDOM_STATE, **p))
    return pipe_kur(LGBMClassifier(n_jobs=1, verbose=-1, random_state=RANDOM_STATE, **p))


adaylar = {"XGBoost · varsayılan": pipe_kur(XGBClassifier(n_jobs=1, random_state=RANDOM_STATE)),
           "LightGBM · varsayılan": pipe_kur(LGBMClassifier(n_jobs=1, verbose=-1, random_state=RANDOM_STATE)),
           "CatBoost · varsayılan": pipe_kur(CatBoostClassifier(iterations=500, thread_count=1, verbose=0, random_seed=RANDOM_STATE))}
for i in range(5):
    adaylar[f"XGBoost · arama #{i + 1}"] = aday_kur("xgb", xgb_arama_t.iloc[i])
    adaylar[f"LightGBM · arama #{i + 1}"] = aday_kur("lgbm", lgbm_arama_t.iloc[i])


def taze_degerlendir(ad, pipe):
    skorlar = []
    for tr, va in TAZE:
        p = clone(pipe).fit(X_egitim.iloc[tr][SENARYO_B], y_egitim[tr])
        skorlar.append(f1_score(y_egitim[va], np.asarray(p.predict(X_egitim.iloc[va][SENARYO_B])).ravel(), average="macro"))
    return {"aday": ad, "taze makro-F1 (ort.)": np.mean(skorlar), "std": np.std(skorlar)}


taze = pd.DataFrame(Parallel(n_jobs=N_ISCI)(delayed(taze_degerlendir)(a, p) for a, p in adaylar.items())).sort_values("taze makro-F1 (ort.)", ascending=False).reset_index(drop=True)
display(taze.round(4))
rapor_kaydet(taze, "final_adaylari_taze_katlar", index=False)

fig, ax = plt.subplots(figsize=(10, 6))
t = taze.iloc[::-1]
ax.barh(t["aday"], t["taze makro-F1 (ort.)"], xerr=t["std"], color=[RENK["turuncu"] if "XGBoost" in a else (RENK["yesil"] if "LightGBM" in a else RENK["mor"]) for a in t["aday"]])
ax.set(xlim=t["taze makro-F1 (ort.)"].min() - 0.01, title="Taze katlarda (3×5) makro-F1: varsayılan mı, ayarlı mı?", xlabel="makro-F1 (ortalama ± std)")
kaydet(fig, "20_final_adaylari_taze_katlar")
plt.show()
````

### 🏆 11.3 · Final Modelin Seçimi

Final model, **taze katlardaki** en iyi XGBoost adayıdır (bu notebook'un konusu XGBoost olduğu için XGBoost ailesi içinden seçiyoruz; LightGBM ve CatBoost karşılaştırma içindir). Ek olarak üç modelin **olasılık ortalaması (Voting)** da bir rakip olarak sınanır.

````python
en_iyi_xgb_ad = taze[taze["aday"].str.startswith("XGBoost")].iloc[0]["aday"]
FINAL_PIPE = clone(adaylar[en_iyi_xgb_ad])
FINAL_AD = en_iyi_xgb_ad
print("Final XGBoost adayı:", FINAL_AD)
xgb_parametre = {k.replace("model__", ""): v for k, v in FINAL_PIPE.get_params().items() if k.startswith("model__") and k.replace("model__", "") in
                 ("n_estimators", "learning_rate", "max_depth", "min_child_weight", "subsample", "colsample_bytree", "reg_lambda", "reg_alpha", "gamma")}
display(pd.Series(xgb_parametre, name="değer").to_frame().round(4))

# Rakip: ayarlı XGBoost + ayarlı LightGBM + CatBoost (soft voting)
en_iyi_lgbm_ad = taze[taze["aday"].str.startswith("LightGBM")].iloc[0]["aday"]
VOTING_PIPE = Pipeline([("renk", FunctionTransformer(renk_indeksleri)), ("model", VotingClassifier(
    estimators=[("xgb", clone(adaylar[FINAL_AD]).named_steps["model"]), ("lgbm", clone(adaylar[en_iyi_lgbm_ad]).named_steps["model"]),
                ("cat", CatBoostClassifier(iterations=500, thread_count=1, verbose=0, random_seed=RANDOM_STATE))], voting="soft"))])


def final_fit(ad, pipe):
    return ad, clone(pipe).fit(X_egitim[SENARYO_B], y_egitim)


modeller_final = dict(Parallel(n_jobs=N_ISCI)(delayed(final_fit)(a, p) for a, p in
                      {"XGBoost · varsayılan": adaylar["XGBoost · varsayılan"], f"XGBoost · ayarlı": FINAL_PIPE, "LightGBM · ayarlı": adaylar[en_iyi_lgbm_ad],
                       "CatBoost · varsayılan": adaylar["CatBoost · varsayılan"], "Voting (XGB + LGBM + Cat, ayarlı)": VOTING_PIPE}.items()))
satir_f = []
for ad, m in modeller_final.items():
    olas = m.predict_proba(X_test[SENARYO_B])
    satir_f.append({"model": ad, **metrikler(y_test, olas.argmax(1), olas)})
final_tablo = pd.DataFrame(satir_f)
display(final_tablo.round(4))
rapor_kaydet(final_tablo, "final_model_karsilastirma_test", index=False)

# Varsayılan ↔ ayarlı: test setinde eşleştirilmiş bootstrap
p_var = modeller_final["XGBoost · varsayılan"].predict(X_test[SENARYO_B])
p_ay = modeller_final["XGBoost · ayarlı"].predict(X_test[SENARYO_B])
rng = np.random.default_rng(RANDOM_STATE)
fark = np.array([(lambda i: (accuracy_score(y_test[i], p_ay[i]) - accuracy_score(y_test[i], p_var[i]),
                             makro_f1_hizli(y_test[i], p_ay[i]) - makro_f1_hizli(y_test[i], p_var[i])))(rng.integers(0, len(y_test), len(y_test))) for _ in range(2000)])
aralik = np.percentile(fark, [2.5, 97.5], axis=0)
display(Markdown(f"""**Ayarlı − varsayılan XGBoost (test, 2.000 nesne):** Δ doğruluk = **{fark[:, 0].mean():+.4f}** (%95 aralık [{aralik[0, 0]:+.4f}; {aralik[1, 0]:+.4f}]) ·
Δ makro-F1 = **{fark[:, 1].mean():+.4f}** (%95 aralık [{aralik[0, 1]:+.4f}; {aralik[1, 1]:+.4f}]). Aralık sıfırı {'içeriyor → fark şanstan ayırt edilemez' if aralik[0, 1] <= 0 <= aralik[1, 1] else 'içermiyor → fark gerçek'}."""))
````

Final adayımız (ayarlı XGBoost) **neye bakarak** karar veriyor? Dört farklı bakış açısı:

| Yöntem | Soruya cevabı |
|:--|:--|
| **`weight`** | Özellik kaç **bölünmede** kullanıldı? |
| **`gain`** | Kullanıldığı bölünmelerin **ortalama kazancı** (kavram dersindeki gain) |

| **Permütasyon** | Özelliği **karıştırırsak** test başarısı ne kadar düşer? |
| **TreeSHAP** | Her tahmin için her özelliğin **katkısı** (Lundberg vd., 2020) |

> ⚠ Önem puanları **nedensellik değil**, modelin **kullanım biçimini** gösterir; birbiriyle ilişkili özelliklerde (renkler!) önem **paylaşılır**.

````python
FINAL_FIT = modeller_final["XGBoost · ayarlı"]
booster = FINAL_FIT.named_steps["model"].get_booster()
OZELLIKLER = list(renk_indeksleri(X_test[SENARYO_B].head(2)).columns)
print("Özellikler:", OZELLIKLER)

onem = pd.DataFrame({t: pd.Series(booster.get_score(importance_type=t)) for t in ["weight", "gain", "cover", "total_gain"]}).reindex(OZELLIKLER).fillna(0)
onem_oran = onem / onem.sum()
display(onem.round(2))
rapor_kaydet(onem, "ozellik_onemi_xgboost")

fig, eksen = plt.subplots(1, 4, figsize=(18, 4.2), sharey=True)
for ax, t in zip(eksen, onem_oran.columns):
    s = onem_oran[t].sort_values()
    ax.barh(s.index, s.values, color=RENK["turuncu"])
    ax.set(title=f"importance_type = {t}", xlabel="pay")
fig.tight_layout()
kaydet(fig, "21_ozellik_onemi_turleri")
plt.show()
````

````python
# Permütasyon önemi: ham sütunları test setinde karıştırarak makro-F1 kaybı (10 tekrar)
pi = permutation_importance(FINAL_FIT, X_test[SENARYO_B], y_test, scoring="f1_macro", n_repeats=10, random_state=RANDOM_STATE, n_jobs=N_ISCI)
perm = pd.DataFrame({"ort. makro-F1 kaybı": pi.importances_mean, "std": pi.importances_std}, index=SENARYO_B).sort_values("ort. makro-F1 kaybı", ascending=False)
display(perm.round(4))
rapor_kaydet(perm, "permutasyon_onemi")

# TreeSHAP: XGBoost'un yerleşik 'pred_contribs' özelliği ile (ek kütüphane gerekmez)
Xt = renk_indeksleri(X_test[SENARYO_B]).reset_index(drop=True)
katki = booster.predict(xgboost.DMatrix(Xt), pred_contribs=True)         # şekil: (nesne, sınıf, özellik + sapma)
ham_marj = booster.predict(xgboost.DMatrix(Xt), output_margin=True)
print("SHAP katkı şekli:", katki.shape, "· toplamsallık kontrolü (Σ katkı = ham çıktı) en büyük fark:", f"{np.abs(katki.sum(2) - ham_marj).max():.2e}")

shap_ort = pd.DataFrame({SINIFLAR[k]: np.abs(katki[:, k, :-1]).mean(0) for k in range(3)}, index=OZELLIKLER)
fig, eksen = plt.subplots(1, 2, figsize=(15, 5.2), gridspec_kw={"width_ratios": [1, 1.5]})
sns.heatmap(shap_ort.loc[shap_ort.sum(1).sort_values(ascending=False).index], annot=True, fmt=".2f", cmap="YlOrRd", ax=eksen[0], cbar=False)
eksen[0].set(title="Ortalama |SHAP|: sınıf başına özellik katkısı")
# QSO sınıfı için 'arı sürüsü' benzeri grafik: x = SHAP değeri, renk = özellik değeri
sira = list(shap_ort["QSO"].sort_values(ascending=False).index[:6])[::-1]
for j, f in enumerate(sira):
    k_f = OZELLIKLER.index(f)
    sv = katki[:, 1, k_f]
    deger = Xt[f].rank(pct=True).to_numpy()
    eksen[1].scatter(sv, j + np.random.default_rng(j).uniform(-0.28, 0.28, len(sv)), c=deger, cmap="coolwarm", s=7, alpha=0.65)
eksen[1].set_yticks(range(len(sira)))
eksen[1].set_yticklabels(sira)
eksen[1].axvline(0, color="black", lw=0.8)
eksen[1].set(title="QSO (kuasar) çıktısı için SHAP dağılımı (renk: özellik değeri, mavi = düşük · kırmızı = yüksek)", xlabel="SHAP (log-odds katkısı)")
eksen[1].title.set_fontsize(10.5)
fig.tight_layout()
kaydet(fig, "22_shap_ozet")
plt.show()
rapor_kaydet(shap_ort, "shap_ortalama_mutlak")
````

### 🔬 13.2 · Segment Denetimi: Hangi Alt Grupta Daha Çok Yanılıyoruz?

Tek bir genel skor, alt gruplardaki **eşitsizlikleri gizler**. Test setini parlaklık dilimlerine, gerçek sınıfa ve gökyüzü bölgesine ayırıp doğruluğu (%95 Wilson aralığıyla) bakıyoruz.

````python
def wilson(k, n, z=1.96):
    if n == 0:
        return np.nan, np.nan
    p = k / n
    d = 1 + z ** 2 / n
    merkez = (p + z ** 2 / (2 * n)) / d
    yari = z * np.sqrt(p * (1 - p) / n + z ** 2 / (4 * n ** 2)) / d
    return merkez - yari, merkez + yari


seg = X_test[SENARYO_B + ["dec"]].copy().reset_index(drop=True)
seg["dogru"] = (tahmin_test == y_test)
seg["gercek"] = [SINIFLAR[i] for i in y_test]
seg["r dilimi"] = pd.qcut(seg["r"], 5, labels=["en parlak", "parlak", "orta", "soluk", "en soluk"])
seg["bölge"] = np.where(seg["dec"] > 10, "kuzey yama (dec > 10°)", "ekvator şeridi (dec ≤ 10°)")
seg["güven"] = pd.cut(olas_test.max(1), [0, 0.6, 0.8, 0.95, 0.99, 1.0], labels=["≤0,60", "0,60–0,80", "0,80–0,95", "0,95–0,99", "> 0,99"])

satir_s = []
for kolon in ["r dilimi", "gercek", "bölge", "güven"]:
    for deger, g in seg.groupby(kolon, observed=True):
        n, k = len(g), int(g["dogru"].sum())
        alt, ust = wilson(k, n)
        satir_s.append({"segment": kolon, "değer": str(deger), "n": n, "doğruluk": k / n, "alt %95": alt, "üst %95": ust, "hata sayısı": n - k})
segment = pd.DataFrame(satir_s)
display(segment.round(4))
rapor_kaydet(segment, "segment_denetimi", index=False)

fig, eksen = plt.subplots(1, 4, figsize=(19, 4.2), sharey=True)
for ax, kolon in zip(eksen, ["r dilimi", "gercek", "bölge", "güven"]):
    d_ = segment[segment["segment"] == kolon]
    ax.bar(d_["değer"], d_["doğruluk"], yerr=[d_["doğruluk"] - d_["alt %95"], d_["üst %95"] - d_["doğruluk"]], capsize=4, color=RENK["turuncu"])
    ax.set(title=kolon, ylim=(0.6, 1.02))
    ax.tick_params(axis="x", labelrotation=20, labelsize=8)
    for i, (v, n) in enumerate(zip(d_["doğruluk"], d_["n"])):
        ax.text(i, 0.62, f"n={n}", ha="center", fontsize=8)
eksen[0].set_ylabel("doğruluk")
fig.suptitle("Segment denetimi: alt grup doğrulukları (%95 Wilson aralığı)", fontweight="bold", y=1.04)
fig.tight_layout()
kaydet(fig, "26_segment_denetimi")
plt.show()
````

### 📐 14.1 · Olasılıklar Güvenilir mi? (Kalibrasyon)

"%90 kuasar" dediğimizde bu nesnelerin gerçekten **%90'ı** kuasar mı? Bunu **güvenilirlik (reliability) diyagramı** ve **ECE** (*expected calibration error*) ile ölçeriz. Gerekirse **isotonic kalibrasyon** ile düzeltiriz (Niculescu-Mizil ve Caruana, 2005).

````python
def ece_top(y, olas, n_kutu=10):
    """Üst-etiket (top-label) beklenen kalibrasyon hatası."""
    guven, tahmin = olas.max(1), olas.argmax(1)
    dogru = (tahmin == y).astype(float)
    kutu = np.minimum((guven * n_kutu).astype(int), n_kutu - 1)
    e = 0.0
    for b in range(n_kutu):
        m = kutu == b
        if m.any():
            e += m.mean() * abs(dogru[m].mean() - guven[m].mean())
    return e


def brier_cok(y, olas):
    Y = np.eye(3)[y]
    return ((olas - Y) ** 2).sum(1).mean()


kal = CalibratedClassifierCV(clone(FINAL_PIPE), method="isotonic", cv=5).fit(X_egitim[SENARYO_B], y_egitim)
olas_kal = kal.predict_proba(X_test[SENARYO_B])

satir_k = []
for ad, o in [("Ham XGBoost (ayarlı)", olas_test), ("İsotonic kalibreli", olas_kal)]:
    satir_k.append({"olasılık": ad, "log-loss": log_loss(y_test, o), "Brier (çok sınıflı)": brier_cok(y_test, o), "ECE (üst-etiket)": ece_top(y_test, o),
                    "doğruluk": accuracy_score(y_test, o.argmax(1))})
kalibrasyon = pd.DataFrame(satir_k)
display(kalibrasyon.round(4))
rapor_kaydet(kalibrasyon, "kalibrasyon", index=False)
display(Markdown(f"**Okuma:** Ham ayarlı XGBoost zaten **iyi kalibre** (ECE = {kalibrasyon.iloc[0]['ECE (üst-etiket)']:.4f}); isotonic kalibrasyon ECE'yi çok az değiştiriyor ({kalibrasyon.iloc[1]['ECE (üst-etiket)']:.4f}) ve log-loss'u **kötüleştiriyor** ({kalibrasyon.iloc[0]['log-loss']:.4f} → {kalibrasyon.iloc[1]['log-loss']:.4f}). Yani burada kalibrasyon **gerekmiyor**: her zaman uygulanacak bir adım değil, **ölçülüp karar verilecek** bir adımdır."))

fig, eksen = plt.subplots(1, 2, figsize=(13, 4.8))
for ad, o, r in [("ham", olas_test, RENK["pembe"]), ("isotonic", olas_kal, RENK["mavi"])]:
    guven, dogru = o.max(1), (o.argmax(1) == y_test)
    kutu = np.minimum((guven * 10).astype(int), 9)
    xs = [guven[kutu == b].mean() for b in range(10) if (kutu == b).sum() >= 5]
    ys = [dogru[kutu == b].mean() for b in range(10) if (kutu == b).sum() >= 5]
    eksen[0].plot(xs, ys, "o-", color=r, lw=2.4, label=f"{ad} (ECE = {ece_top(y_test, o):.4f})")
eksen[0].plot([0.3, 1], [0.3, 1], "--", color=RENK["gri"], label="mükemmel kalibrasyon")
eksen[0].set(title="Güvenilirlik diyagramı (üst-etiket)", xlabel="modelin güveni", ylabel="gerçek doğruluk")
eksen[0].legend()
# Yalnız QSO için bire karşı diğerleri
for ad, o, r in [("ham", olas_test, RENK["pembe"]), ("isotonic", olas_kal, RENK["mavi"])]:
    p_q, y_q = o[:, 1], (y_test == 1).astype(int)
    kutu = np.minimum((p_q * 10).astype(int), 9)
    xs = [p_q[kutu == b].mean() for b in range(10) if (kutu == b).sum() >= 5]
    ys = [y_q[kutu == b].mean() for b in range(10) if (kutu == b).sum() >= 5]
    eksen[1].plot(xs, ys, "o-", color=r, lw=2.4, label=ad)
eksen[1].plot([0, 1], [0, 1], "--", color=RENK["gri"])
eksen[1].set(title="Yalnız kuasar (QSO) olasılığı", xlabel="tahmin edilen P(QSO)", ylabel="gözlenen kuasar oranı")
eksen[1].legend()
fig.tight_layout()
kaydet(fig, "27_kalibrasyon")
plt.show()
````

### 🔭 14.2 · Spektroskopi Bütçesi: Hangi Nesnelere Bakmalı?

Teleskop zamanı kısıtlı: test setindeki 2.000 nesnenin **yalnızca K tanesi** için spektrum alabiliriz ve hedefimiz **kuasar bulmak**. Nesneleri **P(QSO)** değerine göre sıralayıp en üstteki K tanesini seçersek, rastgele seçime göre ne kazanırız? (**Lift**)

````python
p_qso = olas_test[:, 1]
n_qso = int((y_test == 1).sum())
taban_oran = n_qso / len(y_test)
sira = np.argsort(-p_qso)

satir_b = []
for K in [50, 100, 150, n_qso, 250, 400]:
    secilen = sira[:K]
    bulunan = int((y_test[secilen] == 1).sum())
    satir_b.append({"bütçe K": K, "bulunan kuasar": bulunan, "kesinlik (precision@K)": bulunan / K, "geri çağırma (recall@K)": bulunan / n_qso,
                    "rastgele seçimle beklenen": K * taban_oran, "lift (kat)": (bulunan / K) / taban_oran})
butce = pd.DataFrame(satir_b)
display(butce.round(3))
rapor_kaydet(butce, "spektroskopi_butcesi", index=False)

# Kazanç eğrisi (cumulative gain): model ↔ rastgele ↔ ideal
kum = np.cumsum(y_test[sira] == 1) / n_qso
fig, ax = plt.subplots(figsize=(8.5, 4.6))
ax.plot(np.arange(1, len(sira) + 1) / len(sira), kum, color=RENK["pembe"], lw=3, label="XGBoost (P(QSO) sıralaması)")
ax.plot([0, 1], [0, 1], "--", color=RENK["gri"], label="rastgele seçim")
ax.plot([0, taban_oran, 1], [0, 1, 1], ":", color=RENK["altin"], lw=2.4, label="ideal sıralama")
ax.set(title="Kazanç eğrisi: nesnelerin ilk %X'i seçilirse kuasarların kaçı bulunur?", xlabel="seçilen nesne oranı", ylabel="bulunan kuasar oranı")
ax.legend(loc="lower right")
kaydet(fig, "28_kazanc_egrisi")
plt.show()
````

### 💶 14.3 · Maliyete Duyarlı Eşik

Her hatanın bedeli aynı değil: **bir kuasarı kaçırmak** (spektroskopik bir keşfi kaybetmek) ile **yanlış bir kuasar adayı için spektrum harcamak** farklı maliyetlere sahip. Örnek varsayım: **kaçırılan kuasar = 5 birim**, **boşa harcanan spektrum = 1 birim**. P(QSO) için eşik **τ**'yu **eğitim OOF olasılıklarında** (test'e bakmadan) seçip test'te sınıyoruz.

````python
C_KACIRMA, C_BOSA = 5.0, 1.0
oof_olas = cross_val_predict(clone(FINAL_PIPE), X_egitim[SENARYO_B], y_egitim, cv=KATLAR, method="predict_proba", n_jobs=N_ISCI)


def maliyet(y, p_q, tau):
    secilen = p_q >= tau
    fn = int(((y == 1) & ~secilen).sum())
    fp = int(((y != 1) & secilen).sum())
    return C_KACIRMA * fn + C_BOSA * fp, fn, fp


taular = np.linspace(0.02, 0.98, 97)
oof_maliyet = np.array([maliyet(y_egitim, oof_olas[:, 1], t)[0] for t in taular]) / len(y_egitim)
TAU = float(taular[oof_maliyet.argmin()])
tau_varsayilan = 0.5

satir_c = []
for ad, t in [("varsayılan eşik (τ = 0,5)", tau_varsayilan), (f"maliyet-en-iyi eşik (τ = {TAU:.2f})", TAU)]:
    c, fn, fp = maliyet(y_test, p_qso, t)
    secilen = p_qso >= t
    satir_c.append({"karar kuralı": ad, "kaçırılan kuasar (FN)": fn, "boşa spektrum (FP)": fp, "toplam maliyet": c, "kişi başı maliyet": c / len(y_test),
                    "kuasar geri çağırma": ((y_test == 1) & secilen).sum() / n_qso, "kuasar kesinlik": ((y_test == 1) & secilen).sum() / max(secilen.sum(), 1)})
satir_c.append({"karar kuralı": "hiç spektrum alma", "kaçırılan kuasar (FN)": n_qso, "boşa spektrum (FP)": 0, "toplam maliyet": C_KACIRMA * n_qso, "kişi başı maliyet": C_KACIRMA * n_qso / len(y_test),
                "kuasar geri çağırma": 0.0, "kuasar kesinlik": np.nan})
satir_c.append({"karar kuralı": "hepsi için spektrum al", "kaçırılan kuasar (FN)": 0, "boşa spektrum (FP)": len(y_test) - n_qso, "toplam maliyet": C_BOSA * (len(y_test) - n_qso),
                "kişi başı maliyet": C_BOSA * (len(y_test) - n_qso) / len(y_test), "kuasar geri çağırma": 1.0, "kuasar kesinlik": taban_oran})
karar = pd.DataFrame(satir_c)
display(karar.round(3))
rapor_kaydet(karar, "maliyete_duyarli_karar", index=False)

test_maliyet = np.array([maliyet(y_test, p_qso, t)[0] for t in taular]) / len(y_test)
fig, ax = plt.subplots(figsize=(9, 4.4))
ax.plot(taular, oof_maliyet, color=RENK["mavi"], lw=2.8, label="eğitim OOF (τ buradan seçildi)")
ax.plot(taular, test_maliyet, color=RENK["pembe"], lw=2.8, label="test (yalnızca doğrulama)")
ax.axvline(TAU, color=RENK["altin"], lw=2.4, ls="--", label=f"seçilen τ = {TAU:.2f}")
ax.set(title="Eşiğe göre nesne başına beklenen maliyet (kaçırma = 5, boşa spektrum = 1)", xlabel="P(QSO) eşiği τ", ylabel="nesne başına maliyet")
ax.legend()
kaydet(fig, "29_maliyet_esigi")
plt.show()
````

### ❓ Quiz (cevaplar açılır kutularda)

<details><summary><b>1.</b> Redshift neden "neredeyse etiketin kendisi" sayılır ve Senaryo B'de neden kullanılmaz?</summary>

Redshift, etiketin üretildiği **aynı spektrumdan** ölçülür; yıldızlar için ≈ 0'dır. Fotometri senaryosunda **spektrum henüz yoktur**, bu yüzden redshift **bilinmez**; kullanmak gerçek kullanımda olmayacak bir bilgiyi modele vermek olurdu.
</details>

<details><summary><b>2.</b> Plaka bazlı doğrulama ne için yapıldı? Sonuç ne söyledi?</summary>

Gözlem kimlikleri (plate, mjd, fiberid) sınıfı **ezberletiyor mu** diye sınamak için. Aynı plakadaki nesneler aynı katta tutulduğunda skor **düşmedi**; yani bu veride belirgin bir sızıntı **bulunmadı**. Kimlikleri yine de çıkardık, çünkü fiziksel girdi değiller.
</details>

<details><summary><b>3.</b> Dengesiz sınıfta neden yalnızca doğruluğa bakmıyoruz?</summary>

"Hep galaksi de" bile yüksek doğruluk verir; nadir sınıf (kuasar) gözden kaçar. **Makro-F1**, dengeli doğruluk, MCC ve **sınıf bazlı geri çağırma** birlikte izlenir.
</details>

<details><summary><b>4.</b> Üst sıradaki iki modelin makro-F1 farkının bootstrap aralığı sıfırı kesiyorsa ne anlama gelir?</summary>

Fark **şanstan ayırt edilemez**; hangisinin "gerçekten daha iyi" olduğunu söyleyemeyiz. Sıralama bir tahmindir, kesinlik değil.
</details>

<details><summary><b>5.</b> λ, α, γ'nın görevi nedir?</summary>

λ (L2) yaprak değerlerini küçültür; α (L1) küçük değerleri sıfıra iter; γ asgari gain şartıdır (bölünme gain'i γ'dan küçükse yapılmaz). Üçü de aşırı öğrenmeyi azaltır.
</details>

<details><summary><b>6.</b> Maliyete duyarlı eşik neden 0,5'ten farklı çıkabilir?</summary>

Kuasarı kaçırmak yanlış alarmdan **daha pahalı** olduğu için P(QSO) eşiği düşürülür; model daha cömert seçim yapar. Eşik iş maliyetine göre eğitim OOF olasılıklarında seçilir.
</details>

<details><summary><b>7.</b> `TreeSHAP` toplamsallık kontrolü ne söyler?</summary>

Bir nesne için tüm özellik katkıları ve sapma toplanırsa modelin **ham çıktısı (log-odds)** elde edilir; katkılar böylece **tutarlı** bir açıklama sunar.
</details>
