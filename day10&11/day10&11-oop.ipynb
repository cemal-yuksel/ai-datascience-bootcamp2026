## 💼 **YBS ÖRNEĞİ: MÜŞTERİ CLASS'I**

### 🎓 **Adım 1: En Basit Class Tanımlama**

Class tanımlamak için `class` anahtar kelimesini kullanırız:

````python
# 🎨 EN BASİT CLASS TANIMI
# Class ismi her zaman BÜYÜK HARFLE başlar (Python convention)

class Musteri:
    pass  # "pass" = "şimdilik boş bırak" demek

print("✅ İlk class'ımızı oluşturduk!")
print(f"🔍 Musteri class'ının tipi: {type(Musteri)}")
````

---

### 🎓 **Adım 2: Object (Nesne) Oluşturma**

Class'ı tanımladık, şimdi bu şablondan **gerçek nesneler** oluşturalım!

> 💡 **ÖNEMLİ:** Class = Tarif Defteri, Object = Yaptığınız Yemek!

````python
# 🏗️ OBJECT (NESNE) OLUŞTURMA
# Class'tan nesne oluşturmak için class adını fonksiyon gibi çağırırız

musteri1 = Musteri()  # İlk müşteri nesnesi
musteri2 = Musteri()  # İkinci müşteri nesnesi
musteri3 = Musteri()  # Üçüncü müşteri nesnesi

print("🎉 3 adet müşteri nesnesi oluşturduk!")
print(f"\n📦 musteri1'in tipi: {type(musteri1)}")
print(f"📦 musteri2'nin tipi: {type(musteri2)}")
print(f"📦 musteri3'ün tipi: {type(musteri3)}")

print("\n🔍 Her nesne farklı bir bellek adresinde:")
print(f"   musteri1 adresi: {id(musteri1)}")
print(f"   musteri2 adresi: {id(musteri2)}")
print(f"   musteri3 adresi: {id(musteri3)}")

print("\n💡 Aynı kalıptan (class) 3 farklı nesne ürettik!")
````

---

### 🎓 **Adım 3: Nesnelere Özellik (Attribute) Ekleme**

Şu anda nesnelerimiz boş. Onlara **özellikler** ekleyelim!

> 🔥 **ÖZEL NOT:** Python'da nesnelere **sonradan** özellik ekleyebiliriz (dinamik dil avantajı)

````python
# 🎨 NESNELERE ÖZELLİK EKLEME (Manuel yöntem - tavsiye edilmez)

# musteri1'e özellikler ekleyelim
musteri1.isim = "Ayşe Yılmaz"
musteri1.email = "ayse@gmail.com"
musteri1.telefon = "0555-111-2233"
musteri1.bakiye = 5000

# musteri2'ye özellikler ekleyelim  
musteri2.isim = "Mehmet Demir"
musteri2.email = "mehmet@gmail.com"
musteri2.telefon = "0555-444-5566"
musteri2.bakiye = 3000

print("✅ Müşterilere özellikler ekledik!\n")

# Özelliklere erişim
print("👤 MÜŞTERİ 1 BİLGİLERİ:")
print(f"   📛 İsim: {musteri1.isim}")
print(f"   📧 Email: {musteri1.email}")
print(f"   📱 Telefon: {musteri1.telefon}")
print(f"   💰 Bakiye: {musteri1.bakiye} TL")

print("\n👤 MÜŞTERİ 2 BİLGİLERİ:")
print(f"   📛 İsim: {musteri2.isim}")
print(f"   📧 Email: {musteri2.email}")
print(f"   📱 Telefon: {musteri2.telefon}")
print(f"   💰 Bakiye: {musteri2.bakiye} TL")

print("\n⚠️ DİKKAT: Bu yöntem pratik değil! Her nesne için tek tek yazmak gerekiyor.")
````

### 🎓 **Adım 4: Constructor (__init__) ile Otomatik Başlatma**

Her nesne için özellikleri tek tek yazmak zahmetli! **Çözüm:** `__init__` metodu

> 💡 **__init__ NEDİR?**  
> - Constructor (Yapıcı) metod
> - Nesne oluşturulduğunda **otomatik** çalışır
> - Başlangıç değerlerini ayarlar
> - İki tarafında **çift alt çizgi** var (dunder = double underscore)

### 🎨 **VİZYON:**

```python
❌ ÖNCESİ:
musteri = Musteri()
musteri.isim = "Ayşe"
musteri.email = "ayse@gmail.com"
musteri.telefon = "0555-111-2233"
# 4 satır kod!

✅ SONRASI:
musteri = Musteri("Ayşe", "ayse@gmail.com", "0555-111-2233")
# Tek satır! 🎉
```

````python
# 🎯 PROFESYONEL YÖNTEM: __init__ İLE CLASS TANIMI

class MusteriPro:
    """
    🏢 TechShop E-Ticaret Müşteri Sınıfı
    
    Bu class, sistemdeki müşterileri temsil eder.
    Her müşterinin isim, email, telefon ve bakiye bilgileri vardır.
    """
    
    def __init__(self, isim, email, telefon, bakiye=0):
        """
        🎬 Constructor (Yapıcı) Metod
        Nesne oluşturulduğunda otomatik çalışır
        
        Parametreler:
        - self: Nesnenin kendisi (Python otomatik ekler, yazmanıza gerek yok)
        - isim: Müşterinin adı soyadı
        - email: E-posta adresi
        - telefon: Telefon numarası
        - bakiye: Hesap bakiyesi (varsayılan = 0)
        """
        # self.özellik_adı = parametre değeri
        self.isim = isim
        self.email = email
        self.telefon = telefon
        self.bakiye = bakiye
        
        print(f"✅ {isim} adlı müşteri sisteme eklendi!")

print("🎉 Profesyonel Musteri class'ı tanımlandı!\n")
````

---

### 🔍 **self Parametresi Nedir?**

**En çok sorulan soru!** 🤔

> **self** = Nesnenin kendisine referans
> 
> Python'a "Bu özellik **bu nesneye** ait" dediğimizi söyleriz

**Analoji:**
```
🏠 Bir apartman düşünün:
   - Apartman = Class
   - Daireler = Objects
   - self = "Benim dairem" demek
   
Ayşe'nin dairesi: self.renk = "mavi"
Mehmet'in dairesi: self.renk = "beyaz"

Aynı apartman, ama her daire farklı!
```

> ⚠️ **ÖNEMLİ:** `self` parametresini fonksiyon tanımlarken yazarsınız ama çağırırken **yazmassınız**!

````python
# 🚀 ŞİMDİ NESNELER OLUŞTURALIM - ÇOK KOLAY!

# Müşteri 1: Ayşe Yılmaz
musteri_ayse = MusteriPro(
    isim="Ayşe Yılmaz",
    email="ayse.yilmaz@gmail.com", 
    telefon="0555-111-2233",
    bakiye=5000
)

# Müşteri 2: Mehmet Demir (bakiye belirtmedik, varsayılan 0 olacak)
musteri_mehmet = MusteriPro(
    isim="Mehmet Demir",
    email="mehmet.demir@gmail.com",
    telefon="0555-444-5566"
)

# Müşteri 3: Zeynep Kaya
musteri_zeynep = MusteriPro(
    isim="Zeynep Kaya",
    email="zeynep.kaya@hotmail.com",
    telefon="0555-777-8899",
    bakiye=12000
)

print("\n" + "="*50)
print("📊 MÜŞTERİ BİLGİLERİ")
print("="*50)

print(f"\n👤 MÜŞTERİ 1:")
print(f"   📛 İsim: {musteri_ayse.isim}")
print(f"   📧 Email: {musteri_ayse.email}")
print(f"   📱 Telefon: {musteri_ayse.telefon}")
print(f"   💰 Bakiye: {musteri_ayse.bakiye:,} TL")

print(f"\n👤 MÜŞTERİ 2:")
print(f"   📛 İsim: {musteri_mehmet.isim}")
print(f"   📧 Email: {musteri_mehmet.email}")
print(f"   📱 Telefon: {musteri_mehmet.telefon}")
print(f"   💰 Bakiye: {musteri_mehmet.bakiye:,} TL")

print(f"\n👤 MÜŞTERİ 3:")
print(f"   📛 İsim: {musteri_zeynep.isim}")
print(f"   📧 Email: {musteri_zeynep.email}")
print(f"   📱 Telefon: {musteri_zeynep.telefon}")
print(f"   💰 Bakiye: {musteri_zeynep.bakiye:,} TL")

print("\n🎉 3 müşteri sadece 3 satır kodla oluşturuldu!")
````

Şimdi birden fazla müşteri ile çalışalım - **YBS'de çok kullanılır!**

````python
# 📊 TOPLU MÜŞTERİ OLUŞTURMA - GERÇEK YBS SENARYOSU

# Müşteri verilerini bir liste olarak tanımlayalım (normalde database'den gelir)
musteri_verileri = [
    {"isim": "Ali Veli", "email": "ali.veli@gmail.com", "telefon": "0532-111-2233", "bakiye": 3500},
    {"isim": "Fatma Yıldız", "email": "fatma.yildiz@hotmail.com", "telefon": "0533-222-3344", "bakiye": 8000},
    {"isim": "Can Özdemir", "email": "can.ozdemir@outlook.com", "telefon": "0534-333-4455", "bakiye": 1500},
    {"isim": "Elif Şahin", "email": "elif.sahin@gmail.com", "telefon": "0535-444-5566", "bakiye": 12000},
    {"isim": "Burak Aslan", "email": "burak.aslan@yahoo.com", "telefon": "0536-555-6677", "bakiye": 5500},
]

# Boş bir liste oluştur - tüm müşteri nesnelerini burada saklayacağız
musteri_listesi = []

print("🏢 TechShop A.Ş. - Müşteri Kayıt Sistemi")
print("="*50 + "\n")

# Her müşteri verisi için nesne oluştur
for veri in musteri_verileri:
    yeni_musteri = MusteriPro(
        isim=veri["isim"],
        email=veri["email"],
        telefon=veri["telefon"],
        bakiye=veri["bakiye"]
    )
    musteri_listesi.append(yeni_musteri)

print("\n" + "="*50)
print(f"✅ Toplam {len(musteri_listesi)} müşteri sisteme kaydedildi!")
print("="*50)
````

````python
# 🎨 EN BASİT METHOD ÖRNEĞİ

class Kopek:
    """Basit bir köpek sınıfı"""
    
    def __init__(self, isim, yas):
        """Constructor - Başlangıç özellikleri"""
        self.isim = isim
        self.yas = yas
    
    # 📢 İLK METHODUMUZ!
    def havla(self):
        """Köpek havlama metodu"""
        print(f"{self.isim}: Hav hav! 🐕")
    
    def koş(self):
        """Köpek koşma metodu"""
        print(f"{self.isim} koşuyor! 🏃‍♂️💨")
    
    def bilgileriGoster(self):
        """Köpek bilgilerini göster"""
        print(f"🐕 İsim: {self.isim}")
        print(f"🎂 Yaş: {self.yas}")

# Köpek nesnesi oluştur
karabas = Kopek("Karabaş", 3)

print("🎉 Köpek oluşturuldu! Şimdi yeteneklerini görelim:\n")

# Methodları çağır
karabas.havla()
karabas.koş()
karabas.bilgileriGoster()

print("\n💡 Gördünüz mü? Nesne artık şeyler yapabiliyor!")
````

## 💼 **YBS ÖRNEĞİ: MÜŞTERİ İŞLEMLERİ**

Şimdi gerçek bir YBS senaryosu için Müşteri class'ına metodlar ekleyelim!

````python
# 🏢 PROFESYONEL MÜŞTERİ CLASS'I - METHODLARLA GÜÇLENDİRİLMİŞ!

class MusteriV2:
    """
    🏢 TechShop E-Ticaret Müşteri Sınıfı (v2.0)
    
    Bu sınıf müşteri işlemlerini yönetir:
    - Bakiye yükleme
    - Sipariş verme
    - Bilgi görüntüleme
    """
    
    def __init__(self, isim, email, telefon, bakiye=0):
        """Constructor - Müşteri oluştur"""
        self.isim = isim
        self.email = email
        self.telefon = telefon
        self.bakiye = bakiye
        self.siparis_gecmisi = []  # Siparişler burada saklanacak
        print(f"✅ {isim} hesabı oluşturuldu!")
    
    # 💰 METHOD 1: Bakiye Yükleme
    def bakiye_yukle(self, miktar):
        """
        Müşteri hesabına bakiye yükler
        
        Parametreler:
        - miktar: Yüklenecek tutar (TL)
        """
        if miktar <= 0:
            print("⚠️ Hata: Yüklenecek miktar 0'dan büyük olmalı!")
            return False
        
        self.bakiye += miktar
        print(f"💰 {miktar:,} TL yüklendi!")
        print(f"   Yeni bakiye: {self.bakiye:,} TL")
        return True
    
    # 🛒 METHOD 2: Sipariş Verme
    def siparis_ver(self, urun_adi, fiyat):
        """
        Müşteri sipariş verir
        
        Parametreler:
        - urun_adi: Ürün adı
        - fiyat: Ürün fiyatı (TL)
        """
        # Bakiye kontrolü
        if self.bakiye < fiyat:
            print(f"❌ Yetersiz bakiye!")
            print(f"   Gerekli: {fiyat:,} TL - Mevcut: {self.bakiye:,} TL")
            return False
        
        # Siparişi işle
        self.bakiye -= fiyat
        self.siparis_gecmisi.append({
            "urun": urun_adi,
            "fiyat": fiyat,
            "tarih": datetime.now().strftime("%d.%m.%Y %H:%M")
        })
        
        print(f"🎉 Sipariş başarılı!")
        print(f"   Ürün: {urun_adi}")
        print(f"   Fiyat: {fiyat:,} TL")
        print(f"   Kalan bakiye: {self.bakiye:,} TL")
        return True
    
    # 📊 METHOD 3: Bilgileri Göster
    def bilgileri_goster(self):
        """Müşteri bilgilerini ekrana yazdırır"""
        print("\n" + "="*50)
        print("👤 MÜŞTERİ BİLGİLERİ")
        print("="*50)
        print(f"📛 İsim      : {self.isim}")
        print(f"📧 Email     : {self.email}")
        print(f"📱 Telefon   : {self.telefon}")
        print(f"💰 Bakiye    : {self.bakiye:,} TL")
        print(f"🛒 Sipariş   : {len(self.siparis_gecmisi)} adet")
        print("="*50)
    
    # 📜 METHOD 4: Sipariş Geçmişini Göster
    def siparis_gecmisini_goster(self):
        """Müşterinin tüm siparişlerini listeler"""
        if not self.siparis_gecmisi:
            print("📭 Henüz sipariş yok.")
            return
        
        print(f"\n📜 {self.isim} - SİPARİŞ GEÇMİŞİ")
        print("="*60)
        
        for i, siparis in enumerate(self.siparis_gecmisi, 1):
            print(f"\n🛍️ Sipariş #{i}")
            print(f"   Ürün  : {siparis['urun']}")
            print(f"   Fiyat : {siparis['fiyat']:,} TL")
            print(f"   Tarih : {siparis['tarih']}")
        
        print("\n" + "="*60)
        toplam = sum(s['fiyat'] for s in self.siparis_gecmisi)
        print(f"💵 Toplam Harcama: {toplam:,} TL")
    
    # 📈 METHOD 5: Müşteri Değeri Hesapla
    def musteri_degeri_hesapla(self):
        """
        Müşterinin toplam değerini hesaplar
        Returns: Toplam harcama tutarı
        """
        toplam_harcama = sum(s['fiyat'] for s in self.siparis_gecmisi)
        return toplam_harcama

print("🎉 Gelişmiş MusteriV2 class'ı hazır!\n")
````

## 🔥 **İLERİ SEVİYE: CLASS METHODS vs STATIC METHODS**

Şimdiye kadar **instance method** gördük. İki tür method daha var!

### 📊 **Method Türleri:**

| Tür | Decorator | İlk Parametre | Kullanım |
|-----|-----------|---------------|----------|
| **Instance Method** | Yok | `self` | Nesne özellikleriyle çalışır |
| **Class Method** | `@classmethod` | `cls` | Class seviyesi işlemler |
| **Static Method** | `@staticmethod` | Yok | Bağımsız utility fonksiyonlar |

### 💡 **Ne Zaman Hangisi?**

- 🟢 **Instance Method:** Nesnenin özelliklerini kullan/değiştir → `self.bakiye`
- 🟡 **Class Method:** Class değişkenleri veya alternatif constructor → `cls.toplam_musteri`
- 🔵 **Static Method:** Sadece hesaplama/yardımcı fonksiyon → hiç self/cls yok

````python
# 🎓 TÜM METHOD TÜRLERİNİ İÇEREN CLASS

class Urun:
    """
    🏢 E-Ticaret Ürün Sınıfı
    3 farklı method türünü gösterir
    """
    
    # Class variable (tüm nesnelerde ortak)
    toplam_urun_sayisi = 0
    kdv_orani = 0.18  # %18 KDV
    
    def __init__(self, isim, fiyat, stok):
        """Constructor - Instance method"""
        self.isim = isim
        self.fiyat = fiyat
        self.stok = stok
        Urun.toplam_urun_sayisi += 1  # Her yeni ürün için artır
    
    # 🟢 INSTANCE METHOD - Nesne özellikleriyle çalışır
    def kdv_dahil_fiyat(self):
        """Ürünün KDV dahil fiyatını hesaplar"""
        kdv_tutari = self.fiyat * Urun.kdv_orani
        toplam = self.fiyat + kdv_tutari
        return toplam
    
    def stok_durumu(self):
        """Stok durumunu raporlar"""
        if self.stok > 10:
            durum = "✅ Stokta Bol"
        elif self.stok > 0:
            durum = "⚠️ Stok Azalıyor"
        else:
            durum = "❌ Stok Yok"
        
        return f"{self.isim}: {durum} ({self.stok} adet)"
    
    # 🟡 CLASS METHOD - Class seviyesi işlemler
    @classmethod
    def toplam_urun_sayisini_goster(cls):
        """Sistemdeki toplam ürün sayısını gösterir"""
        print(f"📦 Sistemde toplam {cls.toplam_urun_sayisi} ürün var")
    
    @classmethod
    def kdv_oranini_degistir(cls, yeni_oran):
        """KDV oranını değiştirir (tüm ürünler için geçerli)"""
        cls.kdv_orani = yeni_oran
        print(f"✅ KDV oranı %{yeni_oran*100:.0f} olarak güncellendi")
    
    @classmethod
    def ucuz_urun_olustur(cls, isim):
        """
        Alternatif constructor - Ucuz ürün oluşturur
        Class method constructor pattern örneği
        """
        return cls(isim, fiyat=100, stok=50)
    
    # 🔵 STATIC METHOD - Bağımsız yardımcı fonksiyon
    @staticmethod
    def fiyat_araligini_kontrol_et(fiyat):
        """
        Fiyatın makul aralıkta olup olmadığını kontrol eder
        Static - hiç self veya cls kullanmıyor
        """
        if fiyat < 0:
            return "❌ Geçersiz: Fiyat negatif olamaz!"
        elif fiyat < 10:
            return "⚠️ Çok ucuz!"
        elif fiyat > 100000:
            return "⚠️ Çok pahalı!"
        else:
            return "✅ Makul fiyat aralığı"
    
    @staticmethod
    def indirim_hesapla(fiyat, indirim_yuzdesi):
        """
        İndirimli fiyatı hesaplar
        Static - genel bir hesaplama fonksiyonu
        """
        indirim_tutari = fiyat * (indirim_yuzdesi / 100)
        yeni_fiyat = fiyat - indirim_tutari
        return yeni_fiyat

print("🎉 Gelişmiş Urun class'ı (3 method türüyle) hazır!\n")
````

````python
# 🎮 METHOD TÜRLERİNİ TEST EDELİM

print("="*70)
print("🧪 METHOD TÜRLERİ TESTİ")
print("="*70 + "\n")

# Ürünler oluştur
print("📦 Ürünler oluşturuluyor...\n")
laptop = Urun("MacBook Pro M3", 65000, 15)
telefon = Urun("iPhone 15", 45000, 8)
kulaklik = Urun("AirPods Pro", 8500, 3)

print()

# 🟢 INSTANCE METHODS (nesne üzerinden)
print("-"*70)
print("🟢 INSTANCE METHODS - Nesneye özel işlemler")
print("-"*70 + "\n")

print(f"💻 {laptop.isim}")
print(f"   Fiyat (KDV Hariç): {laptop.fiyat:,} TL")
print(f"   Fiyat (KDV Dahil): {laptop.kdv_dahil_fiyat():,.2f} TL")
print(f"   {laptop.stok_durumu()}")

print(f"\n📱 {telefon.isim}")
print(f"   Fiyat (KDV Hariç): {telefon.fiyat:,} TL")
print(f"   Fiyat (KDV Dahil): {telefon.kdv_dahil_fiyat():,.2f} TL")
print(f"   {telefon.stok_durumu()}")

print(f"\n🎧 {kulaklik.isim}")
print(f"   {kulaklik.stok_durumu()}")

# 🟡 CLASS METHODS (class üzerinden)
print("\n" + "-"*70)
print("🟡 CLASS METHODS - Class seviyesi işlemler")
print("-"*70 + "\n")

# Class method ile toplam ürün sayısı
Urun.toplam_urun_sayisini_goster()

# Alternatif constructor ile ürün oluştur
print("\n🎁 Class method ile ucuz ürün oluşturuluyor:")
mouse = Urun.ucuz_urun_olustur("Kablosuz Mouse")
print(f"   {mouse.isim} - {mouse.fiyat} TL - Stok: {mouse.stok}")

# Şimdi toplam ürün sayısı arttı
print()
Urun.toplam_urun_sayisini_goster()

# KDV oranını değiştir
print("\n💼 KDV oranı değiştiriliyor:")
Urun.kdv_oranini_degistir(0.20)  # %20'ye çıkar

print(f"\n💻 {laptop.isim} - Yeni KDV dahil fiyat: {laptop.kdv_dahil_fiyat():,.2f} TL")

# 🔵 STATIC METHODS (class üzerinden, self/cls yok)
print("\n" + "-"*70)
print("🔵 STATIC METHODS - Bağımsız yardımcı fonksiyonlar")
print("-"*70 + "\n")

# Fiyat kontrolü
test_fiyatlar = [50, 5, 150000, -100, 25000]
print("💵 Fiyat kontrolleri:")
for fiyat in test_fiyatlar:
    sonuc = Urun.fiyat_araligini_kontrol_et(fiyat)
    print(f"   {fiyat:>7} TL → {sonuc}")

# İndirim hesaplama
print("\n🎁 İndirim hesaplamaları:")
print(f"   50,000 TL'ye %20 indirim: {Urun.indirim_hesapla(50000, 20):,.2f} TL")
print(f"   10,000 TL'ye %35 indirim: {Urun.indirim_hesapla(10000, 35):,.2f} TL")

# Static method'u nesne üzerinden de çağırabiliriz (ama tavsiye edilmez)
print(f"\n💡 Static method nesne üzerinden: {laptop.indirim_hesapla(5000, 15):,.2f} TL")

print("\n" + "="*70)
print("✅ Tüm method türleri test edildi!")
print("="*70)
````

````python
# 🏢 KAPSAMLI SİPARİŞ YÖNETİM SİSTEMİ

class SiparisYonetimSistemi:
    """
    🏢 TechShop A.Ş. Sipariş Yönetim Sistemi
    
    Bu sistem şunları yönetir:
    - Müşteri kayıtları
    - Ürün kataloğu
    - Sipariş işlemleri
    - Raporlama
    """
    
    # Class variables
    toplam_musteri = 0
    toplam_urun = 0
    toplam_siparis = 0
    
    def __init__(self, sistem_adi):
        """Sistem başlatıcı"""
        self.sistem_adi = sistem_adi
        self.musteriler = {}  # {musteri_id: MusteriV2}
        self.urunler = {}     # {urun_id: Urun}
        self.siparisler = []  # Sipariş listesi
        print(f"🎉 {sistem_adi} başlatıldı!")
    
    # 👤 Müşteri işlemleri
    def musteri_ekle(self, musteri_id, musteri):
        """Sisteme müşteri ekle"""
        if musteri_id in self.musteriler:
            print(f"⚠️ {musteri_id} ID'li müşteri zaten var!")
            return False
        
        self.musteriler[musteri_id] = musteri
        SiparisYonetimSistemi.toplam_musteri += 1
        print(f"✅ {musteri.isim} sisteme eklendi (ID: {musteri_id})")
        return True
    
    # 📦 Ürün işlemleri
    def urun_ekle(self, urun_id, urun):
        """Sisteme ürün ekle"""
        if urun_id in self.urunler:
            print(f"⚠️ {urun_id} ID'li ürün zaten var!")
            return False
        
        self.urunler[urun_id] = urun
        SiparisYonetimSistemi.toplam_urun += 1
        print(f"✅ {urun.isim} kataloğa eklendi (ID: {urun_id})")
        return True
    
    # 🛒 Sipariş işlemleri
    def siparis_olustur(self, musteri_id, urun_id):
        """Sipariş oluştur"""
        # Kontroller
        if musteri_id not in self.musteriler:
            print(f"❌ {musteri_id} ID'li müşteri bulunamadı!")
            return False
        
        if urun_id not in self.urunler:
            print(f"❌ {urun_id} ID'li ürün bulunamadı!")
            return False
        
        musteri = self.musteriler[musteri_id]
        urun = self.urunler[urun_id]
        
        # Stok kontrolü
        if urun.stok <= 0:
            print(f"❌ {urun.isim} stokta yok!")
            return False
        
        # Sipariş ver
        fiyat = urun.kdv_dahil_fiyat()
        if musteri.siparis_ver(urun.isim, fiyat):
            urun.stok -= 1
            self.siparisler.append({
                "musteri": musteri.isim,
                "urun": urun.isim,
                "fiyat": fiyat,
                "tarih": datetime.now()
            })
            SiparisYonetimSistemi.toplam_siparis += 1
            return True
        
        return False
    
    # 📊 Raporlama
    @classmethod
    def sistem_istatistikleri(cls):
        """Genel sistem istatistikleri"""
        print("\n" + "="*50)
        print("📊 SİSTEM İSTATİSTİKLERİ")
        print("="*50)
        print(f"👥 Toplam Müşteri: {cls.toplam_musteri}")
        print(f"📦 Toplam Ürün: {cls.toplam_urun}")
        print(f"🛒 Toplam Sipariş: {cls.toplam_siparis}")
        print("="*50)
    
    def detayli_rapor(self):
        """Bu sistemin detaylı raporu"""
        print(f"\n📋 {self.sistem_adi.upper()} DETAYLI RAPORU")
        print("="*60)
        
        print(f"\n👥 MÜŞTERİLER ({len(self.musteriler)} adet):")
        for mid, m in self.musteriler.items():
            toplam_harcama = m.musteri_degeri_hesapla()
            print(f"   [{mid}] {m.isim} - Bakiye: {m.bakiye:,} TL - Harcama: {toplam_harcama:,} TL")
        
        print(f"\n📦 ÜRÜNLER ({len(self.urunler)} adet):")
        for uid, u in self.urunler.items():
            print(f"   [{uid}] {u.isim} - {u.fiyat:,} TL - Stok: {u.stok}")
        
        print(f"\n🛒 SİPARİŞLER ({len(self.siparisler)} adet):")
        for i, s in enumerate(self.siparisler, 1):
            print(f"   {i}. {s['musteri']} → {s['urun']} ({s['fiyat']:,.2f} TL)")
        
        print("\n" + "="*60)

print("✅ SiparisYonetimSistemi class'ı hazır!")
````

# 🧬 BÖLÜM 5: INHERITANCE (KALITIM) - Kodun Yeniden Kullanımı

---

## 🎯 **INHERITANCE NEDİR?**

### 📚 **Basit Açıklama:**

> **Inheritance (Kalıtım)** = Bir class'ın başka bir class'tan özellik ve metodları **miras alması**

### 🎨 **GERÇEK HAYAT ANALOJİSİ:**

```
👪 AİLE AĞACI
   
   👴 DEDE (Ana Class/Parent/Superclass)
      - Siyah saç
      - Kahverengi göz
      - 170cm boy
   
   👨 BABA (Child Class)
      ✅ Dedesinden aldı: Siyah saç, Kahverengi göz
      ➕ Kendine özgü: Doktor mesleği
   
   👦 OĞUL (Grandchild Class)
      ✅ Babadan aldı: Tüm özellikler
      ➕ Kendine özgü: Öğrenci statüsü
```

### 💼 **YBS'DE NEDEN ÖNEMLİ?**

Gerçek dünyada **benzer ama farklı** nesneler vardır:
- 🏢 **Genel Müşteri** → **Bireysel Müşteri**, **Kurumsal Müşteri**
- 💳 **Ödeme Yöntemi** → **Kredi Kartı**, **Havale**, **Kapıda Ödeme**
- 📦 **Ürün** → **Elektronik**, **Giyim**, **Kitap**

Her biri ortak özelliklere sahip ama **farklılıkları** var!

---

## 🔰 **ADIM ADIM ÖĞRENELIM**

### 🎓 **Adım 1: Parent Class (Ana Sınıf)**

Önce genel bir müşteri sınıfı oluşturalım:

````python
# 👤 PARENT CLASS: GenelMusteri (Temel Müşteri Sınıfı)

class GenelMusteri:
    """
    🏢 Genel Müşteri Sınıfı (Parent/Base Class)
    
    Tüm müşteri tiplerinin ortak özelliklerini içerir.
    """
    
    # Class variable - tüm müşteriler için ortak
    musteri_sayisi = 0
    
    def __init__(self, isim, email, telefon):
        """Constructor - Temel müşteri özellikleri"""
        self.isim = isim
        self.email = email
        self.telefon = telefon
        self.kayit_tarihi = datetime.now()
        self.aktif = True
        
        GenelMusteri.musteri_sayisi += 1
        print(f"✅ {isim} genel müşteri olarak kaydedildi")
    
    def bilgileri_goster(self):
        """Temel müşteri bilgilerini gösterir"""
        print("\n" + "="*50)
        print("👤 MÜŞTERİ BİLGİLERİ")
        print("="*50)
        print(f"📛 İsim     : {self.isim}")
        print(f"📧 Email    : {self.email}")
        print(f"📱 Telefon  : {self.telefon}")
        print(f"📅 Kayıt    : {self.kayit_tarihi.strftime('%d.%m.%Y')}")
        print(f"✅ Durum    : {'Aktif' if self.aktif else 'Pasif'}")
        print("="*50)
    
    def iletisim_gonder(self, mesaj):
        """Müşteriye mesaj gönderir"""
        print(f"📧 {self.isim}'e mesaj gönderiliyor...")
        print(f"   💬 Mesaj: {mesaj}")
        return True
    
    @classmethod
    def toplam_musteri_sayisi(cls):
        """Toplam müşteri sayısını döndürür"""
        return cls.musteri_sayisi

print("🎉 GenelMusteri parent class'ı oluşturuldu!\n")

# Test edelim
genel = GenelMusteri("Ali Veli", "ali@test.com", "0555-123-4567")
genel.bilgileri_goster()
genel.iletisim_gonder("Hoş geldiniz!")
````

### 🎓 **Adım 2: Child Class (Alt Sınıf) - Bireysel Müşteri**

Şimdi GenelMusteri'den **miras alan** bir BireyselMusteri class'ı oluşturalım!

> 💡 **Syntax:** `class ChildClass(ParentClass):`

### 🔑 **Önemli Noktalar:**

1. ✅ **Parantez içinde parent class yazılır:** `class BireyselMusteri(GenelMusteri)`
2. ✅ **Parent'ın tüm özellik/metodları miras alınır**
3. ✅ **Yeni özellik/metodlar eklenebilir**
4. ✅ **Mevcut metodlar override edilebilir (üzerine yazılır)**
5. ✅ **`super()` ile parent'a erişilir**

````python
# 👨 CHILD CLASS 1: BireyselMusteri (GenelMusteri'den miras alır)

class BireyselMusteri(GenelMusteri):
    """
    🏃 Bireysel Müşteri Sınıfı (Child Class)
    
    GenelMusteri'den miras alır ve bireysel müşterilere
    özel özellikler ekler.
    """
    
    def __init__(self, isim, email, telefon, tc_no, dogum_tarihi):
        """
        Constructor - Parent'ın constructor'ını çağırır + ekstra özellikler
        """
        # super() ile parent class'ın __init__ metodunu çağır
        super().__init__(isim, email, telefon)
        
        # Bireysel müşteriye özel özellikler
        self.tc_no = tc_no
        self.dogum_tarihi = dogum_tarihi
        self.bakiye = 0
        self.sadakat_puani = 0
        
        print(f"   🎯 Bireysel müşteri özellikleri eklendi")
    
    # YENİ METHOD - Sadece bireysel müşteride var
    def yas_hesapla(self):
        """Müşterinin yaşını hesaplar"""
        bugun = datetime.now()
        yas = bugun.year - self.dogum_tarihi.year
        
        # Doğum günü henüz gelmediyse 1 eksilt
        if bugun.month < self.dogum_tarihi.month or \
           (bugun.month == self.dogum_tarihi.month and bugun.day < self.dogum_tarihi.day):
            yas -= 1
        
        return yas
    
    def sadakat_puani_ekle(self, puan):
        """Sadakat puanı ekler"""
        self.sadakat_puani += puan
        print(f"⭐ +{puan} sadakat puanı eklendi!")
        print(f"   Toplam puan: {self.sadakat_puani}")
    
    def bakiye_yukle(self, miktar):
        """Hesaba bakiye yükler"""
        if miktar > 0:
            self.bakiye += miktar
            print(f"💰 {miktar:,} TL yüklendi! Yeni bakiye: {self.bakiye:,} TL")
            return True
        return False
    
    # OVERRIDE - Parent'ın metodunu yeniden yazıyoruz
    def bilgileri_goster(self):
        """
        Parent'ın bilgileri_goster metodunu override ediyoruz
        Hem parent bilgileri hem de bireysel müşteri bilgileri gösterilir
        """
        # Önce parent'ın metodunu çağır
        super().bilgileri_goster()
        
        # Sonra ekstra bilgileri ekle
        print("\n👤 BİREYSEL MÜŞTERİ EK BİLGİLERİ:")
        print(f"🆔 TC No           : {self.tc_no}")
        print(f"🎂 Yaş             : {self.yas_hesapla()}")
        print(f"💰 Bakiye          : {self.bakiye:,} TL")
        print(f"⭐ Sadakat Puanı   : {self.sadakat_puani}")
        print("="*50)

print("🎉 BireyselMusteri child class'ı oluşturuldu!\n")
````

### 🎓 **Adım 3: Child Class 2 - Kurumsal Müşteri**

Aynı parent'tan başka bir child class daha oluşturalım!

> 💡 **Çoklu Kalıtım:** Bir parent'tan birden fazla child türetilebilir!

````python
# 🏢 CHILD CLASS 2: KurumsalMusteri (GenelMusteri'den miras alır)

class KurumsalMusteri(GenelMusteri):
    """
    🏢 Kurumsal Müşteri Sınıfı (Child Class)
    
    GenelMusteri'den miras alır ve kurumsal müşterilere
    özel özellikler ekler.
    """
    
    def __init__(self, isim, email, telefon, vergi_no, sektor, calisansayisi):
        """Constructor - Parent + kurumsal özellikler"""
        # Parent constructor'ı çağır
        super().__init__(isim, email, telefon)
        
        # Kurumsal müşteriye özel özellikler
        self.vergi_no = vergi_no
        self.sektor = sektor
        self.calisan_sayisi = calisansayisi
        self.kredi_limiti = self.calisan_sayisi * 1000  # Çalışan başı 1000 TL
        self.kullanilan_kredi = 0
        self.toplam_ciro = 0
        
        print(f"   🏢 Kurumsal müşteri özellikleri eklendi")
    
    # YENİ METODLAR - Sadece kurumsal müşteride var
    def kredi_kullan(self, miktar):
        """Kredi limiti kullanımı"""
        kullanilabilir = self.kredi_limiti - self.kullanilan_kredi
        
        if miktar > kullanilabilir:
            print(f"❌ Yetersiz kredi limiti!")
            print(f"   İstenen: {miktar:,} TL")
            print(f"   Kullanılabilir: {kullanilabilir:,} TL")
            return False
        
        self.kullanilan_kredi += miktar
        print(f"💳 {miktar:,} TL kredi kullanıldı!")
        print(f"   Kalan limit: {self.kredi_limiti - self.kullanilan_kredi:,} TL")
        return True
    
    def odeme_yap(self, miktar):
        """Kredi ödemesi yapar"""
        if miktar > self.kullanilan_kredi:
            miktar = self.kullanilan_kredi
        
        self.kullanilan_kredi -= miktar
        print(f"✅ {miktar:,} TL ödeme yapıldı!")
        print(f"   Kalan borç: {self.kullanilan_kredi:,} TL")
        return True
    
    def ciro_ekle(self, miktar):
        """Toplam ciroya ekleme yapar"""
        self.toplam_ciro += miktar
        print(f"📈 Ciro güncellendi: {self.toplam_ciro:,} TL")
    
    def firma_kategorisi(self):
        """Firma kategorisini belirler"""
        if self.calisan_sayisi < 10:
            return "Mikro İşletme"
        elif self.calisan_sayisi < 50:
            return "Küçük İşletme"
        elif self.calisan_sayisi < 250:
            return "Orta Ölçekli İşletme"
        else:
            return "Büyük İşletme"
    
    # OVERRIDE - Parent'ın metodunu yeniden yazıyoruz
    def bilgileri_goster(self):
        """Kurumsal müşteri bilgilerini gösterir"""
        # Parent'ın metodunu çağır
        super().bilgileri_goster()
        
        # Kurumsal bilgileri ekle
        print("\n🏢 KURUMSAL MÜŞTERİ EK BİLGİLERİ:")
        print(f"🆔 Vergi No        : {self.vergi_no}")
        print(f"🏭 Sektör          : {self.sektor}")
        print(f"👥 Çalışan Sayısı  : {self.calisan_sayisi}")
        print(f"🏷️  Kategori        : {self.firma_kategorisi()}")
        print(f"💳 Kredi Limiti    : {self.kredi_limiti:,} TL")
        print(f"💰 Kullanılan      : {self.kullanilan_kredi:,} TL")
        print(f"📊 Toplam Ciro     : {self.toplam_ciro:,} TL")
        print("="*50)
    
    # OVERRIDE - Kurumsal müşterilere özel mesaj
    def iletisim_gonder(self, mesaj):
        """Kurumsal müşterilere özel mesaj formatı"""
        print(f"📧 {self.isim} şirketine mesaj gönderiliyor...")
        print(f"   🏢 Sayın Yetkili,")
        print(f"   💬 {mesaj}")
        print(f"   📝 Saygılarımızla, TechShop A.Ş.")
        return True

print("🎉 KurumsalMusteri child class'ı oluşturuldu!\n")
````

## 🎯 **POLYMORPHISM NEDİR?**

### 📚 **Basit Açıklama:**

> **Polymorphism** = Aynı method ismi, **farklı nesnelerde farklı davranışlar**

### 🎨 **GERÇEK HAYAT ANALOJİSİ:**

```
🎵 "ÇAL" KOMUTU - FARKLI ENSTRÜMANLARDA FARKLI SES!

🎸 Gitar.cal()    → "Tıng tıng tıng" 🎶
🎹 Piyano.cal()   → "Ding ding ding" 🎵
🥁 Davul.cal()    → "Bum bum bum" 🥁

Aynı komut (cal) ama her enstrüman farklı ses çıkarır!
```

### 🔥 **Inheritance vs Polymorphism:**

| Konsept | Anlam | Örnek |
|---------|-------|-------|
| **Inheritance** | Özellik/metod miras alma | Child, parent'tan özellik alır |
| **Polymorphism** | Aynı arayüz, farklı davranış | Her child aynı metodu farklı çalıştırır |

> 💡 **ÖNEMLİ:** Polymorphism genellikle inheritance ile birlikte kullanılır!

## 💼 **YBS'DE NEDEN ÖNEMLİ?**

Gerçek dünyada **aynı işlem, farklı tiplerle** yapılır:
- 💳 **Ödeme Al:** Kredi kartı, havale, kapıda ödeme → Hepsi farklı işler
- 📧 **Bildirim Gönder:** Email, SMS, Push → Farklı kanallar
- 💰 **İndirim Hesapla:** Bireysel %10, Kurumsal %20 → Farklı oranlar

Polymorphism sayesinde **tek kod** ile farklı tipleri yönetebiliriz!

## 🔰 **ADIM ADIM ÖĞRENELIM**

### 🎓 **Adım 1: Polymorphism Örneği - İndirim Hesaplama**

Her müşteri tipine özel indirim hesaplayalım!

````python
# 🎭 POLYMORPHİSM İÇİN GÜNCELLENMIŞ PARENT CLASS

class MusteriBase:
    """
    👤 Temel Müşteri Sınıfı (Polymorphism için)
    
    Tüm müşteri tiplerinin ortak arayüzünü tanımlar.
    """
    
    def __init__(self, isim, email):
        self.isim = isim
        self.email = email
        self.toplam_alisveris = 0
    
    def indirim_hesapla(self, tutar):
        """
        ⭐ POLYMORPHİC METHOD - Her child farklı uygular!
        
        Bu metod parent'ta temel bir implementasyon sağlar.
        Her child class kendi indirim mantığını override eder.
        """
        # Varsayılan: İndirim yok
        return tutar
    
    def alisveris_yap(self, tutar):
        """Alışveriş işlemi - indirim_hesapla metodunu kullanır"""
        indirimli_tutar = self.indirim_hesapla(tutar)
        indirim = tutar - indirimli_tutar
        
        print(f"\n💰 ALIŞVERİŞ DETAYI - {self.isim}")
        print(f"   📦 Tutar        : {tutar:,.2f} TL")
        print(f"   🎁 İndirim      : {indirim:,.2f} TL")
        print(f"   💳 Ödenecek     : {indirimli_tutar:,.2f} TL")
        
        self.toplam_alisveris += indirimli_tutar
        return indirimli_tutar

print("✅ MusteriBase (Polymorphism için) hazır!\n")
````

````python
# 🥉 CHILD 1: Standart Müşteri - %5 İndirim

class StandartMusteri(MusteriBase):
    """
    👤 Standart Müşteri
    
    Yeni müşteriler için %5 sabit indirim
    """
    
    def __init__(self, isim, email):
        super().__init__(isim, email)
        self.tip = "Standart"
        self.indirim_orani = 0.05  # %5
    
    def indirim_hesapla(self, tutar):
        """
        ⭐ POLYMORPHİC METHOD - Standart indirim
        """
        indirim = tutar * self.indirim_orani
        return tutar - indirim


# 🥈 CHILD 2: VIP Müşteri - %15 İndirim

class VIPMusteri(MusteriBase):
    """
    ⭐ VIP Müşteri
    
    Sadık müşteriler için %15 özel indirim
    """
    
    def __init__(self, isim, email, uyelik_yili):
        super().__init__(isim, email)
        self.tip = "VIP"
        self.uyelik_yili = uyelik_yili
        self.indirim_orani = 0.15  # %15
    
    def indirim_hesapla(self, tutar):
        """
        ⭐ POLYMORPHİC METHOD - VIP indirim
        
        Bonus: 3 yıldan fazla üye ise ekstra %5
        """
        suanki_yil = datetime.now().year
        uyelik_suresi = suanki_yil - self.uyelik_yili
        
        # Temel VIP indirimi
        indirim_orani = self.indirim_orani
        
        # 3 yıldan fazla üyelerde ekstra indirim
        if uyelik_suresi > 3:
            indirim_orani += 0.05  # Ekstra %5
            print(f"   🎉 {uyelik_suresi} yıllık sadakat bonusu: +%5!")
        
        indirim = tutar * indirim_orani
        return tutar - indirim


# 🥇 CHILD 3: Kurumsal Müşteri - Kademeli İndirim

class KurumsalMusteriPoly(MusteriBase):
    """
    🏢 Kurumsal Müşteri
    
    Tutara göre kademeli indirim sistemi
    """
    
    def __init__(self, isim, email, calisan_sayisi):
        super().__init__(isim, email)
        self.tip = "Kurumsal"
        self.calisan_sayisi = calisan_sayisi
    
    def indirim_hesapla(self, tutar):
        """
        ⭐ POLYMORPHİC METHOD - Kademeli kurumsal indirim
        
        Alışveriş tutarına göre değişken indirim:
        - 10,000 TL altı   : %10
        - 10,000-50,000 TL : %20
        - 50,000 TL üstü   : %30
        """
        if tutar < 10000:
            indirim_orani = 0.10
            print("   📊 İndirim kademesi: %10 (Standart)")
        elif tutar < 50000:
            indirim_orani = 0.20
            print("   📊 İndirim kademesi: %20 (Orta)")
        else:
            indirim_orani = 0.30
            print("   📊 İndirim kademesi: %30 (Yüksek)")
        
        indirim = tutar * indirim_orani
        return tutar - indirim


print("🎉 Tüm polymorphic müşteri sınıfları oluşturuldu!\n")
print("✅ Her sınıfta 'indirim_hesapla' metodu var")
print("✅ Ama her biri FARKLI şekilde çalışıyor!")
print("✅ Bu POLYMORPHİSM'dir! 🎭")
````

````python
# 🎬 POLYMORPHİSM DEMONSTRASYONU

print("="*70)
print("🎭 POLYMORPHİSM GÖSTERGESI - AYNI METHOD, FARKLI DAVRANIŞLAR")
print("="*70 + "\n")

# Farklı müşteri tipleri oluştur
musteriler = [
    StandartMusteri("Ali Veli", "ali@gmail.com"),
    VIPMusteri("Ayşe Yıldız", "ayse@gmail.com", 2018),  # 8 yıllık üye
    KurumsalMusteriPoly("TechCorp A.Ş.", "info@techcorp.com", 100)
]

# Alışveriş tutarı
alisveris_tutari = 25000  # 25,000 TL

print(f"🛒 Alışveriş Tutarı: {alisveris_tutari:,} TL\n")
print("-"*70)

# 🔥 İŞTE POLYMORPHİSM!
# Aynı metod (alisveris_yap) ama her müşteri için farklı indirim!

for musteri in musteriler:
    print(f"\n👤 Müşteri Tipi: {musteri.tip}")
    musteri.alisveris_yap(alisveris_tutari)
    print("-"*70)

print("\n" + "="*70)
print("🎯 GÖRDÜNÜZ MÜ? 3 farklı nesne, aynı metod ismi")
print("   Ama her biri FARKLI şekilde çalıştı! Bu POLYMORPHİSM!")
print("="*70)
````

## 💼 **GERÇEK YBS UYGULAMASI: Toplu İşlem Fonksiyonu**

Polymorphism'in asıl gücü: **Tek fonksiyon, tüm tipler!**

````python
# 💼 POLYMORPHİC FONKSİYON - Tüm müşteri tipleri için çalışır!

def toplu_fatura_isle(musteri_listesi, fatura_tutari):
    """
    🎯 POLYMORPHİC FONKSİYON
    
    Herhangi bir müşteri tipi kabul eder.
    Her müşteri kendi indirim_hesapla metodunu kullanır.
    
    Bu fonksiyon müşteri tipini bilmek zorunda değil!
    Duck Typing: "Ördek gibi yürüyorsa, ördek gibi seslenirse, ördek gibidir"
    """
    print("="*70)
    print("📊 TOPLU FATURA İŞLEME SİSTEMİ")
    print("="*70 + "\n")
    
    toplam_satis = 0
    toplam_indirim = 0
    
    for i, musteri in enumerate(musteri_listesi, 1):
        print(f"\n{'─'*70}")
        print(f"📋 Fatura #{i} - {musteri.tip} Müşteri: {musteri.isim}")
        print(f"{'─'*70}")
        
        # 🔥 POLYMORPHİSM: Her müşteri kendi indirim_hesapla'sını çalıştırır!
        indirimli_tutar = musteri.alisveris_yap(fatura_tutari)
        indirim = fatura_tutari - indirimli_tutar
        
        toplam_satis += indirimli_tutar
        toplam_indirim += indirim
    
    print(f"\n{'='*70}")
    print("📈 ÖZET RAPOR")
    print(f"{'='*70}")
    print(f"📦 İşlenen Fatura Sayısı  : {len(musteri_listesi)}")
    print(f"💰 Toplam Satış (İndirimli): {toplam_satis:,.2f} TL")
    print(f"🎁 Toplam İndirim          : {toplam_indirim:,.2f} TL")
    print(f"📊 Ortalama İndirim Oranı  : %{(toplam_indirim/(fatura_tutari*len(musteri_listesi))*100):.1f}")
    print(f"{'='*70}\n")
    
    return toplam_satis, toplam_indirim


# Test edelim!
print("🧪 POLYMORPHİC FONKSİYON TESTİ\n")

# Karışık müşteri listesi
karisik_musteriler = [
    StandartMusteri("Mehmet Öz", "mehmet@gmail.com"),
    VIPMusteri("Zeynep Kaya", "zeynep@gmail.com", 2015),  # 11 yıllık üye
    KurumsalMusteriPoly("DataTech Ltd.", "info@datatech.com", 50),
    StandartMusteri("Can Demir", "can@gmail.com"),
    VIPMusteri("Elif Şahin", "elif@gmail.com", 2022),  # 4 yıllık üye
]

# Tek fonksiyon çağrısı - tüm tipler için çalışır!
toplu_fatura_isle(karisik_musteriler, 15000)

print("✅ Gördünüz mü? Fonksiyon müşteri tipini bilmiyor!")
print("✅ Her müşteri kendi indirim mantığını uyguladı!")
print("✅ Bu POLYMORPHİSM'in gücüdür! 🎭")
````

## 📊 **POLYMORPHİSM + VERİ ANALİZİ**

Polymorphism'i veri analizi ile birleştirelim!

````python
# 📊 POLYMORPHİSM + PANDAS: Müşteri Analizi

print("📊 POLYMORPHİSM İLE MÜŞTERİ ANALİZİ")
print("="*70 + "\n")

# Simülasyon: 100 rastgele müşteri oluştur
np.random.seed(42)

musteri_simule = []
for i in range(100):
    tip = np.random.choice(['Standart', 'VIP', 'Kurumsal'], p=[0.6, 0.25, 0.15])
    
    if tip == 'Standart':
        m = StandartMusteri(f"Müşteri_{i+1}", f"musteri{i+1}@test.com")
    elif tip == 'VIP':
        yil = np.random.randint(2010, 2024)
        m = VIPMusteri(f"VIP_{i+1}", f"vip{i+1}@test.com", yil)
    else:
        calisan = np.random.randint(10, 500)
        m = KurumsalMusteriPoly(f"Şirket_{i+1}", f"sirket{i+1}@test.com", calisan)
    
    musteri_simule.append(m)

print(f"✅ {len(musteri_simule)} müşteri oluşturuldu!\n")

# Her müşteriye rastgele tutarlarda alışveriş yaptır
alisveris_verileri = []

for musteri in musteri_simule:
    tutar = np.random.uniform(5000, 100000)  # 5K - 100K TL arası
    indirimli_tutar = musteri.indirim_hesapla(tutar)
    
    alisveris_verileri.append({
        'musteri_adi': musteri.isim,
        'musteri_tip': musteri.tip,
        'tutar': tutar,
        'indirimli_tutar': indirimli_tutar,
        'indirim': tutar - indirimli_tutar,
        'indirim_orani': ((tutar - indirimli_tutar) / tutar) * 100
    })

# DataFrame oluştur
df_alisveris = pd.DataFrame(alisveris_verileri)

print("📋 İLK 10 ALIŞVERİŞ:")
print(df_alisveris.head(10).to_string(index=False))

print("\n" + "="*70)
print("📈 MÜŞTERİ TİPİNE GÖRE ANALİZ")
print("="*70 + "\n")

# Tip bazında özet
ozet = df_alisveris.groupby('musteri_tip').agg({
    'tutar': ['count', 'sum', 'mean'],
    'indirim': ['sum', 'mean'],
    'indirim_orani': 'mean'
}).round(2)

print(ozet)

print("\n" + "="*70)
print("💡 POLYMORPHİSM ETKİSİ:")
print("="*70)
print("✅ Her müşteri tipi kendi indirim mantığını uyguladı")
print("✅ Aynı kod (indirim_hesapla) farklı sonuçlar üretti")
print("✅ Veri analizi polymorphic nesnelerle yapıldı!")
print("="*70)
````

## 🔰 **ADIM ADIM ÖĞRENELIM**

### 🎓 **Adım 1: Python'da Erişim Seviyeleri**

Python'da 3 tip erişim seviyesi var:

| Prefix | İsim | Erişim | Örnek |
|--------|------|--------|-------|
| Yok | **Public** | Her yerden erişilebilir | `self.isim` |
| `_` | **Protected** | Sadece class ve subclass | `self._email` |
| `__` | **Private** | Sadece class içinden | `self.__sifre` |

> ⚠️ **ÖNEMLİ:** Python'da "gerçek" private yok! Sadece **konvansiyon** var.
> `__attribute` aslında `_ClassName__attribute` olarak mangle edilir.

````python
# 🔓 KÖTÜ ÖRNEK: Encapsulation YOK!

class BankaHesabiKotu:
    """❌ Kötü tasarım - Hiçbir koruma yok!"""
    
    def __init__(self, hesap_no, bakiye):
        self.hesap_no = hesap_no
        self.bakiye = bakiye  # Public - herkes değiştirebilir!
    
    def bilgi_goster(self):
        print(f"Hesap: {self.hesap_no} - Bakiye: {self.bakiye} TL")

print("❌ ENCAPSULATION OLMADAN:")
print("="*60 + "\n")

hesap1 = BankaHesabiKotu("TR001", 10000)
hesap1.bilgi_goster()

# 🚨 PROBLEM: Dışarıdan doğrudan değiştirilebilir!
print("\n🚨 Birisi kötü niyetli kod yazarsa:")
hesap1.bakiye = -999999  # Negatif bakiye!?
hesap1.bilgi_goster()

print("\n💥 Bakiye negatif oldu! Sistem çöktü! 💥")
print("="*60)
````

````python
# 🔒 İYİ ÖRNEK: ENCAPSULATION VAR!

class BankaHesabiIyi:
    """✅ İyi tasarım - Veri koruması var!"""
    
    def __init__(self, hesap_no, bakiye):
        self.hesap_no = hesap_no
        self.__bakiye = bakiye  # Private (çift alt çizgi)
        self.__islem_gecmisi = []  # Private liste
    
    # GETTER - Bakiyeyi okuma (read-only)
    def get_bakiye(self):
        """Bakiyeyi güvenli şekilde döndürür"""
        return self.__bakiye
    
    # SETTER - Bakiyeyi değiştirme (validation ile)
    def para_yatir(self, miktar):
        """Para yatırma - kontrollü erişim"""
        if miktar <= 0:
            print("❌ Hata: Miktar 0'dan büyük olmalı!")
            return False
        
        self.__bakiye += miktar
        self.__islem_gecmisi.append(f"+{miktar} TL yatırıldı")
        print(f"✅ {miktar} TL yatırıldı. Yeni bakiye: {self.__bakiye} TL")
        return True
    
    def para_cek(self, miktar):
        """Para çekme - kontrollü erişim"""
        if miktar <= 0:
            print("❌ Hata: Miktar 0'dan büyük olmalı!")
            return False
        
        if miktar > self.__bakiye:
            print(f"❌ Yetersiz bakiye! Mevcut: {self.__bakiye} TL")
            return False
        
        self.__bakiye -= miktar
        self.__islem_gecmisi.append(f"-{miktar} TL çekildi")
        print(f"✅ {miktar} TL çekildi. Kalan bakiye: {self.__bakiye} TL")
        return True
    
    def islem_gecmisini_goster(self):
        """İşlem geçmişini göster"""
        if not self.__islem_gecmisi:
            print("📭 Henüz işlem yok.")
            return
        
        print("\n📜 İŞLEM GEÇMİŞİ:")
        for i, islem in enumerate(self.__islem_gecmisi, 1):
            print(f"   {i}. {islem}")
    
    def bilgi_goster(self):
        print(f"🏦 Hesap: {self.hesap_no} - Bakiye: {self.__bakiye} TL")

print("✅ ENCAPSULATION İLE:")
print("="*60 + "\n")

hesap2 = BankaHesabiIyi("TR002", 10000)
hesap2.bilgi_goster()

print("\n💰 Para yatırma denemeleri:")
hesap2.para_yatir(5000)
hesap2.para_yatir(-100)  # Negatif - reddedilecek!

print("\n💸 Para çekme denemeleri:")
hesap2.para_cek(3000)
hesap2.para_cek(20000)  # Yetersiz bakiye - reddedilecek!

print("\n🔒 Private attribute'a doğrudan erişim denemesi:")
try:
    print(hesap2.__bakiye)  # Hata verecek!
except AttributeError as e:
    print(f"❌ Hata: {e}")
    print("💡 Doğrudan erişilemez! get_bakiye() kullanmalısınız.")

print(f"\n✅ Doğru yöntem: get_bakiye() = {hesap2.get_bakiye()} TL")

# İşlem geçmişi
hesap2.islem_gecmisini_goster()

print("\n" + "="*60)
print("🎯 Bakiye korundu! Sadece kontrollü erişim mümkün!")
print("="*60)
````

### 🎓 **Adım 2: @property Decorator - Pythonic Yöntem**

Python'da getter/setter için daha şık bir yöntem: **@property**

> 💡 **AVANTAJ:** Method gibi çağırılır ama attribute gibi kullanılır!

````python
# 🎨 @property ile MODERN ENCAPSULATION

class MusteriGuvenli:
    """
    🔒 Güvenli Müşteri Sınıfı
    
    @property decorator ile modern encapsulation
    """
    
    def __init__(self, isim, email, yas):
        self.__isim = isim
        self.__email = email
        self.__yas = yas
        self.__bakiye = 0
        self.__kredi_kart_no = None
    
    # 📖 GETTER - @property decorator ile
    @property
    def isim(self):
        """İsim getter - okuma erişimi"""
        return self.__isim.title()  # İlk harfleri büyük
    
    @property
    def email(self):
        """Email getter"""
        return self.__email.lower()  # Küçük harfe çevir
    
    @property
    def yas(self):
        """Yaş getter"""
        return self.__yas
    
    @property
    def bakiye(self):
        """Bakiye getter - sadece okuma"""
        return self.__bakiye
    
    # ✏️ SETTER - @attribute.setter decorator ile
    @yas.setter
    def yas(self, yeni_yas):
        """Yaş setter - validation ile"""
        if not isinstance(yeni_yas, int):
            raise TypeError("Yaş tam sayı olmalı!")
        
        if yeni_yas < 18:
            raise ValueError("Yaş 18'den küçük olamaz!")
        
        if yeni_yas > 120:
            raise ValueError("Geçersiz yaş!")
        
        self.__yas = yeni_yas
        print(f"✅ Yaş güncellendi: {yeni_yas}")
    
    @email.setter
    def email(self, yeni_email):
        """Email setter - validation ile"""
        if '@' not in yeni_email:
            raise ValueError("Geçersiz email formatı!")
        
        self.__email = yeni_email
        print(f"✅ Email güncellendi: {yeni_email}")
    
    # 💳 Kredi kartı - read-only (setter yok!)
    @property
    def kredi_kart_no(self):
        """Kredi kartı - güvenli gösterim"""
        if self.__kredi_kart_no is None:
            return "Kayıtlı kart yok"
        
        # Son 4 hane hariç gizle
        return f"**** **** **** {self.__kredi_kart_no[-4:]}"
    
    def kredi_karti_ekle(self, kart_no):
        """Kredi kartı ekleme - private metod ile"""
        if len(str(kart_no)) != 16:
            raise ValueError("Kart numarası 16 hane olmalı!")
        
        self.__kredi_kart_no = str(kart_no)
        print(f"✅ Kart eklendi: {self.kredi_kart_no}")
    
    def para_yukle(self, miktar):
        """Bakiye yükleme"""
        if miktar > 0:
            self.__bakiye += miktar
            print(f"💰 {miktar} TL yüklendi")
    
    def bilgileri_goster(self):
        """Müşteri bilgilerini göster"""
        print(f"\n{'='*50}")
        print(f"👤 İsim           : {self.isim}")
        print(f"📧 Email          : {self.email}")
        print(f"🎂 Yaş            : {self.yas}")
        print(f"💰 Bakiye         : {self.bakiye} TL")
        print(f"💳 Kredi Kartı    : {self.kredi_kart_no}")
        print(f"{'='*50}")

print("🎨 @property DECORATOR İLE ENCAPSULATION")
print("="*70 + "\n")

# Müşteri oluştur
ali = MusteriGuvenli("ali veli", "ALI@GMAIL.COM", 25)
ali.bilgileri_goster()

print("\n" + "-"*70)
print("📖 GETTER KULLANIMI - Attribute gibi erişim:")
print("-"*70)
print(f"   İsim: {ali.isim}")  # Method gibi değil, attribute gibi!
print(f"   Email: {ali.email}")
print(f"   Yaş: {ali.yas}")

print("\n" + "-"*70)
print("✏️ SETTER KULLANIMI - Attribute gibi değiştirme:")
print("-"*70)
ali.yas = 30  # Setter otomatik çağrılır
ali.email = "ali.yeni@gmail.com"

print("\n" + "-"*70)
print("❌ GEÇERSİZ DEĞER DENEMELERİ:")
print("-"*70)

try:
    ali.yas = 15  # 18'den küçük - hata!
except ValueError as e:
    print(f"   ❌ Yaş hatası: {e}")

try:
    ali.email = "gecersiz-email"  # @ yok - hata!
except ValueError as e:
    print(f"   ❌ Email hatası: {e}")

print("\n" + "-"*70)
print("🔒 SADECE OKUMA (READ-ONLY) ÖZELLİKLER:")
print("-"*70)

# Bakiye - sadece okuma
print(f"   💰 Bakiye: {ali.bakiye} TL")

# Doğrudan değiştirme denemesi
try:
    ali.bakiye = 999999  # Setter yok - hata!
except AttributeError as e:
    print(f"   ❌ Bakiye değiştirilemez! Sadece para_yukle() ile.")

# Doğru yöntem
ali.para_yukle(5000)
print(f"   ✅ Yeni bakiye: {ali.bakiye} TL")

print("\n" + "-"*70)
print("💳 KREDİ KARTI GİZLİLİĞİ:")
print("-"*70)

ali.kredi_karti_ekle("1234567890123456")
print(f"   Kart No: {ali.kredi_kart_no}")  # Gizli gösterilir

# Güncel bilgiler
ali.bilgileri_goster()

print("\n" + "="*70)
print("✅ @property sayesinde:")
print("   • Attribute gibi kullanım (ali.yas)")
print("   • Otomatik validation")
print("   • Read-only özellikler")
print("   • Veri gizliliği ve güvenlik")
print("="*70)
````

## 💼 **GERÇEK YBS UYGULAMASI: E-Ticaret Güvenlik Sistemi**

Şimdi öğrendiklerimizi gerçek bir YBS senaryosunda kullanalım!

````python
# 🏢 KAPSAMLI YBS UYGULAMASI: Güvenli E-Ticaret Sistemi

import hashlib  # Şifre hashleme için

class GuvenliMusteri:
    """
    🔒 Güvenli E-Ticaret Müşteri Sistemi
    
    KVKK uyumlu, güvenli veri yönetimi
    """
    
    _toplam_musteri = 0  # Protected class variable
    
    def __init__(self, isim, email, sifre, tc_no):
        # Public attributes
        self.musteri_id = f"M{GuvenliMusteri._toplam_musteri + 1:05d}"
        self.kayit_tarihi = datetime.now()
        
        # Protected attributes (konvansiyon)
        self._isim = isim
        self._email = email
        
        # Private attributes (gerçek gizli veri)
        self.__sifre_hash = self.__hash_sifre(sifre)
        self.__tc_no = tc_no
        self.__bakiye = 0
        self.__kredi_kart_bilgileri = []
        self.__siparis_gecmisi = []
        
        GuvenliMusteri._toplam_musteri += 1
        print(f"✅ Müşteri oluşturuldu: {self.musteri_id}")
    
    # 🔐 Private metodlar
    def __hash_sifre(self, sifre):
        """Şifreyi hashler (private metod)"""
        return hashlib.sha256(sifre.encode()).hexdigest()
    
    def __tc_dogrula(self, tc):
        """TC doğrulama algoritması (basitleştirilmiş)"""
        return len(str(tc)) == 11 and str(tc).isdigit()
    
    # 📖 Property'ler - Kontrollü erişim
    @property
    def isim(self):
        """İsim - read-only dışardan"""
        return self._isim.title()
    
    @property
    def email(self):
        """Email - read-only"""
        return self._email
    
    @property
    def bakiye(self):
        """Bakiye - read-only"""
        return self.__bakiye
    
    @property
    def tc_no_maskeli(self):
        """TC'yi maskeli göster (KVKK uyumu)"""
        tc = str(self.__tc_no)
        return f"{tc[:3]}****{tc[-2:]}"
    
    @property
    def siparis_sayisi(self):
        """Sipariş sayısı"""
        return len(self.__siparis_gecmisi)
    
    # 🔐 Şifre işlemleri
    def sifre_dogrula(self, sifre):
        """Şifre kontrolü"""
        girilen_hash = self.__hash_sifre(sifre)
        if girilen_hash == self.__sifre_hash:
            print("✅ Şifre doğru!")
            return True
        else:
            print("❌ Hatalı şifre!")
            return False
    
    def sifre_degistir(self, eski_sifre, yeni_sifre):
        """Şifre değiştirme"""
        if not self.sifre_dogrula(eski_sifre):
            return False
        
        if len(yeni_sifre) < 6:
            print("❌ Yeni şifre en az 6 karakter olmalı!")
            return False
        
        self.__sifre_hash = self.__hash_sifre(yeni_sifre)
        print("✅ Şifre başarıyla değiştirildi!")
        return True
    
    # 💳 Kredi kartı işlemleri
    def kredi_karti_ekle(self, kart_no, kart_sahibi, son_kullanma):
        """Kredi kartı ekleme - güvenli"""
        if len(str(kart_no)) != 16:
            print("❌ Geçersiz kart numarası!")
            return False
        
        # Kartı hashleyerek sakla (güvenlik)
        kart_hash = self.__hash_sifre(str(kart_no))
        
        self.__kredi_kart_bilgileri.append({
            'kart_hash': kart_hash,
            'son_4_hane': str(kart_no)[-4:],
            'kart_sahibi': kart_sahibi,
            'son_kullanma': son_kullanma
        })
        
        print(f"✅ Kart eklendi: **** **** **** {str(kart_no)[-4:]}")
        return True
    
    def kartlari_listele(self):
        """Kayıtlı kartları listele (güvenli)"""
        if not self.__kredi_kart_bilgileri:
            print("💳 Kayıtlı kart yok")
            return
        
        print("\n💳 KAYITLI KARTLAR:")
        for i, kart in enumerate(self.__kredi_kart_bilgileri, 1):
            print(f"   {i}. **** **** **** {kart['son_4_hane']}")
            print(f"      Sahip: {kart['kart_sahibi']}")
            print(f"      S.K.T: {kart['son_kullanma']}")
    
    # 💰 Bakiye işlemleri
    def bakiye_yukle(self, miktar, sifre):
        """Bakiye yükleme - şifre gerekli"""
        if not self.sifre_dogrula(sifre):
            return False
        
        if miktar <= 0:
            print("❌ Geçersiz miktar!")
            return False
        
        self.__bakiye += miktar
        print(f"💰 {miktar:,} TL yüklendi. Yeni bakiye: {self.__bakiye:,} TL")
        return True
    
    # 🛒 Sipariş işlemleri
    def siparis_ver(self, urun, fiyat, sifre):
        """Sipariş verme - şifre ve bakiye kontrolü"""
        if not self.sifre_dogrula(sifre):
            return False
        
        if fiyat > self.__bakiye:
            print(f"❌ Yetersiz bakiye! Gerekli: {fiyat:,} TL")
            return False
        
        self.__bakiye -= fiyat
        self.__siparis_gecmisi.append({
            'urun': urun,
            'fiyat': fiyat,
            'tarih': datetime.now()
        })
        
        print(f"🎉 Sipariş tamamlandı: {urun}")
        print(f"   Fiyat: {fiyat:,} TL")
        print(f"   Kalan bakiye: {self.__bakiye:,} TL")
        return True
    
    def siparis_gecmisi_goster(self, sifre):
        """Sipariş geçmişi - şifre gerekli"""
        if not self.sifre_dogrula(sifre):
            return
        
        if not self.__siparis_gecmisi:
            print("📭 Sipariş geçmişi yok")
            return
        
        print(f"\n📜 SİPARİŞ GEÇMİŞİ - {self.isim}")
        print("="*60)
        for i, sip in enumerate(self.__siparis_gecmisi, 1):
            print(f"\n{i}. {sip['urun']}")
            print(f"   Fiyat: {sip['fiyat']:,} TL")
            print(f"   Tarih: {sip['tarih'].strftime('%d.%m.%Y %H:%M')}")
        
        toplam = sum(s['fiyat'] for s in self.__siparis_gecmisi)
        print(f"\n{'='*60}")
        print(f"💵 Toplam Harcama: {toplam:,} TL")
    
    # 📊 Bilgi gösterme
    def bilgileri_goster(self):
        """Genel bilgiler (hassas veri yok)"""
        print(f"\n{'='*60}")
        print(f"👤 MÜŞTERİ BİLGİLERİ")
        print(f"{'='*60}")
        print(f"🆔 ID             : {self.musteri_id}")
        print(f"📛 İsim           : {self.isim}")
        print(f"📧 Email          : {self.email}")
        print(f"🆔 TC (Maskeli)   : {self.tc_no_maskeli}")
        print(f"💰 Bakiye         : {self.bakiye:,} TL")
        print(f"🛒 Sipariş Sayısı : {self.siparis_sayisi}")
        print(f"📅 Kayıt Tarihi   : {self.kayit_tarihi.strftime('%d.%m.%Y')}")
        print(f"{'='*60}")

print("🔒 GÜVENLİ E-TİCARET SİSTEMİ OLUŞTURULDU!")
print("="*70)
````

## 🔰 **PYTHON'DA ABSTRACTION**

### 🎓 **ABC (Abstract Base Class) Modülü**

Python'da abstraction için `abc` modülü kullanılır:

```python
from abc import ABC, abstractmethod

class AbstractClass(ABC):
    @abstractmethod
    def zorunlu_metod(self):
        pass
```

> ⚠️ **KURAL:** Abstract class'tan **direkt nesne oluşturamazsınız!**  
> Önce bir child class oluşturup abstract metodları implement etmelisiniz.

````python
# 📐 İLK ABSTRACT CLASS ÖRNEĞ

from abc import ABC, abstractmethod

class Sekil(ABC):
    """
    📐 Abstract Base Class: Şekil
    
    Tüm şekillerin uyması gereken sözleşme (contract)
    """
    
    def __init__(self, isim):
        self.isim = isim
    
    @abstractmethod
    def alan_hesapla(self):
        """
        ⭐ ABSTRACT METHOD - Child class'lar MUTLAKA implement etmeli!
        
        Bu metod burada sadece tanımlanır, implementasyon child'da yapılır.
        """
        pass
    
    @abstractmethod
    def cevre_hesapla(self):
        """Abstract method - Child'lar implement etmeli"""
        pass
    
    # Normal metod (abstract değil)
    def bilgi_goster(self):
        """Concrete method - Child'lar kullanabilir"""
        print(f"\n📐 Şekil: {self.isim}")
        print(f"   Alan: {self.alan_hesapla():.2f}")
        print(f"   Çevre: {self.cevre_hesapla():.2f}")

print("✅ Abstract Base Class (Sekil) tanımlandı!")
print("\n⚠️ ÖNEMLİ: Abstract class'tan direkt nesne oluşturamazsınız!")
print("\n❌ Deneme:")

try:
    s = Sekil("Test")  # HATA verecek!
except TypeError as e:
    print(f"   Hata: {e}")
    print("   💡 Abstract metodları implement etmelisiniz!")
````

````python
# ✅ CONCRETE CLASSES - Abstract metodları implement edelim!

class Kare(Sekil):
    """Kare sınıfı - Sekil'den türer"""
    
    def __init__(self, kenar):
        super().__init__("Kare")
        self.kenar = kenar
    
    # Abstract metodları implement etmeliyiz!
    def alan_hesapla(self):
        """Alan = kenar × kenar"""
        return self.kenar * self.kenar
    
    def cevre_hesapla(self):
        """Çevre = 4 × kenar"""
        return 4 * self.kenar


class Daire(Sekil):
    """Daire sınıfı - Sekil'den türer"""
    
    def __init__(self, yaricap):
        super().__init__("Daire")
        self.yaricap = yaricap
    
    def alan_hesapla(self):
        """Alan = π × r²"""
        import math
        return math.pi * (self.yaricap ** 2)
    
    def cevre_hesapla(self):
        """Çevre = 2 × π × r"""
        import math
        return 2 * math.pi * self.yaricap


class Dikdortgen(Sekil):
    """Dikdörtgen sınıfı - Sekil'den türer"""
    
    def __init__(self, uzun_kenar, kisa_kenar):
        super().__init__("Dikdörtgen")
        self.uzun_kenar = uzun_kenar
        self.kisa_kenar = kisa_kenar
    
    def alan_hesapla(self):
        """Alan = uzun × kısa"""
        return self.uzun_kenar * self.kisa_kenar
    
    def cevre_hesapla(self):
        """Çevre = 2 × (uzun + kısa)"""
        return 2 * (self.uzun_kenar + self.kisa_kenar)


print("✅ Concrete class'lar oluşturuldu!")
print("\n🧪 TEST EDELİM:")

# Şekiller oluştur
kare = Kare(5)
daire = Daire(3)
dikdortgen = Dikdortgen(8, 4)

# Bilgileri göster
kare.bilgi_goster()
daire.bilgi_goster()
dikdortgen.bilgi_goster()

print("\n" + "="*60)
print("🎯 Her şekil aynı arayüzü kullandı!")
print("   • alan_hesapla()")
print("   • cevre_hesapla()")
print("   • bilgi_goster()")
print("="*60)
````

## 💼 **GERÇEK YBS UYGULAMASI: Ödeme Sistemi**

E-ticaret sitesinde farklı ödeme yöntemleri olsun. Hepsi aynı arayüzü kullansın!

````python
# 💳 ABSTRACT CLASS: OdemeYontemi

class OdemeYontemi(ABC):
    """
    💳 Abstract Base Class: Ödeme Yöntemi
    
    Tüm ödeme yöntemlerinin uyması gereken sözleşme
    """
    
    def __init__(self, isim):
        self.isim = isim
        self.islem_ucreti = 0
    
    @abstractmethod
    def odeme_al(self, tutar, musteri_bilgi):
        """
        ⭐ ZORUNLU: Ödeme alma işlemi
        Her ödeme yöntemi bunu kendi yöntemine göre uygular
        """
        pass
    
    @abstractmethod
    def iade_yap(self, tutar, islem_id):
        """⭐ ZORUNLU: İade işlemi"""
        pass
    
    @abstractmethod
    def dogrulama_yap(self, bilgiler):
        """⭐ ZORUNLU: Ödeme bilgilerini doğrula"""
        pass
    
    # Concrete metodlar (tüm child'lar kullanabilir)
    def islem_ucreti_hesapla(self, tutar):
        """İşlem ücretini hesapla"""
        return tutar * self.islem_ucreti
    
    def toplam_tutar_hesapla(self, tutar):
        """İşlem ücreti dahil toplam"""
        return tutar + self.islem_ucreti_hesapla(tutar)
    
    def odeme_raporu(self, tutar):
        """Ödeme özet raporu"""
        print(f"\n💳 ÖDEME YÖNTEMİ: {self.isim}")
        print(f"   Tutar: {tutar:,.2f} TL")
        print(f"   İşlem Ücreti: {self.islem_ucreti_hesapla(tutar):,.2f} TL")
        print(f"   Toplam: {self.toplam_tutar_hesapla(tutar):,.2f} TL")

print("✅ OdemeYontemi abstract class'ı tanımlandı!")
````

````python
# 💳 CONCRETE CLASS 1: Kredi Kartı Ödemesi

class KrediKartiOdeme(OdemeYontemi):
    """💳 Kredi kartı ile ödeme"""
    
    def __init__(self):
        super().__init__("Kredi Kartı")
        self.islem_ucreti = 0.025  # %2.5 komisyon
    
    def dogrulama_yap(self, bilgiler):
        """Kart bilgilerini doğrula"""
        kart_no = bilgiler.get('kart_no', '')
        cvv = bilgiler.get('cvv', '')
        
        if len(str(kart_no)) != 16:
            return False, "Geçersiz kart numarası!"
        
        if len(str(cvv)) != 3:
            return False, "Geçersiz CVV!"
        
        return True, "Kart bilgileri geçerli"
    
    def odeme_al(self, tutar, musteri_bilgi):
        """Kredi kartı ile ödeme işlemi"""
        print(f"\n{'='*60}")
        print(f"💳 KREDİ KARTI İLE ÖDEME")
        print(f"{'='*60}")
        
        # Doğrulama
        gecerli, mesaj = self.dogrulama_yap(musteri_bilgi)
        if not gecerli:
            print(f"❌ {mesaj}")
            return False
        
        # Ödeme işlemi
        toplam = self.toplam_tutar_hesapla(tutar)
        self.odeme_raporu(tutar)
        
        # Taksit seçenekleri
        print(f"\n🎁 Taksit Seçenekleri:")
        for taksit in [1, 3, 6, 9, 12]:
            aylik = toplam / taksit
            print(f"   {taksit:>2} Taksit: {aylik:>8,.2f} TL/ay")
        
        print(f"\n✅ Ödeme başarılı! İşlem No: KK{np.random.randint(100000, 999999)}")
        return True
    
    def iade_yap(self, tutar, islem_id):
        """İade işlemi"""
        print(f"\n🔄 İADE İŞLEMİ")
        print(f"   İşlem No: {islem_id}")
        print(f"   Tutar: {tutar:,.2f} TL")
        print(f"   ✅ İade 3-5 iş günü içinde kartınıza yansıyacaktır.")
        return True


# 🏦 CONCRETE CLASS 2: Havale/EFT Ödemesi

class HavaleOdeme(OdemeYontemi):
    """🏦 Havale/EFT ile ödeme"""
    
    def __init__(self):
        super().__init__("Havale/EFT")
        self.islem_ucreti = 0  # Komisyon yok!
    
    def dogrulama_yap(self, bilgiler):
        """IBAN doğrulama"""
        iban = bilgiler.get('iban', '')
        
        if not iban.startswith('TR'):
            return False, "IBAN TR ile başlamalı!"
        
        if len(iban) != 26:
            return False, "IBAN 26 karakter olmalı!"
        
        return True, "IBAN geçerli"
    
    def odeme_al(self, tutar, musteri_bilgi):
        """Havale ile ödeme"""
        print(f"\n{'='*60}")
        print(f"🏦 HAVALE/EFT İLE ÖDEME")
        print(f"{'='*60}")
        
        gecerli, mesaj = self.dogrulama_yap(musteri_bilgi)
        if not gecerli:
            print(f"❌ {mesaj}")
            return False
        
        self.odeme_raporu(tutar)
        
        print(f"\n📝 HAVALE BİLGİLERİ:")
        print(f"   Banka: TechShop Bankası")
        print(f"   IBAN: TR33 0006 1005 1978 6457 8413 26")
        print(f"   Açıklama: Sipariş - {musteri_bilgi.get('siparis_no', 'N/A')}")
        
        print(f"\n⏱️  Havale yapıldıktan sonra siparişiniz işleme alınacaktır.")
        print(f"✅ Ödeme talimatı oluşturuldu!")
        return True
    
    def iade_yap(self, tutar, islem_id):
        """İade işlemi"""
        print(f"\n🔄 İADE İŞLEMİ")
        print(f"   İşlem No: {islem_id}")
        print(f"   Tutar: {tutar:,.2f} TL")
        print(f"   ✅ İade hesabınıza 1-2 iş günü içinde yapılacaktır.")
        return True


# 💵 CONCRETE CLASS 3: Kapıda Ödeme

class KapidaOdeme(OdemeYontemi):
    """💵 Kapıda ödeme (Nakit/Kart)"""
    
    def __init__(self):
        super().__init__("Kapıda Ödeme")
        self.islem_ucreti = 0.015  # %1.5 ekstra ücret
    
    def dogrulama_yap(self, bilgiler):
        """Adres kontrolü"""
        adres = bilgiler.get('adres', '')
        telefon = bilgiler.get('telefon', '')
        
        if len(adres) < 10:
            return False, "Adres çok kısa!"
        
        if len(telefon) < 10:
            return False, "Geçersiz telefon!"
        
        return True, "Bilgiler geçerli"
    
    def odeme_al(self, tutar, musteri_bilgi):
        """Kapıda ödeme"""
        print(f"\n{'='*60}")
        print(f"💵 KAPIDA ÖDEME")
        print(f"{'='*60}")
        
        gecerli, mesaj = self.dogrulama_yap(musteri_bilgi)
        if not gecerli:
            print(f"❌ {mesaj}")
            return False
        
        self.odeme_raporu(tutar)
        
        print(f"\n📍 TESLİMAT ADRESİ:")
        print(f"   {musteri_bilgi.get('adres', 'N/A')}")
        print(f"   Tel: {musteri_bilgi.get('telefon', 'N/A')}")
        
        print(f"\n💡 Kargonuz teslim edilirken ödeme yapabilirsiniz:")
        print(f"   • Nakit")
        print(f"   • Kurye POS cihazı ile kart")
        
        print(f"\n✅ Sipariş onaylandı! Kargo hazırlanıyor...")
        return True
    
    def iade_yap(self, tutar, islem_id):
        """İade işlemi"""
        print(f"\n🔄 İADE İŞLEMİ")
        print(f"   İşlem No: {islem_id}")
        print(f"   Tutar: {tutar:,.2f} TL")
        print(f"   ✅ İade nakit veya hesaba transfer şeklinde yapılacaktır.")
        return True

print("✅ Tüm ödeme yöntemleri oluşturuldu!")
````

````python
# 💻 POLYMORPHİC ÖDEME İŞLEYİCİ

def odeme_isle(odeme_yontemi: OdemeYontemi, tutar, musteri_bilgi):
    """
    🔥 POLYMORPHİC FONKSİYON
    
    Herhangi bir OdemeYontemi alt sınıfını kabul eder.
    Abstraction sayesinde hangi ödeme yöntemi olduğunu bilmesi gerekmez!
    """
    print(f"\n🔄 Ödeme işlemi başlatılıyor...")
    
    # Type hint kontrol (opsiyonel)
    if not isinstance(odeme_yontemi, OdemeYontemi):
        print("❌ Geçersiz ödeme yöntemi!")
        return False
    
    # Ödeme al - hangi tip olduğunu bilmiyoruz, önemli değil!
    basarili = odeme_yontemi.odeme_al(tutar, musteri_bilgi)
    
    if basarili:
        print(f"\n🎉 Ödeme tamamlandı!")
        return True
    else:
        print(f"\n❌ Ödeme başarısız!")
        return False


def toplu_odeme_karsilastir(tutar):
    """Tüm ödeme yöntemlerini karşılaştır"""
    print(f"\n{'='*70}")
    print(f"📊 ÖDEME YÖNTEMLERİ KARŞILAŞTIRMASI")
    print(f"{'='*70}")
    print(f"\n💰 Tutar: {tutar:,.2f} TL\n")
    
    yontemler = [
        KrediKartiOdeme(),
        HavaleOdeme(),
        KapidaOdeme()
    ]
    
    print(f"{'Yöntem':<20} | {'İşlem Ücreti':<15} | {'Toplam Ödeme':<15}")
    print("-"*70)
    
    for yontem in yontemler:
        ucret = yontem.islem_ucreti_hesapla(tutar)
        toplam = yontem.toplam_tutar_hesapla(tutar)
        
        print(f"{yontem.isim:<20} | {ucret:>13,.2f} TL | {toplam:>13,.2f} TL")
    
    print("="*70)


# TEST: Polymorphic fonksiyon
print("🧪 POLYMORPHİC FONKSİYON TESTİ")
print("="*70)

# Farklı ödeme yöntemleri
kk = KrediKartiOdeme()
havale = HavaleOdeme()
kapida = KapidaOdeme()

# Aynı fonksiyon, farklı tipler!
odeme_isle(kk, 1000, {
    'kart_no': '1234567890123456',
    'cvv': '123'
})

odeme_isle(havale, 1000, {
    'iban': 'TR330006100519786457841326',
    'siparis_no': 'SIP001'
})

odeme_isle(kapida, 1500, {
    'adres': 'Atatürk Cad. No:100 Çankaya/Ankara',
    'telefon': '0555-999-8877'
})

# Karşılaştırma
toplu_odeme_karsilastir(10000)
````

## 🎬 **DEMO 2: Matematiksel Operatörler**

Kendi objelerimizi toplayalım, çıkaralım!

````python
# 🛒 E-TİCARET SİPARİŞ SİSTEMİ

class Siparis:
    """
    🛍️ Sipariş sınıfı - Matematiksel operatörlerle!
    """
    
    def __init__(self, siparis_no, tutar, urun_adedi):
        self.siparis_no = siparis_no
        self.tutar = tutar
        self.urun_adedi = urun_adedi
    
    def __str__(self):
        return f"📦 Sipariş #{self.siparis_no}: {self.tutar:,.2f} TL ({self.urun_adedi} ürün)"
    
    def __repr__(self):
        return f"Siparis(no={self.siparis_no}, tutar={self.tutar}, adet={self.urun_adedi})"
    
    # ➕ TOPLAMA OPERATÖRÜ
    def __add__(self, other):
        """
        🔥 İki siparişi birleştir!
        siparis1 + siparis2 çağrıldığında bu çalışır
        """
        if not isinstance(other, Siparis):
            raise TypeError("Sadece Siparis nesneleri toplanabilir!")
        
        # Yeni sipariş oluştur
        yeni_no = f"{self.siparis_no}+{other.siparis_no}"
        yeni_tutar = self.tutar + other.tutar
        yeni_adet = self.urun_adedi + other.urun_adedi
        
        return Siparis(yeni_no, yeni_tutar, yeni_adet)
    
    # ➖ ÇIKARMA OPERATÖRÜ
    def __sub__(self, other):
        """
        🔥 İptal/İade işlemi!
        siparis1 - siparis2
        """
        if not isinstance(other, Siparis):
            raise TypeError("Sadece Siparis nesneleri çıkarılabilir!")
        
        yeni_no = f"{self.siparis_no}-{other.siparis_no}"
        yeni_tutar = max(0, self.tutar - other.tutar)  # Negatif olmasın
        yeni_adet = max(0, self.urun_adedi - other.urun_adedi)
        
        return Siparis(yeni_no, yeni_tutar, yeni_adet)
    
    # ✖️ ÇARPMA OPERATÖRÜ (İndirim için!)
    def __mul__(self, sayi):
        """
        🔥 İndirim uygula!
        siparis * 0.8  # %20 indirim
        """
        if not isinstance(sayi, (int, float)):
            raise TypeError("Sadece sayılarla çarpılabilir!")
        
        yeni_tutar = self.tutar * sayi
        return Siparis(f"{self.siparis_no}*{sayi}", yeni_tutar, self.urun_adedi)


# TEST: Matematiksel operatörler
print("="*70)
print("🧪 MATEMATİKSEL OPERATÖRLER TESTİ")
print("="*70)

# Siparişler
s1 = Siparis("S001", 1500, 3)
s2 = Siparis("S002", 2500, 5)
s3 = Siparis("S003", 800, 2)

print("\n📦 Siparişlerimiz:")
print(s1)
print(s2)
print(s3)

# TOPLAMA
print("\n➕ TOPLAMA: s1 + s2")
birlesik = s1 + s2
print(birlesik)

# ÇIKARMA (İade)
print("\n➖ ÇIKARMA (İade): s2 - s3")
iade_sonrasi = s2 - s3
print(iade_sonrasi)

# ÇARPMA (İndirim)
print("\n✖️ ÇARPMA (İndirim): s1 * 0.8")
indirimli = s1 * 0.8  # %20 indirim
print(f"Orijinal: {s1}")
print(f"İndirimli: {indirimli}")

# ZİNCİRLEME İŞLEMLER!
print("\n🔗 ZİNCİRLEME: (s1 + s2 + s3) * 0.9")
toplam_indirimli = (s1 + s2 + s3) * 0.9  # %10 toplu indirim
print(toplam_indirimli)

print("\n" + "="*70)
````

## 🎬 **DEMO 3: Karşılaştırma Operatörleri**

Müşterileri skorlarına göre karşılaştıralım!

````python
# 🎯 CRM - MÜŞTERİ SKORLAMA SİSTEMİ

class MusteriCRM:
    """
    👤 CRM Müşterisi - Karşılaştırma operatörleriyle!
    """
    
    def __init__(self, ad, alisveris_sayisi, toplam_harcama, uyelik_yili):
        self.ad = ad
        self.alisveris_sayisi = alisveris_sayisi
        self.toplam_harcama = toplam_harcama
        self.uyelik_yili = uyelik_yili
        self.skor = self._skor_hesapla()
    
    def _skor_hesapla(self):
        """📊 Müşteri skoru hesapla"""
        # Basit skorlama algoritması
        skor = 0
        skor += self.alisveris_sayisi * 10  # Her alışveriş 10 puan
        skor += (self.toplam_harcama / 100) * 5  # Her 100 TL = 5 puan
        skor += (2026 - self.uyelik_yili) * 20  # Her yıl üyelik = 20 puan
        return int(skor)
    
    def __str__(self):
        return f"👤 {self.ad:<10} | Skor: {self.skor:>4} | Harcama: {self.toplam_harcama:>8,.0f} TL"
    
    def __repr__(self):
        return f"MusteriCRM('{self.ad}', skor={self.skor})"
    
    # ⚖️ EŞİTLİK OPERATÖRÜ
    def __eq__(self, other):
        """
        🔥 İki müşteri eşit mi?
        musteri1 == musteri2
        """
        if not isinstance(other, MusteriCRM):
            return False
        return self.skor == other.skor
    
    # 📊 KÜÇÜKTÜR OPERATÖRÜ
    def __lt__(self, other):
        """
        🔥 Hangi müşteri daha düşük skorlu?
        musteri1 < musteri2
        """
        if not isinstance(other, MusteriCRM):
            return NotImplemented
        return self.skor < other.skor
    
    # 📊 KÜÇÜK EŞİT OPERATÖRÜ
    def __le__(self, other):
        """musteri1 <= musteri2"""
        return self.skor <= other.skor
    
    # 📊 BÜYÜKTÜR OPERATÖRÜ
    def __gt__(self, other):
        """
        🔥 Hangi müşteri daha yüksek skorlu?
        musteri1 > musteri2
        """
        return self.skor > other.skor
    
    # 📊 BÜYÜK EŞİT OPERATÖRÜ
    def __ge__(self, other):
        """musteri1 >= musteri2"""
        return self.skor >= other.skor
    
    # ❌ EŞİT DEĞİL OPERATÖRÜ
    def __ne__(self, other):
        """musteri1 != musteri2"""
        return not self.__eq__(other)


# TEST: Karşılaştırma operatörleri
print("="*70)
print("🧪 KARŞILAŞTIRMA OPERATÖRLERI TESTİ")
print("="*70)

# Müşteriler
m1 = MusteriCRM("Ahmet", alisveris_sayisi=25, toplam_harcama=15000, uyelik_yili=2020)
m2 = MusteriCRM("Ayşe", alisveris_sayisi=50, toplam_harcama=35000, uyelik_yili=2018)
m3 = MusteriCRM("Mehmet", alisveris_sayisi=15, toplam_harcama=8000, uyelik_yili=2023)
m4 = MusteriCRM("Zeynep", alisveris_sayisi=30, toplam_harcama=20000, uyelik_yili=2021)

musteriler = [m1, m2, m3, m4]

print("\n👥 Müşterilerimiz:")
for m in musteriler:
    print(m)

# KARŞILAŞTIRMALAR
print("\n" + "="*70)
print("⚖️ KARŞILAŞTIRMALAR")
print("="*70)

print(f"\n📊 Ahmet == Mehmet? {m1 == m3}")
print(f"📊 Ahmet < Ayşe? {m1 < m2}")
print(f"📊 Mehmet > Zeynep? {m3 > m4}")
print(f"📊 Ayşe >= Zeynep? {m2 >= m4}")

# SIRALAMA - Python'un sorted() fonksiyonu __lt__ kullanır!
print("\n" + "="*70)
print("🏆 MÜŞTERİ SIRALAMA (Skora Göre)")
print("="*70)

sirali_musteriler = sorted(musteriler, reverse=True)  # Büyükten küçüğe

print("\n🥇 EN YÜKSEK SKORDAN EN DÜŞÜĞE:")
for i, m in enumerate(sirali_musteriler, 1):
    print(f"{i}. {m}")

# VIP MÜŞTERİ SEÇİMİ
print("\n" + "="*70)
print("💎 VIP MÜŞTERİ SEÇİMİ (Skor > 1000)")
print("="*70)

vip_musteriler = [m for m in musteriler if m.skor > 1000]
print(f"\n✨ {len(vip_musteriler)} VIP müşteri bulundu:\n")
for m in vip_musteriler:
    print(m)

print("\n" + "="*70)
````

## 🎬 **DEMO 4: Container Methods**

`__len__`, `__getitem__`, `__contains__` ile sepet oluşturalım!

````python
# 🛒 ALIŞVERİŞ SEPETİ SİSTEMİ

class SepetItem:
    """Sepetteki bir ürün"""
    def __init__(self, ad, fiyat, adet):
        self.ad = ad
        self.fiyat = fiyat
        self.adet = adet
    
    def toplam(self):
        return self.fiyat * self.adet
    
    def __str__(self):
        return f"{self.ad:<20} x{self.adet} = {self.toplam():>8,.2f} TL"


class AlisverisSepeti:
    """
    🛒 Alışveriş Sepeti - Container magic methodlarıyla!
    """
    
    def __init__(self):
        self.items = []  # İç liste
    
    def ekle(self, ad, fiyat, adet=1):
        """Sepete ürün ekle"""
        item = SepetItem(ad, fiyat, adet)
        self.items.append(item)
        print(f"✅ {ad} sepete eklendi!")
    
    # 📏 UZUNLUK OPERATÖRÜ
    def __len__(self):
        """
        🔥 Sepetteki ürün sayısı
        len(sepet) çağrıldığında bu çalışır
        """
        return len(self.items)
    
    # 🔍 İNDEKSLEME OPERATÖRÜ
    def __getitem__(self, index):
        """
        🔥 İndeksle erişim
        sepet[0], sepet[1], sepet[-1] gibi
        """
        return self.items[index]
    
    # 📝 ATAMA OPERATÖRÜ
    def __setitem__(self, index, item):
        """
        🔥 İndeksle atama
        sepet[0] = yeni_item
        """
        if not isinstance(item, SepetItem):
            raise TypeError("Sadece SepetItem eklenebilir!")
        self.items[index] = item
    
    # 🔎 ÜYELİK OPERATÖRÜ
    def __contains__(self, urun_adi):
        """
        🔥 Ürün sepette mi?
        "Laptop" in sepet çağrıldığında bu çalışır
        """
        return any(item.ad == urun_adi for item in self.items)
    
    # 🔄 İTERATOR OPERATÖRÜ
    def __iter__(self):
        """
        🔥 for döngüsünde kullanım
        for item in sepet:
        """
        return iter(self.items)
    
    def toplam_tutar(self):
        """Sepet toplamı"""
        return sum(item.toplam() for item in self.items)
    
    def __str__(self):
        if not self.items:
            return "🛒 Sepet boş!"
        
        output = "\n" + "="*60 + "\n"
        output += "🛒 ALIŞVERİŞ SEPETİ\n"
        output += "="*60 + "\n"
        
        for i, item in enumerate(self.items, 1):
            output += f"{i}. {item}\n"
        
        output += "-"*60 + "\n"
        output += f"{'TOPLAM:':<22} {self.toplam_tutar():>8,.2f} TL\n"
        output += "="*60
        
        return output


# TEST: Container methods
print("="*70)
print("🧪 CONTAINER METHODS TESTİ")
print("="*70)

# Sepet oluştur
sepet = AlisverisSepeti()

# Ürün ekle
print("\n📦 ÜRÜN EKLEME:")
sepet.ekle("Laptop", 25000, 1)
sepet.ekle("Mouse", 450, 2)
sepet.ekle("Klavye", 1200, 1)
sepet.ekle("Monitör", 8500, 2)

# LEN() - Uzunluk
print(f"\n📏 len(sepet) = {len(sepet)} ürün")

# INDEKSLEME - []
print(f"\n🔍 sepet[0] (İlk ürün):")
print(f"   {sepet[0]}")

print(f"\n🔍 sepet[-1] (Son ürün):")
print(f"   {sepet[-1]}")

# CONTAINS - in operatörü
print(f"\n🔎 ÜYELİK KONTROLÜ:")
print(f"   'Laptop' in sepet? {('Laptop' in sepet)}")
print(f"   'Tablet' in sepet? {('Tablet' in sepet)}")

# ITERATION - for döngüsü
print(f"\n🔄 FOR DÖNGÜSÜ:")
print("   Sepetteki tüm ürünler:")
for item in sepet:
    print(f"   - {item.ad}: {item.fiyat:,.2f} TL")

# SLICE - Dilimleme
print(f"\n✂️ DİLİMLEME:")
print("   sepet[1:3] (2. ve 3. ürünler):")
for item in sepet[1:3]:
    print(f"   - {item}")

# SEPET GÖRÜNÜMÜ
print(sepet)

print("\n" + "="*70)
````

## 💼 **YBS Uygulaması: Kapsamlı Sipariş Yönetim Sistemi**

Tüm magic methodları bir arada kullanan gerçek dünya örneği!

````python
# 🏢 KAPSAMLI SİPARİŞ YÖNETİM SİSTEMİ - PANDAS ENTEGRASYONU

class SiparisYonetim:
    """
    📦 Sipariş Yönetim Sistemi
    
    Tüm Magic Methodları kullanan gerçek dünya uygulaması!
    """
    
    def __init__(self, firma_adi):
        self.firma_adi = firma_adi
        self.siparisler = []
        self.toplam_ciro = 0
        self.olusturma_tarihi = pd.Timestamp.now()
    
    # 📊 STRING REPRESENTATION
    def __str__(self):
        return (f"🏢 {self.firma_adi} Sipariş Sistemi\n"
                f"   📦 Toplam Sipariş: {len(self.siparisler)}\n"
                f"   💰 Toplam Ciro: {self.toplam_ciro:,.2f} TL")
    
    def __repr__(self):
        return f"SiparisYonetim('{self.firma_adi}', {len(self.siparisler)} sipariş)"
    
    # 📏 UZUNLUK
    def __len__(self):
        """Toplam sipariş sayısı"""
        return len(self.siparisler)
    
    # 🔍 İNDEKSLEME
    def __getitem__(self, index):
        """sistem[0] ile sipariş erişimi"""
        return self.siparisler[index]
    
    # ➕ TOPLAMA (Sistem birleştirme)
    def __add__(self, other):
        """İki sistemi birleştir"""
        if not isinstance(other, SiparisYonetim):
            raise TypeError("Sadece SiparisYonetim nesneleri toplanabilir!")
        
        yeni_sistem = SiparisYonetim(f"{self.firma_adi}+{other.firma_adi}")
        yeni_sistem.siparisler = self.siparisler + other.siparisler
        yeni_sistem.toplam_ciro = self.toplam_ciro + other.toplam_ciro
        return yeni_sistem
    
    # ⚖️ KARŞILAŞTIRMA
    def __gt__(self, other):
        """Hangi sistem daha büyük ciro?"""
        return self.toplam_ciro > other.toplam_ciro
    
    def __lt__(self, other):
        return self.toplam_ciro < other.toplam_ciro
    
    def __eq__(self, other):
        return self.toplam_ciro == other.toplam_ciro
    
    # 🔄 İTERASYON
    def __iter__(self):
        """for siparis in sistem:"""
        return iter(self.siparisler)
    
    # 🔎 ÜYELİK
    def __contains__(self, siparis_no):
        """siparis_no in sistem"""
        return any(s['siparis_no'] == siparis_no for s in self.siparisler)
    
    # İŞ LOJİĞİ METODLARI
    def siparis_ekle(self, siparis_no, musteri, tutar, urunler):
        """Yeni sipariş ekle"""
        siparis = {
            'siparis_no': siparis_no,
            'musteri': musteri,
            'tutar': tutar,
            'urunler': urunler,
            'tarih': pd.Timestamp.now(),
            'durum': 'Onaylandı'
        }
        self.siparisler.append(siparis)
        self.toplam_ciro += tutar
        return f"✅ Sipariş {siparis_no} eklendi!"
    
    def to_dataframe(self):
        """Pandas DataFrame'e dönüştür"""
        if not self.siparisler:
            return pd.DataFrame()
        
        df_data = []
        for s in self.siparisler:
            df_data.append({
                'Sipariş No': s['siparis_no'],
                'Müşteri': s['musteri'],
                'Tutar': s['tutar'],
                'Ürün Sayısı': len(s['urunler']),
                'Tarih': s['tarih'],
                'Durum': s['durum']
            })
        
        return pd.DataFrame(df_data)
    
    def rapor(self):
        """Detaylı rapor"""
        if not self.siparisler:
            return "📭 Henüz sipariş yok!"
        
        df = self.to_dataframe()
        
        output = f"\n{'='*80}\n"
        output += f"📊 {self.firma_adi} - SİPARİŞ RAPORU\n"
        output += f"{'='*80}\n\n"
        
        output += f"📅 Rapor Tarihi: {pd.Timestamp.now().strftime('%d.%m.%Y %H:%M')}\n"
        output += f"📦 Toplam Sipariş: {len(self.siparisler)}\n"
        output += f"💰 Toplam Ciro: {self.toplam_ciro:,.2f} TL\n"
        output += f"📈 Ortalama Sipariş Değeri: {df['Tutar'].mean():,.2f} TL\n"
        output += f"🔝 En Yüksek Sipariş: {df['Tutar'].max():,.2f} TL\n"
        output += f"🔽 En Düşük Sipariş: {df['Tutar'].min():,.2f} TL\n"
        
        output += f"\n{'-'*80}\n"
        output += "📋 SİPARİŞ DETAYLARI:\n"
        output += f"{'-'*80}\n"
        
        for i, row in df.iterrows():
            output += f"{i+1}. {row['Sipariş No']:<10} | {row['Müşteri']:<15} | "
            output += f"{row['Tutar']:>10,.2f} TL | {row['Durum']}\n"
        
        output += f"{'='*80}\n"
        
        return output


# 🎬 DEMO: Kapsamlı Sistem
print("="*80)
print("🎬 KAPSAMLI SİPARİŞ YÖNETİM SİSTEMİ DEMOsu")
print("="*80)

# İki farklı şube
sistem_istanbul = SiparisYonetim("TechShop İstanbul")
sistem_ankara = SiparisYonetim("TechShop Ankara")

# İstanbul siparişleri
print("\n📦 İSTANBUL ŞUBESİ - Sipariş Ekleme:")
sistem_istanbul.siparis_ekle("IST001", "Ahmet Yılmaz", 15000, ["Laptop", "Mouse"])
sistem_istanbul.siparis_ekle("IST002", "Ayşe Demir", 8500, ["Tablet"])
sistem_istanbul.siparis_ekle("IST003", "Mehmet Kaya", 25000, ["Laptop", "Monitör", "Klavye"])

# Ankara siparişleri
print("\n📦 ANKARA ŞUBESİ - Sipariş Ekleme:")
sistem_ankara.siparis_ekle("ANK001", "Zeynep Arslan", 12000, ["Laptop"])
sistem_ankara.siparis_ekle("ANK002", "Ali Çelik", 3500, ["Mouse", "Klavye", "Kulakık"])

# __STR__ kullanımı
print("\n" + "="*80)
print("📊 SİSTEM DURUMLARI (__str__):")
print("="*80)
print(sistem_istanbul)
print()
print(sistem_ankara)

# __LEN__ kullanımı
print("\n" + "="*80)
print("📏 UZUNLUK (__len__):")
print("="*80)
print(f"İstanbul sipariş sayısı: len(sistem_istanbul) = {len(sistem_istanbul)}")
print(f"Ankara sipariş sayısı: len(sistem_ankara) = {len(sistem_ankara)}")

# __GETITEM__ kullanımı
print("\n" + "="*80)
print("🔍 İNDEKSLEME (__getitem__):")
print("="*80)
print(f"İstanbul'un ilk siparişi: sistem_istanbul[0]")
print(f"   Sipariş No: {sistem_istanbul[0]['siparis_no']}")
print(f"   Müşteri: {sistem_istanbul[0]['musteri']}")
print(f"   Tutar: {sistem_istanbul[0]['tutar']:,.2f} TL")

# __CONTAINS__ kullanımı
print("\n" + "="*80)
print("🔎 ÜYELİK KONTROLÜ (__contains__):")
print("="*80)
print(f"'IST001' in sistem_istanbul? {'IST001' in sistem_istanbul}")
print(f"'ANK999' in sistem_ankara? {'ANK999' in sistem_ankara}")

# __GT__ / __LT__ kullanımı (Karşılaştırma)
print("\n" + "="*80)
print("⚖️ KARŞILAŞTIRMA (__gt__, __lt__):")
print("="*80)
print(f"İstanbul > Ankara? {sistem_istanbul > sistem_ankara}")
print(f"İstanbul Ciro: {sistem_istanbul.toplam_ciro:,.2f} TL")
print(f"Ankara Ciro: {sistem_ankara.toplam_ciro:,.2f} TL")

# __ADD__ kullanımı (Birleştirme)
print("\n" + "="*80)
print("➕ BİRLEŞTİRME (__add__):")
print("="*80)
print("sistem_genel = sistem_istanbul + sistem_ankara")
sistem_genel = sistem_istanbul + sistem_ankara
print(sistem_genel)

# __ITER__ kullanımı
print("\n" + "="*80)
print("🔄 İTERASYON (__iter__):")
print("="*80)
print("for siparis in sistem_genel (İlk 3 sipariş):")
for i, siparis in enumerate(sistem_genel, 1):
    if i > 3:
        break
    print(f"   {i}. {siparis['siparis_no']}: {siparis['musteri']} - {siparis['tutar']:,.2f} TL")

# PANDAS ENTEGRASYONU
print("\n" + "="*80)
print("🐼 PANDAS DATAFRAME DÖNÜŞÜMÜ:")
print("="*80)
df_genel = sistem_genel.to_dataframe()
print(df_genel.to_string(index=False))

# DETAYLI RAPOR
print(sistem_genel.rapor())

print("="*80)
print("✅ Magic Methods ile güçlü bir sistem oluşturduk!")
print("="*80)
````

## 🎴 **BÖLÜM 9 FLASHCARDS**

### 📇 **KART 1: Magic Methods Nedir?**

**Soru:** Special Methods (Magic Methods) nedir ve nasıl tanımlanır?

**Cevap:**
- Çift alt çizgi ile başlayan ve biten özel metodlardır (`__xxx__`)
- "Dunder" (Double UNDERscore) metodlar olarak da bilinir
- Python'da operatörleri ve built-in fonksiyonları özelleştirmek için kullanılır
- Örnek: `__init__`, `__str__`, `__add__`, `__len__`

**YBS Örnek:**
```python
class Musteri:
    def __str__(self):
        return f"👤 {self.ad}"
    
    def __add__(self, other):
        # İki müşteri puanını topla
        return self.puan + other.puan
```

---

### 📇 **KART 2: __str__ vs __repr__**

**Soru:** `__str__` ve `__repr__` arasındaki fark nedir?

**Cevap:**
- **`__str__`**: Human-readable (kullanıcı dostu)
  - `print()`, `str()` tarafından kullanılır
  - Son kullanıcılar için okunabilir format
  
- **`__repr__`**: Developer-friendly (geliştirici dostu)
  - REPL'de ve `repr()` tarafından kullanılır
  - Nesneyi yeniden oluşturmak için bilgi verir
  - Mümkünse valid Python kodu olmalı

**Örnek:**
```python
def __str__(self):
    return f"Sipariş #{self.no}: {self.tutar} TL"

def __repr__(self):
    return f"Siparis(no={self.no}, tutar={self.tutar})"
```

---

### 📇 **KART 3: Matematiksel Operatörler**

**Soru:** Matematiksel operatörler için hangi magic methodlar kullanılır?

**Cevap:**
- `__add__(self, other)` → `+` operatörü
- `__sub__(self, other)` → `-` operatörü
- `__mul__(self, other)` → `*` operatörü
- `__truediv__(self, other)` → `/` operatörü
- `__floordiv__(self, other)` → `//` operatörü
- `__mod__(self, other)` → `%` operatörü
- `__pow__(self, other)` → `**` operatörü

**YBS Kullanımı:**
```python
# Sipariş toplama
siparis_toplam = siparis1 + siparis2

# İndirim uygulama
indirimli = siparis * 0.8  # %20 indirim
```

### 📇 **KART 4: Karşılaştırma Operatörleri**

**Soru:** Karşılaştırma operatörleri için hangi magic methodlar vardır?

**Cevap:**
- `__eq__(self, other)` → `==` (eşit mi?)
- `__ne__(self, other)` → `!=` (eşit değil mi?)
- `__lt__(self, other)` → `<` (küçük mü?)
- `__le__(self, other)` → `<=` (küçük eşit mi?)
- `__gt__(self, other)` → `>` (büyük mü?)
- `__ge__(self, other)` → `>=` (büyük eşit mi?)

**Kullanım:**
```python
# Müşteri skorlarına göre sıralama
sirali = sorted(musteriler)  # __lt__ kullanır

# Karşılaştırma

    print("Müşteri1 daha yüksek skorlu!")
```

### 📇 **KART 5: Container Methods**

**Soru:** Container davranışları için hangi magic methodlar kullanılır?

**Cevap:**
- `__len__(self)` → `len(obj)` - Uzunluk
- `__getitem__(self, key)` → `obj[key]` - İndeksleme
- `__setitem__(self, key, value)` → `obj[key] = val` - Atama
- `__delitem__(self, key)` → `del obj[key]` - Silme
- `__contains__(self, item)` → `item in obj` - Üyelik
- `__iter__(self)` → `for x in obj` - İterasyon

**Sepet Örneği:**
```python
len(sepet)           # __len__
sepet[0]             # __getitem__
"Laptop" in sepet    # __contains__
for item in sepet:   # __iter__
```

### 📇 **KART 8: Common Pitfalls (Yaygın Hatalar)**

**Soru:** Magic methodları kullanırken dikkat edilmesi gereken noktalar nelerdir?

**Cevap:**

**❌ Yapılmaması Gerekenler:**
```python
# 1. Type kontrolü yapmamak
def __add__(self, other):
    return self.value + other.value  # ❌ TypeError riski!

# 2. None döndürmek yerine NotImplemented
def __eq__(self, other):
    if not isinstance(other, MyClass):
        return None  # ❌ Yanlış! False veya NotImplemented olmalı

# 3. __repr__ ile __str__'i karıştırmak
def __repr__(self):
    return "Güzel görünüm"  # ❌ Valid Python kodu değil!
```

**✅ Doğru Yaklaşım:**
```python
def __add__(self, other):
    if not isinstance(other, MyClass):
        return NotImplemented  # ✅ Doğru!
    return MyClass(self.value + other.value)

def __eq__(self, other):
    if not isinstance(other, MyClass):
        return NotImplemented  # ✅ Doğru!
    return self.value == other.value

def __repr__(self):
    return f"MyClass(value={self.value})"  # ✅ Valid Python!
```

**Önemli Kurallar:**
1. **Type kontrolü yapın**: `isinstance()` kullanın
2. **NotImplemented döndürün**: Desteklenmeyen tipler için
3. **Immutability**: `__add__` yeni nesne döndürmeli
4. **Consistency**: `__eq__` varsa `__hash__` da olmalı (set/dict için)

### 🎯 **Tip 8: Descriptor Protocol**

class PositiveNumber:

## 📦 **ADIM 1: Ürün Sistemi**

Inheritance ve Polymorphism ile ürün hiyerarşisi!

````python
# 📦 ÜRÜN SİSTEMİ - Inheritance & Polymorphism

from abc import ABC, abstractmethod
from datetime import datetime, timedelta
from typing import Optional, List, Dict
import pandas as pd
import numpy as np

class UrunBase(ABC):
    """
    🎯 Abstract Base Class: Tüm ürünlerin temel sınıfı
    """
    
    _id_counter = 1000  # Class variable - otomatik ID
    
    def __init__(self, ad: str, fiyat: float, stok: int, kategori: str):
        self.urun_id = UrunBase._id_counter
        UrunBase._id_counter += 1
        
        self.ad = ad
        self._fiyat = fiyat  # Encapsulation!
        self.stok = stok
        self.kategori = kategori
        self.olusturma_tarihi = datetime.now()
        self.satis_sayisi = 0
    
    # 🔐 ENCAPSULATION - Fiyat Property
    @property
    def fiyat(self):
        return self._fiyat
    
    @fiyat.setter
    def fiyat(self, yeni_fiyat):
        if yeni_fiyat < 0:
            raise ValueError("Fiyat negatif olamaz!")
        self._fiyat = yeni_fiyat
    
    # 🎭 ABSTRACTION - Zorunlu metodlar
    @abstractmethod
    def kargo_ucreti_hesapla(self) -> float:
        """Her ürün tipinin kargo ücreti farklı"""
        pass
    
    @abstractmethod
    def iade_kosullari(self) -> str:
        """Her ürün tipinin iade koşulları farklı"""
        pass
    
    # 🔮 MAGIC METHODS
    def __str__(self):
        return f"📦 {self.ad} - {self.fiyat:,.2f} TL (Stok: {self.stok})"
    
    def __repr__(self):
        return f"{self.__class__.__name__}(id={self.urun_id}, ad='{self.ad}', fiyat={self.fiyat})"
    
    def __eq__(self, other):
        """Ürün eşitliği - ID'ye göre"""
        if not isinstance(other, UrunBase):
            return False
        return self.urun_id == other.urun_id
    
    def __lt__(self, other):
        """Sıralama - Fiyata göre"""
        if not isinstance(other, UrunBase):
            return NotImplemented
        return self.fiyat < other.fiyat
    
    # İŞ LOJİĞİ
    def stok_guncelle(self, miktar: int):
        """Stok güncelle (+ ekleme, - azaltma)"""
        yeni_stok = self.stok + miktar
        if yeni_stok < 0:
            raise ValueError("Yetersiz stok!")
        self.stok = yeni_stok
    
    def satis_yap(self, adet: int = 1):
        """Satış gerçekleştir"""
        self.stok_guncelle(-adet)
        self.satis_sayisi += adet
    
    def toplam_deger(self) -> float:
        """Stok değeri"""
        return self.fiyat * self.stok


class FizikselUrun(UrunBase):
    """
    📦 Fiziksel Ürün - Kargo gerektirir
    """
    
    def __init__(self, ad: str, fiyat: float, stok: int, kategori: str, 
                 agirlik: float, boyutlar: tuple):
        super().__init__(ad, fiyat, stok, kategori)
        self.agirlik = agirlik  # kg
        self.boyutlar = boyutlar  # (en, boy, yükseklik) cm
    
    def kargo_ucreti_hesapla(self) -> float:
        """Ağırlık ve boyuta göre kargo"""
        temel_ucret = 25.0
        
        # Ağırlık faktörü
        if self.agirlik > 10:
            temel_ucret += (self.agirlik - 10) * 5
        
        # Hacim faktörü
        hacim = self.boyutlar[0] * self.boyutlar[1] * self.boyutlar[2]
        if hacim > 100000:  # 100000 cm³
            temel_ucret += 50
        
        return temel_ucret
    
    def iade_kosullari(self) -> str:
        return "14 gün içinde iade edilebilir. Ürün kullanılmamış olmalıdır."


class DijitalUrun(UrunBase):
    """
    💾 Dijital Ürün - Kargo yok, anında teslimat
    """
    
    def __init__(self, ad: str, fiyat: float, stok: int, kategori: str,
                 dosya_boyutu: float, indirme_linki: str):
        super().__init__(ad, fiyat, stok, kategori)
        self.dosya_boyutu = dosya_boyutu  # MB
        self.indirme_linki = indirme_linki
        self.indirme_sayisi = 0
    
    def kargo_ucreti_hesapla(self) -> float:
        """Dijital ürünlerde kargo yok"""
        return 0.0
    
    def iade_kosullari(self) -> str:
        return "Dijital ürünlerde iade yapılamaz."
    
    def indir(self):
        """İndirme gerçekleştir"""
        self.indirme_sayisi += 1
        return f"İndirme başlatıldı: {self.indirme_linki}"


# TEST: Ürün sistemi
print("="*80)
print("🧪 ÜRÜN SİSTEMİ TESTİ")
print("="*80)

# Fiziksel ürünler
laptop = FizikselUrun(
    ad="Gaming Laptop XPro",
    fiyat=45000,
    stok=15,
    kategori="Elektronik",
    agirlik=2.5,
    boyutlar=(40, 30, 5)
)

kitap = FizikselUrun(
    ad="Python ile Veri Bilimi",
    fiyat=250,
    stok=100,
    kategori="Kitap",
    agirlik=0.5,
    boyutlar=(20, 15, 3)
)

# Dijital ürün
kurs = DijitalUrun(
    ad="Python Masterclass Video Kursu",
    fiyat=499,
    stok=999,  # Sınırsız
    kategori="Eğitim",
    dosya_boyutu=15000,  # 15 GB
    indirme_linki="https://platform.com/download/python-masterclass"
)

urunler = [laptop, kitap, kurs]

print("\n📦 ÜRÜN LİSTESİ:")
for urun in urunler:
    print(f"\n{urun}")
    print(f"   ID: {urun.urun_id}")
    print(f"   Kategori: {urun.kategori}")
    print(f"   Kargo: {urun.kargo_ucreti_hesapla():,.2f} TL")
    print(f"   İade: {urun.iade_kosullari()}")
    print(f"   Stok Değeri: {urun.toplam_deger():,.2f} TL")

# POLYMORPHISM - Aynı metod, farklı davranışlar
print("\n" + "="*80)
print("🎭 POLYMORPHİSM DEMO")
print("="*80)

def urun_ozeti(urun: UrunBase):
    """Polymorphic fonksiyon - herhangi bir ürün tipini kabul eder"""
    print(f"\n📊 {urun.ad}")
    print(f"   Tip: {urun.__class__.__name__}")
    print(f"   Fiyat: {urun.fiyat:,.2f} TL")
    print(f"   Kargo: {urun.kargo_ucreti_hesapla():,.2f} TL")  # Polymorphism!
    print(f"   Toplam: {urun.fiyat + urun.kargo_ucreti_hesapla():,.2f} TL")

for urun in urunler:
    urun_ozeti(urun)

# MAGIC METHODS
print("\n" + "="*80)
print("🔮 MAGIC METHODS DEMO")
print("="*80)

# Sıralama (__lt__)
print("\n📊 Fiyata göre sıralı ürünler:")
sirali = sorted(urunler)
for i, urun in enumerate(sirali, 1):
    print(f"   {i}. {urun}")

# Karşılaştırma (__eq__)
print(f"\n⚖️ laptop == kitap? {laptop == kitap}")

print("\n" + "="*80)
````

## 👤 **ADIM 2: Müşteri Sistemi**

Inheritance, Polymorphism ve Encapsulation ile müşteri hiyerarşisi!

````python
# 👤 MÜŞTERİ SİSTEMİ - Inheritance, Polymorphism, Encapsulation

import hashlib

class MusteriBase(ABC):
    """
    👤 Abstract Base Class: Tüm müşterilerin temel sınıfı
    """
    
    _id_counter = 5000
    
    def __init__(self, ad: str, soyad: str, email: str, telefon: str):
        self.musteri_id = MusteriBase._id_counter
        MusteriBase._id_counter += 1
        
        self.ad = ad
        self.soyad = soyad
        self._email = email  # Encapsulation
        self._telefon = telefon  # Encapsulation
        self.kayit_tarihi = datetime.now()
        
        # İstatistikler
        self.toplam_harcama = 0.0
        self.siparis_sayisi = 0
        self.puan = 0
    
    # 🔐 ENCAPSULATION - Email Property
    @property
    def email(self):
        return self._email
    
    @email.setter
    def email(self, yeni_email):
        if "@" not in yeni_email:
            raise ValueError("Geçersiz email!")
        self._email = yeni_email
    
    # 🔐 ENCAPSULATION - Telefon Property
    @property
    def telefon(self):
        return self._telefon
    
    @telefon.setter
    def telefon(self, yeni_telefon):
        # Sadece rakamları al
        temiz_telefon = ''.join(filter(str.isdigit, yeni_telefon))
        if len(temiz_telefon) != 10:
            raise ValueError("Telefon 10 haneli olmalı!")
        self._telefon = temiz_telefon
    
    # 🎭 ABSTRACTION - Zorunlu metodlar
    @abstractmethod
    def indirim_hesapla(self, tutar: float) -> float:
        """Her müşteri tipinin indirimi farklı"""
        pass
    
    @abstractmethod
    def segment(self) -> str:
        """Müşteri segmenti"""
        pass
    
    # 🔮 MAGIC METHODS
    def __str__(self):
        return f"👤 {self.ad} {self.soyad} (#{self.musteri_id})"
    
    def __repr__(self):
        return f"{self.__class__.__name__}(id={self.musteri_id}, ad='{self.ad} {self.soyad}')"
    
    def __eq__(self, other):
        if not isinstance(other, MusteriBase):
            return False
        return self.musteri_id == other.musteri_id
    
    def __lt__(self, other):
        """Toplam harcamaya göre sırala"""
        if not isinstance(other, MusteriBase):
            return NotImplemented
        return self.toplam_harcama < other.toplam_harcama
    
    # İŞ LOJİĞİ
    def siparis_tamamla(self, tutar: float):
        """Sipariş tamamlandığında güncelle"""
        indirimli_tutar = self.indirim_hesapla(tutar)
        self.toplam_harcama += indirimli_tutar
        self.siparis_sayisi += 1
        
        # Puan kazan (Her 10 TL = 1 puan)
        self.puan += int(indirimli_tutar / 10)
        
        return indirimli_tutar
    
    def profil_bilgisi(self) -> dict:
        """Müşteri profil bilgisi"""
        return {
            'ID': self.musteri_id,
            'Ad Soyad': f"{self.ad} {self.soyad}",
            'Email': self.email,
            'Segment': self.segment(),
            'Toplam Harcama': self.toplam_harcama,
            'Sipariş Sayısı': self.siparis_sayisi,
            'Puan': self.puan,
            'Kayıt Tarihi': self.kayit_tarihi.strftime('%d.%m.%Y')
        }


class BireyselMusteri(MusteriBase):
    """
    🙋 Bireysel Müşteri - Standart müşteri
    """
    
    def __init__(self, ad: str, soyad: str, email: str, telefon: str, tc_no: str):
        super().__init__(ad, soyad, email, telefon)
        self._tc_no = self._hash_tc(tc_no)  # Güvenli saklama!
    
    def _hash_tc(self, tc: str) -> str:
        """TC'yi hash'le - güvenlik!"""
        return hashlib.sha256(tc.encode()).hexdigest()
    
    def indirim_hesapla(self, tutar: float) -> float:
        """Bireysel müşteri: %5 standart indirim"""
        indirim_orani = 0.05
        
        # Sadık müşteri bonusu
        if self.siparis_sayisi > 10:
            indirim_orani = 0.10  # %10
        
        indirim = tutar * indirim_orani
        return tutar - indirim
    
    def segment(self) -> str:
        if self.toplam_harcama > 10000:
            return "Altın Müşteri"
        elif self.toplam_harcama > 5000:
            return "Gümüş Müşteri"
        else:
            return "Bronz Müşteri"


class KurumsalMusteri(MusteriBase):
    """
    🏢 Kurumsal Müşteri - Şirketler için
    """
    
    def __init__(self, ad: str, soyad: str, email: str, telefon: str,
                 sirket_adi: str, vergi_no: str):
        super().__init__(ad, soyad, email, telefon)
        self.sirket_adi = sirket_adi
        self.vergi_no = vergi_no
        self.kredi_limiti = 50000  # 50.000 TL kredi
    
    def indirim_hesapla(self, tutar: float) -> float:
        """Kurumsal müşteri: Kademeli indirim"""
        if tutar > 10000:
            indirim_orani = 0.20  # %20
        elif tutar > 5000:
            indirim_orani = 0.15  # %15
        else:
            indirim_orani = 0.10  # %10
        
        indirim = tutar * indirim_orani
        return tutar - indirim
    
    def segment(self) -> str:
        return "Kurumsal"
    
    def __str__(self):
        return f"🏢 {self.sirket_adi} ({self.ad} {self.soyad}) (#{self.musteri_id})"


class VIPMusteri(BireyselMusteri):
    """
    💎 VIP Müşteri - Özel ayrıcalıklı müşteriler
    """
    
    def __init__(self, ad: str, soyad: str, email: str, telefon: str, tc_no: str):
        super().__init__(ad, soyad, email, telefon, tc_no)
        self.vip_level = 1
        self.ozel_temsilci = "Müşteri Hizmetleri"
    
    def indirim_hesapla(self, tutar: float) -> float:
        """VIP müşteri: %25 + Puan indirimi"""
        base_indirim = tutar * 0.25  # %25 sabit indirim
        
        # Puan indirimi (1 puan = 1 TL)
        puan_indirimi = min(self.puan, tutar * 0.10)  # Max %10
        
        toplam_indirim = base_indirim + puan_indirimi
        return tutar - toplam_indirim
    
    def segment(self) -> str:
        return f"VIP Level {self.vip_level}"
    
    def vip_seviye_atla(self):
        """VIP seviyesini yükselt"""
        if self.toplam_harcama > 50000:
            self.vip_level = 3
        elif self.toplam_harcama > 25000:
            self.vip_level = 2


# TEST: Müşteri sistemi
print("="*80)
print("🧪 MÜŞTERİ SİSTEMİ TESTİ")
print("="*80)

# Müşteri oluştur
bireysel = BireyselMusteri(
    ad="Ahmet",
    soyad="Yılmaz",
    email="ahmet.yilmaz@email.com",
    telefon="5551234567",
    tc_no="12345678901"
)

kurumsal = KurumsalMusteri(
    ad="Ayşe",
    soyad="Demir",
    email="ayse.demir@sirket.com",
    telefon="5559876543",
    sirket_adi="TechCorp A.Ş.",
    vergi_no="1234567890"
)

vip = VIPMusteri(
    ad="Mehmet",
    soyad="Kaya",
    email="mehmet.kaya@email.com",
    telefon="5556547890",
    tc_no="98765432109"
)

musteriler_liste = [bireysel, kurumsal, vip]

print("\n👥 MÜŞTERİ LİSTESİ:")
for m in musteriler_liste:
    print(f"\n{m}")
    print(f"   Email: {m.email}")
    print(f"   Segment: {m.segment()}")

# POLYMORPHISM - İndirim hesaplama
print("\n" + "="*80)
print("🎭 POLYMORPHİSM: İNDİRİM HESAPLAMA")
print("="*80)

test_tutar = 10000

for musteri in musteriler_liste:
    indirimli = musteri.indirim_hesapla(test_tutar)
    indirim = test_tutar - indirimli
    print(f"\n{musteri.ad} ({musteri.__class__.__name__}):")
    print(f"   Orijinal: {test_tutar:,.2f} TL")
    print(f"   İndirim: {indirim:,.2f} TL")
    print(f"   Ödenecek: {indirimli:,.2f} TL")

# SİPARİŞ SİMÜLASYONU
print("\n" + "="*80)
print("🛍️ SİPARİŞ SİMÜLASYONU")
print("="*80)

# Siparişler yap
bireysel.siparis_tamamla(5000)
bireysel.siparis_tamamla(3000)
kurumsal.siparis_tamamla(15000)
vip.siparis_tamamla(8000)
vip.siparis_tamamla(12000)

print("\n📊 Sipariş sonrası durum:")
for musteri in musteriler_liste:
    print(f"\n{musteri}")
    print(f"   Toplam Harcama: {musteri.toplam_harcama:,.2f} TL")
    print(f"   Sipariş Sayısı: {musteri.siparis_sayisi}")
    print(f"   Puan: {musteri.puan}")
    print(f"   Segment: {musteri.segment()}")

# MAGIC METHODS - Sıralama
print("\n" + "="*80)
print("🔮 MAGIC METHODS: SIRALAMA")
print("="*80)

print("\n🏆 En çok harcayan müşteriler:")
sirali_musteriler = sorted(musteriler_liste, reverse=True)
for i, m in enumerate(sirali_musteriler, 1):
    print(f"   {i}. {m} - {m.toplam_harcama:,.2f} TL")

print("\n" + "="*80)
````

## 🛒 **ADIM 3: Sepet ve Sipariş Sistemi**

Container magic methods ve composition pattern!

````python
# 🛒 SEPET VE SİPARİŞ SİSTEMİ - Composition & Magic Methods

class SepetItem:
    """
    📦 Sepetteki tek bir ürün
    """
    
    def __init__(self, urun: UrunBase, adet: int):
        if adet <= 0:
            raise ValueError("Adet 0'dan büyük olmalı!")
        
        if urun.stok < adet:
            raise ValueError(f"Yetersiz stok! Mevcut: {urun.stok}")
        
        self.urun = urun
        self.adet = adet
        self.ekleme_tarihi = datetime.now()
    
    def ara_toplam(self) -> float:
        """Bu item'ın toplam fiyatı"""
        return self.urun.fiyat * self.adet
    
    def kargo_toplami(self) -> float:
        """Bu item'ın kargo toplamı"""
        return self.urun.kargo_ucreti_hesapla()
    
    def __str__(self):
        return f"{self.urun.ad} x{self.adet} = {self.ara_toplam():,.2f} TL"
    
    def __repr__(self):
        return f"SepetItem(urun_id={self.urun.urun_id}, adet={self.adet})"


class Sepet:
    """
    🛒 Alışveriş Sepeti - Container Magic Methods
    """
    
    def __init__(self, musteri: MusteriBase):
        self.musteri = musteri
        self.items: List[SepetItem] = []
        self.olusturma_tarihi = datetime.now()
    
    # 🔮 MAGIC METHODS - Container
    
    def __len__(self):
        """Sepetteki ürün çeşit sayısı"""
        return len(self.items)
    
    def __getitem__(self, index):
        """Sepet[0] ile erişim"""
        return self.items[index]
    
    def __iter__(self):
        """for item in sepet:"""
        return iter(self.items)
    
    def __contains__(self, urun_id: int):
        """urun_id in sepet"""
        return any(item.urun.urun_id == urun_id for item in self.items)
    
    def __str__(self):
        if not self.items:
            return "🛒 Sepet boş!"
        
        output = f"\n{'='*70}\n"
        output += f"🛒 {self.musteri.ad}'ın Sepeti\n"
        output += f"{'='*70}\n"
        
        for i, item in enumerate(self.items, 1):
            output += f"{i}. {item}\n"
        
        output += f"{'-'*70}\n"
        output += f"Ara Toplam: {self.ara_toplam():>10,.2f} TL\n"
        output += f"Kargo: {self.toplam_kargo():>10,.2f} TL\n"
        output += f"{'='*70}\n"
        output += f"TOPLAM: {self.genel_toplam():>10,.2f} TL\n"
        output += f"{'='*70}"
        
        return output
    
    # İŞ LOJİĞİ
    
    def ekle(self, urun: UrunBase, adet: int = 1):
        """Sepete ürün ekle"""
        # Ürün zaten sepette mi?
        for item in self.items:
            if item.urun.urun_id == urun.urun_id:
                # Adet artır
                yeni_adet = item.adet + adet
                if urun.stok < yeni_adet:
                    raise ValueError(f"Yetersiz stok! Mevcut: {urun.stok}")
                item.adet = yeni_adet
                return f"✅ {urun.ad} adedi güncellendi: {item.adet}"
        
        # Yeni ürün ekle
        item = SepetItem(urun, adet)
        self.items.append(item)
        return f"✅ {urun.ad} sepete eklendi!"
    
    def cikar(self, urun_id: int):
        """Ürünü sepetten çıkar"""
        self.items = [item for item in self.items if item.urun.urun_id != urun_id]
        return "✅ Ürün sepetten çıkarıldı!"
    
    def bosalt(self):
        """Sepeti boşalt"""
        self.items.clear()
        return "✅ Sepet boşaltıldı!"
    
    def ara_toplam(self) -> float:
        """Ürünlerin toplamı (kargo hariç)"""
        return sum(item.ara_toplam() for item in self.items)
    
    def toplam_kargo(self) -> float:
        """Toplam kargo ücreti"""
        # Her ürün için 1 kez kargo
        return sum(item.kargo_toplami() for item in self.items)
    
    def genel_toplam(self) -> float:
        """Genel toplam (ürün + kargo)"""
        return self.ara_toplam() + self.toplam_kargo()
    
    def toplam_adet(self) -> int:
        """Sepetteki toplam ürün adedi"""
        return sum(item.adet for item in self.items)


class Siparis:
    """
    📦 Sipariş - Tamamlanmış bir alışveriş
    """
    
    _siparis_counter = 10000
    
    def __init__(self, musteri: MusteriBase, sepet: Sepet):
        if not sepet.items:
            raise ValueError("Sepet boş olamaz!")
        
        self.siparis_no = f"SIP{Siparis._siparis_counter}"
        Siparis._siparis_counter += 1
        
        self.musteri = musteri
        self.items = sepet.items.copy()  # Sepet kopyası
        self.siparis_tarihi = datetime.now()
        
        # Fiyat hesapla
        self.ara_toplam = sepet.ara_toplam()
        self.kargo_ucreti = sepet.toplam_kargo()
        self.indirimli_tutar = musteri.indirim_hesapla(self.ara_toplam)
        self.indirim = self.ara_toplam - self.indirimli_tutar
        self.genel_toplam = self.indirimli_tutar + self.kargo_ucreti
        
        # Durum
        self.durum = "Onaylandı"
        self.kargo_takip_no = None
    
    def __str__(self):
        output = f"\n{'='*70}\n"
        output += f"📦 Sipariş #{self.siparis_no}\n"
        output += f"{'='*70}\n"
        output += f"Müşteri: {self.musteri}\n"
        output += f"Tarih: {self.siparis_tarihi.strftime('%d.%m.%Y %H:%M')}\n"
        output += f"Durum: {self.durum}\n"
        output += f"{'-'*70}\n"
        
        for i, item in enumerate(self.items, 1):
            output += f"{i}. {item}\n"
        
        output += f"{'-'*70}\n"
        output += f"Ara Toplam: {self.ara_toplam:>10,.2f} TL\n"
        output += f"İndirim: {self.indirim:>10,.2f} TL\n"
        output += f"Kargo: {self.kargo_ucreti:>10,.2f} TL\n"
        output += f"{'='*70}\n"
        output += f"TOPLAM: {self.genel_toplam:>10,.2f} TL\n"
        output += f"{'='*70}"
        
        return output
    
    def kargoya_ver(self):
        """Siparişi kargoya ver"""
        import random
        self.durum = "Kargoda"
        self.kargo_takip_no = f"KRG{random.randint(100000, 999999)}"
        return f"✅ Sipariş kargoya verildi! Takip No: {self.kargo_takip_no}"
    
    def teslim_et(self):
        """Siparişi teslim et"""
        self.durum = "Teslim Edildi"
        
        # Stok güncelle
        for item in self.items:
            item.urun.satis_yap(item.adet)
        
        # Müşteri istatistiklerini güncelle
        self.musteri.siparis_tamamla(self.ara_toplam)
        
        return f"✅ Sipariş teslim edildi!"


# TEST: Sepet ve Sipariş
print("="*80)
print("🧪 SEPET VE SİPARİŞ SİSTEMİ TESTİ")
print("="*80)

# Müşteri ve ürünler (önceki testlerden)
test_musteri = BireyselMusteri(
    ad="Elif",
    soyad="Yıldız",
    email="elif.yildiz@email.com",
    telefon="5554445566",
    tc_no="11223344556"
)

# Sepet oluştur
sepet = Sepet(test_musteri)

print("\n🛒 SEPET İŞLEMLERİ:")
print(sepet.ekle(laptop, 1))
print(sepet.ekle(kitap, 3))
print(sepet.ekle(kurs, 2))

# MAGIC METHODS TESTİ
print(f"\n📏 len(sepet) = {len(sepet)} çeşit ürün")
print(f"📊 Toplam adet: {sepet.toplam_adet()} ürün")
print(f"🔍 Laptop sepette mi? {laptop.urun_id in sepet}")

# Sepet görünümü
print(sepet)

# İTERASYON
print("\n🔄 Sepetteki ürünler (iteration):")
for item in sepet:
    print(f"   - {item}")

# SİPARİŞ OLUŞTUR
print("\n" + "="*80)
print("📦 SİPARİŞ OLUŞTURMA")
print("="*80)

siparis = Siparis(test_musteri, sepet)
print(siparis)

# SİPARİŞ İŞLEMLERİ
print("\n📦 Sipariş İşlemleri:")
print(siparis.kargoya_ver())
print(siparis.teslim_et())

print(f"\n✅ Sipariş durumu: {siparis.durum}")
print(f"📊 Müşteri toplam harcama: {test_musteri.toplam_harcama:,.2f} TL")
print(f"🎁 Müşteri puan: {test_musteri.puan}")

print("\n" + "="*80)
````

````python
# 🎯 ANA E-TİCARET PLATFORM - Orchestrator Pattern

class ETicaretPlatform:
    """
    🏢 E-Ticaret Platform - Ana orchestrator sınıfı
    
    Tüm OOP prensiplerini kullanan kapsamlı sistem!
    """
    
    def __init__(self, platform_adi: str):
        self.platform_adi = platform_adi
        self.musteriler: Dict[int, MusteriBase] = {}
        self.urunler: Dict[int, UrunBase] = {}
        self.siparisler: List[Siparis] = []
        self.aktif_sepetler: Dict[int, Sepet] = {}
        self.baslangic_tarihi = datetime.now()
    
    # 🔮 MAGIC METHODS
    
    def __str__(self):
        return (f"🏢 {self.platform_adi}\n"
                f"   👥 {len(self.musteriler)} Müşteri\n"
                f"   📦 {len(self.urunler)} Ürün\n"
                f"   🛍️ {len(self.siparisler)} Sipariş")
    
    def __len__(self):
        """Toplam sipariş sayısı"""
        return len(self.siparisler)
    
    def __getitem__(self, siparis_no: str):
        """Sipariş no ile erişim"""
        for siparis in self.siparisler:
            if siparis.siparis_no == siparis_no:
                return siparis
        raise KeyError(f"Sipariş bulunamadı: {siparis_no}")
    
    # MÜŞTERİ YÖNETİMİ
    
    def musteri_ekle(self, musteri: MusteriBase):
        """Platform'a müşteri ekle"""
        self.musteriler[musteri.musteri_id] = musteri
        return f"✅ Müşteri eklendi: {musteri}"
    
    def musteri_bul(self, musteri_id: int) -> Optional[MusteriBase]:
        """Müşteri ID ile ara"""
        return self.musteriler.get(musteri_id)
    
    # ÜRÜN YÖNETİMİ
    
    def urun_ekle(self, urun: UrunBase):
        """Platform'a ürün ekle"""
        self.urunler[urun.urun_id] = urun
        return f"✅ Ürün eklendi: {urun}"
    
    def urun_bul(self, urun_id: int) -> Optional[UrunBase]:
        """Ürün ID ile ara"""
        return self.urunler.get(urun_id)
    
    def urun_ara(self, anahtar_kelime: str) -> List[UrunBase]:
        """İsimde arama yap"""
        sonuclar = []
        for urun in self.urunler.values():
            if anahtar_kelime.lower() in urun.ad.lower():
                sonuclar.append(urun)
        return sonuclar
    
    # SEPET YÖNETİMİ
    
    def sepet_olustur(self, musteri: MusteriBase) -> Sepet:
        """Müşteri için sepet oluştur"""
        sepet = Sepet(musteri)
        self.aktif_sepetler[musteri.musteri_id] = sepet
        return sepet
    
    def sepet_getir(self, musteri_id: int) -> Optional[Sepet]:
        """Müşterinin sepetini getir"""
        return self.aktif_sepetler.get(musteri_id)
    
    # SİPARİŞ YÖNETİMİ
    
    def siparis_olustur(self, musteri: MusteriBase) -> Siparis:
        """Sepetten sipariş oluştur"""
        sepet = self.sepet_getir(musteri.musteri_id)
        if not sepet or not sepet.items:
            raise ValueError("Sepet boş veya bulunamadı!")
        
        siparis = Siparis(musteri, sepet)
        self.siparisler.append(siparis)
        
        # Sepeti temizle
        sepet.bosalt()
        
        return siparis
    
    # ANALİZ & RAPORLAMA - PANDAS ENTEGRASYONU
    
    def musteriler_df(self) -> pd.DataFrame:
        """Müşterileri DataFrame'e çevir"""
        if not self.musteriler:
            return pd.DataFrame()
        
        data = []
        for m in self.musteriler.values():
            profil = m.profil_bilgisi()
            data.append(profil)
        
        return pd.DataFrame(data)
    
    def urunler_df(self) -> pd.DataFrame:
        """Ürünleri DataFrame'e çevir"""
        if not self.urunler:
            return pd.DataFrame()
        
        data = []
        for u in self.urunler.values():
            data.append({
                'ID': u.urun_id,
                'Ad': u.ad,
                'Kategori': u.kategori,
                'Fiyat': u.fiyat,
                'Stok': u.stok,
                'Satış Sayısı': u.satis_sayisi,
                'Stok Değeri': u.toplam_deger(),
                'Tip': u.__class__.__name__
            })
        
        return pd.DataFrame(data)
    
    def siparisler_df(self) -> pd.DataFrame:
        """Siparişleri DataFrame'e çevir"""
        if not self.siparisler:
            return pd.DataFrame()
        
        data = []
        for s in self.siparisler:
            data.append({
                'Sipariş No': s.siparis_no,
                'Müşteri ID': s.musteri.musteri_id,
                'Müşteri': f"{s.musteri.ad} {s.musteri.soyad}",
                'Tarih': s.siparis_tarihi,
                'Ara Toplam': s.ara_toplam,
                'İndirim': s.indirim,
                'Kargo': s.kargo_ucreti,
                'Toplam': s.genel_toplam,
                'Durum': s.durum,
                'Ürün Sayısı': len(s.items)
            })
        
        return pd.DataFrame(data)
    
    def dashboard(self):
        """📊 Yönetici Dashboard'u"""
        print("\n" + "="*90)
        print(f"📊 {self.platform_adi} - YÖNETİCİ DASHBOARD")
        print("="*90)
        
        # Genel İstatistikler
        print("\n🎯 GENEL İSTATİSTİKLER")
        print("-"*90)
        print(f"👥 Toplam Müşteri: {len(self.musteriler)}")
        print(f"📦 Toplam Ürün: {len(self.urunler)}")
        print(f"🛍️ Toplam Sipariş: {len(self.siparisler)}")
        
        if self.siparisler:
            toplam_ciro = sum(s.genel_toplam for s in self.siparisler)
            ortalama_sepet = toplam_ciro / len(self.siparisler)
            print(f"💰 Toplam Ciro: {toplam_ciro:,.2f} TL")
            print(f"📈 Ortalama Sepet: {ortalama_sepet:,.2f} TL")
        
        # En Çok Satan Ürünler
        if self.urunler:
            print("\n🏆 EN ÇOK SATAN ÜRÜNLER")
            print("-"*90)
            df_urunler = self.urunler_df()
            top_urunler = df_urunler.nlargest(5, 'Satış Sayısı')[['Ad', 'Satış Sayısı', 'Fiyat']]
            print(top_urunler.to_string(index=False))
        
        # En Değerli Müşteriler
        if self.musteriler:
            print("\n💎 EN DEĞERLİ MÜŞTERİLER")
            print("-"*90)
            df_musteriler = self.musteriler_df()
            top_musteriler = df_musteriler.nlargest(5, 'Toplam Harcama')[['Ad Soyad', 'Segment', 'Toplam Harcama', 'Sipariş Sayısı']]
            print(top_musteriler.to_string(index=False))
        
        # Kategori Analizi
        if self.siparisler:
            print("\n📊 KATEGORİ ANALİZİ")
            print("-"*90)
            kategori_data = []
            for urun in self.urunler.values():
                if urun.satis_sayisi > 0:
                    kategori_data.append({
                        'Kategori': urun.kategori,
                        'Satış': urun.satis_sayisi
                    })
            
            if kategori_data:
                df_kategori = pd.DataFrame(kategori_data)
                kategori_ozet = df_kategori.groupby('Kategori')['Satış'].sum().sort_values(ascending=False)
                print(kategori_ozet.to_string())
        
        # Müşteri Segmentasyonu
        if self.musteriler:
            print("\n👥 MÜŞTERİ SEGMENTASYONU")
            print("-"*90)
            df_musteriler = self.musteriler_df()
            segment_dagilim = df_musteriler['Segment'].value_counts()
            print(segment_dagilim.to_string())
        
        print("\n" + "="*90)


# 🎬 BÜYÜK DEMO: TÜM SİSTEM BİR ARADA!
print("="*90)
print("🎬 KAPSAMLI E-TİCARET PLATFORM DEMOsu")
print("="*90)

# Platform oluştur
platform = ETicaretPlatform("TechShop Online")

# ÜRÜNLER EKLE
print("\n📦 ÜRÜN EKLEME:")
urunler_ekle = [
    FizikselUrun("MacBook Pro 16", 85000, 10, "Laptop", 2.1, (35, 25, 2)),
    FizikselUrun("iPhone 15 Pro", 65000, 25, "Telefon", 0.2, (15, 8, 1)),
    FizikselUrun("AirPods Pro", 8500, 50, "Aksesuar", 0.05, (10, 10, 5)),
    FizikselUrun("Python for Data Science", 350, 100, "Kitap", 0.6, (20, 15, 3)),
    DijitalUrun("Complete Python Course", 799, 9999, "Eğitim", 25000, "https://platform.com/python"),
    DijitalUrun("Data Science Bootcamp", 1499, 9999, "Eğitim", 50000, "https://platform.com/ds"),
    FizikselUrun("Magic Mouse", 2500, 30, "Aksesuar", 0.1, (12, 7, 4)),
    FizikselUrun("4K Monitor", 15000, 15, "Monitör", 5.5, (60, 40, 10))
]

for urun in urunler_ekle:
    platform.urun_ekle(urun)

print(f"✅ {len(platform.urunler)} ürün eklendi!")

# MÜŞTERİLER EKLE
print("\n👥 MÜŞTERİ EKLEME:")
musteriler_ekle = [
    BireyselMusteri("Ahmet", "Yılmaz", "ahmet@email.com", "5551111111", "11111111111"),
    BireyselMusteri("Ayşe", "Demir", "ayse@email.com", "5552222222", "22222222222"),
    KurumsalMusteri("Mehmet", "Kaya", "mehmet@sirket.com", "5553333333", "TechCorp", "1234567890"),
    VIPMusteri("Fatma", "Şahin", "fatma@email.com", "5554444444", "33333333333"),
    BireyselMusteri("Ali", "Çelik", "ali@email.com", "5555555555", "44444444444"),
    KurumsalMusteri("Zeynep", "Arslan", "zeynep@firma.com", "5556666666", "DataSoft", "9876543210"),
]

for musteri in musteriler_ekle:
    platform.musteri_ekle(musteri)

print(f"✅ {len(platform.musteriler)} müşteri eklendi!")

# SİMÜLASYON: SİPARİŞLER OLUŞTUR
print("\n🛍️ SİPARİŞ SİMÜLASYONU:")
print("-"*90)

import random

for musteri in list(platform.musteriler.values())[:5]:  # İlk 5 müşteri
    # Sepet oluştur
    sepet = platform.sepet_olustur(musteri)
    
    # Random ürünler ekle
    urun_sayisi = random.randint(1, 4)
    eklenen_urunler = random.sample(list(platform.urunler.values()), urun_sayisi)
    
    for urun in eklenen_urunler:
        if urun.stok > 0:
            adet = random.randint(1, min(3, urun.stok))
            sepet.ekle(urun, adet)
    
    # Sipariş oluştur
    if sepet.items:
        siparis = platform.siparis_olustur(musteri)
        siparis.kargoya_ver()
        siparis.teslim_et()
        
        print(f"✅ {musteri.ad} - Sipariş: {siparis.siparis_no} - {siparis.genel_toplam:,.2f} TL")

print(f"\n✅ {len(platform.siparisler)} sipariş oluşturuldu!")

# DASHBOARD GÖSTER
platform.dashboard()

# PANDAS ANALİZ
print("\n" + "="*90)
print("🐼 PANDAS DETAYLI ANALİZ")
print("="*90)

# Siparişler DataFrame
df_siparisler = platform.siparisler_df()
print("\n📦 SİPARİŞLER:")
print(df_siparisler[['Sipariş No', 'Müşteri', 'Toplam', 'Durum']].to_string(index=False))

# İstatistikler
print("\n📊 SİPARİŞ İSTATİSTİKLERİ:")
print(df_siparisler[['Ara Toplam', 'İndirim', 'Kargo', 'Toplam']].describe())

# Müşteri analizi
df_musteriler = platform.musteriler_df()
print("\n👥 MÜŞTERİ ANALİZİ:")
print(df_musteriler[['Ad Soyad', 'Segment', 'Toplam Harcama', 'Puan']].to_string(index=False))

print("\n" + "="*90)
print("✅ KAPSAMLI E-TİCARET SİSTEMİ BAŞARIYLA ÇALIŞIYOR!")
print("="*90)
````

### 💡 **Design Patterns Kullanılanlar**

1. **Factory Pattern** (Implicit)
   - Farklı müşteri/ürün tipleri oluşturma

2. **Composite Pattern**
   - Sepet → SepetItem → Ürün hiyerarşisi

3. **Strategy Pattern**
   - Farklı indirim stratejileri (polymorphism ile)

4. **Facade Pattern**
   - `ETicaretPlatform` tüm karmaşıklığı saklar

5. **Observer Pattern** (Partial)
   - Stok güncellemeleri
