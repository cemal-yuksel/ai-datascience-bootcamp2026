## 🌟 ADIM 1: Giriş ve Motivasyon

### 📖 JOIN Nedir ve Neden Önemlidir?

<div style="background: linear-gradient(135deg, #f093fb 0%, #f5576c 100%); padding: 25px; border-radius: 12px; color: white; margin: 20px 0;">
    <h3 style="color: white; margin-top: 0;">💡 Temel Fikir</h3>
    <p style="font-size: 16px; line-height: 1.8;">
        Veritabanında tablolar genellikle <strong>"normalize"</strong> edilir; yani tekrar eden bilgileri tek tabloda şişirmek yerine <strong>parçalara ayırırız</strong>. 
    </p>
    <p style="font-size: 16px; line-height: 1.8;">
        Örneğin <strong>"Müşteri"</strong> bilgisi ayrı, <strong>"Sipariş"</strong> bilgisi ayrı tabloda tutulur. Bu mantıklıdır; ama rapor alırken <em>"müşteri adı + sipariş tarihi + tutar"</em> gibi birleşik bir çıktı isteriz.
    </p>
    <p style="font-size: 18px; font-weight: bold; margin-top: 15px;">
        🌉 İşte <span style="background: #fff; color: #f5576c; padding: 5px 10px; border-radius: 5px;">JOIN</span>, rapor ihtiyacı ile normalize tasarım arasındaki <strong>köprüdür</strong>.
    </p>
</div>

### ⚙️ Yardımcı Fonksiyonlar

JOIN işlemlerini daha kolay yönetmek için kullanışlı fonksiyonlar oluşturalım:

````python
# ⚙️ Yardımcı Fonksiyonlar Tanımlama

def run_query(query, params=None, fetch=True):
    """
    SQL sorgusunu çalıştır ve sonuçları DataFrame olarak döndür
    
    Args:
        query (str): Çalıştırılacak SQL sorgusu
        params (tuple): Sorgu parametreleri (opsiyonel)
        fetch (bool): Sonuçları getir mi? (SELECT için True, INSERT/UPDATE için False)
    
    Returns:
        pandas.DataFrame veya None
    """
    try:
        if fetch:
            # SELECT sorguları için - Pandas DataFrame döndür
            if params:
                df = pd.read_sql_query(query, conn, params=params)
            else:
                df = pd.read_sql_query(query, conn)
            return df
        else:
            # INSERT, UPDATE, DELETE için
            if params:
                cursor.execute(query, params)
            else:
                cursor.execute(query)
            conn.commit()
            return None
    except Exception as e:
        print(f"❌ Hata: {e}")
        return None


def display_table(table_name, limit=10):
    """
    Tablo içeriğini güzel bir şekilde göster
    
    Args:
        table_name (str): Gösterilecek tablo adı
        limit (int): Gösterilecek satır sayısı
    """
    query = f"SELECT * FROM {table_name} LIMIT {limit}"
    df = run_query(query)
    
    if df is not None and not df.empty:
        print(f"\n📊 {table_name} Tablosu (İlk {limit} satır)")
        print("=" * 70)
        
        # Styled display
        styled_df = df.style.set_properties(**{
            'background-color': '#f8f9fa',
            'color': '#333',
            'border': '1px solid #ddd',
            'padding': '10px',
            'text-align': 'left'
        }).set_table_styles([
            {'selector': 'th', 'props': [
                ('background', 'linear-gradient(135deg, #5e72e4 0%, #825ee4 100%)'),
                ('color', 'white'),
                ('font-weight', 'bold'),
                ('padding', '12px'),
                ('text-align', 'left')
            ]}
        ])
        
        display(styled_df)
        print(f"\n✅ Toplam {len(df)} satır gösteriliyor\n")
    else:
        print(f"⚠️ {table_name} tablosu boş veya bulunamadı.")


def get_table_info(table_name):
    """
    Tablo yapısını göster (sütunlar, tipler, vb.)
    
    Args:
        table_name (str): İncelenecek tablo adı
    """
    query = f"PRAGMA table_info({table_name})"
    df = run_query(query)
    
    if df is not None:
        html_info = f"""
        <div style="background: #f8f9fa; padding: 20px; border-radius: 10px; margin: 20px 0; border-left: 5px solid #11cdef;">
            <h4 style="color: #11cdef; margin-top: 0;">🔍 {table_name} - Tablo Yapısı</h4>
        """
        
        for idx, row in df.iterrows():
            pk_badge = " <span style='background: #f5365c; color: white; padding: 2px 8px; border-radius: 3px; font-size: 11px;'>PK</span>" if row['pk'] == 1 else ""
            html_info += f"""
            <div style="background: white; padding: 12px; margin: 8px 0; border-radius: 6px; border-left: 3px solid #11cdef;">
                <strong style="color: #5e72e4;">{row['name']}</strong>{pk_badge}<br/>
                <small style="color: #666;">Tip: <code>{row['type']}</code> | NULL: {'Hayır' if row['notnull'] else 'Evet'}</small>
            </div>
            """
        
        html_info += "</div>"
        display(HTML(html_info))


print("✅ Yardımcı fonksiyonlar tanımlandı!")
print("   📌 run_query() - SQL sorgusu çalıştır")
print("   📌 display_table() - Tablo içeriğini göster")
print("   📌 get_table_info() - Tablo yapısını incele")
````

## 🏗️ ADIM 4: SQL DDL - Tablo Oluşturma

Artık **veritabanı yapımızı** SQL komutları ile oluşturacağız! Bu adımda **DDL (Data Definition Language)** komutlarını öğreneceğiz.

### 📚 Bu Adımda Öğrenecekleriniz:

✅ **CREATE TABLE** komutu ile tablo oluşturma  
✅ **PRIMARY KEY** (Birincil Anahtar) tanımlama  
✅ **FOREIGN KEY** (Yabancı Anahtar) tanımlama  
✅ **Veri tipleri** (INTEGER, VARCHAR, DECIMAL, DATE)  
✅ **NOT NULL** constraint'leri  
✅ **Referential Integrity** (İlişkisel Bütünlük)

### 🎯 Oluşturacağımız Tablolar:

1. **departments** - Departman bilgileri
2. **employees** - Çalışan bilgileri
3. **projects** - Proje bilgileri
4. **tasks** - Görev bilgileri
5. **payments** - Ödeme bilgileri

> **💡 İpucu:** CREATE TABLE komutları ile tabloların iskeletini oluşturuyoruz. Daha sonra bu tablolara veri ekleyeceğiz.

````python
# 🏗️ ADIM 4: TechnoVision ERP Tablolarını Oluşturma
# SQL DDL (Data Definition Language) komutları ile tablo oluşturma

print("🏗️ TechnoVision ERP - Tablo Oluşturma İşlemi Başlıyor...")
print("=" * 70)

# 1️⃣ DEPARTMENTS Tablosu
print("\n📋 1. DEPARTMENTS tablosu oluşturuluyor...")

create_departments = """
CREATE TABLE IF NOT EXISTS departments (
    department_id INTEGER PRIMARY KEY AUTOINCREMENT,
    department_name VARCHAR(100) NOT NULL UNIQUE,
    budget DECIMAL(15, 2) NOT NULL,
    location VARCHAR(50) NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
"""

cursor.execute(create_departments)
print("   ✅ DEPARTMENTS tablosu başarıyla oluşturuldu!")

# 2️⃣ EMPLOYEES Tablosu
print("\n👥 2. EMPLOYEES tablosu oluşturuluyor...")

create_employees = """
CREATE TABLE IF NOT EXISTS employees (
    employee_id INTEGER PRIMARY KEY AUTOINCREMENT,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    department_id INTEGER NOT NULL,
    email VARCHAR(100) NOT NULL UNIQUE,
    hire_date DATE NOT NULL,
    salary DECIMAL(12, 2) NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (department_id) REFERENCES departments(department_id)
        ON DELETE RESTRICT
        ON UPDATE CASCADE
);
"""

cursor.execute(create_employees)
print("   ✅ EMPLOYEES tablosu başarıyla oluşturuldu!")

# 3️⃣ PROJECTS Tablosu
print("\n📊 3. PROJECTS tablosu oluşturuluyor...")

create_projects = """
CREATE TABLE IF NOT EXISTS projects (
    project_id INTEGER PRIMARY KEY AUTOINCREMENT,
    project_name VARCHAR(100) NOT NULL,
    department_id INTEGER NOT NULL,
    start_date DATE NOT NULL,
    end_date DATE,
    budget DECIMAL(15, 2) NOT NULL,
    status VARCHAR(20) DEFAULT 'ACTIVE',
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (department_id) REFERENCES departments(department_id)
        ON DELETE RESTRICT
        ON UPDATE CASCADE,
    CHECK (end_date IS NULL OR end_date >= start_date)
);
"""

cursor.execute(create_projects)
print("   ✅ PROJECTS tablosu başarıyla oluşturuldu!")

# 4️⃣ TASKS Tablosu
print("\n✅ 4. TASKS tablosu oluşturuluyor...")

create_tasks = """
CREATE TABLE IF NOT EXISTS tasks (
    task_id INTEGER PRIMARY KEY AUTOINCREMENT,
    project_id INTEGER NOT NULL,
    assigned_to INTEGER NOT NULL,
    task_name VARCHAR(200) NOT NULL,
    description TEXT,
    status VARCHAR(20) DEFAULT 'PENDING',
    due_date DATE,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (project_id) REFERENCES projects(project_id)
        ON DELETE CASCADE
        ON UPDATE CASCADE,
    FOREIGN KEY (assigned_to) REFERENCES employees(employee_id)
        ON DELETE RESTRICT
        ON UPDATE CASCADE
);
"""

cursor.execute(create_tasks)
print("   ✅ TASKS tablosu başarıyla oluşturuldu!")

# 5️⃣ PAYMENTS Tablosu
print("\n💰 5. PAYMENTS tablosu oluşturuluyor...")

create_payments = """
CREATE TABLE IF NOT EXISTS payments (
    payment_id INTEGER PRIMARY KEY AUTOINCREMENT,
    employee_id INTEGER NOT NULL,
    amount DECIMAL(12, 2) NOT NULL,
    payment_date DATE NOT NULL,
    payment_type VARCHAR(50) DEFAULT 'SALARY',
    description TEXT,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (employee_id) REFERENCES employees(employee_id)
        ON DELETE RESTRICT
        ON UPDATE CASCADE,
    CHECK (amount > 0)
);
"""

cursor.execute(create_payments)
print("   ✅ PAYMENTS tablosu başarıyla oluşturuldu!")

# Değişiklikleri kaydet
conn.commit()

print("\n" + "=" * 70)
print("🎉 TÜM TABLOLAR BAŞARIYLA OLUŞTURULDU!")
print("\n📊 Oluşturulan Tablolar:")
print("   1️⃣ departments  - Departman bilgileri")
print("   2️⃣ employees    - Çalışan bilgileri")
print("   3️⃣ projects     - Proje bilgileri")
print("   4️⃣ tasks        - Görev bilgileri")
print("   5️⃣ payments     - Ödeme bilgileri")
print("\n✨ İlişkisel bütünlük (Referential Integrity) sağlandı!")
print("🔗 Foreign Key constraint'leri aktif!")
print("=" * 70)
````

## 🔗 ADIM 6: INNER JOIN - İki Tabloyu Birleştirme

Artık **JOIN** işlemlerine başlıyoruz! JOIN, SQL'in **en güçlü özelliğidir** çünkü farklı tablolardaki verileri birleştirerek **anlamlı sonuçlar** elde etmemizi sağlar.

### 🎯 INNER JOIN Nedir?

**INNER JOIN**, iki tabloda **ortak olan kayıtları** birleştirir. Sadece **her iki tabloda da eşleşen** kayıtları getirir.

### 📚 Bu Adımda Öğrenecekleriniz:

✅ **INNER JOIN** syntax'ı ve kullanımı  
✅ **ON** clause ile eşleştirme koşulu  
✅ **Tablo aliasları** (kısaltmalar) kullanımı  
✅ **Sütun seçimi** ve isimlendirme  
✅ **WHERE** ile filtreleme  
✅ **ORDER BY** ile sıralama

### 🔍 INNER JOIN Mantığı (Adım Adım):

```
TABLO A          TABLO B          SONUÇ (INNER JOIN)
-------          -------          -------------------
  1 ────────────── A                1 - A
  2                B                2 - C
  3 ────────────── C                
  4                                 
```

**Açıklama:**
1. Tablo A'da ID=1, Tablo B'de ID=1 var → **Eşleşti! ✅**
2. Tablo A'da ID=2 yok, Tablo B'de var → **Eşleşmedi ❌**
3. Tablo A'da ID=3 var, Tablo B'de ID=3 var → **Eşleşti! ✅**
4. Tablo A'da ID=4 var, Tablo B'de yok → **Eşleşmedi ❌**

**Sonuç:** Sadece 1 ve 3 numaralı kayıtlar döner!

### 💡 INNER JOIN Kuralları:

1. **Her iki tabloda da bulunmalı** - Sadece eşleşen kayıtlar
2. **ON ile ilişki belirtilmeli** - Hangi sütunlar eşleşecek?
3. **Foreign Key genelde kullanılır** - department_id gibi

### 🎯 Gerçek Hayat Örneği:

**Soru:** "Her çalışanın adı, soyadı ve **hangi departmanda çalıştığını** göster"

**Problem:** 
- Çalışan bilgileri **employees** tablosunda
- Departman isimleri **departments** tablosunda
- employees'da sadece **department_id** var (sayı)
- Departman ismini görmek için **JOIN** gerekli!

> **💡 Önemli:** INNER JOIN sadece **ilişkili kayıtları** gösterir. Eğer bir çalışanın department_id'si departments tablosunda yoksa, o çalışan sonuçta **görünmez**!

### 🎯 Örnek 1: Temel INNER JOIN (Çalışan - Departman)

**Soru:** Her çalışanın adı, soyadı ve departman ismini göster.

**Tablo Analizi:**
- **employees** tablosu: first_name, last_name, department_id (sayı)
- **departments** tablosu: department_id (PK), department_name

**Problem:** employees'da sadece department_id var (örn: 1, 2, 3...). Departman ismini görmek için departments tablosuna JOIN yapmalıyız!

````python
# 🎯 Örnek 1: Temel INNER JOIN - Çalışan ve Departman Bilgileri
# Her çalışanın adı, soyadı ve departman ismini göster

print("🎯 Örnek 1: Temel INNER JOIN - Çalışan ve Departman")
print("=" * 70)

# ADIM 1: Önce employees tablosuna bakalım (JOIN yapmadan)
print("\n📋 ADIM 1: employees tablosundaki veriler (JOIN olmadan):")
print("-" * 70)

query_before = """
SELECT 
    employee_id,
    first_name,
    last_name,
    department_id
FROM employees
LIMIT 5;
"""

df_before = pd.read_sql_query(query_before, conn)
print(df_before.to_string(index=False))
print("\n❌ Problem: department_id sadece sayı (1, 2, 3...). Departman ismi yok!")

# ADIM 2: Şimdi INNER JOIN ile departments tablosunu birleştirelim
print("\n" + "=" * 70)
print("📋 ADIM 2: INNER JOIN ile departman isimlerini ekleyelim:")
print("-" * 70)

query_join = """
SELECT 
    e.employee_id,
    e.first_name,
    e.last_name,
    e.department_id,
    d.department_name,
    d.location
FROM employees e
INNER JOIN departments d
    ON e.department_id = d.department_id
ORDER BY e.employee_id
LIMIT 10;
"""

df_join = pd.read_sql_query(query_join, conn)

# Sonucu güzelce göster
styled_result = df_join.style.set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#172b4d',
    'border': '1px solid #dee2e6',
    'padding': '8px',
    'text-align': 'left'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#5e72e4'),
        ('color', 'white'),
        ('padding', '10px'),
        ('text-align', 'left'),
        ('font-weight', 'bold')
    ]}
]).apply(lambda x: ['background-color: #e3f2fd' if x.name % 2 == 0 else '' for i in x], axis=1)

display(styled_result)

print("\n✅ Artık her çalışanın departman ismi görünüyor!")

# ADIM 3: SQL sorgusunu adım adım açıklayalım
print("\n" + "=" * 70)
print("📖 ADIM 3: SQL Sorgusu Açıklaması:")
print("-" * 70)

explanation_html = """
<div style="background: white; border-radius: 15px; padding: 25px; margin: 20px 0; box-shadow: 0 5px 20px rgba(0,0,0,0.1);">
    <h3 style="color: #5e72e4; margin-top: 0;">🔍 SQL Sorgusu Satır Satır Açıklama</h3>
    
    <div style="background: #f8f9fa; padding: 20px; border-radius: 10px; border-left: 5px solid #5e72e4; margin: 15px 0;">
        <pre style="margin: 0; color: #172b4d; font-size: 14px; line-height: 2;">
<strong style="color: #5e72e4;">SELECT</strong>
    <span style="color: #2dce89;">e.employee_id</span>,        <span style="color: #666;">-- employees tablosundan çalışan ID</span>
    <span style="color: #2dce89;">e.first_name</span>,         <span style="color: #666;">-- employees tablosundan ad</span>
    <span style="color: #2dce89;">e.last_name</span>,          <span style="color: #666;">-- employees tablosundan soyad</span>
    <span style="color: #2dce89;">e.department_id</span>,      <span style="color: #666;">-- employees tablosundan departman ID (sayı)</span>
    <span style="color: #11cdef;">d.department_name</span>,    <span style="color: #666;">-- departments tablosundan departman ismi (INNER JOIN sayesinde!)</span>
    <span style="color: #11cdef;">d.location</span>            <span style="color: #666;">-- departments tablosundan lokasyon</span>
<strong style="color: #5e72e4;">FROM</strong> <span style="color: #fb6340;">employees e</span>           <span style="color: #666;">-- Ana tablo (e = employees için kısaltma/alias)</span>
<strong style="color: #5e72e4;">INNER JOIN</strong> <span style="color: #11cdef;">departments d</span>  <span style="color: #666;">-- Birleştireceğimiz tablo (d = departments için alias)</span>
    <strong style="color: #5e72e4;">ON</strong> <span style="color: #fb6340;">e</span>.<span style="color: #f5365c;">department_id</span> = <span style="color: #11cdef;">d</span>.<span style="color: #2dce89;">department_id</span>  <span style="color: #666;">-- Eşleştirme koşulu (Foreign Key = Primary Key)</span>
<strong style="color: #5e72e4;">ORDER BY</strong> <span style="color: #2dce89;">e.employee_id</span>    <span style="color: #666;">-- Çalışan ID'ye göre sırala</span>
<strong style="color: #5e72e4;">LIMIT</strong> 10;                   <span style="color: #666;">-- İlk 10 kaydı getir</span></pre>
    </div>
    
    <div style="display: grid; grid-template-columns: repeat(2, 1fr); gap: 20px; margin-top: 20px;">
        
        <div style="background: #e3f2fd; padding: 20px; border-radius: 10px; border-left: 5px solid #5e72e4;">
            <h4 style="color: #5e72e4; margin-top: 0;">🔑 Önemli Kavramlar</h4>
            <ul style="color: #666; line-height: 2;">
                <li><strong>e</strong> ve <strong>d</strong> = <strong>Alias</strong> (takma ad)</li>
                <li><strong>e.department_id</strong> = employees'daki FK</li>
                <li><strong>d.department_id</strong> = departments'taki PK</li>
                <li><strong>ON</strong> = Eşleştirme koşulu</li>
            </ul>
        </div>
        
        <div style="background: #fff3e0; padding: 20px; border-radius: 10px; border-left: 5px solid #fb6340;">
            <h4 style="color: #fb6340; margin-top: 0;">💡 Neden Alias Kullanıyoruz?</h4>
            <ul style="color: #666; line-height: 2;">
                <li>Kod daha kısa ve okunabilir</li>
                <li><code>employees.first_name</code> yerine <code>e.first_name</code></li>
                <li>Uzun tablo isimleri için zorunlu</li>
                <li>Aynı tabloyu 2 kez kullanınca gerekli</li>
            </ul>
        </div>
        
    </div>
    
    <div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 20px; border-radius: 10px; margin-top: 20px; color: white;">
        <h4 style="color: white; margin-top: 0; text-align: center;">🎯 INNER JOIN Nasıl Çalışıyor?</h4>
        <ol style="line-height: 2; font-size: 14px;">
            <li><strong>employees</strong> tablosundan her satırı al</li>
            <li>Her satırın <strong>department_id</strong> değerine bak</li>
            <li><strong>departments</strong> tablosunda aynı department_id'yi bul</li>
            <li>Eşleşen satırları <strong>birleştir</strong></li>
            <li>Eşleşmezse o satırı <strong>gösterme</strong> (INNER JOIN özelliği!)</li>
        </ol>
    </div>
    
</div>
"""

display(HTML(explanation_html))

print("\n" + "=" * 70)
print("✅ Temel INNER JOIN örneği tamamlandı!")
print("🎯 İlk JOIN sorgunuzu başarıyla çalıştırdınız!")
print("=" * 70)
````

### 🎯 Örnek 2: INNER JOIN + WHERE (Filtreleme)

**Soru:** "Yazılım Geliştirme" departmanında çalışan kişilerin adı, soyadı ve maaşlarını göster.

**Yeni Kavram:** JOIN yaptıktan sonra **WHERE** ile filtreleme yapabiliriz!

**Mantık:**
1. Önce INNER JOIN ile tabloları birleştir
2. Sonra WHERE ile sadece istediğin kayıtları filtrele

````python
# 🎯 Örnek 2: INNER JOIN + WHERE - Filtreleme ile Birleştirme
# Belirli bir departmandaki çalışanları göster

print("🎯 Örnek 2: INNER JOIN + WHERE ile Filtreleme")
print("=" * 70)
print("\n💼 Soru: 'Yazılım Geliştirme' departmanındaki çalışanları göster\n")

# SQL Sorgusu
query_filtered = """
SELECT 
    e.employee_id AS 'Çalışan ID',
    e.first_name AS 'Ad',
    e.last_name AS 'Soyad',
    e.email AS 'E-posta',
    d.department_name AS 'Departman',
    e.salary AS 'Maaş (TL)',
    e.hire_date AS 'İşe Giriş'
FROM employees e
INNER JOIN departments d
    ON e.department_id = d.department_id
WHERE d.department_name = 'Yazılım Geliştirme'
ORDER BY e.salary DESC;
"""

df_filtered = pd.read_sql_query(query_filtered, conn)

# Sonucu göster
styled_filtered = df_filtered.style.set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#172b4d',
    'border': '1px solid #dee2e6',
    'padding': '10px',
    'text-align': 'left'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#5e72e4'),
        ('color', 'white'),
        ('padding', '12px'),
        ('text-align', 'left'),
        ('font-weight', 'bold')
    ]},
    {'selector': 'tbody tr:hover', 'props': [
        ('background-color', '#e3f2fd')
    ]}
]).format({'Maaş (TL)': '{:,.2f}'})

display(styled_filtered)

print(f"\n✅ Sonuç: {len(df_filtered)} çalışan bulundu!")
print(f"💰 En yüksek maaş: {df_filtered['Maaş (TL)'].max():,.2f} TL")
print(f"💰 Ortalama maaş: {df_filtered['Maaş (TL)'].mean():,.2f} TL")

# SQL Açıklaması
where_explanation = """
<div style="background: white; border-radius: 15px; padding: 25px; margin: 20px 0; box-shadow: 0 5px 20px rgba(0,0,0,0.1);">
    <h3 style="color: #11cdef; margin-top: 0;">🔍 WHERE ile Filtreleme Nasıl Çalışır?</h3>
    
    <div style="background: #f8f9fa; padding: 20px; border-radius: 10px; border-left: 5px solid #11cdef; margin: 15px 0;">
        <h4 style="color: #172b4d; margin-top: 0;">📝 SQL Sorgusu:</h4>
        <pre style="margin: 0; color: #172b4d; font-size: 14px; line-height: 2;">
<strong style="color: #5e72e4;">SELECT</strong> e.first_name, e.last_name, e.salary, d.department_name
<strong style="color: #5e72e4;">FROM</strong> employees e
<strong style="color: #5e72e4;">INNER JOIN</strong> departments d
    <strong style="color: #5e72e4;">ON</strong> e.department_id = d.department_id
<strong style="color: #11cdef;">WHERE</strong> d.department_name = <strong style="color: #f5365c;">'Yazılım Geliştirme'</strong>    <span style="color: #666;">← Filtreleme!</span>
<strong style="color: #5e72e4;">ORDER BY</strong> e.salary DESC;</pre>
    </div>
    
    <div style="display: grid; grid-template-columns: repeat(2, 1fr); gap: 20px; margin-top: 20px;">
        
        <div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 20px; border-radius: 10px; color: white;">
            <h4 style="color: white; margin-top: 0;">🔄 İşlem Sırası</h4>
            <ol style="line-height: 2; margin: 0;">
                <li><strong>FROM</strong> employees → Ana tablo</li>
                <li><strong>INNER JOIN</strong> departments → Birleştir</li>
                <li><strong>WHERE</strong> filtrele → Sadece "Yazılım Geliştirme"</li>
                <li><strong>ORDER BY</strong> sırala → Maaşa göre</li>
            </ol>
        </div>
        
        <div style="background: #e3f2fd; padding: 20px; border-radius: 10px; border-left: 5px solid #11cdef;">
            <h4 style="color: #11cdef; margin-top: 0;">💡 Önemli Notlar</h4>
            <ul style="color: #666; line-height: 2; margin: 0;">
                <li><strong>WHERE</strong> her zaman <strong>JOIN'den sonra</strong> gelir</li>
                <li>Birleştirme sonrası filtreleme yapar</li>
                <li>Performans için önce JOIN, sonra WHERE</li>
                <li>WHERE'de hem e hem d sütunlarını kullanabilirsiniz</li>
            </ul>
        </div>
        
    </div>
    
    <div style="background: linear-gradient(135deg, #2dce89 0%, #2dcecc 100%); padding: 20px; border-radius: 10px; margin-top: 20px; color: white;">
        <h4 style="color: white; margin-top: 0; text-align: center;">🎯 WHERE vs ON Farkı</h4>
        <div style="display: grid; grid-template-columns: repeat(2, 1fr); gap: 20px; margin-top: 15px;">
            <div>
                <strong>ON Clause:</strong><br/>
                <small>→ Tabloları <strong>nasıl birleştireceğini</strong> belirtir<br/>
                → Eşleştirme koşulu<br/>
                → Örnek: ON e.dept_id = d.dept_id</small>
            </div>
            <div>
                <strong>WHERE Clause:</strong><br/>
                <small>→ Birleşmiş veriyi <strong>nasıl filtreleyeceğini</strong> belirtir<br/>
                → Filtreleme koşulu<br/>
                → Örnek: WHERE d.name = 'IT'</small>
            </div>
        </div>
    </div>
    
</div>
"""

display(HTML(where_explanation))

print("\n" + "=" * 70)
print("✅ INNER JOIN + WHERE örneği tamamlandı!")
print("=" * 70)
````

### 🎯 Örnek 3: 3 Tablo Birleştirme (Çoklu JOIN)

**Soru:** Her görevin adı, hangi projede olduğu, hangi departmana ait olduğu ve kime atandığını göster.

**Yeni Kavram:** **Birden fazla JOIN** kullanarak 3 veya daha fazla tabloyu birleştirebiliriz!

**Tablo İlişkileri:**
- **tasks** → **projects** (project_id ile)
- **projects** → **departments** (department_id ile)
- **tasks** → **employees** (assigned_to ile)

Bu 3 tabloyu birleştirerek tüm bilgileri göreceğiz!

````python
# 🎯 Örnek 3: 3 Tablo Birleştirme - Çoklu INNER JOIN
# tasks + projects + departments + employees birleştirmesi

print("🎯 Örnek 3: 3 Tablo Birleştirme - Çoklu INNER JOIN")
print("=" * 70)
print("\n📊 Soru: Her görevin detaylı bilgilerini göster\n")

# 4 Tablo birleştirme!
query_multi = """
SELECT 
    t.task_id AS 'Görev ID',
    t.task_name AS 'Görev Adı',
    t.status AS 'Durum',
    p.project_name AS 'Proje',
    d.department_name AS 'Departman',
    e.first_name || ' ' || e.last_name AS 'Atanan Kişi',
    e.email AS 'E-posta'
FROM tasks t
INNER JOIN projects p
    ON t.project_id = p.project_id
INNER JOIN departments d
    ON p.department_id = d.department_id
INNER JOIN employees e
    ON t.assigned_to = e.employee_id
ORDER BY t.task_id
LIMIT 10;
"""

df_multi = pd.read_sql_query(query_multi, conn)

# Durum renklendirmesi için fonksiyon
def color_status(val):
    if val == 'COMPLETED':
        return 'background-color: #d4edda; color: #155724; font-weight: bold;'
    elif val == 'IN_PROGRESS':
        return 'background-color: #fff3cd; color: #856404; font-weight: bold;'
    else:  # PENDING
        return 'background-color: #f8d7da; color: #721c24; font-weight: bold;'

styled_multi = df_multi.style.set_properties(**{
    'border': '1px solid #dee2e6',
    'padding': '10px',
    'text-align': 'left'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#2dce89'),
        ('color', 'white'),
        ('padding', '12px'),
        ('text-align', 'left'),
        ('font-weight', 'bold')
    ]},
    {'selector': 'tbody tr:nth-child(even)', 'props': [
        ('background-color', '#f8f9fa')
    ]}
]).applymap(color_status, subset=['Durum'])

display(styled_multi)

print(f"\n✅ Toplam {len(df_multi)} görev gösteriliyor (ilk 10)")

# SQL Açıklaması
multi_join_explanation = """
<div style="background: white; border-radius: 15px; padding: 25px; margin: 20px 0; box-shadow: 0 5px 20px rgba(0,0,0,0.1);">
    <h3 style="color: #2dce89; margin-top: 0;">🔗 Çoklu JOIN Nasıl Çalışır?</h3>
    
    <div style="background: #f8f9fa; padding: 20px; border-radius: 10px; border-left: 5px solid #2dce89; margin: 15px 0;">
        <h4 style="color: #172b4d; margin-top: 0;">📝 SQL Sorgusu (4 Tablo):</h4>
        <pre style="margin: 0; color: #172b4d; font-size: 13px; line-height: 2;">
<strong style="color: #5e72e4;">SELECT</strong> t.task_name, p.project_name, d.department_name, e.first_name
<strong style="color: #5e72e4;">FROM</strong> <span style="color: #fb6340;">tasks t</span>                           <span style="color: #666;">← 1. Ana tablo</span>
<strong style="color: #2dce89;">INNER JOIN</strong> <span style="color: #11cdef;">projects p</span>                 <span style="color: #666;">← 2. İlk birleştirme</span>
    <strong style="color: #5e72e4;">ON</strong> <span style="color: #fb6340;">t</span>.<strong>project_id</strong> = <span style="color: #11cdef;">p</span>.<strong>project_id</strong>     <span style="color: #666;">   (tasks → projects)</span>
<strong style="color: #2dce89;">INNER JOIN</strong> <span style="color: #f5365c;">departments d</span>           <span style="color: #666;">← 3. İkinci birleştirme</span>
    <strong style="color: #5e72e4;">ON</strong> <span style="color: #11cdef;">p</span>.<strong>department_id</strong> = <span style="color: #f5365c;">d</span>.<strong>department_id</strong>  <span style="color: #666;">   (projects → departments)</span>
<strong style="color: #2dce89;">INNER JOIN</strong> <span style="color: #2dce89;">employees e</span>             <span style="color: #666;">← 4. Üçüncü birleştirme</span>
    <strong style="color: #5e72e4;">ON</strong> <span style="color: #fb6340;">t</span>.<strong>assigned_to</strong> = <span style="color: #2dce89;">e</span>.<strong>employee_id</strong>    <span style="color: #666;">   (tasks → employees)</span></pre>
    </div>
    
    <div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 25px; border-radius: 10px; margin: 20px 0; color: white;">
        <h4 style="color: white; margin-top: 0; text-align: center;">🔄 Adım Adım İşlem Akışı</h4>
        <div style="text-align: center; line-height: 2.5; font-size: 16px;">
            <strong>1️⃣ tasks</strong> (Ana Tablo)<br/>
            ↓ <em>INNER JOIN</em> (project_id)<br/>
            <strong>2️⃣ tasks + projects</strong><br/>
            ↓ <em>INNER JOIN</em> (department_id)<br/>
            <strong>3️⃣ tasks + projects + departments</strong><br/>
            ↓ <em>INNER JOIN</em> (assigned_to)<br/>
            <strong>4️⃣ tasks + projects + departments + employees</strong><br/>
            ✅ <strong>Sonuç!</strong>
        </div>
    </div>
    
    <div style="display: grid; grid-template-columns: repeat(2, 1fr); gap: 20px; margin-top: 20px;">
        
        <div style="background: #e3f2fd; padding: 20px; border-radius: 10px; border-left: 5px solid #11cdef;">
            <h4 style="color: #11cdef; margin-top: 0;">💡 Önemli Kurallar</h4>
            <ul style="color: #666; line-height: 2; margin: 0;">
                <li>Her JOIN için ayrı <strong>ON</strong> clause gerekir</li>
                <li>Sıra önemli! Sol → Sağ'a doğru</li>
                <li>İstediğin kadar JOIN yapabilirsin</li>
                <li>Performance: Fazla JOIN yavaşlatır</li>
            </ul>
        </div>
        
        <div style="background: #fff3cd; padding: 20px; border-radius: 10px; border-left: 5px solid #ffc107;">
            <h4 style="color: #856404; margin-top: 0;">⚠️ Dikkat Edilecekler</h4>
            <ul style="color: #666; line-height: 2; margin: 0;">
                <li>Her tablo için <strong>alias</strong> kullan (t, p, d, e)</li>
                <li>Hangi sütun hangi tablodan? Karışmasın!</li>
                <li>NULL değerler kayıplar yaratabilir</li>
                <li>Test ederken LIMIT kullan</li>
            </ul>
        </div>
        
    </div>
    
    <div style="background: linear-gradient(135deg, #2dce89 0%, #2dcecc 100%); padding: 20px; border-radius: 10px; margin-top: 20px; color: white; text-align: center;">
        <h4 style="color: white; margin-top: 0;">🎯 Gerçek Hayatta Kullanım</h4>
        <p style="line-height: 1.8; margin: 0;">
            E-ticaret sistemlerinde: <strong>Sipariş → Ürün → Kategori → Tedarikçi</strong><br/>
            CRM sistemlerinde: <strong>Müşteri → Sipariş → Ürün → Satış Temsilcisi</strong><br/>
            ERP sistemlerinde: <strong>Görev → Proje → Departman → Çalışan</strong> (Bizim örneğimiz!)
        </p>
    </div>
    
</div>
"""

display(HTML(multi_join_explanation))

# İstatistikler
print("\n" + "=" * 70)
print("📊 Görev Durumu İstatistikleri:")
print("-" * 70)

status_query = """
SELECT 
    t.status AS 'Durum',
    COUNT(*) AS 'Toplam',
    COUNT(*) * 100.0 / (SELECT COUNT(*) FROM tasks) AS 'Yüzde'
FROM tasks t
GROUP BY t.status
ORDER BY COUNT(*) DESC;
"""

df_status = pd.read_sql_query(status_query, conn)
for _, row in df_status.iterrows():
    status_icon = '✅' if row['Durum'] == 'COMPLETED' else '🔄' if row['Durum'] == 'IN_PROGRESS' else '⏳'
    print(f"   {status_icon} {row['Durum']:15s}: {row['Toplam']:2d} görev ({row['Yüzde']:5.1f}%)")

print("\n" + "=" * 70)
print("✅ Çoklu INNER JOIN örneği tamamlandı!")
print("=" * 70)
````

### 🎯 Örnek 4: INNER JOIN + GROUP BY (Aggregate Fonksiyonlar)

**Soru:** Her departmandaki çalışan sayısını ve ortalama maaşı göster.

**Yeni Kavram:** JOIN ile **GROUP BY** ve **Aggregate fonksiyonları** (COUNT, AVG, SUM, MAX, MIN) kullanabiliriz!

**Öğreneceğimiz:**
- COUNT() - Sayma
- AVG() - Ortalama
- MAX() - En yüksek
- MIN() - En düşük
- SUM() - Toplam

````python
# 🎯 Örnek 4: INNER JOIN + GROUP BY - Aggregate Fonksiyonlar
# Departman bazında istatistikler

print("🎯 Örnek 4: INNER JOIN + GROUP BY + Aggregate Fonksiyonlar")
print("=" * 70)
print("\n📊 Soru: Her departmanın çalışan istatistiklerini göster\n")

# Aggregate fonksiyonları ile JOIN
query_aggregate = """
SELECT 
    d.department_name AS 'Departman',
    d.location AS 'Lokasyon',
    COUNT(e.employee_id) AS 'Çalışan Sayısı',
    ROUND(AVG(e.salary), 2) AS 'Ortalama Maaş',
    MIN(e.salary) AS 'En Düşük Maaş',
    MAX(e.salary) AS 'En Yüksek Maaş',
    ROUND(SUM(e.salary), 2) AS 'Toplam Maaş Gideri'
FROM departments d
INNER JOIN employees e
    ON d.department_id = e.department_id
GROUP BY d.department_name, d.location
ORDER BY COUNT(e.employee_id) DESC;
"""

df_aggregate = pd.read_sql_query(query_aggregate, conn)

# Sonucu güzelce göster
styled_aggregate = df_aggregate.style.set_properties(**{
    'border': '1px solid #dee2e6',
    'padding': '10px',
    'text-align': 'left'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#fb6340'),
        ('color', 'white'),
        ('padding', '12px'),
        ('text-align', 'left'),
        ('font-weight', 'bold')
    ]},
    {'selector': 'tbody tr:nth-child(even)', 'props': [
        ('background-color', '#fff3e0')
    ]},
    {'selector': 'tbody tr:hover', 'props': [
        ('background-color', '#ffe5e5')
    ]}
]).format({
    'Ortalama Maaş': '{:,.2f} TL',
    'En Düşük Maaş': '{:,.2f} TL',
    'En Yüksek Maaş': '{:,.2f} TL',
    'Toplam Maaş Gideri': '{:,.2f} TL'
})

display(styled_aggregate)

print(f"\n✅ Toplam {len(df_aggregate)} departman")
print(f"💰 Şirket geneli toplam maaş gideri: {df_aggregate['Toplam Maaş Gideri'].sum():,.2f} TL")

# SQL Açıklaması
aggregate_explanation = """
<div style="background: white; border-radius: 15px; padding: 25px; margin: 20px 0; box-shadow: 0 5px 20px rgba(0,0,0,0.1);">
    <h3 style="color: #fb6340; margin-top: 0;">📊 Aggregate Fonksiyonlar ile JOIN</h3>
    
    <div style="background: #f8f9fa; padding: 20px; border-radius: 10px; border-left: 5px solid #fb6340; margin: 15px 0;">
        <h4 style="color: #172b4d; margin-top: 0;">📝 SQL Sorgusu:</h4>
        <pre style="margin: 0; color: #172b4d; font-size: 13px; line-height: 2;">
<strong style="color: #5e72e4;">SELECT</strong> 
    d.department_name,
    <strong style="color: #fb6340;">COUNT</strong>(e.employee_id) AS 'Çalışan Sayısı',      <span style="color: #666;">← Kaç çalışan?</span>
    <strong style="color: #fb6340;">AVG</strong>(e.salary) AS 'Ortalama Maaş',            <span style="color: #666;">← Ortalama ne kadar?</span>
    <strong style="color: #fb6340;">MIN</strong>(e.salary) AS 'En Düşük Maaş',            <span style="color: #666;">← En düşük ne kadar?</span>
    <strong style="color: #fb6340;">MAX</strong>(e.salary) AS 'En Yüksek Maaş',           <span style="color: #666;">← En yüksek ne kadar?</span>
    <strong style="color: #fb6340;">SUM</strong>(e.salary) AS 'Toplam Maaş'               <span style="color: #666;">← Toplam ne kadar?</span>
<strong style="color: #5e72e4;">FROM</strong> departments d
<strong style="color: #5e72e4;">INNER JOIN</strong> employees e
    <strong style="color: #5e72e4;">ON</strong> d.department_id = e.department_id
<strong style="color: #11cdef;">GROUP BY</strong> d.department_name                   <span style="color: #666;">← Departman bazında grupla!</span>
<strong style="color: #5e72e4;">ORDER BY</strong> COUNT(e.employee_id) DESC;           <span style="color: #666;">← En kalabalık departman üstte</span></pre>
    </div>
    
    <div style="display: grid; grid-template-columns: repeat(3, 1fr); gap: 15px; margin-top: 20px;">
        
        <div style="background: linear-gradient(135deg, #5e72e4 0%, #825ee4 100%); padding: 20px; border-radius: 10px; text-align: center; color: white;">
            <div style="font-size: 36px; margin-bottom: 10px;">📊</div>
            <h4 style="color: white; margin: 10px 0; font-size: 15px;">COUNT()</h4>
            <small style="opacity: 0.9;">Kayıt sayısını sayar<br/>Örnek: Kaç çalışan?</small>
        </div>
        
        <div style="background: linear-gradient(135deg, #11cdef 0%, #1171ef 100%); padding: 20px; border-radius: 10px; text-align: center; color: white;">
            <div style="font-size: 36px; margin-bottom: 10px;">📈</div>
            <h4 style="color: white; margin: 10px 0; font-size: 15px;">AVG()</h4>
            <small style="opacity: 0.9;">Ortalama hesaplar<br/>Örnek: Ortalama maaş?</small>
        </div>
        
        <div style="background: linear-gradient(135deg, #2dce89 0%, #2dcecc 100%); padding: 20px; border-radius: 10px; text-align: center; color: white;">
            <div style="font-size: 36px; margin-bottom: 10px;">💰</div>
            <h4 style="color: white; margin: 10px 0; font-size: 15px;">SUM()</h4>
            <small style="opacity: 0.9;">Toplam hesaplar<br/>Örnek: Toplam maaş?</small>
        </div>
        
        <div style="background: linear-gradient(135deg, #fb6340 0%, #fa3e3e 100%); padding: 20px; border-radius: 10px; text-align: center; color: white;">
            <div style="font-size: 36px; margin-bottom: 10px;">⬇️</div>
            <h4 style="color: white; margin: 10px 0; font-size: 15px;">MIN()</h4>
            <small style="opacity: 0.9;">En küçük değer<br/>Örnek: En düşük maaş?</small>
        </div>
        
        <div style="background: linear-gradient(135deg, #f5365c 0%, #f50057 100%); padding: 20px; border-radius: 10px; text-align: center; color: white;">
            <div style="font-size: 36px; margin-bottom: 10px;">⬆️</div>
            <h4 style="color: white; margin: 10px 0; font-size: 15px;">MAX()</h4>
            <small style="opacity: 0.9;">En büyük değer<br/>Örnek: En yüksek maaş?</small>
        </div>
        
        <div style="background: linear-gradient(135deg, #ffc107 0%, #ff9800 100%); padding: 20px; border-radius: 10px; text-align: center; color: white;">
            <div style="font-size: 36px; margin-bottom: 10px;">🎯</div>
            <h4 style="color: white; margin: 10px 0; font-size: 15px;">ROUND()</h4>
            <small style="opacity: 0.9;">Yuvarlama yapar<br/>Örnek: 12.456 → 12.46</small>
        </div>
        
    </div>
    
    <div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 25px; border-radius: 10px; margin: 20px 0; color: white;">
        <h4 style="color: white; margin-top: 0; text-align: center;">🔄 GROUP BY Nasıl Çalışır?</h4>
        <ol style="line-height: 2; font-size: 14px; margin: 0;">
            <li><strong>JOIN</strong> ile tabloları birleştir → Her çalışan için departman ismi gelir</li>
            <li><strong>GROUP BY</strong> ile departman bazında grupla → Aynı departmandakileri bir araya getir</li>
            <li><strong>Aggregate fonksiyonlar</strong> her grup için hesaplama yapar → COUNT, AVG, SUM, ...</li>
            <li>Her departman için <strong>tek bir satır</strong> döner</li>
        </ol>
    </div>
    
    <div style="display: grid; grid-template-columns: repeat(2, 1fr); gap: 20px; margin-top: 20px;">
        
        <div style="background: #e3f2fd; padding: 20px; border-radius: 10px; border-left: 5px solid #11cdef;">
            <h4 style="color: #11cdef; margin-top: 0;">💡 GROUP BY Kuralları</h4>
            <ul style="color: #666; line-height: 2; margin: 0;">
                <li>SELECT'te <strong>aggregate olmayan</strong> sütunlar GROUP BY'da olmalı</li>
                <li>Örnek: SELECT dept_name → GROUP BY dept_name</li>
                <li>Aggregate fonksiyonları (COUNT, AVG, ...) GROUP BY'da olmaz</li>
                <li>Birden fazla sütuna göre gruplama yapılabilir</li>
            </ul>
        </div>
        
        <div style="background: #fff3cd; padding: 20px; border-radius: 10px; border-left: 5px solid #ffc107;">
            <h4 style="color: #856404; margin-top: 0;">⚠️ Sık Yapılan Hatalar</h4>
            <ul style="color: #666; line-height: 2; margin: 0;">
                <li><strong>Hata:</strong> SELECT e.first_name, COUNT(*) GROUP BY dept</li>
                <li>→ first_name GROUP BY'da yok!</li>
                <li><strong>Doğru:</strong> SELECT dept, COUNT(*) GROUP BY dept</li>
                <li>Ya aggregate yap, ya da GROUP BY'a ekle!</li>
            </ul>
        </div>
        
    </div>
    
    <div style="background: linear-gradient(135deg, #2dce89 0%, #2dcecc 100%); padding: 20px; border-radius: 10px; margin-top: 20px; color: white; text-align: center;">
        <h4 style="color: white; margin-top: 0;">🎯 Gerçek Hayat Örnekleri</h4>
        <div style="line-height: 2;">
            <strong>Satış Raporları:</strong> Ürün kategorisi bazında toplam satış<br/>
            <strong>İnsan Kaynakları:</strong> Departman bazında ortalama maaş<br/>
            <strong>E-Ticaret:</strong> Müşteri bazında toplam sipariş tutarı<br/>
            <strong>Finans:</strong> Ay bazında gelir/gider özeti
        </div>
    </div>
    
</div>
"""

display(HTML(aggregate_explanation))

print("\n" + "=" * 70)
print("✅ INNER JOIN + GROUP BY + Aggregate örneği tamamlandı!")
print("=" * 70)
````

## 🎯 ADIM 7: LEFT JOIN (LEFT OUTER JOIN) - Soldaki Tablonun Tüm Kayıtları

### 📚 LEFT JOIN Nedir?

**LEFT JOIN**, soldaki (sol taraftaki) tablonun **TÜM kayıtlarını** getirir ve sağdaki tablodan eşleşenleri ekler. Eğer sağda eşleşme yoksa, o sütunlar **NULL** olarak gelir.

```
┌─────────────────────┐
│   SOL TABLO (Ana)   │  ← Buradaki HER kayıt sonuçta olacak!
└──────────┬──────────┘
           │
           │ LEFT JOIN
           │
┌──────────▼──────────┐
│   SAĞ TABLO         │  ← Burada eşleşen varsa gelir, yoksa NULL
└─────────────────────┘
```

### 🤔 INNER JOIN ile Farkı Nedir?

| Özellik | INNER JOIN | LEFT JOIN |
|---------|------------|-----------|
| **Sol tablodaki kayıtlar** | Sadece eşleşenler | **HEPSİ** (eşleşen + eşleşmeyen) |
| **Sağ tabloda eşleşme yoksa** | O satır sonuçta yok | Satır var ama sağ sütunlar **NULL** |
| **Kullanım alanı** | İlişkili kayıtlar | "Hangilerinde eksik?" sorguları |

### 💡 Gerçek Hayat Örneği

**Senaryo:** TechnoVision'da tüm departmanları görmek istiyoruz ama bazı departmanlarda henüz proje olmayabilir.

- **INNER JOIN kullanırsak:** Sadece projesi olan departmanları görürüz
- **LEFT JOIN kullanırsak:** TÜM departmanları görürüz (projesi olmayanlar NULL ile gelir)

Bu bize **"Hangi departmanlarda proje yok?"** sorusunun cevabını verir!

### 🎯 Örnek 1: Tüm Departmanlar ve Projeleri (Projesi Olmayanlar da Dahil)

````python
# 🎯 Örnek 1: Departmanlar ve Projeleri - LEFT JOIN

# 📊 ADIM 1: INNER JOIN ile görelim (Karşılaştırma için)
query_inner = """
SELECT 
    d.department_name AS 'Departman',
    d.location AS 'Lokasyon',
    COUNT(p.project_id) AS 'Proje Sayısı'
FROM departments d
INNER JOIN projects p ON d.department_id = p.department_id
GROUP BY d.department_name, d.location
ORDER BY COUNT(p.project_id) DESC;
"""

df_inner = run_query(query_inner)
print("🔗 INNER JOIN Sonucu:")
print(f"📌 Toplam Departman Sayısı: {len(df_inner)}")
print()
display(df_inner.style.set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#2c3e50',
    'border-color': '#dee2e6',
    'text-align': 'center',
    'font-size': '14px'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#5e72e4'),
        ('color', 'white'),
        ('font-weight', 'bold'),
        ('text-align', 'center'),
        ('padding', '12px')
    ]}
]))

print("\n" + "="*80 + "\n")

# 📊 ADIM 2: LEFT JOIN ile görelim
query_left = """
SELECT 
    d.department_name AS 'Departman',
    d.location AS 'Lokasyon',
    COUNT(p.project_id) AS 'Proje Sayısı',
    CASE 
        WHEN COUNT(p.project_id) = 0 THEN '❌ Projesi Yok'
        ELSE '✅ Proje Var'
    END AS 'Durum'
FROM departments d
LEFT JOIN projects p ON d.department_id = p.department_id
GROUP BY d.department_name, d.location
ORDER BY COUNT(p.project_id) DESC;
"""

df_left = run_query(query_left)
print("⬅️ LEFT JOIN Sonucu:")
print(f"📌 Toplam Departman Sayısı: {len(df_left)}")
print()

# Renklendirme fonksiyonu
def color_status(val):
    if '✅' in str(val):
        return 'background-color: #d4edda; color: #155724; font-weight: bold;'
    elif '❌' in str(val):
        return 'background-color: #f8d7da; color: #721c24; font-weight: bold;'
    return ''

styled_left = df_left.style.applymap(color_status, subset=['Durum']).set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#2c3e50',
    'border-color': '#dee2e6',
    'text-align': 'center',
    'font-size': '14px'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#11cdef'),
        ('color', 'white'),
        ('font-weight', 'bold'),
        ('text-align', 'center'),
        ('padding', '12px')
    ]}
])

display(styled_left)

# 📊 ADIM 3: Farkları vurgulayalım
print("\n" + "="*80)
print("🔍 KARŞILAŞTIRMA:")
print("="*80)
print(f"🔗 INNER JOIN → {len(df_inner)} departman (Sadece projesi olanlar)")
print(f"⬅️ LEFT JOIN  → {len(df_left)} departman (TÜM departmanlar)")
print(f"📌 Fark: {len(df_left) - len(df_inner)} departmanın projesi yok!")

# Projesi olmayan departmanları bulalım
no_projects = df_left[df_left['Proje Sayısı'] == 0]
if len(no_projects) > 0:
    print(f"\n❌ Projesi Olmayan Departmanlar:")
    for idx, row in no_projects.iterrows():
        print(f"   • {row['Departman']} ({row['Lokasyon']})")

# 💡 Detaylı Açıklama
explanation_html = """
<div style='background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 30px; border-radius: 15px; color: white; box-shadow: 0 10px 40px rgba(0,0,0,0.3); margin-top: 30px;'>
    <h2 style='text-align: center; margin-bottom: 25px; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);'>
        💡 LEFT JOIN - Adım Adım Açıklama
    </h2>
    
    <div style='background: rgba(255,255,255,0.1); padding: 25px; border-radius: 10px; backdrop-filter: blur(10px);'>
        <h3 style='margin-top: 0; border-bottom: 2px solid rgba(255,255,255,0.3); padding-bottom: 10px;'>
            🔍 SQL Sorgusu Analizi
        </h3>
        
        <div style='background: rgba(0,0,0,0.3); padding: 20px; border-radius: 10px; margin: 20px 0; font-family: "Courier New", monospace;'>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>SELECT</span>
                <div style='margin-left: 30px; color: #98fb98;'>
                    d.department_name <span style='color: #87ceeb;'>AS</span> 'Departman',<br>
                    d.location <span style='color: #87ceeb;'>AS</span> 'Lokasyon',<br>
                    <span style='color: #ffa500;'>COUNT</span>(p.project_id) <span style='color: #87ceeb;'>AS</span> 'Proje Sayısı'
                </div>
            </div>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>FROM</span> departments d
            </div>
            <div style='margin: 10px 0; font-weight: bold; font-size: 1.1em;'>
                <span style='color: #ff69b4;'>LEFT JOIN</span> projects p
                <div style='margin-left: 30px;'><span style='color: #ffd700;'>ON</span> d.department_id = p.department_id</div>
            </div>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>GROUP BY</span> d.department_name, d.location
            </div>
        </div>
        
        <div style='display: grid; grid-template-columns: repeat(2, 1fr); gap: 20px; margin-top: 25px;'>
            <div style='background: rgba(46, 213, 115, 0.2); padding: 20px; border-radius: 10px; border: 2px solid #2ed573;'>
                <h4 style='margin-top: 0;'>✅ LEFT JOIN'in Gücü</h4>
                <ul style='line-height: 1.8; margin: 10px 0;'>
                    <li><strong>departments</strong> tablosu SOL tarafta</li>
                    <li>Sol tablonun <strong>TÜM kayıtları</strong> gelir</li>
                    <li>Eşleşmeyen <code style='background: rgba(0,0,0,0.3); padding: 3px 8px; border-radius: 5px;'>projects</code> NULL olur</li>
                    <li><code style='background: rgba(0,0,0,0.3); padding: 3px 8px; border-radius: 5px;'>COUNT(p.project_id)</code> NULL'ları sayMAZ (0 döner)</li>
                </ul>
            </div>
            
            <div style='background: rgba(255, 193, 7, 0.2); padding: 20px; border-radius: 10px; border: 2px solid #ffc107;'>
                <h4 style='margin-top: 0;'>🎯 Neden Bu Sorgu?</h4>
                <ul style='line-height: 1.8; margin: 10px 0;'>
                    <li><strong>"Hangi departmanlarda proje yok?"</strong></li>
                    <li>Boştaki kaynakları görmek</li>
                    <li>İş dağılımını analiz etmek</li>
                    <li>Yeni projeler için uygun departman bulmak</li>
                </ul>
            </div>
        </div>
        
        <div style='background: rgba(255, 71, 87, 0.2); padding: 20px; border-radius: 10px; margin-top: 20px; border: 2px solid #ff4757;'>
            <h4 style='margin-top: 0;'>⚠️ Önemli: COUNT() Davranışı</h4>
            <p style='line-height: 1.8; font-size: 1.05em;'>
                <strong>COUNT(*)</strong> → NULL değerleri de sayar (her satırı sayar)<br>
                <strong>COUNT(sütun_adı)</strong> → NULL değerleri saymaz! ✅<br>
                <br>
                <code style='background: rgba(0,0,0,0.3); padding: 8px 12px; border-radius: 5px; display: inline-block; margin-top: 10px;'>
                    COUNT(p.project_id)
                </code> kullandık çünkü NULL olan projeleri <strong>saymamalıyız</strong>!
            </p>
        </div>
    </div>
</div>
"""

display(HTML(explanation_html))

print("\n✅ LEFT JOIN ile TÜM departmanları gördük (projesi olmasa bile)!")
print("🎯 INNER JOIN sadece 'ilişkili olanları' gösterirdi.")
````

### 🎯 Örnek 3: Çoklu LEFT JOIN - Ödeme Durumu Analizi

**İş Problemi:** Her çalışan için toplam maaş, aldığı ödeme ve kalan borç tutarını görmek istiyoruz. Bazı çalışanlar henüz ödeme almamış olabilir.

````python
# 🎯 Örnek 3: Ödeme Durumu Analizi - Çoklu LEFT JOIN

query_payments = """
SELECT 
    e.employee_id AS 'ID',
    e.first_name || ' ' || e.last_name AS 'Çalışan',
    d.department_name AS 'Departman',
    e.salary AS 'Aylık Maaş',
    COALESCE(SUM(p.amount), 0) AS 'Alınan Ödeme',
    e.salary - COALESCE(SUM(p.amount), 0) AS 'Kalan Borç',
    COUNT(p.payment_id) AS 'Ödeme Sayısı',
    CASE 
        WHEN SUM(p.amount) IS NULL THEN '❌ Ödeme Yok'
        WHEN e.salary - COALESCE(SUM(p.amount), 0) <= 0 THEN '✅ Tam Ödendi'
        WHEN e.salary - COALESCE(SUM(p.amount), 0) <= e.salary * 0.3 THEN '⚠️ Az Kaldı'
        ELSE '🔴 Çok Borç'
    END AS 'Durum'
FROM employees e
LEFT JOIN departments d ON e.department_id = d.department_id
LEFT JOIN payments p ON e.employee_id = p.employee_id
GROUP BY e.employee_id, e.first_name, e.last_name, d.department_name, e.salary
ORDER BY (e.salary - COALESCE(SUM(p.amount), 0)) DESC;
"""

df_payments = run_query(query_payments)
print("💰 Çalışan Ödeme Durumu Analizi:")
print(f"📌 Toplam Çalışan: {len(df_payments)}")
print()

# Para formatı
df_payments_display = df_payments.copy()
df_payments_display['Aylık Maaş'] = df_payments_display['Aylık Maaş'].apply(lambda x: f"{x:,.2f} TL")
df_payments_display['Alınan Ödeme'] = df_payments_display['Alınan Ödeme'].apply(lambda x: f"{x:,.2f} TL")
df_payments_display['Kalan Borç'] = df_payments_display['Kalan Borç'].apply(lambda x: f"{x:,.2f} TL")

# Renklendirme
def color_payment_status(val):
    if '✅' in str(val):
        return 'background-color: #d4edda; color: #155724; font-weight: bold;'
    elif '❌' in str(val):
        return 'background-color: #f8d7da; color: #721c24; font-weight: bold;'
    elif '⚠️' in str(val):
        return 'background-color: #fff3cd; color: #856404; font-weight: bold;'
    elif '🔴' in str(val):
        return 'background-color: #f5c6cb; color: #721c24; font-weight: bold;'
    return ''

styled_payments = df_payments_display.style.applymap(color_payment_status, subset=['Durum']).set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#2c3e50',
    'border-color': '#dee2e6',
    'text-align': 'center',
    'font-size': '13px'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#2dce89'),
        ('color', 'white'),
        ('font-weight', 'bold'),
        ('text-align', 'center'),
        ('padding', '12px'),
        ('font-size', '14px')
    ]}
])

display(styled_payments)

# 📊 İstatistikler
print("\n" + "="*80)
print("📊 GENEL İSTATİSTİKLER:")
print("="*80)

total_salary = df_payments['Aylık Maaş'].sum()
total_paid = df_payments['Alınan Ödeme'].sum()
total_debt = df_payments['Kalan Borç'].sum()

print(f"💰 Toplam Maaş Yükümlülüğü: {total_salary:,.2f} TL")
print(f"✅ Toplam Ödenen: {total_paid:,.2f} TL")
print(f"🔴 Toplam Kalan Borç: {total_debt:,.2f} TL")
print(f"📈 Ödeme Oranı: %{(total_paid/total_salary*100):.1f}")

# Ödeme alamayanlar
no_payment = df_payments[df_payments['Alınan Ödeme'] == 0]
print(f"\n❌ Hiç Ödeme Almayan: {len(no_payment)} kişi")
if len(no_payment) > 0:
    for idx, row in no_payment.iterrows():
        print(f"   • {row['Çalışan']} - {row['Departman']}")

# Tam ödenenler
full_paid = df_payments[df_payments['Kalan Borç'] <= 0]
print(f"\n✅ Tam Ödenen: {len(full_paid)} kişi")

# 💡 COALESCE Açıklaması
coalesce_html = """
<div style='background: linear-gradient(135deg, #4facfe 0%, #00f2fe 100%); padding: 30px; border-radius: 15px; color: white; box-shadow: 0 10px 40px rgba(0,0,0,0.3); margin-top: 30px;'>
    <h2 style='text-align: center; margin-bottom: 25px; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);'>
        💡 COALESCE() Fonksiyonu - NULL Değerleri Yönetme
    </h2>
    
    <div style='background: rgba(255,255,255,0.1); padding: 25px; border-radius: 10px; backdrop-filter: blur(10px);'>
        <h3 style='margin-top: 0; border-bottom: 2px solid rgba(255,255,255,0.3); padding-bottom: 10px;'>
            🎯 COALESCE Nedir?
        </h3>
        
        <p style='font-size: 1.1em; line-height: 1.8; margin: 20px 0;'>
            <strong>COALESCE(değer1, değer2, ...)</strong> fonksiyonu ilk NULL olmayan değeri döndürür.
            Eğer tüm değerler NULL ise, NULL döner.
        </p>
        
        <div style='background: rgba(0,0,0,0.3); padding: 20px; border-radius: 10px; margin: 20px 0; font-family: "Courier New", monospace;'>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>-- Ödeme yoksa NULL gelir, bunu 0'a çevir:</span><br>
                <span style='color: #ffa500;'>COALESCE</span>(SUM(p.amount), <span style='color: #98fb98;'>0</span>) <span style='color: #87ceeb;'>AS</span> 'Alınan Ödeme'
            </div>
            <br>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>-- Borç hesaplama (NULL olmasın):</span><br>
                e.salary - <span style='color: #ffa500;'>COALESCE</span>(SUM(p.amount), <span style='color: #98fb98;'>0</span>) <span style='color: #87ceeb;'>AS</span> 'Kalan Borç'
            </div>
        </div>
        
        <div style='display: grid; grid-template-columns: 1fr 1fr; gap: 20px; margin-top: 25px;'>
            <div style='background: rgba(255, 71, 87, 0.2); padding: 20px; border-radius: 10px; border: 2px solid #ff4757;'>
                <h4 style='margin-top: 0; text-align: center;'>❌ COALESCE Kullanmazsan</h4>
                <div style='background: rgba(0,0,0,0.3); padding: 15px; border-radius: 8px; margin: 15px 0;'>
                    <code style='font-size: 1em;'>
                        15000 - NULL = NULL ❌
                    </code>
                </div>
                <p style='margin: 10px 0; text-align: center;'>
                    Matematiksel işlem NULL döner!<br>
                    Sonuç kullanılamaz hale gelir.
                </p>
            </div>
            
            <div style='background: rgba(46, 213, 115, 0.2); padding: 20px; border-radius: 10px; border: 2px solid #2ed573;'>
                <h4 style='margin-top: 0; text-align: center;'>✅ COALESCE Kullanırsan</h4>
                <div style='background: rgba(0,0,0,0.3); padding: 15px; border-radius: 8px; margin: 15px 0;'>
                    <code style='font-size: 1em;'>
                        15000 - COALESCE(NULL, 0) = 15000 ✅
                    </code>
                </div>
                <p style='margin: 10px 0; text-align: center;'>
                    NULL değer 0'a dönüştürülür!<br>
                    İşlem doğru çalışır.
                </p>
            </div>
        </div>
        
        <div style='background: rgba(138, 43, 226, 0.2); padding: 20px; border-radius: 10px; margin-top: 25px; border: 2px solid #8a2be2;'>
            <h4 style='margin-top: 0;'>🎓 COALESCE Kullanım Örnekleri</h4>
            <div style='display: grid; grid-template-columns: 1fr 1fr; gap: 15px;'>
                <div>
                    <p style='margin: 5px 0; font-weight: bold;'>1️⃣ NULL'u 0'a çevir:</p>
                    <code style='background: rgba(0,0,0,0.3); padding: 8px; border-radius: 5px; display: block; font-size: 0.9em;'>
                        COALESCE(price, 0)
                    </code>
                </div>
                <div>
                    <p style='margin: 5px 0; font-weight: bold;'>2️⃣ Varsayılan metin:</p>
                    <code style='background: rgba(0,0,0,0.3); padding: 8px; border-radius: 5px; display: block; font-size: 0.9em;'>
                        COALESCE(name, 'Bilinmiyor')
                    </code>
                </div>
                <div>
                    <p style='margin: 5px 0; font-weight: bold;'>3️⃣ İlk dolu alan:</p>
                    <code style='background: rgba(0,0,0,0.3); padding: 8px; border-radius: 5px; display: block; font-size: 0.9em;'>
                        COALESCE(phone1, phone2, 'Yok')
                    </code>
                </div>
                <div>
                    <p style='margin: 5px 0; font-weight: bold;'>4️⃣ Hesaplamada güvenli:</p>
                    <code style='background: rgba(0,0,0,0.3); padding: 8px; border-radius: 5px; display: block; font-size: 0.9em;'>
                        price * COALESCE(discount, 1)
                    </code>
                </div>
            </div>
        </div>
        
        <div style='background: rgba(255, 193, 7, 0.2); padding: 20px; border-radius: 10px; margin-top: 20px; border: 2px solid #ffc107;'>
            <h4 style='margin-top: 0;'>⚠️ Önemli Notlar</h4>
            <ul style='line-height: 1.8; margin: 10px 0;'>
                <li><strong>COALESCE</strong> birden fazla parametre alabilir: <code style='background: rgba(0,0,0,0.3); padding: 3px 8px; border-radius: 5px;'>COALESCE(a, b, c, 0)</code></li>
                <li>LEFT JOIN'de sık kullanılır (NULL değerler yaygın olduğu için)</li>
                <li>Aggregate fonksiyonlarla (SUM, AVG vb.) birlikte kullan</li>
                <li>SQLite'da <code style='background: rgba(0,0,0,0.3); padding: 3px 8px; border-radius: 5px;'>IFNULL(değer, varsayılan)</code> da kullanılabilir (sadece 2 parametre)</li>
            </ul>
        </div>
    </div>
</div>
"""

display(HTML(coalesce_html))

print("\n✅ Çoklu LEFT JOIN + COALESCE ile ödeme analizini tamamladık!")
print("🎯 NULL değerleri yönetmeyi öğrendik!")
````

## 🎯 ADIM 8: RIGHT JOIN (RIGHT OUTER JOIN) - Sağdaki Tablonun Tüm Kayıtları

### 📚 RIGHT JOIN Nedir?

**RIGHT JOIN**, LEFT JOIN'in tam tersidir! Sağdaki (sağ taraftaki) tablonun **TÜM kayıtlarını** getirir ve soldaki tablodan eşleşenleri ekler.

```
┌─────────────────────┐
│   SOL TABLO         │  ← Burada eşleşen varsa gelir, yoksa NULL
└──────────┬──────────┘
           │
           │ RIGHT JOIN
           │
┌──────────▼──────────┐
│  SAĞ TABLO (Ana)    │  ← Buradaki HER kayıt sonuçta olacak!
└─────────────────────┘
```

### 🤔 LEFT JOIN ile Farkı Nedir?

| Özellik | LEFT JOIN | RIGHT JOIN |
|---------|-----------|------------|
| **Hangi tablo tamamen gelir?** | Sol tablo | **Sağ tablo** |
| **Hangi tarafta NULL olur?** | Sağ tarafta | **Sol tarafta** |
| **Kullanım sıklığı** | Çok yaygın (⭐⭐⭐⭐⭐) | Daha az (⭐⭐) |

### 💡 Önemli Not

> ⚠️ **RIGHT JOIN çok nadir kullanılır!** Neden?
> 
> Çünkü her RIGHT JOIN, LEFT JOIN'e çevrilebilir! Sadece tabloların yerini değiştirin:
> 
> ```sql
> -- Bunlar AYNI sonucu verir:
> A LEFT JOIN B  ≡  B RIGHT JOIN A
> ```
> 
> **Tavsiye:** Kodunuzu daha okunabilir yapmak için hep LEFT JOIN kullanın!

### 🎯 RIGHT JOIN Kullanım Durumu

**Senaryo:** Tüm projeleri ve hangi departmanlara ait olduklarını görmek istiyoruz. Bazı projeler departman ataması olmadan oluşturulmuş olabilir (orphan kayıtlar).

## 🎯 ADIM 9: FULL OUTER JOIN - Her İki Tablonun Tüm Kayıtları

### 📚 FULL OUTER JOIN Nedir?

**FULL OUTER JOIN**, LEFT JOIN + RIGHT JOIN birleşimidir! **Her iki tablonun da TÜM kayıtlarını** getirir.

```
┌─────────────────────┐
│   SOL TABLO         │  ← Tüm kayıtlar gelir (eşleşen + eşleşmeyen)
└──────────┬──────────┘
           │
           │ FULL OUTER JOIN
           │
┌──────────▼──────────┐
│   SAĞ TABLO         │  ← Tüm kayıtlar gelir (eşleşen + eşleşmeyen)
└─────────────────────┘
```

### 🤔 Diğer JOIN'lerle Karşılaştırma

| JOIN Tipi | Sol Tablo | Sağ Tablo | Sonuç |
|-----------|-----------|-----------|-------|
| **INNER JOIN** | Sadece eşleşenler | Sadece eşleşenler | Minimum kayıt |
| **LEFT JOIN** | **HEPSİ** ✅ | Sadece eşleşenler | Sol tablo kadar |
| **RIGHT JOIN** | Sadece eşleşenler | **HEPSİ** ✅ | Sağ tablo kadar |
| **FULL OUTER JOIN** | **HEPSİ** ✅ | **HEPSİ** ✅ | **Maksimum kayıt** |

### ⚠️ ÖNEMLİ: SQLite'da FULL OUTER JOIN YOK!

> SQLite, FULL OUTER JOIN'i doğrudan desteklemez! 😱
> 
> **Çözüm:** LEFT JOIN + UNION + RIGHT JOIN (WHERE IS NULL)
> 
> Veya daha kolay: LEFT JOIN UNION ALL (RIGHT JOIN WHERE IS NULL)

### 🎯 Kullanım Alanları

- Her iki tablodaki TÜM kayıtları görmek
- "Eşleşen + Eşleşmeyen" tam analiz
- Veri tutarsızlıklarını bulmak
- Yedek sistemler arası veri karşılaştırma

### 🎯 Örnek: SQLite'da FULL OUTER JOIN Simülasyonu

````python
# 🎯 FULL OUTER JOIN Simülasyonu (SQLite'da UNION kullanarak)

# 📊 ADIM 1: INNER JOIN (Sadece eşleşenler)
query_inner_compare = """
SELECT 
    d.department_name AS 'Departman',
    p.project_name AS 'Proje',
    'Eşleşti' AS 'Durum'
FROM departments d
INNER JOIN projects p ON d.department_id = p.department_id;
"""

df_inner_c = run_query(query_inner_compare)
print("🔗 INNER JOIN Sonucu:")
print(f"📌 Kayıt Sayısı: {len(df_inner_c)}")
print("🎯 Sadece ilişkili olanlar\n")

print("="*80 + "\n")

# 📊 ADIM 2: FULL OUTER JOIN Simülasyonu
# SQLite FULL OUTER JOIN'i desteklemediği için UNION kullanacağız
query_full_outer = """
-- Sol tarafın tümü (LEFT JOIN)
SELECT 
    d.department_name AS 'Departman',
    p.project_name AS 'Proje',
    CASE 
        WHEN p.project_id IS NULL THEN '⬅️ Sadece Sol'
        ELSE '✅ Eşleşti'
    END AS 'Durum'
FROM departments d
LEFT JOIN projects p ON d.department_id = p.department_id

UNION

-- Sağ tarafın eşleşmeyenleri (RIGHT JOIN kısmı)
SELECT 
    d.department_name AS 'Departman',
    p.project_name AS 'Proje',
    '➡️ Sadece Sağ' AS 'Durum'
FROM departments d
RIGHT JOIN projects p ON d.department_id = p.department_id
WHERE d.department_id IS NULL;
"""

df_full_outer = run_query(query_full_outer)
print("🔄 FULL OUTER JOIN (Simülasyon):")
print(f"📌 Toplam Kayıt: {len(df_full_outer)}")
print("🎯 Her iki tablonun TÜM kayıtları")
print()

# Renklendirme
def color_full_outer(val):
    if '⬅️' in str(val):
        return 'background-color: #d1ecf1; color: #0c5460; font-weight: bold;'
    elif '➡️' in str(val):
        return 'background-color: #f8d7da; color: #721c24; font-weight: bold;'
    elif '✅' in str(val):
        return 'background-color: #d4edda; color: #155724; font-weight: bold;'
    return ''

styled_full = df_full_outer.style.applymap(color_full_outer, subset=['Durum']).set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#2c3e50',
    'border-color': '#dee2e6',
    'text-align': 'center',
    'font-size': '14px'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#6f42c1'),
        ('color', 'white'),
        ('font-weight', 'bold'),
        ('text-align', 'center'),
        ('padding', '12px')
    ]}
])

display(styled_full)

# İstatistikler
matched = len(df_full_outer[df_full_outer['Durum'] == '✅ Eşleşti'])
left_only = len(df_full_outer[df_full_outer['Durum'] == '⬅️ Sadece Sol'])
right_only = len(df_full_outer[df_full_outer['Durum'] == '➡️ Sadece Sağ'])

print("\n" + "="*80)
print("📊 DURUM ANALİZİ:")
print("="*80)
print(f"✅ Eşleşen Kayıtlar: {matched}")
print(f"⬅️ Sadece Sol Tabloda (Proje olmayan departmanlar): {left_only}")
print(f"➡️ Sadece Sağ Tabloda (Departman olmayan projeler): {right_only}")
print(f"📌 Toplam: {len(df_full_outer)}")

# 💡 Açıklama
explanation_html = """
<div style='background: linear-gradient(135deg, #6f42c1 0%, #9b59b6 100%); padding: 30px; border-radius: 15px; color: white; box-shadow: 0 10px 40px rgba(0,0,0,0.3); margin-top: 30px;'>
    <h2 style='text-align: center; margin-bottom: 25px; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);'>
        💡 SQLite'da FULL OUTER JOIN Nasıl Yapılır?
    </h2>
    
    <div style='background: rgba(255,255,255,0.1); padding: 25px; border-radius: 10px; backdrop-filter: blur(10px);'>
        <h3 style='margin-top: 0; border-bottom: 2px solid rgba(255,255,255,0.3); padding-bottom: 10px;'>
            🔧 UNION Tekniği
        </h3>
        
        <div style='background: rgba(0,0,0,0.3); padding: 20px; border-radius: 10px; margin: 20px 0; font-family: "Courier New", monospace; font-size: 0.95em;'>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>-- ADIM 1: Sol tarafın TÜM kayıtları</span><br>
                <span style='color: #ffd700;'>SELECT</span> ...<br>
                <span style='color: #ffd700;'>FROM</span> tablo1<br>
                <span style='color: #ff69b4;'>LEFT JOIN</span> tablo2 ...<br>
            </div>
            <div style='margin: 20px 0; text-align: center; font-size: 1.2em; font-weight: bold;'>
                <span style='color: #ffa500;'>UNION</span>
            </div>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>-- ADIM 2: Sağ taraftaki eşleşmeyenler</span><br>
                <span style='color: #ffd700;'>SELECT</span> ...<br>
                <span style='color: #ffd700;'>FROM</span> tablo1<br>
                <span style='color: #ff69b4;'>RIGHT JOIN</span> tablo2 ...<br>
                <span style='color: #ffd700;'>WHERE</span> tablo1.id <span style='color: #ff6b6b;'>IS NULL</span><br>
            </div>
        </div>
        
        <div style='display: grid; grid-template-columns: repeat(3, 1fr); gap: 15px; margin-top: 25px;'>
            <div style='background: rgba(46, 213, 115, 0.2); padding: 15px; border-radius: 10px; border: 2px solid #2ed573;'>
                <h4 style='margin: 0 0 10px 0; text-align: center;'>1️⃣ LEFT JOIN</h4>
                <p style='margin: 0; font-size: 0.9em; text-align: center; line-height: 1.6;'>
                    Sol tablonun HEPSİ<br>+ Eşleşenler
                </p>
            </div>
            
            <div style='background: rgba(255, 193, 7, 0.2); padding: 15px; border-radius: 10px; border: 2px solid #ffc107;'>
                <h4 style='margin: 0 0 10px 0; text-align: center;'>➕ UNION</h4>
                <p style='margin: 0; font-size: 0.9em; text-align: center; line-height: 1.6;'>
                    İki sorguyu<br>birleştir
                </p>
            </div>
            
            <div style='background: rgba(255, 71, 87, 0.2); padding: 15px; border-radius: 10px; border: 2px solid #ff4757;'>
                <h4 style='margin: 0 0 10px 0; text-align: center;'>2️⃣ RIGHT (NULL)</h4>
                <p style='margin: 0; font-size: 0.9em; text-align: center; line-height: 1.6;'>
                    Sağ tabloda olup<br>solda olmayan
                </p>
            </div>
        </div>
        
        <div style='background: rgba(255, 255, 255, 0.1); padding: 20px; border-radius: 10px; margin-top: 25px; border: 2px dashed rgba(255,255,255,0.3);'>
            <h4 style='margin-top: 0;'>🎯 UNION vs UNION ALL</h4>
            <div style='display: grid; grid-template-columns: 1fr 1fr; gap: 20px;'>
                <div>
                    <h5 style='margin: 10px 0; color: #2ed573;'>✅ UNION</h5>
                    <ul style='line-height: 1.8; margin: 5px 0; font-size: 0.95em;'>
                        <li>Tekrarlı kayıtları temizler</li>
                        <li>Daha yavaş (DISTINCT yapıyor)</li>
                        <li>FULL OUTER JOIN için ideal ✅</li>
                    </ul>
                </div>
                <div>
                    <h5 style='margin: 10px 0; color: #ffc107;'>⚡ UNION ALL</h5>
                    <ul style='line-height: 1.8; margin: 5px 0; font-size: 0.95em;'>
                        <li>Tekrarlı kayıtlar kalır</li>
                        <li>Daha hızlı</li>
                        <li>Dikkat: Çift kayıt olabilir ⚠️</li>
                    </ul>
                </div>
            </div>
        </div>
        
        <div style='background: rgba(138, 43, 226, 0.2); padding: 20px; border-radius: 10px; margin-top: 20px; border: 2px solid #8a2be2;'>
            <h4 style='margin-top: 0;'>💡 PostgreSQL, MySQL, SQL Server'da:</h4>
            <div style='background: rgba(0,0,0,0.3); padding: 15px; border-radius: 8px; font-family: "Courier New", monospace;'>
                <span style='color: #ffd700;'>SELECT</span> *<br>
                <span style='color: #ffd700;'>FROM</span> tablo1<br>
                <span style='color: #ff69b4;'>FULL OUTER JOIN</span> tablo2<br>
                &nbsp;&nbsp;<span style='color: #ffd700;'>ON</span> tablo1.id = tablo2.foreign_key;<br>
            </div>
            <p style='margin: 15px 0 0 0; font-size: 0.95em; text-align: center;'>
                ✅ Doğrudan FULL OUTER JOIN desteklenir! (SQLite hariç)
            </p>
        </div>
    </div>
</div>
"""

display(HTML(explanation_html))

print("\n✅ SQLite'da FULL OUTER JOIN'i UNION ile simüle ettik!")
print("🎯 Her iki tablonun TÜM kayıtlarını gördük!")
````

## 🎯 ADIM 10: CROSS JOIN - Kartezyen Çarpım (Her Kombinasyon)

### 📚 CROSS JOIN Nedir?

**CROSS JOIN**, iki tablonun **her kaydını birbirleriyle eşleştirir**! Bu Kartezyen Çarpım (Cartesian Product) olarak bilinir.

**Formül:** Sonuç Sayısı = Sol Tablo × Sağ Tablo

```
Sol Tablo: 3 kayıt
Sağ Tablo: 4 kayıt
CROSS JOIN Sonucu: 3 × 4 = 12 kayıt! 🚀
```

### ⚠️ DİKKAT: CROSS JOIN Tehlikelidir!

> 1000 kayıt × 1000 kayıt = **1,000,000 kayıt!** 💥
> 
> Çok dikkatli kullanın, yoksa veritabanı kilitlenebilir!

### 🎯 Ne Zaman Kullanılır?

- Tüm kombinasyonları görmek istediğinizde
- Test verileri oluşturmak için
- Çok nadir! (günlük işte %1'den az)

### 🎯 Örnek: CROSS JOIN ile Tüm Kombinasyonlar

````python
# 🎯 CROSS JOIN Örneği - Her Departman × Her Proje

# Küçük bir örnek için sadece ilk 3 departman ve 3 projeyi kullanalım
query_cross = """
SELECT 
    d.department_name AS 'Departman',
    p.project_name AS 'Proje',
    '🔄 Her Kombinasyon' AS 'Tip'
FROM (SELECT * FROM departments LIMIT 3) d
CROSS JOIN (SELECT * FROM projects LIMIT 3) p
ORDER BY d.department_name, p.project_name;
"""

df_cross = run_query(query_cross)
print("🔄 CROSS JOIN Sonucu:")
print(f"📌 Kayıt Sayısı: {len(df_cross)}")
print(f"🧮 Hesaplama: 3 departman × 3 proje = {len(df_cross)} kayıt")
print()

styled_cross = df_cross.style.set_properties(**{
    'background-color': '#fff3cd',
    'color': '#856404',
    'border-color': '#ffeaa7',
    'text-align': 'center',
    'font-size': '14px'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#ff6348'),
        ('color', 'white'),
        ('font-weight', 'bold'),
        ('text-align', 'center'),
        ('padding', '12px')
    ]}
])

display(styled_cross)

# Açıklama
explanation_html = """
<div style='background: linear-gradient(135deg, #ff6348 0%, #ff9f43 100%); padding: 30px; border-radius: 15px; color: white; box-shadow: 0 10px 40px rgba(0,0,0,0.3); margin-top: 30px;'>
    <h2 style='text-align: center; margin-bottom: 25px; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);'>
        💡 CROSS JOIN - Kartezyen Çarpım
    </h2>
    
    <div style='background: rgba(255,255,255,0.1); padding: 25px; border-radius: 10px; backdrop-filter: blur(10px);'>
        <h3 style='margin-top: 0; border-bottom: 2px solid rgba(255,255,255,0.3); padding-bottom: 10px;'>
            🎯 Syntax
        </h3>
        
        <div style='background: rgba(0,0,0,0.3); padding: 20px; border-radius: 10px; margin: 20px 0; font-family: "Courier New", monospace;'>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>SELECT</span> *<br>
                <span style='color: #ffd700;'>FROM</span> tablo1<br>
                <span style='color: #ff69b4;'>CROSS JOIN</span> tablo2;<br>
            </div>
            <div style='margin: 20px 0; padding-top: 15px; border-top: 2px dashed rgba(255,255,255,0.3);'>
                <span style='color: #ffd700;'>-- Alternatif (eski stil):</span><br>
                <span style='color: #ffd700;'>SELECT</span> *<br>
                <span style='color: #ffd700;'>FROM</span> tablo1, tablo2;<br>
            </div>
        </div>
        
        <div style='background: rgba(255, 71, 87, 0.3); padding: 20px; border-radius: 10px; margin-top: 25px; border: 2px solid #ff4757;'>
            <h4 style='margin-top: 0;'>⚠️ TEHLİKE UYARISI!</h4>
            <div style='display: grid; grid-template-columns: repeat(2, 1fr); gap: 15px; margin-top: 15px;'>
                <div style='text-align: center;'>
                    <h5 style='margin: 10px 0;'>🟢 Güvenli</h5>
                    <p style='margin: 5px 0; font-size: 1.2em; font-weight: bold;'>10 × 10 = 100</p>
                    <p style='margin: 5px 0; font-size: 0.9em;'>Kabul edilebilir</p>
                </div>
                <div style='text-align: center;'>
                    <h5 style='margin: 10px 0;'>🔴 Tehlikeli</h5>
                    <p style='margin: 5px 0; font-size: 1.2em; font-weight: bold;'>1000 × 1000 = 1M</p>
                    <p style='margin: 5px 0; font-size: 0.9em;'>VERİTABANI KİTLENİR! 💥</p>
                </div>
            </div>
            <p style='margin: 15px 0 0 0; text-align: center; font-size: 1.05em;'>
                <strong>Mutlaka LIMIT kullanın veya WHERE ile filtreleyin!</strong>
            </p>
        </div>
        
        <div style='background: rgba(46, 213, 115, 0.2); padding: 20px; border-radius: 10px; margin-top: 20px; border: 2px solid #2ed573;'>
            <h4 style='margin-top: 0;'>✅ Kullanım Alanları</h4>
            <ul style='line-height: 1.8; columns: 2; column-gap: 20px;'>
                <li>Tüm kombinasyonları test etmek</li>
                <li>Takvim × Çalışan = Vardiya planı</li>
                <li>Ürün × Renk = Stok kombinasyonları</li>
                <li>Test verisi oluşturmak</li>
            </ul>
        </div>
    </div>
</div>
"""

display(HTML(explanation_html))

print("\n✅ CROSS JOIN: Her kombinasyonu gördük!")
print("⚠️ Gerçek projede çok dikkatli kullanın!")
````

## 🎯 ADIM 11: SELF JOIN - Tabloyu Kendisiyle Birleştirme

### 📚 SELF JOIN Nedir?

**SELF JOIN**, bir tablonun **kendisiyle join edilmesidir**! Aynı tablo, farklı alias'larla iki kez kullanılır.

```
employees tablosu:
├── employee_id
├── first_name
└── manager_id (→ employee_id'ye referans!)

SELF JOIN ile:
"Her çalışanın yöneticisinin adını getir"
```

### 🤔 Ne Zaman Kullanılır?

**Hiyerarşik İlişkiler:**
- Çalışan → Yönetici
- Kategori → Üst Kategori
- Şehir → Bağlı Olduğu İl
- Dosya → Üst Klasör

**Karşılaştırmalar:**
- Aynı şehirdeki çalışanlar
- Aynı maaş alan kişiler
- Benzer ürünler

### 🎯 Örnek 1: Aynı Departmandaki Çalışan Çiftleri (Karşılaştırma)

````python
# 🎯 SELF JOIN Örnek 1: Aynı Departmandaki Çalışanlar

query_self = """
SELECT 
    e1.first_name || ' ' || e1.last_name AS 'Çalışan 1',
    e2.first_name || ' ' || e2.last_name AS 'Çalışan 2',
    d.department_name AS 'Ortak Departman',
    e1.salary AS 'Maaş 1',
    e2.salary AS 'Maaş 2',
    ABS(e1.salary - e2.salary) AS 'Maaş Farkı'
FROM employees e1
INNER JOIN employees e2 
    ON e1.department_id = e2.department_id 
    AND e1.employee_id < e2.employee_id  -- Çift tekrarını önle
INNER JOIN departments d 
    ON e1.department_id = d.department_id
ORDER BY d.department_name, e1.first_name;
"""

df_self = run_query(query_self)
print("👥 SELF JOIN: Aynı Departmandaki Çalışan Çiftleri")
print(f"📌 Toplam Çift: {len(df_self)}")
print("🎯 Her departmandaki çalışanlar birbirleriyle eşleştirildi")
print()

# Maaş formatı
df_self_display = df_self.copy()
df_self_display['Maaş 1'] = df_self_display['Maaş 1'].apply(lambda x: f"{x:,.0f} TL")
df_self_display['Maaş 2'] = df_self_display['Maaş 2'].apply(lambda x: f"{x:,.0f} TL")
df_self_display['Maaş Farkı'] = df_self_display['Maaş Farkı'].apply(lambda x: f"{x:,.0f} TL")

styled_self = df_self_display.style.set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#2c3e50',
    'border-color': '#dee2e6',
    'text-align': 'center',
    'font-size': '13px'
}).set_table_styles([
    {'selector': 'thead th', 'props': [
        ('background-color', '#5e72e4'),
        ('color', 'white'),
        ('font-weight', 'bold'),
        ('text-align', 'center'),
        ('padding', '12px')
    ]}
])

display(styled_self)

# Açıklama
explanation_html = """
<div style='background: linear-gradient(135deg, #5e72e4 0%, #825ee4 100%); padding: 30px; border-radius: 15px; color: white; box-shadow: 0 10px 40px rgba(0,0,0,0.3); margin-top: 30px;'>
    <h2 style='text-align: center; margin-bottom: 25px; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);'>
        💡 SELF JOIN - Nasıl Çalışır?
    </h2>
    
    <div style='background: rgba(255,255,255,0.1); padding: 25px; border-radius: 10px; backdrop-filter: blur(10px);'>
        <h3 style='margin-top: 0; border-bottom: 2px solid rgba(255,255,255,0.3); padding-bottom: 10px;'>
            🎯 Adım Adım Açıklama
        </h3>
        
        <div style='background: rgba(0,0,0,0.3); padding: 20px; border-radius: 10px; margin: 20px 0; font-family: "Courier New", monospace;'>
            <div style='margin: 10px 0;'>
                <span style='color: #ffd700;'>FROM</span> employees <span style='color: #98fb98;'>e1</span>  <span style='color: #87ceeb;'>-- İlk çalışan</span><br>
                <span style='color: #ff69b4;'>INNER JOIN</span> employees <span style='color: #98fb98;'>e2</span>  <span style='color: #87ceeb;'>-- İkinci çalışan (AYNI TABLO!)</span><br>
                &nbsp;&nbsp;<span style='color: #ffd700;'>ON</span> e1.department_id = e2.department_id  <span style='color: #87ceeb;'>-- Aynı departman</span><br>
                &nbsp;&nbsp;<span style='color: #ffd700;'>AND</span> e1.employee_id <span style='color: #ff6b6b;'>&lt;</span> e2.employee_id  <span style='color: #87ceeb;'>-- Tekrar önleme!</span>
            </div>
        </div>
        
        <div style='display: grid; grid-template-columns: repeat(3, 1fr); gap: 15px; margin-top: 25px;'>
            <div style='background: rgba(46, 213, 115, 0.2); padding: 15px; border-radius: 10px; border: 2px solid #2ed573;'>
                <h4 style='margin: 0 0 10px 0; text-align: center;'>1️⃣ İki Alias</h4>
                <p style='margin: 0; font-size: 0.9em; text-align: center; line-height: 1.6;'>
                    <code style='background: rgba(0,0,0,0.3); padding: 5px 10px; border-radius: 5px;'>e1</code> ve 
                    <code style='background: rgba(0,0,0,0.3); padding: 5px 10px; border-radius: 5px;'>e2</code><br>
                    Aynı tablo, iki kez!
                </p>
            </div>
            
            <div style='background: rgba(255, 193, 7, 0.2); padding: 15px; border-radius: 10px; border: 2px solid #ffc107;'>
                <h4 style='margin: 0 0 10px 0; text-align: center;'>2️⃣ ON Şartı</h4>
                <p style='margin: 0; font-size: 0.9em; text-align: center; line-height: 1.6;'>
                    Aynı departman olanları eşleştir
                </p>
            </div>
            
            <div style='background: rgba(255, 71, 87, 0.2); padding: 15px; border-radius: 10px; border: 2px solid #ff4757;'>
                <h4 style='margin: 0 0 10px 0; text-align: center;'>3️⃣ Tekrar Önleme</h4>
                <p style='margin: 0; font-size: 0.9em; text-align: center; line-height: 1.6;'>
                    <code style='background: rgba(0,0,0,0.3); padding: 5px 10px; border-radius: 5px;'>e1.id &lt; e2.id</code><br>
                    (Ahmet-Mehmet) = (Mehmet-Ahmet) önlenir
                </p>
            </div>
        </div>
        
        <div style='background: rgba(255, 255, 255, 0.1); padding: 20px; border-radius: 10px; margin-top: 25px; border: 2px dashed rgba(255,255,255,0.3);'>
            <h4 style='margin-top: 0;'>🎓 Neden <code style='background: rgba(0,0,0,0.3); padding: 5px 10px; border-radius: 5px;'>e1.id &lt; e2.id</code> ?</h4>
            <div style='display: grid; grid-template-columns: 1fr 1fr; gap: 20px;'>
                <div>
                    <h5 style='margin: 10px 0; color: #ff4757;'>❌ KULLANMAZSAN:</h5>
                    <ul style='line-height: 1.8; margin: 5px 0; font-size: 0.95em;'>
                        <li>Ahmet - Mehmet ✅</li>
                        <li>Mehmet - Ahmet ❌ (Tekrar!)</li>
                        <li>Ahmet - Ahmet ❌ (Kendisiyle!)</li>
                    </ul>
                </div>
                <div>
                    <h5 style='margin: 10px 0; color: #2ed573;'>✅ KULLANIRSAN:</h5>
                    <ul style='line-height: 1.8; margin: 5px 0; font-size: 0.95em;'>
                        <li>Ahmet - Mehmet ✅</li>
                        <li>Her çift sadece 1 kez gelir</li>
                        <li>Kendisiyle eşleşme olmaz</li>
                    </ul>
                </div>
            </div>
        </div>
        
        <div style='background: rgba(138, 43, 226, 0.2); padding: 20px; border-radius: 10px; margin-top: 20px; border: 2px solid #8a2be2;'>
            <h4 style='margin-top: 0;'>🎯 SELF JOIN Kullanım Örnekleri</h4>
            <ul style='line-height: 1.8; columns: 2; column-gap: 20px;'>
                <li>Aynı departmandaki çalışanlar</li>
                <li>Aynı maaşı alanlar</li>
                <li>Benzer ürünler (fiyat aralığı)</li>
                <li>Kardeş kategoriler</li>
                <li>Yönetici-çalışan ilişkisi</li>
                <li>Arkadaş önerileri (sosyal ağ)</li>
            </ul>
        </div>
    </div>
</div>
"""

display(HTML(explanation_html))

print("\n✅ SELF JOIN ile aynı tablonun kayıtlarını birbirleriyle eşleştirdik!")
print("🎯 Bu teknik hiyerarşik ilişkilerde çok kullanışlı!")
````
