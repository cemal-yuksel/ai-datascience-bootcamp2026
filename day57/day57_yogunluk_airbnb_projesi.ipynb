### 🔍 3.2 · Şüpheli Değerlere Yakından Bakış

Altı kontrol yapıyoruz: (1) tam kopya satır, (2) **aynı ev sahibi + aynı koordinat + aynı oda tipi** (olası tekrar ilan), (3) **sıfır fiyat**, (4) bir yıldan uzun **en az gece**, (5) **yorumsuz ilanda eksik "aylık yorum"** (yapısal eksiklik), (6) koordinatın düştüğü ilçe poligonu ile **beyan edilen ilçenin** uyuşmaması.

```python
# İlçe poligonları (derece) ile nokta-poligon testi
sinir = json.loads(Path(HARITA_YOLU).read_text(encoding="utf-8"))["features"]
poligon = {f["properties"]["ilce"]: [MplPath(np.asarray(p[0])) for p in f["geometry"]["coordinates"]] for f in sinir}
nokta = ham[["Boylam", "Enlem"]].to_numpy()
bulunan = np.full(len(ham), "Sınır dışı/kıyı", dtype=object)
for ad, pl in poligon.items():
    icinde = np.zeros(len(ham), bool)
    for p in pl: icinde |= p.contains_points(nokta)
    bulunan[icinde] = ad
ham["Koordinattaki ilçe"] = bulunan

ayni_yer = ham.duplicated(subset=["Ev sahibi no", "Enlem", "Boylam", "Oda tipi"], keep=False)
supheli = pd.DataFrame([
    ("Tam kopya satır (ilan no hariç)", int(ham.drop(columns="İlan no").duplicated().sum())),
    ("Aynı ev sahibi + aynı koordinat + aynı oda tipi", int(ayni_yer.sum())),
    ("Fiyat = 0", int((ham["Fiyat"] == 0).sum())),
    ("Fiyat > 2.000 $", int((ham["Fiyat"] > 2000).sum())),
    ("En az gece > 365", int((ham["En az gece"] > 365).sum())),
    ("Yorum sayısı 0 → Aylık yorum eksik", int(((ham["Yorum sayısı"] == 0) & ham["Aylık yorum"].isna()).sum())),
    ("Yorum var ama Aylık yorum eksik", int(((ham["Yorum sayısı"] > 0) & ham["Aylık yorum"].isna()).sum())),
    ("Koordinat hiçbir ilçe poligonunda değil (kıyı/iskele)", int((bulunan == "Sınır dışı/kıyı").sum())),
    ("Koordinat başka ilçede (beyan ≠ poligon)", int(((bulunan != ham["İlçe"]) & (bulunan != "Sınır dışı/kıyı")).sum())),
], columns=["Kontrol", "Satır"])
display(supheli)
uyumsuz = ham[(bulunan != ham["İlçe"]) & (bulunan != "Sınır dışı/kıyı")]
display(pd.crosstab(uyumsuz["İlçe"], uyumsuz["Koordinattaki ilçe"]).rename_axis(index="Beyan", columns="Poligon"))
```

### 🧾 3.3 · Karar Defteri

```python
karar = pd.DataFrame([
    ("Tam kopya satır", "Yok", "İşlem gerekmedi"),
    ("Aynı ev sahibi + aynı nokta + aynı tip", f"{int(ayni_yer.sum())} ilan", "KORUNDU: ayrı ilanlar (aynı binada birden çok oda/daire); 14.3'te 'tek ev sahibi odakları' olarak izlenecek"),
    ("Fiyat = 0", f"{int((ham['Fiyat'] == 0).sum())} ilan", "Konum geçerli → kümelemede KORUNDU; fiyat analizlerinde eksik (NaN) sayıldı"),
    ("Fiyat > 2.000 $", f"{int((ham['Fiyat'] > 2000).sum())} ilan", "KORUNDU: lüks ilan olabilir; fiyat özetlerinde medyan ve log ölçek kullanıldı"),
    ("En az gece > 365", f"{int((ham['En az gece'] > 365).sum())} ilan", "KORUNDU: kümelemeye girmez; 'riskli ilan' tanımında zaten < 30 gece aranır"),
    ("Yorumsuz ilanda Aylık yorum eksik", f"{int(ham['Aylık yorum'].isna().sum())} ilan", "Yapısal eksik → 0 ile dolduruldu (yorum yoksa aylık yorum 0'dır)"),
    ("Beyan ≠ poligon ilçesi", f"{len(uyumsuz)} ilan", "Kümeleme yalnızca koordinatı kullanır → etkisiz; ilçe analizlerinde BEYAN edilen ilçe kullanıldı"),
], columns=["Durum", "Kapsam", "Karar"])
karar.to_csv(REP_DIR / "02_karar_defteri.csv", index=False); display(karar)

d = ham.copy()
d["Fiyat"] = d["Fiyat"].replace(0, np.nan)
d["Aylık yorum"] = d["Aylık yorum"].fillna(0)
# Riskli ilan (Kural 5): tüm ev, 30 geceden kısa kiralanabilir, yılda en az 90 gün müsait
d["Riskli"] = (d["Oda tipi"] == "Tüm ev") & (d["En az gece"] < 30) & (d["Yıllık müsaitlik"] >= 90)
d["Çoklu ilan sahibi"] = d["Ev sahibinin ilan sayısı"] >= 2
print("Analize giren ilan:", len(d), "| riskli ilan oranı:", yuzde(100 * d["Riskli"].mean()))
```

### 🌗 5.2 · Yoğunluk Gerçekten Ne Kadar Farklı?

Her ilçenin ilan yoğunluğunu (ilan / km²) ve bir ilanın **20. en yakın komşusuna** olan tipik uzaklığını ölçüyoruz. İkincisi, DBSCAN'in eps'i için "o ilçede kalabalık ne demek?" sorusunun cevabıdır.

```python
def poligon_alan_km2(ad):
    toplam = 0
    for p in [f for f in sinir if f["properties"]["ilce"] == ad][0]["geometry"]["coordinates"]:
        P = PROJ.transform(np.asarray(p[0])[:, ::-1]); x, y = P[:, 0], P[:, 1]
        toplam += .5 * abs(np.dot(x, np.roll(y, 1)) - np.dot(y, np.roll(x, 1)))
    return toplam
k20 = NearestNeighbors(n_neighbors=21).fit(Ztr).kneighbors(Ztr)[0][:, -1]
doku = pd.DataFrame({"İlan": tr["İlçe"].value_counts(), "Alan (km²)": pd.Series({a: poligon_alan_km2(a) for a in ILCE_SIRA})}).reindex(ILCE_SIRA)
doku["İlan / km²"] = doku["İlan"] / doku["Alan (km²)"]
doku["20. komşu medyan (m)"] = pd.Series(k20 * 1000).groupby(tr["İlçe"]).median().reindex(ILCE_SIRA)
doku = doku.round(1); doku.to_csv(REP_DIR / "03_yogunluk_dokusu.csv"); display(doku)

fig, ax = plt.subplots(1, 3, figsize=(19, 6), gridspec_kw={"width_ratios": [1.2, 1, 1]})
harita(ax[0], "Altıgen başına ilan (log)")
hb = ax[0].hexbin(*Ztr.T, gridsize=75, bins="log", cmap="magma_r", mincnt=1, zorder=3)
fig.colorbar(hb, ax=ax[0], shrink=.6)
doku["İlan / km²"].plot.bar(ax=ax[1], color=[ILCE_RENK[i] for i in ILCE_SIRA], rot=0, logy=True)
ax[1].set(title="İlan yoğunluğu (ilan / km², log)", xlabel="")
for i, ilce in enumerate(ILCE_SIRA):
    v = np.sort(k20[tr["İlçe"].to_numpy() == ilce] * 1000)
    ax[2].plot(np.linspace(0, 100, len(v)), v, c=ILCE_RENK[ilce], lw=2.2, label=ilce)
ax[2].set(yscale="log", xlabel="ilanların yüzdesi (sıralı)", ylabel="20. komşuya uzaklık (m, log)", title="İlçe ilçe k-mesafe eğrileri")
ax[2].legend(); kaydet(fig, "02_yogunluk_dokusu"); plt.show()
```

### 📏 6.1 · Aynı HDBSCAN, Dört Farklı Uzay

Aynı algoritmayı (HDBSCAN · yaprak, `min_cluster_size = 100`, `min_samples = 20`) dört farklı özellik uzayında çalıştırıyoruz. Amaç "hangi uzay daha yüksek skor verir?" değil, **hangi uzayın iş sorusuna uygun olduğu** ve seçimin sonucu ne kadar değiştirdiği.

| Uzay | Ne içerir? |
|:--|:--|
| **Kilometre** | x_km, y_km (fiziksel mesafe) |
| **Derece** | enlem, boylam (1° boylam ≠ 1° enlem) |
| **z-skor** | standartlaştırılmış km (iki eksen ayrı ölçeklenir) |
| **km + fiyat** | x_km, y_km, log(fiyat) × 1 km (fiyatı 1 km'lik bir eksen gibi tartar) |

```python
log_fiyat = np.log(tr["Fiyat"].fillna(tr["Fiyat"].median())).to_numpy()
uzay = {"Kilometre": Ztr, "Derece": tr[["Boylam", "Enlem"]].to_numpy(), "z-skor": StandardScaler().fit_transform(Ztr),
        "km + fiyat": np.c_[Ztr, (log_fiyat - log_fiyat.mean()) / log_fiyat.std()]}
uzay_et, uzay_satir = {}, []
for ad, U in uzay.items():
    t0 = time.time(); et = HDBSCAN(min_cluster_size=100, min_samples=20, cluster_selection_method="leaf", copy=True).fit_predict(U)
    uzay_et[ad] = et
    uzay_satir.append({"Uzay": ad, **kume_ozeti(et), "DBCV (kendi uzayında)": dbcv_skoru(U[DEG], et[DEG]), "Süre (sn)": time.time() - t0})
uzay_tab = pd.DataFrame(uzay_satir).set_index("Uzay")
uzay_tab["ARI (km'ye göre)"] = [adjusted_rand_score(uzay_et["Kilometre"], uzay_et[a]) for a in uzay_tab.index]
uzay_tab = uzay_tab.round(3); uzay_tab.to_csv(REP_DIR / "04_uzay_deneyi.csv"); display(uzay_tab)

fig, ax = plt.subplots(1, 4, figsize=(22, 6.2))
for a, (ad, et) in zip(ax, uzay_et.items()):
    harita(a); a.scatter(*Ztr.T, s=.6, c=renkler(et), lw=0, zorder=3)
    a.set_title(f"{ad}\n{kume_ozeti(et)['Odak sayısı']} odak · gürültü {yuzde(kume_ozeti(et)['Gürültü %'], 0)} · ARI(km) {uzay_tab.loc[ad, 'ARI (km\'ye göre)']:.2f}")
fig.suptitle("Aynı algoritma, dört uzay: odaklar her zaman km haritasında çizildi", fontweight="bold")
kaydet(fig, "04_uzay_deneyi"); plt.show()
```

### 🏎️ 8.2 · Yarışı Koşalım

```python
def etiket_al(model, Z):
    m = model.fit(Z)
    return np.asarray(m.labels_ if hasattr(m, "labels_") else m.predict(Z))

def degerlendir(ad, aile, et, sure, strateji="Tam veri"):
    o = kume_ozeti(et); e = et[DEG]; mk = e >= 0
    sil = silhouette_score(Ztr[DEG][mk], e[mk]) if len(set(e[mk])) > 1 else np.nan
    ki = et >= 0
    ami = adjusted_mutual_info_score(tr["Mahalle"][ki], et[ki]) if ki.sum() > 0 and len(set(et[ki])) > 1 else np.nan
    ihlal = [k for k, v in [("K1", o["Gürültü %"] > 50), ("K2", o["En büyük odak %"] > 15), ("K3", not 10 <= o["Odak sayısı"] <= 150)] if v]
    return {"Algoritma": ad, "Aile": aile, "Strateji": strateji, **o, "DBCV": dbcv_skoru(Ztr[DEG], e), "Silüet": sil,
            "Mahalle AMI": ami, "Süre (sn)": sure, "Uygun": not ihlal, "İhlal": ", ".join(ihlal) or "—"}

ETIKET, satir = {}, []
for ad, (aile, model, strateji) in YARISCILAR.items():
    t0 = time.time(); et = etiket_al(clone(model), Ztr); sure = time.time() - t0
    ETIKET[ad] = et; satir.append(degerlendir(ad, aile, et, sure, strateji))
    print(f"{ad:<28} {sure:6.1f} sn · {satir[-1]['Odak sayısı']:>4} odak · gürültü %{satir[-1]['Gürültü %']:5.1f} · DBCV {satir[-1]['DBCV']:+.3f}")
yaris = (pd.DataFrame(satir).set_index("Algoritma").sort_values(["Uygun", "DBCV"], ascending=[False, False]))
yaris.round(3).to_csv(REP_DIR / "06_buyuk_yaris.csv")
display(yaris.round(3))
```

### 🔁 9.1 · Altı Aday, Sekiz Alt Örneklem

Adaylar: yarışta kurallara uyan **ilk 3 algoritma** + DBCV'si en yüksek **3 yoğunluk algoritması** (kural ihlali parametreden kaynaklanabilir; onlara 10. bölümde ayar şansı verilecek). Her aday eğitim ilanlarının rastgele **%90'ı** üzerinde 8 kez yeniden kurulur ve ortak ilanlarda tam eğitimdeki etiketlerle **ARI** hesaplanır. Yoğunluk eşikleri (minPts, min_cluster_size, …) alt örneklem oranında küçültülür ki "kalabalık" tanımı değişmesin.

```python
SAYIM_PARAM = ["min_samples", "min_cluster_size", "esik", "en_az", "gurultu_rho"]
def olcekle(model, f):
    """Yoğunluk 'sayım' parametrelerini alt örneklem oranında küçültür (ör. minPts 20 → 18)."""
    m = clone(model); p = m.get_params(deep=False)
    yeni = {k: max(2, int(round(p[k] * f))) for k in SAYIM_PARAM if k in p and isinstance(p[k], (int, np.integer))}
    if "min_cluster_size" in yeni and isinstance(m, OPTICS): yeni.pop("min_cluster_size")
    return m.set_params(**yeni)

def kararlilik(model, et_tam, B=8, oran=.9, tohum=0):
    rng = np.random.default_rng(tohum); ari, k = [], []
    for b in range(B):
        idx = np.sort(rng.choice(len(Ztr), int(oran * len(Ztr)), replace=False))
        et = etiket_al(olcekle(model, oran), Ztr[idx])
        ari.append(adjusted_rand_score(et_tam[idx], et)); k.append(kume_ozeti(et)["Odak sayısı"])
    return np.array(ari), np.array(k)

uygun_ilk3 = [a for a in yaris.index if yaris.loc[a, "Uygun"]][:3]
yog_ilk3 = [a for a in yaris[yaris["Aile"] == "Yoğunluk"].sort_values("DBCV", ascending=False).index if a not in uygun_ilk3][:3]
ADAY = uygun_ilk3 + yog_ilk3
kar, kar_dag = [], {}
for ad in ADAY:
    t0 = time.time(); a_, k_ = kararlilik(YARISCILAR[ad][1], ETIKET[ad]); kar_dag[ad] = a_
    kar.append({"Algoritma": ad, "Aile": YARISCILAR[ad][0], "Uygun (yarış)": yaris.loc[ad, "Uygun"], "ARI ort.": a_.mean(), "ARI std": a_.std(),
                "Odak sayısı ort.": k_.mean(), "Odak sayısı std": k_.std(), "Süre (sn)": time.time() - t0})
    print(f"{ad:<26} ARI {a_.mean():.3f} ± {a_.std():.3f}")
kar = pd.DataFrame(kar).set_index("Algoritma").round(3); kar.to_csv(REP_DIR / "07_kararlilik.csv"); display(kar)
fig, ax = plt.subplots(figsize=(12, 5))
ax.boxplot([kar_dag[a] for a in ADAY], tick_labels=ADAY, patch_artist=True,
           boxprops=dict(facecolor=RENK["altin"], alpha=.6)); ax.set(ylabel="ARI (alt örneklem ↔ tam eğitim)", title="Kararlılık: 8 alt örneklem")
ax.tick_params(axis="x", rotation=15); kaydet(fig, "11_kararlilik"); plt.show()
```

Ayar ızgarası **yalnızca eğitimde** ve **aynı ölçütlerle** değerlendirilir. Dört aday ailesi:

### 🧭 11.1 · Final Boru Hattı ve Odak Profilleri

Final boru hattı: `KmProjeksiyon → YogunlukAtayici(final kümeleyici)`.

**Önemli bir uygulama ayrıntısı:** scikit-learn'deki `HDBSCAN` yeni gelen noktayı tahmin edemez. Algoritmanın özgün geliştiricilerinin `hdbscan` kütüphanesi ise aynı algoritmaya ek olarak **`approximate_predict`** sunar: yeni ilan, eğitimde kurulan yoğunluk ağacında hangi odağın "yaşadığı" seviyeye düşüyorsa o odağa katılır, düşmüyorsa gürültü olur (McInnes ve ark., 2017). Final boru hattında bu yüzden **aynı ayarlarla** özgün kütüphaneyi kullanıyoruz (`ozgun_hdbscan`). İki uygulama eşit yoğunluklu noktalarda farklı eşitlik bozma (tie-break) kuralları kullandığı için etiketler birebir aynı değildir; aşağıda örtüşmeyi ölçüyor ve **kuralları final model için yeniden denetliyoruz**.

Her odağın profili **eğitim** ilanlarıyla çıkarılır. Alan, odağa ait ilanların **dışbükey zarfının** (convex hull) yüzölçümüdür.

**Adlandırma kuralı:** odak adı = odaktaki **en kalabalık mahalle** (aynı mahalle birden çok odakta baskınsa yanına sıra numarası). **Karakter** etiketi önceden belirlenmiş eşiklerle verilir (odaklar arası dağılıma göre):

| Karakter | Kural |
|:--|:--|
| 💎 Turist vitrini | medyan fiyat ≥ odakların %75'lik dilimi **ve** tüm ev payı ≥ %60 |
| 🏢 Ticari yoğunlaşma | çoklu ilan sahibi payı ≥ odakların %75'lik dilimi |
| 🛏️ Oda kiralayan yerel | tüm ev payı < %40 |
| 🌆 Karma konut | diğerleri |

```python
FINAL_OZGUN = ozgun_hdbscan(FINAL_KUMELEYICI)                       # HDBSCAN ise özgün kütüphane (approximate_predict)
FINAL = Pipeline([("km", KmProjeksiyon()), ("odak", YogunlukAtayici(clone(FINAL_OZGUN)))]).fit(tr[KONUM])
seg_tr = FINAL["odak"].labels_
assert np.allclose(FINAL["km"].transform(tr[KONUM]), Ztr)            # aynı merkez → aynı ızgara
o_fin = kume_ozeti(seg_tr)
print("Final uygulama:", type(FINAL["odak"].model_).__module__.split(".")[0], "·", FINAL_SATIR["Ayar"])
print(f"Yarıştaki (scikit-learn) etiketlerle örtüşme ARI = {adjusted_rand_score(etiket_al(clone(FINAL_KUMELEYICI), Ztr), seg_tr):.3f}")
print(f"Final model: {o_fin['Odak sayısı']} odak · gürültü {yuzde(o_fin['Gürültü %'])} · en büyük {yuzde(o_fin['En büyük odak %'])} · "
      f"DBCV {dbcv_skoru(Ztr[DEG], seg_tr[DEG]):+.3f} · Kural 1–3: {'SAĞLANIYOR' if uygun_mu(o_fin) else 'İHLAL'}")

def alan_km2(P):
    return ConvexHull(P).volume if len(P) >= 3 else 0.0
p = tr.assign(Odak=seg_tr)
odaklar = sorted(set(seg_tr) - {-1})
profil = pd.DataFrame({
    "İlan": p[p["Odak"] >= 0].groupby("Odak").size(),
    "Alan (km²)": pd.Series({k: alan_km2(Ztr[seg_tr == k]) for k in odaklar}),
    "Medyan fiyat": p.groupby("Odak")["Fiyat"].median(),
    "Tüm ev %": p.groupby("Odak")["Oda tipi"].apply(lambda s: 100 * (s == "Tüm ev").mean()),
    "Riskli ilan": p.groupby("Odak")["Riskli"].sum(),
    "Çoklu ilan sahibi %": p.groupby("Odak")["Çoklu ilan sahibi"].mean() * 100,
    "Aylık yorum (medyan)": p.groupby("Odak")["Aylık yorum"].median(),
    "En büyük ev sahibi payı %": p.groupby("Odak")["Ev sahibi no"].apply(lambda s: 100 * s.value_counts(normalize=True).iloc[0]),
    "Baskın ilçe": p.groupby("Odak")["İlçe"].agg(lambda s: s.value_counts().index[0]),
    "Baskın mahalle": p.groupby("Odak")["Mahalle"].agg(lambda s: s.value_counts().index[0]),
    "Baskın mahalle %": p.groupby("Odak")["Mahalle"].agg(lambda s: 100 * s.value_counts(normalize=True).iloc[0]),
}).loc[odaklar]
profil["İlan %"] = 100 * profil["İlan"] / len(tr)
profil["İlan / km²"] = profil["İlan"] / profil["Alan (km²)"].clip(lower=.01)
profil["Riskli / km²"] = profil["Riskli ilan"] / profil["Alan (km²)"].clip(lower=.01)
q_fiyat, q_coklu = profil["Medyan fiyat"].quantile(.75), profil["Çoklu ilan sahibi %"].quantile(.75)
def karakter(r):
    if r["Medyan fiyat"] >= q_fiyat and r["Tüm ev %"] >= 60: return "💎 Turist vitrini"
    if r["Çoklu ilan sahibi %"] >= q_coklu: return "🏢 Ticari yoğunlaşma"
    if r["Tüm ev %"] < 40: return "🛏️ Oda kiralayan yerel"
    return "🌆 Karma konut"
profil["Karakter"] = profil.apply(karakter, axis=1)
profil = profil.sort_values("İlan", ascending=False)
sayac, ad = {}, {}
for k, r in profil.iterrows():
    m = r["Baskın mahalle"]; sayac[m] = sayac.get(m, 0) + 1
    ad[k] = m if sayac[m] == 1 else f"{m} ({sayac[m]})"
profil.insert(0, "Odak adı", pd.Series(ad))
ODAK_AD = {int(k): v for k, v in ad.items()}
profil.round(2).to_csv(REP_DIR / "10_odak_profili.csv")
print(f"{len(odaklar)} odak · eğitim ilanlarının {yuzde(100 * (seg_tr >= 0).mean())} kadarı bir odakta")
display(profil.head(15).round(1))
display(profil["Karakter"].value_counts().rename("Odak sayısı").to_frame())
```

### 🏆 12.1 · Test Seti Açılıyor (Tek Sefer)

Kilitli 9.779 ilan, eğitimde öğrenilen boru hattından geçer: kilometre ızgarası (eğitim merkezi) → **HDBSCAN `approximate_predict`**. Dört soru soruyoruz:

1. Gürültü oranı ve odak payları eğitimdekine benziyor mu?
2. Test ilanlarındaki odak yapısı yoğunluk açısından geçerli mi (DBCV)?
3. **Tutarlılık:** Eğitim + test birlikte sıfırdan kümelenseydi (yoğunluk eşikleri veri büyüdüğü oranda büyütülerek), test ilanları aynı odaklara mı düşerdi?
4. **Dış doğrulama:** Odaklar yalnızca konumdan bulundu. Fiyatı, ilçeden ve mahalleden **daha iyi** açıklıyorlar mı?

```python
seg_te = FINAL.predict(te[KONUM])
o_tr, o_te = kume_ozeti(seg_tr), kume_ozeti(seg_te)
pay = pd.DataFrame({"Eğitim %": pd.Series(seg_tr[seg_tr >= 0]).value_counts(normalize=True) * 100,
                    "Test %": pd.Series(seg_te[seg_te >= 0]).value_counts(normalize=True) * 100}).fillna(0)
# Tutarlılık: eğitim+test birlikte, eşikler 1/0,8 = 1,25 kat büyütülerek sıfırdan kümelenir
tum_Z = np.r_[Ztr, Zte]
et_tum = etiket_al(olcekle(FINAL_OZGUN, len(tum_Z) / len(Ztr)), tum_Z)
et_tum_te, et_tum_tr = et_tum[len(Ztr):], et_tum[:len(Ztr)]
deg_te = np.random.default_rng(RS).choice(len(Zte), min(6000, len(Zte)), replace=False)
test_olc = pd.DataFrame({
    "Eğitim": [o_tr["Odak sayısı"], o_tr["Gürültü %"], o_tr["En büyük odak %"], dbcv_skoru(Ztr[DEG], seg_tr[DEG]), np.nan, np.nan],
    "Test": [len(set(seg_te) - {-1}), o_te["Gürültü %"], o_te["En büyük odak %"], dbcv_skoru(Zte[deg_te], seg_te[deg_te]),
             adjusted_rand_score(et_tum_te, seg_te), 100 * ((seg_te >= 0) == (et_tum_te >= 0)).mean()],
}, index=["Odak sayısı (görülen)", "Gürültü %", "En büyük odak %", "DBCV", "ARI (atama ↔ birlikte kümeleme)", "Odakta mı? uyumu %"])
test_olc.round(3).to_csv(REP_DIR / "11_test_olcutleri.csv"); display(test_olc.round(3))
print("Odak payları korelasyonu (eğitim ↔ test):", round(pay.corr().iloc[0, 1], 3))
print("Eğitim etiketleri ↔ birlikte kümeleme (eğitim ilanları) ARI:", round(adjusted_rand_score(seg_tr, et_tum_tr), 3))
```

```python
# Dış doğrulama: eğitimde öğrenilen grup ortalaması ile test ilanlarının log(fiyat)'ını tahmin et (R², test)
def grup_r2(grup_tr, grup_te):
    y_tr = np.log(tr["Fiyat"]); y_te = np.log(te["Fiyat"]); ort = y_tr.groupby(grup_tr).mean()
    tahmin = pd.Series(grup_te).map(ort).fillna(y_tr.mean()).to_numpy()
    m = y_te.notna().to_numpy()
    return 1 - np.sum((y_te[m] - tahmin[m]) ** 2) / np.sum((y_te[m] - y_te[m].mean()) ** 2)
odak_grup_tr = np.where(seg_tr >= 0, seg_tr, -1 - tr["İlçe"].map({a: i for i, a in enumerate(ILCE_SIRA)}).to_numpy())
odak_grup_te = np.where(seg_te >= 0, seg_te, -1 - te["İlçe"].map({a: i for i, a in enumerate(ILCE_SIRA)}).to_numpy())
dis = pd.DataFrame({"Grup sayısı": [5, tr["Mahalle"].nunique(), len(odaklar) + 5, K_OFIS],
                    "Test R² (log fiyat)": [grup_r2(tr["İlçe"].to_numpy(), te["İlçe"].to_numpy()),
                                            grup_r2(tr["Mahalle"].to_numpy(), te["Mahalle"].to_numpy()),
                                            grup_r2(odak_grup_tr, odak_grup_te),
                                            grup_r2(ETIKET["K-Means"], YARISCILAR["K-Means"][1].fit(Ztr).predict(Zte))]},
                   index=["İlçe (5)", f"Mahalle ({tr['Mahalle'].nunique()})", "Odak + ilçe bazında dağınık", "K-Means bölgeleri (k = 50)"]).round(4)
dis.to_csv(REP_DIR / "12_dis_dogrulama_fiyat.csv"); display(dis)

fig, ax = plt.subplots(1, 3, figsize=(21, 6.5), gridspec_kw={"width_ratios": [1.1, 1, 1]})
harita(ax[0], f"Test ilanları: {yuzde(100 - o_te['Gürültü %'])} bir odağa atandı")
ax[0].scatter(*Zte[seg_te < 0].T, s=1, c=GURULTU, lw=0, zorder=2); ax[0].scatter(*Zte[seg_te >= 0].T, s=2.5, c=renkler(seg_te[seg_te >= 0]), lw=0, zorder=3)
ax[1].scatter(pay["Eğitim %"], pay["Test %"], s=40, c=RENK["mavi"], edgecolor="k"); mx = pay.max().max() * 1.05
ax[1].plot([0, mx], [0, mx], ls="--", c="k"); ax[1].set(xlabel="Eğitimde odak payı %", ylabel="Testte odak payı %", title="Odak payları: eğitim ↔ test")
dis["Test R² (log fiyat)"].plot.barh(ax=ax[2], color=[RENK["gri"], RENK["mavi"], RENK["altin"], RENK["cyan"]]); ax[2].set(title="Dış doğrulama: fiyatı ne açıklıyor? (test R²)", xlabel="R²")
kaydet(fig, "15_test_seti"); plt.show()
```

### 🕵️ 13.1 · Alan Bütçesiyle Hedefleme (Kural 5)
<p align="center">
  <img src="assets/diyagram_denetim_karari.svg" width="100%" alt="Odaktan karara">

Ekiplerin taradığı alan sınırlı. Üç strateji, **aynı alan bütçesiyle** (km²) karşılaştırılır:

- **Odak tabanlı:** odaklar eğitimdeki **riskli ilan / km²** değerine göre sıralanır, bütçe dolana kadar seçilir.
- **Mahalle tabanlı:** 221 mahalle aynı ölçüte göre sıralanır (mahalle alanı = eğitim ilanlarının dışbükey zarfı; adil karşılaştırma için odaklarla aynı ölçüm).
- **Rastgele mahalle:** 300 rastgele sıralamanın ortalaması (taban).

Sıralama **eğitimde** yapılır; başarı, **test** setindeki riskli ilanların yüzde kaçının seçilen bölgelere düştüğüyle ölçülür.

```python
oncelik = profil[["Odak adı", "Karakter", "İlan", "Alan (km²)", "Riskli ilan", "Riskli / km²"]].sort_values("Riskli / km²", ascending=False)
oncelik.round(2).to_csv(REP_DIR / "13_denetim_oncelik.csv"); display(oncelik.head(12).round(2))

mah_alan = {m: alan_km2(Ztr[(tr["Mahalle"] == m).to_numpy()]) for m in tr["Mahalle"].unique()}
mah_tab = pd.DataFrame({"Alan (km²)": pd.Series(mah_alan), "Riskli ilan": tr.groupby("Mahalle")["Riskli"].sum()})
mah_tab = mah_tab[mah_tab["Alan (km²)"] > 0]
mah_tab["Riskli / km²"] = mah_tab["Riskli ilan"] / mah_tab["Alan (km²)"].clip(lower=.01)

riskli_te = te["Riskli"].to_numpy(); n_riskli = riskli_te.sum()
def egri(sirali_birim, alanlar, uyelik_te):
    """Bütçe arttıkça yakalanan test riskli ilan oranı."""
    birikim, yakalanan, sonuc = 0, set(), [(0, 0)]
    for b in sirali_birim:
        birikim += alanlar[b]; yakalanan.add(b)
        sonuc.append((birikim, 100 * (riskli_te & np.isin(uyelik_te, list(yakalanan))).sum() / n_riskli))
    return np.array(sonuc)
e_odak = egri(oncelik.index, profil["Alan (km²)"], seg_te)
e_mah = egri(mah_tab.sort_values("Riskli / km²", ascending=False).index, mah_tab["Alan (km²)"], te["Mahalle"].to_numpy())
BUTCE = np.linspace(0, 60, 121)
def ara(e): return np.interp(BUTCE, e[:, 0], e[:, 1], right=e[-1, 1])
rng = np.random.default_rng(RS)
e_rast = np.mean([ara(egri(rng.permutation(mah_tab.index), mah_tab["Alan (km²)"], te["Mahalle"].to_numpy())) for _ in range(300)], axis=0)
butce_tab = pd.DataFrame({"Bütçe (km²)": BUTCE, "Odak tabanlı %": ara(e_odak), "Mahalle tabanlı %": ara(e_mah), "Rastgele mahalle %": e_rast})
sec = butce_tab[butce_tab["Bütçe (km²)"].isin([5, 10, 15, 20, 30, 45, 60])].round(1).set_index("Bütçe (km²)")
sec.to_csv(REP_DIR / "14_butce_senaryosu.csv"); display(sec)
```

### 🧪 14.2 · Parametre Duyarlılığı

Final ayarın düğmelerini ±%40 oynatıp odak haritasının ne kadar değiştiğine bakıyoruz (ARI, finale göre).

```python
fp = FINAL_OZGUN.get_params(); duy = []
dugmeler = [k for k in ["eps", "min_samples", "min_cluster_size"] if k in fp and fp[k] is not None]
for k in dugmeler:
    for f in [.6, .8, 1.2, 1.4]:
        v = fp[k] * f if k == "eps" else max(2, int(round(fp[k] * f)))
        et = etiket_al(clone(FINAL_OZGUN).set_params(**{k: v}), Ztr); o = kume_ozeti(et)
        duy.append({"Düğme": k, "Çarpan": f, "Değer": round(v, 3), **o, "ARI (finale göre)": adjusted_rand_score(seg_tr, et), "Uygun": uygun_mu(o)})
duy = pd.DataFrame(duy).round(3); duy.to_csv(REP_DIR / "16_parametre_duyarliligi.csv", index=False); display(duy)
```
