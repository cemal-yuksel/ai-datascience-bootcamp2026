```python
# ── Kütüphaneler ─────────────────────────────────────────────
import hashlib
import json
import os
import platform
import subprocess
import sys
import time
import warnings
from pathlib import Path

os.environ["PYTHONWARNINGS"] = "ignore"                     # paralel çalışan süreçlerde de uyarı gürültüsünü kapatır

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
import seaborn as sns
import sklearn
import xgboost
import lightgbm
import catboost
from IPython.display import Image, Markdown, display
from joblib import Parallel, delayed
from matplotlib.animation import FuncAnimation, PillowWriter
from scipy.stats import loguniform, randint, uniform

from sklearn.base import clone
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor, make_column_selector
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import (AdaBoostRegressor, BaggingRegressor, ExtraTreesRegressor, GradientBoostingRegressor,
                              HistGradientBoostingRegressor, RandomForestRegressor, StackingRegressor, VotingRegressor)
from sklearn.feature_selection import mutual_info_regression
from sklearn.impute import SimpleImputer
from sklearn.inspection import permutation_importance
from sklearn.kernel_approximation import Nystroem
from sklearn.linear_model import (BayesianRidge, ElasticNet, GammaRegressor, HuberRegressor, Lasso, LinearRegression,
                                  Ridge, RidgeCV, SGDRegressor)
from sklearn.metrics import mean_absolute_error, r2_score, root_mean_squared_error
from sklearn.model_selection import (GroupKFold, KFold, RandomizedSearchCV, RepeatedKFold, cross_val_predict,
                                     train_test_split)
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (FunctionTransformer, OneHotEncoder, PolynomialFeatures, QuantileTransformer,
                                   RobustScaler, StandardScaler)
from sklearn.svm import LinearSVR, SVR
from sklearn.tree import DecisionTreeRegressor, ExtraTreeRegressor
from xgboost import DMatrix, XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor

from konut_yardimci import SEHIRLER, KonutOzellikleri, isaretli_log, uzaklik_km      # pipeline içindeki özellik mühendisliği (ayrı dosyada)

warnings.filterwarnings("ignore")

# ── Sabitler ve klasörler ───────────────────────────────────
RANDOM_STATE = 42
N_ISCI = 4                                                  # paralel çalışacak süreç sayısı
VERI_YOLU = Path("21-housing.csv")
HEDEF = "median_house_value"
FIG_DIR, RAPOR_DIR, MODEL_DIR = Path("outputs/figures"), Path("outputs/reports"), Path("outputs/model")
for klasor in (FIG_DIR, RAPOR_DIR, MODEL_DIR):
    klasor.mkdir(parents=True, exist_ok=True)

# ── Görsel tema (notebook boyunca aynı renkler) ──────────────
RENK = {"altin": "#EDAE49", "mor": "#6C5CE7", "pembe": "#D1495B", "mavi": "#2B6CB0",
        "yesil": "#2E9E6B", "turuncu": "#E8833A", "gri": "#8A94A6", "koyu": "#2E4057", "cyan": "#0E9AA7"}
AILE_RENK = {"Taban": "#B8C0CC", "Doğrusal": RENK["altin"], "Komşuluk ve çekirdek": RENK["pembe"], "Sinir ağı": RENK["mavi"],
             "Ağaç ve orman": RENK["yesil"], "Boosting": RENK["turuncu"], "Topluluk": RENK["mor"]}

plt.rcParams.update({
    "figure.dpi": 100, "savefig.dpi": 150, "savefig.bbox": "tight",
    "axes.titleweight": "bold", "axes.titlesize": 13, "axes.grid": True, "grid.alpha": 0.25,
    "axes.spines.top": False, "axes.spines.right": False,
})
sns.set_theme(style="whitegrid", rc={"axes.spines.top": False, "axes.spines.right": False, "grid.alpha": 0.25})


def kaydet(fig, ad):
    """Grafiği outputs/figures altına PNG olarak kaydeder."""
    yol = FIG_DIR / f"{ad}.png"
    fig.savefig(yol)
    print(f"💾 Kaydedildi → {yol.as_posix()}")


def rapor_kaydet(df, ad, index=True):
    """Tabloyu outputs/reports altına CSV olarak kaydeder."""
    yol = RAPOR_DIR / f"{ad}.csv"
    df.to_csv(yol, index=index, encoding="utf-8")
    print(f"💾 Rapor → {yol.as_posix()}  ({len(df)} satır)")


def gif_kaydet(kare_ciz, n_kare, yol, figsize=(8, 4.5), fps=2, dpi=70):
    """kare_ciz(fig, i) fonksiyonuyla her kareyi çizip animasyonlu GIF olarak kaydeder."""
    fig = plt.figure(figsize=figsize)
    anim = FuncAnimation(fig, lambda i: kare_ciz(fig, i), frames=n_kare)
    anim.save(yol, writer=PillowWriter(fps=fps), dpi=dpi)
    plt.close(fig)
    print(f"💾 Animasyon → {Path(yol).as_posix()} ({Path(yol).stat().st_size / 1024:.0f} KB)")


def usd(x, ondalik=0):
    """Sayıyı Türkçe biçimde dolar olarak yazar: 45954 → '45.954 $'."""
    return f"{x:,.{ondalik}f}".replace(",", "X").replace(".", ",").replace("X", ".") + " $"


def rmse(y, t):
    """Kök ortalama kare hata (dolar): tahminler gerçekten ortalama kaç dolar uzak? Büyük hataları daha çok cezalandırır."""
    return float(root_mean_squared_error(y, t))


def olcutler(y, t):
    """Regresyon ölçütleri: RMSE ve MAE (dolar), R², MAPE ve medyan yüzde hata (%)."""
    y, t = np.asarray(y, float), np.asarray(t, float)
    ape = np.abs(y - t) / y
    return {"RMSE": rmse(y, t), "MAE": float(mean_absolute_error(y, t)), "R²": float(r2_score(y, t)),
            "MAPE %": float(100 * ape.mean()), "Medyan APE %": float(100 * np.median(ape))}


def on_isleme(olcek="yok", doldur=True):
    """Sayısal sütunlar (isteğe bağlı doldurma + ölçekleme) ve kategorik sütun (one-hot) için ColumnTransformer."""
    adimlar = [("doldur", SimpleImputer(strategy="median"))] if doldur else []
    if olcek == "standart":
        adimlar.append(("olcek", StandardScaler()))
    elif olcek == "saglam":
        adimlar.append(("olcek", RobustScaler()))
    elif olcek == "log":
        adimlar += [("log", FunctionTransformer(isaretli_log, feature_names_out="one-to-one")), ("olcek", StandardScaler())]
    elif olcek == "kuantil":
        adimlar.append(("olcek", QuantileTransformer(output_distribution="normal", n_quantiles=200, random_state=RANDOM_STATE)))
    sayisal = Pipeline(adimlar) if adimlar else "passthrough"
    return ColumnTransformer([("sayisal", sayisal, make_column_selector(dtype_include=np.number)),
                              ("kategori", OneHotEncoder(handle_unknown="ignore"), ["ocean_proximity"])])


def pipe_kur(model, olcek="yok", hedef_olcek=False, ozellik="tam", doldur=True):
    """Özellik üretimi → ön işleme → model. hedef_olcek=True ise hedef de standartlaştırılır (SVR, MLP, SGD gibi modeller için)."""
    p = Pipeline([("ozellik", KonutOzellikleri(ozellik)), ("on", on_isleme(olcek, doldur)), ("model", model)])
    return TransformedTargetRegressor(regressor=p, transformer=StandardScaler()) if hedef_olcek else p


print(f"Python {platform.python_version()} · scikit-learn {sklearn.__version__} · pandas {pd.__version__} · numpy {np.__version__}")
print(f"XGBoost {xgboost.__version__} · LightGBM {lightgbm.__version__} · CatBoost {catboost.__version__} · SciPy {scipy.__version__}")
```

### 🏡 2.1 · İş Problemi

Bir **emlak değerleme platformu** düşünün. Bir mahallede gerçekleşen satışları bilmeden, yalnızca **nüfus sayımı özetlerinden** (gelir, hane sayısı, konum, evlerin yaşı …) o mahallenin **ortanca ev değerini** tahmin etmek istiyoruz. İyi bir model:

- 🏦 Kredi ve sigorta ekiplerine **hızlı bir ön değerleme** verir,
- 🗺️ Yeni açılan bir bölgede **fiyat seviyesini** kabaca gösterir,
- 📏 Ne kadar **emin** olduğunu da söyler (tek sayı yerine fiyat aralığı).

> **Soru:** Bir mahallenin ortanca ev değerini **kaç dolar** yanılarak tahmin edebiliyoruz? Ve bu hata **yeni bir bölgede** de aynı mı kalıyor?

Kaggle'da paylaşılan sürümde **20.640 satır** vardır; her satır bir **mahalle bloğudur** (*block group*, ortalama ~1.425 kişi).

| Ölçüt | Anlamı |

### 🔒 2.4 · Test Setini Kilitliyoruz

Verinin **%20'sini** (4.128 mahalle) şimdiden ayırıp **kilitliyoruz**; model seçimi, ayar ve keşif **yalnızca eğitim setiyle** yapılacak. Test seti en sonda bir kez “sınav” olarak açılır. Ayırırken hedefi 10 dilime bölüp (*stratified*) her dilimden orantılı örnek alıyoruz; böylece eğitim ve test fiyat dağılımları birbirine benzer.

```python
fiyat_dilimi = pd.qcut(ham[HEDEF], 10, labels=False)                              # hedefi 10 eşit dilime böl
egitim_idx, test_idx = train_test_split(np.arange(len(ham)), test_size=0.20, stratify=fiyat_dilimi, random_state=RANDOM_STATE)

X_tum, y_tum = ham.drop(columns=HEDEF), ham[HEDEF]
X_egitim, y_egitim = X_tum.iloc[egitim_idx].reset_index(drop=True), y_tum.iloc[egitim_idx].reset_index(drop=True)
X_test, y_test = X_tum.iloc[test_idx].reset_index(drop=True), y_tum.iloc[test_idx].reset_index(drop=True)

print(f"Eğitim: {len(X_egitim):,} mahalle · Test (kilitli): {len(X_test):,} mahalle")
karsilastir = pd.DataFrame({"eğitim": y_egitim.describe(), "test": y_test.describe()}).round(0)
display(karsilastir.T)

taban = float(y_egitim.mean())                                                    # taban model: herkese eğitim ortalaması
taban_olcut = olcutler(y_test, np.full(len(y_test), taban))
print(f"\nTaban model (herkese {usd(taban)}) → test RMSE = {usd(taban_olcut['RMSE'])} · MAE = {usd(taban_olcut['MAE'])} · R² = {taban_olcut['R²']:.3f}")
```

### 🧱 5.1 · Hücre Haritası (Coğrafi Gruplar)

Haritayı **0,5° × 0,5°**'lik (yaklaşık 45 × 55 km) hücrelere bölüyoruz. Aynı hücredeki mahalleler **aynı gruba** aittir; coğrafi doğrulamada bir grup ya tamamen eğitimde ya tamamen sınavdadır.

```python
def hucre_no(X, boyut=0.5):
    """Her mahalleye bulunduğu 0,5°'lik hücrenin numarasını verir (coğrafi grup)."""
    return (np.floor(X["latitude"] / boyut).astype(int) * 1000 + np.floor(X["longitude"] / boyut).astype(int)).to_numpy()


GRUP = hucre_no(X_egitim)
CV5 = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)                    # rastgele 5 kat
print(f"Hücre sayısı: {len(np.unique(GRUP))} · en kalabalık hücre: {pd.Series(GRUP).value_counts().iloc[0]:,} mahalle")

# Coğrafi katlar: GroupKFold hücreleri katlara dağıtır; hangi hücre hangi katta?
kat_no = np.zeros(len(X_egitim), dtype=int)
for k, (_, va) in enumerate(GroupKFold(n_splits=5).split(X_egitim, y_egitim, groups=GRUP)):
    kat_no[va] = k
fig, ax = plt.subplots(figsize=(6.5, 6.2))
for k in range(5):
    m = kat_no == k
    ax.scatter(X_egitim["longitude"][m], X_egitim["latitude"][m], s=4, label=f"kat {k + 1} ({m.sum():,} mahalle)")
ax.set(title="Coğrafi katlar: her renk bir sınav bölgesi", xlabel="boylam", ylabel="enlem")
ax.set_aspect(1.25)
ax.legend(markerscale=4, fontsize=8)
kaydet(fig, "06_cografi_katlar")
plt.show()
```

### 🔭 5.2 · Komşu Ne Kadar Yakın?

Her sınav mahallesinin, eğitim kümesindeki **en yakın komşusuna** olan uzaklığını ölçelim. Rastgele bölmede bu mesafe **çok kısadır**; coğrafi bölmede **kilometrelerce** artar.

```python
from scipy.spatial import cKDTree

KM = np.c_[X_egitim["latitude"] * 111.0, X_egitim["longitude"] * 111.0 * np.cos(np.radians(37))]   # derece → yaklaşık km


def en_yakin_mesafe(bolmeler):
    """Her doğrulama noktası için eğitim kümesindeki en yakın komşunun uzaklığı (km)."""
    sonuc = []
    for tr, va in bolmeler:
        d, _ = cKDTree(KM[tr]).query(KM[va])
        sonuc.append(d)
    return np.concatenate(sonuc)


mesafe_rastgele = en_yakin_mesafe(list(CV5.split(X_egitim)))
mesafe_cografi = en_yakin_mesafe(list(GroupKFold(n_splits=5).split(X_egitim, y_egitim, groups=GRUP)))
mesafe = pd.DataFrame({"rastgele": pd.Series(mesafe_rastgele).describe(percentiles=[0.5, 0.9]),
                       "coğrafi": pd.Series(mesafe_cografi).describe(percentiles=[0.5, 0.9])}).loc[["mean", "50%", "90%", "max"]].round(2)
mesafe.index = ["ortalama", "medyan", "%90", "en büyük"]
display(mesafe.rename_axis("en yakın eğitim komşusu (km)"))

fig, ax = plt.subplots(figsize=(8.5, 3.8))
ax.hist(np.clip(mesafe_rastgele, 0, 60), bins=60, alpha=0.75, label="rastgele bölme", color=RENK["mavi"])
ax.hist(np.clip(mesafe_cografi, 0, 60), bins=60, alpha=0.75, label="coğrafi bölme", color=RENK["pembe"])
ax.set(title="Sınav mahallesinin en yakın eğitim komşusuna uzaklığı", xlabel="km (60'ta kesildi)", ylabel="mahalle sayısı")
ax.set_yscale("log")
ax.legend()
kaydet(fig, "07_komsu_mesafesi")
plt.show()
```

### 🧪 5.3 · Deney: Aynı Model, İki Sınav

Varsayılan **XGBoost** ve basit bir **Ridge** modelini iki doğrulama stiliyle sınayalım. Model, pipeline ve veri **aynı**; değişen tek şey **sınavın nasıl kurulduğu**.

```python
def cv_tahmin(model_pipe, bolmeler, X=None, y=None):
    """Verilen bölmelerle (liste) çapraz doğrulama tahmini (OOF). Katlar paralel çalışır."""
    X = X_egitim if X is None else X
    y = y_egitim if y is None else y
    return cross_val_predict(model_pipe, X, y, cv=bolmeler, n_jobs=N_ISCI)


rastgele_b = list(CV5.split(X_egitim))
cografi_b = list(GroupKFold(n_splits=5).split(X_egitim, y_egitim, groups=GRUP))
satirlar = []
for ad, mod in [("XGBoost · varsayılan", XGBRegressor(random_state=RANDOM_STATE, n_jobs=1)),
                ("Ridge (log + ölçek)", Ridge())]:
    for sinav, b in [("rastgele 5 kat", rastgele_b), ("coğrafi 5 kat", cografi_b)]:
        pipe = pipe_kur(mod, olcek="log" if "Ridge" in ad else "yok")
        oof = cv_tahmin(pipe, b)
        satirlar.append({"model": ad, "sınav": sinav, **olcutler(y_egitim, oof)})
sinav_tablo = pd.DataFrame(satirlar)
display(sinav_tablo.round(3))
rapor_kaydet(sinav_tablo.round(4), "sinav_stili_deneyi", index=False)

fig, eksen = plt.subplots(1, 2, figsize=(12, 4))
for ax, metrik in zip(eksen, ["RMSE", "R²"]):
    sns.barplot(data=sinav_tablo, x="model", y=metrik, hue="sınav", palette=[RENK["mavi"], RENK["pembe"]], ax=ax)
    ax.set_title(f"{metrik}: rastgele ↔ coğrafi sınav")
    ax.set_xlabel("")
fig.tight_layout()
kaydet(fig, "08_sinav_stili_deneyi")
plt.show()
```

Ağaç modelinin rastgele sınavdaki üstünlüğünün büyük kısmı **komşuyu hatırlamaktan** geliyor; yeni bir bölgede avantaj **küçülüyor** (ama Ridge'den hâlâ 7.800 $ iyi). ### 🧬 7.1 · Yarışçıların Listesi

Tüm modeller **varsayılan ya da makul ayarlarla** başlar (ayar 11. bölümde). Hiçbirine ayrıcalık yok: **ayar eşitliği**, adil karşılaştırmanın ilk şartıdır. Süre karşılaştırması için her model **tek iş parçacığıyla** çalışır, **modeller** ise paralel koşar.

```python
def model_listesi():
    """{ad: (aile, tahminleyici, ölçek, hedef_ölçekle?)}  ·  hepsi tek iş parçacığıyla çalışır."""
    rs = RANDOM_STATE
    xgb = lambda **k: XGBRegressor(random_state=rs, n_jobs=1, **k)
    lgbm = lambda **k: LGBMRegressor(random_state=rs, n_jobs=1, verbose=-1, **k)
    cat = lambda **k: CatBoostRegressor(random_seed=rs, thread_count=1, verbose=0, allow_writing_files=False, **k)
    L = {}

    def ekle(ad, aile, model, olcek="yok", hedef=False):
        L[ad] = (aile, model, olcek, hedef)

    ekle("Taban · ortalama", "Taban", DummyRegressor())
    # ── Doğrusal aile (log ölçek; hedef standartlaşır) ──
    ekle("Doğrusal Regresyon", "Doğrusal", LinearRegression(), "log")
    ekle("Ridge", "Doğrusal", Ridge(alpha=1.0), "log", True)
    ekle("Lasso", "Doğrusal", Lasso(alpha=0.01, max_iter=5000), "log", True)
    ekle("ElasticNet", "Doğrusal", ElasticNet(alpha=0.01, l1_ratio=0.5, max_iter=5000), "log", True)
    ekle("Huber", "Doğrusal", HuberRegressor(max_iter=500), "log", True)
    ekle("Bayesian Ridge", "Doğrusal", BayesianRidge(), "log", True)
    ekle("SGD Regressor", "Doğrusal", SGDRegressor(random_state=rs), "log", True)
    ekle("Gamma GLM", "Doğrusal", GammaRegressor(alpha=1e-4, max_iter=1000), "log")
    ekle("Polinom (d=2) + Ridge", "Doğrusal", Pipeline([("poly", PolynomialFeatures(2, include_bias=False)), ("ridge", Ridge(alpha=10.0))]), "log", True)
    # ── Komşuluk ve çekirdek (kuantil ölçek) ──
    ekle("k-En Yakın Komşu", "Komşuluk ve çekirdek", KNeighborsRegressor(), "kuantil", True)
    ekle("SVR (RBF)", "Komşuluk ve çekirdek", SVR(), "kuantil", True)
    ekle("Doğrusal SVR", "Komşuluk ve çekirdek", LinearSVR(random_state=rs, max_iter=5000), "kuantil", True)
    ekle("Nystroem + Ridge", "Komşuluk ve çekirdek", Pipeline([("ny", Nystroem(n_components=300, random_state=rs)), ("ridge", Ridge())]), "kuantil", True)
    # ── Sinir ağı ──
    ekle("MLP (64-32)", "Sinir ağı", MLPRegressor(hidden_layer_sizes=(64, 32), early_stopping=True, max_iter=300, random_state=rs), "kuantil", True)
    # ── Ağaç ve orman ──
    ekle("Karar Ağacı", "Ağaç ve orman", DecisionTreeRegressor(random_state=rs))
    ekle("Tek Rastgele Ağaç", "Ağaç ve orman", ExtraTreeRegressor(random_state=rs))
    ekle("Random Forest", "Ağaç ve orman", RandomForestRegressor(random_state=rs, n_jobs=1))
    ekle("Extra Trees", "Ağaç ve orman", ExtraTreesRegressor(random_state=rs, n_jobs=1))
    ekle("Bagging", "Ağaç ve orman", BaggingRegressor(random_state=rs, n_jobs=1))
    # ── Boosting ──
    ekle("AdaBoost", "Boosting", AdaBoostRegressor(random_state=rs))
    ekle("Gradient Boosting", "Boosting", GradientBoostingRegressor(random_state=rs))
    ekle("HistGradientBoosting", "Boosting", HistGradientBoostingRegressor(random_state=rs))
    ekle("XGBoost · varsayılan", "Boosting", xgb())
    ekle("XGBoost · DART", "Boosting", xgb(booster="dart", rate_drop=0.1, skip_drop=0.5))
    ekle("XGBoost · orman modu", "Boosting", xgb(n_estimators=1, num_parallel_tree=200, learning_rate=1.0, subsample=0.8, colsample_bynode=0.8, reg_lambda=1e-5, max_depth=12))
    ekle("LightGBM", "Boosting", lgbm())
    ekle("CatBoost", "Boosting", cat())
    # ── Topluluk ──
    ekle("Voting (XGB + LGBM + Cat)", "Topluluk", VotingRegressor([("xgb", xgb()), ("lgbm", lgbm()), ("cat", cat())]))
    ekle("Stacking (ET + XGB + LGBM → Ridge)", "Topluluk",
         StackingRegressor([("et", ExtraTreesRegressor(random_state=rs, n_jobs=1)), ("xgb", xgb()), ("lgbm", lgbm())], final_estimator=RidgeCV(), cv=5, n_jobs=1))
    return L


MODELLER = model_listesi()
aile_sayim = pd.Series({a: sum(v[0] == a for v in MODELLER.values()) for a in AILE_RENK})
print(f"Toplam: {len(MODELLER) - 1} algoritma + 1 taban model")
display(aile_sayim.to_frame("algoritma sayısı").T)
```

### 🏎️ 7.2 · Yarışı Koşalım

Her model **5 katlı çapraz doğrulama** ile değerlendirilir: eğitim setinin 5 parçasından her biri sırayla “sınav” olur. Her mahalle **tam bir kez** sınav görür ve tahmini saklanır (**OOF** tahmini); bu tahminlerden hem skor hem de sonraki bölümdeki **bootstrap** hesaplanır.

```python
def yaris_kos(ad, aile, model, olcek, hedef, bolmeler):
    """Tek modeli verilen katlarla koşar; OOF tahmin, süre ve ölçütleri döndürür."""
    t0 = time.perf_counter()
    pipe = pipe_kur(clone(model), olcek, hedef)
    oof = cross_val_predict(pipe, X_egitim, y_egitim, cv=bolmeler, n_jobs=1)
    sure = time.perf_counter() - t0
    kat = [rmse(y_egitim.iloc[va], oof[va]) for _, va in bolmeler]
    o = olcutler(y_egitim, oof)
    return {"algoritma": ad, "aile": aile, "CV RMSE": o["RMSE"], "kat std": float(np.std(kat)), "CV MAE": o["MAE"], "CV R²": o["R²"],
            "CV MAPE %": o["MAPE %"], "süre (sn)": sure}, oof


def yaris(bolmeler, modeller=None):
    modeller = MODELLER if modeller is None else modeller
    sonuc = Parallel(n_jobs=N_ISCI)(delayed(yaris_kos)(ad, *v, bolmeler) for ad, v in modeller.items())
    tablo = pd.DataFrame([s[0] for s in sonuc]).sort_values("CV RMSE").reset_index(drop=True)
    tablo.index = tablo.index + 1
    return tablo, {s[0]["algoritma"]: s[1] for s in sonuc}


t0 = time.perf_counter()
tablo_yaris, oof_yaris = yaris(rastgele_b)
print(f"Yarış {time.perf_counter() - t0:.0f} sn sürdü ({len(tablo_yaris)} model × 5 kat).")
display(tablo_yaris.round({"CV RMSE": 0, "kat std": 0, "CV MAE": 0, "CV R²": 3, "CV MAPE %": 1, "süre (sn)": 1}))
rapor_kaydet(tablo_yaris.round(4), "yaris_cv", index=True)
```

### 🔬 8.1 · Yarışın Test Karnesi

Her modeli **tüm eğitim setiyle** eğitip hem **eğitim** hem **test** setinde değerlendiriyoruz. (Model seçimi hâlâ **CV'ye** göre yapıldı; test yalnızca **bağımsız doğrulama** amaçlı.)

```python
def test_kos(ad, aile, model, olcek, hedef):
    t0 = time.perf_counter()
    pipe = pipe_kur(clone(model), olcek, hedef).fit(X_egitim, y_egitim)
    sure = time.perf_counter() - t0
    p_egitim, p_test = pipe.predict(X_egitim), pipe.predict(X_test)
    o = olcutler(y_test, p_test)
    return {"algoritma": ad, "aile": aile, "eğitim RMSE": rmse(y_egitim, p_egitim), "test RMSE": o["RMSE"], "test MAE": o["MAE"], "test R²": o["R²"],
            "test MAPE %": o["MAPE %"], "eğitim süresi (sn)": sure}, p_test


sonuc = Parallel(n_jobs=N_ISCI)(delayed(test_kos)(ad, *v) for ad, v in MODELLER.items())
tablo_test = pd.DataFrame([s[0] for s in sonuc]).merge(tablo_yaris[["algoritma", "CV RMSE"]], on="algoritma")
tablo_test["fark (test − CV)"] = tablo_test["test RMSE"] - tablo_test["CV RMSE"]
tablo_test["aşırı öğrenme payı"] = tablo_test["test RMSE"] - tablo_test["eğitim RMSE"]
tablo_test = tablo_test.sort_values("test RMSE").reset_index(drop=True)
tablo_test.index += 1
tahmin_test = {s[0]["algoritma"]: s[1] for s in sonuc}
display(tablo_test[["algoritma", "aile", "eğitim RMSE", "CV RMSE", "test RMSE", "test R²", "test MAPE %", "eğitim süresi (sn)"]].round({"eğitim RMSE": 0, "CV RMSE": 0, "test RMSE": 0, "test R²": 3, "test MAPE %": 1, "eğitim süresi (sn)": 1}))
rapor_kaydet(tablo_test.round(4), "yaris_test", index=True)
sira_cv = tablo_yaris.set_index("algoritma")["CV RMSE"].rank()
sira_test = tablo_test.set_index("algoritma")["test RMSE"].rank()
from scipy.stats import spearmanr
print(f"CV sırası ile test sırası arasındaki Spearman ρ = {spearmanr(sira_cv, sira_test.reindex(sira_cv.index))[0]:.3f}")
```

### 🎲 8.3 · Fark Gerçek mi, Şans mı? (Bootstrap)

16.512 mahallenin OOF tahminleri üzerinde **mahalleleri yeniden örnekleyerek** (2.000 kez) iki modelin **RMSE farkının** %95 güven aralığını buluyoruz. Aralık **sıfırı içeriyorsa**, fark şanstan ayırt edilemez.

```python
def bootstrap_fark(oof_a, oof_b, B=2000, seed=RANDOM_STATE):
    """RMSE(a) − RMSE(b) farkının bootstrap dağılımı (negatif = a daha iyi)."""
    rng = np.random.default_rng(seed)
    ea, eb = (y_egitim.to_numpy() - oof_a) ** 2, (y_egitim.to_numpy() - oof_b) ** 2
    fark = np.empty(B)
    for i in range(B):
        idx = rng.integers(0, len(ea), len(ea))
        fark[i] = np.sqrt(ea[idx].mean()) - np.sqrt(eb[idx].mean())
    return fark


REF = "XGBoost · varsayılan"
adaylar = [a for a in tablo_yaris["algoritma"].head(9) if a != REF]
satirlar = []
for ad in adaylar:
    f = bootstrap_fark(oof_yaris[ad], oof_yaris[REF])
    lo, hi = np.percentile(f, [2.5, 97.5])
    satirlar.append({"algoritma": ad, "RMSE farkı (aday − XGB varsayılan)": f.mean(), "%95 alt": lo, "%95 üst": hi, "sıfırı içeriyor mu?": "evet (şans)" if lo <= 0 <= hi else "hayır (gerçek)"})
t83 = pd.DataFrame(satirlar)
display(t83.round(0))
rapor_kaydet(t83.round(1), "bootstrap_fark", index=False)

fig, ax = plt.subplots(figsize=(10, 0.5 * len(t83) + 1.6))
y = np.arange(len(t83))[::-1]
renkler = [RENK["gri"] if r["sıfırı içeriyor mu?"].startswith("evet") else (RENK["yesil"] if r["RMSE farkı (aday − XGB varsayılan)"] < 0 else RENK["pembe"]) for _, r in t83.iterrows()]
ax.hlines(y, t83["%95 alt"], t83["%95 üst"], color=renkler, lw=5)
ax.scatter(t83["RMSE farkı (aday − XGB varsayılan)"], y, color="black", zorder=3, s=25)
ax.axvline(0, color="black", lw=1)
ax.set_yticks(y)
ax.set_yticklabels(t83["algoritma"])
ax.set(title=f"Varsayılan XGBoost'a göre RMSE farkı (%95 bootstrap aralığı)", xlabel="RMSE farkı ($) · negatif = aday daha iyi")
fig.tight_layout()
kaydet(fig, "11_bootstrap_fark")
plt.show()
```

### 🧰 10.1 · Hazırlık

```python
on_hazir = Pipeline([("ozellik", KonutOzellikleri("tam")), ("on", on_isleme("yok"))])
idx_e, idx_d = train_test_split(np.arange(len(X_egitim)), test_size=0.2, random_state=RANDOM_STATE)
X_e = on_hazir.fit_transform(X_egitim.iloc[idx_e])                              # ön işleme yalnızca eğitim kısmından öğrenilir
X_d = on_hazir.transform(X_egitim.iloc[idx_d])
y_e, y_d = y_egitim.iloc[idx_e].to_numpy(), y_egitim.iloc[idx_d].to_numpy()
print(f"Eğitim: {X_e.shape} · Doğrulama: {X_d.shape}")


def xgb_egit(izle=True, **k):
    """XGBoost'u eğitir; her turdaki eğitim/doğrulama RMSE'sini kaydeder."""
    m = XGBRegressor(random_state=RANDOM_STATE, n_jobs=N_ISCI, eval_metric="rmse", **k)
    m.fit(X_e, y_e, eval_set=[(X_e, y_e), (X_d, y_d)], verbose=False)
    return m


def egriler(m):
    g = m.evals_result()
    return np.array(g["validation_0"]["rmse"]), np.array(g["validation_1"]["rmse"])
```

### 🪜 10.4 · Derinlik × Çocuk Ağırlığı

`max_depth` ağacın ne kadar **karmaşık** soru sorabileceğini, `min_child_weight` ise bir yaprakta **en az kaç örnek ağırlığı** bulunması gerektiğini belirler (küçük yapraklara frendir).

```python
satirlar = []
for d in [2, 3, 4, 6, 8, 10, 12]:
    for mcw in [1, 5, 20]:
        m = xgb_egit(n_estimators=300, learning_rate=0.1, max_depth=d, min_child_weight=mcw)
        satirlar.append({"derinlik": d, "min_child_weight": mcw, "eğitim RMSE": rmse(y_e, m.predict(X_e)), "doğrulama RMSE": rmse(y_d, m.predict(X_d))})
t_d = pd.DataFrame(satirlar)
isi = t_d.pivot(index="derinlik", columns="min_child_weight", values="doğrulama RMSE")
fark = t_d.assign(makas=t_d["doğrulama RMSE"] - t_d["eğitim RMSE"]).pivot(index="derinlik", columns="min_child_weight", values="makas")
fig, eksen = plt.subplots(1, 2, figsize=(12, 4.2))
sns.heatmap(isi, annot=True, fmt=".0f", cmap="viridis_r", ax=eksen[0], cbar=False)
eksen[0].set_title("Doğrulama RMSE ($)")
sns.heatmap(fark, annot=True, fmt=".0f", cmap="rocket_r", ax=eksen[1], cbar=False)
eksen[1].set_title("Makas: doğrulama − eğitim ($)")
fig.tight_layout()
kaydet(fig, "17_xgb_derinlik_cocuk_agirligi")
plt.show()
rapor_kaydet(t_d.round(1), "xgb_derinlik_cocuk_agirligi", index=False)
```

### 🌲 10.7 · `booster` ve `tree_method` Seçenekleri

```python
satirlar = []
secenekler = [("gbtree · hist (varsayılan)", dict(booster="gbtree", tree_method="hist")),
              ("gbtree · exact", dict(booster="gbtree", tree_method="exact")),
              ("gbtree · approx", dict(booster="gbtree", tree_method="approx")),
              ("gbtree · lossguide (yaprak odaklı)", dict(booster="gbtree", tree_method="hist", grow_policy="lossguide", max_leaves=63, max_depth=0)),
              ("DART (dropout)", dict(booster="dart", rate_drop=0.1, skip_drop=0.5)),
              ("gblinear (doğrusal)", dict(booster="gblinear"))]
for ad, ek in secenekler:
    t0 = time.perf_counter()
    m = XGBRegressor(n_estimators=300, learning_rate=0.1, random_state=RANDOM_STATE, n_jobs=N_ISCI, **({"max_depth": 6} if "gblinear" not in ad and "lossguide" not in ad else {}), **ek).fit(X_e, y_e)
    sure = time.perf_counter() - t0
    satirlar.append({"yöntem": ad, "doğrulama RMSE": rmse(y_d, m.predict(X_d)), "süre (sn)": sure})
t_b = pd.DataFrame(satirlar)
display(t_b.round(1))
rapor_kaydet(t_b.round(2), "xgb_booster_yontemleri", index=False)
```

### 🔍 11.2 · Taze Katlarda Doğrulama ve Final Adayları

Aramada **en iyi çıkan** ayar, aynı katlarda seçildiği için **iyimser** olabilir (*kazananın laneti*). Bu yüzden ilk üç XGBoost adayını ve diğer finalistleri, aramada hiç kullanılmamış **taze katlarla** (5 kat × 2 tekrar, farklı tohum) yeniden sınıyoruz.

```python
def parametre_sozlugu(satir, on_ek="param_model__"):
    """Arama tablosundaki bir satırdan model parametrelerini okur."""
    p = {c.replace(on_ek, ""): satir[c] for c in satir.index if c.startswith(on_ek)}
    for k, v in p.items():
        if k in ("n_estimators", "max_depth", "min_child_weight", "num_leaves", "min_child_samples", "subsample_freq"):
            p[k] = int(v)
        else:
            p[k] = float(v)
    return p


TAZE = list(RepeatedKFold(n_splits=5, n_repeats=2, random_state=2024).split(X_egitim))
xgb_p = [parametre_sozlugu(ara_xgb.loc[i]) for i in range(3)]
lgbm_p = parametre_sozlugu(ara_lgbm.loc[0])
xgb_k = lambda p=None: XGBRegressor(random_state=RANDOM_STATE, n_jobs=1, **(p or {}))
lgbm_k = lambda p=None: LGBMRegressor(random_state=RANDOM_STATE, n_jobs=1, verbose=-1, **(p or {}))
cat_k = lambda: CatBoostRegressor(random_seed=RANDOM_STATE, thread_count=1, verbose=0, allow_writing_files=False)

adaylar = {
    "XGBoost · varsayılan": xgb_k(),
    "XGBoost · ayarlı #1": xgb_k(xgb_p[0]), "XGBoost · ayarlı #2": xgb_k(xgb_p[1]), "XGBoost · ayarlı #3": xgb_k(xgb_p[2]),
    "LightGBM · varsayılan": lgbm_k(), "LightGBM · ayarlı": lgbm_k(lgbm_p),
    "CatBoost · varsayılan": cat_k(),
    "Voting (ayarlı XGB + ayarlı LGBM + Cat)": VotingRegressor([("xgb", xgb_k(xgb_p[0])), ("lgbm", lgbm_k(lgbm_p)), ("cat", cat_k())]),
}


def taze_kos(ad, model):
    skorlar = []
    for tr, va in TAZE:
        p = pipe_kur(clone(model)).fit(X_egitim.iloc[tr], y_egitim.iloc[tr]).predict(X_egitim.iloc[va])
        skorlar.append(rmse(y_egitim.iloc[va], p))
    return {"aday": ad, "taze CV RMSE": np.mean(skorlar), "std": np.std(skorlar)}


t0 = time.perf_counter()
taze = pd.DataFrame(Parallel(n_jobs=N_ISCI)(delayed(taze_kos)(a, m) for a, m in adaylar.items())).sort_values("taze CV RMSE").reset_index(drop=True)
print(f"Taze katlar {time.perf_counter() - t0:.0f} sn sürdü.")
display(taze.round(0))
rapor_kaydet(taze.round(2), "final_adaylari_taze_katlar", index=False)
```

Taze katlarda **ayarlı XGBoost #1: 43.261 ± 561 $** ve varsayılan **46.453 ± 949 $**: gerçek kazanç ≈ **3.200 $ (%6,9)**. **Voting** 42.971 $ ile en iyi, ama XGBoost #1'den farkı (≈ 290 $) katlar arası dalgalanmanın (≈ 670 $) altında. **Ayarsız CatBoost** (43.919 $) ayarlı XGBoost'un ≈ 660 $ gerisinde; yani **ayar, algoritma seçimi kadar fark yaratıyor**. Proje **XGBoost** odaklıdır; final model, **taze katlardaki en iyi XGBoost adayıdır** (varsayılan dâhil). Ayarın kazancını **doğrulama kümesinde** değil, **taze katlarda** ölçüyoruz; kazanç küçükse bunu dürüstçe söylemek gerekir.

- **Permütasyon önemi:** test setinde sütunu **karıştırınca** hata ne kadar artıyor? (modelden bağımsız, *dış* görüş)
- **TreeSHAP** (Lundberg vd., 2020): her tahmini sütun katkılarına **toplamsal** olarak ayırır.

```python
FINAL_PIPE.fit(X_egitim, y_egitim)
ozellik_adlari = [a.split("__", 1)[1].replace("<", "alti_").replace(" ", "_") for a in FINAL_PIPE[:-1].get_feature_names_out()]     # XGBoost "<" ve boşluk sevmez
booster = FINAL_PIPE.named_steps["model"].get_booster()
booster.feature_names = ozellik_adlari
print("Modelin girdileri:", len(ozellik_adlari), "özellik")

onem_tablosu = pd.DataFrame({tur: pd.Series(booster.get_score(importance_type=tur)) for tur in ["weight", "gain", "cover"]}).reindex(ozellik_adlari).fillna(0)
for c in onem_tablosu:
    onem_tablosu[c] = onem_tablosu[c] / onem_tablosu[c].sum()
onem_tablosu = onem_tablosu.sort_values("gain", ascending=False)
display(onem_tablosu.round(3))
rapor_kaydet(onem_tablosu.round(4), "ozellik_onemi_xgboost")

fig, eksen = plt.subplots(1, 3, figsize=(16, 4.6), sharey=True)
t = onem_tablosu.iloc[::-1]
for ax, tur, renk, ad in zip(eksen, ["weight", "gain", "cover"], [RENK["gri"], RENK["mavi"], RENK["turuncu"]], ["weight: kaç kez bölünmede kullanıldı", "gain: bölünmelere toplam katkı", "cover: kaç örneği etkiledi"]):
    ax.barh(t.index, t[tur], color=renk)
    ax.set_title(ad, fontsize=10)
fig.tight_layout()
kaydet(fig, "21_ozellik_onemi_turleri")
plt.show()
```

### 🧪 12.4 · Ablasyon: Bilgiyi Gerçekten Çıkarınca Ne Oluyor?

Önem puanları **neyin kullanıldığını** söyler; ama bir sütun grubunu **gerçekten çıkarıp** modeli yeniden eğitmek, **neye ihtiyaç duyulduğunu** gösterir. Aynı final parametrelerle, sütun **gruplarını** tek tek çıkarıp 5 katlı CV yapıyoruz (seçim, pipeline **içinde** yapıldığı için sızıntı yoktur).

```python
tam_adlar = ozellik_adlari
gruplar = {
    "Tümü": [],
    "− Konum (boylam, enlem)": ["longitude", "latitude"],
    "− Okyanus yakınlığı": [a for a in tam_adlar if a.startswith("ocean_proximity")],
    "− Şehre uzaklık": ["en_yakin_sehir_km"],
    "− Tüm konum bilgisi": ["longitude", "latitude", "en_yakin_sehir_km"] + [a for a in tam_adlar if a.startswith("ocean_proximity")],
    "− Gelir": ["median_income"],
    "− Konut yaşı": ["housing_median_age"],
    "− Ham toplam sayılar (oda, yatak, nüfus, hane)": ["total_rooms", "total_bedrooms", "population", "households"],
    "− Oran özellikleri": ["oda_hane", "yatak_oda_orani", "kisi_hane"],
}


def ablasyon_kos(ad, cikar):
    tut = [i for i, a in enumerate(tam_adlar) if a not in cikar]
    model = clone(FINAL_MODEL).set_params(n_jobs=1)
    pipe = Pipeline([("ozellik", KonutOzellikleri("tam")), ("on", on_isleme("yok")), ("sec", ColumnTransformer([("tut", "passthrough", tut)])), ("model", model)])
    oof = cross_val_predict(pipe, X_egitim, y_egitim, cv=rastgele_b, n_jobs=1)
    return {"grup": ad, "çıkarılan sütun": len(cikar), "CV RMSE": rmse(y_egitim, oof)}


abl = pd.DataFrame(Parallel(n_jobs=N_ISCI)(delayed(ablasyon_kos)(a, c) for a, c in gruplar.items()))
abl["kayıp ($)"] = abl["CV RMSE"] - abl.loc[0, "CV RMSE"]
display(abl.round(0))
rapor_kaydet(abl.round(1), "ablasyon", index=False)
fig, ax = plt.subplots(figsize=(9, 4.2))
a = abl.iloc[1:].sort_values("kayıp ($)")
ax.barh(a["grup"], a["kayıp ($)"], color=[RENK["pembe"] if v > 1500 else RENK["gri"] for v in a["kayıp ($)"]])
ax.set(title="Bir grubu çıkarınca CV RMSE ne kadar bozuluyor?", xlabel="RMSE artışı ($)")
fig.tight_layout()
kaydet(fig, "25_ablasyon")
plt.show()
```

### 🔬 13.3 · Segment Denetimi: Hangi Alt Grupta Daha Çok Yanılıyoruz?

Tek bir genel skor, alt gruplardaki **eşitsizlikleri gizler**. Test mahallelerini **okyanus yakınlığı, gelir dilimi, en yakın şehir, tavan durumu ve konut yaşına** göre ayırıp her grupta RMSE, MAE, **yüzde hata** ve **yanlılığı** (*bias*: ortalama tahmin − gerçek) ölçüyoruz. Pozitif yanlılık = model o grupta **fazla pahalı** tahmin ediyor.

```python
d = X_test.copy()
d["gercek"], d["tahmin"] = y_test.to_numpy(), p_test
sehir_ad = list(SEHIRLER)
mesafe = np.column_stack([uzaklik_km(d["latitude"], d["longitude"], *SEHIRLER[s]) for s in sehir_ad])
d["en yakın şehir"] = np.array(sehir_ad)[mesafe.argmin(1)]
d.loc[mesafe.min(1) > 100, "en yakın şehir"] = "100 km'den uzak (kırsal)"
gelir_sinir = X_egitim["median_income"].quantile([0.25, 0.5, 0.75]).to_numpy()                 # sınırlar eğitimden
d["gelir dilimi"] = pd.cut(d["median_income"], [-np.inf, *gelir_sinir, np.inf], labels=["Ç1 (en düşük)", "Ç2", "Ç3", "Ç4 (en yüksek)"])
d["fiyat tavanı"] = np.where(d["gercek"] >= 500001, "tavanlı (500.001 $)", "tavansız")
d["konut yaşı"] = pd.cut(d["housing_median_age"], [0, 15, 30, 45, 60], labels=["≤15", "16–30", "31–45", "46–52"])


def segment_tablosu(sutun):
    t = d.groupby(sutun, observed=True).apply(lambda g: pd.Series({"n": len(g), "RMSE": rmse(g["gercek"], g["tahmin"]), "MAE": mean_absolute_error(g["gercek"], g["tahmin"]),
                                                                      "MAPE %": 100 * (np.abs(g["gercek"] - g["tahmin"]) / g["gercek"]).mean(),
                                                                      "yanlılık ($)": (g["tahmin"] - g["gercek"]).mean()}))
    t.insert(0, "segment türü", sutun)
    return t.reset_index().rename(columns={sutun: "segment"})


segmentler = pd.concat([segment_tablosu(s) for s in ["ocean_proximity", "gelir dilimi", "en yakın şehir", "fiyat tavanı", "konut yaşı"]], ignore_index=True)
display(segmentler.round({"n": 0, "RMSE": 0, "MAE": 0, "MAPE %": 1, "yanlılık ($)": 0}))
rapor_kaydet(segmentler.round(2), "segment_denetimi", index=False)

fig, eksen = plt.subplots(1, 2, figsize=(16, 6.2))
s = segmentler[segmentler["n"] >= 30].reset_index(drop=True)
renk_s = [RENK["pembe"] if m > m_final["MAPE %"] * 1.2 else (RENK["yesil"] if m < m_final["MAPE %"] * 0.8 else RENK["gri"]) for m in s["MAPE %"]]
eksen[0].barh(s["segment türü"] + " · " + s["segment"].astype(str), s["MAPE %"], color=renk_s)
eksen[0].axvline(m_final["MAPE %"], color="black", ls="--", label=f"genel MAPE %{m_final['MAPE %']:.1f}")
eksen[0].invert_yaxis()
eksen[0].legend()
eksen[0].set(title="Segmentlere göre ortalama yüzde hata", xlabel="MAPE (%)")
eksen[0].tick_params(axis="y", labelsize=8)
eksen[1].barh(s["segment türü"] + " · " + s["segment"].astype(str), s["yanlılık ($)"] / 1000, color=[RENK["pembe"] if v > 0 else RENK["mavi"] for v in s["yanlılık ($)"]])
eksen[1].invert_yaxis()
eksen[1].axvline(0, color="black")
eksen[1].set(title="Yanlılık (+ fazla pahalı tahmin, − fazla ucuz tahmin)", xlabel="ortalama (tahmin − gerçek), bin $")
eksen[1].tick_params(axis="y", labelsize=8)
fig.tight_layout()
kaydet(fig, "28_segment_denetimi")
plt.show()
```

### 📐 14.1 · Kuantil Regresyonla Fiyat Aralığı

XGBoost'ta `objective="reg:quantileerror"` ile birden çok kuantil **tek modelde** öğrenilir. Aralığın doğruluğunu ölçmek için eğitim setini ikiye ayırıyoruz: **%75 öğrenme**, **%25 kalibrasyon**. Kalibrasyon kısmı, bir sonraki adımda aralığı **ayarlamak** için kullanılacak.

```python
ALFA = 0.20                                                                          # hedef: %80 kapsama
Xp = FINAL_PIPE[:-1].transform(X_egitim)                                             # ön işleme (final pipeline'dan, eğitimden öğrenilmiş)
Xp_test = FINAL_PIPE[:-1].transform(X_test)
i_ogren, i_kal = train_test_split(np.arange(len(Xp)), test_size=0.25, random_state=RANDOM_STATE)

final_p = {k: v for k, v in FINAL_MODEL.get_params().items() if k in ("n_estimators", "learning_rate", "max_depth", "min_child_weight", "subsample", "colsample_bytree", "reg_lambda", "reg_alpha", "gamma")}
kuantil_model = XGBRegressor(objective="reg:quantileerror", quantile_alpha=np.array([ALFA / 2, 0.5, 1 - ALFA / 2]), random_state=RANDOM_STATE, n_jobs=N_ISCI, **final_p)
kuantil_model.fit(Xp[i_ogren], y_egitim.to_numpy()[i_ogren])

q_kal, q_test = kuantil_model.predict(Xp[i_kal]), kuantil_model.predict(Xp_test)
q_kal.sort(axis=1)
q_test.sort(axis=1)                                                                  # kuantillerin çaprazlamasını önle
y_kal, yt = y_egitim.to_numpy()[i_kal], y_test.to_numpy()


def kapsama(q, y):
    return float(np.mean((y >= q[:, 0]) & (y <= q[:, 2])))


print(f"Ham aralık (10.–90. kuantil): kalibrasyon kapsaması %{100 * kapsama(q_kal, y_kal):.1f} · test kapsaması %{100 * kapsama(q_test, yt):.1f} · ortalama genişlik {usd((q_test[:, 2] - q_test[:, 0]).mean())}")
```

### 🎚️ 14.2 · Uyumlu Düzeltme (CQR)

Kalibrasyon kümesinde her mahalle için “aralığın dışına **ne kadar taştığını**” ölçüyoruz (içindeyse negatif). Bu skorların $\lceil (1-\alpha)(n+1) \rceil / n$ **kuantili**, aralığın **her iki yana ne kadar genişletilmesi** gerektiğini söyler. Bu yöntem, belirli varsayımlar altında **sonlu örneklemde kapsama garantisi** verir (Romano vd., 2019).

```python
skor = np.maximum(q_kal[:, 0] - y_kal, y_kal - q_kal[:, 2])                          # aralık dışına taşma miktarı (içindeyse negatif)
n_kal = len(skor)
QHAT = float(np.quantile(skor, min(1.0, np.ceil((1 - ALFA) * (n_kal + 1)) / n_kal), method="higher"))
print(f"Düzeltme miktarı: aralığın her iki ucu {usd(QHAT)} kadar genişletiliyor")

aralik = np.c_[q_test[:, 0] - QHAT, q_test[:, 1], q_test[:, 2] + QHAT]
aralik[:, 0] = np.maximum(aralik[:, 0], 0)
sonuc = pd.DataFrame({
    "ham kuantil aralığı": {"test kapsaması %": 100 * kapsama(q_test, yt), "ort. genişlik ($)": (q_test[:, 2] - q_test[:, 0]).mean(), "medyan göreli genişlik %": 100 * np.median((q_test[:, 2] - q_test[:, 0]) / q_test[:, 1])},
    "CQR düzeltmeli aralık": {"test kapsaması %": 100 * kapsama(aralik, yt), "ort. genişlik ($)": (aralik[:, 2] - aralik[:, 0]).mean(), "medyan göreli genişlik %": 100 * np.median((aralik[:, 2] - aralik[:, 0]) / aralik[:, 1])},
}).T
display(sonuc.round(1))
rapor_kaydet(sonuc.round(2), "aralik_kapsama")

# Segmentlere göre kapsama
d["alt"], d["ust"], d["medyan tahmin"] = aralik[:, 0], aralik[:, 2], aralik[:, 1]
d["içinde"] = (d["gercek"] >= d["alt"]) & (d["gercek"] <= d["ust"])
kaps = pd.concat([d.groupby(s, observed=True)["içinde"].agg(["mean", "size"]).rename(columns={"mean": "kapsama", "size": "n"}).assign(tür=s) for s in ["ocean_proximity", "gelir dilimi", "fiyat tavanı"]])
kaps["kapsama %"] = 100 * kaps["kapsama"]
display(kaps[["tür", "n", "kapsama %"]].round(1))
rapor_kaydet(kaps[["tür", "n", "kapsama %"]].round(2), "aralik_kapsama_segment")
```

### 🧮 14.3 · Belirsizliği Karara Çevirmek: “Hangi Tahmine İnsan Bakmalı?”

Aralığı **görece genişliğe** (genişlik ÷ medyan tahmin) çevirip tahminleri **güvenilirlik sırasına** koyuyoruz. Senaryo: değerleme ekibi **en belirsiz tahminleri elle inceler**; geri kalan tahminler **otomatik** onaylanır. Elle incelenen oran arttıkça otomatik kısımdaki hata nasıl değişiyor?

```python
gorece = (d["ust"] - d["alt"]) / d["medyan tahmin"]
sira = np.argsort(gorece.to_numpy())                                                # en güvenilirden en belirsize
yy, pp = d["gercek"].to_numpy(), p_test
satirlar = []
for oran in [0.3, 0.5, 0.7, 0.8, 0.9, 1.0]:
    k = int(oran * len(d))
    ic = sira[:k]
    ape = np.abs(yy[ic] - pp[ic]) / yy[ic]
    satirlar.append({"otomatik onaylanan oran": f"%{100 * oran:.0f}", "RMSE ($)": rmse(yy[ic], pp[ic]), "medyan yüzde hata": 100 * np.median(ape), "±%20 içinde (%)": 100 * (ape <= 0.20).mean()})
oto = pd.DataFrame(satirlar)
display(oto.round(1))
rapor_kaydet(oto.round(2), "otomatik_onay_butcesi", index=False)

kes = np.linspace(0.2, 1.0, 40)
rm = [rmse(yy[sira[: int(k * len(d))]], pp[sira[: int(k * len(d))]]) for k in kes]
isabet = [100 * (np.abs(yy[sira[: int(k * len(d))]] - pp[sira[: int(k * len(d))]]) / yy[sira[: int(k * len(d))]] <= 0.2).mean() for k in kes]
fig, ax = plt.subplots(figsize=(9, 4.3))
ax.plot(100 * kes, rm, color=RENK["pembe"], lw=3, label="RMSE ($)")
ax.set(title="Otomatik onay oranı arttıkça hata", xlabel="otomatik onaylanan mahalle oranı (%) · belirsizliğe göre sıralı", ylabel="RMSE ($)")
ax2 = ax.twinx()
ax2.plot(100 * kes, isabet, color=RENK["mavi"], lw=3, label="±%20 içinde (%)")
ax2.set_ylabel("±%20 isabet oranı (%)")
ax2.grid(False)
fig.legend(loc="upper center", ncol=2, bbox_to_anchor=(0.5, 0.0))
kaydet(fig, "31_otomatik_onay_butcesi")
plt.show()
```

### 💶 14.4 · Maliyete Duyarlı Tahmin: Asimetrik Hata

Her yanlışın **bedeli aynı değildir**. Bir kredi ekibi için bir mahalleyi **fazla pahalı** değerlemek (teminatı olduğundan değerli sanmak), **fazla ucuz** değerlemekten daha **maliyetlidir**. Varsayalım **fazla tahminin dolar başına maliyeti 2**, **eksik tahminin 1**. Böyle asimetrik maliyetin en iyi tahmini ortalama değil, **$\tau = c_{eksik}/(c_{eksik}+c_{fazla}) = 1/3$ kuantilidir**. Bunu sınayalım.

```python
C_FAZLA, C_EKSIK = 2.0, 1.0
tau = C_EKSIK / (C_EKSIK + C_FAZLA)
tau_model = XGBRegressor(objective="reg:quantileerror", quantile_alpha=tau, random_state=RANDOM_STATE, n_jobs=N_ISCI, **final_p).fit(Xp, y_egitim.to_numpy())
p_tau = tau_model.predict(Xp_test)


def toplam_maliyet(y, p):
    return float(np.mean(C_FAZLA * np.maximum(p - y, 0) + C_EKSIK * np.maximum(y - p, 0)))


kar = pd.DataFrame({
    "ortalama tahmin (final XGBoost)": p_test, "medyan tahmin (kuantil 0,5)": q_test[:, 1], f"asimetrik kuantil (τ = {tau:.2f})": p_tau,
}).apply(lambda p: pd.Series({"ort. maliyet ($/mahalle)": toplam_maliyet(yt, p.to_numpy()), "RMSE ($)": rmse(yt, p.to_numpy()),
                              "fazla tahmin oranı (%)": 100 * np.mean(p.to_numpy() > yt)})).T
display(kar.round(1))
rapor_kaydet(kar.round(2), "maliyete_duyarli_karar")
fig, ax = plt.subplots(figsize=(8.5, 3.6))
ax.barh(kar.index[::-1], kar["ort. maliyet ($/mahalle)"][::-1], color=[RENK["yesil"], RENK["gri"], RENK["gri"]])
for y_, v in enumerate(kar["ort. maliyet ($/mahalle)"][::-1]):
    ax.text(v + 80, y_, usd(v), va="center", fontsize=9)
ax.set(title=f"Fazla tahmin {C_FAZLA:g}×, eksik tahmin {C_EKSIK:g}× pahalıysa ortalama maliyet", xlabel="ortalama maliyet ($ / mahalle)", xlim=(0, kar["ort. maliyet ($/mahalle)"].max() * 1.2))
fig.tight_layout()
kaydet(fig, "32_maliyete_duyarli_karar")
plt.show()
```

### 🧰 15.2 · Kullanıma Hazır Tahmin Fonksiyonu

Değerleme ekibinin kullanacağı basit bir arayüz: mahallenin dokuz bilgisini ver, **fiyat tahminini**, **%80 aralığı** ve gerekirse bir **uyarıyı** al. Eğitim verisinin kapsamı dışındaki girdiler için uyarı da üretir.

```python
def konut_fiyati_tahmin(longitude, latitude, housing_median_age, total_rooms, total_bedrooms, population, households, median_income, ocean_proximity, model_paketi=None):
    """Bir mahalle için ortanca ev değeri tahmini ($), %80 aralık ve uyarılar döndürür."""
    p = joblib.load(MODEL_YOLU) if model_paketi is None else model_paketi
    x = pd.DataFrame([dict(longitude=longitude, latitude=latitude, housing_median_age=housing_median_age, total_rooms=total_rooms, total_bedrooms=total_bedrooms,
                           population=population, households=households, median_income=median_income, ocean_proximity=ocean_proximity)])[p["girdi_sutunlari"]]
    tahmin = float(p["nokta_modeli"].predict(x)[0])
    q = np.sort(p["aralik_modeli"].predict(x)[0])
    alt, ust = max(q[0] - p["cqr_duzeltme"], 0.0), q[2] + p["cqr_duzeltme"]
    uyarilar = []
    if not (32.5 <= latitude <= 42.0 and -124.5 <= longitude <= -114.1):
        uyarilar.append("Konum Kaliforniya sınırları dışında: tahmin güvenilmez.")
    if households < 1 or total_rooms <= 0:
        uyarilar.append("Hane veya oda sayısı geçersiz.")
    if ust >= 500001:
        uyarilar.append("Aralığın üst ucu veri setindeki 500.001 $ tavanına ulaşıyor: gerçek değer daha yüksek olabilir.")
    return {"tahmin ($)": round(tahmin), "%80 aralık ($)": (round(alt), round(ust)), "uyarilar": uyarilar}


yuklu = joblib.load(MODEL_YOLU)
for i in [3, 11, 20]:
    s = X_test.iloc[i]
    sonuc = konut_fiyati_tahmin(*s[["longitude", "latitude", "housing_median_age", "total_rooms", "total_bedrooms", "population", "households", "median_income", "ocean_proximity"]], model_paketi=yuklu)
    print(f"Test mahallesi {i}: gerçek {usd(y_test.iloc[i])} → tahmin {usd(sonuc['tahmin ($)'])} · aralık {usd(sonuc['%80 aralık ($)'][0])} – {usd(sonuc['%80 aralık ($)'][1])} · {sonuc['uyarilar'] or 'uyarı yok'}")
print("Kaliforniya dışı örnek:", konut_fiyati_tahmin(-100.0, 45.0, 20, 2000, 400, 1000, 400, 4.0, "INLAND", yuklu))
```

### ❓ 16.3 · Quiz (cevaplar açılır kutularda)

<details><summary><b>1.</b> Neden bu projede ana ölçüt olarak RMSE seçtik, tek başına yeterli mi?</summary>

Büyük hataları (pahalı mahallelerde birkaç yüz bin $'lık sapmalar) daha çok cezalandırdığı ve **doğrudan dolar** cinsinden yorumlandığı için. Tek başına yeterli değil: **MAE** uç değerlere daha az duyarlı, **MAPE/medyan yüzde hata** ucuz ve pahalı mahalleleri adil karşılaştırır, **R²** tabana göre iyileşmeyi gösterir.
</details>

<details><summary><b>2.</b> Fiyattaki 500.001 $ değeri neden sorun?</summary>

Gerçek fiyat değil, **kesilmiş** (sansürlü) bir değerdir: “en az bu kadar”. Model bu mahalleleri sistematik **düşük** tahmin eder (testte yanlılık ≈ −51 bin $, RMSE 86 bin $). Çıkarmak tavansız mahalleleri iyileştirir ama toplam hatayı artırır: karar **iş tanımına** bağlıdır.
</details>

<details><summary><b>3.</b> Rastgele ve coğrafi çapraz doğrulama arasında neden bu kadar fark çıktı?</summary>

Komşu mahalleler birbirine çok benzer. Rastgele bölmede sınav mahallesinin eğitimde **ortalama 1,2 km** uzakta bir komşusu var; coğrafi bölmede **11,1 km**. Model komşusunu “hatırlayarak” kolay sınav verir; yeni bölgede bu avantaj kaybolur (XGBoost: 45,9 → 63,3 bin $).
</details>

<details><summary><b>4.</b> SGD Regressor neden standart ölçekle “patladı”?</summary>

`kisi_hane` gibi sütunlarda **1.243'lük uç değerler** var; standartlaştırınca yüzlerce standart sapma uzaklıkta kalıyorlar ve gradyan inişi **ıraksıyor**. Log veya kuantil ölçek bu uçları sıkıştırır.
</details>

<details><summary><b>5.</b> Bir modelin XGBoost'tan farkının bootstrap aralığı sıfırı içeriyorsa ne anlama gelir?</summary>

Farkın **şanstan ayırt edilemediği**; mevcut örneklemde o modelin XGBoost'tan daha iyi/daha kötü olduğuna güvenle söylenemez (örnek: LightGBM −15 $, HistGB +145 $).
</details>

<details><summary><b>6.</b> Öğrenme hızını 0,1'den 0,01'e düşürürsek ne yapmalıyız?</summary>

Her ağaç daha küçük adım attığı için **kabaca 10 kat daha çok ağaç** gerekir; erken durdurma bunu otomatik bulur. Bu projede η = 0,01 1.500 turda bile bitmemişti.
</details>

<details><summary><b>7.</b> Kuantil regresyonla ürettiğimiz “%80 aralık” testte %67'yi kapsadı. Ne yaparız?</summary>

Aralık **fazla dar** (aşırı güvenli). Kalibrasyon kümesinde **uyumlu düzeltme (CQR)** ile aralığı iki yana genişletiriz: kapsama %79'a çıktı. Ama alt gruplarda (yüksek gelir, tavanlı) kapsama hâlâ eşit değil.
</details>

<details><summary><b>8.</b> Fazla tahminin maliyeti eksik tahminin 2 katıysa en iyi nokta tahmini hangisidir?</summary>

**τ = c_eksik / (c_eksik + c_fazla) = 1/3 kuantili.** Testte ortalama maliyet %9 düştü; fazla tahmin oranı %57 → %37,5.
</details>
