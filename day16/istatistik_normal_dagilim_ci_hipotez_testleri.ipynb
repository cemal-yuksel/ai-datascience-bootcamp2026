## 🏢 KampüsSepeti Veri Setinin Oluşturulması

<div style="background: linear-gradient(to right, #E3F2FD, #BBDEFB); padding: 20px; border-radius: 10px; margin: 20px 0;">
    <h3 style="color: #1565C0;">🎯 Gerçekçi Simülasyon Stratejisi</h3>
    <p style="font-size: 15px; line-height: 1.8;">
        Gerçek bir YBS sisteminden geliyormuş gibi <b>3 farklı veri seti</b> oluşturacağız:
    </p>
    <ol style="font-size: 15px; line-height: 1.8;">
        <li><b>🚚 Teslimat Süreleri:</b> Normal dağılıma YAKIN (μ=28 dk, σ=6 dk)</li>
        <li><b>💰 Sepet Tutarları:</b> Sağa ÇARPIK (log-normal) - gerçekçi olmayan!</li>
        <li><b>⏱️ Restoran Hazırlık:</b> Normal dağılım (μ=15 dk, σ=4 dk)</li>
    </ol>
    <p style="background-color: #FFF9C4; padding: 10px; border-radius: 5px; margin-top: 10px;">
        <b>💡 YBS İçgörüsü:</b> Bu farklılıklar sayesinde "her veri normal dağılmaz" gerçeğini göreceğiz!
    </p>
</div>

````python
# ============================================================================
# 🏗️ KAMPÜSSEPETİ VERİ SETİ OLUŞTURMA
# ============================================================================

# Simülasyon parametreleri
n_siparisler = 5000  # 5000 sipariş simüle edeceğiz (1 aylık veri)

print("🏗️ KampüsSepeti veri seti oluşturuluyor...\n")
print("="*70)

# -------------------------------------------------------------------------
# 1️⃣ TESLİMAT SÜRELERİ (Normal Dağılıma YAKIN)
# -------------------------------------------------------------------------
# Ortalama: 28 dakika, Standart Sapma: 6 dakika
# Mantık: Çoğu teslimat 22-34 dk arasında, uçlar nadir

teslimat_ortalama = 28
teslimat_std = 6

teslimat_sureleri = np.random.normal(
    loc=teslimat_ortalama,      # μ (mu) - ortalama
    scale=teslimat_std,          # σ (sigma) - standart sapma
    size=n_siparisler            # Üretilecek gözlem sayısı
)

# Gerçekçilik için: Negatif değerleri 5 dk'ya, çok yüksekleri 60 dk'ya sınırla
# (Gerçek dünyada teslimat 0'dan küçük olamaz!)
teslimat_sureleri = np.clip(teslimat_sureleri, 5, 60)

print("✅ 1. TESLİMAT SÜRELERİ oluşturuldu")
print(f"   📊 Hedef: μ={teslimat_ortalama} dk, σ={teslimat_std} dk")
print(f"   📈 Gerçek: μ={teslimat_sureleri.mean():.2f} dk, σ={teslimat_sureleri.std():.2f} dk")
print(f"   📏 Min: {teslimat_sureleri.min():.1f} dk | Max: {teslimat_sureleri.max():.1f} dk\n")

# -------------------------------------------------------------------------
# 2️⃣ SEPET TUTARLARI (SAĞA ÇARPIK - Log-Normal)
# -------------------------------------------------------------------------
# Gerçek dünyada sepet tutarları normal DEĞİLDİR!
# Çoğu öğrenci küçük harcama yapar, az sayıda büyük sipariş

# Log-normal dağılım kullanıyoruz (sağa çarpık)
sepet_median = 120  # Medyan sepet tutarı (TL)
sepet_std_log = 0.5  # Log ölçeğinde standart sapma

sepet_tutarlari = np.random.lognormal(
    mean=np.log(sepet_median),   # Log ölçeğinde ortalama
    sigma=sepet_std_log,         # Log ölçeğinde std sapma
    size=n_siparisler
)

# Gerçekçi sınırlar: Min 30 TL, Max 500 TL
sepet_tutarlari = np.clip(sepet_tutarlari, 30, 500)

print("✅ 2. SEPET TUTARLARI oluşturuldu")
print(f"   💰 Hedef Medyan: {sepet_median} TL")
print(f"   📊 Gerçek: Ortalama={sepet_tutarlari.mean():.2f} TL, Medyan={np.median(sepet_tutarlari):.2f} TL")
print(f"   📏 Min: {sepet_tutarlari.min():.1f} TL | Max: {sepet_tutarlari.max():.1f} TL")
print(f"   ⚠️ DİKKAT: Bu veri SAĞA ÇARPIK - Normal dağılım DEĞİL!\n")

# -------------------------------------------------------------------------
# 3️⃣ RESTORAN HAZIRLIK SÜRELERİ (Normal Dağılım)
# -------------------------------------------------------------------------
# Restoranlar standart çalışırsa bu metrik normale yakın olabilir

hazirlik_ortalama = 15
hazirlik_std = 4

hazirlik_sureleri = np.random.normal(
    loc=hazirlik_ortalama,
    scale=hazirlik_std,
    size=n_siparisler
)

# Gerçekçi sınırlar: Min 5 dk, Max 35 dk
hazirlik_sureleri = np.clip(hazirlik_sureleri, 5, 35)

print("✅ 3. RESTORAN HAZIRLIK SÜRELERİ oluşturuldu")
print(f"   🍳 Hedef: μ={hazirlik_ortalama} dk, σ={hazirlik_std} dk")
print(f"   📈 Gerçek: μ={hazirlik_sureleri.mean():.2f} dk, σ={hazirlik_sureleri.std():.2f} dk")
print(f"   📏 Min: {hazirlik_sureleri.min():.1f} dk | Max: {hazirlik_sureleri.max():.1f} dk\n")

# -------------------------------------------------------------------------
# 📊 VERİLERİ PANDAS DATAFRAME'E DÖNÜŞTÜRME
# -------------------------------------------------------------------------

df_kampussepeti = pd.DataFrame({
    'siparis_id': range(1, n_siparisler + 1),
    'teslimat_suresi_dk': teslimat_sureleri,
    'sepet_tutari_tl': sepet_tutarlari,
    'hazirlik_suresi_dk': hazirlik_sureleri
})

# Ek hesaplanan değişkenler
df_kampussepeti['toplam_sure_dk'] = (
    df_kampussepeti['hazirlik_suresi_dk'] + 
    df_kampussepeti['teslimat_suresi_dk']
)

print("="*70)
print("\n✨ VERİ SETİ HAZIR!\n")
print(f"📦 Toplam Sipariş: {len(df_kampussepeti):,}")
print(f"📊 Değişken Sayısı: {df_kampussepeti.shape[1]}")
print("\n🔍 İlk 5 satıra göz atalım:\n")
print(df_kampussepeti.head())
````

## 📊 Temel İstatistiksel Özet

<div style="background-color: #E8EAF6; padding: 15px; border-left: 5px solid #3F51B5; border-radius: 5px; margin: 15px 0;">
    <h4>📈 Tanımlayıcı İstatistikler</h4>
    <p>Veri setimizin genel özelliklerini inceleyelim. Bu adım <b>her analizin başlangıcıdır</b>!</p>

````python
# ============================================================================
# 📊 TANIMLAYICI İSTATİSTİKLER
# ============================================================================

print("📊 KAMPÜSSEPETİ VERİ SETİ - TANIMLAYICI İSTATİSTİKLER")
print("="*70)

# Pandas describe() fonksiyonu: Temel istatistikleri hesaplar
# count, mean, std, min, %25, %50 (medyan), %75, max
istatistikler = df_kampussepeti.describe()

print("\n", istatistikler.round(2))

print("\n" + "="*70)
print("\n💡 ÖNEMLİ GÖZLEMLER:\n")

# Teslimat süreleri analizi
tes_mean = df_kampussepeti['teslimat_suresi_dk'].mean()
tes_median = df_kampussepeti['teslimat_suresi_dk'].median()
print(f"🚚 TESLİMAT SÜRELERİ:")
print(f"   • Ortalama ≈ Medyan? {tes_mean:.2f} ≈ {tes_median:.2f} → {'✅ SİMETRİK (Normal olabilir)' if abs(tes_mean - tes_median) < 1 else '⚠️ ÇARPIK'}")

# Sepet tutarları analizi
sep_mean = df_kampussepeti['sepet_tutari_tl'].mean()
sep_median = df_kampussepeti['sepet_tutari_tl'].median()
print(f"\n💰 SEPET TUTARLARI:")
print(f"   • Ortalama > Medyan? {sep_mean:.2f} > {sep_median:.2f} → {'⚠️ SAĞA ÇARPIK (Normal DEĞİL!)' if sep_mean > sep_median + 5 else '✅ Simetrik'}")

# Hazırlık süreleri analizi
haz_mean = df_kampussepeti['hazirlik_suresi_dk'].mean()
haz_median = df_kampussepeti['hazirlik_suresi_dk'].median()
print(f"\n🍳 HAZIRLIK SÜRELERİ:")
print(f"   • Ortalama ≈ Medyan? {haz_mean:.2f} ≈ {haz_median:.2f} → {'✅ SİMETRİK (Normal olabilir)' if abs(haz_mean - haz_median) < 0.5 else '⚠️ ÇARPIK'}")

print("\n" + "="*70)
````

## 📊 İlk Görselleştirme: Histogram Karşılaştırması

<div style="background-color: #FFF3E0; padding: 15px; border-left: 5px solid #FF6F00; border-radius: 5px; margin: 15px 0;">
    <h4>🎨 Görselleştirme Hedefi</h4>
    <p>3 değişkenimizi yan yana histogram ile görselleştirerek <b>hangisinin normal dağılıma benzediğini</b> göreceğiz.</p>
    <ul>
        <li>🟢 <b>Çan eğrisi</b> görüyorsak → Normal dağılıma YAKIN</li>
        <li>🔴 <b>Sağa kuyruk</b> görüyorsak → ÇARPIK dağılım</li>
    </ul>
</div>

````python
# ============================================================================
# 📊 HİSTOGRAM GÖRSELLEŞTİRMESİ - 3 DEĞİŞKEN KARŞILAŞTIRMASI
# ============================================================================

# 1 satır, 3 sütunluk subplot yapısı oluştur
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Pastel renkler tanımla (manuel kontrol için)
renkler = ['#B8E6B8', '#FFD6A5', '#CFBAF0']

# -------------------------------------------------------------------------
# 1️⃣ TESLİMAT SÜRELERİ
# -------------------------------------------------------------------------
axes[0].hist(df_kampussepeti['teslimat_suresi_dk'], 
             bins=40,                    # Histogram bar sayısı
             color=renkler[0], 
             edgecolor='darkgreen',      # Bar kenarı rengi
             alpha=0.7,                  # Şeffaflık
             density=True)               # Olasılık yoğunluğu olarak göster

# Normal dağılım eğrisi ekle (teorik karşılaştırma için)
mu_tes = df_kampussepeti['teslimat_suresi_dk'].mean()
sigma_tes = df_kampussepeti['teslimat_suresi_dk'].std()
x_tes = np.linspace(mu_tes - 4*sigma_tes, mu_tes + 4*sigma_tes, 100)
axes[0].plot(x_tes, stats.norm.pdf(x_tes, mu_tes, sigma_tes), 
             'r-', linewidth=2.5, label='Teorik Normal Dağılım')

axes[0].axvline(mu_tes, color='red', linestyle='--', linewidth=2, label=f'μ={mu_tes:.1f} dk')
axes[0].set_title('Teslimat Süreleri\n(Normal Dağılıma Yakın)', fontsize=13, fontweight='bold', color='darkgreen')
axes[0].set_xlabel('Süre (dakika)', fontsize=11)
axes[0].set_ylabel('Yoğunluk', fontsize=11)
axes[0].legend(fontsize=9)
axes[0].grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# 2️⃣ SEPET TUTARLARI
# -------------------------------------------------------------------------
axes[1].hist(df_kampussepeti['sepet_tutari_tl'], 
             bins=40, 
             color=renkler[1], 
             edgecolor='darkorange', 
             alpha=0.7,
             density=True)

mu_sep = df_kampussepeti['sepet_tutari_tl'].mean()
median_sep = df_kampussepeti['sepet_tutari_tl'].median()

axes[1].axvline(mu_sep, color='red', linestyle='--', linewidth=2, label=f'Ortalama={mu_sep:.1f} TL')
axes[1].axvline(median_sep, color='blue', linestyle='--', linewidth=2, label=f'Medyan={median_sep:.1f} TL')
axes[1].set_title('Sepet Tutarları\n(SAĞA ÇARPIK - Normal Değil!)', fontsize=13, fontweight='bold', color='darkorange')
axes[1].set_xlabel('Tutar (TL)', fontsize=11)
axes[1].set_ylabel('Yoğunluk', fontsize=11)
axes[1].legend(fontsize=9)
axes[1].grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# 3️⃣ HAZIRLIK SÜRELERİ
# -------------------------------------------------------------------------
axes[2].hist(df_kampussepeti['hazirlik_suresi_dk'], 
             bins=35, 
             color=renkler[2], 
             edgecolor='darkviolet', 
             alpha=0.7,
             density=True)

mu_haz = df_kampussepeti['hazirlik_suresi_dk'].mean()
sigma_haz = df_kampussepeti['hazirlik_suresi_dk'].std()
x_haz = np.linspace(mu_haz - 4*sigma_haz, mu_haz + 4*sigma_haz, 100)
axes[2].plot(x_haz, stats.norm.pdf(x_haz, mu_haz, sigma_haz), 
             'r-', linewidth=2.5, label='Teorik Normal Dağılım')

axes[2].axvline(mu_haz, color='red', linestyle='--', linewidth=2, label=f'μ={mu_haz:.1f} dk')
axes[2].set_title('Hazırlık Süreleri\n(Normal Dağılıma Yakın)', fontsize=13, fontweight='bold', color='darkviolet')
axes[2].set_xlabel('Süre (dakika)', fontsize=11)
axes[2].set_ylabel('Yoğunluk', fontsize=11)
axes[2].legend(fontsize=9)
axes[2].grid(True, alpha=0.3, linestyle=':')

# Genel başlık ve layout ayarları
plt.suptitle('KampüsSepeti: Hangi Metrikler Normal Dağılıyor?', 
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("\n📊 GRAFİK YORUMU:")
print("="*70)
print("✅ Sol ve Sağ grafikler → ÇAN EĞRİSİ şeklinde (simetrik)")
print("⚠️ Orta grafik → SAĞ KUYRUK uzun (çarpık - ortalama > medyan)")
print("\n💡 YBS İçgörüsü: Sepet tutarlarına normal dağılım varsayımı yapmak YANLIŞ olur!")
````

## 🎨 Python ile 68-95-99.7 Kuralını Görselleştirme

<div style="background-color: #FFF3E0; padding: 15px; border-left: 5px solid #FF6F00; border-radius: 5px; margin: 15px 0;">
    <h4>🎯 Bu Grafikte Göreceğimiz:</h4>
    <ul>
        <li>📊 Normal dağılım eğrisi (çan şekli)</li>
        <li>🎨 Her aralık için <b>farklı renkli gölgelendirme</b></li>
        <li>📏 μ ± 1σ, ± 2σ, ± 3σ sınırları</li>
        <li>💯 Her bölgenin kapladığı <b>yüzde değerleri</b></li>
    </ul>
</div>

````python
# ============================================================================
# 🎨 68-95-99.7 KURALI GÖRSELLEŞTİRME
# ============================================================================

# KampüsSepeti teslimat verileri kullanarak gerçek örnek
mu = df_kampussepeti['teslimat_suresi_dk'].mean()    # Ortalama
sigma = df_kampussepeti['teslimat_suresi_dk'].std()  # Standart sapma

print("📊 KAMPÜSSEPETİ TESLİMAT SÜRELERİ")
print("="*70)
print(f"μ (Ortalama): {mu:.2f} dakika")
print(f"σ (Std Sapma): {sigma:.2f} dakika\n")

# Normal dağılım için x değerleri oluştur (μ ± 4σ aralığında)
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
# Normal dağılım yoğunluk fonksiyonu (PDF)
y = stats.norm.pdf(x, mu, sigma)

# Grafik oluştur
plt.figure(figsize=(14, 8))

# -------------------------------------------------------------------------
# ANA NORMAL DAĞILIM EĞRİSİ
# -------------------------------------------------------------------------
plt.plot(x, y, 'black', linewidth=3, label='Normal Dağılım Eğrisi', zorder=5)

# -------------------------------------------------------------------------
# 🟢 μ ± 1σ (68%) - YEŞİL ALAN
# -------------------------------------------------------------------------
x_1sigma = x[(x >= mu - sigma) & (x <= mu + sigma)]
y_1sigma = stats.norm.pdf(x_1sigma, mu, sigma)
plt.fill_between(x_1sigma, y_1sigma, alpha=0.6, color='#81C784', 
                 label=f'μ ± 1σ: [{mu-sigma:.1f}, {mu+sigma:.1f}] dk (68%)', zorder=3)

# -------------------------------------------------------------------------
# 🟡 μ ± 2σ (95%) - SARI ALAN (1σ ile 2σ arası)
# -------------------------------------------------------------------------
# Sol taraf: μ - 2σ ile μ - 1σ arası
x_2sigma_left = x[(x >= mu - 2*sigma) & (x <= mu - sigma)]
y_2sigma_left = stats.norm.pdf(x_2sigma_left, mu, sigma)
plt.fill_between(x_2sigma_left, y_2sigma_left, alpha=0.5, color='#FFF176',
                 zorder=2)

# Sağ taraf: μ + 1σ ile μ + 2σ arası
x_2sigma_right = x[(x >= mu + sigma) & (x <= mu + 2*sigma)]
y_2sigma_right = stats.norm.pdf(x_2sigma_right, mu, sigma)
plt.fill_between(x_2sigma_right, y_2sigma_right, alpha=0.5, color='#FFF176',
                 label=f'μ ± 2σ: [{mu-2*sigma:.1f}, {mu+2*sigma:.1f}] dk (95%)', zorder=2)

# -------------------------------------------------------------------------
# 🔴 μ ± 3σ (99.7%) - KIRMIZI ALAN (2σ ile 3σ arası)
# -------------------------------------------------------------------------
# Sol taraf: μ - 3σ ile μ - 2σ arası
x_3sigma_left = x[(x >= mu - 3*sigma) & (x <= mu - 2*sigma)]
y_3sigma_left = stats.norm.pdf(x_3sigma_left, mu, sigma)
plt.fill_between(x_3sigma_left, y_3sigma_left, alpha=0.4, color='#FF8A80',
                 zorder=1)

# Sağ taraf: μ + 2σ ile μ + 3σ arası
x_3sigma_right = x[(x >= mu + 2*sigma) & (x <= mu + 3*sigma)]
y_3sigma_right = stats.norm.pdf(x_3sigma_right, mu, sigma)
plt.fill_between(x_3sigma_right, y_3sigma_right, alpha=0.4, color='#FF8A80',
                 label=f'μ ± 3σ: [{mu-3*sigma:.1f}, {mu+3*sigma:.1f}] dk (99.7%)', zorder=1)

# -------------------------------------------------------------------------
# DİKEY ÇIZGILER - Ortalama ve standart sapma işaretleri
# -------------------------------------------------------------------------
plt.axvline(mu, color='darkred', linestyle='-', linewidth=2.5, 
            label=f'μ = {mu:.1f} dk', zorder=4)

# ±1σ çizgileri
plt.axvline(mu - sigma, color='green', linestyle='--', linewidth=2, alpha=0.7)
plt.axvline(mu + sigma, color='green', linestyle='--', linewidth=2, alpha=0.7)

# ±2σ çizgileri
plt.axvline(mu - 2*sigma, color='orange', linestyle='--', linewidth=2, alpha=0.7)
plt.axvline(mu + 2*sigma, color='orange', linestyle='--', linewidth=2, alpha=0.7)

# ±3σ çizgileri
plt.axvline(mu - 3*sigma, color='red', linestyle='--', linewidth=2, alpha=0.7)
plt.axvline(mu + 3*sigma, color='red', linestyle='--', linewidth=2, alpha=0.7)

# -------------------------------------------------------------------------
# YÜZDE DEĞERLERİNİ GRAFİĞE YAZI OLARAK EKLEME
# -------------------------------------------------------------------------
max_y = max(y)

# %68 yazısı (yeşil bölge)
plt.text(mu, max_y * 0.4, '68%', fontsize=20, fontweight='bold', 
         ha='center', va='center', color='darkgreen',
         bbox=dict(boxstyle='round,pad=0.5', facecolor='white', edgecolor='darkgreen', linewidth=2))

# %95 yazıları (sarı bölgeler)
plt.text(mu - 1.5*sigma, max_y * 0.15, '13.5%', fontsize=14, fontweight='bold',
         ha='center', color='#F57C00')
plt.text(mu + 1.5*sigma, max_y * 0.15, '13.5%', fontsize=14, fontweight='bold',
         ha='center', color='#F57C00')

# %99.7 yazıları (kırmızı bölgeler)
plt.text(mu - 2.5*sigma, max_y * 0.05, '2.35%', fontsize=12, fontweight='bold',
         ha='center', color='#D32F2F')
plt.text(mu + 2.5*sigma, max_y * 0.05, '2.35%', fontsize=12, fontweight='bold',
         ha='center', color='#D32F2F')

# -------------------------------------------------------------------------
# GRAFİK AYARLARI
# -------------------------------------------------------------------------
plt.title('68-95-99.7 Kuralı: KampüsSepeti Teslimat Süreleri\n' + 
          f'(μ={mu:.1f} dk, σ={sigma:.1f} dk)', 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Teslimat Süresi (dakika)', fontsize=13, fontweight='bold')
plt.ylabel('Olasılık Yoğunluğu', fontsize=13, fontweight='bold')
plt.legend(loc='upper left', fontsize=10, framealpha=0.95)
plt.grid(True, alpha=0.3, linestyle=':', linewidth=0.5)
plt.xlim(mu - 4*sigma, mu + 4*sigma)

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# GERÇEK VERİYLE DOĞRULAMA
# -------------------------------------------------------------------------
print("\n" + "="*70)
print("✅ GERÇEK VERİYLE DOĞRULAMA (KampüsSepeti verileri)")
print("="*70)

# Verimizdeki gerçek yüzdeleri hesapla
total = len(df_kampussepeti)

# μ ± 1σ aralığında kaç gözlem var?
count_1sigma = len(df_kampussepeti[
    (df_kampussepeti['teslimat_suresi_dk'] >= mu - sigma) & 
    (df_kampussepeti['teslimat_suresi_dk'] <= mu + sigma)
])
pct_1sigma = (count_1sigma / total) * 100

# μ ± 2σ aralığında kaç gözlem var?
count_2sigma = len(df_kampussepeti[
    (df_kampussepeti['teslimat_suresi_dk'] >= mu - 2*sigma) & 
    (df_kampussepeti['teslimat_suresi_dk'] <= mu + 2*sigma)
])
pct_2sigma = (count_2sigma / total) * 100

# μ ± 3σ aralığında kaç gözlem var?
count_3sigma = len(df_kampussepeti[
    (df_kampussepeti['teslimat_suresi_dk'] >= mu - 3*sigma) & 
    (df_kampussepeti['teslimat_suresi_dk'] <= mu + 3*sigma)
])
pct_3sigma = (count_3sigma / total) * 100

print(f"\n🟢 μ ± 1σ [{mu-sigma:.1f}, {mu+sigma:.1f}] dk:")
print(f"   Teorik: %68 | Gerçek: %{pct_1sigma:.1f} → {'✅ Uyumlu!' if abs(pct_1sigma - 68) < 5 else '⚠️ Sapma var'}")

print(f"\n🟡 μ ± 2σ [{mu-2*sigma:.1f}, {mu+2*sigma:.1f}] dk:")
print(f"   Teorik: %95 | Gerçek: %{pct_2sigma:.1f} → {'✅ Uyumlu!' if abs(pct_2sigma - 95) < 3 else '⚠️ Sapma var'}")

print(f"\n🔴 μ ± 3σ [{mu-3*sigma:.1f}, {mu+3*sigma:.1f}] dk:")
print(f"   Teorik: %99.7 | Gerçek: %{pct_3sigma:.1f} → {'✅ Uyumlu!' if abs(pct_3sigma - 99.7) < 1 else '⚠️ Sapma var'}")

print("\n" + "="*70)
print("💡 Yorum: Gerçek verilerimiz teorik normal dağılıma ne kadar yakın!")
print("   Bu, teslimat sürelerinin normal dağılım varsayımını destekliyor.")
````

### 🎨 Görsel Karşılaştırma: Farklı μ ve σ Değerleri

<div style="background-color: #FFF3E0; padding: 15px; border-left: 5px solid #FF6F00; border-radius: 5px; margin: 15px 0;">
    <h4>📊 3 Senaryo Göreceğiz:</h4>

````python
# ============================================================================
# 🎨 PARAMETRE ETKİSİ GÖRSELLEŞTİRMESİ - 3 SENARYO
# ============================================================================

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Normal Dağılımda μ ve σ Parametrelerinin Etkisi\n', 
             fontsize=18, fontweight='bold', y=0.995)

# x ekseni için geniş bir aralık
x_range = np.linspace(0, 60, 1000)

# -------------------------------------------------------------------------
# 1️⃣ FARKLI μ, AYNI σ - Kampüs A vs Kampüs B
# -------------------------------------------------------------------------
ax1 = axes[0, 0]

# Kampüs A: Şehir merkezi (hızlı teslimat)
mu_A = 22
sigma_A = 5
y_A = stats.norm.pdf(x_range, mu_A, sigma_A)
ax1.plot(x_range, y_A, linewidth=3, color='#66BB6A', label=f'Kampüs A: μ={mu_A}, σ={sigma_A}')
ax1.fill_between(x_range, y_A, alpha=0.3, color='#66BB6A')
ax1.axvline(mu_A, color='#2E7D32', linestyle='--', linewidth=2, alpha=0.8)

# Kampüs B: Kırsal alan (yavaş teslimat)
mu_B = 35
sigma_B = 5
y_B = stats.norm.pdf(x_range, mu_B, sigma_B)
ax1.plot(x_range, y_B, linewidth=3, color='#42A5F5', label=f'Kampüs B: μ={mu_B}, σ={sigma_B}')
ax1.fill_between(x_range, y_B, alpha=0.3, color='#42A5F5')
ax1.axvline(mu_B, color='#1565C0', linestyle='--', linewidth=2, alpha=0.8)

ax1.set_title('Senaryo 1: Farklı Ortalama (μ), Aynı Dağınıklık (σ)\n' + 
              'Kampüs A daha hızlı, ama değişkenlik aynı', 
              fontsize=12, fontweight='bold', pad=10)
ax1.set_xlabel('Teslimat Süresi (dk)', fontsize=11)
ax1.set_ylabel('Yoğunluk', fontsize=11)
ax1.legend(loc='upper right', fontsize=10)
ax1.grid(True, alpha=0.3, linestyle=':')
ax1.set_xlim(0, 55)

# -------------------------------------------------------------------------
# 2️⃣ AYNI μ, FARKLI σ - Normal Gün vs Kampanya Günü
# -------------------------------------------------------------------------
ax2 = axes[0, 1]

# Normal gün: Düşük değişkenlik
mu_normal = 28
sigma_normal = 4
y_normal = stats.norm.pdf(x_range, mu_normal, sigma_normal)
ax2.plot(x_range, y_normal, linewidth=3, color='#9575CD', 
         label=f'Normal Gün: σ={sigma_normal} (Düşük değişkenlik)')
ax2.fill_between(x_range, y_normal, alpha=0.4, color='#9575CD')

# Kampanya günü: Yüksek değişkenlik
sigma_kampanya = 10
y_kampanya = stats.norm.pdf(x_range, mu_normal, sigma_kampanya)
ax2.plot(x_range, y_kampanya, linewidth=3, color='#FF7043', 
         label=f'Kampanya Günü: σ={sigma_kampanya} (Yüksek değişkenlik)')
ax2.fill_between(x_range, y_kampanya, alpha=0.4, color='#FF7043')

# Ortak ortalama çizgisi
ax2.axvline(mu_normal, color='black', linestyle='-', linewidth=2.5, 
            label=f'μ={mu_normal} (Her ikisi de aynı)', alpha=0.7)

ax2.set_title('Senaryo 2: Aynı Ortalama (μ), Farklı Dağınıklık (σ)\n' + 
              'Kampanyada ortalama aynı ama öngörülebilirlik azalıyor!', 
              fontsize=12, fontweight='bold', pad=10)
ax2.set_xlabel('Teslimat Süresi (dk)', fontsize=11)
ax2.set_ylabel('Yoğunluk', fontsize=11)
ax2.legend(loc='upper right', fontsize=9)
ax2.grid(True, alpha=0.3, linestyle=':')
ax2.set_xlim(0, 60)

# Açıklama kutusu
ax2.text(45, max(y_normal)*0.7, 
         '💡 Mor eğri daha dar:\nTahmin edilebilir\n\n🔥 Turuncu daha geniş:\nRiskli, değişken',
         bbox=dict(boxstyle='round,pad=0.8', facecolor='#FFF9C4', 
                   edgecolor='#F57C00', linewidth=2),
         fontsize=9, verticalalignment='top')

# -------------------------------------------------------------------------
# 3️⃣ GERÇEK VERİ: Teslimat vs Hazırlık Süreleri
# -------------------------------------------------------------------------
ax3 = axes[1, 0]

# Teslimat süreleri
mu_tes = df_kampussepeti['teslimat_suresi_dk'].mean()
sigma_tes = df_kampussepeti['teslimat_suresi_dk'].std()
y_tes = stats.norm.pdf(x_range, mu_tes, sigma_tes)
ax3.plot(x_range, y_tes, linewidth=3, color='#26A69A', 
         label=f'Teslimat: μ={mu_tes:.1f}, σ={sigma_tes:.1f}')
ax3.fill_between(x_range, y_tes, alpha=0.4, color='#26A69A')
ax3.axvline(mu_tes, color='#00695C', linestyle='--', linewidth=2)

# Hazırlık süreleri
mu_haz = df_kampussepeti['hazirlik_suresi_dk'].mean()
sigma_haz = df_kampussepeti['hazirlik_suresi_dk'].std()
y_haz = stats.norm.pdf(x_range, mu_haz, sigma_haz)
ax3.plot(x_range, y_haz, linewidth=3, color='#EC407A', 
         label=f'Hazırlık: μ={mu_haz:.1f}, σ={sigma_haz:.1f}')
ax3.fill_between(x_range, y_haz, alpha=0.4, color='#EC407A')
ax3.axvline(mu_haz, color='#AD1457', linestyle='--', linewidth=2)

ax3.set_title('Senaryo 3: KampüsSepeti Gerçek Verileri\n' + 
              'Teslimat daha uzun ve daha değişken', 
              fontsize=12, fontweight='bold', pad=10)
ax3.set_xlabel('Süre (dakika)', fontsize=11)
ax3.set_ylabel('Yoğunluk', fontsize=11)
ax3.legend(loc='upper right', fontsize=10)
ax3.grid(True, alpha=0.3, linestyle=':')
ax3.set_xlim(0, 50)

# -------------------------------------------------------------------------
# 4️⃣ ÖZETLEYİCİ TABLO
# -------------------------------------------------------------------------
ax4 = axes[1, 1]
ax4.axis('off')  # Eksen çizgilerini kapat

# Tablo verisi
tablo_data = [
    ['Senaryo', 'Parametre', 'Etki', 'YBS Yorumu'],
    ['Kampüs A vs B', 'μ farklı', 'Grafik sağa/sola kayar', 'Konum etkisi'],
    ['Normal vs Kampanya', 'σ farklı', 'Grafik daralır/genişler', 'Risk değişir'],
    ['Teslimat vs Hazırlık', 'Her ikisi farklı', 'Hem yer hem genişlik', 'Farklı süreçler']
]

# Tablo oluştur
table = ax4.table(cellText=tablo_data, cellLoc='left', loc='center',
                  colWidths=[0.25, 0.2, 0.3, 0.25])

# Tablo stilini ayarla
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 3)

# Başlık satırını renklendir
for i in range(4):
    cell = table[(0, i)]
    cell.set_facecolor('#7E57C2')
    cell.set_text_props(weight='bold', color='white', fontsize=11)

# Diğer satırları pastel renklerle
renkler_satir = ['#E1BEE7', '#C5CAE9', '#B2DFDB']
for i in range(1, 4):
    for j in range(4):
        cell = table[(i, j)]
        cell.set_facecolor(renkler_satir[i-1])
        cell.set_edgecolor('#9E9E9E')
        cell.set_linewidth(1.5)

ax4.set_title('📊 Özet: μ ve σ\'nın YBS\'deki Anlamı', 
              fontsize=13, fontweight='bold', pad=20)

# Açıklama kutusu
ax4.text(0.5, 0.15, 
         '💡 Temel Prensip:\n' +
         '• μ artarsa → Süreçler yavaşlıyor\n' +
         '• σ artarsa → Süreçler öngörülemez oluyor\n' +
         '• YBS hedefi: μ\'yü azalt, σ\'yı kontrol et!',
         transform=ax4.transAxes,
         bbox=dict(boxstyle='round,pad=1', facecolor='#FFF9C4', 
                   edgecolor='#F57C00', linewidth=2),
         fontsize=11, verticalalignment='top', ha='center')

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# YORUM ve ANALİZ
# -------------------------------------------------------------------------
print("\n" + "="*70)
print("📊 PARAMETRE ANALİZİ SONUÇLARI")
print("="*70)

print("\n🎯 SENARYO 1: Farklı Kampüsler")
print(f"   • Kampüs A (μ={mu_A}): Merkeze yakın, hızlı teslimat")
print(f"   • Kampüs B (μ={mu_B}): Uzak konum, yavaş teslimat")
print(f"   • σ aynı ({sigma_A}): Her iki kampüste de değişkenlik aynı seviyede")

print("\n🎯 SENARYO 2: Normal vs Kampanya")
print(f"   • μ aynı ({mu_normal}): Ortalama teslimat süresi değişmedi")
print(f"   • Normal gün σ={sigma_normal}: Dar eğri, öngörülebilir")
print(f"   • Kampanya σ={sigma_kampanya}: Geniş eğri, belirsiz")
print("   ⚠️ YBS Riski: Müşteri memnuniyeti σ ile ters orantılı!")

print("\n🎯 SENARYO 3: Gerçek KampüsSepeti")
print(f"   • Teslimat: μ={mu_tes:.1f}, σ={sigma_tes:.1f}")
print(f"   • Hazırlık: μ={mu_haz:.1f}, σ={sigma_haz:.1f}")
print(f"   • Teslimat daha uzun ve {sigma_tes/sigma_haz:.1f}x daha değişken")
print("   💡 İçgörü: Teslimat süreci daha fazla iyileştirme gerektirir!")

print("\n" + "="*70)
````

## 💼 YBS Vaka Çalışması: SLA Hedefi Belirleme

<div style="background: linear-gradient(to right, #B2EBF2, #80DEEA); padding: 20px; border-radius: 10px; margin: 20px 0;">

### 📊 Mermaid: SLA Karar Süreci


````python
# ============================================================================
# 💼 YBS VAKA: SLA HEDEFİ ANALİZİ
# ============================================================================

print("🎯 KAMPÜSSEPETİ SLA ANALİZİ")
print("="*70)

# Mevcut parametreler
mu = df_kampussepeti['teslimat_suresi_dk'].mean()
sigma = df_kampussepeti['teslimat_suresi_dk'].std()

print(f"\n📊 Mevcut Durum:")
print(f"   μ (Ortalama): {mu:.2f} dakika")
print(f"   σ (Std Sapma): {sigma:.2f} dakika")

# Hedef: 35 dakikanın altında %90 teslimat
sla_limit = 35
hedef_oran = 90

print(f"\n🎯 SLA Hedefi:")
print(f"   Siparişlerin %{hedef_oran}'ı {sla_limit} dakikanın altında olmalı")

# -------------------------------------------------------------------------
# MEVCUT DURUMU HESAPLA
# -------------------------------------------------------------------------

# 35 dakikanın altında olan sipariş sayısı (gerçek veri)
gercek_altinda = len(df_kampussepeti[df_kampussepeti['teslimat_suresi_dk'] < sla_limit])
gercek_oran = (gercek_altinda / len(df_kampussepeti)) * 100

# Teorik hesaplama (normal dağılım varsayımı)
# Z-skoru: (X - μ) / σ
z_skor = (sla_limit - mu) / sigma
# P(X < 35) - kümülatif dağılım fonksiyonu (CDF)
teorik_oran = stats.norm.cdf(z_skor) * 100

print(f"\n📈 Mevcut Performans:")
print(f"   • Gerçek veriden: %{gercek_oran:.2f} sipariş {sla_limit} dk altında")
print(f"   • Teorik (Normal dağ.): %{teorik_oran:.2f}")
print(f"   • Z-skoru: {z_skor:.2f}")

# Hedefe ulaşıldı mı?
if gercek_oran >= hedef_oran:
    print(f"\n✅ HEDEF ULAŞILDI! %{gercek_oran:.1f} ≥ %{hedef_oran}")
    basari_durumu = "BAŞARILI"
    renk_basari = '#4CAF50'
else:
    eksik = hedef_oran - gercek_oran
    print(f"\n⚠️ HEDEF TUTMADI! %{gercek_oran:.1f} < %{hedef_oran}")
    print(f"   Hedefe %{eksik:.1f} puan eksik")
    basari_durumu = "İYİLEŞTİRME GEREKLİ"
    renk_basari = '#FF5722'

# -------------------------------------------------------------------------
# GÖRSELLEŞTİRME: SLA Hedefi
# -------------------------------------------------------------------------

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# SOL PANEL: Normal dağılım eğrisi ve SLA sınırı
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, mu, sigma)

ax1.plot(x, y, 'black', linewidth=3, label='Teslimat Süresi Dağılımı')
ax1.fill_between(x, y, alpha=0.3, color='lightblue')

# SLA sınırının solunu (hedef bölge) yeşil ile gölgele
x_hedef = x[x <= sla_limit]
y_hedef = stats.norm.pdf(x_hedef, mu, sigma)
ax1.fill_between(x_hedef, y_hedef, alpha=0.6, color='#81C784', 
                 label=f'{sla_limit} dk altı: %{teorik_oran:.1f}')

# SLA sınırının sağını (başarısız bölge) kırmızı ile gölgele
x_basarisiz = x[x > sla_limit]
y_basarisiz = stats.norm.pdf(x_basarisiz, mu, sigma)
ax1.fill_between(x_basarisiz, y_basarisiz, alpha=0.5, color='#EF5350',
                 label=f'{sla_limit} dk üstü: %{100-teorik_oran:.1f}')

# SLA sınır çizgisi
ax1.axvline(sla_limit, color='red', linestyle='--', linewidth=3, 
            label=f'SLA Sınırı = {sla_limit} dk')

# Ortalama çizgisi
ax1.axvline(mu, color='blue', linestyle='-', linewidth=2, 
            label=f'μ = {mu:.1f} dk', alpha=0.7)

ax1.set_title(f'SLA Hedefi: {sla_limit} Dakika\n(Z={z_skor:.2f}, P={teorik_oran:.1f}%)', 
              fontsize=13, fontweight='bold')
ax1.set_xlabel('Teslimat Süresi (dakika)', fontsize=11)
ax1.set_ylabel('Olasılık Yoğunluğu', fontsize=11)
ax1.legend(loc='upper right', fontsize=10)
ax1.grid(True, alpha=0.3, linestyle=':')

# SAĞ PANEL: Hedef vs Gerçek Karşılaştırma
kategoriler = ['Hedef SLA', 'Mevcut Durum']
degerler = [hedef_oran, gercek_oran]
renkler_bar = ['#FFC107', renk_basari]

bars = ax2.bar(kategoriler, degerler, color=renkler_bar, alpha=0.8, 
               edgecolor='black', linewidth=2)

# Her bar üzerine değer yaz
for bar, deger in zip(bars, degerler):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 1,
             f'%{deger:.1f}',
             ha='center', va='bottom', fontsize=16, fontweight='bold')

# Hedef çizgisi
ax2.axhline(hedef_oran, color='red', linestyle='--', linewidth=2.5, 
            label=f'Hedef: %{hedef_oran}', alpha=0.7)

# Başarı durumu etiketi
if gercek_oran >= hedef_oran:
    durum_text = f"✅ BAŞARILI\nHedef tutturuldu!"
    text_color = '#2E7D32'
    bg_color = '#C8E6C9'
else:
    durum_text = f"⚠️ İYİLEŞTİRME\nGEREKLİ\n{eksik:.1f} puan eksik"
    text_color = '#C62828'
    bg_color = '#FFCDD2'

ax2.text(0.5, 0.7, durum_text,
         transform=ax2.transAxes, fontsize=14, fontweight='bold',
         ha='center', va='center', color=text_color,
         bbox=dict(boxstyle='round,pad=1', facecolor=bg_color, 
                   edgecolor=text_color, linewidth=3))

ax2.set_title(f'SLA Hedef Karşılaştırması\n({sla_limit} dk altında olma oranı)', 
              fontsize=13, fontweight='bold')
ax2.set_ylabel('Yüzde (%)', fontsize=11)
ax2.set_ylim(0, 100)
ax2.legend(fontsize=10)
ax2.grid(True, alpha=0.3, axis='y', linestyle=':')

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# İYİLEŞTİRME ÖNERİLERİ
# -------------------------------------------------------------------------
print("\n" + "="*70)
print("💡 İYİLEŞTİRME ÖNERİLERİ")
print("="*70)

if gercek_oran < hedef_oran:
    print("\n🔧 Seçenek 1: ORTALAMA TESLİMAT SÜRESİNİ AZALT (μ'yü küçült)")
    
    # Hedef için gereken ortalama
    # %90 için Z ≈ 1.28 (tek taraflı)
    z_hedef = stats.norm.ppf(hedef_oran / 100)  # Hedef için gereken Z-skoru
    # X = μ + Z*σ  →  μ = X - Z*σ
    mu_gerekli = sla_limit - z_hedef * sigma
    mu_azaltma = mu - mu_gerekli
    
    print(f"   • Mevcut μ: {mu:.2f} dk")
    print(f"   • Gerekli μ: {mu_gerekli:.2f} dk")
    print(f"   • Azaltma gerekli: {mu_azaltma:.2f} dakika")
    print(f"   📍 Yöntem: Rota optimizasyonu, daha fazla kurye, depoya yakın yerleşim")
    
    print(f"\n🔧 Seçenek 2: DEĞİŞKENLİĞİ AZALT (σ'yı küçült)")
    
    # Hedef için gereken sigma
    # σ = (X - μ) / Z
    sigma_gerekli = (sla_limit - mu) / z_hedef
    sigma_azaltma = sigma - sigma_gerekli
    
    print(f"   • Mevcut σ: {sigma:.2f} dk")
    print(f"   • Gerekli σ: {sigma_gerekli:.2f} dk")
    print(f"   • Azaltma gerekli: {sigma_azaltma:.2f} dakika")
    print(f"   📍 Yöntem: Süreç standardizasyonu, makine öğrenmesi tahmin, trafik analizi")
    
    print(f"\n🔧 Seçenek 3: SLA LİMİTİNİ AYARLA")
    
    # %90 için gereken SLA limiti (mevcut μ ve σ ile)
    sla_gerekli = mu + z_hedef * sigma
    
    print(f"   • Mevcut SLA hedefi: {sla_limit} dk")
    print(f"   • Gerçekçi SLA (%90 için): {sla_gerekli:.1f} dk")
    print(f"   📍 Yöntem: SLA'yı {sla_gerekli:.0f} dakikaya çıkar (pazarlama ile koordine)")

else:
    print("\n✅ Mevcut performans zaten hedefin üzerinde!")
    print("   💡 Öneriler:")
    print("   • Bu başarıyı sürdürülebilir kılmak için süreçleri dokümante et")
    print("   • Daha agresif hedef belirle (örn: %95 - 35 dk altında)")
    print("   • Müşteri memnuniyeti anketleriyle doğrula")

print("\n" + "="*70)
````

## 🔢 Python ile Z-Skoru Hesaplama

<div style="background-color: #E8EAF6; padding: 15px; border-left: 5px solid #3F51B5; border-radius: 5px; margin: 15px 0;">
    <h4>🎯 Bu Bölümde Yapacaklarımız:</h4>
    <ul>
        <li>📊 KampüsSepeti verilerinden <b>farklı teslimat sürelerinin</b> Z-skorlarını hesapla</li>
        <li>🎨 Z-skorlarını <b>normal dağılım üzerinde</b> görselleştir</li>
        <li>💡 Her Z-skorunun <b>ne anlama geldiğini</b> yorumla</li>
    </ul>
</div>

````python
# ============================================================================
# 🔢 Z-SKORU HESAPLAMA ve GÖRSELLEŞTİRME
# ============================================================================

# Parametreler
mu = df_kampussepeti['teslimat_suresi_dk'].mean()
sigma = df_kampussepeti['teslimat_suresi_dk'].std()

print("🎯 KAMPÜSSEPETİ Z-SKORU ANALİZİ")
print("="*70)
print(f"μ (Ortalama): {mu:.2f} dakika")
print(f"σ (Std Sapma): {sigma:.2f} dakika\n")

# -------------------------------------------------------------------------
# FARKLI TESLİMAT SÜRELERİ İÇİN Z-SKORU HESAPLAMA
# -------------------------------------------------------------------------

# Analiz edeceğimiz teslimat süreleri
teslimat_durumlari = {
    '⚡ Çok Hızlı': 15,
    '✅ Hızlı': 22,
    '📊 Ortalama': 28,
    '⚠️ Yavaş': 35,
    '🔴 Çok Yavaş': 45,
    '🚨 Anomali': 55
}

print("📊 Z-SKORU HESAPLAMALARI")
print("="*70)

z_skorlari = {}
for durum, sure in teslimat_durumlari.items():
    # Z = (X - μ) / σ
    z = (sure - mu) / sigma
    z_skorlari[durum] = {'sure': sure, 'z': z}
    
    # Yorumlama
    if abs(z) < 1:
        yorum = "Normal aralık (68% bandında)"
        emoji = "✅"
    elif abs(z) < 2:
        yorum = "Kabul edilebilir (95% bandında)"
        emoji = "⚠️"
    elif abs(z) < 3:
        yorum = "Sınır değer (99.7% bandında)"
        emoji = "🔴"
    else:
        yorum = "ANOMALİ - İnceleme gerekli!"
        emoji = "🚨"
    
    print(f"\n{durum}: {sure} dakika")
    print(f"   Z-skoru: {z:+.2f}")
    print(f"   {emoji} Yorum: {yorum}")

# -------------------------------------------------------------------------
# GÖRSELLEŞTİRME: Z-skorlarını Normal Dağılım Üzerinde Göster
# -------------------------------------------------------------------------

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 12))

# ÜST PANEL: Orijinal Değerler (Ham Süre)
# -------------------------------------------------------------------------
x_original = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y_original = stats.norm.pdf(x_original, mu, sigma)

ax1.plot(x_original, y_original, 'black', linewidth=3, label='Normal Dağılım')
ax1.fill_between(x_original, y_original, alpha=0.2, color='lightblue')

# Ortalama çizgisi
ax1.axvline(mu, color='blue', linestyle='-', linewidth=2.5, 
            label=f'μ = {mu:.1f} dk', alpha=0.7)

# ±1σ, ±2σ, ±3σ çizgileri
for i in range(1, 4):
    ax1.axvline(mu - i*sigma, color='gray', linestyle='--', linewidth=1, alpha=0.5)
    ax1.axvline(mu + i*sigma, color='gray', linestyle='--', linewidth=1, alpha=0.5)

# Farklı teslimat durumlarını işaretle
renkler = ['#4CAF50', '#8BC34A', '#2196F3', '#FF9800', '#FF5722', '#D32F2F']
for (durum, veri), renk in zip(z_skorlari.items(), renkler):
    sure = veri['sure']
    z = veri['z']
    y_pos = stats.norm.pdf(sure, mu, sigma)
    
    # Nokta işareti
    ax1.scatter(sure, y_pos, s=300, color=renk, edgecolors='black', 
                linewidth=2, zorder=5, alpha=0.8)
    
    # Etiket
    ax1.annotate(f'{durum}\n{sure} dk\nZ={z:.2f}',
                xy=(sure, y_pos),
                xytext=(sure, y_pos + max(y_original)*0.15),
                fontsize=9, fontweight='bold',
                ha='center',
                bbox=dict(boxstyle='round,pad=0.5', facecolor=renk, 
                         edgecolor='black', linewidth=1.5, alpha=0.85),
                arrowprops=dict(arrowstyle='->', lw=2, color='black'))

ax1.set_title('Teslimat Süreleri - Orijinal Ölçekte (dakika)\n' + 
              f'μ={mu:.1f} dk, σ={sigma:.1f} dk',
              fontsize=14, fontweight='bold', pad=15)
ax1.set_xlabel('Teslimat Süresi (dakika)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Olasılık Yoğunluğu', fontsize=12)
ax1.legend(loc='upper left', fontsize=10)
ax1.grid(True, alpha=0.3, linestyle=':')

# ALT PANEL: Standardize Edilmiş (Z-skoru)
# -------------------------------------------------------------------------
# Standart normal dağılım: μ=0, σ=1
x_standard = np.linspace(-4, 4, 1000)
y_standard = stats.norm.pdf(x_standard, 0, 1)

ax2.plot(x_standard, y_standard, 'black', linewidth=3, label='Standart Normal (μ=0, σ=1)')
ax2.fill_between(x_standard, y_standard, alpha=0.2, color='lightgreen')

# Z=0 çizgisi (ortalama)
ax2.axvline(0, color='blue', linestyle='-', linewidth=2.5, 
            label='Z = 0 (Ortalama)', alpha=0.7)

# Z = ±1, ±2, ±3 çizgileri
for i in range(1, 4):
    ax2.axvline(-i, color='gray', linestyle='--', linewidth=1, alpha=0.5)
    ax2.axvline(i, color='gray', linestyle='--', linewidth=1, alpha=0.5)
    # Etiketler
    ax2.text(-i, -0.02, f'Z={-i}', ha='center', fontsize=9, color='gray')
    ax2.text(i, -0.02, f'Z={+i}', ha='center', fontsize=9, color='gray')

# Farklı durumları Z-skoru üzerinde işaretle
for (durum, veri), renk in zip(z_skorlari.items(), renkler):
    z = veri['z']
    sure = veri['sure']
    y_pos = stats.norm.pdf(z, 0, 1)
    
    # Nokta işareti
    ax2.scatter(z, y_pos, s=300, color=renk, edgecolors='black',
                linewidth=2, zorder=5, alpha=0.8)
    
    # Etiket
    ax2.annotate(f'{durum}\nZ={z:.2f}',
                xy=(z, y_pos),
                xytext=(z, y_pos + 0.08),
                fontsize=9, fontweight='bold',
                ha='center',
                bbox=dict(boxstyle='round,pad=0.5', facecolor=renk,
                         edgecolor='black', linewidth=1.5, alpha=0.85),
                arrowprops=dict(arrowstyle='->', lw=2, color='black'))

# 68-95-99.7 bölgelerini gölgele
z_1 = x_standard[(x_standard >= -1) & (x_standard <= 1)]
ax2.fill_between(z_1, 0, stats.norm.pdf(z_1, 0, 1), 
                 alpha=0.3, color='green', label='68% (±1σ)')

z_2 = x_standard[((x_standard >= -2) & (x_standard < -1)) | 
                 ((x_standard > 1) & (x_standard <= 2))]
ax2.fill_between(z_2, 0, stats.norm.pdf(z_2, 0, 1),
                 alpha=0.3, color='yellow', label='95% (±2σ)')

z_3 = x_standard[((x_standard >= -3) & (x_standard < -2)) |
                 ((x_standard > 2) & (x_standard <= 3))]
ax2.fill_between(z_3, 0, stats.norm.pdf(z_3, 0, 1),
                 alpha=0.3, color='orange', label='99.7% (±3σ)')

ax2.set_title('Standardize Edilmiş Dağılım - Z-Skoru Ölçeğinde\n' +
              'Tüm normal dağılımlar bu forma dönüştürülebilir!',
              fontsize=14, fontweight='bold', pad=15)
ax2.set_xlabel('Z-Skoru (Standart Sapma Cinsinden)', fontsize=12, fontweight='bold')
ax2.set_ylabel('Olasılık Yoğunluğu', fontsize=12)
ax2.legend(loc='upper right', fontsize=9)
ax2.grid(True, alpha=0.3, linestyle=':')
ax2.set_xlim(-4, 4)

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# YORUM ve ANALİZ
# -------------------------------------------------------------------------
print("\n" + "="*70)
print("Z-SKORU YORUM REHBERİ")
print("="*70)

print("\n🎯 GÖZLEMLENENLERİN ANLAMLARI:")

for durum, veri in z_skorlari.items():
    sure = veri['sure']
    z = veri['z']
    
    print(f"\n{durum}: {sure} dk (Z = {z:+.2f})")
    
    if z > 0:
        print(f"   📍 Konum: Ortalamanın {z:.2f}σ ÜSTÜNDE")
    elif z < 0:
        print(f"   📍 Konum: Ortalamanın {abs(z):.2f}σ ALTINDA")
    else:
        print(f"   📍 Konum: TAM ORTALAMADA")
    
    # Yüzdelik dilim
    percentile = stats.norm.cdf(z) * 100
    print(f"   📊 Yüzdelik: Teslimatların %{percentile:.1f}'inden daha yavaş")
    
    # YBS yorumu
    if abs(z) > 3:
        print(f"   🚨 YBS Aksiyonu: ACIL İNCELEME! Sistematik problem olabilir")
    elif abs(z) > 2:
        print(f"   ⚠️ YBS Aksiyonu: İzleme listesine al, müşteri ile iletişime geç")
    elif abs(z) > 1:
        print(f"   📋 YBS Aksiyonu: Normal varyasyon içinde, kayıt tut")
    else:
        print(f"   ✅ YBS Aksiyonu: Standart performans, sorun yok")

print("\n" + "="*70)
print("💡 ÖNEMLİ ÇIKARILAR:")
print("   • Z-skoru, farklı ölçeklerdeki verileri karşılaştırılabilir yapar")
print("   • |Z| > 3 → Veri anormal, incelenmeli")
print("   • Z = 0 → Tam ortalama performans")
print("   • Pozitif Z → Ortalamanın üstü (yavaş teslimat)")
print("   • Negatif Z → Ortalamanın altı (hızlı teslimat)")
print("="*70)
````

## 📊 Olasılık Hesaplamaları: Z-Tablosu Kullanımı

<div style="background: linear-gradient(to right, #FFE0B2, #FFCCBC); padding: 20px; border-radius: 10px; margin: 20px 0;">
    <h3 style="color: #E64A19;">💡 Z-Tablosu: Olasılık Hesaplama Makinesi</h3>
    <p style="font-size: 15px; line-height: 1.8;">
        Z-skoru ile <b>olasılık</b> arasında köprü kuran araç <b>Z-tablosu</b> (veya Python'da <code>scipy.stats.norm</code>)dur:
    </p>
    <ul style="font-size: 15px; line-height: 2;">
        <li>📊 <b>CDF (Cumulative Distribution Function):</b> P(X ≤ a) - "X değerinin a'dan küçük veya eşit olma olasılığı"</li>
        <li>📈 <b>SF (Survival Function):</b> P(X > a) = 1 - CDF(a) - "X değerinin a'dan büyük olma olasılığı"</li>
        <li>📏 <b>Aralık:</b> P(a < X < b) = CDF(b) - CDF(a)</li>
    </ul>
</div>

    <ol style="font-size: 15px; line-height: 2;">
        <li><b>Soru 1:</b> Bir öğrenci "45 dakikayı aşarsa iptal ederim" diyor. Bu olasılık nedir?</li>
        <li><b>Soru 2:</b> Müşteri memnuniyeti için teslimat 35 dakikanın altında olmalı. Başarı oranı?</li>
        <li><b>Soru 3:</b> İdeal teslimat penceresi 25-35 dakika. Bu aralıkta olma olasılığı?</li>
        <li><b>Soru 4:</b> En hızlı %10'luk dilim hangi süreden kısa teslimat yapıyor?</li>
    </ol>
</div>

````python
# ============================================================================
# 💼 KAMPÜSSEPETI OLASILIK HESAPLAMALARI
# ============================================================================

mu = df_kampussepeti['teslimat_suresi_dk'].mean()
sigma = df_kampussepeti['teslimat_suresi_dk'].std()

print("📊 KAMPÜSSEPETİ OLASILIK ANALİZİ")
print("="*70)
print(f"μ (Ortalama): {mu:.2f} dakika")
print(f"σ (Std Sapma): {sigma:.2f} dakika")
print("="*70)

# -------------------------------------------------------------------------
# SORU 1: P(X > 45) - 45 dakikayı aşma olasılığı
# -------------------------------------------------------------------------
print("\n🔴 SORU 1: '45 dakikayı aşarsa iptal ederim' olasılığı nedir?")
print("-"*70)

sure_1 = 45
# Z-skoru
z_1 = (sure_1 - mu) / sigma
# P(X > 45) = 1 - P(X ≤ 45) = SF(45)
prob_1 = stats.norm.sf(sure_1, mu, sigma)  # Survival Function
prob_1_pct = prob_1 * 100

print(f"   📐 Z-skoru: {z_1:.2f}")
print(f"   📊 P(Teslimat > {sure_1} dk) = {prob_1:.4f}")
print(f"   📈 Yüzde: %{prob_1_pct:.2f}")
print(f"\n   💡 Yorum: Siparişlerin yaklaşık %{prob_1_pct:.1f}'i {sure_1} dakikayı aşıyor")
print(f"   💼 YBS Kararı: {'⚠️ Yüksek risk - İyileştirme gerekli!' if prob_1_pct > 5 else '✅ Kabul edilebilir risk'}")

# Gerçek veriyle doğrula
gercek_1 = (df_kampussepeti['teslimat_suresi_dk'] > sure_1).sum()
gercek_1_pct = (gercek_1 / len(df_kampussepeti)) * 100
print(f"   ✅ Gerçek veriden: %{gercek_1_pct:.2f} ({gercek_1} sipariş)")

# -------------------------------------------------------------------------
# SORU 2: P(X ≤ 35) - 35 dakika altı başarı oranı
# -------------------------------------------------------------------------
print("\n\n🟢 SORU 2: Teslimat 35 dakikanın altında olma olasılığı?")
print("-"*70)

sure_2 = 35
z_2 = (sure_2 - mu) / sigma
# P(X ≤ 35) = CDF(35)
prob_2 = stats.norm.cdf(sure_2, mu, sigma)  # Cumulative Distribution Function
prob_2_pct = prob_2 * 100

print(f"   📐 Z-skoru: {z_2:.2f}")
print(f"   📊 P(Teslimat ≤ {sure_2} dk) = {prob_2:.4f}")
print(f"   📈 Yüzde: %{prob_2_pct:.2f}")
print(f"\n   💡 Yorum: Siparişlerin %{prob_2_pct:.1f}'i {sure_2} dakikanın altında teslim ediliyor")
print(f"   💼 YBS Kararı: {'✅ Hedef tutturuldu!' if prob_2_pct >= 80 else '⚠️ Hedefe ulaşılamadı'}")

# Gerçek veriyle doğrula
gercek_2 = (df_kampussepeti['teslimat_suresi_dk'] <= sure_2).sum()
gercek_2_pct = (gercek_2 / len(df_kampussepeti)) * 100
print(f"   ✅ Gerçek veriden: %{gercek_2_pct:.2f} ({gercek_2} sipariş)")

# -------------------------------------------------------------------------
# SORU 3: P(25 < X < 35) - İdeal aralık olasılığı
# -------------------------------------------------------------------------
print("\n\n🟡 SORU 3: İdeal teslimat penceresi (25-35 dk) olasılığı?")
print("-"*70)

sure_3a = 25
sure_3b = 35
z_3a = (sure_3a - mu) / sigma
z_3b = (sure_3b - mu) / sigma

# P(25 < X < 35) = CDF(35) - CDF(25)
prob_3 = stats.norm.cdf(sure_3b, mu, sigma) - stats.norm.cdf(sure_3a, mu, sigma)
prob_3_pct = prob_3 * 100

print(f"   📐 Z-skorları: {z_3a:.2f} ile {z_3b:.2f} arası")
print(f"   📊 P({sure_3a} < Teslimat < {sure_3b} dk) = {prob_3:.4f}")
print(f"   📈 Yüzde: %{prob_3_pct:.2f}")
print(f"\n   💡 Yorum: Siparişlerin %{prob_3_pct:.1f}'i ideal aralıkta")
print(f"   💼 YBS Kararı: {'✅ Çok iyi performans!' if prob_3_pct >= 50 else '⚠️ İyileştirme alanı var'}")

# Gerçek veriyle doğrula
gercek_3 = ((df_kampussepeti['teslimat_suresi_dk'] > sure_3a) & 
            (df_kampussepeti['teslimat_suresi_dk'] < sure_3b)).sum()
gercek_3_pct = (gercek_3 / len(df_kampussepeti)) * 100
print(f"   ✅ Gerçek veriden: %{gercek_3_pct:.2f} ({gercek_3} sipariş)")

# -------------------------------------------------------------------------
# SORU 4: En hızlı %10'luk dilim - TERS İŞLEM (Percentile Point Function)
# -------------------------------------------------------------------------
print("\n\n⚡ SORU 4: En hızlı %10'luk dilim hangi süreden kısa?")
print("-"*70)

percentile_4 = 10  # Alt %10
# PPF (Percent Point Function) = Ters CDF
sure_4 = stats.norm.ppf(percentile_4 / 100, mu, sigma)
z_4 = (sure_4 - mu) / sigma

print(f"   🎯 Aranan: Alt %{percentile_4} (en hızlı dilim)")
print(f"   📐 Z-skoru: {z_4:.2f}")
print(f"   ⏱️ Süre: {sure_4:.2f} dakika")
print(f"\n   💡 Yorum: En hızlı %{percentile_4}'luk teslimatlar {sure_4:.1f} dakikanın altında")
print(f"   💼 YBS Kararı: Bu kuryeler/rotalar örnek alınmalı, en iyi pratikler dokümante edilmeli")

# Gerçek veriyle doğrula
gercek_4 = df_kampussepeti['teslimat_suresi_dk'].quantile(percentile_4 / 100)
print(f"   ✅ Gerçek veriden {percentile_4}. persentil: {gercek_4:.2f} dakika")

# -------------------------------------------------------------------------
# GÖRSELLEŞTİRME: Tüm Olasılıkları Grafikte Göster
# -------------------------------------------------------------------------

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('KampüsSepeti Olasılık Hesaplamaları - Görsel Analiz', 
             fontsize=16, fontweight='bold', y=0.995)

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, mu, sigma)

# -------------------------------------------------------------------------
# PANEL 1: P(X > 45) - 45 dk'yı aşma
# -------------------------------------------------------------------------
ax1 = axes[0, 0]
ax1.plot(x, y, 'black', linewidth=2)
ax1.fill_between(x, y, alpha=0.2, color='lightblue')

# 45'in üstünü kırmızı ile gölgele
x_red = x[x > sure_1]
y_red = stats.norm.pdf(x_red, mu, sigma)
ax1.fill_between(x_red, y_red, alpha=0.6, color='#EF5350', label=f'P(X > {sure_1}) = {prob_1_pct:.1f}%')

ax1.axvline(sure_1, color='red', linestyle='--', linewidth=3, label=f'{sure_1} dk sınırı')
ax1.axvline(mu, color='blue', linestyle='-', linewidth=2, alpha=0.5)

ax1.set_title(f' 45+ Dakika Aşma Riski\n(İptal olasılığı)', fontsize=12, fontweight='bold')
ax1.set_xlabel('Teslimat Süresi (dk)', fontsize=10)
ax1.set_ylabel('Yoğunluk', fontsize=10)
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# PANEL 2: P(X ≤ 35) - 35 dk altı başarı
# -------------------------------------------------------------------------
ax2 = axes[0, 1]
ax2.plot(x, y, 'black', linewidth=2)
ax2.fill_between(x, y, alpha=0.2, color='lightblue')

# 35'in altını yeşil ile gölgele
x_green = x[x <= sure_2]
y_green = stats.norm.pdf(x_green, mu, sigma)
ax2.fill_between(x_green, y_green, alpha=0.6, color='#66BB6A', label=f'P(X ≤ {sure_2}) = {prob_2_pct:.1f}%')

ax2.axvline(sure_2, color='green', linestyle='--', linewidth=3, label=f'{sure_2} dk hedefi')
ax2.axvline(mu, color='blue', linestyle='-', linewidth=2, alpha=0.5)

ax2.set_title(f'35 Dakika Altı Başarı\n(Müşteri memnuniyeti)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Teslimat Süresi (dk)', fontsize=10)
ax2.set_ylabel('Yoğunluk', fontsize=10)
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# PANEL 3: P(25 < X < 35) - İdeal aralık
# -------------------------------------------------------------------------
ax3 = axes[1, 0]
ax3.plot(x, y, 'black', linewidth=2)
ax3.fill_between(x, y, alpha=0.2, color='lightblue')

# 25-35 arası sarı ile gölgele
x_yellow = x[(x > sure_3a) & (x < sure_3b)]
y_yellow = stats.norm.pdf(x_yellow, mu, sigma)
ax3.fill_between(x_yellow, y_yellow, alpha=0.6, color='#FFF176', 
                 label=f'P({sure_3a} < X < {sure_3b}) = {prob_3_pct:.1f}%')

ax3.axvline(sure_3a, color='orange', linestyle='--', linewidth=2, label=f'{sure_3a} dk')
ax3.axvline(sure_3b, color='orange', linestyle='--', linewidth=2, label=f'{sure_3b} dk')
ax3.axvline(mu, color='blue', linestyle='-', linewidth=2, alpha=0.5)

ax3.set_title(f'İdeal Teslimat Penceresi\n({sure_3a}-{sure_3b} dakika)', fontsize=12, fontweight='bold')
ax3.set_xlabel('Teslimat Süresi (dk)', fontsize=10)
ax3.set_ylabel('Yoğunluk', fontsize=10)
ax3.legend(fontsize=9)
ax3.grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# PANEL 4: En hızlı %10 - Percentile
# -------------------------------------------------------------------------
ax4 = axes[1, 1]
ax4.plot(x, y, 'black', linewidth=2)
ax4.fill_between(x, y, alpha=0.2, color='lightblue')

# En hızlı %10'u mor ile gölgele
x_purple = x[x <= sure_4]
y_purple = stats.norm.pdf(x_purple, mu, sigma)
ax4.fill_between(x_purple, y_purple, alpha=0.6, color='#AB47BC', 
                 label=f'En hızlı %{percentile_4}: ≤ {sure_4:.1f} dk')

ax4.axvline(sure_4, color='purple', linestyle='--', linewidth=3, label=f'{percentile_4}. persentil')
ax4.axvline(mu, color='blue', linestyle='-', linewidth=2, alpha=0.5)

ax4.set_title(f'En Hızlı %{percentile_4} Dilim\n(Benchmark performans)', fontsize=12, fontweight='bold')
ax4.set_xlabel('Teslimat Süresi (dk)', fontsize=10)
ax4.set_ylabel('Yoğunluk', fontsize=10)
ax4.legend(fontsize=9)
ax4.grid(True, alpha=0.3, linestyle=':')

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# ÖZET TABLO
# -------------------------------------------------------------------------
print("\n" + "="*70)
print("📊 OLASILIK HESAPLAMALARI ÖZETİ")
print("="*70)

ozet_data = {
    'Soru': [
        'P(X > 45 dk)',
        'P(X ≤ 35 dk)',
        'P(25 < X < 35)',
        'Alt %10 sınırı'
    ],
    'Teorik': [
        f'{prob_1_pct:.2f}%',
        f'{prob_2_pct:.2f}%',
        f'{prob_3_pct:.2f}%',
        f'{sure_4:.2f} dk'
    ],
    'Gerçek': [
        f'{gercek_1_pct:.2f}%',
        f'{gercek_2_pct:.2f}%',
        f'{gercek_3_pct:.2f}%',
        f'{gercek_4:.2f} dk'
    ],
    'YBS Yorumu': [
        '🔴 İptal riski',
        '🟢 Başarı oranı',
        '🟡 İdeal performans',
        '⚡ Benchmark'
    ]
}

df_ozet = pd.DataFrame(ozet_data)
print("\n", df_ozet.to_string(index=False))

print("\n" + "="*70)
print("💡 ÖNEMLİ ÇIKARILAR:")
print("   • Normal dağılım varsayımı teorik ve gerçek değerleri yaklaştırıyor ✅")
print("   • Olasılık hesapları risk yönetimi için somut metrikler sağlıyor")
print("   • PPF (ters fonksiyon) ile hedefe göre sınır belirleyebiliyoruz")
print("   • YBS kararlarında 'ne kadar olasılıkla' sorusu cevaplanıyor")
print("="*70)
````

## 🚨 Anomali Tespiti: 3σ Kuralı ile Outlier Bulma

<div style="background: linear-gradient(to right, #FFEBEE, #FFCDD2); padding: 20px; border-radius: 10px; margin: 20px 0;">
    <h3 style="color: #C62828;">⚠️ Anomali Tespiti: Veri Kalitesi Kontrolü</h3>
    <p style="font-size: 15px; line-height: 1.8;">
        <b>3σ Kuralı:</b> Normal dağılımda değerlerin %99.7'si <b>μ ± 3σ</b> aralığında olmalıdır.<br>
        Bu aralığın <b>dışındaki</b> değerler <b>anomali</b> (outlier) olarak kabul edilir!
    </p>
    <ul style="font-size: 15px; line-height: 2;">
        <li>📊 <b>|Z| > 3:</b> Çok olağandışı (olasılık < %0.3)</li>
        <li>🚨 <b>Anlamı:</b> Sistematik problem, veri hatası, veya gerçek istisnai durum</li>
        <li>💼 <b>YBS Aksiyonu:</b> İnceleme, kök neden analizi, süreç düzeltme</li>
    </ul>
</div>

````python
# ============================================================================
# 🚨 ANOMALİ TESPİTİ: 3σ KURALI UYGULAMASI
# ============================================================================

mu = df_kampussepeti['teslimat_suresi_dk'].mean()
sigma = df_kampussepeti['teslimat_suresi_dk'].std()

print("🚨 KAMPÜSSEPETİ ANOMALİ TESPİT SİSTEMİ")
print("="*70)
print(f"μ (Ortalama): {mu:.2f} dakika")
print(f"σ (Std Sapma): {sigma:.2f} dakika")

# -------------------------------------------------------------------------
# ANOMALİ EŞİKLERİNİ BELİRLE
# -------------------------------------------------------------------------

# 3σ kuralı sınırları
alt_sinir = mu - 3 * sigma
ust_sinir = mu + 3 * sigma

print(f"\n📊 3σ Kuralı Sınırları:")
print(f"   Alt Sınır: {alt_sinir:.2f} dakika (μ - 3σ)")
print(f"   Üst Sınır: {ust_sinir:.2f} dakika (μ + 3σ)")
print(f"   Normal Aralık: [{alt_sinir:.1f}, {ust_sinir:.1f}] dakika")

# -------------------------------------------------------------------------
# Z-SKORLARINI HESAPLA
# -------------------------------------------------------------------------

# Her sipariş için Z-skoru hesapla
df_kampussepeti['z_score'] = (df_kampussepeti['teslimat_suresi_dk'] - mu) / sigma

# Anomali etiketleme
df_kampussepeti['anomali'] = np.abs(df_kampussepeti['z_score']) > 3

# Anomali kategorileri
def anomali_kategori(row):
    z = row['z_score']
    if abs(z) <= 3:
        return 'Normal'
    elif z > 3:
        return 'Aşırı Yavaş (Üst)'
    else:
        return 'Aşırı Hızlı (Alt)'

df_kampussepeti['anomali_tipi'] = df_kampussepeti.apply(anomali_kategori, axis=1)

# -------------------------------------------------------------------------
# ANOMALİ İSTATİSTİKLERİ
# -------------------------------------------------------------------------

total_siparisler = len(df_kampussepeti)
anomali_sayisi = df_kampussepeti['anomali'].sum()
anomali_orani = (anomali_sayisi / total_siparisler) * 100

normal_sayisi = total_siparisler - anomali_sayisi
normal_orani = (normal_sayisi / total_siparisler) * 100

print(f"\n📈 ANOMALİ İSTATİSTİKLERİ:")
print(f"   ✅ Normal: {normal_sayisi:,} sipariş (%{normal_orani:.2f})")
print(f"   🚨 Anomali: {anomali_sayisi:,} sipariş (%{anomali_orani:.2f})")

# Teorik beklenti: Normal dağılımda %0.3 anomali beklenir
teorik_anomali = 0.3
fark = abs(anomali_orani - teorik_anomali)

print(f"\n   💡 Teorik Beklenti: %{teorik_anomali} anomali (3σ kuralı)")
print(f"   📊 Fark: {fark:.2f} puan")

if fark < 0.5:
    print(f"   ✅ Veri normal dağılıma çok uygun!")
elif fark < 1.5:
    print(f"   ⚠️ Küçük sapma var, kabul edilebilir")
else:
    print(f"   🔴 Büyük sapma! Normal dağılım varsayımı zayıf olabilir")

# Anomali tipleri dağılımı
if anomali_sayisi > 0:
    print(f"\n   🔍 ANOMALİ TİPLERİ:")
    anomali_df = df_kampussepeti[df_kampussepeti['anomali'] == True]
    
    asiri_yavas = (anomali_df['z_score'] > 3).sum()
    asiri_hizli = (anomali_df['z_score'] < -3).sum()
    
    print(f"      • Aşırı Yavaş (Z > 3): {asiri_yavas} sipariş")
    print(f"      • Aşırı Hızlı (Z < -3): {asiri_hizli} sipariş")

# -------------------------------------------------------------------------
# GÖRSELLEŞTİRME: Anomali Tespiti
# -------------------------------------------------------------------------

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Anomali Tespiti: 3σ Kuralı Uygulaması', 
             fontsize=16, fontweight='bold', y=0.995)

# -------------------------------------------------------------------------
# PANEL 1: Histogram + Anomali İşaretleme
# -------------------------------------------------------------------------
ax1 = axes[0, 0]

# Normal değerler
normal_data = df_kampussepeti[df_kampussepeti['anomali'] == False]['teslimat_suresi_dk']
ax1.hist(normal_data, bins=50, color='#81C784', alpha=0.7, 
         label=f'Normal ({len(normal_data)} sipariş)', edgecolor='darkgreen')

# Anomaliler
if anomali_sayisi > 0:
    anomali_data = df_kampussepeti[df_kampussepeti['anomali'] == True]['teslimat_suresi_dk']
    ax1.hist(anomali_data, bins=20, color='#EF5350', alpha=0.9,
             label=f'Anomali ({len(anomali_data)} sipariş)', edgecolor='darkred')

# Sınırlar
ax1.axvline(alt_sinir, color='red', linestyle='--', linewidth=3, 
            label=f'Alt Sınır: {alt_sinir:.1f} dk')
ax1.axvline(ust_sinir, color='red', linestyle='--', linewidth=3,
            label=f'Üst Sınır: {ust_sinir:.1f} dk')
ax1.axvline(mu, color='blue', linestyle='-', linewidth=2.5, alpha=0.7)

ax1.set_title('Teslimat Süreleri: Normal vs Anomali', 
              fontsize=12, fontweight='bold')
ax1.set_xlabel('Teslimat Süresi (dakika)', fontsize=10)
ax1.set_ylabel('Frekans', fontsize=10)
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# PANEL 2: Z-Skor Dağılımı
# -------------------------------------------------------------------------
ax2 = axes[0, 1]

# Z-skorları histogram
colors_z = ['#EF5350' if abs(z) > 3 else '#81C784' 
            for z in df_kampussepeti['z_score']]

ax2.hist(df_kampussepeti['z_score'], bins=60, color='#90CAF9', 
         alpha=0.7, edgecolor='darkblue', label='Z-skorları')

# Anomali eşikleri
ax2.axvline(-3, color='red', linestyle='--', linewidth=3, label='Z = ±3 (Anomali)')
ax2.axvline(3, color='red', linestyle='--', linewidth=3)
ax2.axvline(0, color='blue', linestyle='-', linewidth=2.5, alpha=0.7, label='Z = 0 (Ortalama)')

# Standart sapma işaretleri
for i in [-2, -1, 1, 2]:
    ax2.axvline(i, color='gray', linestyle=':', linewidth=1, alpha=0.5)

ax2.set_title('Z-Skor Dağılımı\n(Standardize Edilmiş)', 
              fontsize=12, fontweight='bold')
ax2.set_xlabel('Z-Skoru', fontsize=10)
ax2.set_ylabel('Frekans', fontsize=10)
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# PANEL 3: Scatter Plot - Sipariş ID vs Z-skor
# -------------------------------------------------------------------------
ax3 = axes[1, 0]

# Normal ve anomali siparişleri farklı renkle
normal_mask = df_kampussepeti['anomali'] == False
anomali_mask = df_kampussepeti['anomali'] == True

# Örnekleme (çok veri varsa)
sample_size = min(1000, len(df_kampussepeti))
sample_idx = np.random.choice(df_kampussepeti.index, sample_size, replace=False)
df_sample = df_kampussepeti.loc[sample_idx]

normal_sample = df_sample[df_sample['anomali'] == False]
anomali_sample = df_sample[df_sample['anomali'] == True]

ax3.scatter(normal_sample.index, normal_sample['z_score'], 
           c='#81C784', s=30, alpha=0.6, label='Normal', edgecolors='none')

if len(anomali_sample) > 0:
    ax3.scatter(anomali_sample.index, anomali_sample['z_score'],
               c='#EF5350', s=100, alpha=0.9, label='Anomali', 
               edgecolors='black', linewidth=1.5, marker='X')

# Anomali bandı
ax3.axhspan(-3, 3, alpha=0.1, color='green', label='Normal Band (±3σ)')
ax3.axhspan(-4, -3, alpha=0.2, color='red', label='Anomali Bölgesi')
ax3.axhspan(3, 4, alpha=0.2, color='red')

ax3.axhline(3, color='red', linestyle='--', linewidth=2)
ax3.axhline(-3, color='red', linestyle='--', linewidth=2)
ax3.axhline(0, color='blue', linestyle='-', linewidth=1, alpha=0.5)

ax3.set_title('Zaman Serisi: Anomali Tespiti\n(Örnek: 1000 sipariş)', 
              fontsize=12, fontweight='bold')
ax3.set_xlabel('Sipariş ID', fontsize=10)
ax3.set_ylabel('Z-Skoru', fontsize=10)
ax3.legend(fontsize=9, loc='upper right')
ax3.grid(True, alpha=0.3, linestyle=':')
ax3.set_ylim(-4, 4)

# -------------------------------------------------------------------------
# PANEL 4: Pasta Grafik - Normal vs Anomali
# -------------------------------------------------------------------------
ax4 = axes[1, 1]

categories = ['Normal\n(±3σ içinde)', 'Anomali\n(±3σ dışında)']
sizes = [normal_sayisi, anomali_sayisi]
colors_pie = ['#81C784', '#EF5350']
explode = (0, 0.1)  # Anomali dilimini vurgula

wedges, texts, autotexts = ax4.pie(sizes, explode=explode, labels=categories,
                                     colors=colors_pie, autopct='%1.2f%%',
                                     shadow=True, startangle=90,
                                     textprops={'fontsize': 11, 'weight': 'bold'})

# Yüzde değerlerini beyaz ve bold yap
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(14)
    autotext.set_weight('bold')

ax4.set_title(f'Anomali Oranı\n(Toplam: {total_siparisler:,} sipariş)', 
              fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# ANOMALİ ÖRNEKLERİ - DETAYLI ANALİZ
# -------------------------------------------------------------------------

if anomali_sayisi > 0:
    print("\n" + "="*70)
    print("🔍 ANOMALİ DETAYLI ANALİZ")
    print("="*70)
    
    # En aşırı 5 anomali
    anomaliler_sirali = df_kampussepeti[df_kampussepeti['anomali'] == True].sort_values(
        'z_score', key=abs, ascending=False
    ).head(10)
    
    print(f"\n🚨 En Aşırı 10 Anomali:")
    print("-"*70)
    
    for idx, (i, row) in enumerate(anomaliler_sirali.iterrows(), 1):
        sure = row['teslimat_suresi_dk']
        z = row['z_score']
        
        if z > 0:
            yorum = f"Çok YAVAŞ - Ortalamanın {z:.2f}σ üstünde"
            emoji = "🐢"
        else:
            yorum = f"Çok HIZLI - Ortalamanın {abs(z):.2f}σ altında"
            emoji = "⚡"
        
        print(f"{idx:2d}. Sipariş #{i+1:5d}: {sure:5.1f} dk | Z = {z:+6.2f} | {emoji} {yorum}")
    
    print("\n💼 YBS ÖNERİLERİ:")
    print("   1. Bu siparişlerin detaylarını manuel incele")
    print("   2. GPS kayıtlarını, rota bilgilerini kontrol et")
    print("   3. Kurye performans raporlarıyla çapraz kontrol yap")
    print("   4. Hava durumu, trafik gibi dış faktörleri değerlendir")
    print("   5. Tekrarlanan anomaliler varsa süreç iyileştirme başlat")

else:
    print("\n✅ Hiç anomali tespit edilmedi! Veri çok temiz.")

print("\n" + "="*70)
````

## 📊 Python ile QQ-Plot Oluşturma

<div style="background-color: #E1F5FE; padding: 15px; border-left: 5px solid #0277BD; border-radius: 5px; margin: 15px 0;">
    <h4>🎯 KampüsSepeti Metrikleri QQ-Plot Analizi</h4>
    <p>3 farklı metriği QQ-Plot ile test edeceğiz:</p>
    <ul>
        <li>🚚 <b>Teslimat Süreleri</b> - Normal olması bekleniyor</li>
        <li>💰 <b>Sepet Tutarları</b> - Sağa çarpık olduğunu biliyoruz</li>
        <li>🍳 <b>Hazırlık Süreleri</b> - Normal olması bekleniyor</li>
    </ul>

## 🧪 **SHAPIRO-WILK TESTİ: İSTATİSTİKSEL NORMALLİK KONTROLÜ**

QQ-Plot **görsel** bir kontroldü, şimdi **matematiksel/istatistiksel** bir test yapalım!

### 📐 Shapiro-Wilk Testi Nedir?

- **Hipotez Testi**: Veri normal dağılıma uyuyor mu?
- **H₀ (Null Hypothesis)**: Veri normal dağılıma uyuyor
- **H₁ (Alternative Hypothesis)**: Veri normal dağılıma uymuyor

### 🎯 P-Value Yorumlama:

```
p-value > 0.05  →  H₀ kabul  →  ✅ Normal dağılıma UYGUN
p-value ≤ 0.05  →  H₀ red    →  ⚠️ Normal dağılıma UYGUN DEĞİL
```

### 🔍 Karar Akış Şeması:

```mermaid
flowchart TD
    A["🧪 Shapiro-Wilk Testi Uygula<br/>stats.shapiro(data)"] --> B["p-value hesapla"]
    
    B --> C{"p-value > 0.05?"}
    
    C -->|Evet| D["✅ H₀ KABUL<br/>Normal Dağılıma Uygun"]
    C -->|Hayır| E["⚠️ H₀ RED<br/>Normal Dağılıma Uygun Değil"]
    
    D --> F["💼 Parametrik Testler<br/>Kullanılabilir<br/>(t-test, ANOVA)"]
    
    E --> G["💼 Alternatif Yöntemler"]
    G --> G1["1️⃣ Log Dönüşümü Dene"]
    G --> G2["2️⃣ Non-Parametrik Testler<br/>(Mann-Whitney, Kruskal-Wallis)"]
    G --> G3["3️⃣ Robust Yöntemler<br/>(Medyan, IQR)"]
    
    style D fill:#C8E6C9,stroke:#388E3C,stroke-width:3px
    style E fill:#FFCCBC,stroke:#D84315,stroke-width:3px
    style F fill:#B3E5FC,stroke:#0277BD,stroke-width:2px
    style G fill:#FFF9C4,stroke:#F57F17,stroke-width:2px
```

### 📋 **FLASHCARD #8: SHAPIRO-WILK TEST YORUMLAMA**

| **p-value Aralığı** | **Karar** | **Yorum** | **YBS Aksiyonu** |
|:---:|:---:|:---|:---|
| **p > 0.05** | ✅ **Normal** | Veri normal dağılıma uyuyor | Parametrik testlere devam et |
| **0.01 < p ≤ 0.05** | ⚠️ **Sınırda** | Hafif sapma var | Log dönüşümü dene |
| **0.001 < p ≤ 0.01** | ⚠️ **Normal Değil** | Anlamlı sapma | Non-parametrik testlere geç |
| **p ≤ 0.001** | ❌ **Kesinlikle Normal Değil** | Çok büyük sapma | Medyan/Robust yöntemler kullan |

---

````python
# ============================================================================
# 🧪 SHAPIRO-WILK NORMALLİK TESTİ
# ============================================================================

print("🧪 SHAPIRO-WILK NORMALLİK TESTİ")
print("="*70)
print("H₀: Veri normal dağılıma uyuyor")
print("H₁: Veri normal dağılıma uymuyor")
print("Karar Kuralı: p-value > 0.05 ise H₀ kabul (Normal)")
print("="*70)

# Test fonksiyonu
def normallik_testi(data, metrik_adi, renk):
    """
    Shapiro-Wilk testi uygula ve sonuçları yorumla
    """
    statistic, p_value = stats.shapiro(data)
    
    # Karar
    if p_value > 0.05:
        karar = "✅ NORMAL DAĞILIMA UYGUN"
        yorum = "Parametrik testler kullanılabilir"
        sembol = "✅"
    else:
        karar = "⚠️ NORMAL DAĞILIMA UYGUN DEĞİL"
        yorum = "Log dönüşümü veya non-parametrik testler gerekli"
        sembol = "⚠️"
    
    print(f"\n{sembol} {metrik_adi}")
    print("-" * 70)
    print(f"   Test İstatistiği (W): {statistic:.6f}")
    print(f"   P-Value             : {p_value:.6f}")
    print(f"   Karar               : {karar}")
    print(f"   💼 YBS Yorumu       : {yorum}")
    
    return statistic, p_value

# -------------------------------------------------------------------------
# 3 METRİK İÇİN TEST
# -------------------------------------------------------------------------

# 1. Teslimat Süreleri
stat1, pval1 = normallik_testi(
    df_kampussepeti['teslimat_suresi_dk'], 
    "🚚 TESLİMAT SÜRELERİ", 
    '#81C784'
)

# 2. Sepet Tutarları
stat2, pval2 = normallik_testi(
    df_kampussepeti['sepet_tutari_tl'], 
    "💰 SEPET TUTARLARI", 
    '#FFB74D'
)

# 3. Hazırlık Süreleri
stat3, pval3 = normallik_testi(
    df_kampussepeti['hazirlik_suresi_dk'], 
    "🍳 HAZIRLIK SÜRELERİ", 
    '#BA68C8'
)

print("\n" + "="*70)

# -------------------------------------------------------------------------
# GÖRSELLEŞTİRME: P-VALUE KARŞILAŞTIRMA
# -------------------------------------------------------------------------

fig, ax = plt.subplots(figsize=(12, 7))

metriks = ['Teslimat\nSüreleri', 'Sepet\nTutarları', 'Hazırlık\nSüreleri']
p_values = [pval1, pval2, pval3]
renkler = ['#81C784', '#FFB74D', '#BA68C8']

# Bar chart
bars = ax.bar(metriks, p_values, color=renkler, alpha=0.8, edgecolor='black', linewidth=2)

# Kritik çizgi (α = 0.05)
ax.axhline(y=0.05, color='red', linestyle='--', linewidth=3, label='α = 0.05 (Kritik Değer)')

# p-value değerlerini bar üzerine yaz
for i, (bar, pval) in enumerate(zip(bars, p_values)):
    height = bar.get_height()
    
    # Karar simgesi
    if pval > 0.05:
        simge = "✅ Normal"
        ypos = height + 0.02
    else:
        simge = "⚠️ Normal Değil"
        ypos = height + 0.02
    
    ax.text(bar.get_x() + bar.get_width()/2., ypos,
            f'p={pval:.4f}\n{simge}',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

ax.set_xlabel('KampüsSepeti Metrikleri', fontsize=12, fontweight='bold')
ax.set_ylabel('P-Value', fontsize=12, fontweight='bold')
ax.set_title('Shapiro-Wilk Normallik Testi Sonuçları\nP-Value Karşılaştırması', 
             fontsize=14, fontweight='bold')
ax.set_ylim(0, max(p_values) + 0.15)
ax.legend(fontsize=11, loc='upper right')
ax.grid(True, alpha=0.3, axis='y', linestyle=':')

# Arka plan renklendirme
ax.axhspan(0, 0.05, alpha=0.1, color='red', label='Red Bölgesi')
ax.axhspan(0.05, ax.get_ylim()[1], alpha=0.1, color='green')

plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("📊 TEST SONUÇLARI ÖZETİ")
print("="*70)
print("\n✅ NORMAL DAĞILAN METRİKLER:")
if pval1 > 0.05:
    print(f"   • Teslimat Süreleri (p={pval1:.4f})")
if pval3 > 0.05:
    print(f"   • Hazırlık Süreleri (p={pval3:.4f})")

print("\n⚠️ NORMAL DAĞILMAYAN METRİKLER:")
if pval2 <= 0.05:
    print(f"   • Sepet Tutarları (p={pval2:.4f})")
    print("     → Log dönüşümü gerekiyor!")

print("\n" + "="*70)
print("💡 YBS AKSIYON PLANI:")
print("   1. Normal metriklere → Parametrik testler (t-test, ANOVA) uygula")
print("   2. Normal olmayan → Log dönüşümü yap veya non-parametrik testler kullan")
print("   3. SLA hedefleri → Normal metriklere göre belirle")
print("="*70)
````

## 🔄 **LOG DÖNÜŞÜMÜ: ÇARPIK VERİYİ NORMAL HALE GETİRME**

**Sepet Tutarları** çarpık çıktı (p-value < 0.05). Log dönüşümü ile bunu düzeltelim!

### 🎯 Log Dönüşümü Ne Zaman Kullanılır?

1. ⚠️ **Sağa çarpık dağılımlar** (right-skewed)
2. 💰 **Para, gelir, maliyet verileri** (doğal olarak çarpık)
3. 📈 **Üssel büyüme gösteren veriler**
4. 🔬 **Varyansı stabilize etmek için**

### 📐 Formül:

```
Dönüşüm:  y = log(x)  veya  y = log(x + 1)  (sıfır varsa)
```

### 💼 **YBS Senaryosu: Fiyatlandırma Stratejisi**

KampüsSepeti **premium müşteri segmentasyonu** yapıyor:
- Hangi müşteriler yüksek sepet tutarına sahip?
- Log dönüşümü sonrası **normal dağılım** varsayımı ile **%95 güven aralığı** hesaplayabilir miyiz?

Şimdi bunu Python ile yapalım! 👇

````python
# ============================================================================
# 🔄 LOG DÖNÜŞÜMÜ UYGULAMASI
# ============================================================================

print("🔄 LOG DÖNÜŞÜMÜ: SEPET TUTARLARINI NORMALIZE ETME")
print("="*70)

# Orijinal veri
sepet_orijinal = df_kampussepeti['sepet_tutari_tl'].copy()

# Log dönüşümü uygula
sepet_log = np.log(sepet_orijinal)

# İstatistikler
print("\n📊 ÖNCESİ (Orijinal Sepet Tutarları):")
print(f"   Ortalama: {sepet_orijinal.mean():.2f} TL")
print(f"   Std Sapma: {sepet_orijinal.std():.2f} TL")
print(f"   Medyan: {sepet_orijinal.median():.2f} TL")
print(f"   Çarpıklık: {stats.skew(sepet_orijinal):.4f} (Sağa çarpık!)")

print("\n📊 SONRASI (Log Dönüşümlü):")
print(f"   Ortalama: {sepet_log.mean():.4f}")
print(f"   Std Sapma: {sepet_log.std():.4f}")
print(f"   Medyan: {sepet_log.median():.4f}")
print(f"   Çarpıklık: {stats.skew(sepet_log):.4f} (0'a yaklaştı!)")

print("\n" + "="*70)

# -------------------------------------------------------------------------
# SHAPIRO-WILK TESTİ: ÖNCESİ vs SONRASI
# -------------------------------------------------------------------------

stat_oncesi, pval_oncesi = stats.shapiro(sepet_orijinal)
stat_sonrasi, pval_sonrasi = stats.shapiro(sepet_log)

print("\n🧪 SHAPIRO-WILK TESTİ KARŞILAŞTIRMASI")
print("="*70)
print(f"\n⚠️ ÖNCESİ (Orijinal):")
print(f"   P-Value: {pval_oncesi:.6f}")
if pval_oncesi > 0.05:
    print(f"   Karar: ✅ Normal (ama çok düşük p-value!)")
else:
    print(f"   Karar: ⚠️ Normal DEĞİL")

print(f"\n✅ SONRASI (Log Dönüşümlü):")
print(f"   P-Value: {pval_sonrasi:.6f}")
if pval_sonrasi > 0.05:
    print(f"   Karar: ✅ Normal DAĞILIMA UYGUN")
else:
    print(f"   Karar: ⚠️ Hala normal değil")

print("\n📈 İyileştirme: p-value {:.4f} → {:.4f}".format(pval_oncesi, pval_sonrasi))

# -------------------------------------------------------------------------
# GÖRSELLEŞTİRME: ÖNCESİ vs SONRASI
# -------------------------------------------------------------------------

fig = plt.figure(figsize=(18, 10))

# SATIR 1: HİSTOGRAM KARŞILAŞTIRMASI
# -------------------------------------

# Orijinal Histogram
ax1 = plt.subplot(2, 3, 1)
ax1.hist(sepet_orijinal, bins=50, density=True, alpha=0.7, color='#FFB74D',
         edgecolor='darkorange', label='Gerçek Veri')
mu_o = sepet_orijinal.mean()
sigma_o = sepet_orijinal.std()
x_o = np.linspace(sepet_orijinal.min(), sepet_orijinal.max(), 100)
ax1.plot(x_o, stats.norm.pdf(x_o, mu_o, sigma_o), 'r--', linewidth=3,
         label='Teorik Normal (Uyumsuz)', alpha=0.7)
ax1.set_title('ÖNCESİ: Sepet Tutarları\n(Sağa Çarpık)', 
              fontsize=12, fontweight='bold', color='darkorange')
ax1.set_xlabel('Sepet Tutarı (TL)', fontsize=10)
ax1.set_ylabel('Yoğunluk', fontsize=10)
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3, linestyle=':')

# Log Dönüşümlü Histogram
ax2 = plt.subplot(2, 3, 2)
ax2.hist(sepet_log, bins=40, density=True, alpha=0.7, color='#81C784',
         edgecolor='darkgreen', label='Log Dönüşümlü')
mu_l = sepet_log.mean()
sigma_l = sepet_log.std()
x_l = np.linspace(sepet_log.min(), sepet_log.max(), 100)
ax2.plot(x_l, stats.norm.pdf(x_l, mu_l, sigma_l), 'r-', linewidth=3,
         label='Teorik Normal (Uyumlu)')
ax2.set_title('SONRASI: Log(Sepet Tutarları)\n(Normal)', 
              fontsize=12, fontweight='bold', color='darkgreen')
ax2.set_xlabel('Log(Sepet Tutarı)', fontsize=10)
ax2.set_ylabel('Yoğunluk', fontsize=10)
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3, linestyle=':')

# P-Value Karşılaştırma Bar Chart
ax3 = plt.subplot(2, 3, 3)
labels = ['Orijinal', 'Log Dönüşümlü']
p_vals = [pval_oncesi, pval_sonrasi]
colors = ['#FFB74D', '#81C784']
bars = ax3.bar(labels, p_vals, color=colors, alpha=0.8, edgecolor='black', linewidth=2)
ax3.axhline(y=0.05, color='red', linestyle='--', linewidth=2, label='α = 0.05')

for bar, pval in zip(bars, p_vals):
    height = bar.get_height()
    simge = "✅" if pval > 0.05 else "⚠️"
    ax3.text(bar.get_x() + bar.get_width()/2., height + 0.02,
             f'{simge}\np={pval:.4f}',
             ha='center', va='bottom', fontsize=10, fontweight='bold')

ax3.set_title('🧪 Shapiro-Wilk P-Value\nKarşılaştırması', 
              fontsize=12, fontweight='bold')
ax3.set_ylabel('P-Value', fontsize=10)
ax3.set_ylim(0, max(p_vals) + 0.15)
ax3.legend(fontsize=9)
ax3.grid(True, alpha=0.3, axis='y', linestyle=':')

# SATIR 2: QQ-PLOT KARŞILAŞTIRMASI
# -------------------------------------

# Orijinal QQ-Plot
ax4 = plt.subplot(2, 3, 4)
stats.probplot(sepet_orijinal, dist="norm", plot=ax4)
ax4.get_lines()[0].set_markerfacecolor('#FFB74D')
ax4.get_lines()[0].set_markeredgecolor('darkorange')
ax4.get_lines()[0].set_markersize(5)
ax4.get_lines()[1].set_color('red')
ax4.get_lines()[1].set_linewidth(2)
ax4.set_title('ÖNCESİ: QQ-Plot\n(S şekli - Sağa Çarpık)', 
              fontsize=12, fontweight='bold', color='darkorange')
ax4.set_xlabel('Teorik Kuantiller', fontsize=9)
ax4.set_ylabel('Gözlenen Kuantiller', fontsize=9)
ax4.grid(True, alpha=0.3, linestyle=':')

# Log Dönüşümlü QQ-Plot
ax5 = plt.subplot(2, 3, 5)
stats.probplot(sepet_log, dist="norm", plot=ax5)
ax5.get_lines()[0].set_markerfacecolor('#81C784')
ax5.get_lines()[0].set_markeredgecolor('darkgreen')
ax5.get_lines()[0].set_markersize(5)
ax5.get_lines()[1].set_color('red')
ax5.get_lines()[1].set_linewidth(2)
ax5.set_title('SONRASI: QQ-Plot\n(Düz çizgi - Normal)', 
              fontsize=12, fontweight='bold', color='darkgreen')
ax5.set_xlabel('Teorik Kuantiller', fontsize=9)
ax5.set_ylabel('Gözlenen Kuantiller', fontsize=9)
ax5.grid(True, alpha=0.3, linestyle=':')

# Box Plot Karşılaştırma
ax6 = plt.subplot(2, 3, 6)
bp = ax6.boxplot([sepet_orijinal, sepet_log], 
                  labels=['Orijinal\n(Sağa Çarpık)', 'Log Dönüşümlü\n(Simetrik)'],
                  patch_artist=True,
                  widths=0.6)

bp['boxes'][0].set_facecolor('#FFB74D')
bp['boxes'][1].set_facecolor('#81C784')

for patch in bp['boxes']:
    patch.set_alpha(0.7)
    patch.set_edgecolor('black')
    patch.set_linewidth(2)

ax6.set_title('Box Plot Karşılaştırma\n(Çarpıklık Görünümü)', 
              fontsize=12, fontweight='bold')
ax6.set_ylabel('Değer', fontsize=10)
ax6.grid(True, alpha=0.3, axis='y', linestyle=':')

plt.suptitle('LOG DÖNÜŞÜMÜ ETKİSİ: SEPET TUTARLARI\nÖncesi vs Sonrası Normallik Karşılaştırması',
             fontsize=14, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("💡 YBS SONUÇ ve AKSIYON")
print("="*70)
print("\nLOG DÖNÜŞÜMÜ BAŞARILI:")
print("   • Çarpıklık 0'a yaklaştı")
print("   • P-value arttı (normal dağılıma yaklaştı)")
print("   • QQ-Plot düzeldi (S şekli → düz çizgi)")

print("\n💼 İŞ KARARI:")
print("   1. Fiyatlandırma analizlerinde log(sepet_tutari) kullan")
print("   2. Premium segment için %95 güven aralığı hesapla")
print("   3. A/B testlerinde parametrik testler artık güvenli")
print("   4. Tahmine dayalı modellerde log-transformed feature ekle")

print("\n🎯 KAMPÜSSEPETI STRATEJİSİ:")
print("   • Normal dağılım varsayımı → Daha güvenilir istatistikler")
print("   • Outlier etkisi azaldı → Robust kararlar")
print("   • Segmentasyon için %95 CI hesaplanabilir")

print("="*70)
````

## 🔮 **MERKEZİ LİMİT TEOREMİ NEDİR?**

### 📖 Tanım:

> **"Popülasyonun dağılımı ne olursa olsun (normal, çarpık, uniform, her ne olursa), yeterince büyük örneklemlerden hesaplanan ortalamaların dağılımı NORMAL DAĞILIM gösterir."**

### 🎯 Basit Dille:

Diyelim ki:
- 💰 **Sepet tutarları SAĞA ÇARPIK** (normal değil)
- Her gün **30 sipariş** alıyorsun
- Her günün **ortalama sepet tutarını** hesaplıyorsun
- 100 gün boyunca bu günlük ortalamaları topluyorsun

**CLT der ki**: Bu 100 günlük ortalamaların dağılımı **NORMAL DAĞILIM** olacaktır! 🤯

---

### 🧮 Matematiksel Formül:

Popülasyon: `X ~ herhangi bir dağılım (μ, σ²)`  
Örneklem büyüklüğü: `n` (örneğin 30)

Örneklem ortalaması: 

$$
\bar{X} = \frac{1}{n} \sum_{i=1}^{n} X_i
$$

**CLT'ye göre:**

$$
\bar{X} \sim N\left(\mu, \frac{\sigma^2}{n}\right)
$$

veya standart hata ile:

$$
\bar{X} \sim N\left(\mu, \frac{\sigma}{\sqrt{n}}\right)
$$

### 🔑 Anahtar Kavramlar:

| **Terim** | **Türkçe** | **Anlamı** |
|:---|:---|:---|
| **Population Mean** | Popülasyon Ortalaması (μ) | Tüm verinin gerçek ortalaması |
| **Sample Mean** | Örneklem Ortalaması ($\bar{X}$) | n elemanlı örneğin ortalaması |
| **Standard Error (SE)** | Standart Hata | $\frac{\sigma}{\sqrt{n}}$ - Örneklem ortalamalarının standart sapması |
| **Sampling Distribution** | Örnekleme Dağılımı | Örneklem ortalamalarının oluşturduğu dağılım |

| **n arttıkça SE?** | Azalır → Daha hassas tahminler |
| **Minimum n?** | n ≥ 30 (popülasyon normal değilse) |
| **YBS Kullanımı?** | Günlük/haftalık ortalama analizleri |

---

## 🎲 **PYTHON İLE CLT SİMÜLASYONU**

Şimdi CLT'yi **gerçek KampüsSepeti verisi** ile kanıtlayalım!

### 📋 Deney Planı:

1. **Popülasyon**: 5000 sipariş (sepet_tutari_tl) - SAĞA ÇARPIK
2. **Örneklem Büyüklüğü**: n = 30 sipariş
3. **Simülasyon**: 10,000 kez rastgele 30 sipariş seç, ortalamasını al
4. **Sonuç**: Bu 10,000 ortalamanın dağılımı NORMAL olmalı!

Hazır mısın? Gidelim! 🚀

---

````python
# ============================================================================
# 🎲 MERKEZİ LİMİT TEOREMİ SİMÜLASYONU
# ============================================================================

print("🎲 CENTRAL LIMIT THEOREM SİMÜLASYONU")
print("="*70)

# Parametreler
populasyon = df_kampussepeti['sepet_tutari_tl'].values  # 5000 sipariş
n_orneklem = 30          # Her örneklemde 30 sipariş
n_simulasyon = 10000     # 10,000 kez tekrarla

print(f"📊 Popülasyon Büyüklüğü: {len(populasyon):,}")
print(f"📦 Örneklem Büyüklüğü (n): {n_orneklem}")
print(f"🔄 Simülasyon Sayısı: {n_simulasyon:,}")
print("\n" + "="*70)

# Popülasyon istatistikleri
mu_populasyon = populasyon.mean()
sigma_populasyon = populasyon.std()

print("\n🎯 POPÜLASYON İSTATİSTİKLERİ:")
print(f"   μ (Ortalama)    : {mu_populasyon:.2f} TL")
print(f"   σ (Std Sapma)   : {sigma_populasyon:.2f} TL")
print(f"   Çarpıklık       : {stats.skew(populasyon):.4f} (SAĞA ÇARPIK!)")

# -------------------------------------------------------------------------
# SİMÜLASYON: 10,000 ÖRNEKLEM ORTALAMASI HESAPLA
# -------------------------------------------------------------------------

print("\n🔄 Simülasyon başlıyor...")
print("   Her iterasyonda 30 rastgele sipariş seçilip ortalaması hesaplanıyor...")

orneklem_ortalamalari = []

np.random.seed(42)  # Tekrarlanabilirlik için

for i in range(n_simulasyon):
    # Rastgele 30 sipariş seç
    orneklem = np.random.choice(populasyon, size=n_orneklem, replace=True)
    
    # Ortalamasını hesapla
    orneklem_ort = orneklem.mean()
    orneklem_ortalamalari.append(orneklem_ort)
    
    # İlerleme göstergesi
    if (i + 1) % 2000 == 0:
        print(f"   ✓ {i+1:,} / {n_simulasyon:,} tamamlandı...")

orneklem_ortalamalari = np.array(orneklem_ortalamalari)

print("\n✅ Simülasyon tamamlandı!")
print("="*70)

# -------------------------------------------------------------------------
# CLT TAHMİNLERİ vs GERÇEK SONUÇLAR
# -------------------------------------------------------------------------

# CLT'ye göre teorik değerler
teorik_ortalama = mu_populasyon
teorik_std = sigma_populasyon / np.sqrt(n_orneklem)  # Standart Hata (SE)

# Gerçek (simülasyondan elde edilen) değerler
gercek_ortalama = orneklem_ortalamalari.mean()
gercek_std = orneklem_ortalamalari.std()

print("\n📊 CLT TAHMİNİ vs GERÇEK SONUÇLAR:")
print("-" * 70)
print(f"{'Metrik':<30} {'CLT Tahmini':<20} {'Simülasyon Sonucu':<20}")
print("-" * 70)
print(f"{'Ortalama (μ)':<30} {teorik_ortalama:<20.2f} {gercek_ortalama:<20.2f}")
print(f"{'Standart Sapma (SE)':<30} {teorik_std:<20.2f} {gercek_std:<20.2f}")
print("-" * 70)

fark_ort = abs(teorik_ortalama - gercek_ortalama)
fark_std = abs(teorik_std - gercek_std)

print(f"\n✅ Ortalama Farkı: {fark_ort:.4f} TL (Neredeyse sıfır!)")
print(f"✅ Std Sapma Farkı: {fark_std:.4f} TL (CLT mükemmel çalıştı!)")

# -------------------------------------------------------------------------
# NORMALLİK KONTROLÜ: SHAPIRO-WILK
# -------------------------------------------------------------------------

# Popülasyon normallik testi
stat_pop, pval_pop = stats.shapiro(populasyon[:1000])  # İlk 1000 (test limiti)

# Örneklem ortalamaları normallik testi
stat_sample, pval_sample = stats.shapiro(orneklem_ortalamalari[:1000])  # İlk 1000

print("\n🧪 SHAPIRO-WILK NORMALLİK TESTİ:")
print("="*70)
print(f"📦 POPÜLASYON (Sepet Tutarları - Orijinal):")
print(f"   P-Value: {pval_pop:.6f}")
if pval_pop > 0.05:
    print(f"   Karar: ✅ Normal")
else:
    print(f"   Karar: ⚠️ Normal DEĞİL (Çarpık!)")

print(f"\n📊 ÖRNEKLEM ORTALAMALARI (CLT sonrası):")
print(f"   P-Value: {pval_sample:.6f}")
if pval_sample > 0.05:
    print(f"   Karar: ✅ NORMAL DAĞILIM (CLT çalıştı!)")
else:
    print(f"   Karar: ⚠️ Normal değil")

print("\n" + "="*70)
print("🎉 CLT KANITLANDI!")
print("   • Popülasyon: ÇARPIK → Normal DEĞİL")
print("   • Örneklem Ortalamaları: NORMAL → CLT doğrulandı!")
print("="*70)
````

## ⚡ **STANDART HATA (SE) ve ÖRNEKLEM BÜYÜKLÜĞÜ ETKİSİ**

### 📐 Standart Hata Formülü:

$$
SE = \frac{\sigma}{\sqrt{n}}
$$

### 🔍 Ne Anlama Gelir?

- **σ**: Popülasyon standart sapması (sabit)
- **n**: Örneklem büyüklüğü (değişken)
- **SE**: Örneklem ortalamalarının standart sapması

### 🎯 Kritik İçgörü:

```
n ARTTIKÇA → SE AZALIR → Tahminler daha hassas!
```

### 💡 Örnek:

KampüsSepeti için σ = 45 TL (sepet tutarları):

| **n** | **SE = 45/√n** | **Yorum** |
|:---:|:---:|:---|
| **n=10** | 14.2 TL | Çok geniş aralık, güvensiz |
| **n=30** | **8.2 TL** | ✅ CLT altın standart |
| **n=100** | 4.5 TL | Daha hassas tahmin |
| **n=400** | 2.25 TL | Çok hassas, ama maliyetli |

### 💼 **YBS Senaryosu: Günlük Raporlama**

KampüsSepeti her gün **n sipariş** alıyor. Genel Müdür soruyor:

> "Günlük ortalama sepet tutarını ne kadar kesinlikle tahmin edebiliriz?"

**Cevap**: n'e bağlı!

- **Az sipariş (n=10)**: SE = 14.2 TL → ±28.4 TL aralık (güvensiz!)
- **Orta sipariş (n=30)**: SE = 8.2 TL → ±16.4 TL aralık (makul)
- **Çok sipariş (n=100)**: SE = 4.5 TL → ±9 TL aralık (hassas!)

Şimdi bunu Python ile görelim! 👇

````python
# ============================================================================
# ⚡ ÖRNEKLEM BÜYÜKLÜĞÜ ETKİSİ SİMÜLASYONU
# ============================================================================

print("⚡ ÖRNEKLEM BÜYÜKLÜĞÜ ETKİSİ ANALİZİ")
print("="*70)

# Test edilecek örneklem büyüklükleri
orneklem_buyuklukleri = [5, 10, 20, 30, 50, 100, 200]
n_simulasyon = 5000  # Her n için 5000 simülasyon

# Popülasyon parametreleri
mu = mu_populasyon
sigma = sigma_populasyon

print(f"📊 Popülasyon: μ={mu:.2f} TL, σ={sigma:.2f} TL")
print(f"🔄 Her n için {n_simulasyon:,} simülasyon yapılacak\n")
print("="*70)

# Sonuçları sakla
sonuclar = {
    'n': [],
    'teorik_SE': [],
    'gercek_SE': [],
    'ortalamalarin_ortalamasi': [],
    'dagilim_genisligi': []
}

np.random.seed(42)

for n in orneklem_buyuklukleri:
    print(f"\n📦 n = {n} için simülasyon yapılıyor...")
    
    # Teorik SE
    teorik_SE = sigma / np.sqrt(n)
    
    # Simülasyon
    orneklem_orts = []
    for _ in range(n_simulasyon):
        orneklem = np.random.choice(populasyon, size=n, replace=True)
        orneklem_orts.append(orneklem.mean())
    
    orneklem_orts = np.array(orneklem_orts)
    
    # Gerçek SE
    gercek_SE = orneklem_orts.std()
    
    # Ortalama
    ort_ort = orneklem_orts.mean()
    
    # %95 aralık genişliği (2*SE)
    genislik = 2 * gercek_SE
    
    # Kaydet
    sonuclar['n'].append(n)
    sonuclar['teorik_SE'].append(teorik_SE)
    sonuclar['gercek_SE'].append(gercek_SE)
    sonuclar['ortalamalarin_ortalamasi'].append(ort_ort)
    sonuclar['dagilim_genisligi'].append(genislik)
    
    print(f"   Teorik SE: {teorik_SE:.2f} TL")
    print(f"   Gerçek SE: {gercek_SE:.2f} TL")
    print(f"   ±2SE Aralık: {genislik:.2f} TL (%95 güven için)")

print("\n" + "="*70)
print("✅ Tüm simülasyonlar tamamlandı!")
print("="*70)

# DataFrame'e çevir
df_sonuc = pd.DataFrame(sonuclar)

print("\n📊 SONUÇ TABLOSU:")
print(df_sonuc.to_string(index=False))

# -------------------------------------------------------------------------
# GÖRSELLEŞTİRME: SE'nin n ile Değişimi
# -------------------------------------------------------------------------

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# PANEL 1: SE vs n (Log-Log Scale)
ax1 = axes[0, 0]
ax1.plot(df_sonuc['n'], df_sonuc['teorik_SE'], 'ro-', linewidth=3, 
         markersize=10, label='Teorik SE (σ/√n)', alpha=0.7)
ax1.plot(df_sonuc['n'], df_sonuc['gercek_SE'], 'b^--', linewidth=2, 
         markersize=8, label='Simülasyon SE', alpha=0.7)

ax1.set_xlabel('Örneklem Büyüklüğü (n)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Standart Hata (SE) - TL', fontsize=12, fontweight='bold')
ax1.set_title('Standart Hata vs Örneklem Büyüklüğü\n(SE = σ/√n İlişkisi)', 
              fontsize=13, fontweight='bold')
ax1.legend(fontsize=11, loc='upper right')
ax1.grid(True, alpha=0.4, linestyle=':')
ax1.set_xscale('log')
ax1.set_yscale('log')

# n=30 vurgusu
ax1.axvline(30, color='green', linestyle=':', linewidth=2, alpha=0.7)
ax1.text(30, ax1.get_ylim()[1]*0.7, 'n=30\nCLT Altın\nStandart', 
         ha='center', fontsize=9, fontweight='bold',
         bbox=dict(boxstyle='round', facecolor='#C8E6C9', alpha=0.8))

# PANEL 2: %95 Güven Aralığı Genişliği
ax2 = axes[0, 1]
bars = ax2.bar(range(len(df_sonuc)), df_sonuc['dagilim_genisligi'], 
               color='#FFB74D', alpha=0.7, edgecolor='darkorange', linewidth=2)

# n=30'u vurgula
bars[3].set_color('#81C784')
bars[3].set_edgecolor('darkgreen')

ax2.set_xticks(range(len(df_sonuc)))
ax2.set_xticklabels([f'n={n}' for n in df_sonuc['n']], fontsize=10)
ax2.set_xlabel('Örneklem Büyüklüğü', fontsize=12, fontweight='bold')
ax2.set_ylabel('±2SE Aralık (TL)', fontsize=12, fontweight='bold')
ax2.set_title('%95 Güven Aralığı Genişliği\n(Daha büyük n → Daha dar aralık)', 
              fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y', linestyle=':')

# Değerleri bar üzerine yaz
for i, (bar, val) in enumerate(zip(bars, df_sonuc['dagilim_genisligi'])):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 0.5,
             f'±{val:.1f} TL', ha='center', va='bottom', 
             fontsize=9, fontweight='bold')

# PANEL 3: Farklı n'ler için Dağılım Karşılaştırması
ax3 = axes[1, 0]

# n=10, 30, 100 için histogramlar
secili_n = [10, 30, 100]
renkler = ['#FFCCBC', '#81C784', '#B3E5FC']
labels_list = []

for n_val, renk in zip(secili_n, renkler):
    # Simülasyon
    orts = []
    for _ in range(3000):
        orn = np.random.choice(populasyon, size=n_val, replace=True)
        orts.append(orn.mean())
    
    ax3.hist(orts, bins=30, density=True, alpha=0.5, color=renk,
             edgecolor='black', linewidth=1.5, 
             label=f'n={n_val} (SE={sigma/np.sqrt(n_val):.1f})')

ax3.axvline(mu, color='red', linestyle='--', linewidth=3, label='μ (Gerçek)')
ax3.set_xlabel('Örneklem Ortalaması (TL)', fontsize=12, fontweight='bold')
ax3.set_ylabel('Yoğunluk', fontsize=12, fontweight='bold')
ax3.set_title('Farklı n için Örneklem Ortalamaları Dağılımı\n(Büyük n → Daha dar dağılım)', 
              fontsize=13, fontweight='bold')
ax3.legend(fontsize=10, loc='upper right')
ax3.grid(True, alpha=0.3, linestyle=':')

# PANEL 4: Hassasiyet Tablosu
ax4 = axes[1, 1]
ax4.axis('off')

# Tablo verisi
tablo_data = []
for i, row in df_sonuc.iterrows():
    n_val = int(row['n'])
    se_val = row['gercek_SE']
    aralik = row['dagilim_genisligi']
    
    if n_val < 30:
        yorum = "⚠️ Az güvenilir"
        renk_kodu = '#FFCCBC'
    elif n_val == 30:
        yorum = "✅ İdeal"
        renk_kodu = '#C8E6C9'
    elif n_val <= 100:
        yorum = "✅ Çok iyi"
        renk_kodu = '#B3E5FC'
    else:
        yorum = "🎯 Mükemmel (maliyetli)"
        renk_kodu = '#E1BEE7'
    
    tablo_data.append([f'n={n_val}', f'{se_val:.2f}', f'±{aralik:.1f}', yorum])

# Tablo çiz
tablo = ax4.table(cellText=tablo_data,
                  colLabels=['Örneklem\nBüyüklüğü', 'Standart\nHata (TL)', 
                            '%95 Güven\nAralığı', 'YBS Değerlendirme'],
                  cellLoc='center',
                  loc='center',
                  colWidths=[0.20, 0.20, 0.25, 0.35])

tablo.auto_set_font_size(False)
tablo.set_fontsize(10)
tablo.scale(1, 2.5)

# Başlık satırı formatla
for i in range(4):
    cell = tablo[(0, i)]
    cell.set_facecolor('#667eea')
    cell.set_text_props(weight='bold', color='white')

# Veri satırları formatla
for i in range(1, len(tablo_data) + 1):
    for j in range(4):
        cell = tablo[(i, j)]
        if df_sonuc.iloc[i-1]['n'] == 30:
            cell.set_facecolor('#C8E6C9')
            cell.set_text_props(weight='bold')
        else:
            cell.set_facecolor('white')

ax4.set_title('Örneklem Büyüklüğü Karar Tablosu\nKampüsSepeti İçin Öneriler', 
              fontsize=13, fontweight='bold', pad=20)

plt.suptitle('ÖRNEKLEM BÜYÜKLÜĞÜ ETKİSİ: CLT ANALİZİ\nStandart Hata (SE) ve Hassasiyet İlişkisi',
             fontsize=15, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("💡 YBS SONUÇ ve ÖNERILER")
print("="*70)
print("\n🎯 ÖRNEKLEM BÜYÜKLÜĞÜ STRATEJİSİ:")
print(f"   • n < 30: Güvensiz, CLT tam çalışmıyor")
print(f"   • n = 30: ✅ Altın standart, SE={sigma/np.sqrt(30):.2f} TL")
print(f"   • n = 100: Çok hassas, SE={sigma/np.sqrt(100):.2f} TL")
print(f"   • n > 200: Marjinal iyileşme, maliyet artışı fazla")

print("\n💼 KAMPÜSSEPETI İÇİN TAVSİYE:")
print("   • Günlük raporlama: Minimum n=30 sipariş")
print("   • Haftalık analiz: n=100+ sipariş (daha hassas)")
print("   • A/B testi: Her grup n≥30 olmalı")
print("   • SLA izleme: n=50 ideal (hassasiyet + maliyet dengesi)")

print("\n📐 FORMÜL HATIRLATMA:")
print("   SE = σ / √n")
print("   n 4 kat artarsa → SE yarıya iner")
print("   n 9 kat artarsa → SE 1/3'e iner")
print("="*70)
````

## 🎯 **NOKTA TAHMİNİ vs ARALIK TAHMİNİ**

### 📖 İki Tür Tahmin:

#### 1️⃣ **Nokta Tahmini (Point Estimate)**

> Tek bir sayı ile tahmin yaparız.

**Örnek**: KampüsSepeti'nin ortalama sepet tutarı **131 TL**'dir.

**Problem**:
- ❌ Ne kadar emin olduğumuzu söyleyemiyoruz
- ❌ Hata payını bilmiyoruz
- ❌ "131 TL" kesin mi? ±1 TL mi? ±10 TL mi?

---

> Bir aralık ile tahmin yaparız + güven seviyesi belirtiriz.

**Örnek**: KampüsSepeti'nin ortalama sepet tutarı **%95 güvenle [126, 136] TL** arasındadır.

**Avantajlar**:
- ✅ Belirsizliği ifade ediyoruz
- ✅ Hata payını gösteriyoruz
- ✅ Güven seviyesini bildiriyoruz

## 📊 **GÜVEN ARALIKLARI (CI) NEDİR?**

### 📖 Tanım:

> **Güven Aralığı (Confidence Interval)**: Popülasyon parametresinin (μ) **belirli bir güven seviyesinde** içinde bulunacağı değer aralığıdır.

### 🎯 Basit Dille:

Diyelim ki 100 kere örneklem aldık ve her biri için %95 güven aralığı hesapladık:
- **95 tanesinde** gerçek μ değeri aralığın içinde olur
- **5 tanesinde** aralık dışında kalır

Bu %95 **güven seviyesi** demektir!

---

### 📐 Formül (Popülasyon σ biliniyorsa - z-distribution):

$$
CI = \bar{x} \pm z_{\alpha/2} \times \frac{\sigma}{\sqrt{n}}
$$

**Veya**:

$$
CI = \bar{x} \pm z_{\alpha/2} \times SE
$$

### 🔑 Formül Bileşenleri:

| **Terim** | **Anlamı** | **Örnek** |
|:---|:---|:---|
| $\bar{x}$ | Örneklem ortalaması | 131 TL |
| $z_{\alpha/2}$ | Kritik z-değeri | 1.96 (%95 için) |
| $\sigma$ | Popülasyon std sapması | 45 TL |
| $n$ | Örneklem büyüklüğü | 100 sipariş |
| $SE$ | Standart Hata | σ/√n = 4.5 TL |

---

### 📊 Güven Seviyeleri ve Kritik Değerler:

| **Güven Seviyesi** | **α (Alpha)** | **z-kritik** | **Yorum** |
|:---:|:---:|:---:|:---|
| **90%** | 0.10 | **1.645** | Geniş aralık, az hassas |
| **95%** | 0.05 | **1.96** | ⭐ Standart (en yaygın) |
| **99%** | 0.01 | **2.576** | Dar hassasiyet, çok geniş aralık |

## 🔢 **PYTHON İLE GÜVEN ARALIKLARI HESAPLAMA**

Şimdi **KampüsSepeti verisi** ile %90, %95, %99 güven aralıkları hesaplayalım!

### 📋 Senaryo:

- **Veri**: Teslimat süreleri (5000 sipariş)
- **Örneklem**: Rastgele 100 sipariş seçeceğiz
- **Soru**: Gerçek ortalama teslimat süresi nedir?
- **Cevap**: %95 güvenle [Alt, Üst] aralığı

Başlayalım! 🚀

````python
# ============================================================================
# 📊 GÜVEN ARALIKLARI HESAPLAMA - TESLİMAT SÜRELERİ
# ============================================================================

print("📊 GÜVEN ARALIKLARI HESAPLAMA: TESLİMAT SÜRELERİ")
print("="*70)

# Popülasyon
populasyon = df_kampussepeti['teslimat_suresi_dk'].values
mu_gercek = populasyon.mean()
sigma_gercek = populasyon.std()

print(f"🎯 GERÇEK POPÜLASYON PARAMETRELERİ:")
print(f"   μ (Gerçek Ortalama): {mu_gercek:.2f} dakika")
print(f"   σ (Gerçek Std Sapma): {sigma_gercek:.2f} dakika")
print(f"   N (Popülasyon): {len(populasyon):,} sipariş")

# Örneklem al
np.random.seed(42)
n_orneklem = 100
orneklem = np.random.choice(populasyon, size=n_orneklem, replace=False)

# Örneklem istatistikleri
x_bar = orneklem.mean()
s_orneklem = orneklem.std(ddof=1)  # Örneklem std sapması
SE = sigma_gercek / np.sqrt(n_orneklem)  # Teorik SE (σ biliniyorsa)

print(f"\n📦 ÖRNEKLEM İSTATİSTİKLERİ:")
print(f"   x̄ (Örneklem Ortalaması): {x_bar:.2f} dakika")
print(f"   s (Örneklem Std Sapma): {s_orneklem:.2f} dakika")
print(f"   n (Örneklem Büyüklüğü): {n_orneklem}")
print(f"   SE (Standart Hata): {SE:.2f} dakika")

print("\n" + "="*70)

# -------------------------------------------------------------------------
# 3 FARKLI GÜVEN SEVİYESİ İÇİN CI HESAPLA
# -------------------------------------------------------------------------

guven_seviyeleri = [90, 95, 99]
z_kritik_degerler = [1.645, 1.96, 2.576]
renkler = ['#FFE0B2', '#C8E6C9', '#B3E5FC']

print("\n📐 GÜVEN ARALIKLARI HESAPLAMALARI:")
print("="*70)

ci_sonuclari = []

for guven, z_krit, renk in zip(guven_seviyeleri, z_kritik_degerler, renkler):
    # Marjin of Error (Hata Payı)
    margin_of_error = z_krit * SE
    
    # Güven Aralığı
    ci_alt = x_bar - margin_of_error
    ci_ust = x_bar + margin_of_error
    
    # Genişlik
    genislik = ci_ust - ci_alt
    
    # Gerçek μ aralıkta mı?
    icinde_mi = ci_alt <= mu_gercek <= ci_ust
    
    ci_sonuclari.append({
        'guven': guven,
        'z_krit': z_krit,
        'margin': margin_of_error,
        'alt': ci_alt,
        'ust': ci_ust,
        'genislik': genislik,
        'icinde': icinde_mi,
        'renk': renk
    })
    
    print(f"\n{'='*70}")
    print(f"🎯 %{guven} GÜVEN ARALIĞI:")
    print(f"{'='*70}")
    print(f"   z-kritik: {z_krit}")
    print(f"   Margin of Error (±): {margin_of_error:.2f} dakika")
    print(f"   Alt Sınır: {ci_alt:.2f} dakika")
    print(f"   Üst Sınır: {ci_ust:.2f} dakika")
    print(f"   Aralık Genişliği: {genislik:.2f} dakika")
    print(f"   Gerçek μ ({mu_gercek:.2f}) aralıkta mı? {'✅ EVET' if icinde_mi else '❌ HAYIR'}")
    print(f"\n   💼 YBS İfadesi:")
    print(f"   'Ortalama teslimat süresi %{guven} güvenle")
    print(f"    [{ci_alt:.1f}, {ci_ust:.1f}] dakika arasındadır.'")

print("\n" + "="*70)

# -------------------------------------------------------------------------
# KARŞILAŞTIRMA TABLOSU
# -------------------------------------------------------------------------

print("\n📊 GÜVEN SEVİYELERİ KARŞILAŞTIRMA TABLOSU:")
print("="*70)
print(f"{'Güven':<10} {'z-kritik':<12} {'Margin':<12} {'Alt':<10} {'Üst':<10} {'Genişlik':<12}")
print("-"*70)

for sonuc in ci_sonuclari:
    print(f"%{sonuc['guven']:<9} {sonuc['z_krit']:<12.3f} ±{sonuc['margin']:<10.2f} "
          f"{sonuc['alt']:<10.2f} {sonuc['ust']:<10.2f} {sonuc['genislik']:<12.2f}")

print("="*70)

print("\n💡 GÖZLEMLER:")
print("   1. Güven seviyesi ↑ → Aralık genişliği ↑")
print("   2. %90: En dar, ama daha az güvenli")
print("   3. %95: ⭐ Standart tercih (dengeli)")
print("   4. %99: En geniş, ama çok güvenli")
print("   5. Tüm aralıklar gerçek μ'yü içeriyor! ✅")
print("="*70)
````

````python
# ============================================================================
# 📊 GÜVEN ARALIKLARI GÖRSELLEŞTİRME
# ============================================================================

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# -------------------------------------------------------------------------
# PANEL 1: Normal Dağılım ile CI Gösterimi
# -------------------------------------------------------------------------

ax1 = axes[0, 0]

# Normal dağılım eğrisi
x_range = np.linspace(mu_gercek - 4*SE, mu_gercek + 4*SE, 1000)
y_normal = stats.norm.pdf(x_range, mu_gercek, SE)

ax1.plot(x_range, y_normal, 'k-', linewidth=2, label='Teorik Dağılım')

# Gerçek μ
ax1.axvline(mu_gercek, color='red', linestyle='--', linewidth=3, 
            label=f'Gerçek μ = {mu_gercek:.2f}', alpha=0.7)

# Örneklem ortalaması
ax1.axvline(x_bar, color='darkgreen', linestyle=':', linewidth=3,
            label=f'Örneklem x̄ = {x_bar:.2f}', alpha=0.7)

# %95 CI gölgeleme
ci_95 = ci_sonuclari[1]  # %95
x_fill = x_range[(x_range >= ci_95['alt']) & (x_range <= ci_95['ust'])]
y_fill = stats.norm.pdf(x_fill, mu_gercek, SE)
ax1.fill_between(x_fill, 0, y_fill, alpha=0.3, color='green', 
                 label='%95 CI Bölgesi')

ax1.axvline(ci_95['alt'], color='green', linestyle='-', linewidth=2, alpha=0.7)
ax1.axvline(ci_95['ust'], color='green', linestyle='-', linewidth=2, alpha=0.7)

ax1.set_xlabel('Teslimat Süresi (dakika)', fontsize=11, fontweight='bold')
ax1.set_ylabel('Yoğunluk', fontsize=11, fontweight='bold')
ax1.set_title('%95 Güven Aralığı Görselleştirme\nNormal Dağılım Üzerinde',
              fontsize=12, fontweight='bold')
ax1.legend(fontsize=9, loc='upper right')
ax1.grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# PANEL 2: 3 Güven Seviyesi Karşılaştırması
# -------------------------------------------------------------------------

ax2 = axes[0, 1]

y_pozisyonlar = [3, 2, 1]

for i, (sonuc, y_pos) in enumerate(zip(ci_sonuclari, y_pozisyonlar)):
    # Aralık çizgisi
    ax2.plot([sonuc['alt'], sonuc['ust']], [y_pos, y_pos], 
             linewidth=8, color=sonuc['renk'], alpha=0.7,
             label=f"%{sonuc['guven']} CI")
    
    # Uç noktalar
    ax2.plot([sonuc['alt'], sonuc['ust']], [y_pos, y_pos], 
             'ko', markersize=8)
    
    # Merkez (x_bar)
    ax2.plot(x_bar, y_pos, 'g^', markersize=12)
    
    # Etiketler
    ax2.text(sonuc['alt'] - 0.3, y_pos, f"{sonuc['alt']:.1f}", 
             ha='right', va='center', fontsize=9, fontweight='bold')
    ax2.text(sonuc['ust'] + 0.3, y_pos, f"{sonuc['ust']:.1f}", 
             ha='left', va='center', fontsize=9, fontweight='bold')
    ax2.text(x_bar, y_pos + 0.15, f"x̄={x_bar:.1f}", 
             ha='center', va='bottom', fontsize=8, fontweight='bold', color='darkgreen')

# Gerçek μ çizgisi
ax2.axvline(mu_gercek, color='red', linestyle='--', linewidth=2,
            label=f'Gerçek μ = {mu_gercek:.2f}', alpha=0.7)

ax2.set_yticks(y_pozisyonlar)
ax2.set_yticklabels(['%90 CI', '%95 CI', '%99 CI'], fontsize=10, fontweight='bold')
ax2.set_xlabel('Teslimat Süresi (dakika)', fontsize=11, fontweight='bold')
ax2.set_title('Güven Seviyeleri Karşılaştırması\nAralık Genişlikleri',
              fontsize=12, fontweight='bold')
ax2.legend(fontsize=9, loc='upper right')
ax2.grid(True, alpha=0.3, axis='x', linestyle=':')
ax2.set_xlim(mu_gercek - 3, mu_gercek + 3)

# -------------------------------------------------------------------------
# PANEL 3: Aralık Genişliği Bar Chart
# -------------------------------------------------------------------------

ax3 = axes[1, 0]

labels = [f'%{s["guven"]}' for s in ci_sonuclari]
genislikler = [s['genislik'] for s in ci_sonuclari]
renkler_bar = [s['renk'] for s in ci_sonuclari]

bars = ax3.bar(labels, genislikler, color=renkler_bar, alpha=0.7,
               edgecolor='black', linewidth=2)

# Değerleri bar üzerine yaz
for bar, gen in zip(bars, genislikler):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2., height + 0.05,
             f'{gen:.2f} dk', ha='center', va='bottom',
             fontsize=11, fontweight='bold')

ax3.set_ylabel('Aralık Genişliği (dakika)', fontsize=11, fontweight='bold')
ax3.set_xlabel('Güven Seviyesi', fontsize=11, fontweight='bold')
ax3.set_title('Güven Seviyesi vs Aralık Genişliği\n(Yüksek Güven = Geniş Aralık)',
              fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='y', linestyle=':')

# Ok ve açıklama
ax3.annotate('Güven ↑\nGenişlik ↑', xy=(2, genislikler[2]), 
             xytext=(1.5, genislikler[2] + 0.4),
             arrowprops=dict(arrowstyle='->', color='red', linewidth=2),
             fontsize=10, fontweight='bold', color='red')

# -------------------------------------------------------------------------
# PANEL 4: Margin of Error Karşılaştırması
# -------------------------------------------------------------------------

ax4 = axes[1, 1]

labels_me = [f'%{s["guven"]}' for s in ci_sonuclari]
margins = [s['margin'] for s in ci_sonuclari]
z_values = [s['z_krit'] for s in ci_sonuclari]

# Dual y-axis
ax4_twin = ax4.twinx()

# Margin bar chart
bars_me = ax4.bar(labels_me, margins, color='#FFB74D', alpha=0.7,
                  edgecolor='darkorange', linewidth=2, label='Margin of Error')

# z-critical çizgi grafiği
line_z = ax4_twin.plot(labels_me, z_values, 'ro-', linewidth=3,
                       markersize=10, label='z-critical', alpha=0.7)

# Değerleri yaz
for i, (bar, margin, z) in enumerate(zip(bars_me, margins, z_values)):
    height = bar.get_height()
    ax4.text(bar.get_x() + bar.get_width()/2., height + 0.03,
             f'±{margin:.2f}', ha='center', va='bottom',
             fontsize=10, fontweight='bold', color='darkorange')
    
    ax4_twin.text(i, z + 0.05, f'z={z:.3f}', ha='center', va='bottom',
                  fontsize=9, fontweight='bold', color='darkred')

ax4.set_xlabel('Güven Seviyesi', fontsize=11, fontweight='bold')
ax4.set_ylabel('Margin of Error (dakika)', fontsize=11, fontweight='bold', color='darkorange')
ax4_twin.set_ylabel('z-critical Değeri', fontsize=11, fontweight='bold', color='darkred')

ax4.set_title('Margin of Error ve z-critical İlişkisi\n(z↑ → Margin↑)',
              fontsize=12, fontweight='bold')
ax4.grid(True, alpha=0.3, axis='y', linestyle=':')

# Legend
lines1, labels1 = ax4.get_legend_handles_labels()
lines2, labels2 = ax4_twin.get_legend_handles_labels()
ax4.legend(lines1 + lines2, labels1 + labels2, fontsize=9, loc='upper left')

plt.suptitle('GÜVEN ARALIKLARI ANALİZİ: KAMPÜSSEPETİ TESLİMAT SÜRELERİ\n'
             f'Örneklem: n={n_orneklem}, x̄={x_bar:.2f} dk, SE={SE:.2f} dk',
             fontsize=14, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("📊 GÖRSEL YORUMLARI:")
print("="*70)
print("\n📍 PANEL 1 (Üst Sol): Normal dağılım üzerinde %95 CI")
print("   • Yeşil alan: Gerçek μ'nün %95 ihtimalle olduğu bölge")
print("   • Kırmızı çizgi: Gerçek μ (28.00 dk)")
print("   • Yeşil noktalı: Örneklem ortalaması x̄")

print("\n📏 PANEL 2 (Üst Sağ): 3 güven seviyesi yan yana")
print("   • %99 en geniş, %90 en dar")
print("   • Hepsi gerçek μ'yü içeriyor ✅")

print("\n📊 PANEL 3 (Alt Sol): Genişlik karşılaştırma")
print("   • Güven arttıkça genişlik artar")
print("   • Trade-off: Güven vs Hassasiyet")

print("\n📐 PANEL 4 (Alt Sağ): Margin ve z-critical")
print("   • z-critical arttıkça margin artar")
print("   • Formül: Margin = z × SE")
print("="*70)
````

## 📊 **ÖRNEKLEM BÜYÜKLÜĞÜNÜN CI'YE ETKİSİ**

### 🎯 Soru:

KampüsSepeti'nin **farklı günlerde** farklı sipariş sayıları olduğunu düşün:
- **Pazartesi**: 30 sipariş (az)
- **Çarşamba**: 100 sipariş (orta)
- **Cuma**: 300 sipariş (çok)

Her gün için %95 CI hesaplarsak, hangi gün **daha hassas** tahmin verir?

### 💡 Tahmin:

n ↑ → SE ↓ → CI daralır → **Daha hassas!**

Şimdi bunu Python ile kanıtlayalım! 👇

````python
# ============================================================================
# 📊 FARKLI n DEĞERLERİ İLE CI KARŞILAŞTIRMASI
# ============================================================================

print("📊 ÖRNEKLEM BÜYÜKLÜĞÜ ETKİSİ: CI KARŞILAŞTIRMASI")
print("="*70)

# Popülasyon
populasyon = df_kampussepeti['teslimat_suresi_dk'].values
mu_gercek = populasyon.mean()
sigma = populasyon.std()

print(f"🎯 POPÜLASYON: μ={mu_gercek:.2f} dk, σ={sigma:.2f} dk")
print("="*70)

# Farklı n değerleri (farklı günler)
n_degerler = [30, 50, 100, 200, 300, 500]
gunler = ["Pazartesi", "Salı", "Çarşamba", "Perşembe", "Cuma", "Cumartesi"]
renkler = ['#FFCCBC', '#FFE0B2', '#FFF9C4', '#C8E6C9', '#B3E5FC', '#E1BEE7']

ci_sonuclari_n = []

np.random.seed(42)

for n, gun, renk in zip(n_degerler, gunler, renkler):
    # Örneklem al
    orneklem = np.random.choice(populasyon, size=n, replace=False)
    x_bar = orneklem.mean()
    
    # SE hesapla
    SE = sigma / np.sqrt(n)
    
    # %95 CI hesapla
    z_95 = 1.96
    margin = z_95 * SE
    ci_alt = x_bar - margin
    ci_ust = x_bar + margin
    genislik = ci_ust - ci_alt
    
    # Gerçek μ içinde mi?
    icinde = ci_alt <= mu_gercek <= ci_ust
    
    ci_sonuclari_n.append({
        'n': n,
        'gun': gun,
        'x_bar': x_bar,
        'SE': SE,
        'margin': margin,
        'alt': ci_alt,
        'ust': ci_ust,
        'genislik': genislik,
        'icinde': icinde,
        'renk': renk
    })
    
    print(f"\n📅 {gun} (n={n}):")
    print(f"   x̄ = {x_bar:.2f} dk")
    print(f"   SE = {SE:.2f} dk")
    print(f"   %95 CI: [{ci_alt:.2f}, {ci_ust:.2f}] dk")
    print(f"   Genişlik: {genislik:.2f} dk")
    print(f"   μ içinde mi? {'✅ EVET' if icinde else '❌ HAYIR'}")

print("\n" + "="*70)
print("📊 KARŞILAŞTIRMA TABLOSU:")
print("="*70)
print(f"{'Gün':<12} {'n':<8} {'SE':<10} {'Genişlik':<12} {'Hassasiyet':<15}")
print("-"*70)

for sonuc in ci_sonuclari_n:
    hassasiyet = "Çok Düşük" if sonuc['genislik'] > 2.5 else \
                 "Düşük" if sonuc['genislik'] > 2.0 else \
                 "Orta" if sonuc['genislik'] > 1.5 else \
                 "İyi" if sonuc['genislik'] > 1.0 else "Mükemmel"
    
    print(f"{sonuc['gun']:<12} {sonuc['n']:<8} {sonuc['SE']:<10.2f} "
          f"{sonuc['genislik']:<12.2f} {hassasiyet:<15}")

print("="*70)

print("\n💡 GÖZLEMLER:")
print("   1. n arttıkça SE azalır (σ/√n formülü)")
print("   2. SE azaldıkça CI genişliği azalır")
print("   3. n=30: ±2.14 dk (geniş aralık)")
print("   4. n=300: ±0.68 dk (dar aralık)")
print("   5. n 4 kat artarsa → Genişlik yarıya iner!")
print("="*70)
````

## 🔄 **BOOTSTRAP YÖNTEMI İLE CI**

### 🤔 Problem:

Şimdiye kadar **σ (popülasyon std sapması) biliniyordu**. Ama gerçek hayatta:
- ❌ σ'yı her zaman bilmeyiz
- ❌ Popülasyonun tamamına erişemeyiz
- ❌ Sadece örneklemimiz var

### 💡 Çözüm: BOOTSTRAP!

> **Bootstrap**: Mevcut örneklemden **tekrar tekrar örneklem alarak** (resampling) CI hesaplama yöntemi.

---

### 📐 Bootstrap Algoritması:

```mermaid
flowchart TD
    A["📦 Orijinal Örneklem<br/>n=100 gözlem"] --> B["🔄 Resampling<br/>(Yerine koyarak seçim)"]
    
    B --> C["Bootstrap Örnek 1<br/>n=100"]
    B --> D["Bootstrap Örnek 2<br/>n=100"]
    B --> E["Bootstrap Örnek 3<br/>n=100"]
    B --> F["...<br/>10,000 örnek daha"]
    
    C --> G["Ortalama 1"]
    D --> H["Ortalama 2"]
    E --> I["Ortalama 3"]
    F --> J["10,000 ortalama"]
    
    G --> K["📊 Bootstrap Dağılımı<br/>10,000 ortalama"]
    H --> K
    I --> K
    J --> K
    
    K --> L["%95 CI:<br/>2.5 ve 97.5 yüzdelik"]
    
    style A fill:#FFF9C4,stroke:#F57F17,stroke-width:2px
    style B fill:#FFE0B2,stroke:#EF6C00,stroke-width:2px
    style K fill:#C8E6C9,stroke:#388E3C,stroke-width:3px
    style L fill:#B3E5FC,stroke:#0277BD,stroke-width:3px
```

---

### 🎯 Avantajları:

1. ✅ **σ bilmeden** CI hesaplanabilir
2. ✅ **Dağılım varsayımı yok** (normal olmasa da olur)
3. ✅ **Küçük örneklemler** için bile çalışır
4. ✅ **Pratik ve esnek**

---

### 📋 **FLASHCARD #15: BOOTSTRAP**

| **Özellik** | **Parametrik CI** | **Bootstrap CI** |
|:---|:---:|:---:|
| **σ gerekli mi?** | ✅ Evet | ❌ Hayır |
| **Normal dağılım varsayımı?** | ✅ Evet | ❌ Hayır |
| **Hesaplama?** | Formül | Simülasyon |
| **Kullanım?** | σ biliniyorsa | σ bilinmiyorsa |
| **Hız?** | Hızlı | Yavaş (10K iterasyon) |

---

Şimdi Python ile bootstrap CI hesaplayalım! 👇

---

````python
# ============================================================================
# 🔄 BOOTSTRAP YÖNTEM İLE CI HESAPLAMA
# ============================================================================

print("🔄 BOOTSTRAP YÖNTEMİ İLE GÜVEN ARALIĞI")
print("="*70)

# Orijinal örneklem al
np.random.seed(42)
n_orneklem = 100
orneklem_orijinal = np.random.choice(df_kampussepeti['teslimat_suresi_dk'].values,
                                     size=n_orneklem, replace=False)

x_bar_orijinal = orneklem_orijinal.mean()
s_orijinal = orneklem_orijinal.std(ddof=1)

print(f"📦 ORİJİNAL ÖRNEKLEM:")
print(f"   n = {n_orneklem}")
print(f"   x̄ = {x_bar_orijinal:.2f} dakika")
print(f"   s = {s_orijinal:.2f} dakika")
print(f"   ⚠️ NOT: σ (popülasyon std) bilinmiyor!")

# -------------------------------------------------------------------------
# BOOTSTRAP: 10,000 İTERASYON
# -------------------------------------------------------------------------

n_bootstrap = 10000
bootstrap_ortalamalar = []

print(f"\n🔄 BOOTSTRAP SİMÜLASYONU:")
print(f"   {n_bootstrap:,} kez resampling yapılıyor...")

for i in range(n_bootstrap):
    # Yerine koyarak örneklem al (with replacement)
    bootstrap_sample = np.random.choice(orneklem_orijinal, size=n_orneklem, replace=True)
    bootstrap_ortalamalar.append(bootstrap_sample.mean())
    
    if (i + 1) % 2000 == 0:
        print(f"   ✓ {i+1:,} / {n_bootstrap:,} tamamlandı...")

bootstrap_ortalamalar = np.array(bootstrap_ortalamalar)

print("\n✅ Bootstrap tamamlandı!")

# -------------------------------------------------------------------------
# BOOTSTRAP CI HESAPLAMA: Percentile Metodu
# -------------------------------------------------------------------------

# %95 CI: 2.5 ve 97.5 yüzdelik dilimler
ci_alt_bootstrap = np.percentile(bootstrap_ortalamalar, 2.5)
ci_ust_bootstrap = np.percentile(bootstrap_ortalamalar, 97.5)
genislik_bootstrap = ci_ust_bootstrap - ci_alt_bootstrap

print("\n" + "="*70)
print("📊 BOOTSTRAP %95 GÜVEN ARALIĞI:")
print("="*70)
print(f"   Alt Sınır (2.5 yüzdelik): {ci_alt_bootstrap:.2f} dakika")
print(f"   Üst Sınır (97.5 yüzdelik): {ci_ust_bootstrap:.2f} dakika")
print(f"   CI: [{ci_alt_bootstrap:.2f}, {ci_ust_bootstrap:.2f}] dakika")
print(f"   Genişlik: {genislik_bootstrap:.2f} dakika")

# -------------------------------------------------------------------------
# KARŞILAŞTIRMA: Parametrik vs Bootstrap
# -------------------------------------------------------------------------

# Parametrik CI (σ biliniyorsa)
mu_gercek = df_kampussepeti['teslimat_suresi_dk'].mean()
sigma_gercek = df_kampussepeti['teslimat_suresi_dk'].std()
SE_parametrik = sigma_gercek / np.sqrt(n_orneklem)
ci_alt_parametrik = x_bar_orijinal - 1.96 * SE_parametrik
ci_ust_parametrik = x_bar_orijinal + 1.96 * SE_parametrik
genislik_parametrik = ci_ust_parametrik - ci_alt_parametrik

print("\n" + "="*70)
print("📊 PARAMETRİK vs BOOTSTRAP KARŞILAŞTIRMA:")
print("="*70)
print(f"\n🔢 PARAMETRİK CI (σ biliniyorsa):")
print(f"   Formula: x̄ ± 1.96 × σ/√n")
print(f"   CI: [{ci_alt_parametrik:.2f}, {ci_ust_parametrik:.2f}] dakika")
print(f"   Genişlik: {genislik_parametrik:.2f} dakika")

print(f"\n🔄 BOOTSTRAP CI (σ bilinmiyorsa):")
print(f"   Metod: 10,000 resampling + percentile")
print(f"   CI: [{ci_alt_bootstrap:.2f}, {ci_ust_bootstrap:.2f}] dakika")
print(f"   Genişlik: {genislik_bootstrap:.2f} dakika")

fark_alt = abs(ci_alt_parametrik - ci_alt_bootstrap)
fark_ust = abs(ci_ust_parametrik - ci_ust_bootstrap)

print(f"\n📏 FARKLAR:")
print(f"   Alt sınır farkı: {fark_alt:.2f} dakika")
print(f"   Üst sınır farkı: {fark_ust:.2f} dakika")
print(f"   ✅ Bootstrap parametrik yönteme çok yakın!")

# Gerçek μ her iki aralıkta da mı?
icinde_parametrik = ci_alt_parametrik <= mu_gercek <= ci_ust_parametrik
icinde_bootstrap = ci_alt_bootstrap <= mu_gercek <= ci_ust_bootstrap

print(f"\n🎯 GERÇEK μ ({mu_gercek:.2f}) KARŞILAŞTIRMASI:")
print(f"   Parametrik CI içinde mi? {'✅ EVET' if icinde_parametrik else '❌ HAYIR'}")
print(f"   Bootstrap CI içinde mi? {'✅ EVET' if icinde_bootstrap else '❌ HAYIR'}")

print("\n" + "="*70)
print("💡 BOOTSTRAP AVANTAJLARI:")
print("   1. σ bilmeden CI hesaplanabiliyor")
print("   2. Normal dağılım varsayımı gerekmiyor")
print("   3. Küçük örneklemler için güvenilir")
print("   4. Parametrik yönteme çok yakın sonuçlar")
print("\n⚠️ DEZAVANTAJ:")
print("   • Hesaplama yoğun (10,000 iterasyon)")
print("   • Parametrik yöntemden yavaş")
print("="*70)
````

````python
# ============================================================================
# 📊 BOOTSTRAP vs PARAMETRİK GÖRSELLEŞTİRME
# ============================================================================

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# -------------------------------------------------------------------------
# PANEL 1: Bootstrap Dağılımı Histogram
# -------------------------------------------------------------------------

ax1 = axes[0, 0]

# Histogram
ax1.hist(bootstrap_ortalamalar, bins=50, density=True, alpha=0.7,
         color='#81C784', edgecolor='darkgreen', linewidth=1.5,
         label='Bootstrap Dağılımı')

# Normal eğri ekle (karşılaştırma için)
x_range = np.linspace(bootstrap_ortalamalar.min(), bootstrap_ortalamalar.max(), 100)
ax1.plot(x_range, stats.norm.pdf(x_range, x_bar_orijinal, bootstrap_ortalamalar.std()),
         'r-', linewidth=3, label='Normal Fit', alpha=0.7)

# Merkez ve CI çizgileri
ax1.axvline(x_bar_orijinal, color='blue', linestyle=':', linewidth=2,
            label=f'x̄ = {x_bar_orijinal:.2f}')
ax1.axvline(ci_alt_bootstrap, color='green', linestyle='--', linewidth=2,
            label=f'2.5% = {ci_alt_bootstrap:.2f}')
ax1.axvline(ci_ust_bootstrap, color='green', linestyle='--', linewidth=2,
            label=f'97.5% = {ci_ust_bootstrap:.2f}')
ax1.axvline(mu_gercek, color='red', linestyle='-', linewidth=2,
            label=f'Gerçek μ = {mu_gercek:.2f}', alpha=0.7)

# CI bölgesini gölgele
x_fill = x_range[(x_range >= ci_alt_bootstrap) & (x_range <= ci_ust_bootstrap)]
y_fill = stats.norm.pdf(x_fill, x_bar_orijinal, bootstrap_ortalamalar.std())
ax1.fill_between(x_fill, 0, y_fill, alpha=0.3, color='green',
                 label='%95 CI Bölgesi')

ax1.set_xlabel('Teslimat Süresi (dakika)', fontsize=11, fontweight='bold')
ax1.set_ylabel('Yoğunluk', fontsize=11, fontweight='bold')
ax1.set_title('Bootstrap Dağılımı (10,000 Resampling)\n%95 CI Görselleştirme',
              fontsize=13, fontweight='bold')
ax1.legend(fontsize=8, loc='upper right')
ax1.grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# PANEL 2: Parametrik vs Bootstrap CI Karşılaştırma
# -------------------------------------------------------------------------

ax2 = axes[0, 1]

y_pozisyonlar = [2, 1]
labels = ['Parametrik\n(σ biliniyorsa)', 'Bootstrap\n(σ bilinmiyorsa)']
ci_ranges = [
    (ci_alt_parametrik, ci_ust_parametrik),
    (ci_alt_bootstrap, ci_ust_bootstrap)
]
colors = ['#FFB74D', '#81C784']

for i, (y_pos, label, ci_range, color) in enumerate(zip(y_pozisyonlar, labels, ci_ranges, colors)):
    # CI çizgisi
    ax2.plot([ci_range[0], ci_range[1]], [y_pos, y_pos],
             linewidth=10, color=color, alpha=0.7)
    
    # Uç noktalar
    ax2.plot([ci_range[0], ci_range[1]], [y_pos, y_pos],
             'ko', markersize=8)
    
    # Merkez
    ax2.plot(x_bar_orijinal, y_pos, 'b^', markersize=12)
    
    # Etiketler
    ax2.text(ci_range[0] - 0.2, y_pos, f'{ci_range[0]:.2f}',
             ha='right', va='center', fontsize=10, fontweight='bold')
    ax2.text(ci_range[1] + 0.2, y_pos, f'{ci_range[1]:.2f}',
             ha='left', va='center', fontsize=10, fontweight='bold')
    
    # Metod etiketi
    ax2.text(mu_gercek - 2, y_pos, label,
             ha='right', va='center', fontsize=9, fontweight='bold')

# Gerçek μ
ax2.axvline(mu_gercek, color='red', linestyle='--', linewidth=2,
            label=f'Gerçek μ = {mu_gercek:.2f}', alpha=0.7)

ax2.set_yticks([])
ax2.set_xlabel('Teslimat Süresi (dakika)', fontsize=11, fontweight='bold')
ax2.set_title('Parametrik vs Bootstrap CI Karşılaştırma\n(İki yöntem de gerçek μ\'yü içeriyor)',
              fontsize=13, fontweight='bold')
ax2.legend(fontsize=10, loc='upper right')
ax2.grid(True, alpha=0.3, axis='x', linestyle=':')
ax2.set_xlim(mu_gercek - 2.5, mu_gercek + 2.5)

# -------------------------------------------------------------------------
# PANEL 3: Bootstrap İterasyon Yakınsaması
# -------------------------------------------------------------------------

ax3 = axes[1, 0]

# İlk 1000 iterasyon için kümülatif CI
iterasyonlar = [100, 500, 1000, 2000, 5000, 10000]
ci_alt_iter = []
ci_ust_iter = []

for n_iter in iterasyonlar:
    alt = np.percentile(bootstrap_ortalamalar[:n_iter], 2.5)
    ust = np.percentile(bootstrap_ortalamalar[:n_iter], 97.5)
    ci_alt_iter.append(alt)
    ci_ust_iter.append(ust)

# Grafik
ax3.plot(iterasyonlar, ci_alt_iter, 'ro-', linewidth=2, markersize=8,
         label='Alt Sınır', alpha=0.7)
ax3.plot(iterasyonlar, ci_ust_iter, 'bo-', linewidth=2, markersize=8,
         label='Üst Sınır', alpha=0.7)

# Parametrik CI referans çizgileri
ax3.axhline(ci_alt_parametrik, color='red', linestyle='--', linewidth=1.5,
            alpha=0.5, label='Parametrik Alt')
ax3.axhline(ci_ust_parametrik, color='blue', linestyle='--', linewidth=1.5,
            alpha=0.5, label='Parametrik Üst')

ax3.set_xlabel('Bootstrap İterasyon Sayısı', fontsize=11, fontweight='bold')
ax3.set_ylabel('CI Sınırları (dakika)', fontsize=11, fontweight='bold')
ax3.set_title('Bootstrap Yakınsaması\n(İterasyon arttıkça stabilleşir)',
              fontsize=13, fontweight='bold')
ax3.legend(fontsize=9, loc='best')
ax3.grid(True, alpha=0.3, linestyle=':')
ax3.set_xscale('log')

# -------------------------------------------------------------------------
# PANEL 4: Özet Tablo
# -------------------------------------------------------------------------

ax4 = axes[1, 1]
ax4.axis('off')

# Karşılaştırma tablosu
tablo_data = [
    ['Formül', 'x̄ ± z×SE', '2.5%-97.5% percentile'],
    ['σ Gerekli mi?', '✅ Evet', '❌ Hayır'],
    ['Normal Varsayımı?', '✅ Evet', '❌ Hayır'],
    ['Alt Sınır', f'{ci_alt_parametrik:.2f} dk', f'{ci_alt_bootstrap:.2f} dk'],
    ['Üst Sınır', f'{ci_ust_parametrik:.2f} dk', f'{ci_ust_bootstrap:.2f} dk'],
    ['Genişlik', f'{genislik_parametrik:.2f} dk', f'{genislik_bootstrap:.2f} dk'],
    ['μ İçinde mi?', '✅ Evet' if icinde_parametrik else '❌ Hayır',
     '✅ Evet' if icinde_bootstrap else '❌ Hayır'],
    ['Hız', '⚡ Hızlı', '🐌 Yavaş (10K iter)'],
    ['Kullanım', 'σ biliniyorsa', 'σ bilinmiyorsa']
]

tablo = ax4.table(cellText=tablo_data,
                  colLabels=['Özellik', 'Parametrik CI', 'Bootstrap CI'],
                  cellLoc='center',
                  loc='center',
                  colWidths=[0.30, 0.35, 0.35])

tablo.auto_set_font_size(False)
tablo.set_fontsize(9)
tablo.scale(1, 2.0)

# Başlık satırı
for i in range(3):
    cell = tablo[(0, i)]
    if i == 0:
        cell.set_facecolor('#9E9E9E')
    elif i == 1:
        cell.set_facecolor('#FFB74D')
    else:
        cell.set_facecolor('#81C784')
    cell.set_text_props(weight='bold', color='white')

# Veri satırları
for i in range(1, len(tablo_data) + 1):
    for j in range(3):
        cell = tablo[(i, j)]
        if j == 0:
            cell.set_facecolor('#EEEEEE')
        else:
            cell.set_facecolor('white')

ax4.set_title('Parametrik vs Bootstrap CI: Detaylı Karşılaştırma\nKampüsSepeti için Karar Matrisi',
              fontsize=13, fontweight='bold', pad=20)

plt.suptitle('BOOTSTRAP YÖNTEMİ ANALİZİ\nσ Bilinmiyorsa Güven Aralığı Hesaplama',
             fontsize=14, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("💼 YBS ÖNERİSİ: BOOTSTRAP NE ZAMAN KULLANILMALI?")
print("="*70)
print("\n✅ BOOTSTRAP KULLAN:")
print("   1. Popülasyon σ bilinmiyorsa")
print("   2. Veri normal dağılmıyorsa (çarpık, heavy-tail)")
print("   3. Küçük örneklemler (n<30)")
print("   4. Karmaşık istatistikler için (medyan, quantile)")
print("\n🔢 PARAMETRİK KULLAN:")
print("   1. σ biliniyorsa veya n çok büyükse")
print("   2. Veri normal dağıldığı biliniyorsa")
print("   3. Hız önemliyse (gerçek zamanlı rapor)")
print("\n🎯 KAMPÜSSEPETI İÇİN:")
print("   • Tarihsel σ var → Parametrik kullan")
print("   • Yeni metrik → Bootstrap ile başla")
print("   • A/B testi → Bootstrap daha güvenilir")
print("="*70)
````

## 🎯 **HİPOTEZ TESTLERİ NEDİR?**

### 📖 Tanım:

> **Hipotez Testi**: Veri kullanarak bir iddianın (hipotezin) doğruluğunu test etme sürecidir.

### 🤔 Gerçek Hayat Örneği:

**İddia**: "KampüsSepeti'nin yeni kurye sistemi teslimat süresini azalttı!"

**Soru**: Bu iddia verilerle destekleniyor mu?

**Yöntem**: Hipotez testi! 🔬

---

### 🎯 **İki Tip Hipotez:**

#### 1️⃣ **Null Hypothesis (H₀) - Sıfır Hipotezi**

> "Bir değişiklik/fark YOK" iddiası.

**Örnek**: 
- H₀: Yeni sistem teslimat süresini DEĞİŞTİRMEDİ (μ₁ = μ₂)
- H₀: İlaç etkisiz (μ = μ₀)
- H₀: Kampanya sepet tutarını artırmadı

**Özellik**: Varsayılan durum, "status quo"

---

#### 2️⃣ **Alternative Hypothesis (H₁ veya Hₐ) - Alternatif Hipotez**

> "Bir değişiklik/fark VAR" iddiası.

**Örnek**:
- H₁: Yeni sistem teslimat süresini AZALTTI (μ₁ < μ₂)
- H₁: İlaç etkili (μ ≠ μ₀)
- H₁: Kampanya sepet tutarını artırdı (μ₁ > μ₂)

**Özellik**: İspatlamak istediğimiz durum

---

### 📊 **3 Tip Alternatif Hipotez:**

| **Tip** | **Sembol** | **Anlamı** | **Örnek** |
|:---|:---:|:---|:---|
| **İki Kuyruklu** | μ ≠ μ₀ | Fark var (hangi yönde olursa) | "Teslimat süresi değişti" |
| **Sol Kuyruklu** | μ < μ₀ | Azaldı | "Teslimat süresi azaldı" |
| **Sağ Kuyruklu** | μ > μ₀ | Arttı | "Sepet tutarı arttı" |

## 📊 **p-VALUE NEDİR?**

### 📖 Tanım:

> **p-value**: H₀ doğruyken, gözlemlediğimiz veya daha aşırı sonuçları elde etme olasılığıdır.

### 🎯 Basit Dille:

**Düşük p-value** → "Bu kadar fark şans eseri olamaz!" → H₀ red  
**Yüksek p-value** → "Bu fark şans eseri olabilir" → H₀ kabul

---

### 📐 p-value Yorumlama:

| **p-value** | **Yorum** | **Karar** |
|:---:|:---|:---|
| **p < 0.01** | Çok güçlü kanıt | H₀ kesinlikle red |
| **0.01 ≤ p < 0.05** | Güçlü kanıt | H₀ red (standart) |
| **0.05 ≤ p < 0.10** | Zayıf kanıt | H₀ kabul (ama sınırda) |
| **p ≥ 0.10** | Kanıt yok | H₀ kabul |

---

### ⚠️ **Alpha (α) - Anlamlılık Düzeyi:**

**Tanım**: H₀'ı reddetmek için kabul ettiğimiz maksimum hata oranı.

**Standart**: α = 0.05 (%5)

**Karar Kuralı**:
```
p-value < α  →  H₀ RED (Anlamlı fark VAR)
p-value ≥ α  →  H₀ KABUL (Anlamlı fark YOK)
```

## ⚠️ **TYPE I ve TYPE II HATALAR**

### 📖 İki Tür Yanlış Karar:

Hipotez testinde **4 olası durum** var:

| **Gerçek Durum** | **Kararımız: H₀ Kabul** | **Kararımız: H₀ Red** |
|:---|:---:|:---:|
| **H₀ Doğru** | ✅ Doğru Karar | ❌ **Type I Error (α)** |
| **H₀ Yanlış (H₁ Doğru)** | ❌ **Type II Error (β)** | ✅ Doğru Karar |

---

### 🔴 **Type I Error (α - Alpha Error)**

> **"Yanlış Alarm!"** - Aslında fark yok ama "var" dedik.

**Tanım**: H₀ doğruyken yanlışlıkla H₀'ı reddetmek.

**Örnek**:
- 🏥 **Tıp**: Sağlıklı hastaya "hasta" teşhisi
- 🔬 **KampüsSepeti**: Etkisiz kampanyayı "etkili" sandık
- ⚖️ **Hukuk**: Suçsuz kişiyi "suçlu" bulduk

**Kontrol**: α değeri ile (genelde α=0.05)

**Risk**: Gereksiz maliyet, yanlış aksiyonlar

---

### 🔵 **Type II Error (β - Beta Error)**

> **"Fırsatı Kaçırma!"** - Aslında fark var ama "yok" dedik.

**Tanım**: H₀ yanlışken (H₁ doğru) yanlışlıkla H₀'ı kabul etmek.

**Örnek**:
- 🏥 **Tıp**: Hasta olan kişiye "sağlıklı" teşhisi
- 🔬 **KampüsSepeti**: Etkili kampanyayı "etkisiz" sandık
- ⚖️ **Hukuk**: Suçlu kişiyi "suçsuz" bulduk

**Kontrol**: Örneklem büyüklüğü (n↑ → β↓)

**Risk**: Fırsatı kaçırma, iyileşmeyi görememe

---

### 📊 **Power (Güç) = 1 - β**

**Tanım**: Gerçek bir farkı doğru şekilde tespit etme gücü.

- **Yüksek Power (0.80-0.90)**: İyi test, farkları yakalıyor
- **Düşük Power (<0.70)**: Zayıf test, farkları kaçırıyor

**Power Artırma Yolları**:
1. n'i artır (daha fazla veri)
2. α'yı artır (ama Type I riski artar!)
3. Etki büyüklüğü artarsa (doğal olarak artar)

### 📋 **FLASHCARD #19: HATA TÜRLERİ**

| **Özellik** | **Type I Error (α)** | **Type II Error (β)** |
|:---|:---:|:---:|
| **Tanım** | H₀ doğru ama red | H₀ yanlış ama kabul |
| **Alternatif adı** | False Positive | False Negative |
| **Metafor** | Yanlış alarm | Fırsatı kaçırma |
| **Olasılık** | α (genelde 0.05) | β (hedef <0.20) |
| **Kontrol** | α düşür | n artır |
| **Trade-off** | α↓ → β↑ | β↓ → α↑ veya n↑ |

---

### 💼 **KampüsSepeti Örnekleri:**

**Senaryo 1: Yeni Kurye Sistemi**

- **Type I Error**: Aslında etkisiz ama "etkili" dedik → Gereksiz yatırım
- **Type II Error**: Aslında etkili ama "etkisiz" dedik → İyileşmeyi kaçırdık

**Hangi hata daha kötü?**
- Maliyet düşükse: Type I tolere edilebilir (pilot dene)
- Maliyet yüksekse: Type I riskini minimize et (α=0.01 kullan)

---

**Senaryo 2: Kampanya Test**

- **Type I Error (α=0.05)**: %5 ihtimalle etkisiz kampanyayı "etkili" sanacağız
- **Type II Error**: Etkili kampanyayı kaçırma riski → n=200+ kullan

**YBS Kararı**: n=300, α=0.05 ile %80+ power sağla!

## 🆚 **KAMPÜSSEPETI A/B TESTİ: PYTHON UYGULAMASI**

### 🎯 Senaryo:

KampüsSepeti **iki farklı teslimat stratejisi** test ediyor:

- **Grup A (Kontrol)**: Mevcut sistem (tek kuryeli teslimat)
- **Grup B (Deney)**: Yeni sistem (akıllı rotalama + çift kurye)

**Soru**: Yeni sistem (B) teslimat süresini **anlamlı şekilde** azalttı mı?

---

### 📐 Hipotezler:

- **H₀**: μ_A = μ_B (İki sistem arasında fark YOK)
- **H₁**: μ_A > μ_B (Yeni sistem daha HIZLI - sol kuyruklu test)
- **α**: 0.05 (standart)

---

### 📊 Test: **İki Örneklem t-testi (Independent Samples t-test)**

Şimdi Python ile test edelim! 🚀

---

````python
# ============================================================================
# 🆚 KAMPÜSSEPETI A/B TESTİ: İKİ GRUP KARŞILAŞTIRMASI
# ============================================================================

print("🆚 KAMPÜSSEPETI A/B TESTİ: TESLİMAT SİSTEMLERİ")
print("="*70)

# Veri simülasyonu
np.random.seed(42)

# GRUP A: Mevcut sistem (kontrol)
n_A = 150
mu_A = 28.0  # Ortalama 28 dk
sigma_A = 6.0
grup_A = np.random.normal(mu_A, sigma_A, n_A)
grup_A = np.clip(grup_A, 10, 50)  # Gerçekçi sınırlar

# GRUP B: Yeni sistem (deney) - daha hızlı
n_B = 150
mu_B = 25.5  # 2.5 dk daha hızlı!
sigma_B = 5.5
grup_B = np.random.normal(mu_B, sigma_B, n_B)
grup_B = np.clip(grup_B, 10, 50)

# İstatistikler
x_bar_A = grup_A.mean()
x_bar_B = grup_B.mean()
s_A = grup_A.std(ddof=1)
s_B = grup_B.std(ddof=1)

print("📊 GRUP İSTATİSTİKLERİ:")
print("="*70)
print(f"\n🅰️  GRUP A (Mevcut Sistem):")
print(f"   n: {n_A}")
print(f"   x̄: {x_bar_A:.2f} dakika")
print(f"   s: {s_A:.2f} dakika")
print(f"   Min-Max: [{grup_A.min():.1f}, {grup_A.max():.1f}]")

print(f"\n🅱️  GRUP B (Yeni Sistem):")
print(f"   n: {n_B}")
print(f"   x̄: {x_bar_B:.2f} dakika")
print(f"   s: {s_B:.2f} dakika")
print(f"   Min-Max: [{grup_B.min():.1f}, {grup_B.max():.1f}]")

print(f"\n📏 FARK:")
print(f"   x̄_A - x̄_B = {x_bar_A - x_bar_B:.2f} dakika")
print(f"   Yüzde: {((x_bar_A - x_bar_B) / x_bar_A * 100):.1f}% iyileşme")

print("\n" + "="*70)

# -------------------------------------------------------------------------
# HİPOTEZ TESTİ: İKİ ÖRNEKLEM t-TESTİ
# -------------------------------------------------------------------------

print("\n🔬 HİPOTEZ TESTİ:")
print("="*70)
print("   H₀: μ_A = μ_B  (İki sistem arasında fark YOK)")
print("   H₁: μ_A > μ_B  (Yeni sistem daha HIZLI)")
print("   α:  0.05 (standart anlamlılık düzeyi)")
print("   Test: İki örneklem t-testi (independent samples)")
print("="*70)

# İki örneklem t-testi (tek kuyruklu)
# alternative='greater' : grup A > grup B olduğunu test et
t_statistic, p_value = stats.ttest_ind(grup_A, grup_B, alternative='greater')

print(f"\n📊 TEST SONUÇLARI:")
print(f"   t-istatistik: {t_statistic:.4f}")
print(f"   p-value: {p_value:.6f}")
print(f"   α (eşik): 0.05")

print("\n" + "="*70)

# Karar
if p_value < 0.05:
    karar = "✅ H₀ RED, H₁ KABUL"
    yorum = "Yeni sistem ANLAMLı şekilde daha hızlı!"
    aksiyon = "🚀 Yeni sistemi tüm filosuna uygula!"
    renk_kod = "BAŞARILI"
else:
    karar = "⚠️ H₀ KABUL"
    yorum = "İki sistem arasında anlamlı fark YOK"
    aksiyon = "🔄 Daha fazla veri topla veya sistem gözden geçir"
    renk_kod = "BAŞARISIZ"

print(f"💡 KARAR: {karar}")
print(f"   {yorum}")
print(f"\n📊 YORDaha fazla veri topla veya sistem gözden geçirUMLAMA:")
if p_value < 0.01:
    print(f"   Çok güçlü kanıt (p={p_value:.6f} << 0.05)")
elif p_value < 0.05:
    print(f"   Güçlü kanıt (p={p_value:.6f} < 0.05)")
elif p_value < 0.10:
    print(f"   Zayıf kanıt (p={p_value:.6f} sınırda)")
else:
    print(f"   Kanıt yok (p={p_value:.6f} > 0.05)")

print(f"\n💼 YBS AKSİYONU:")
print(f"   {aksiyon}")

# -------------------------------------------------------------------------
# ETKI BÜYÜKLÜĞÜ (Effect Size): Cohen's d
# -------------------------------------------------------------------------

# Cohen's d hesaplama
pooled_std = np.sqrt(((n_A - 1) * s_A**2 + (n_B - 1) * s_B**2) / (n_A + n_B - 2))
cohens_d = (x_bar_A - x_bar_B) / pooled_std

print("\n" + "="*70)
print("📐 ETKİ BÜYÜKLÜĞÜ (Cohen's d):")
print("="*70)
print(f"   Cohen's d: {cohens_d:.4f}")

if abs(cohens_d) < 0.2:
    etki = "Çok Küçük (negligible)"
elif abs(cohens_d) < 0.5:
    etki = "Küçük (small)"
elif abs(cohens_d) < 0.8:
    etki = "Orta (medium)"
else:
    etki = "Büyük (large)"

print(f"   Yorumu: {etki}")
print(f"\n💡 Anlamı:")
print(f"   İki grup ortalaması arasında {cohens_d:.2f} standart sapma fark var")

# -------------------------------------------------------------------------
# %95 GÜVEN ARALIKLARI
# -------------------------------------------------------------------------

# Her grup için %95 CI
ci_A = stats.t.interval(0.95, n_A-1, loc=x_bar_A, scale=s_A/np.sqrt(n_A))
ci_B = stats.t.interval(0.95, n_B-1, loc=x_bar_B, scale=s_B/np.sqrt(n_B))

print("\n" + "="*70)
print("📊 %95 GÜVEN ARALIKLARI:")
print("="*70)
print(f"   Grup A: [{ci_A[0]:.2f}, {ci_A[1]:.2f}] dakika")
print(f"   Grup B: [{ci_B[0]:.2f}, {ci_B[1]:.2f}] dakika")

# Çakışma kontrolü
cakisma = not (ci_A[1] < ci_B[0] or ci_B[1] < ci_A[0])
print(f"\n💡 CI Çakışması: {'✅ VAR (overlap)' if cakisma else '❌ YOK (no overlap)'}")

if not cakisma:
    print("   → Gruplar açıkça farklı!")
else:
    print("   → Çakışma olsa da istatistiksel test kesin karar verir")

print("="*70)

# -------------------------------------------------------------------------
# POWER ANALİZİ (POST-HOC)
# -------------------------------------------------------------------------

# Basitleştirilmiş power hesabı
from scipy.stats import norm

# Z-test yaklaşımı (n büyük)
se_diff = np.sqrt((s_A**2 / n_A) + (s_B**2 / n_B))
z_observed = (x_bar_A - x_bar_B) / se_diff
critical_z = 1.645  # Tek kuyruklu %95 için

# Power = P(Z > critical_z | gerçek fark var)
power = 1 - norm.cdf(critical_z - z_observed)

print(f"\n⚡ POWER ANALİZİ (POST-HOC):")
print(f"   Testin gücü (power): {power:.4f} ({power*100:.1f}%)")

if power >= 0.80:
    print(f"   ✅ Yüksek power! Gerçek farkları yakalıyor")
elif power >= 0.70:
    print(f"   ⚠️ Orta power. Kabul edilebilir")
else:
    print(f"   ❌ Düşük power. Daha fazla veri gerekli")

print("\n" + "="*70)
print("🎉 A/B TESTİ TAMAMLANDI!")
print("="*70)
````

````python
# ============================================================================
# 📊 A/B TESTİ GÖRSELLEŞTİRME
# ============================================================================

fig, axes = plt.subplots(2, 3, figsize=(18, 12))

# -------------------------------------------------------------------------
# PANEL 1: Histogram Karşılaştırma
# -------------------------------------------------------------------------

ax1 = axes[0, 0]

# Grup A
ax1.hist(grup_A, bins=30, alpha=0.6, color='#FFB74D', edgecolor='darkorange',
         linewidth=1.5, label=f'Grup A (x̄={x_bar_A:.1f})', density=True)

# Grup B
ax1.hist(grup_B, bins=30, alpha=0.6, color='#81C784', edgecolor='darkgreen',
         linewidth=1.5, label=f'Grup B (x̄={x_bar_B:.1f})', density=True)

# Ortalama çizgileri
ax1.axvline(x_bar_A, color='darkorange', linestyle='--', linewidth=2, alpha=0.8)
ax1.axvline(x_bar_B, color='darkgreen', linestyle='--', linewidth=2, alpha=0.8)

ax1.set_xlabel('Teslimat Süresi (dakika)', fontsize=11, fontweight='bold')
ax1.set_ylabel('Yoğunluk', fontsize=11, fontweight='bold')
ax1.set_title('A vs B  Histogram Karşılaştırma\nDağılım Görünümü',
              fontsize=12, fontweight='bold')
ax1.legend(fontsize=10, loc='upper right')
ax1.grid(True, alpha=0.3, linestyle=':')

# -------------------------------------------------------------------------
# PANEL 2: Box Plot Karşılaştırma
# -------------------------------------------------------------------------

ax2 = axes[0, 1]

bp = ax2.boxplot([grup_A, grup_B], labels=['Grup A\n(Mevcut)', 'Grup B\n(Yeni)'],
                 patch_artist=True, widths=0.6)

bp['boxes'][0].set_facecolor('#FFB74D')
bp['boxes'][1].set_facecolor('#81C784')

for patch in bp['boxes']:
    patch.set_alpha(0.7)
    patch.set_edgecolor('black')
    patch.set_linewidth(2)

ax2.set_ylabel('Teslimat Süresi (dakika)', fontsize=11, fontweight='bold')
ax2.set_title('Box Plot Karşılaştırma\nMedyan ve Varyans',
              fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y', linestyle=':')

# İstatistik kutuları
textstr_A = f'x̄={x_bar_A:.1f}\ns={s_A:.1f}'
textstr_B = f'x̄={x_bar_B:.1f}\ns={s_B:.1f}'

props_A = dict(boxstyle='round', facecolor='#FFE0B2', alpha=0.8)
props_B = dict(boxstyle='round', facecolor='#C8E6C9', alpha=0.8)

ax2.text(1, ax2.get_ylim()[1]*0.95, textstr_A, transform=ax2.transData,
         fontsize=9, verticalalignment='top', bbox=props_A)
ax2.text(2, ax2.get_ylim()[1]*0.95, textstr_B, transform=ax2.transData,
         fontsize=9, verticalalignment='top', bbox=props_B)

# -------------------------------------------------------------------------
# PANEL 3: CI Karşılaştırma
# -------------------------------------------------------------------------

ax3 = axes[0, 2]

y_pozisyonlar = [2, 1]
labels = ['Grup A\n(Mevcut)', 'Grup B\n(Yeni)']
cis = [ci_A, ci_B]
means = [x_bar_A, x_bar_B]
colors = ['#FFB74D', '#81C784']

for i, (y_pos, label, ci, mean, color) in enumerate(zip(y_pozisyonlar, labels, cis, means, colors)):
    # CI çizgisi
    ax3.plot([ci[0], ci[1]], [y_pos, y_pos], linewidth=10, color=color, alpha=0.7)
    
    # Uç noktalar
    ax3.plot([ci[0], ci[1]], [y_pos, y_pos], 'ko', markersize=8)
    
    # Merkez
    ax3.plot(mean, y_pos, 'r^', markersize=12)
    
    # Etiketler
    ax3.text(ci[0] - 0.3, y_pos, f'{ci[0]:.1f}', ha='right', va='center',
             fontsize=10, fontweight='bold')
    ax3.text(ci[1] + 0.3, y_pos, f'{ci[1]:.1f}', ha='left', va='center',
             fontsize=10, fontweight='bold')

ax3.set_yticks([])
ax3.set_xlabel('Teslimat Süresi (dakika)', fontsize=11, fontweight='bold')
ax3.set_title('%95 Güven Aralıkları\nÇakışma Kontrolü',
              fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='x', linestyle=':')

# Çakışma vurgusu
if not cakisma:
    ax3.text(0.5, 0.5, '❌ ÇAKIŞMA YOK\nGruplar Farklı!',
             transform=ax3.transAxes, fontsize=11, fontweight='bold',
             ha='center', va='center', color='green',
             bbox=dict(boxstyle='round,pad=0.5', facecolor='#C8E6C9', alpha=0.8))

# -------------------------------------------------------------------------
# PANEL 4: p-value Görselleştirme
# -------------------------------------------------------------------------

ax4 = axes[1, 0]

# Normal dağılım eğrisi (H₀ altında)
x_range = np.linspace(-4, 4, 1000)
y_normal = stats.t.pdf(x_range, df=n_A+n_B-2)

ax4.plot(x_range, y_normal, 'k-', linewidth=2, label='H₀ Dağılımı')

# Kritik değer (α=0.05, tek kuyruklu)
t_critical = stats.t.ppf(0.95, df=n_A+n_B-2)
ax4.axvline(t_critical, color='red', linestyle='--', linewidth=2,
            label=f't-kritik = {t_critical:.2f} (α=0.05)', alpha=0.7)

# Gözlenen t-istatistik
ax4.axvline(t_statistic, color='green', linestyle='-', linewidth=3,
            label=f't-gözlenen = {t_statistic:.2f}', alpha=0.7)

# Red bölgesi gölgeleme
x_reject = x_range[x_range >= t_critical]
y_reject = stats.t.pdf(x_reject, df=n_A+n_B-2)
ax4.fill_between(x_reject, 0, y_reject, alpha=0.3, color='red',
                 label=f'Red Bölgesi (α=0.05)')

# p-value bölgesi
if t_statistic > t_critical:
    x_pvalue = x_range[x_range >= t_statistic]
    y_pvalue = stats.t.pdf(x_pvalue, df=n_A+n_B-2)
    ax4.fill_between(x_pvalue, 0, y_pvalue, alpha=0.5, color='darkgreen',
                     label=f'p-value = {p_value:.4f}')

ax4.set_xlabel('t-istatistik', fontsize=11, fontweight='bold')
ax4.set_ylabel('Yoğunluk', fontsize=11, fontweight='bold')
ax4.set_title('🔬 Hipotez Testi: p-value Görselleştirme\nTek Kuyruklu Test (H₁: μ_A > μ_B)',
              fontsize=12, fontweight='bold')
ax4.legend(fontsize=8, loc='upper right')
ax4.grid(True, alpha=0.3, linestyle=':')
ax4.set_xlim(-1, 5)

# -------------------------------------------------------------------------
# PANEL 5: Etki Büyüklüğü
# -------------------------------------------------------------------------

ax5 = axes[1, 1]

# Cohen's d kategorileri
categories = ['Çok Küçük\n(<0.2)', 'Küçük\n(0.2-0.5)', 
              'Orta\n(0.5-0.8)', 'Büyük\n(>0.8)']
boundaries = [0, 0.2, 0.5, 0.8, 2.0]
colors_d = ['#FFCCBC', '#FFE0B2', '#FFF9C4', '#C8E6C9']

# Bar plot
for i, (cat, color) in enumerate(zip(categories, colors_d)):
    ax5.barh(i, boundaries[i+1] - boundaries[i], left=boundaries[i],
             color=color, alpha=0.7, edgecolor='black', linewidth=1.5)

# Gözlenen Cohen's d
ax5.plot(cohens_d, 2, 'r*', markersize=30, label=f"Cohen's d = {cohens_d:.3f}")

ax5.set_xlabel("Cohen's d", fontsize=11, fontweight='bold')
ax5.set_yticks(range(4))
ax5.set_yticklabels(categories, fontsize=9)
ax5.set_title('Etki Büyüklüğü (Effect Size)\nCohen\'s d Kategorileri',
              fontsize=12, fontweight='bold')
ax5.legend(fontsize=10, loc='upper right')
ax5.grid(True, alpha=0.3, axis='x', linestyle=':')
ax5.set_xlim(0, 1.5)

# -------------------------------------------------------------------------
# PANEL 6: Özet Tablo
# -------------------------------------------------------------------------

ax6 = axes[1, 2]
ax6.axis('off')

# Karar özeti tablosu
tablo_data = [
    ['Grup', 'Grup A (Mevcut)', 'Grup B (Yeni)'],
    ['n', f'{n_A}', f'{n_B}'],
    ['Ortalama (x̄)', f'{x_bar_A:.2f} dk', f'{x_bar_B:.2f} dk'],
    ['Std Sapma (s)', f'{s_A:.2f} dk', f'{s_B:.2f} dk'],
    ['%95 CI', f'[{ci_A[0]:.1f}, {ci_A[1]:.1f}]', f'[{ci_B[0]:.1f}, {ci_B[1]:.1f}]'],
    ['', '', ''],
    ['Test Sonuçları', '', ''],
    ['t-istatistik', f'{t_statistic:.4f}', ''],
    ['p-value', f'{p_value:.6f}', ''],
    ['Karar (α=0.05)', karar, ''],
    ['Cohen\'s d', f'{cohens_d:.3f} ({etki})', ''],
    ['Power', f'{power:.3f} ({power*100:.0f}%)', '']
]

tablo = ax6.table(cellText=tablo_data, cellLoc='center', loc='center',
                  colWidths=[0.35, 0.30, 0.30])

tablo.auto_set_font_size(False)
tablo.set_fontsize(8)
tablo.scale(1, 2.0)

# Renklendirme
for i in range(len(tablo_data)):
    for j in range(3):
        cell = tablo[(i, j)]
        
        if i == 0:  # Başlık
            cell.set_facecolor('#fa709a')
            cell.set_text_props(weight='bold', color='white')
        elif i == 6:  # Alt başlık
            cell.set_facecolor('#fee140')
            cell.set_text_props(weight='bold')
        elif i == 9:  # Karar satırı
            if "RED" in karar:
                cell.set_facecolor('#C8E6C9')
            else:
                cell.set_facecolor('#FFCCBC')
            cell.set_text_props(weight='bold')
        else:
            cell.set_facecolor('white')

ax6.set_title('📊 A/B Testi Özet Raporu\nKampüsSepeti Teslimat Sistemi',
              fontsize=12, fontweight='bold', pad=20)

plt.suptitle('KAMPÜSSEPETI A/B TESTİ: İKİ GRUP KARŞILAŞTIRMASI\n'
             f'H₀ {"RED" if p_value < 0.05 else "KABUL"}: Yeni Sistem '
             f'{"Anlamlı Şekilde Daha Hızlı! ✅" if p_value < 0.05 else "Anlamlı Fark Yok ⚠️"}',
             fontsize=14, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("💼 FİNAL YBS RAPORU:")
print("="*70)
print(f"\n📊 TEST ÖZETİ:")
print(f"   • Grup A (Mevcut): {x_bar_A:.1f} ± {s_A:.1f} dk (n={n_A})")
print(f"   • Grup B (Yeni): {x_bar_B:.1f} ± {s_B:.1f} dk (n={n_B})")
print(f"   • Fark: {x_bar_A - x_bar_B:.2f} dk ({((x_bar_A - x_bar_B) / x_bar_A * 100):.1f}% iyileşme)")

print(f"\n🔬 İSTATİSTİKSEL SONUÇ:")
print(f"   • p-value: {p_value:.6f}")
print(f"   • Karar: {karar}")
print(f"   • Effect Size: {cohens_d:.3f} ({etki})")
print(f"   • Power: {power:.1%}")

print(f"\n💼 İŞ KARARI:")
if p_value < 0.05:
    print(f"   ✅ Yeni sistem istatistiksel olarak anlamlı şekilde daha hızlı!")
    print(f"   ✅ Ortalama {x_bar_A - x_bar_B:.1f} dakika kazanç")
    print(f"   ✅ Günde 1000 sipariş için: ~{(x_bar_A - x_bar_B) * 1000 / 60:.0f} saat tasarruf!")
    print(f"   🚀 ÖNERİ: Yeni sistemi derhal tüm filosuna uygula!")
    print(f"   💰 ROI Hesaplama: Kurye maliyeti × zaman tasarrufu")
else:
    print(f"   ⚠️ İstatistiksel olarak anlamlı fark bulunamadı")
    print(f"   🔄 ÖNERİ: Daha fazla veri topla veya sistemi gözden geçir")
    print(f"   📊 Alternatif: Pilot uygulamayı n=300+ ile tekrarla")

print("="*70)
````
