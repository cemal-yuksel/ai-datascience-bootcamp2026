````python
# ============================================================
#  📈 3.2 — ÖZELLİK DAĞILIMLARI: HİSTOGRAM + KDE
#  Her sayısal değişkenin dağılımını görselleştiriyoruz.
#  KDE (Kernel Density Estimation) = sürekli dağılım tahmini
# ============================================================

# Sayısal özellikler (Id sütunu analiz dışı, Species hedef değişken)
features = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
feature_names_tr = {
    'SepalLengthCm': 'Çanak Uzunluğu (cm)',
    'SepalWidthCm' : 'Çanak Genişliği (cm)',
    'PetalLengthCm': 'Taç Uzunluğu (cm)',
    'PetalWidthCm' : 'Taç Genişliği (cm)'
}

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.patch.set_facecolor('#0f0025')
axes = axes.flatten()

# Türlere göre renk eşlemesi
species_colors = {
    'Iris-setosa'    : '#e879f9',
    'Iris-versicolor': '#a78bfa',
    'Iris-virginica' : '#818cf8'
}

for idx, feature in enumerate(features):
    ax = axes[idx]
    ax.set_facecolor('#1a0033')

    # Her tür için ayrı KDE histogram
    for species, color in species_colors.items():
        subset = df[df['Species'] == species][feature]
        ax.hist(
            subset, bins=15, alpha=0.4,
            color=color, edgecolor='white', linewidth=0.5,
            label=species
        )
        # KDE eğrisi
        from scipy import stats
        kde = stats.gaussian_kde(subset)
        x_range = np.linspace(subset.min() - 0.5, subset.max() + 0.5, 200)
        ax.plot(x_range, kde(x_range) * len(subset) * (subset.max() - subset.min()) / 15,
                color=color, linewidth=2.5, linestyle='-')

    ax.set_title(f'📊 {feature_names_tr[feature]}', color='#e879f9', fontweight='bold', fontsize=11, pad=10)
    ax.set_xlabel('Değer (cm)', color='#c4b5fd', fontsize=9)
    ax.set_ylabel('Frekans', color='#c4b5fd', fontsize=9)
    ax.tick_params(colors='#a78bfa', labelsize=8)
    ax.legend(fontsize=8, framealpha=0.2, labelcolor='white')
    for spine in ax.spines.values():
        spine.set_edgecolor('#4c1d95')

plt.suptitle('🌸 Iris Özellikleri — Dağılım Analizi (Histogram + KDE)',
             color='#c084fc', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

print("📌 Önemli Gözlem: Petal özellikleri (taç yaprak) sınıfları çok daha")
print("   iyi ayırt ediyor — özellikle Setosa diğerlerinden belirgin şekilde ayrışıyor!")
````

````python
# ============================================================
#  📦 3.3 — BOXPLOT: AYKIRI DEĞER ANALİZİ
#  Kutu grafik ile her özelliğin türlere göre dağılımını
#  ve potansiyel aykırı değerleri tespit ediyoruz.
#  IQR yöntemi: Q1-1.5*IQR altı ve Q3+1.5*IQR üstü = aykırı
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.patch.set_facecolor('#0f0025')
axes = axes.flatten()

for idx, feature in enumerate(features):
    ax = axes[idx]
    ax.set_facecolor('#1a0033')

    # Türlere göre renkli boxplot
    df_melted = df[['Species', feature]].copy()

    bp = ax.boxplot(
        [df[df['Species'] == sp][feature].values for sp in df['Species'].unique()],
        labels=df['Species'].unique(),
        patch_artist=True,   # Kutuları renklendir
        notch=False,
        widths=0.4,
        medianprops=dict(color='white', linewidth=2.5),
        whiskerprops=dict(color='#c4b5fd', linewidth=1.5),
        capprops=dict(color='#a78bfa', linewidth=2),
        flierprops=dict(marker='o', color='#f472b6', markersize=7, alpha=0.7)
    )

    # Her kutu için renk ata
    for patch, color in zip(bp['boxes'], PASTEL_COLORS):
        patch.set_facecolor(color)
        patch.set_alpha(0.6)
        patch.set_edgecolor('white')
        patch.set_linewidth(1.5)

    ax.set_title(f'📦 {feature_names_tr[feature]}', color='#e879f9', fontweight='bold', fontsize=11, pad=10)
    ax.set_xlabel('Iris Türü', color='#c4b5fd', fontsize=9)
    ax.set_ylabel('Değer (cm)', color='#c4b5fd', fontsize=9)
    ax.tick_params(colors='#a78bfa', labelsize=7)
    for spine in ax.spines.values():
        spine.set_edgecolor('#4c1d95')

plt.suptitle('📦 Boxplot — Sınıf Bazlı Dağılım & Aykırı Değer Analizi',
             color='#c084fc', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("📌 Boxplot Okuma Rehberi:")
print("   • Kutunun ortasındaki çizgi = Medyan (Q2)")
print("   • Kutunun alt kenarı = Q1 (1. çeyreklik)")
print("   • Kutunun üst kenarı = Q3 (3. çeyreklik)")
print("   • Bıyıklar = Q1-1.5*IQR ve Q3+1.5*IQR arası")
print("   • Noktalar = Potansiyel aykırı değerler")
````

      Diagonal = dağılım, Off-diagonal = scatter plot
      <br>📌 <em>Sınıf ayrımını keşfetmek için ideal!</em>
    </p>
  </div>

  <span style="color:#c4b5fd; font-size:0.88em;">
    Korelasyon analizi, bir bankanın <em>"Müşteri geliri ile kredi limiti arasında ilişki var mı?"</em>
    sorusunu cevaplamasına benzer. Yüksek korelasyon → multicollinearity problemi → model hatası!
  </span>

````python
# ============================================================
#  🌡️ 4.1 — KORELASYON MATRİSİ (ISIL HARİTA)
#  Sayısal değişkenler arasındaki doğrusal ilişkiyi ölçüyoruz.
#  Yüksek korelasyon: bir değişkeni bilince diğerini tahmin edebiliriz.
# ============================================================

# Id sütununu çıkar - anlamsız sıra numarası
df_numeric = df.drop(columns=['Id', 'Species'])

# Pearson korelasyon matrisi hesapla
corr_matrix = df_numeric.corr(method='pearson')

fig, ax = plt.subplots(figsize=(9, 7))
fig.patch.set_facecolor('#0f0025')
ax.set_facecolor('#1a0033')

# Mor tonlu özel korelasyon renk haritası
import matplotlib.colors as mcolors
cmap_custom = plt.cm.get_cmap('RdPu')   # Kırmızı-Mor renk geçişi

mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)  # Üst üçgeni gizle

im = sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='RdPu',
    center=0,
    vmin=-1, vmax=1,
    linewidths=2,
    linecolor='#1a0033',
    annot_kws={'size': 13, 'weight': 'bold', 'color': '#ffffff'},
    cbar_kws={'shrink': 0.8},
    ax=ax
)

ax.set_title('🌡️ Pearson Korelasyon Matrisi\n(Sayısal Özellikler)',
             color='#e879f9', fontweight='bold', fontsize=13, pad=15)
ax.tick_params(colors='#c4b5fd', labelsize=10)
plt.setp(ax.get_xticklabels(), rotation=30, ha='right')
plt.tight_layout()
plt.show()

print("📌 Korelasyon Yorumu:")
print(f"   📌 PetalLength ↔ PetalWidth : {corr_matrix.loc['PetalLengthCm','PetalWidthCm']:.3f} → Çok Yüksek Korelasyon!")
print(f"   📌 SepalLength ↔ PetalLength: {corr_matrix.loc['SepalLengthCm','PetalLengthCm']:.3f} → Yüksek Korelasyon")
print(f"   📌 SepalWidth ↔ PetalLength : {corr_matrix.loc['SepalWidthCm','PetalLengthCm']:.3f} → Negatif Korelasyon")
print("\n⚠️  Not: Naive Bayes, değişkenler BAĞIMSIZ olduğunu varsayar.")
print("    Bu yüzden 'Naive' (saf/basit) olarak adlandırılır!")
````

    StandardScaler, her ölçütü eşit önemde değerlendirmemizi sağlar!

````python
# ============================================================
#  ⚙️ 5.1 — VERİ ÖN İŞLEME ADIMLARI
# ============================================================

print("=" * 65)
print("⚙️  VERİ ÖN İŞLEME BAŞLIYOR...")
print("=" * 65)

# -----------------------------------------------------------
#  ADIM 1: Id sütununu kaldır (anlamsız sıra numarası)
# -----------------------------------------------------------
df_clean = df.drop(columns=['Id'])
print(f"\n✅ ADIM 1: 'Id' sütunu kaldırıldı.")
print(f"   Yeni shape: {df_clean.shape}")

# -----------------------------------------------------------
#  ADIM 2: Bağımsız değişkenler (X) ve hedef (y) ayır
# -----------------------------------------------------------
X = df_clean.drop(columns=['Species'])     # Özellikler (4 sütun)
y = df_clean['Species']                    # Hedef değişken

print(f"\n✅ ADIM 2: X (özellikler) ve y (hedef) ayrıldı.")
print(f"   X shape: {X.shape}  →  {list(X.columns)}")
print(f"   y shape: {y.shape}  →  Sınıflar: {y.unique()}")

# -----------------------------------------------------------
#  ADIM 3: Label Encoding — Metinsel sınıf etiketini sayıya çevir
#  Iris-setosa=0, Iris-versicolor=1, Iris-virginica=2
# -----------------------------------------------------------
le = LabelEncoder()
y_encoded = le.fit_transform(y)

print(f"\n✅ ADIM 3: Label Encoding tamamlandı.")
print(f"   Etiket Haritası:")
for i, cls in enumerate(le.classes_):
    print(f"   {cls:25s} → {i}")
print(f"   y_encoded (ilk 10): {y_encoded[:10]}")

# -----------------------------------------------------------
#  ADIM 4: Train/Test Split — %80 eğitim, %20 test
#  random_state=42 → Tekrarlanabilirlik (reproducibility)
#  stratify=y → Sınıf dengesini koruyarak böl
# -----------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded,
    test_size=0.2,       # %20 test seti
    random_state=42,     # Tekrarlanabilirlik için sabit tohum
    stratify=y_encoded   # Sınıf oranını koru
)

print(f"\n✅ ADIM 4: Train/Test Split tamamlandı.")
print(f"   Eğitim seti : {X_train.shape[0]} örnek (%{X_train.shape[0]/len(X)*100:.0f})")
print(f"   Test seti   : {X_test.shape[0]} örnek  (%{X_test.shape[0]/len(X)*100:.0f})")

# -----------------------------------------------------------
#  ADIM 5: StandardScaler ile özellik ölçeklendirme
#  z = (x - ortalama) / standart_sapma
#  ÖNEMLI: Scaler sadece eğitim setine fit etmeliyiz!
#  Test seti sızıntısını (data leakage) önlemek için!
# -----------------------------------------------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)    # Eğitim: fit + transform
X_test_scaled  = scaler.transform(X_test)         # Test: SADECE transform!

print(f"\n✅ ADIM 5: StandardScaler uygulandı.")
print(f"   Eğitim seti özellik ortalamaları (ölçekleme sonrası):")
means_after = X_train_scaled.mean(axis=0)
stds_after  = X_train_scaled.std(axis=0)
for i, feat in enumerate(X.columns):
    print(f"   {feat:20s}: μ={means_after[i]:.6f}  σ={stds_after[i]:.6f}")

print("\n" + "=" * 65)
print("🎉 Veri ön işleme TAMAMLANDI! Modeller için hazır.")
print("=" * 65)
print(f"\n📊 Final Özet:")
print(f"   X_train shape: {X_train_scaled.shape}")
print(f"   X_test shape : {X_test_scaled.shape}")
print(f"   y_train dağılımı: {dict(zip(*np.unique(y_train, return_counts=True)))}")
print(f"   y_test  dağılımı: {dict(zip(*np.unique(y_test,  return_counts=True)))}")
````

````python
# ============================================================
#  📏 6.1 — BASELINE MODEL: DummyClassifier
#  3 farklı strateji ile baz çizgisi belirliyoruz.
# ============================================================

import time

print("=" * 65)
print("📏  BASELINE MODEL — DummyClassifier ile Baz Çizgisi")
print("=" * 65)

# 3 farklı DummyClassifier stratejisi test et
strategies = {
    'stratified'   : 'Sınıf dağılımına göre rastgele tahmin',
    'most_frequent': 'En sık sınıfı hep tahmin et',
    'uniform'      : 'Tüm sınıflara eşit olasılıkla tahmin'
}

baseline_results = {}

for strategy, desc in strategies.items():
    # Model oluştur ve eğit
    dummy = DummyClassifier(strategy=strategy, random_state=42)
    dummy.fit(X_train_scaled, y_train)
    y_pred_dummy = dummy.predict(X_test_scaled)

    # Metrikleri hesapla
    acc = accuracy_score(y_test, y_pred_dummy)
    f1  = f1_score(y_test, y_pred_dummy, average='weighted', zero_division=0)
    prec = precision_score(y_test, y_pred_dummy, average='weighted', zero_division=0)
    rec  = recall_score(y_test, y_pred_dummy, average='weighted', zero_division=0)

    baseline_results[strategy] = {
        'Strateji': strategy,
        'Açıklama': desc,
        'Accuracy': acc,
        'F1-Score': f1,
        'Precision': prec,
        'Recall': rec
    }

    print(f"\n  Strateji: {strategy:15s} | {desc}")
    print(f"  Accuracy  : {acc:.4f} ({acc*100:.2f}%)")
    print(f"  F1-Score  : {f1:.4f}")
    print(f"  Precision : {prec:.4f}")
    print(f"  Recall    : {rec:.4f}")

# Baseline referans değeri olarak 'stratified' kullan
baseline_acc = baseline_results['most_frequent']['Accuracy']
print(f"\n{'='*65}")
print(f"🎯 Baseline Referans Değeri (most_frequent): {baseline_acc:.4f} ({baseline_acc*100:.2f}%)")
print(f"   ➡️ Modellerimiz bu değerden MUTLAKA iyi olmalı!")
print(f"{'='*65}")

# ============================================================
#  📊 Baseline Konfüzyon Matrisi (most_frequent)
# ============================================================
dummy_best = DummyClassifier(strategy='most_frequent', random_state=42)
dummy_best.fit(X_train_scaled, y_train)
y_pred_dummy_best = dummy_best.predict(X_test_scaled)

fig, ax = plt.subplots(figsize=(7, 5))
fig.patch.set_facecolor('#0f0025')
ax.set_facecolor('#1a0033')

cm_dummy = confusion_matrix(y_test, y_pred_dummy_best)
sns.heatmap(
    cm_dummy, annot=True, fmt='d',
    cmap='YlOrRd',
    xticklabels=le.classes_, yticklabels=le.classes_,
    linewidths=2, linecolor='#1a0033',
    annot_kws={'size': 14, 'weight': 'bold'},
    ax=ax
)
ax.set_title('📏 Baseline (most_frequent) — Konfüzyon Matrisi',
             color='#fbbf24', fontweight='bold', fontsize=12, pad=15)
ax.set_xlabel('Tahmin Edilen', color='#fde68a', fontsize=10)
ax.set_ylabel('Gerçek', color='#fde68a', fontsize=10)
ax.tick_params(colors='#fcd34d', labelsize=8)
plt.tight_layout()
plt.show()

print("💡 Not: Baseline model sadece tek bir sınıfı tahmin ediyor —")
print("   Bu yüzden Versicolor ve Virginica tahminleri tamamen boş!")
````

````python
# ============================================================
#  ⚡ 8.1 — 12 SINIFLANDIRMA ALGORİTMASI — EĞİTİM & DEĞERLENDİRME
#  Döngü içinde tüm modelleri eğitip metriklerini topluyoruz.
# ============================================================

print("=" * 70)
print("⚡  12 ALGORİTMA KARŞILAŞTIRMASI BAŞLIYOR...")
print("=" * 70)

# -----------------------------------------------------------
#  Tüm model tanımlamaları — isim, nesne, açıklama (Türkçe)
# -----------------------------------------------------------
models = {
    "Logistic Regression" : LogisticRegression(max_iter=1000, random_state=42),
    "K-NN (K=5)"          : KNeighborsClassifier(n_neighbors=5),
    "Decision Tree"       : DecisionTreeClassifier(random_state=42),
    "Random Forest"       : RandomForestClassifier(n_estimators=100, random_state=42),
    "Gradient Boosting"   : GradientBoostingClassifier(n_estimators=100, random_state=42),
    "SVM (RBF)"           : SVC(kernel='rbf', probability=True, random_state=42),
    "LDA"                 : LinearDiscriminantAnalysis(),
    "AdaBoost"            : AdaBoostClassifier(n_estimators=100, random_state=42),
    "Extra Trees"         : ExtraTreesClassifier(n_estimators=100, random_state=42),
    "Ridge Classifier"    : RidgeClassifier(),
    "Bagging Classifier"  : BaggingClassifier(n_estimators=50, random_state=42),
    "MLP Neural Network"  : MLPClassifier(hidden_layer_sizes=(100,50), max_iter=1000, random_state=42)
}

# -----------------------------------------------------------
#  Sonuçları depolamak için liste
# -----------------------------------------------------------
results_list = []

print(f"\n{'Algoritma':<22} {'Accuracy':>10} {'F1-Score':>10} {'Precision':>10} {'Recall':>10} {'Süre(ms)':>10}")
print("-" * 76)

for model_name, model in models.items():
    start_time = time.time()             # Eğitim süresi ölçümü başlat

    # Modeli eğit (only train set)
    model.fit(X_train_scaled, y_train)

    # Test seti üzerinde tahmin yap
    y_pred = model.predict(X_test_scaled)

    end_time = time.time()               # Süre ölçümü bitir
    elapsed_ms = (end_time - start_time) * 1000   # Milisaniyeye çevir

    # Performans metrikleri hesapla
    acc   = accuracy_score(y_test, y_pred)
    f1    = f1_score(y_test, y_pred, average='weighted')
    prec  = precision_score(y_test, y_pred, average='weighted', zero_division=0)
    rec   = recall_score(y_test, y_pred, average='weighted', zero_division=0)

    # Cross-Validation (5-katlı) — daha güvenilir performans tahmini
    cv_scores = cross_val_score(
        model, X_train_scaled, y_train,
        cv=5, scoring='accuracy', n_jobs=-1
    )
    cv_mean = cv_scores.mean()
    cv_std  = cv_scores.std()

    # Sonucu listeye ekle
    results_list.append({
        'Algoritma'      : model_name,
        'Accuracy'       : acc,
        'F1-Score'       : f1,
        'Precision'      : prec,
        'Recall'         : rec,
        'CV Accuracy'    : cv_mean,
        'CV Std'         : cv_std,
        'Süre (ms)'      : elapsed_ms
    })

    # Terminale yazdır
    print(f"{model_name:<22} {acc:>10.4f} {f1:>10.4f} {prec:>10.4f} {rec:>10.4f} {elapsed_ms:>10.1f}")

# Sonuçları DataFrame'e dönüştür
results_df = pd.DataFrame(results_list)

# Accuracy'ye göre sırala (büyükten küçüğe)
results_df = results_df.sort_values('Accuracy', ascending=False).reset_index(drop=True)
results_df.index = results_df.index + 1   # 1'den başlasın

print(f"\n{'='*70}")
print("📊 SIRALI SONUÇLAR TABLOSU:")
print(f"{'='*70}")
print(results_df[['Algoritma','Accuracy','F1-Score','CV Accuracy','CV Std','Süre (ms)']].to_string())
print(f"\n🏆 EN İYİ MODEL: {results_df.iloc[0]['Algoritma']}")
print(f"   Accuracy: {results_df.iloc[0]['Accuracy']:.4f}")
print(f"   F1-Score: {results_df.iloc[0]['F1-Score']:.4f}")
print(f"\n📌 Naive Bayes Sıralaması:")
nb_rank = results_df[results_df['Algoritma'] == 'Gaussian NB'].index
print(f"   Not: GaussianNB ayrı eğitilmiştir — Accuracy: {acc_gnb:.4f}")
````

````python
# ============================================================
#  📊 9.1 — 12 ALGORİTMA KARŞILAŞTIRMA GÖRSELLEŞTİRMESİ
#  Pastel renkli çubuk grafikler ile performans karşılaştırması
# ============================================================

# Naive Bayes sonucunu ekle ve tam listeyi hazırla
nb_row = pd.DataFrame([{
    'Algoritma'  : 'Gaussian NB',
    'Accuracy'   : acc_gnb,
    'F1-Score'   : f1_gnb,
    'Precision'  : prec_gnb,
    'Recall'     : rec_gnb,
    'CV Accuracy': cross_val_score(gnb, X_train_scaled, y_train, cv=5).mean(),
    'CV Std'     : cross_val_score(gnb, X_train_scaled, y_train, cv=5).std(),
    'Süre (ms)'  : 0.5  # Çok hızlı
}])

full_results = pd.concat([results_df, nb_row], ignore_index=True)
full_results = full_results.sort_values('Accuracy', ascending=False).reset_index(drop=True)
full_results.index += 1

# ============================================================
#  🎨 GÖRSEL 1: Accuracy Karşılaştırma Bar Chart
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(18, 14))
fig.patch.set_facecolor('#0f0025')

# Renk paleti (13 model için)
n_models = len(full_results)
colors_list = plt.cm.cool(np.linspace(0.2, 0.9, n_models))

# ---- Grafik 1: Accuracy Bar Chart ----
ax1 = axes[0, 0]
ax1.set_facecolor('#1a0033')

bars = ax1.barh(
    full_results['Algoritma'][::-1],
    full_results['Accuracy'][::-1],
    color=colors_list[::-1],
    edgecolor='white', linewidth=0.8,
    height=0.65
)

# Baseline çizgisi ekle
ax1.axvline(x=baseline_acc, color='#fbbf24', linestyle='--',
            linewidth=2, alpha=0.8, label=f'Baseline ({baseline_acc:.2f})')

# Değerleri çubuk üzerine ekle
for bar, val in zip(bars, full_results['Accuracy'][::-1]):
    ax1.text(val + 0.002, bar.get_y() + bar.get_height()/2,
             f'{val:.3f}', va='center', ha='left',
             color='white', fontweight='bold', fontsize=9)

ax1.set_title('🎯 Accuracy Karşılaştırması', color='#22d3ee', fontweight='bold', fontsize=11, pad=10)
ax1.set_xlabel('Accuracy', color='#a78bfa', fontsize=9)
ax1.tick_params(colors='#c4b5fd', labelsize=8)
ax1.legend(fontsize=8, framealpha=0.3, labelcolor='white')
ax1.set_xlim([0, 1.08])
for spine in ax1.spines.values():
    spine.set_edgecolor('#4c1d95')

# ---- Grafik 2: F1-Score Bar Chart ----
ax2 = axes[0, 1]
ax2.set_facecolor('#1a0033')

f1_sorted = full_results.sort_values('F1-Score', ascending=True)
colors_f1 = plt.cm.PuRd(np.linspace(0.3, 0.9, n_models))

ax2.barh(
    f1_sorted['Algoritma'],
    f1_sorted['F1-Score'],
    color=colors_f1,
    edgecolor='white', linewidth=0.8, height=0.65
)
for i, (_, row) in enumerate(f1_sorted.iterrows()):
    ax2.text(row['F1-Score'] + 0.002, i, f"{row['F1-Score']:.3f}",
             va='center', color='white', fontweight='bold', fontsize=9)

ax2.set_title('🏆 F1-Score Karşılaştırması', color='#e879f9', fontweight='bold', fontsize=11, pad=10)
ax2.set_xlabel('F1-Score (Weighted)', color='#a78bfa', fontsize=9)
ax2.tick_params(colors='#c4b5fd', labelsize=8)
ax2.set_xlim([0, 1.08])
for spine in ax2.spines.values():
    spine.set_edgecolor('#4c1d95')

# ---- Grafik 3: CV Karşılaştırma (Error Bar) ----
ax3 = axes[1, 0]
ax3.set_facecolor('#1a0033')

colors_cv = plt.cm.viridis(np.linspace(0.3, 0.9, n_models))
full_results_sorted_cv = full_results.sort_values('CV Accuracy', ascending=True)

ax3.barh(
    full_results_sorted_cv['Algoritma'],
    full_results_sorted_cv['CV Accuracy'],
    xerr=full_results_sorted_cv['CV Std'],
    color=colors_cv,
    edgecolor='white', linewidth=0.8, height=0.65,
    error_kw={'ecolor':'white', 'capsize':4, 'linewidth':1.5}
)
for i, (_, row) in enumerate(full_results_sorted_cv.iterrows()):
    ax3.text(row['CV Accuracy'] + 0.008, i,
             f"{row['CV Accuracy']:.3f}±{row['CV Std']:.3f}",
             va='center', color='white', fontsize=7.5)

ax3.set_title('📊 5-Fold CV Accuracy (± Std)', color='#34d399', fontweight='bold', fontsize=11, pad=10)
ax3.set_xlabel('Cross-Validation Accuracy', color='#a78bfa', fontsize=9)
ax3.tick_params(colors='#c4b5fd', labelsize=8)
ax3.set_xlim([0, 1.15])
for spine in ax3.spines.values():
    spine.set_edgecolor('#4c1d95')

# ---- Grafik 4: Hız vs Accuracy Scatter ----
ax4 = axes[1, 1]
ax4.set_facecolor('#1a0033')

speed_data = full_results[full_results['Süre (ms)'] > 0]  # 0 olmayan süreleri al
sc = ax4.scatter(
    speed_data['Süre (ms)'],
    speed_data['Accuracy'],
    c=speed_data['Accuracy'],
    cmap='cool',
    s=200,
    edgecolor='white', linewidth=1.5,
    alpha=0.9,
    zorder=5
)
for _, row in speed_data.iterrows():
    ax4.annotate(
        row['Algoritma'],
        (row['Süre (ms)'], row['Accuracy']),
        xytext=(5, 5), textcoords='offset points',
        color='#c4b5fd', fontsize=7
    )

cbar4 = plt.colorbar(sc, ax=ax4)
cbar4.ax.tick_params(colors='#c4b5fd')
cbar4.set_label('Accuracy', color='#c4b5fd', fontsize=9)

ax4.set_title('⚡ Hız vs Accuracy (Trade-off)', color='#fbbf24', fontweight='bold', fontsize=11, pad=10)
ax4.set_xlabel('Eğitim Süresi (ms)', color='#a78bfa', fontsize=9)
ax4.set_ylabel('Test Accuracy', color='#a78bfa', fontsize=9)
ax4.tick_params(colors='#c4b5fd')
for spine in ax4.spines.values():
    spine.set_edgecolor('#4c1d95')

plt.suptitle('⚡ 13 Sınıflandırma Modeli — Kapsamlı Performans Karşılaştırması',
             color='#e879f9', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

# ============================================================
#  📋 SON SIRALAMA TABLOSU
# ============================================================
print("\n" + "=" * 80)
print("🏆 FİNAL SIRALAMA TABLOSU (Accuracy'ye Göre)")
print("=" * 80)
display_df = full_results[['Algoritma','Accuracy','F1-Score','Precision','Recall','CV Accuracy','Süre (ms)']].copy()
display_df.columns = ['Algoritma', 'Test Acc.', 'F1-Score', 'Precision', 'Recall', 'CV Acc.', 'Süre(ms)']
print(display_df.round(4).to_string())

best_model_name = full_results.iloc[0]['Algoritma']
best_model_acc  = full_results.iloc[0]['Accuracy']
print(f"\n{'='*80}")
print(f"🥇 1. Sıra : {best_model_name} — Accuracy: {best_model_acc:.4f}")
print(f"🥈 2. Sıra : {full_results.iloc[1]['Algoritma']} — Accuracy: {full_results.iloc[1]['Accuracy']:.4f}")
print(f"🥉 3. Sıra : {full_results.iloc[2]['Algoritma']} — Accuracy: {full_results.iloc[2]['Accuracy']:.4f}")
print(f"{'='*80}")
print(f"\n🔍 Gaussian NB: Accuracy = {acc_gnb:.4f} | Baseline: {baseline_acc:.4f}")
print(f"   NB Üstünlüğü Baseline'a Göre: +{(acc_gnb - baseline_acc)*100:.2f} puan 🚀")
````
