### 🤔 **Pandas'ın Tanımı**

**Pandas**, Python'da **tablo formatındaki verileri** (Excel, CSV, veritabanı) yönetmek için kullanılan en güçlü kütüphanedir.

#### 💡 **Excel vs Pandas Karşılaştırması**

| Özellik | Excel 📊 | Pandas 🐼 |
|---------|----------|----------|
| **Manuel İşlem** | Her seferinde tekrar edersiniz | Bir kez yazarsınız, sürekli kullanırsınız |
| **Hata Riski** | Kopyala-yapıştır hataları çok | Kod kontrol edilebilir, tekrar üretilebilir |
| **Büyük Veri** | Yavaşlar, kilitlenir | Milyonlarca satırı hızlı işler |
| **Otomasyon** | Makrolar karmaşık | Python ile kolay entegre |
| **Versiyon Kontrolü** | Dosya çoğaltma | Git ile versiyon yönetimi |

### 🧩 **NumPy ile İlişkisi**

<div style="background: linear-gradient(135deg, #f093fb 0%, #f5576c 100%); padding: 15px; border-radius: 10px; color: white; margin: 20px 0;">
    <strong>NumPy:</strong> Ham sayı motoru (matematiksel işlemler için hızlı dizi işlemleri)<br>
    <strong>Pandas:</strong> İş dünyası tablosu motoru (anlamlı sütunlar, satırlar, etiketler)

- **NumPy:** `[120, 150, 80, 200]` → Sadece sayılar

| Ürün | Adet |

| Tost | 150 |
| Çorba | 80 |
| Su | 200 |

Görüldüğü gibi Pandas, **anlamı korur**! Bu da karar vermeyi mümkün kılar.

### 🎯 **Pratik Örnek: Kampüs Veri Sistemi Simülasyonu**

Şimdi gerçek bir YBS senaryosu simüle edelim!

````python
# 📊 SENARYO: Kampüs Kütüphane Günlük Raporu
# Farklı veri kaynaklarından gelen bilgileri Pandas ile birleştiriyoruz

# Kütüphane giriş verileri
kutuphane_giris = pd.DataFrame({
    'Öğrenci_ID': [1001, 1002, 1003, 1004, 1005],
    'İsim': ['Ayşe Yılmaz', 'Mehmet Demir', 'Zeynep Kaya', 'Ahmet Çelik', 'Fatma Şahin'],
    'Giriş_Saati': ['09:00', '09:30', '10:15', '11:00', '13:45'],
    'Bölüm': ['YBS', 'Bilgisayar Müh.', 'YBS', 'Endüstri Müh.', 'YBS']
})

print("📚 KAMPÜS KÜTÜPHANESİ - GÜNLÜK GİRİŞ RAPORU")
print("="*60)
print(kutuphane_giris)
print("="*60)

# 📈 Hızlı istatistik
ybs_ogrenci = kutuphane_giris[kutuphane_giris['Bölüm'] == 'YBS']
print(f"\n✅ Toplam giriş: {len(kutuphane_giris)} öğrenci")
print(f"🎯 YBS öğrenci sayısı: {len(ybs_ogrenci)} kişi")
print(f"📊 YBS oranı: %{(len(ybs_ogrenci)/len(kutuphane_giris)*100):.1f}")
````

Bu tablonun **"Adet" sütunu** tek başına bir **Series**'tir:
- **Değerler:** `[120, 200, 45]`
- **İndeks (Etiketler):** `['Kahve', 'Çay', 'Tost']`

### 🎓 **Örnek 1: Sözlükten Series Oluşturma**

Kampüs senaryosu: Öğrenci notları sistemi

````python
# 📚 SENARYO: YBS Veri Analitiği Dersi Vize Notları
# Sözlükten Series oluşturma

vize_notlari = {
    'Ayşe Yılmaz': 85,
    'Mehmet Demir': 72,
    'Zeynep Kaya': 91,
    'Ahmet Çelik': 68,
    'Fatma Şahin': 88
}

# Series oluşturma - Anahtar = İndeks, Değer = Values
notlar_series = pd.Series(vize_notlari)

print("📊 YBS VERİ ANALİTİĞİ DERSİ - VİZE NOTLARI")
print("="*50)
print(notlar_series)
print("="*50)

# 🔍 İndeks ile erişim (İsimle çağırma)
print(f"\n✅ Zeynep'in notu: {notlar_series['Zeynep Kaya']}")

# 🔍 Sıra numarasıyla erişim
print(f"✅ İlk öğrencinin notu: {notlar_series[0]}")
print(f"✅ Son öğrencinin notu: {notlar_series[-1]}")

# 📈 Hızlı istatistik
print(f"\n📊 Ortalama: {notlar_series.mean():.2f}")
print(f"📊 En yüksek not: {notlar_series.max()}")
print(f"📊 En düşük not: {notlar_series.min()}")
````

### 🎓 **Örnek 2: Listeden Series Oluşturma**

Liste kullanırsanız indeks otomatik olarak 0, 1, 2... şeklinde oluşur.

````python
# 🏠 SENARYO: Kampüs Yurtlarındaki Günlük Arıza Sayıları
# Listeden Series oluşturma

haftalik_ariza = [3, 5, 2, 8, 4, 6, 1]

# İndeks vermeden Series oluşturma
ariza_series = pd.Series(haftalik_ariza)

print("🏠 KAMPÜS YURTLARI - HAFTALIK ARIZA SAYISI (İndekssiz)")
print("="*50)
print(ariza_series)
print("="*50)

# ❌ Problem: Hangi gün kaç arıza oldu? Belli değil!
print("\n⚠️ PROBLEM: İndeks 0,1,2... → Hangi gün olduğu belli değil!")

print("\n" + "="*50)

# ✅ Çözüm: Manuel indeks verme
gunler = ['Pazartesi', 'Salı', 'Çarşamba', 'Perşembe', 'Cuma', 'Cumartesi', 'Pazar']
ariza_series_etiketli = pd.Series(haftalik_ariza, index=gunler)

print("\n🏠 KAMPÜS YURTLARI - HAFTALIK ARIZA SAYISI (Etiketli)")
print("="*50)
print(ariza_series_etiketli)
print("="*50)

# ✅ Artık anlamlı erişim yapabiliriz
print(f"\n✅ Çarşamba günü arıza sayısı: {ariza_series_etiketli['Çarşamba']}")
print(f"✅ Hafta sonu toplam arıza: {ariza_series_etiketli['Cumartesi'] + ariza_series_etiketli['Pazar']}")

# 📊 En çok arıza olan gün
en_cok_gun = ariza_series_etiketli.idxmax()
print(f"\n📈 En çok arıza olan gün: {en_cok_gun} ({ariza_series_etiketli[en_cok_gun]} arıza)")
````

### 🎓 **Örnek 3: NumPy Array'den Series**

NumPy dizilerini de Series'e dönüştürebiliriz.

````python
# 🌡️ SENARYO: Kampüs Sıcaklık Sensörü Verileri
# NumPy array'den Series oluşturma

# NumPy ile rastgele sıcaklık değerleri üretiyoruz (15-30 derece arası)
np.random.seed(42)  # Sonuçlar her çalıştırmada aynı olsun
sicaklik_degerleri = np.random.randint(15, 31, size=7)

# Gün isimleriyle indeks oluşturma
gunler = ['Pazartesi', 'Salı', 'Çarşamba', 'Perşembe', 'Cuma', 'Cumartesi', 'Pazar']
sicaklik_series = pd.Series(sicaklik_degerleri, index=gunler, name='Sıcaklık (°C)')

print("🌡️ KAMPÜS SICAKLIK SENSÖRÜ - HAFTALIK VERİ")
print("="*50)
print(sicaklik_series)
print("="*50)

# 📊 İstatistiksel analiz
print(f"\n📊 Ortalama sıcaklık: {sicaklik_series.mean():.1f}°C")
print(f"📊 En sıcak gün: {sicaklik_series.idxmax()} ({sicaklik_series.max()}°C)")
print(f"📊 En soğuk gün: {sicaklik_series.idxmin()} ({sicaklik_series.min()}°C)")
print(f"📊 Standart sapma: {sicaklik_series.std():.2f}°C")

# 🔍 Koşullu filtreleme
sicak_gunler = sicaklik_series[sicaklik_series > 25]
print(f"\n🔥 25°C üstü günler:")
print(sicak_gunler)
````

### 🎯 **Series İşlemleri ve Metodları**

Series üzerinde yapabileceğiniz temel işlemler:

````python
# 🏪 SENARYO: Kafeterya Günlük Satış Adedi
# Series işlemlerini öğrenelim

gunluk_satis = pd.Series({
    'Pazartesi': 145,
    'Salı': 132,
    'Çarşamba': 168,
    'Perşembe': 155,
    'Cuma': 189,
    'Cumartesi': 98,
    'Pazar': 87
})

print("🏪 KAFETERYA GÜNLÜK SATIŞ (ADET)")
print("="*50)
print(gunluk_satis)
print("="*50)

# 📊 İstatistiksel Metodlar
print("\n📊 İSTATİSTİKLER:")
print(f"   • Toplam satış: {gunluk_satis.sum()} adet")
print(f"   • Ortalama: {gunluk_satis.mean():.1f} adet/gün")
print(f"   • Medyan: {gunluk_satis.median()} adet")
print(f"   • Standart sapma: {gunluk_satis.std():.2f}")
print(f"   • Min: {gunluk_satis.min()} adet ({gunluk_satis.idxmin()})")
print(f"   • Max: {gunluk_satis.max()} adet ({gunluk_satis.idxmax()})")

# 🔍 Koşullu Filtreleme
print("\n🔍 YÜKSEK SATIŞ GÜNLERİ (150+ adet):")
yuksek_satis = gunluk_satis[gunluk_satis >= 150]
print(yuksek_satis)

# 📈 Sıralama
print("\n📈 SATIŞLARA GÖRE SIRALAMA (Azalan):")
sirali = gunluk_satis.sort_values(ascending=False)
print(sirali)

# 🎯 İndekse göre sıralama
print("\n🎯 GÜN ADINA GÖRE SIRALAMA (Alfabetik):")
sirali_gun = gunluk_satis.sort_index()
print(sirali_gun)
````

### 🎓 **Örnek 1: Mükemmel Hizalama (Tüm İndeksler Aynı)**

````python
# 📚 SENARYO: Kampüs Hafta İçi Yoğunluk Analizi
# Her iki sistemde de aynı günler var

# Kütüphane günlük giriş sayıları
kutuphane = pd.Series({
    'Pazartesi': 120,
    'Salı': 145,
    'Çarşamba': 98,
    'Perşembe': 167,
    'Cuma': 189
}, name='Kütüphane')

# Etüt Salonu günlük giriş sayıları
etut_salonu = pd.Series({
    'Pazartesi': 80,
    'Salı': 95,
    'Çarşamba': 110,
    'Perşembe': 88,
    'Cuma': 75
}, name='Etüt Salonu')

print("📊 KAMPÜS YOĞUNLUK VERİLERİ")
print("="*60)
print("\n📚 Kütüphane:")
print(kutuphane)
print("\n🎓 Etüt Salonu:")
print(etut_salonu)
print("="*60)

# ➕ İKİ SERİES'İ TOPLAMA - Otomatik hizalama!
toplam_yogunluk = kutuphane + etut_salonu
toplam_yogunluk.name = 'Toplam Yoğunluk'

print("\n🎯 TOPLAM GÜNLÜK YOĞUNLUK:")
print(toplam_yogunluk)
print("="*60)

# 📊 Analiz
print(f"\n✅ En yoğun gün: {toplam_yogunluk.idxmax()} ({toplam_yogunluk.max()} kişi)")
print(f"✅ En sakin gün: {toplam_yogunluk.idxmin()} ({toplam_yogunluk.min()} kişi)")
print(f"✅ Hafta içi ortalama: {toplam_yogunluk.mean():.1f} kişi")

# 💡 YBS Yorumu
print("\n💡 YBS YORUMU:")
print("   Tüm günler her iki sistemde de kayıtlı → Mükemmel veri entegrasyonu!")
print("   Hiç NaN yok → Veri kalitesi yüksek!")
````

### 🎓 **Örnek 2: Kısmi Hizalama (Bazı İndeksler Farklı)**

Gerçek hayatta sistemler her zaman senkronize çalışmaz! Bazı günler bir sistemde kayıt var, diğerinde yok.

````python
# ⚠️ SENARYO: Gerçekçi Durum - Bazı Günler Kayıt Yok
# Kütüphane Cumartesi-Pazar açık, Etüt Salonu hafta içi açık

# Kütüphane: 7 gün açık
kutuphane_haftalik = pd.Series({
    'Pazartesi': 120,
    'Salı': 145,
    'Çarşamba': 98,
    'Perşembe': 167,
    'Cuma': 189,
    'Cumartesi': 85,
    'Pazar': 62
}, name='Kütüphane')

# Etüt Salonu: Sadece hafta içi açık (5 gün)
etut_haftalik = pd.Series({
    'Pazartesi': 80,
    'Salı': 95,
    'Çarşamba': 110,
    'Perşembe': 88,
    'Cuma': 75
}, name='Etüt Salonu')

print("📊 KAMPÜS HAFTALIK YOĞUNLUK (Farklı Çalışma Günleri)")
print("="*60)
print("\n📚 Kütüphane (7 gün):")
print(kutuphane_haftalik)
print("\n🎓 Etüt Salonu (5 gün):")
print(etut_haftalik)
print("="*60)

# ➕ TOPLAMA - Pandas otomatik hizalar!
toplam_haftalik = kutuphane_haftalik + etut_haftalik

print("\n🎯 TOPLAM YOĞUNLUK (Otomatik Hizalama):")
print(toplam_haftalik)
print("="*60)

# ⚠️ NaN ANALİZİ
print("\n⚠️ NaN (Eksik Veri) ANALİZİ:")
print(f"   • Toplam NaN sayısı: {toplam_haftalik.isna().sum()}")
print(f"   • NaN oranı: %{(toplam_haftalik.isna().sum() / len(toplam_haftalik) * 100):.1f}")

# NaN olan günleri göster
nan_gunler = toplam_haftalik[toplam_haftalik.isna()]
print(f"\n   📅 NaN olan günler:")
print(nan_gunler)

# 💡 YBS Yorumu
print("\n💡 YBS YORUMU:")
print("   ⚠️ Cumartesi-Pazar için NaN var!")
print("   ➡️ Sebep: Etüt salonu hafta sonu kapalı, veri yok.")
print("   ➡️ Bu NORMAL bir durum → Sistem hatası DEĞİL!")
print("\n   ✅ Pazartesi-Cuma için toplam var → Veri entegrasyonu başarılı!")
````

### 🛠️ **NaN ile Çalışma: Tespit ve Yönetim**

````python
# 🔧 NaN Tespit ve Yönetim Teknikleri

print("🔍 NaN TESPİT VE YÖNETİM")
print("="*60)

# 1️⃣ NaN tespiti (Boolean mask)
print("\n1️⃣ Hangi günler NaN?")
print(toplam_haftalik.isna())

# 2️⃣ NaN olmayan değerler
print("\n2️⃣ Hangi günler NaN DEĞİL?")
print(toplam_haftalik.notna())

# 3️⃣ NaN olan değerleri filtreleme
print("\n3️⃣ Sadece NaN değerler:")
nan_degerler = toplam_haftalik[toplam_haftalik.isna()]
print(nan_degerler)

# 4️⃣ NaN olmayan değerleri filtreleme
print("\n4️⃣ Sadece geçerli değerler:")
gecerli_degerler = toplam_haftalik[toplam_haftalik.notna()]
print(gecerli_degerler)

# 5️⃣ NaN'leri atma (dropna)
print("\n5️⃣ NaN'leri SİL (dropna):")
temiz_veri = toplam_haftalik.dropna()
print(temiz_veri)
print(f"   → {len(toplam_haftalik)} günden {len(temiz_veri)} gün kaldı")

# 6️⃣ NaN'leri doldurma (fillna) - Sıfır ile
print("\n6️⃣ NaN'leri 0 ile DOLDUR (fillna):")
sifir_doldurulmus = toplam_haftalik.fillna(0)
print(sifir_doldurulmus)

# 7️⃣ NaN'leri doldurma - Ortalama ile
print("\n7️⃣ NaN'leri ORTALAMA ile DOLDUR:")
ortalama = toplam_haftalik.mean()  # NaN'ler hariç ortalama
ortalama_doldurulmus = toplam_haftalik.fillna(ortalama)
print(ortalama_doldurulmus)
print(f"   → Ortalama kullanıldı: {ortalama:.1f}")

print("\n" + "="*60)
print("💡 YBS KARARI: Hangi yöntem doğru?")
print("   • dropna() → Eğer NaN normal ise (sistem kapalı)")
print("   • fillna(0) → Eğer 'veri yok = 0' demek istiyorsanız")
print("   • fillna(mean) → Eğer tahmin yapmak istiyorsanız")
print("   ⚠️ Yanlış seçim = Yanlış rapor!")
````

### 🎯 **Diğer Aritmetik İşlemler**

Series'ler sadece toplanmaz! Çıkarma, çarpma, bölme de yapabilirsiniz.

````python
# 🧮 SENARYO: Kafeterya Gelir Analizi
# Haftalık satış adedi ve birim fiyat karşılaştırması

hafta1_satis = pd.Series({
    'Kahve': 120,
    'Çay': 200,
    'Tost': 85,
    'Sandviç': 60
}, name='Hafta 1')

hafta2_satis = pd.Series({
    'Kahve': 135,
    'Çay': 180,
    'Tost': 95,
    'Sandviç': 72
}, name='Hafta 2')

birim_fiyat = pd.Series({
    'Kahve': 25,
    'Çay': 10,
    'Tost': 35,
    'Sandviç': 45
}, name='Birim Fiyat (₺)')

print("🏪 KAFETERYA ARİTMETİK İŞLEMLER")
print("="*60)

# ➕ TOPLAMA - Toplam haftalık satış
print("\n➕ HAFTA 1 + HAFTA 2 (Toplam Satış):")
toplam_satis = hafta1_satis + hafta2_satis
print(toplam_satis)

# ➖ ÇIKARMA - Hafta 2'de değişim
print("\n➖ HAFTA 2 - HAFTA 1 (Değişim):")
degisim = hafta2_satis - hafta1_satis
print(degisim)
print(f"\n   📈 En çok artan: {degisim.idxmax()} (+{degisim.max()} adet)")
print(f"   📉 En çok azalan: {degisim.idxmin()} ({degisim.min()} adet)")

# ✖️ ÇARPMA - Hafta 1 gelir hesabı
print("\n✖️ HAFTA 1 SATIŞ × FİYAT (Gelir):")
hafta1_gelir = hafta1_satis * birim_fiyat
print(hafta1_gelir)
print(f"\n   💰 Toplam gelir: {hafta1_gelir.sum():,.0f} ₺")

# ➗ BÖLME - Ortalama günlük satış (5 gün)
print("\n➗ HAFTA 1 SATIŞ ÷ 5 (Günlük Ortalama):")
gunluk_ortalama = hafta1_satis / 5
print(gunluk_ortalama)

# 📊 Yüzdesel değişim
print("\n📊 HAFTA 2 DEĞİŞİM YÜZDESI:")
yuzde_degisim = ((hafta2_satis - hafta1_satis) / hafta1_satis) * 100
print(yuzde_degisim.round(1))

print("\n" + "="*60)
print("💡 YBS İçgörü:")
print(f"   • En karlı ürün: {hafta1_gelir.idxmax()} ({hafta1_gelir.max():,.0f} ₺)")
print(f"   • Büyüme lideri: {yuzde_degisim.idxmax()} (%{yuzde_degisim.max():.1f})")
````

### 🎓 **Örnek 2: NumPy Array'den DataFrame (İndeks ve Sütun İsimleri ile)**

### 🎓 **Önce Test Verisi Oluşturalım**

````python
# 🏪 SENARYO: Kampüs Kafeterya Çalışan Performans Tablosu
# Anlamlı indekslerle DataFrame oluşturma

performans_data = {
    'Şube': ['Merkez', 'Merkez', 'Kuzey', 'Kuzey', 'Güney'],
    'Pozisyon': ['Kasiyer', 'Barista', 'Kasiyer', 'Barista', 'Kasiyer'],
    'Günlük_Satış': [450, 380, 520, 410, 390],
    'Müşteri_Puan': [4.2, 4.8, 4.5, 4.7, 4.3],
    'Deneyim_Yıl': [2, 5, 3, 4, 1]
}

# İndeks olarak çalışan isimleri kullanalım
calisanlar_index = ['Ayşe Yılmaz', 'Mehmet Demir', 'Zeynep Kaya', 'Ahmet Çelik', 'Fatma Şahin']

df_performans = pd.DataFrame(performans_data, index=calisanlar_index)

print("🏪 KAMPÜS KAFETERYA - ÇALIŞAN PERFORMANS TABLOSU")
print("="*70)
print(df_performans)
print("="*70)
print(f"\n📊 Boyut: {df_performans.shape[0]} çalışan × {df_performans.shape[1]} metrik")
print(f"📊 İndeks (Satır etiketleri): {list(df_performans.index)}")
print(f"📊 Sütunlar: {list(df_performans.columns)}")
````

### 🏷️ **loc: Etiket ile Seçim**

`loc` ile satır ve sütun **isimlerini** kullanarak seçim yaparsınız.

````python
# 🏷️ loc KULLANIMI - ETİKET İLE SEÇİM

print("🏷️ loc İLE SEÇİM ÖRNEKLERİ")
print("="*70)

# 1️⃣ TEK HÜCRE - Belirli satır ve sütun
print("\n1️⃣ Zeynep Kaya'nın günlük satış sayısı:")
zeynep_satis = df_performans.loc['Zeynep Kaya', 'Günlük_Satış']
print(f"   → {zeynep_satis} adet")

# 2️⃣ TEK SATIR - Bir çalışanın tüm bilgileri
print("\n2️⃣ Mehmet Demir'in tüm bilgileri (Series):")
mehmet_bilgi = df_performans.loc['Mehmet Demir']
print(mehmet_bilgi)
print(f"   • Tip: {type(mehmet_bilgi)}")

# 3️⃣ TEK SÜTUN - Tüm çalışanların bir özelliği
print("\n3️⃣ Tüm çalışanların müşteri puanları:")
musteri_puanlari = df_performans.loc[:, 'Müşteri_Puan']
print(musteri_puanlari)

# 4️⃣ ÇOKLU SATIR - Birden fazla çalışan
print("\n4️⃣ Ayşe ve Fatma'nın bilgileri:")
secili_calisanlar = df_performans.loc[['Ayşe Yılmaz', 'Fatma Şahin']]
print(secili_calisanlar)

# 5️⃣ ÇOKLU SÜTUN - Belirli metrikleri
print("\n5️⃣ Tüm çalışanların şube ve pozisyon bilgileri:")
sube_pozisyon = df_performans.loc[:, ['Şube', 'Pozisyon']]
print(sube_pozisyon)

# 6️⃣ ALT KÜMESİ - Belirli satır ve sütunlar
print("\n6️⃣ Zeynep ve Ahmet'in satış ve puan bilgileri:")
alt_kume = df_performans.loc[['Zeynep Kaya', 'Ahmet Çelik'], ['Günlük_Satış', 'Müşteri_Puan']]
print(alt_kume)

# 7️⃣ DİLİMLEME (Slicing) - Aralık seçimi
print("\n7️⃣ Ayşe'den Ahmet'e kadar (dahil) tüm satırlar:")
aralik = df_performans.loc['Ayşe Yılmaz':'Ahmet Çelik']
print(aralik)
print("   ⚠️ Not: loc'ta slicing SON değeri de dahil eder!")
````

### 🔢 **iloc: Pozisyon ile Seçim**

`iloc` ile satır ve sütun **sıra numaralarını** (0'dan başlar) kullanarak seçim yaparsınız.

````python
# 🔢 iloc KULLANIMI - POZİSYON İLE SEÇİM

print("🔢 iloc İLE SEÇİM ÖRNEKLERİ")
print("="*70)

# 1️⃣ TEK HÜCRE - Pozisyon ile
print("\n1️⃣ 3. satır, 3. sütun (2, 2 - 0'dan başlar):")
hucre = df_performans.iloc[2, 2]
print(f"   → {hucre}")

# 2️⃣ TEK SATIR - İlk satır
print("\n2️⃣ İlk satır (index 0):")
ilk_satir = df_performans.iloc[0]
print(ilk_satir)

# 3️⃣ TEK SÜTUN - İkinci sütun
print("\n3️⃣ İkinci sütun (index 1 - Pozisyon):")
ikinci_sutun = df_performans.iloc[:, 1]
print(ikinci_sutun)

# 4️⃣ ÇOKLU SATIR - İlk 3 satır
print("\n4️⃣ İlk 3 çalışan (0, 1, 2):")
ilk_uc = df_performans.iloc[[0, 1, 2]]
print(ilk_uc)

# 5️⃣ ÇOKLU SÜTUN - İlk 2 sütun
print("\n5️⃣ İlk 2 sütun:")
ilk_iki_sutun = df_performans.iloc[:, [0, 1]]
print(ilk_iki_sutun)

# 6️⃣ ALT KÜMESİ - Belirli satır ve sütunlar
print("\n6️⃣ İlk 2 satır, son 2 sütun:")
alt_kume_iloc = df_performans.iloc[0:2, -2:]
print(alt_kume_iloc)

# 7️⃣ DİLİMLEME (Slicing) - Aralık
print("\n7️⃣ Satır 1'den 3'e kadar (1 dahil, 3 hariç):")
aralik_iloc = df_performans.iloc[1:3]
print(aralik_iloc)
print("   ⚠️ Not: iloc'ta slicing SON değeri HARİÇ tutar!")

# 8️⃣ TERS SIRALAMA - Son satır
print("\n8️⃣ Son satır (negatif indeks):")
son_satir = df_performans.iloc[-1]
print(son_satir)
````

    <h4 style="color: #333; margin-top: 0;">💡 between() Kullanım Örnekleri</h4>
    <p style="color: #333; line-height: 1.8;">
        <strong>Sözdizimi:</strong> <code>Series.between(sol, sag, inclusive='both')</code><br><br>

````python
# 📌 Örnek 1: Orta Seviye Notlar (70-85 arası)
orta_seviye = df_ogrenciler[df_ogrenciler['Ortalama'].between(70, 85)]
print("✅ Orta Seviye Notlar (70-85):")
print("=" * 90)
print(orta_seviye[['Isim', 'Ortalama']])

# 📌 Karşılaştırma: between vs Manuel Koşul
# YÖNTEM 1: Manuel koşul (uzun)
manuel = df_ogrenciler[(df_ogrenciler['Ortalama'] >= 70) & (df_ogrenciler['Ortalama'] <= 85)]

# YÖNTEM 2: between (kısa ve okunabilir)
between_metod = df_ogrenciler[df_ogrenciler['Ortalama'].between(70, 85)]

print(f"\n📊 Her iki yöntem de aynı: {len(manuel)} = {len(between_metod)} öğrenci")

# 📌 Örnek 2: Geçer Not Aralığı (50-70)
gecer_not = df_ogrenciler[df_ogrenciler['Ortalama'].between(50, 70)]
print("\n✅ Geçer Not Aralığı (50-70):")
print("=" * 90)
print(gecer_not[['Isim', 'Bolum', 'Ortalama']])

# 📌 Örnek 3: inclusive='left' - Sol dahil, sağ hariç [70, 85)
sol_dahil = df_ogrenciler[df_ogrenciler['Ortalama'].between(70, 85, inclusive='left')]
print("\n✅ Sol Dahil [70, 85):")
print("=" * 90)
print(sol_dahil[['Isim', 'Ortalama']])

# 📌 Örnek 4: Kabul Edilebilir Devamsızlık (1-3 gün)
kabul_edilebilir = df_ogrenciler[df_ogrenciler['Devamsizlik'].between(1, 3)]
print("\n✅ Kabul Edilebilir Devamsızlık (1-3 gün):")
print("=" * 90)
print(kabul_edilebilir[['Isim', 'Devamsizlik', 'Ortalama']])

# 📌 Örnek 5: Çoklu Koşul ile Kombine
# Ortalama 70-85 arası VE YBS bölümü
ybs_orta = df_ogrenciler[
    df_ogrenciler['Ortalama'].between(70, 85) & 
    (df_ogrenciler['Bolum'] == 'YBS')
]
print("\n✅ YBS Bölümü + Orta Seviye Not:")
print("=" * 90)
print(ybs_orta[['Isim', 'Bolum', 'Ortalama']])
````

            <th style="padding: 10px; border: 1px solid #ddd;">query() Yöntemi</th>
        </tr>
        <tr>
            <td style="padding: 10px; border: 1px solid #ddd;"><code>df[(df['A'] > 5) & (df['B'] < 10)]</code></td>
            <td style="padding: 10px; border: 1px solid #ddd;"><code>df.query('A > 5 and B < 10')</code></td>
        </tr>
        <tr>
            <td style="padding: 10px; border: 1px solid #ddd;">Çok parantez, tekrar eden df[]</td>
            <td style="padding: 10px; border: 1px solid #ddd;">Temiz, okunabilir, SQL-benzeri</td>
        </tr>
    </table>
    <p style="color: #333; margin-top: 15px; line-height: 1.6;">
        ✅ <strong>and</strong>, <strong>or</strong>, <strong>not</strong> kelimelerini kullanabilirsiniz<br>
        ✅ Değişkenlere <strong>@</strong> ile erişim: <code>@min_not</code><br>
        ✅ SQL bilen geliştiriciler için doğal
    </p>
</div>

---

````python
# 📌 Örnek 1: Basit Sorgu
basarili_query = df_ogrenciler.query('Ortalama > 70')
print("✅ Başarılı Öğrenciler (query ile):")
print("=" * 90)
print(basarili_query[['Isim', 'Ortalama']])

# 📌 Örnek 2: AND Koşulu
query_and = df_ogrenciler.query('Ortalama > 70 and Devamsizlik < 3')
print("\n✅ Başarılı VE Düzenli (query ile):")
print("=" * 90)
print(query_and[['Isim', 'Ortalama', 'Devamsizlik']])

# 📌 Karşılaştırma: Klasik vs query()
# YÖNTEM 1: Klasik (uzun)
klasik = df_ogrenciler[(df_ogrenciler['Ortalama'] > 70) & (df_ogrenciler['Devamsizlik'] < 3)]

# YÖNTEM 2: query() (kısa ve temiz)
query_metod = df_ogrenciler.query('Ortalama > 70 and Devamsizlik < 3')

print(f"\n📊 Sonuç aynı: {len(klasik)} = {len(query_metod)} öğrenci")

# 📌 Örnek 3: String Eşitlik Kontrolü
ybs_query = df_ogrenciler.query('Bolum == "YBS"')
print("\n✅ YBS Bölümü (query ile):")
print("=" * 90)
print(ybs_query[['Isim', 'Bolum', 'Ortalama']])

# 📌 Örnek 4: Değişken Kullanımı (@)
min_not = 75
max_devamsizlik = 3
degiskenli = df_ogrenciler.query('Ortalama > @min_not and Devamsizlik < @max_devamsizlik')
print(f"\n✅ Değişkenli Sorgu (Ortalama > {min_not}, Devamsızlık < {max_devamsizlik}):")
print("=" * 90)
print(degiskenli[['Isim', 'Ortalama', 'Devamsizlik']])

# 📌 Örnek 5: Karmaşık Sorgu - OR ve Parantez
karmasik = df_ogrenciler.query('(Ortalama > 80 or Final > 85) and Devamsizlik < 3')
print("\n✅ Karmaşık Sorgu (Yüksek Not VEYA Final) VE Düşük Devamsızlık:")
print("=" * 90)
print(karmasik[['Isim', 'Ortalama', 'Final', 'Devamsizlik']])

# 📌 Örnek 6: Liste Kontrolü (in operatörü)
liste_query = df_ogrenciler.query('Bolum in ["YBS", "Bilgisayar"]')
print("\n✅ Teknik Bölümler (query + in):")
print("=" * 90)
print(liste_query[['Isim', 'Bolum']])

# 📌 Örnek 7: NOT Operatörü
not_query = df_ogrenciler.query('not (Devamsizlik > 5)')
print("\n✅ Devamsızlığı Yüksek OLMAYAN (query + not):")
print("=" * 90)
print(not_query[['Isim', 'Devamsizlik']])
````

````python
# Özel index'li bir DataFrame oluşturalım
df_ozel_index = pd.DataFrame({
    'Isim': ['Ahmet', 'Ayşe', 'Mehmet'],
    'Not': [85, 92, 78]
}, index=['A', 'B', 'C'])

print("📊 Özel Index'li DataFrame:")
print("=" * 60)
print(df_ozel_index)

# 📌 Örnek 1: reset_index (drop=False - varsayılan)
df_reset1 = df_ozel_index.reset_index()
print("\n✅ reset_index() - Eski index sütun oldu:")
print("=" * 60)
print(df_reset1)

# 📌 Örnek 2: reset_index(drop=True) - Eski index atılır
df_reset2 = df_ozel_index.reset_index(drop=True)
print("\n✅ reset_index(drop=True) - Eski index atıldı:")
print("=" * 60)
print(df_reset2)

# 📌 YBS Senaryosu: Gruplama sonrası reset_index
# Öğrenci notlarını bölüme göre grupla
df_bolum = pd.DataFrame({
    'Bolum': ['YBS', 'YBS', 'Bilgisayar', 'Bilgisayar', 'Matematik'],
    'Ogrenci': ['Ahmet', 'Ayşe', 'Mehmet', 'Can', 'Zeynep'],
    'Not': [85, 92, 78, 88, 95]
})

# Gruplama yap - index bölüm olur
ortalamalar = df_bolum.groupby('Bolum')['Not'].mean()
print("\n📊 Gruplama Sonucu (Bölüm index'te):")
print("=" * 60)
print(ortalamalar)
print(f"\nIndex: {ortalamalar.index.tolist()}")

# reset_index ile sütun yap
ortalamalar_reset = ortalamalar.reset_index()
print("\n✅ reset_index sonrası:")
print("=" * 60)
print(ortalamalar_reset)
print(f"\nArtık 'Bolum' bir sütun: {ortalamalar_reset.columns.tolist()}")
````

````python
# 📌 Örnek 1: MultiIndex Oluşturma
df_multi_data = pd.DataFrame({
    'Bolum': ['YBS', 'YBS', 'YBS', 'Bilgisayar', 'Bilgisayar', 'Bilgisayar'],
    'Sinif': [1, 1, 2, 1, 1, 2],
    'Ogrenci': ['Ahmet', 'Ayşe', 'Mehmet', 'Can', 'Elif', 'Burak'],
    'Not': [85, 92, 78, 88, 95, 82]
})

# İki sütunu index yap
df_multi = df_multi_data.set_index(['Bolum', 'Sinif'])
print("📊 MultiIndex DataFrame (Bölüm → Sınıf):")
print("=" * 80)
print(df_multi)
print(f"\nIndex seviyeleri: {df_multi.index.nlevels}")
print(f"Index isimleri: {df_multi.index.names}")

# 📌 Örnek 2: MultiIndex ile Erişim
print("\n🔍 YBS bölümündeki tüm öğrenciler:")
print("=" * 80)
print(df_multi.loc['YBS'])

print("\n🔍 YBS bölümü, 1. sınıf:")
print("=" * 80)
print(df_multi.loc[('YBS', 1)])

# 📌 Örnek 3: Gruplama ile MultiIndex
df_satis = pd.DataFrame({
    'Sube': ['Merkez', 'Merkez', 'Merkez', 'Kuzey', 'Kuzey', 'Kuzey'],
    'Kategori': ['Elektronik', 'Giyim', 'Elektronik', 'Elektronik', 'Giyim', 'Gıda'],
    'Satis': [50000, 30000, 45000, 35000, 25000, 20000]
})

# Şube ve kategoriye göre grupla
satis_gruplanmis = df_satis.groupby(['Sube', 'Kategori'])['Satis'].sum()
print("\n📊 Şube ve Kategoriye Göre Satışlar (MultiIndex):")
print("=" * 80)
print(satis_gruplanmis)

# 📌 Örnek 4: MultiIndex'ten Normal DataFrame'e
satis_reset = satis_gruplanmis.reset_index()
print("\n✅ reset_index() ile normal DataFrame:")
print("=" * 80)
print(satis_reset)

# 📌 Örnek 5: MultiIndex Seviye Değiştirme
print("\n🔄 Seviye 0 (Bölüm) değerleri:")
print(df_multi.index.get_level_values(0).unique())

print("\n🔄 Seviye 1 (Sınıf) değerleri:")
print(df_multi.index.get_level_values(1).unique())
````

        1️⃣ <strong>head():</strong> İlk satırları göster (varsayılan 5)<br>
        2️⃣ <strong>tail():</strong> Son satırları göster<br>
        3️⃣ <strong>sample():</strong> Rastgele örneklem al<br><br>
        
        <strong>Kullanım Alanları:</strong><br>
        ✅ Veri formatını kontrol et<br>
        ✅ Beklenmedik değerleri fark et<br>
        ✅ Sütun içeriklerini anla<br>
        ✅ Hızlı doğrulama yap
    </p>
</div>

````python
# 📊 Örnek Veri Seti: Öğrenci Başarı Sistemi
df_basari = pd.DataFrame({
    'OgrenciNo': ['2021001', '2021002', '2021003', '2021004', '2021005', 
                  '2021006', '2021007', '2021008', '2021009', '2021010'],
    'Isim': ['Ahmet Yılmaz', 'Ayşe Demir', 'Mehmet Kaya', 'Zeynep Şahin', 'Can Öztürk',
             'Elif Arslan', 'Mustafa Çelik', 'Fatma Koç', 'Ali Yıldız', 'Selin Aydın'],
    'Bolum': ['YBS', 'Bilgisayar', 'YBS', 'Matematik', 'YBS',
              'Bilgisayar', 'YBS', 'Matematik', 'YBS', 'Bilgisayar'],
    'Vize': [85, 92, 78, 95, 88, 82, 91, 87, 79, 94],
    'Final': [88, 95, None, 98, 90, 85, None, 92, 82, 97],
    'Proje': [90, None, 85, 95, 92, 88, 93, 90, 80, 96],
    'Devamsizlik': [2, 1, 5, 0, 3, 2, 1, 2, 6, 1]
})

print("🎓 Öğrenci Başarı Sistemi - Ham Veri")
print("=" * 100)

# 📌 Örnek 1: head() - İlk 5 satır (varsayılan)
print("\n1️⃣ İLK 5 SATIR (head()):")
print("-" * 100)
print(df_basari.head())

# 📌 Örnek 2: head(3) - İlk 3 satır
print("\n2️⃣ İLK 3 SATIR (head(3)):")
print("-" * 100)
print(df_basari.head(3))

# 📌 Örnek 3: tail() - Son 5 satır
print("\n3️⃣ SON 5 SATIR (tail()):")
print("-" * 100)
print(df_basari.tail())

# 📌 Örnek 4: tail(2) - Son 2 satır
print("\n4️⃣ SON 2 SATIR (tail(2)):")
print("-" * 100)
print(df_basari.tail(2))

# 📌 Örnek 5: sample() - Rastgele 3 satır
print("\n5️⃣ RASTGELE 3 SATIR (sample(3)):")
print("-" * 100)
print(df_basari.sample(3, random_state=42))  # random_state sabit sonuçlar için

print("\n" + "=" * 100)
print("💡 İpucu: sample() ile veri setinin genel yapısını hızlıca anlayabilirsiniz!")
````

````python
print("📊 VERİ YAPISI ANALİZİ")
print("=" * 100)

# 📌 Örnek 1: info() - Detaylı bilgi
print("\n1️⃣ DETAYLI BİLGİ (info()):")
print("-" * 100)
print(df_basari.info())

# 📌 Örnek 2: shape - Boyut bilgisi
print("\n2️⃣ VERİ SETİ BOYUTU (shape):")
print("-" * 100)
print(f"Satır sayısı: {df_basari.shape[0]}")
print(f"Sütun sayısı: {df_basari.shape[1]}")
print(f"Toplam hücre: {df_basari.shape[0] * df_basari.shape[1]}")

# 📌 Örnek 3: dtypes - Veri tipleri
print("\n3️⃣ SÜTUN VERİ TİPLERİ (dtypes):")
print("-" * 100)
print(df_basari.dtypes)

# 📌 Örnek 4: columns - Sütun isimleri
print("\n4️⃣ SÜTUN İSİMLERİ (columns):")
print("-" * 100)
print(df_basari.columns.tolist())

# 📌 Örnek 5: index - Index bilgisi
print("\n5️⃣ INDEX BİLGİSİ:")
print("-" * 100)
print(f"Index tipi: {type(df_basari.index)}")
print(f"Index aralığı: {df_basari.index[0]} - {df_basari.index[-1]}")

# 📌 Örnek 6: Bellek kullanımı
print("\n6️⃣ BELLEK KULLANIMI:")
print("-" * 100)
bellek_mb = df_basari.memory_usage(deep=True).sum() / 1024
print(f"Toplam bellek: {bellek_mb:.2f} KB")

print("\n" + "=" * 100)
print("💡 YBS İpucu: info() her analiz işleminin ilk adımıdır!")
````

````python
print("📈 İSTATİSTİKSEL ÖZET")
print("=" * 100)

# 📌 Örnek 1: describe() - Temel istatistikler
print("\n1️⃣ TEMEL İSTATİSTİKLER (describe()):")
print("-" * 100)
print(df_basari.describe())

# 📌 Örnek 2: describe() tüm sütunlar
print("\n2️⃣ TÜM SÜTUNLAR DAHİL (include='all'):")
print("-" * 100)
print(df_basari.describe(include='all'))

# 📌 Örnek 3: Belirli istatistikleri çek
print("\n3️⃣ ORTALAMALAR:")
print("-" * 100)
print(df_basari[['Vize', 'Final', 'Proje', 'Devamsizlik']].mean())

print("\n4️⃣ STANDART SAPMALAR (Değişkenlik):")
print("-" * 100)
print(df_basari[['Vize', 'Final', 'Proje']].std())

print("\n5️⃣ MİNİMUM ve MAKSİMUM:")
print("-" * 100)
print("Minimumlar:")
print(df_basari[['Vize', 'Final', 'Proje', 'Devamsizlik']].min())
print("\nMaksimumlar:")
print(df_basari[['Vize', 'Final', 'Proje', 'Devamsizlik']].max())

# 📌 Örnek 4: value_counts() - Frekans analizi
print("\n6️⃣ BÖLÜM DAĞILIMI (value_counts()):")
print("-" * 100)
print(df_basari['Bolum'].value_counts())

print("\n7️⃣ BÖLÜM YÜZDE DAĞILIMI:")
print("-" * 100)
print(df_basari['Bolum'].value_counts(normalize=True) * 100)

print("\n" + "=" * 100)
print("💡 İpucu: describe() ile veri setinin genel sağlığını hızlıca kontrol edin!")
````

````python
print("❌ EKSİK VERİ YÖNETİMİ")
print("=" * 100)

# Mevcut veri setinde eksik veriler var
print("\n📊 Mevcut Veri Seti:")
print(df_basari)

# 📌 Örnek 1: Eksik veri tespiti
print("\n1️⃣ EKSİK VERİ TESPİTİ (isnull()):")
print("-" * 100)
print(df_basari.isnull())

print("\n2️⃣ EKSİK VERİ SAYILARI:")
print("-" * 100)
print(df_basari.isnull().sum())

print("\n3️⃣ EKSİK VERİ YÜZDE:")
print("-" * 100)
eksik_yuzde = (df_basari.isnull().sum() / len(df_basari)) * 100
print(eksik_yuzde)

# 📌 Örnek 2: Eksik olmayan veriler
print("\n4️⃣ EKSİK OLMAYAN VERİ SAYILARI (notnull()):")
print("-" * 100)
print(df_basari.notnull().sum())

# 📌 Örnek 3: Eksik veri olan satırları bul
print("\n5️⃣ EKSİK VERİ OLAN SATIRLAR:")
print("-" * 100)
eksik_satirlar = df_basari[df_basari.isnull().any(axis=1)]
print(eksik_satirlar)

# 📌 Örnek 4: dropna() - Eksik veri içeren satırları sil
print("\n6️⃣ EKSİK SATIRLARI SİL (dropna()):")
print("-" * 100)
df_temiz = df_basari.dropna()
print(df_temiz)
print(f"\nÖnceki satır sayısı: {len(df_basari)}")
print(f"Sonraki satır sayısı: {len(df_temiz)}")
print(f"Silinen satır: {len(df_basari) - len(df_temiz)}")

# 📌 Örnek 5: dropna(subset=[]) - Belirli sütunlarda eksik olanları sil
print("\n7️⃣ SADECE 'Final' EKSİK OLANLARI SİL:")
print("-" * 100)
df_final_temiz = df_basari.dropna(subset=['Final'])
print(df_final_temiz)
print(f"Kalan satır sayısı: {len(df_final_temiz)}")

print("\n" + "=" * 100)
print("💡 Dikkat: dropna() orijinal DataFrame'i değiştirmez! inplace=True kullanın veya yeni değişkene atayın.")
````

````python
print("✅ EKSİK VERİ DOLDURMA (fillna)")
print("=" * 100)

# 📌 Örnek 1: Sabit değer ile doldurma
print("\n1️⃣ SABİT DEĞER İLE DOLDURMA (fillna(0)):")
print("-" * 100)
df_sifir = df_basari.fillna(0)
print(df_sifir[['Isim', 'Final', 'Proje']])

# 📌 Örnek 2: Ortalama ile doldurma
print("\n2️⃣ ORTALAMA İLE DOLDURMA:")
print("-" * 100)
df_ortalama = df_basari.copy()
df_ortalama['Final'] = df_ortalama['Final'].fillna(df_ortalama['Final'].mean())
df_ortalama['Proje'] = df_ortalama['Proje'].fillna(df_ortalama['Proje'].mean())
print(df_ortalama[['Isim', 'Final', 'Proje']])
print(f"\nFinal ortalaması: {df_basari['Final'].mean():.2f}")
print(f"Proje ortalaması: {df_basari['Proje'].mean():.2f}")

# 📌 Örnek 3: Medyan ile doldurma
print("\n3️⃣ MEDYAN İLE DOLDURMA:")
print("-" * 100)
df_medyan = df_basari.copy()
df_medyan['Final'] = df_medyan['Final'].fillna(df_medyan['Final'].median())
df_medyan['Proje'] = df_medyan['Proje'].fillna(df_medyan['Proje'].median())
print(df_medyan[['Isim', 'Final', 'Proje']])

# 📌 Örnek 4: İleri doldurma (forward fill)
print("\n4️⃣ İLERİ DOLDURMA (ffill):")
print("-" * 100)
df_ileri = df_basari[['Isim', 'Final']].copy()
df_ileri['Final_ffill'] = df_ileri['Final'].fillna(method='ffill')
print(df_ileri)

# 📌 Örnek 5: Geri doldurma (backward fill)
print("\n5️⃣ GERİ DOLDURMA (bfill):")
print("-" * 100)
df_geri = df_basari[['Isim', 'Final']].copy()
df_geri['Final_bfill'] = df_geri['Final'].fillna(method='bfill')
print(df_geri)

# 📌 Örnek 6: Sütuna göre farklı değerler
print("\n6️⃣ SÜTUNLARA ÖZEL DEĞERLER:")
print("-" * 100)
degerler = {'Final': 70, 'Proje': 80}  # Final için 70, Proje için 80
df_ozel = df_basari.fillna(degerler)
print(df_ozel[['Isim', 'Final', 'Proje']])

print("\n" + "=" * 100)
print("💡 YBS İpucu: Sayısal veriler için ortalama/medyan, kategorik veriler için mod kullanın!")
````

````python
print("🔄 TEKRARLAYAN VERİ YÖNETİMİ")
print("=" * 100)

# Duplikasyon örnekleri için yeni veri seti
df_duplike = pd.DataFrame({
    'OgrenciNo': ['2021001', '2021002', '2021003', '2021001', '2021004', '2021002'],
    'Isim': ['Ahmet', 'Ayşe', 'Mehmet', 'Ahmet', 'Zeynep', 'Ayşe'],
    'Bolum': ['YBS', 'Bilgisayar', 'YBS', 'YBS', 'Matematik', 'Bilgisayar'],
    'Not': [85, 92, 78, 85, 95, 92]
})

print("📊 Duplikasyon İçeren Veri Seti:")
print(df_duplike)

# 📌 Örnek 1: duplicated() - Duplike satırları tespit
print("\n1️⃣ DUPLİKE SATIRLAR (duplicated()):")
print("-" * 100)
print(df_duplike.duplicated())

print("\n2️⃣ DUPLİKE SATIR SAYISI:")
print("-" * 100)
print(f"Duplike satır sayısı: {df_duplike.duplicated().sum()}")

# 📌 Örnek 2: Duplike satırları göster
print("\n3️⃣ DUPLİKE SATIRLARI GÖSTER:")
print("-" * 100)
duplike_satirlar = df_duplike[df_duplike.duplicated(keep=False)]
print(duplike_satirlar)

# 📌 Örnek 3: drop_duplicates() - Duplike satırları kaldır
print("\n4️⃣ DUPLİKE SATIRLARI KALDIR (drop_duplicates()):")
print("-" * 100)
df_benzersiz = df_duplike.drop_duplicates()
print(df_benzersiz)
print(f"\nÖnceki satır sayısı: {len(df_duplike)}")
print(f"Sonraki satır sayısı: {len(df_benzersiz)}")
print(f"Kaldırılan satır: {len(df_duplike) - len(df_benzersiz)}")

# 📌 Örnek 4: Belirli sütuna göre duplikasyon kontrolü
print("\n5️⃣ SADECE 'OgrenciNo'ya GÖRE DUPLİKE KONTROLÜ:")
print("-" * 100)
df_ogrenci_benzersiz = df_duplike.drop_duplicates(subset=['OgrenciNo'])
print(df_ogrenci_benzersiz)

# 📌 Örnek 5: keep parametresi
print("\n6️⃣ KEEP PARAMETRESİ:")
print("-" * 100)
print("keep='first' (varsayılan) - İlk kaydı tut:")
print(df_duplike.drop_duplicates(keep='first'))
print("\nkeep='last' - Son kaydı tut:")
print(df_duplike.drop_duplicates(keep='last'))

# 📌 Örnek 6: Gerçek senaryo - Log temizleme
print("\n7️⃣ YBS Senaryo: Sistem Log Temizleme")
print("-" * 100)
df_log_duplike = pd.DataFrame({
    'Tarih': ['2026-02-01', '2026-02-01', '2026-02-01', '2026-02-02'],
    'Kullanici': ['ahmet', 'ahmet', 'ayse', 'ahmet'],
    'Islem': ['Login', 'Login', 'Login', 'Login']
})
print("Ham Log:")
print(df_log_duplike)
print("\nTemizlenmiş Log (aynı gün aynı kullanıcı tek kayıt):")
df_log_temiz = df_log_duplike.drop_duplicates(subset=['Tarih', 'Kullanici'])
print(df_log_temiz)

print("\n" + "=" * 100)
print("💡 İpucu: Duplikasyon kontrolü her veri analizi projesinin önemli bir adımıdır!")
````

    G -.astype.-> I

    style F fill:#4ECDC4,stroke:#333,stroke-width:2px,color:#fff

````python
print("🔄 VERİ TİPİ DÖNÜŞÜMLERİ")
print("=" * 100)

# Test verisi
df_tip_donusum = pd.DataFrame({
    'OgrenciNo': ['001', '002', '003', '004'],
    'Isim': ['Ahmet', 'Ayşe', 'Mehmet', 'Zeynep'],
    'Yas': ['22', '21', '23', '22'],  # String olarak
    'Not': [85.5, 92.3, 78.0, 95.8],
    'Tarih': ['2026-02-01', '2026-02-02', '2026-02-03', '2026-02-04'],
    'Bolum': ['YBS', 'Bilgisayar', 'YBS', 'Matematik']
})

print("📊 Orijinal Veri ve Tipleri:")
print(df_tip_donusum)
print("\nVeri Tipleri:")
print(df_tip_donusum.dtypes)

# 📌 Örnek 1: String'den Integer'a
print("\n1️⃣ STRING → INTEGER (astype(int)):")
print("-" * 100)
df_tip_donusum['Yas'] = df_tip_donusum['Yas'].astype(int)
print(df_tip_donusum[['Isim', 'Yas']])
print(f"Yas sütunu tipi: {df_tip_donusum['Yas'].dtype}")

# 📌 Örnek 2: Float'tan Integer'a
print("\n2️⃣ FLOAT → INTEGER:")
print("-" * 100)
df_tip_donusum['Not_Int'] = df_tip_donusum['Not'].astype(int)
print(df_tip_donusum[['Isim', 'Not', 'Not_Int']])

# 📌 Örnek 3: String'den Datetime'a
print("\n3️⃣ STRING → DATETIME (pd.to_datetime()):")
print("-" * 100)
df_tip_donusum['Tarih'] = pd.to_datetime(df_tip_donusum['Tarih'])
print(df_tip_donusum[['Isim', 'Tarih']])
print(f"Tarih sütunu tipi: {df_tip_donusum['Tarih'].dtype}")

# 📌 Örnek 4: Kategorik veri
print("\n4️⃣ STRING → CATEGORY (Memory Optimization):")
print("-" * 100)
print(f"Önce - Bolum bellek kullanımı: {df_tip_donusum['Bolum'].memory_usage(deep=True)} bytes")
df_tip_donusum['Bolum'] = df_tip_donusum['Bolum'].astype('category')
print(f"Sonra - Bolum bellek kullanımı: {df_tip_donusum['Bolum'].memory_usage(deep=True)} bytes")
print(f"Bolum tipi: {df_tip_donusum['Bolum'].dtype}")

# 📌 Örnek 5: to_numeric() - Hatalı dönüşümleri yönet
print("\n5️⃣ HATA YÖNETİMİ (pd.to_numeric()):")
print("-" * 100)
hata_verisi = pd.Series(['100', '200', 'N/A', '400', 'Eksik'])
print("Ham veri:")
print(hata_verisi)

# errors='coerce' ile hatalı değerler NaN olur
temiz_veri = pd.to_numeric(hata_verisi, errors='coerce')
print("\nTemizlenmiş (errors='coerce'):")
print(temiz_veri)

# 📌 Örnek 6: Toplu tip dönüşümü
print("\n6️⃣ TOPLU TİP DÖNÜŞÜMLERİ:")
print("-" * 100)
print("\nGüncel Veri Tipleri:")
print(df_tip_donusum.dtypes)

print("\n" + "=" * 100)
print("💡 İpucu: Kategorik veriler için 'category' tipi bellek kullanımını önemli ölçüde azaltır!")
````
