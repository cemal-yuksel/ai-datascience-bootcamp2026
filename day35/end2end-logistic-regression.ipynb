## 🎯 **Adım 1: Proje Tanıtımı ve CRISP-DM Metodolojisi**

### 📌** Lojistik Regresyon Nedir?**

**Lojistik Regresyon**, <span style='background-color: #FFE5CC; padding: 2px 6px; border-radius: 3px; font-weight: bold;'>sınıflandırma problemleri</span> için kullanılan denetimli bir makine öğrenmesi algoritmasıdır.

#### 🔑 **Temel Özellikler:**

- **İşlevi:** Kategorik sonuçları tahmin etmek
- **Çıktı:** Olasılık değerleri (0 ile 1 arası)
- **Kullanım Alanları:** İkili sınıflandırma (binary classification) veya çoklu sınıflandırma
- **Matematiksel Temel:** Sigmoid (lojistik) fonksiyonu

#### 💼 **YBS Perspektifinden Örnekler:**

| İş Problemi | Hedef Değişken | YBS Uygulaması |
|------------|----------------|----------------|
| 🏦 **Kredi Risk Analizi** | Kredi onaylanacak mı? (Evet/Hayır) | Risk yönetimi sistemleri |
| 🛒 **Müşteri Churn** | Müşteri ayrılacak mı? (Evet/Hayır) | CRM sistemleri |
| 📧 **Spam Tespiti** | Email spam mi? (Spam/Normal) | Email filtreleme sistemleri |
| 🏥 **Hastalık Teşhisi** | Hastalık var mı? (Evet/Hayır) | Sağlık bilgi sistemleri |
| 🎓 **Öğrenci Başarısı** | Sınavı geçecek mi? (Geçer/Kalır) | Eğitim yönetim sistemleri |

### 🔬 **CRISP-DM Metodolojisi**

**CRISP-DM** (Cross-Industry Standard Process for Data Mining), veri madenciliği ve makine öğrenmesi projelerinde kullanılan <span style='background-color: #E6F4EA; padding: 2px 6px; border-radius: 3px; font-weight: bold;'>endüstri standardı bir süreç modelidir</span>.

#### 📊 **CRISP-DM'nin 6 Aşaması:**

```mermaid
graph TB
    A[1. Business Understanding<br/>İş Anlayışı] --> B[2. Data Understanding<br/>Veri Anlayışı]
    B --> C[3. Data Preparation<br/>Veri Hazırlama]

    D --> E[5. Evaluation<br/>Değerlendirme]
```

### 📋 **CRISP-DM Aşamalarının Detaylı Açıklaması**

#### <span style='background-color: #FFE5CC; padding: 4px 8px; border-radius: 4px;'>1️⃣ Business Understanding (İş Anlayışı)</span>

**Amaç:** Projenin iş hedeflerini anlamak ve makine öğrenmesi problemine dönüştürmek.

**Yapılacaklar:**
- 🎯 İş probleminin net tanımı
- 📈 Başarı kriterlerinin belirlenmesi
- 💰 Maliyet-fayda analizi
- 📊 KPI (Key Performance Indicator) tanımları

**YBS Örneği:** Bir e-ticaret şirketi müşteri kaybını azaltmak istiyor → Churn tahmini modeli geliştirilmeli.

#### <span style='background-color: #E6F4EA; padding: 4px 8px; border-radius: 4px;'>2️⃣ Data Understanding (Veri Anlayışı)</span>

**Amaç:** Veriyi toplamak, keşfetmek ve kalite problemlerini tespit etmek.

**Yapılacaklar:**
- 📥 Veri toplama ve yükleme
- 🔍 Keşifsel Veri Analizi (EDA)
- 📊 İstatistiksel özetler
- 🎨 Görselleştirmeler
- ❓ Veri kalitesi değerlendirmesi

#### <span style='background-color: #E8F4F8; padding: 4px 8px; border-radius: 4px;'>3️⃣ Data Preparation (Veri Hazırlama)</span>

**Amaç:** Veriyi modelleme için hazır hale getirmek.

**Yapılacaklar:**
- 🧹 Veri temizleme (eksik değerler, aykırı değerler)
- 🔧 Feature Engineering (yeni öznitelik türetme)
- 🔄 Dönüşümler (encoding, scaling)
- ✂️ Feature Selection (özellik seçimi)
- 📦 Train-Test split

**Not:** Bu aşama genellikle en çok zaman alan aşamadır (%60-80).

#### <span style='background-color: #F4E6FF; padding: 4px 8px; border-radius: 4px;'>4️⃣ Modeling (Modelleme)</span>

**Amaç:** Makine öğrenmesi modelini seçmek, eğitmek ve optimize etmek.

**Yapılacaklar:**
- 🤖 Model seçimi (Lojistik Regresyon, KNN, SVM, vb.)
- 🎓 Model eğitimi
- 🔧 Hiperparametre optimizasyonu
- ✅ Cross-validation
- 📊 Farklı modellerin karşılaştırılması

#### <span style='background-color: #FFE6E6; padding: 4px 8px; border-radius: 4px;'>5️⃣ Evaluation (Değerlendirme)</span>

**Amaç:** Modelin performansını değerlendirmek ve iş hedeflerini karşılayıp karşılamadığını kontrol etmek.

**Yapılacaklar:**
- 📈 Performans metrikleri (Accuracy, Precision, Recall, F1-Score, ROC-AUC)
- 🔍 Confusion Matrix analizi
- 💼 İş değeri analizi
- ✔️ Model validasyonu
- 🔄 Gerekirse önceki aşamalara geri dönüş

#### <span style='background-color: #FFF9E6; padding: 4px 8px; border-radius: 4px;'>6️⃣ Deployment (Canlıya Alma)</span>

**Amaç:** Modeli üretim ortamına almak ve kullanıma sunmak.

**Yapılacaklar:**
- 💾 Model kaydetme (pickle, joblib)
- 🚀 Model deployment (API, web servisi)
- 📡 Monitoring sistemi kurma
- 📝 Dokümantasyon
- 🔄 Model güncelleme stratejisi

**Not:** Bu projede Deployment aşamasına kadar ilerleyeceğiz, model kaydedilerek hazır hale getirilecek.

### 🎓 **YBS Öğrencileri İçin Önemli Notlar**

> **💡 Not 1:** CRISP-DM doğrusal bir süreç değildir. İteratif (yinelemeli) bir yaklaşımdır. Değerlendirme aşamasında yetersiz sonuçlar alırsanız, önceki aşamalara geri dönebilirsiniz.

> **💡 Not 2:** İş dünyasında makine öğrenmesi projeleri her zaman teknik bir problem değil, öncelikle bir **iş problemi**dir. Bu yüzden Business Understanding aşaması kritik öneme sahiptir.

> **💡 Not 3:** Veri bilimciler zamanlarının çoğunu (%60-80) Data Preparation aşamasında harcarlar. Temiz veri = Başarılı model!

### 🌐 **Veri Setini İndirme ve Kaydetme**

Veri setini URL üzerinden pandas ile yükleyip, lokale kaydedeceğiz:

````python
# Veri setini yükleme - Akıllı separator algılama ile
# Not: Farklı kaynaklardaki veri setleri farklı ayırıcılar kullanabilir (virgül veya noktalı virgül)

data_folder = "data"
if not os.path.exists(data_folder):
    os.makedirs(data_folder)
    print(f"📁 '{data_folder}' klasörü oluşturuldu")

local_file_path = os.path.join(data_folder, "bank_marketing.csv")

# Veriyi yüklemek için yardımcı fonksiyon - doğru separator'ı otomatik algılar
def load_bank_data(file_path_or_url, is_url=False):
    """Veriyi yükler ve doğru separator'ı otomatik algılar"""
    separators = [';', ',']  # Olası ayırıcılar
    
    for sep in separators:
        try:
            if is_url:
                df_test = pd.read_csv(file_path_or_url, sep=sep, nrows=5)
            else:
                df_test = pd.read_csv(file_path_or_url, sep=sep, nrows=5)
            
            # Eğer çok fazla sütun varsa (>15), doğru separator bulunmuştur
            if df_test.shape[1] > 15:
                print(f"✅ Doğru ayırıcı bulundu: '{sep}'")
                if is_url:
                    return pd.read_csv(file_path_or_url, sep=sep)
                else:
                    return pd.read_csv(file_path_or_url, sep=sep)
        except:
            continue
    
    # Hiçbir separator çalışmazsa, varsayılan olarak oku
    print("⚠️ Separator otomatik algılanamadı, varsayılan ayarlarla yükleniyor...")
    return pd.read_csv(file_path_or_url)

# Eğer daha önce indirilmediyse, alternatif kaynaktan indir
if not os.path.exists(local_file_path):
    print("📥 Veri seti indiriliyor...")
    
    # Alternatif URL'ler - sırayla denenecek
    urls = [
        "https://raw.githubusercontent.com/madmashup/targeted-marketing-predictive-engine/master/banking.csv",
        "https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/master/bank.csv"
    ]
    
    df = None
    for i, url in enumerate(urls):
        try:
            print(f"🔗 Kaynak {i+1} deneniyor...")
            df = load_bank_data(url, is_url=True)
            
            # Veri setinin geçerli olup olmadığını kontrol et
            if df is not None and df.shape[1] > 10:
                print(f"✅ Veri seti başarıyla yüklendi! ({df.shape[0]} satır, {df.shape[1]} sütun)")
                break
        except Exception as e:
            print(f"❌ Kaynak {i+1} başarısız: {str(e)[:50]}...")
            continue
    
    # Eğer hiçbir URL çalışmazsa
    if df is None or df.shape[1] <= 10:
        print("\n" + "="*70)
        print("⚠️ OTOMATİK İNDİRME BAŞARISIZ!")
        print("="*70)
        print("\n📌 Lütfen veri setini manuel olarak indirin:")
        print("   1. https://www.kaggle.com/datasets/janiobachmann/bank-marketing-dataset")
        print("   2. 'bank.csv' dosyasını indirin")
        print(f"   3. '{data_folder}' klasörüne 'bank_marketing.csv' adıyla kaydedin")
        print("="*70)
        raise FileNotFoundError("Veri seti indirilemedi. Lütfen manuel olarak indirin.")
    
    # CSV olarak kaydetme
    df.to_csv(local_file_path, index=False)
    print(f"💾 Veri seti lokale kaydedildi: {local_file_path}")
    
else:
    # Daha önce indirilmiş veriyi yükle
    print("📂 Lokalde kaydedilmiş veri seti bulundu, yükleniyor...")
    df = load_bank_data(local_file_path, is_url=False)
    
    # Eğer yanlış parse edilmişse (tek sütun varsa), dosyayı sil ve yeniden indir
    if df.shape[1] <= 2:
        print("⚠️ Veri seti yanlış formatlanmış, yeniden indiriliyor...")
        os.remove(local_file_path)
        
        # Yeniden indir
        urls = [
            "https://raw.githubusercontent.com/madmashup/targeted-marketing-predictive-engine/master/banking.csv",
            "https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/master/bank.csv"
        ]
        
        for i, url in enumerate(urls):
            try:
                print(f"🔗 Kaynak {i+1} deneniyor...")
                df = load_bank_data(url, is_url=True)
                if df.shape[1] > 10:
                    print(f"✅ Veri seti başarıyla yüklendi!")
                    df.to_csv(local_file_path, index=False)
                    print(f"💾 Veri seti lokale kaydedildi: {local_file_path}")
                    break
            except Exception as e:
                continue
    else:
        print("✅ Veri seti başarıyla yüklendi!")

# Veri seti hakkında temel bilgiler
print("\n" + "="*60)
print("📊 VERİ SETİ TEMEL BİLGİLERİ")
print("="*60)
print(f"📏 Satır Sayısı    : {df.shape[0]:,}")
print(f"📏 Sütun Sayısı    : {df.shape[1]}")
print(f"💾 Bellek Kullanımı: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
print(f"🎯 Hedef Değişken  : 'y' (Vadeli mevduat açacak mı?)")
print("="*60)
````

````python
# Hedef değişkenin dağılımı
target_counts = df['y'].value_counts()
target_percentages = df['y'].value_counts(normalize=True) * 100

# Hedef değişkenin değerlerini otomatik algılama
target_values = target_counts.index.tolist()
# Genellikle 'no', 0, 'n' gibi değerler negatif, 'yes', 1 gibi değerler pozitif
negative_value = target_values[0]  # İlk değer (genellikle çoğunluk sınıfı)
positive_value = target_values[1] if len(target_values) > 1 else target_values[0]

print("📊 HEDEF DEĞİŞKEN (y) DAĞILIMI")
print("="*60)
print(f"❌ {negative_value} (Ürünü Almayan)  : {target_counts.iloc[0]:,} ({target_percentages.iloc[0]:.2f}%)")
print(f"✅ {positive_value} (Ürünü Alan)      : {target_counts.iloc[1]:,} ({target_percentages.iloc[1]:.2f}%)")
print("="*60)

# Görselleştirme
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Hedef Degisken Dagilimi', 'Yuzdelik Oran'),
    specs=[[{"type": "bar"}, {"type": "pie"}]]
)

# Bar chart
fig.add_trace(
    go.Bar(
        x=[f'{negative_value} (Almayan)', f'{positive_value} (Alan)'],
        y=[target_counts.iloc[0], target_counts.iloc[1]],
        marker_color=['#FFB3BA', '#BAFFC9'],
        text=[f"{target_counts.iloc[0]:,}", f"{target_counts.iloc[1]:,}"],
        textposition='outside',
        showlegend=False
    ),
    row=1, col=1
)

# Pie chart
fig.add_trace(
    go.Pie(
        labels=[f'{negative_value} (Almayan)', f'{positive_value} (Alan)'],
        values=[target_counts.iloc[0], target_counts.iloc[1]],
        marker_colors=['#FFB3BA', '#BAFFC9'],
        textinfo='label+percent',
        textfont_size=12
    ),
    row=1, col=2
)

fig.update_layout(
    title_text="Vadeli Mevduat Urun Satin Alma Durumu",
    title_font_size=16,
    height=400,
    showlegend=False
)

fig.show()

# YBS perspektifinden değerlendirme
print("\n💼 YBS PERSPEKTİFİNDEN DEĞERLENDİRME:")
print("="*60)
print(f"⚠️ Veri dengesizlik var! (Class Imbalance)")
print(f"   - Çoğunluk sınıfı: {negative_value} ({target_percentages.iloc[0]:.1f}%)")
print(f"   - Azınlık sınıfı : {positive_value} ({target_percentages.iloc[1]:.1f}%)")
print(f"\n📌 İş Etkisi:")
print(f"   Müşterilerin sadece %{target_percentages.iloc[1]:.1f}'i ürünü satın alıyor.")
print(f"   Bu da call center operasyonlarının verimliliğinin düşük olduğunu gösteriyor.")
print("="*60)
````

### 🔗 **Korelasyon Analizi**

Korelasyon analizi, değişkenler arasındaki ilişkinin gücünü ve yönünü ölçer. Pearson korelasyon katsayısı **-1 ile +1** arasında değer alır:

- **+1**: Mükemmel pozitif ilişki
- **0**: İlişki yok  
- **-1**: Mükemmel negatif ilişki

Makine öğrenmesinde **yüksek korelasyonlu değişkenler** (multicollinearity) model performansını olumsuz etkileyebilir.

````python
# Korelasyon matrisi hesaplama
correlation_matrix = df[numerical_features + ['y']].corr()

# İnteraktif korelasyon ısı haritası (heatmap)
fig = go.Figure(data=go.Heatmap(
    z=correlation_matrix.values,
    x=correlation_matrix.columns,
    y=correlation_matrix.columns,
    colorscale='RdYlGn',  # Kırmızı-Sarı-Yeşil
    zmid=0,  # Orta nokta 0'da
    text=correlation_matrix.round(2).values,
    texttemplate='%{text}',
    textfont={"size": 10},
    colorbar=dict(
        title=dict(text="Korelasyon", side="right"),
        tickmode="linear",
        tick0=-1,
        dtick=0.5
    )
))

fig.update_layout(
    title="<b>Numerik Değişkenler ve Hedef Değişken Korelasyon Matrisi</b>",
    title_font=dict(size=18, color='#2F4F4F'),
    width=900,
    height=800,
    xaxis=dict(tickangle=-45),
    plot_bgcolor='white',
    paper_bgcolor='#F5F5F5'
)

fig.show()

print("✅ Korelasyon matrisi başarıyla oluşturuldu!")
print("\n📊 HEDEF DEĞİŞKEN (y) İLE EN YÜKSEK KORELASYONLAR:")
print("="*70)

# Hedef değişken ile korelasyonları sırala
target_corr = correlation_matrix['y'].drop('y').abs().sort_values(ascending=False)

for idx, (feature, corr_value) in enumerate(target_corr.items(), 1):
    original_corr = correlation_matrix['y'][feature]
    direction = "Pozitif ↑" if original_corr > 0 else "Negatif ↓"
    bar = "█" * int(abs(original_corr) * 50)  # Görsel bar
    print(f"{idx:2d}. {feature:20s}: {original_corr:7.4f} ({direction:12s}) {bar}")
````

## 🔧 **Adım 5: Data Preparation - Veri Hazırlama**

Veri hazırlama, ham veriyi makine öğrenmesi modellerine uygun hale getirme sürecidir. Bu aşamada:

- **Kategorik değişkenleri** sayısal değerlere dönüştüreceğiz (Encoding)

- Veriyi **eğitim ve test** setlerine ayıracağız
- **Class imbalance** sorununu ele alacağız

Bu adımlar, modelin doğru ve tutarlı öğrenmesi için kritik öneme sahiptir.

### 🔄 **Hedef Değişkeni Ayırma**

İlk adım olarak hedef değişkeni (y) bağımsız değişkenlerden (X) ayırmalıyız.

````python
# Hedef değişken ve bağımsız değişkenleri ayır
X = df.drop('y', axis=1)
y = df['y']

print("🎯 HEDEF DEĞİŞKEN VE BAĞIMSIZ DEĞİŞKENLER AYRILDI")
print("="*70)
print(f"📊 Bağımsız Değişkenler (X) Boyutu: {X.shape}")
print(f"🎯 Hedef Değişken (y) Boyutu      : {y.shape}")
print(f"📝 Toplam Özellik Sayısı          : {X.shape[1]}")
print("="*70)

print("\n📋 X (Bağımsız Değişkenler) - İlk 5 satır:")
print(X.head())
````

### 🔤 **Kategorik Değişkenlerin Encoding İşlemi**

Makine öğrenmesi algoritmaları sadece sayısal verilerle çalışabilir. Bu nedenle kategorik değişkenleri sayısal değerlere dönüştürmeliyiz.

**Encoding Yöntemleri:**

1. **Label Encoding**: Kategorileri 0, 1, 2... şeklinde sayılara çevirir (sıralı kategoriler için)
2. **One-Hot Encoding**: Her kategori için yeni bir binary (0/1) sütun oluşturur (sırasız kategoriler için)

Bu projede **One-Hot Encoding** kullanacağız çünkü kategorilerimiz arasında doğal bir sıralama yok.

````python
# Kategorik değişkenleri tespit et (X'ten)
categorical_features = X.select_dtypes(include=['object']).columns.tolist()
numerical_features_X = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("🔤 KATEGORİK DEĞİŞKENLER - ONE-HOT ENCODING")
print("="*70)
print(f"📝 Kategorik Değişken Sayısı: {len(categorical_features)}")
print("   Kategorik Değişkenler:", categorical_features)
print(f"\n🔢 Numerik Değişken Sayısı: {len(numerical_features_X)}")
print("="*70)

# One-Hot Encoding uygula
X_encoded = pd.get_dummies(X, columns=categorical_features, drop_first=True, dtype=int)

print(f"\n✅ ONE-HOT ENCODING TAMAMLANDI!")
print("="*70)
print(f"📊 Encoding Öncesi Özellik Sayısı : {X.shape[1]}")
print(f"📊 Encoding Sonrası Özellik Sayısı: {X_encoded.shape[1]}")
print(f"➕ Eklenen Özellik Sayısı         : {X_encoded.shape[1] - X.shape[1]}")
print("="*70)

print("\n📋 X_encoded (Encoded Veriler) - İlk 5 satır, ilk 15 sütun:")
print(X_encoded.iloc[:5, :15])
````

### ✂️ **Train-Test Split (Veri Bölme)**

Makine öğrenmesinde veriyi **eğitim (train)** ve **test** setlerine ayırmak çok önemlidir:

- **Eğitim Seti (Training Set)**: Model bu veriden öğrenir
- **Test Seti (Test Set)**: Modelin performansını değerlendirmek için kullanılır (modelin görmediği veri)

**Stratified Split**: Hedef değişkenin dağılımını koruyarak bölme yapar. Class imbalance durumlarında kritik öneme sahiptir.

**Oran**: Genellikle 80% eğitim, 20% test kullanılır.

````python
from sklearn.model_selection import train_test_split

# Stratified split ile veriyi ayır (hedef değişkenin dağılımını koru)
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, 
    y, 
    test_size=0.2,      # %20 test
    random_state=42,    # Tekrar edilebilirlik için
    stratify=y          # Hedef değişken dağılımını koru
)

print("✂️ TRAIN-TEST SPLIT TAMAMLANDI")
print("="*70)
print(f"📊 Toplam Veri Sayısı             : {len(X_encoded):,}")
print(f"📚 Eğitim Seti Boyutu (80%)       : {X_train.shape}")
print(f"🧪 Test Seti Boyutu (20%)         : {X_test.shape}")
print("="*70)

# Hedef değişkenin dağılımını kontrol et
train_dist = y_train.value_counts(normalize=True) * 100
test_dist = y_test.value_counts(normalize=True) * 100

print("\n🎯 HEDEF DEĞİŞKEN DAĞILIMI:")
print("-"*70)
print(f"   Eğitim Seti  - 0 (Hayır): {train_dist[0]:.2f}% | 1 (Evet): {train_dist[1]:.2f}%")
print(f"   Test Seti    - 0 (Hayır): {test_dist[0]:.2f}% | 1 (Evet): {test_dist[1]:.2f}%")
print("-"*70)
print("   ✅ Stratified split sayesinde dağılım korundu!")
print("="*70)
````

### ⚖️ **Feature Scaling (Özellik Ölçeklendirme)**

Farklı ölçeklerdeki değişkenler (örn: yaş 18-98, nr_employed 4963-5228) modelin öğrenmesini olumsuz etkileyebilir. Ölçeklendirme, tüm özellikleri benzer bir aralığa getirir.

**Scaling Yöntemleri:**

1. **StandardScaler**: Verileri ortalama=0, standart sapma=1 olacak şekilde dönüştürür
   - Formül: z = (x - μ) / σ
   - Outlier'lara duyarlı
   
2. **MinMaxScaler**: Verileri [0, 1] aralığına dönüştürür
   - Formül: x' = (x - min) / (max - min)
   - Outlier'lara çok duyarlı

**Lojistik Regresyon** için genellikle **StandardScaler** tercih edilir.

**ÖNEMLİ**: Scaling işlemi SADECE eğitim verisi üzerinden öğrenilmeli, test verisine uygulanmalı (data leakage'i önlemek için).

````python
from sklearn.preprocessing import StandardScaler

# StandardScaler oluştur
scaler = StandardScaler()

# Sadece eğitim verisinden öğren ve dönüştür
X_train_scaled = scaler.fit_transform(X_train)

# Test verisini eğitim setinden öğrenilen parametrelerle dönüştür
X_test_scaled = scaler.transform(X_test)

print("⚖️ FEATURE SCALING TAMAMLANDI")
print("="*70)
print(f"📚 X_train_scaled Boyutu: {X_train_scaled.shape}")
print(f"🧪 X_test_scaled Boyutu : {X_test_scaled.shape}")
print("="*70)

# Ölçeklendirme öncesi ve sonrası karşılaştırma
print("\n📊 ÖLÇEKLENDİRME ÖNCESİ VE SONRASI KARŞILAŞTIRMA:")
print("-"*70)
print("   İlk 5 özellik için istatistikler:\n")

comparison_df = pd.DataFrame({
    'Özellik': X_train.columns[:5],
    'Önce Ort.': X_train.iloc[:, :5].mean().round(2).values,
    'Önce Std.': X_train.iloc[:, :5].std().round(2).values,
    'Sonra Ort.': X_train_scaled[:, :5].mean(axis=0).round(2),
    'Sonra Std.': X_train_scaled[:, :5].std(axis=0).round(2)
})

print(comparison_df.to_string(index=False))
print("-"*70)
print("   ✅ Tüm özellikler ortalama=0, std=1 olarak ölçeklendirildi!")
print("="*70)
````

## 🤖 **Adım 6: Modeling - Lojistik Regresyon**

### 📚 **Lojistik Regresyon Nedir?**

**Lojistik Regresyon**, ikili (binary) sınıflandırma problemleri için kullanılan bir **denetimli öğrenme (supervised learning)** algoritmasıdır. Adında "regresyon" geçmesine rağmen aslında bir **sınıflandırma** algoritmasıdır.

**Kullanım Alanları:**
- Hastalık teşhisi (hasta/sağlıklı)
- Spam tespiti (spam/değil)
- Müşteri churn tahmini (ayrılır/ayrılmaz)
- Kredi riski değerlendirmesi (riskli/değil)
- **Bizim projemiz**: Müşteri term deposit alır mı? (evet/hayır)

### 🧮 **Matematiksel Temel**

#### **1. Lineer Regresyon vs Lojistik Regresyon**

**Lineer Regresyon:**
$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n$$

- Sonuç: $(-\infty, +\infty)$ aralığında herhangi bir değer
- Problem: Olasılık için uygun değil (0-1 arası olmalı)

**Lojistik Regresyon Çözümü:**
$$z = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n$$

Bu lineer kombinasyonu **Sigmoid fonksiyonu** ile [0, 1] aralığına dönüştürürüz:

#### **2. Sigmoid Fonksiyonu**

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

**Özellikler:**
- Girdi: $(-\infty, +\infty)$
- Çıktı: $(0, 1)$ → Olasılık olarak yorumlanabilir
- $z = 0$ için $\sigma(z) = 0.5$
- $z \to +\infty$ için $\sigma(z) \to 1$
- $z \to -\infty$ için $\sigma(z) \to 0$

#### **3. Tahmin Formülü**

$$P(y=1|x) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 x_1 + ... + \beta_n x_n)}}$$

Bu olasılık değerini bir sınıfa dönüştürmek için **threshold (eşik değeri)** kullanırız:

- $P(y=1|x) \geq 0.5$ → Sınıf = 1 (Pozitif)
- $P(y=1|x) < 0.5$ → Sınıf = 0 (Negatif)

#### **4. Cost Function (Maliyet Fonksiyonu)**

Lojistik regresyon **Log Loss (Binary Cross-Entropy)** kullanır:

$$J(\beta) = -\frac{1}{m} \sum_{i=1}^{m} [y_i \log(h(x_i)) + (1-y_i) \log(1-h(x_i))]$$

Burada:
- $m$: Örnek sayısı
- $y_i$: Gerçek etiket (0 veya 1)
- $h(x_i)$: Tahmin edilen olasılık

#### **5. Optimizasyon**

Cost function'ı minimize etmek için **Gradient Descent** veya **Newton-Raphson** gibi optimizasyon algoritmaları kullanılır.

### 📊 **Sigmoid Fonksiyonu Görselleştirme**

````python
# Sigmoid fonksiyonu tanımla
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# z değerleri oluştur
z_values = np.linspace(-10, 10, 200)
sigmoid_values = sigmoid(z_values)

# İnteraktif sigmoid grafiği
fig = go.Figure()

# Sigmoid eğrisi
fig.add_trace(go.Scatter(
    x=z_values,
    y=sigmoid_values,
    mode='lines',
    name='Sigmoid Function',
    line=dict(color='#FF69B4', width=3),
    hovertemplate='z: %{x:.2f}<br>σ(z): %{y:.3f}<extra></extra>'
))

# Threshold çizgisi (0.5)
fig.add_hline(y=0.5, line_dash="dash", line_color="red", 
              annotation_text="Threshold = 0.5", annotation_position="right")

# z=0 dikey çizgisi
fig.add_vline(x=0, line_dash="dash", line_color="gray",
              annotation_text="z = 0", annotation_position="top")

# Decision boundary bölgeleri
fig.add_hrect(y0=0.5, y1=1.0, fillcolor="#98FB98", opacity=0.2, 
              annotation_text="Sınıf 1 (Pozitif)", annotation_position="top right")
fig.add_hrect(y0=0.0, y1=0.5, fillcolor="#FFB6C1", opacity=0.2,
              annotation_text="Sınıf 0 (Negatif)", annotation_position="bottom right")

# Layout
fig.update_layout(
    title="<b>Sigmoid Fonksiyonu: σ(z) = 1 / (1 + e^(-z))</b>",
    title_font=dict(size=18, color='#2F4F4F'),
    xaxis_title="<b>z = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ</b>",
    yaxis_title="<b>σ(z) = P(y=1|x)</b>",
    height=500,
    plot_bgcolor='white',
    paper_bgcolor='#F5F5F5',
    showlegend=True,
    hovermode='x unified'
)

fig.update_xaxes(showgrid=True, gridcolor='lightgray', zeroline=True)
fig.update_yaxes(showgrid=True, gridcolor='lightgray', range=[-0.05, 1.05])

fig.show()

print("✅ Sigmoid fonksiyonu görselleştirildi!")
print("\n💡 ÖNEMLİ NOKTALAR:")
print("   • z = 0 için σ(z) = 0.5 (exact)")
print("   • z > 0 için σ(z) > 0.5 → Sınıf 1")
print("   • z < 0 için σ(z) < 0.5 → Sınıf 0")
print("   • z → +∞ için σ(z) → 1")
print("   • z → -∞ için σ(z) → 0")
````

### 🎓 **Model Eğitimi (Training)**

Şimdi Scikit-learn kütüphanesini kullanarak Lojistik Regresyon modelimizi eğiteceğiz.

**Önemli Parametreler:**

- **C**: Regularization gücünün tersi (inverse). Düşük C = güçlü regularization
- **penalty**: L1 veya L2 regularization ('l1', 'l2', 'elasticnet', 'none')
- **solver**: Optimizasyon algoritması ('lbfgs', 'liblinear', 'saga', 'newton-cg')
- **max_iter**: Maksimum iterasyon sayısı
- **class_weight**: Class imbalance için ağırlık ('balanced' veya None)

**Class Imbalance Çözümü:** Veri setimizde %88.7 negatif, %11.3 pozitif var. `class_weight='balanced'` parametresi ile azınlık sınıfına daha fazla ağırlık vereceğiz.

````python
from sklearn.linear_model import LogisticRegression
import time

print("🎓 LOJİSTİK REGRESYON MODELİ EĞİTİMİ BAŞLADI...")
print("="*70)

# Model oluştur
model = LogisticRegression(
    penalty='l2',              # L2 regularization
    C=1.0,                     # Regularization gücü (default)
    solver='lbfgs',            # Optimizasyon algoritması
    max_iter=1000,             # Maksimum iterasyon
    class_weight='balanced',   # Class imbalance için
    random_state=42            # Tekrar edilebilirlik
)

# Eğitim süresini ölç
start_time = time.time()

# Modeli eğit
model.fit(X_train_scaled, y_train)

end_time = time.time()
training_time = end_time - start_time

print(f"✅ MODEL EĞİTİMİ TAMAMLANDI!")
print("="*70)
print(f"⏱️  Eğitim Süresi              : {training_time:.3f} saniye")
print(f"🔢 Özellik Sayısı             : {model.n_features_in_}")
print(f"📊 Eğitim Veri Sayısı         : {X_train_scaled.shape[0]:,}")
print(f"🎯 Hedef Sınıf Sayısı         : {len(model.classes_)}")
print(f"🔄 Kullanılan İterasyon       : {model.n_iter_[0]}")
print("="*70)

# Eğitim seti üzerinde accuracy
train_accuracy = model.score(X_train_scaled, y_train)
print(f"\n📈 Eğitim Seti Doğruluğu      : {train_accuracy:.4f} ({train_accuracy*100:.2f}%)")

# Test seti üzerinde accuracy
test_accuracy = model.score(X_test_scaled, y_test)
print(f"🧪 Test Seti Doğruluğu         : {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print("="*70)
````

### 🔮 **Tahmin Yapma (Prediction)**

Lojistik regresyon modeli iki tip tahmin yapar:

1. **predict()**: Sınıf etiketi tahmin eder (0 veya 1)

## 📊 **Adım 7: Model Evaluation - Model Değerlendirme**

Modelin performansını sadece accuracy ile değerlendirmek yetersizdir, özellikle dengesiz veri setlerinde. Aşağıdaki metrikleri kullanarak kapsamlı bir değerlendirme yapacağız:

### 🎯 **Değerlendirme Metrikleri**

#### **1. Confusion Matrix (Karmaşıklık Matrisi)**
- **True Positive (TP)**: Doğru tahmin edilen pozitif örnekler
- **True Negative (TN)**: Doğru tahmin edilen negatif örnekler
- **False Positive (FP)**: Yanlış pozitif (Type I Error)
- **False Negative (FN)**: Yanlış negatif (Type II Error)

#### **2. Classification Metrikleri**
- **Accuracy**: Genel doğruluk = (TP + TN) / (TP + TN + FP + FN)
- **Precision**: Kesinlik = TP / (TP + FP) - "Pozitif dediğimizin ne kadarı gerçekten pozitif?"
- **Recall (Sensitivity)**: Duyarlılık = TP / (TP + FN) - "Gerçek pozitiflerin ne kadarını bulduk?"
- **F1-Score**: Precision ve Recall'un harmonik ortalaması = 2 × (Precision × Recall) / (Precision + Recall)

#### **3. ROC-AUC**
- ROC Curve: True Positive Rate vs False Positive Rate grafiği
- AUC (Area Under Curve): 0.5-1.0 arası, yüksek = iyi model

### 📈 **Confusion Matrix ve Classification Report**

````python
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay

# Confusion Matrix hesapla
cm = confusion_matrix(y_test, y_pred)

print("📊 CONFUSION MATRIX")
print("="*70)
print(cm)
print("="*70)

# Metrik değerlerini çıkar
tn, fp, fn, tp = cm.ravel()

print(f"\n📈 CONFUSION MATRIX BILEŞENLERI:")
print("-"*70)
print(f"   True Negative  (TN): {tn:>6,}  ✅ Doğru negatif tahminler")
print(f"   False Positive (FP): {fp:>6,}  ❌ Yanlış pozitif (Type I Error)")
print(f"   False Negative (FN): {fn:>6,}  ❌ Yanlış negatif (Type II Error)")
print(f"   True Positive  (TP): {tp:>6,}  ✅ Doğru pozitif tahminler")
print("-"*70)

# Manuel metrik hesaplama
accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print(f"\n📊 MANUEL HESAPLANAN METRİKLER:")
print("-"*70)
print(f"   Accuracy  : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"   Precision : {precision:.4f} ({precision*100:.2f}%)")
print(f"   Recall    : {recall:.4f} ({recall*100:.2f}%)")
print(f"   F1-Score  : {f1:.4f} ({f1*100:.2f}%)")
print("="*70)

# Sklearn classification report
print(f"\n📋 CLASSIFICATION REPORT (SKLEARN):")
print("-"*70)
print(classification_report(y_test, y_pred, target_names=['Hayır (0)', 'Evet (1)']))
````

### 📉 **ROC Curve ve AUC Score**

**ROC (Receiver Operating Characteristic) Curve**, farklı threshold değerleri için True Positive Rate (TPR) ve False Positive Rate (FPR) değerlerini gösteren bir grafiktir.

- **TPR (Sensitivity/Recall)**: TP / (TP + FN) - Gerçek pozitifleri bulma oranı
- **FPR**: FP / (FP + TN) - Yanlış pozitif oranı
- **AUC (Area Under Curve)**: 0.5-1.0 arası, 1.0'a yakın = mükemmel model

````python
from sklearn.metrics import roc_curve, roc_auc_score

# ROC Curve için FPR, TPR hesapla
y_pred_proba_pos = y_pred_proba[:, 1]  # Pozitif sınıf olasılıkları
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba_pos)

# AUC skorunu hesapla
auc_score = roc_auc_score(y_test, y_pred_proba_pos)

print("📉 ROC-AUC ANALİZİ")
print("="*70)
print(f"📊 AUC Score: {auc_score:.4f}")
print("="*70)

# ROC Curve Görselleştirme
fig = go.Figure()

# ROC Curve
fig.add_trace(go.Scatter(
    x=fpr,
    y=tpr,
    mode='lines',
    name=f'ROC Curve (AUC = {auc_score:.4f})',
    line=dict(color='#FF69B4', width=3),
    hovertemplate='FPR: %{x:.3f}<br>TPR: %{y:.3f}<extra></extra>'
))

# Rastgele tahmin çizgisi (diagonal)
fig.add_trace(go.Scatter(
    x=[0, 1],
    y=[0, 1],
    mode='lines',
    name='Rastgele Tahmin (AUC = 0.50)',
    line=dict(color='gray', width=2, dash='dash'),
    hovertemplate='Rastgele<extra></extra>'
))

# AUC alanını vurgula
fig.add_trace(go.Scatter(
    x=fpr,
    y=tpr,
    fill='tozeroy',
    fillcolor='rgba(255, 182, 193, 0.3)',
    line=dict(width=0),
    showlegend=False,
    hoverinfo='skip'
))

# Layout
fig.update_layout(
    title=f"<b>ROC Curve - AUC Score: {auc_score:.4f}</b>",
    title_font=dict(size=18, color='#2F4F4F'),
    xaxis_title="<b>False Positive Rate (FPR)</b>",
    yaxis_title="<b>True Positive Rate (TPR) / Recall</b>",
    height=600,
    width=700,
    plot_bgcolor='white',
    paper_bgcolor='#F5F5F5',
    showlegend=True,
    legend=dict(x=0.6, y=0.1),
    hovermode='closest'
)

fig.update_xaxes(showgrid=True, gridcolor='lightgray', range=[-0.02, 1.02])
fig.update_yaxes(showgrid=True, gridcolor='lightgray', range=[-0.02, 1.02])

fig.show()

print("\n✅ ROC Curve görselleştirildi!")
print("\n💡 AUC SCORE YORUMLAMA:")
print("-"*70)
print("   • 0.90 - 1.00: Mükemmel")
print("   • 0.80 - 0.90: İyi")
print("   • 0.70 - 0.80: Orta")
print("   • 0.60 - 0.70: Zayıf")
print("   • 0.50 - 0.60: Başarısız")
print("-"*70)
print(f"   Modelimizin AUC'si: {auc_score:.4f} → İYİ SEVİYEDE!")
print("="*70)
````

````python
# Katsayıları çıkar
coefficients = model.coef_[0]
feature_names = X_encoded.columns

# DataFrame oluştur
feature_importance_df = pd.DataFrame({
    'Özellik': feature_names,
    'Katsayı': coefficients,
    'Mutlak Değer': np.abs(coefficients)
}).sort_values('Mutlak Değer', ascending=False)

# En önemli 15 özellik
top_15 = feature_importance_df.head(15)

print("🔑 FEATURE IMPORTANCE ANALİZİ")
print("="*70)
print(f"📊 Toplam Özellik Sayısı: {len(feature_importance_df)}")
print("="*70)

print("\n📈 EN ÖNEMLİ 15 ÖZELLİK:")
print("-"*70)
print(f"{'Sıra':<5} {'Özellik':<35} {'Katsayı':>12} {'Etki':>10}")
print("-"*70)

for idx, row in top_15.iterrows():
    effect = "POZITIF ↑" if row['Katsayı'] > 0 else "NEGATIF ↓"
    print(f"{feature_importance_df.index.tolist().index(idx)+1:<5} {row['Özellik']:<35} {row['Katsayı']:>12.4f} {effect:>10}")

print("="*70)

# Görselleştirme: En önemli 15 özellik
fig = go.Figure()

colors = ['#98FB98' if c > 0 else '#FFB6C1' for c in top_15['Katsayı']]

fig.add_trace(go.Bar(
    y=top_15['Özellik'][::-1],  # Ters sıralama (en büyük üstte)
    x=top_15['Katsayı'][::-1],
    orientation='h',
    marker=dict(color=colors[::-1]),
    text=top_15['Katsayı'][::-1].round(3),
    textposition='outside',
    hovertemplate='%{y}<br>Katsayı: %{x:.4f}<extra></extra>'
))

fig.update_layout(
    title="<b>En Önemli 15 Özellik (Katsayılarına Göre)</b>",
    title_font=dict(size=18, color='#2F4F4F'),
    xaxis_title="<b>Katsayı Değeri</b>",
    yaxis_title="<b>Özellik Adı</b>",
    height=700,
    plot_bgcolor='white',
    paper_bgcolor='#F5F5F5',
    showlegend=False
)

fig.add_vline(x=0, line_width=2, line_color="black")
fig.update_xaxes(showgrid=True, gridcolor='lightgray', zeroline=True)
fig.update_yaxes(showgrid=False)

fig.show()

print("\n✅ Feature importance görselleştirildi!")
print("\n💡 YORUMLAMA:")
print("-"*70)
print("   🟢 Yeşil (Pozitif): Özellik arttıkça 'Evet' olasılığı artar")
print("   🔴 Pembe (Negatif): Özellik arttıkça 'Hayır' olasılığı artar")
print("="*70)
````

## 💾 **Adım 8: Model Kaydetme (Model Saving)**

Eğitilmiş modeli production ortamında kullanmak için kaydetmeliyiz. Python'da iki popüler yöntem vardır:

### **1. Pickle (Python Standart Kütüphanesi)**
- Python'un built-in modülü
- Herhangi bir Python objesini serialize edebilir
- Dosya uzantısı: `.pkl` veya `.pickle`

### **2. Joblib (Scikit-learn Tavsiyesi)**
- Büyük NumPy array'leri için daha verimli
- Scikit-learn modelleri için özellikle optimize edilmiş
- Dosya uzantısı: `.joblib` veya `.pkl`

**Production için gerekli objeler:**
- ✅ Eğitilmiş model (LogisticRegression)
- ✅ Scaler objesi (StandardScaler)
- ✅ Feature names (sütun isimleri)

````python
import pickle
import joblib
import os

# Model klasörü oluştur
model_dir = 'models'
os.makedirs(model_dir, exist_ok=True)

print("💾 MODEL KAYDETME İŞLEMİ BAŞLADI...")
print("="*70)

# 1. Pickle ile kaydetme
pickle_model_path = os.path.join(model_dir, 'logistic_regression_model.pkl')
pickle_scaler_path = os.path.join(model_dir, 'scaler.pkl')

with open(pickle_model_path, 'wb') as f:
    pickle.dump(model, f)
    
with open(pickle_scaler_path, 'wb') as f:
    pickle.dump(scaler, f)

print(f"✅ Pickle ile kaydedildi:")
print(f"   Model  : {pickle_model_path}")
print(f"   Scaler : {pickle_scaler_path}")
print()

# 2. Joblib ile kaydetme
joblib_model_path = os.path.join(model_dir, 'logistic_regression_model.joblib')
joblib_scaler_path = os.path.join(model_dir, 'scaler.joblib')

joblib.dump(model, joblib_model_path)
joblib.dump(scaler, joblib_scaler_path)

print(f"✅ Joblib ile kaydedildi:")
print(f"   Model  : {joblib_model_path}")
print(f"   Scaler : {joblib_scaler_path}")
print()

# 3. Feature names kaydetme
feature_names_path = os.path.join(model_dir, 'feature_names.pkl')
with open(feature_names_path, 'wb') as f:
    pickle.dump(list(X_encoded.columns), f)

print(f"✅ Feature names kaydedildi:")
print(f"   Features: {feature_names_path}")
print("="*70)

# Dosya boyutlarını kontrol et
pickle_size = os.path.getsize(pickle_model_path) / 1024  # KB
joblib_size = os.path.getsize(joblib_model_path) / 1024  # KB

print(f"\n📊 DOSYA BOYUTLARI:")
print("-"*70)
print(f"   Pickle Model : {pickle_size:.2f} KB")
print(f"   Joblib Model : {joblib_size:.2f} KB")
print("-"*70)
print("="*70)
````
